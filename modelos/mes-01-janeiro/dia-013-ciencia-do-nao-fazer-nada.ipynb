### 📖 A História

Sexta-feira à tarde, a culpa aparece. Aquela sensação de que o tempo livre devia ser usado para alguma coisa — organizar a casa, adiantar trabalho, fazer algo que pareça produtivo. Ficar olhando para o nada parece desperdício.

A ciência sugere o oposto.

Pesquisadores da Universidade da Califórnia em Santa Barbara conduziram um experimento simples: pediram para pessoas resolverem um problema clássico de criatividade, depois as separaram em grupos diferentes durante um intervalo, e mediram quem voltava com soluções mais criativas. Quem teve uma pausa que permitia a mente vagar livremente — sem nenhuma exigência cognitiva — teve desempenho muito melhor do que quem ficou ocupado, descansou em silêncio, ou nem teve pausa alguma.

Não fazer nada, especificamente, não é o problema. É a solução.

---

### 📚 O Conceito: Incubação Criativa e Mind-Wandering

A "mente vagando" (*mind-wandering*) é definida como o deslocamento da atenção de uma tarefa externa para pensamentos internos não relacionados a ela. Estudos de larga escala — incluindo o trabalho de Killingsworth e Gilbert (2010), já abordado em outro dia deste projeto — estimam que isso ocupa entre **30% e 50% do tempo em que estamos acordados**.

O experimento de Baird et al. (2012), publicado na *Psychological Science*, testou algo específico: será que esse estado mental, normalmente associado a distração e até a menor felicidade, tem alguma função construtiva? A resposta foi sim — mas só sob uma condição precisa. Não é qualquer pausa que ajuda. É especificamente a pausa que permite a mente vagar **sem esforço cognitivo nenhum** — nem uma tarefa difícil, nem o silêncio absoluto do descanso passivo, nem continuar trabalhando sem parar.

### 🧮 O Modelo
**Fontes:**
- Baird, B., Smallwood, J., Mrazek, M.D., Kam, J.W.Y., Franklin, M.S. & Schooler, J.W. (2012) — *Inspired by Distraction: Mind Wandering Facilitates Creative Incubation* — Psychological Science, 23(10) — N=145 participantes
- Killingsworth, M.A. & Gilbert, D.T. (2010) — *A Wandering Mind Is an Unhappy Mind* — Science, 330(6006) — N=15.000 (já utilizado no Dia #007 deste projeto)


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print('✅ Bibliotecas carregadas')
```

```python
# --- DADOS DA LITERATURA ---

# Baird et al. 2012 — Psychological Science — N=145 (35 homens, 110 mulheres, 19-32 anos)
n_baird = 145

# Melhoria percentual no teste de criatividade (UUT) para problemas de exposicao repetida
# por condicao de incubacao (valores aproximados extraidos do estudo original, Figura 1)
melhoria_undemanding = 0.32   # tarefa pouco exigente — permite mind-wandering
melhoria_demanding    = -0.03  # tarefa exigente — pouco mind-wandering
melhoria_rest         = 0.02   # descanso passivo (so sentar e nao fazer nada estimulado)
melhoria_no_break     = 0.05   # sem pausa nenhuma

# Mind-wandering — proporcao do tempo acordado (faixa da literatura, incluindo Killingsworth & Gilbert 2010)
p_mind_wandering_min = 0.30
p_mind_wandering_max = 0.50
n_killingsworth = 15000  # ja utilizado no Dia #007 deste projeto

fator_correcao = 0.80

print('=' * 65)
print('  DADOS DA LITERATURA — INCUBAÇÃO CRIATIVA E MIND-WANDERING')
print('=' * 65)
print(f'\n  Baird et al. 2012 (N={n_baird}):')
print(f'  → Melhoria criativa — tarefa pouco exigente : {melhoria_undemanding*100:+.0f}%')
print(f'  → Melhoria criativa — tarefa exigente       : {melhoria_demanding*100:+.0f}%')
print(f'  → Melhoria criativa — descanso passivo      : {melhoria_rest*100:+.0f}%')
print(f'  → Melhoria criativa — sem pausa              : {melhoria_no_break*100:+.0f}%')
print(f'\n  Mind-wandering — faixa da literatura:')
print(f'  → Proporção do tempo acordado : {p_mind_wandering_min*100:.0f}% a {p_mind_wandering_max*100:.0f}%')
print(f'\n  Fator de correção aplicado: x{fator_correcao}')
print('=' * 65)
```

```python
# --- O MODELO ---
# Nota metodologica: o N do estudo Baird et al. (145) e pequeno mas real e documentado
# diretamente no paper original — diferente de estimativas de outros dias, aqui o N
# vem com precisao da fonte primaria.

# Vantagem da tarefa pouco exigente sobre as demais condicoes
vantagem_vs_demanding = melhoria_undemanding - melhoria_demanding
vantagem_vs_rest      = melhoria_undemanding - melhoria_rest
vantagem_vs_no_break  = melhoria_undemanding - melhoria_no_break

# Distribuicao Beta para a faixa de mind-wandering (usando o ponto central da faixa)
p_mw_central = (p_mind_wandering_min + p_mind_wandering_max) / 2
alpha_mw = p_mw_central * n_killingsworth
beta_mw  = (1 - p_mw_central) * n_killingsworth
dist_mw  = stats.beta(alpha_mw, beta_mw)
ic_mw    = dist_mw.interval(0.95)

p_mw_corrigido = p_mw_central * fator_correcao

print('=' * 65)
print('  MODELO — RESULTADO')
print('=' * 65)
print(f'  Vantagem da tarefa pouco exigente sobre:')
print(f'  → Tarefa exigente : +{vantagem_vs_demanding*100:.0f}pp de melhoria criativa')
print(f'  → Descanso passivo: +{vantagem_vs_rest*100:.0f}pp de melhoria criativa')
print(f'  → Sem pausa        : +{vantagem_vs_no_break*100:.0f}pp de melhoria criativa')
print()
print(f'  Mind-wandering — ponto central da faixa: {p_mw_central*100:.0f}%')
print(f'  → IC 95% (N={n_killingsworth:,}): [{ic_mw[0]*100:.1f}%, {ic_mw[1]*100:.1f}%]')
print(f'  → Corrigido (x0.80): {p_mw_corrigido*100:.1f}%')
print('=' * 65)
```

```python
# --- VISUALIZACAO ---

# GRAFICO 1 — Melhoria criativa por tipo de pausa
fig1, ax1 = plt.subplots(figsize=(12, 8))

condicoes = ['Tarefa pouco\nexigente\n(mente vagando)', 'Sem pausa', 'Descanso\npassivo', 'Tarefa\nexigente']
melhorias = [melhoria_undemanding * 100, melhoria_no_break * 100,
             melhoria_rest * 100, melhoria_demanding * 100]
cores1 = ['#2a8a82', '#888888', '#c8a84b', '#c0392b']

bars = ax1.bar(condicoes, melhorias, color=cores1, alpha=0.85, width=0.55)
ax1.axhline(y=0, color='#333333', linewidth=1)
ax1.set_ylabel('Melhoria no teste de criatividade (%)', fontsize=12)
ax1.set_title('Não fazer nada de exigente é o que realmente ajuda a criar\nBaird et al. (2012), Psychological Science — N=145',
              fontsize=13, pad=15)

for bar, m in zip(bars, melhorias):
    offset = 1.5 if m >= 0 else -2.5
    ax1.text(bar.get_x() + bar.get_width() / 2, m + offset,
             f'{m:+.0f}%', ha='center', fontweight='bold', fontsize=13)

plt.tight_layout()
plt.savefig('dia-013-grafico-01-condicoes.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Grafico 1 salvo!')


# GRAFICO 2 — Vantagem da mente vagando sobre as demais condicoes
fig2, ax2 = plt.subplots(figsize=(12, 8))

comparacoes = ['vs. Tarefa\nexigente', 'vs. Descanso\npassivo', 'vs. Sem\npausa']
vantagens = [vantagem_vs_demanding * 100, vantagem_vs_rest * 100, vantagem_vs_no_break * 100]
cores2 = ['#2a8a82', '#2a8a82', '#2a8a82']

bars2 = ax2.barh(comparacoes, vantagens, color=cores2, alpha=0.85, height=0.5)
ax2.set_xlabel('Vantagem da mente vagando (pontos percentuais)', fontsize=12)
ax2.set_title('A pausa que permite a mente vagar vence todas as outras\nBaird et al. (2012) — N=145',
              fontsize=13, pad=15)

for bar, v in zip(bars2, vantagens):
    ax2.text(bar.get_width() + 0.5, bar.get_y() + bar.get_height() / 2,
             f'+{v:.0f}pp', va='center', fontweight='bold', fontsize=13)

plt.tight_layout()
plt.savefig('dia-013-grafico-02-vantagem.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Grafico 2 salvo!')


# GRAFICO 3 — Distribuicao Beta (mind-wandering)
fig3, ax3 = plt.subplots(figsize=(12, 8))

x = np.linspace(ic_mw[0] * 0.998, ic_mw[1] * 1.002, 1000)
y = dist_mw.pdf(x)

ax3.plot(x * 100, y, color='#2a8a82', linewidth=3)
ax3.fill_between(x * 100, y, alpha=0.2, color='#2a8a82')
ax3.axvline(x=ic_mw[0] * 100, color='#c8a84b', linestyle='--', linewidth=2,
            label=f'IC 95%: [{ic_mw[0]*100:.1f}%, {ic_mw[1]*100:.1f}%]')
ax3.axvline(x=ic_mw[1] * 100, color='#c8a84b', linestyle='--', linewidth=2)
ax3.axvline(x=p_mw_central * 100, color='#2a8a82', linewidth=2,
            label=f'Ponto central da faixa: {p_mw_central*100:.0f}%')
ax3.axvline(x=p_mw_corrigido * 100, color='#c0392b', linestyle='-.',
            linewidth=2, label=f'Corrigido (x0.80): {p_mw_corrigido*100:.1f}%')

ax3.set_xlabel('Proporção do tempo acordado com a mente vagando (%)', fontsize=12)
ax3.set_ylabel('Densidade', fontsize=12)
ax3.set_title('Distribuição Beta — Mind-wandering no dia a dia\nKillingsworth & Gilbert (2010) — N=15.000',
              fontsize=13, pad=15)
ax3.legend(fontsize=11)

plt.tight_layout()
plt.savefig('dia-013-grafico-03-bayesiano.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Grafico 3 salvo!')
```

### 💡 O Insight

Num experimento controlado com 145 participantes, quem teve uma pausa com tarefa pouco exigente — que permitia a mente vagar livremente — teve uma melhoria de **32% no desempenho criativo**. Quem ficou sem pausa, descansou em silêncio passivo, ou fez uma tarefa cognitivamente exigente, não teve esse mesmo ganho — em alguns casos, o desempenho até caiu.

A vantagem da pausa certa chegou a **35 pontos percentuais** acima da tarefa exigente, no mesmo experimento.

Isso acontece porque a mente, durante 30% a 50% do tempo acordado, já está naturalmente "vagando" — segundo a pesquisa de Killingsworth e Gilbert (2010). O que o experimento de Baird mostra é que esse estado, normalmente tratado como distração ou perda de tempo, tem uma função real: deixar associações inconscientes trabalharem em problemas que a mente consciente já vinha tentando resolver sem sucesso.

Sexta-feira à tarde, a mente vagando sem culpa não é desperdício. É, possivelmente, o momento exato em que uma solução que você procurava sem encontrar finalmente aparece.

*Da próxima vez que sentir vontade de simplesmente não fazer nada — você vai se permitir, ou vai preencher o tempo com mais uma tarefa?*

### ⚠️ Limitações do Modelo
- O estudo de Baird et al. (2012) usou uma amostra pequena (N=145) de estudantes universitários de uma única instituição — generalização para outras populações requer cautela
- O efeito foi observado especificamente para problemas de criatividade já apresentados antes da pausa (exposição repetida) — não houve o mesmo efeito para problemas completamente novos
- A medida de mind-wandering (30%-50% do tempo) é uma faixa consolidada de múltiplos estudos com metodologias distintas, não um número único e preciso
- O fator de correção (×0,80) foi aplicado apenas à estimativa de mind-wandering, já que os dados do experimento de Baird são resultados experimentais diretos, não estimativas populacionais a serem corrigidas
- Os valores de melhoria percentual por condição foram extraídos visualmente do gráfico original do estudo (Figura 1), não de uma tabela numérica explícita no texto
