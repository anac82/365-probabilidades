## Qual a probabilidade de "não tenho tempo" ser a sua desculpa para não se exercitar?

**Tipo:** Comportamental
**Data de publicação:** 2026-10-26
**Ferramenta:** Python
**Decisão analisada:** A falta de tempo é a sua desculpa?
**Hashtag:** #365Probabilidades #Dia135

---

### 📖 A História

[UMA LINHA DA ANA]

Todo mundo tem a frase pronta. "Não tenho tempo." Ela serve para a academia, para a
corrida, para a caminhada que ia começar na segunda. É a desculpa mais aceita que existe,
porque ninguém discute com ela.

Hoje começa uma trilha de dias sobre tempo, que vai até o fim do ano. E a primeira
pergunta é a mais simples: quando alguém no Brasil explica por que não se exercita, a
resposta é "não tenho tempo"?

---

### 📚 O Conceito

A epidemiologia chama de **barreira percebida** o motivo que a própria pessoa dá para não
fazer algo. Não é a causa medida; é a explicação declarada. Isso importa para a conta de
hoje por dois motivos. O primeiro é que a barreira percebida é autorrelato, e por isso o
projeto aplica o fator de correção ×0,80. O segundo é que uma barreira pode ser muito
citada e pouco associada à inatividade, ou o contrário; o estudo testa as duas coisas
separadamente.

A referência de atividade física é a da OMS: 150 minutos por semana de atividade
moderada no tempo livre. Em uma semana de 168 horas, 150 minutos são 1,5% do tempo.

---

### 🧮 O Modelo

**Fonte central:**
- **Reichert, F. F., Barros, A. J. D., Domingues, M. R. & Hallal, P. C. (2007).** The role
  of perceived personal barriers to engagement in leisure-time physical activity.
  *American Journal of Public Health*, 97(3), 515-519. Pelotas (RS), 2003, amostragem em
  estágios, **N = 3.100 adultos de 20 anos ou mais**, IPAQ longo, oito barreiras.
  Dissertação de mestrado correspondente (UFPel, 2004) para os percentuais das barreiras.

**Fonte de contraste:**
- **Cruz, M. S., Garcia, L. M. T., Andrade, D. R. & Florindo, A. A. (2018).** Barreiras à
  prática de atividade física no tempo de lazer em adultos de área de baixo nível
  socioeconômico. *Revista Brasileira de Atividade Física & Saúde*, 23, e0041. Ermelino
  Matarazzo (SP), 2007, **N = 889 adultos**; barreiras perguntadas só a quem era inativo
  no lazer.

**Modelo próprio:**
- Posterior Beta (Jeffreys) para cada barreira de Pelotas e comparação entre elas por
  simulação (semente 42): qual é a desculpa mais citada, com que certeza.
- O mesmo para São Paulo, com o denominador diferente declarado.
- A sala de 30: quantas pessoas, numa sala sorteada, dizem que o problema é tempo.
- A conta de bolso dos 150 minutos.
- **×0,80 sim**, nas proporções de barreira (autorrelato). O 26,8% de ativos também é
  autorrelato (IPAQ) e recebe o fator na tabela, declarado.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

VERMELHO, DOURADO, VERDE, CINZA = '#c0392b', '#c8a84b', '#1a5f5a', '#6b6a64'
SEMENTE = 42
ger = np.random.default_rng(SEMENTE)

def vir(x, casas=1):
    return f"{x:.{casas}f}".replace(".", ",")

def pct(x, casas=1):
    return vir(100 * x, casas) + "%"

def mil(n):
    return f"{int(round(n)):,}".replace(",", ".")

def beta_j(k, n):
    d = stats.beta(k + 0.5, n - k + 0.5)
    return d, d.mean(), d.interval(0.95)

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
FATOR = 0.80

# Reichert, Barros, Domingues & Hallal (2007), AJPH 97(3). Pelotas, 2003.
N_PEL = 3100
P_ATIVOS_PEL = 0.268          # cumprem 150 min/semana no lazer (IPAQ)
P_ALGUMA_BARREIRA = 0.85      # relatam ao menos uma das oito barreiras
BARREIRAS_PEL = {             # percentuais publicados (as outras quatro não estão em fonte aberta)
    "falta de dinheiro": 0.403,
    "cansaço": 0.381,
    "falta de companhia": 0.322,
    "falta de tempo": 0.315,
}
# No ajuste, cinco das oito barreiras ficaram associadas à inatividade:
ASSOCIADAS_PEL = ["não gostar de exercício", "cansaço", "falta de dinheiro", "falta de companhia", "falta de tempo"]

# Cruz, Garcia, Andrade & Florindo (2018), RBAFS 23:e0041. Ermelino Matarazzo (SP), 2007.
N_SP = 889
P_INATIVOS_SP = 0.685         # inativos no lazer
BARREIRAS_SP = {              # perguntadas só aos inativos; percentuais ponderados
    "falta de tempo": 0.397,
    "falta de vontade": 0.184,
    "não gosta de exercício": 0.063,
    "problema de saúde": 0.056,
    "falta de dinheiro": 0.036,
}
IC_TEMPO_SP = (0.338, 0.459)  # IC 95% publicado
N_INATIVOS_SP = int(round(N_SP * P_INATIVOS_SP))

MIN_OMS = 150                 # minutos por semana
MIN_SEMANA = 7 * 24 * 60

print("=" * 70)
print("  DADOS DA LITERATURA — A DESCULPA PARA NÃO SE EXERCITAR")
print("=" * 70)
print(f"\n  Pelotas, 2003 (Reichert et al. 2007), N = {mil(N_PEL)} adultos de 20+:")
print(f"  → cumprem 150 min/semana no lazer:   {pct(P_ATIVOS_PEL)}")
print(f"  → relatam ao menos uma barreira:      {pct(P_ALGUMA_BARREIRA, 0)}")
for k, v in BARREIRAS_PEL.items():
    print(f"  → {k:<22s} {pct(v)}")
print(f"  → associadas à inatividade no ajuste: {', '.join(ASSOCIADAS_PEL)}")
print(f"\n  São Paulo, 2007 (Cruz et al. 2018), N = {N_SP}, barreiras só entre os {pct(P_INATIVOS_SP)} inativos:")
for k, v in BARREIRAS_SP.items():
    print(f"  → {k:<22s} {pct(v)}")
print(f"  → falta de tempo, IC 95% publicado:    [{pct(IC_TEMPO_SP[0])}; {pct(IC_TEMPO_SP[1])}]")
print(f"\n  Referência da OMS: {MIN_OMS} min/semana = {pct(MIN_OMS / MIN_SEMANA)} da semana")
print("=" * 70)
```

```python
# --- O MODELO ---

# 1. Posterior de cada desculpa em Pelotas, com e sem o fator
post = {}
for k, p in BARREIRAS_PEL.items():
    d, m, ic = beta_j(round(p * N_PEL), N_PEL)
    post[k] = {"dist": d, "media": m, "ic": ic, "corrigida": m * FATOR, "ic_corr": (ic[0] * FATOR, ic[1] * FATOR)}
d_tempo, m_tempo, ic_tempo = beta_j(round(BARREIRAS_PEL["falta de tempo"] * N_PEL), N_PEL)

# 2. Qual é a desculpa mais citada? Simulação das posteriores (semente 42)
SIMS = 200000
amostras = {k: v["dist"].rvs(SIMS, random_state=ger) for k, v in post.items()}
mat = np.column_stack([amostras[k] for k in BARREIRAS_PEL])
nomes = list(BARREIRAS_PEL)
vencedora = np.array(nomes)[mat.argmax(axis=1)]
p_mais_citada = {k: (vencedora == k).mean() for k in nomes}
p_dinheiro_acima_tempo = (amostras["falta de dinheiro"] > amostras["falta de tempo"]).mean()
p_tempo_acima_companhia = (amostras["falta de tempo"] > amostras["falta de companhia"]).mean()

# 3. São Paulo: posterior sobre os inativos (n efetivo = 609, aproximação declarada)
d_sp, m_sp, ic_sp = beta_j(round(BARREIRAS_SP["falta de tempo"] * N_INATIVOS_SP), N_INATIVOS_SP)
# tradução para "entre todos os adultos": só quem é inativo foi perguntado
p_tempo_sp_todos = P_INATIVOS_SP * BARREIRAS_SP["falta de tempo"]

# 4. A sala de 30 (Pelotas, proporção corrigida)
p_sala = m_tempo * FATOR
sala = stats.binom(30, p_sala)
esperado_sala = sala.mean()
ic_sala = sala.interval(0.95)
p_ao_menos_10 = 1 - sala.cdf(9)
# o mesmo com a proporção de SP entre inativos
sala_sp = stats.binom(30, m_sp * FATOR)

# 5. A conta de bolso: 150 minutos na semana
fracao_semana = MIN_OMS / MIN_SEMANA
horas_acordado = 7 * 16  # premissa declarada: 8 h de sono
fracao_acordado = (MIN_OMS / 60) / horas_acordado

print("=" * 70)
print("  MODELO — A DESCULPA PARA NÃO SE EXERCITAR")
print("=" * 70)
print(f"\n  1. Pelotas, posterior Beta (Jeffreys), N = {mil(N_PEL)}:")
for k in nomes:
    v = post[k]
    print(f"     {k:<20s} {pct(v['media'])} [{pct(v['ic'][0])}; {pct(v['ic'][1])}]  ×0,80 → {pct(v['corrigida'])} [{pct(v['ic_corr'][0])}; {pct(v['ic_corr'][1])}]")
print(f"\n  2. Qual é a desculpa mais citada ({mil(SIMS)} sorteios das posteriores):")
for k in nomes:
    print(f"     {k:<20s} {pct(p_mais_citada[k])}")
print(f"     P(dinheiro acima de tempo) = {pct(p_dinheiro_acima_tempo)}; P(tempo acima de companhia) = {pct(p_tempo_acima_companhia)}")
print(f"\n  3. São Paulo, falta de tempo entre os inativos (n ≈ {N_INATIVOS_SP}):")
print(f"     posterior {pct(m_sp)} [{pct(ic_sp[0])}; {pct(ic_sp[1])}]  (publicado [{pct(IC_TEMPO_SP[0])}; {pct(IC_TEMPO_SP[1])}])  ×0,80 → {pct(m_sp * FATOR)}")
print(f"     entre todos os adultos de lá (inativo E cita tempo): {pct(p_tempo_sp_todos)}")
print(f"\n  4. A sala de 30, com {pct(p_sala)} dizendo que é tempo:")
print(f"     esperado {vir(esperado_sala)} pessoas, IC 95% [{int(ic_sala[0])}; {int(ic_sala[1])}]; P(10 ou mais) = {pct(p_ao_menos_10)}")
print(f"     com a proporção de SP entre inativos: esperado {vir(sala_sp.mean())}")
print(f"\n  5. 150 minutos: {pct(fracao_semana)} da semana; {pct(fracao_acordado)} das horas acordado (8 h de sono, premissa)")
print("=" * 70)
```

```python
# --- VISUALIZAÇÃO ---

# ══ GRÁFICO 1 · as quatro desculpas de Pelotas ══
fig1, ax1 = plt.subplots(figsize=(11, 6))
ordem = sorted(nomes, key=lambda k: post[k]["media"], reverse=True)
ys = np.arange(len(ordem))[::-1]
for y, k in zip(ys, ordem):
    v = post[k]; cor = VERMELHO if k == "falta de tempo" else CINZA
    ax1.barh(y, v["media"] * 100, color=cor, alpha=0.9, height=0.6)
    ax1.errorbar(v["media"] * 100, y, xerr=[[(v["media"] - v["ic"][0]) * 100], [(v["ic"][1] - v["media"]) * 100]], fmt="none", ecolor="#222222", capsize=4, lw=1.4)
    ax1.text(v["ic"][1] * 100 + 0.6, y, f"{pct(v['media'])}  [{pct(v['ic'][0])}; {pct(v['ic'][1])}]", va="center", fontsize=11, color="#222222")
ax1.set_yticks(ys); ax1.set_yticklabels(ordem, fontsize=12)
ax1.set_xlim(0, 56); ax1.set_xlabel("adultos que citam a barreira (%)")
ax1.set_title("Em Pelotas, a desculpa mais citada para não se exercitar é dinheiro, não tempo\n"
              f"falta de tempo é a quarta, {pct(post['falta de tempo']['media'])}; N = {mil(N_PEL)}, posterior Beta (Jeffreys), IC 95%", fontsize=12.5, pad=14)
plt.figtext(0.5, 0.005, "Fonte: Reichert, Barros, Domingues & Hallal (2007), AJPH 97(3); percentuais da dissertação (UFPel, 2004)  |  #365Probabilidades",
            ha="center", fontsize=8.5, color="gray")
plt.tight_layout(rect=[0, 0.03, 1, 1])
plt.savefig("dia-135-grafico-01-as-desculpas.png", dpi=150, bbox_inches="tight"); plt.close()
print("✅ Gráfico 1 salvo!")

# ══ GRÁFICO 2 · duas cidades, dois denominadores ══
fig2, ax2 = plt.subplots(figsize=(11, 6))
xs = np.arange(3)
vals = [m_tempo * 100, m_sp * 100, p_tempo_sp_todos * 100]
errs = [[(m_tempo - ic_tempo[0]) * 100, (m_sp - ic_sp[0]) * 100, 0], [(ic_tempo[1] - m_tempo) * 100, (ic_sp[1] - m_sp) * 100, 0]]
ax2.bar(xs, vals, color=[VERMELHO, DOURADO, CINZA], alpha=0.9, width=0.55)
ax2.errorbar(xs[:2], vals[:2], yerr=[errs[0][:2], errs[1][:2]], fmt="none", ecolor="#222222", capsize=5, lw=1.4)
for x, v, e in zip(xs, vals, errs[1]):
    ax2.text(x, v + e + 1.5, pct(v / 100), ha="center", fontsize=13, fontweight="bold", color="#222222")
ax2.set_xticks(xs); ax2.set_xticklabels(["Pelotas 2003\ntodos os adultos\n(N = 3.100)", "São Paulo 2007\nsó os inativos\n(n ≈ 609)", "São Paulo 2007\ntraduzido para\ntodos os adultos"], fontsize=10.5)
ax2.set_ylim(0, 52); ax2.set_ylabel("citam falta de tempo (%)")
ax2.set_title("A mesma desculpa em duas cidades: um em três, ou quatro em dez entre quem não se exercita\n"
              "a diferença é o denominador, e por isso as barras não se comparam ponto a ponto", fontsize=12.5, pad=14)
plt.figtext(0.5, 0.005, "Fontes: Reichert et al. (2007); Cruz, Garcia, Andrade & Florindo (2018), RBAFS 23:e0041  |  #365Probabilidades",
            ha="center", fontsize=8.5, color="gray")
plt.tight_layout(rect=[0, 0.03, 1, 1])
plt.savefig("dia-135-grafico-02-duas-cidades.png", dpi=150, bbox_inches="tight"); plt.close()
print("✅ Gráfico 2 salvo!")

# ══ GRÁFICO 3 · a sala de 30 ══
fig3, ax3 = plt.subplots(figsize=(11, 6))
ks = np.arange(0, 31)
ax3.bar(ks, sala.pmf(ks) * 100, color=VERMELHO, alpha=0.9, label=f"Pelotas, {pct(p_sala)} corrigido")
ax3.bar(ks + 0.0, sala_sp.pmf(ks) * 100, color=DOURADO, alpha=0.45, label=f"São Paulo, inativos, {pct(m_sp * FATOR)} corrigido")
ax3.axvline(esperado_sala, color="#222222", lw=1.4, ls="--")
ax3.text(16.5, max(sala.pmf(ks)) * 100 * 0.62, f"Pelotas: esperado {vir(esperado_sala)}\nIC 95% [{int(ic_sala[0])}; {int(ic_sala[1])}]\nP(10 ou mais) = {pct(p_ao_menos_10)}", fontsize=10.5, color="#222222")
ax3.set_xlim(-0.5, 24.5); ax3.set_xlabel("pessoas, em 30, que dizem que o problema é tempo"); ax3.set_ylabel("probabilidade (%)")
ax3.legend(fontsize=10, loc="upper right")
ax3.set_title(f"Numa sala de 30 pessoas sorteadas, {vir(esperado_sala)} dizem que não se exercitam por falta de tempo\n"
              f"P(10 ou mais) = {pct(p_ao_menos_10)}; binomial sobre a proporção corrigida", fontsize=12.5, pad=14)
plt.figtext(0.5, 0.005, "Modelo próprio, binomial(30, p) com p = proporção corrigida ×0,80  |  #365Probabilidades", ha="center", fontsize=8.5, color="gray")
plt.tight_layout(rect=[0, 0.03, 1, 1])
plt.savefig("dia-135-grafico-03-a-sala.png", dpi=150, bbox_inches="tight"); plt.close()
print("✅ Gráfico 3 salvo!")

# ══ GRÁFICO 4 · 150 minutos na semana ══
fig4, ax4 = plt.subplots(figsize=(11, 3.8))
ax4.barh([0], [168], color="#e6e3dc", height=0.5)
ax4.barh([0], [56], color=CINZA, alpha=0.6, height=0.5, label="sono, 8 h por noite (premissa)")
ax4.barh([0], [MIN_OMS / 60], left=[56], color=VERMELHO, height=0.5, label=f"150 minutos da OMS = {pct(fracao_semana)} da semana")
ax4.set_ylim(-0.45, 0.55)
ax4.text(56 + 1.25, 0.31, "2,5 h", ha="center", fontsize=11, fontweight="bold", color=VERMELHO)
ax4.text(28, 0, "56 h de sono", ha="center", va="center", fontsize=10.5, color="white")
ax4.text(113, 0, "109,5 h acordado, fora os 150 minutos", ha="center", va="center", fontsize=10.5, color="#444444")
ax4.set_xlim(0, 168); ax4.set_yticks([]); ax4.set_xlabel("horas da semana")
ax4.legend(fontsize=9.5, loc="lower right", ncol=2)
ax4.set_title(f"A conta de bolso: 150 minutos são {pct(fracao_semana)} da semana e {pct(fracao_acordado)} das horas acordado", fontsize=12.5, pad=12)
plt.figtext(0.5, 0.01, "Referência da OMS para atividade moderada no lazer; sono de 8 h é premissa declarada  |  #365Probabilidades", ha="center", fontsize=8.5, color="gray")
plt.tight_layout(rect=[0, 0.05, 1, 1])
plt.savefig("dia-135-grafico-04-a-semana.png", dpi=150, bbox_inches="tight"); plt.close()
print("✅ Gráfico 4 salvo!")
```

### 💡 O Insight

**Um em cada três adultos diz que não se exercita por falta de tempo. E não é a desculpa
mais comum.**

Em Pelotas, com 3.100 pessoas, a falta de tempo foi a quarta barreira citada, 31,5%, que
o fator de correção do projeto leva a 25%. Na frente dela vieram dinheiro, cansaço e
falta de companhia. Num bairro de baixa renda de São Paulo, entre quem não se exercita,
o tempo sobe para a primeira posição, quatro em dez. A desculpa muda de lugar conforme
quem responde e o que a cidade tem para oferecer.

Os 150 minutos da OMS são 1,5% da semana. Isso não desmente a desculpa de ninguém. Só
mede o tamanho dela.

*Qual é a sua desculpa, e ela sobreviveria a uma semana de 168 horas desenhada no papel?*

---

### ⚠️ Limitações do Modelo

- Barreira percebida é o motivo que a pessoa declara, não a causa medida. Por isso o
  ×0,80, aplicado a todas as proporções de barreira e à de ativos (IPAQ).
- Pelotas é uma cidade, em 2003; o bairro de São Paulo é um bairro, em 2007. Não há
  pesquisa nacional de barreiras com a mesma pergunta. O dia não generaliza para o Brasil.
- As duas fontes usam denominadores diferentes (todos os adultos em Pelotas; só os
  inativos em São Paulo). A tradução para "todos os adultos" em São Paulo é conta nossa.
- Os percentuais das outras quatro barreiras de Pelotas (não gostar, sentir-se velho,
  lesão ou doença, medo de lesão) não estão em fonte aberta e não entram no gráfico.
- As razões de chances por barreira (ajuste multivariável) não foram lidas no texto
  completo; o dia usa a direção do achado, não a magnitude.
- A sala de 30 supõe sorteio independente da população de Pelotas, o que é régua, não
  descrição de uma sala real.

*A ciência é honesta sobre o que não sabe. O modelo também.*
