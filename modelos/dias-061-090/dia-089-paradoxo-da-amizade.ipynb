## Qual a probabilidade de os seus amigos terem mais amigos que você?

**Decisão analisada:** A minha vida social é pequena, ou a régua está quebrada?

---

### 📖 A História

Você abre o feed e lá estão eles.

O jantar para o qual você não foi chamado. O grupo que viajou junto.
A festa com quarenta pessoas em que todo mundo parece íntimo de todo
mundo. Você fecha o aplicativo com a mesma conta de sempre:
a vida social dos outros é maior que a sua.

Aqui vai a parte estranha: essa conta provavelmente está certa.
E a culpa não é sua. É da matemática.

Em 1991, antes de existir Instagram, feed ou like, o sociólogo
Scott Feld provou um teorema desconfortável: **na média, os seus
amigos têm mais amigos que você.** Não porque você seja menos
interessante. Porque amizade é uma amostra enviesada por construção.

Nos dados clássicos que ele usou, de 146 meninas de uma escola
americana, **80 tinham menos amigos que a média dos próprios amigos**.
Quase duas para cada uma na situação contrária.

No Facebook, décadas depois, com 721 milhões de pessoas:
**92,7%** estão abaixo da média dos amigos.

O feed não mente sobre os outros. Ele mente sobre a régua.

---

### 📚 O Conceito: A Régua Que Nasce Quebrada

O mecanismo é viés de amostragem, e cabe numa frase: **pessoas
populares aparecem na lista de amigos de muita gente.**

Quem tem 40 amigos aparece em 40 listas, e faz 40 pessoas se
sentirem socialmente inadequadas. Quem tem 1 amigo aparece em uma
lista só. Quando você olha para os seus amigos, você não está
olhando para uma amostra aleatória da humanidade: está olhando
para uma amostra que superamostra os populares.

Feld mostrou que isso vira uma equação exata. A média de amigos
dos amigos é sempre:

**média + variância/média**

Sempre maior ou igual à sua média. E o castigo cresce com a
desigualdade: quanto mais variada a popularidade na rede, mais
quebrada fica a régua. É por isso que o paradoxo, que já valia
numa escola de 1961, ficou brutal nas redes sociais, onde a
variância explodiu.

No Dia #083, a rede que encolhe; hoje, o par declarado: a rede que,
do tamanho que for, sempre parece menor que a dos outros.

E o paradoxo tem família: turmas de faculdade parecem maiores que
são (a maioria dos alunos está nas turmas grandes), praias parecem
mais cheias (a maioria das pessoas vai nos dias lotados), a maioria
das cidades é pequena, mas a maioria das pessoas vive nas grandes.

### 🧮 O Modelo

**Fontes:**
- Feld, S. L. (1991) - Why Your Friends Have More Friends Than You
  Do - American Journal of Sociology, 96(6), 1464-1477 - dados de
  Coleman (1961), The Adolescent Society, 12 escolas. Marketville,
  meninas: **N=146** com amizades mútuas; 80 abaixo da média dos
  amigos, 41 acima, 25 empatadas; médias 2,7 (indivíduos) contra
  3,4 (amigos); 74% abaixo da média geral de amigos-de-amigos.
  Equação: média de amigos dos amigos = Σx²/Σx = média + var/média.
  **Fonte central.**
- Ugander, J., Karrer, B., Backstrom, L. & Marlow, C. (2011) - The
  Anatomy of the Facebook Social Graph - N=721 milhões de usuários:
  **92,7%** abaixo da média dos amigos; 83,6% abaixo da mediana.
  Apoio moderno.

**Fator ×0,80: não se aplica.** Propriedade matemática demonstrada
sobre redes sociométricas completas, não survey de autorrelato.

**Suposição declarada:** a simulação abaixo (semente 42) usa um
modelo de configuração com graus sorteados de uma distribuição
assimétrica, para ilustrar o teorema em uma rede sintética de
1.000 pessoas. É ilustração do mecanismo, não estimativa nova.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Feld (1991), AJS 96(6), sobre Coleman (1961), The Adolescent Society

# Marketville, meninas com amizades mútuas
n_meninas = 146
n_abaixo, n_acima, n_empate = 80, 41, 25
media_amigos_individuos = 2.7     # média de amigos por menina
media_amigos_dos_amigos = 3.4     # média de amigos dos amigos
pct_abaixo_media_geral = 0.74     # se comparadas à distribuição geral

# Micro-exemplo do artigo (8 meninas da figura 1)
micro_media_individuos = 2.5
micro_media_amigos = 3.0

# O paradoxo nas plataformas modernas
fb_n = 721e6
fb_pct_abaixo_media = 0.927       # Ugander et al. 2011
fb_pct_abaixo_mediana = 0.836
tw_pct_abaixo = 0.98              # Twitter (Hodas et al. 2013)

# Fator de correção: NÃO se aplica (propriedade matemática, rede completa)

print("=" * 65)
print("  DADOS DA LITERATURA - O PARADOXO DA AMIZADE")
print("=" * 65)
print(f"\n  Marketville, 1961 (N={n_meninas} meninas):")
print(f"  → Menos amigos que a média dos amigos:  {n_abaixo}  ({n_abaixo/n_meninas*100:.1f}%)")
print(f"  → Mais amigos que a média dos amigos:   {n_acima}  ({n_acima/n_meninas*100:.1f}%)")
print(f"  → Empatadas:                            {n_empate}  ({n_empate/n_meninas*100:.1f}%)")
print(f"  → Quase 2 para 1 no lado de baixo")
print(f"\n  As duas médias:")
print(f"  → Amigos por menina:                    {media_amigos_individuos}")
print(f"  → Amigos dos amigos:                    {media_amigos_dos_amigos}")
print(f"  → Contra a régua geral dos amigos:      {pct_abaixo_media_geral*100:.0f}% ficariam abaixo")
print(f"\n  O paradoxo escalou com as plataformas:")
print(f"  → Facebook (N=721 milhões):  {fb_pct_abaixo_media*100:.1f}% abaixo da média dos amigos")
print(f"  →                            {fb_pct_abaixo_mediana*100:.1f}% abaixo da mediana")
print(f"  → Twitter:                   {tw_pct_abaixo*100:.0f}% abaixo")
print(f"\n  Fator ×0,80: NÃO aplicado (propriedade matemática)")
print("=" * 65)
```

```python
# --- O MODELO ---
# A equação de Feld, verificada numa rede sintética (semente 42)
# média de amigos dos amigos = Σx²/Σx = média + variância/média

rng = np.random.default_rng(42)
n = 1000

# Graus sorteados de distribuição assimétrica (mínimo 1 amigo)
graus = rng.negative_binomial(2, 0.35, size=n) + 1
if graus.sum() % 2 == 1:
    graus[0] += 1

# A equação, direto da sequência de graus (identidade exata)
media = graus.mean()
variancia = graus.var()
media_amigos_eq = media + variancia / media
media_amigos_soma = (graus**2).sum() / graus.sum()

# Modelo de configuração: parear pontas de aresta ao acaso
pontas = np.repeat(np.arange(n), graus)
rng.shuffle(pontas)
a, b = pontas[0::2], pontas[1::2]

# Grau médio dos amigos de cada pessoa (multigrafo simples, ilustrativo)
soma_grau_vizinhos = np.zeros(n)
n_vizinhos = np.zeros(n)
np.add.at(soma_grau_vizinhos, a, graus[b])
np.add.at(soma_grau_vizinhos, b, graus[a])
np.add.at(n_vizinhos, a, 1)
np.add.at(n_vizinhos, b, 1)
media_dos_amigos_por_pessoa = soma_grau_vizinhos / np.maximum(n_vizinhos, 1)

pct_abaixo_sim = (graus < media_dos_amigos_por_pessoa).mean()

print("=" * 65)
print("  MODELO - A EQUAÇÃO DE FELD, VERIFICADA (semente 42)")
print("=" * 65)
print(f"\n  Rede sintética: {n} pessoas, {graus.sum()//2} amizades")
print(f"\n  média(x)                       = {media:.2f}")
print(f"  variância(x)                   = {variancia:.2f}")
print(f"\n  média + variância/média        = {media_amigos_eq:.2f}")
print(f"  Σx²/Σx (conferindo)            = {media_amigos_soma:.2f}")
print(f"  → A equação fecha exatamente, como o teorema promete")
print(f"\n  Na rede simulada:")
print(f"  → Seus amigos têm, em média, {media_amigos_eq/media - 1:.0%} mais amigos que você")
print(f"  → {pct_abaixo_sim*100:.0f}% das pessoas têm menos amigos que a média dos seus amigos")
print(f"\n  A LEITURA:")
print(f"  → Ninguém escolheu isso. É consequência de existir variância.")
print(f"  → Quanto mais desigual a popularidade, mais quebrada a régua.")
print(f"  → A escola de 1961 tinha pouca variância: 55% abaixo.")
print(f"     O Facebook tem variância enorme: 93% abaixo.")
print("=" * 65)
```

```python
# --- VISUALIZAÇÃO - GRÁFICOS SEPARADOS ---

# ── GRÁFICO 1 - As duas distribuições: a sua e a da régua ──
fig1, ax1 = plt.subplots(figsize=(10, 6))

# Distribuição de amigos dos indivíduos e dos amigos (ponderada pelo grau)
max_g = min(graus.max(), 20)
bins = np.arange(0.5, max_g + 1.5)
ax1.hist(graus, bins=bins, density=True, alpha=0.55, color='#2a8a82',
         label=f'Amigos das pessoas (média {media:.1f})')
ax1.hist(graus, bins=bins, weights=graus, density=True, alpha=0.55,
         color='#c0392b', label=f'Amigos dos amigos (média {media_amigos_eq:.1f})')
ax1.axvline(media, color='#2a8a82', linestyle='--', linewidth=2)
ax1.axvline(media_amigos_eq, color='#c0392b', linestyle='--', linewidth=2)
ax1.set_xlabel('Número de amigos')
ax1.set_ylabel('Densidade')
ax1.legend(loc='upper right', frameon=False, fontsize=11)
ax1.set_title('A régua quebrada: μ(amigos dos amigos) = μ + σ²/μ\n'
              'Rede sintética, semente 42: os populares entram na conta muitas vezes',
              fontsize=12.5, pad=15)

plt.figtext(0.5, 0.01,
            'Equação: Feld (1991) | simulação ilustrativa, suposição declarada | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-089-grafico-01-equacao.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 - Marketville, 1961: o dado original ──
fig2, ax2 = plt.subplots(figsize=(10, 6))

grupos = ['Menos amigos que\na média dos amigos', 'Empatadas', 'Mais amigos que\na média dos amigos']
valores = [n_abaixo, n_empate, n_acima]
cores2 = ['#c0392b', '#6b6a64', '#2a8a82']

bars = ax2.bar(grupos, valores, color=cores2, alpha=0.88, width=0.5)
ax2.set_ylabel('Meninas (de 146)')
ax2.set_ylim(0, 95)
ax2.set_title('Marketville, 1961: o paradoxo no dado original\n'
              '146 meninas · médias: 2,7 amigos por menina, 3,4 por amigo (Feld, 1991)',
              fontsize=12.5, pad=15)

for bar, v in zip(bars, valores):
    ax2.text(bar.get_x() + bar.get_width()/2, v + 1.8,
             f'{v}\n({v/n_meninas*100:.0f}%)', ha='center', fontweight='bold', fontsize=13)

ax2.annotate('quase 2 para 1', xy=(0, 80), xytext=(1.1, 82),
             fontsize=12, color='#c0392b', fontweight='bold',
             arrowprops=dict(arrowstyle='->', color='#c0392b', lw=1.5))

plt.figtext(0.5, 0.01,
            'Fonte: Feld (1991), AJS 96(6), dados de Coleman (1961) | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-089-grafico-02-marketville.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 - O paradoxo escalou com as redes ──
fig3, ax3 = plt.subplots(figsize=(10, 6))

eras = ['Escola americana\n1961\n(N=146)', 'Rede sintética\nsemente 42\n(N=1.000)',
        'Facebook 2011\n(N=721 milhões)', 'Twitter\n(estudos de 2012-13)']
pcts = [n_abaixo/n_meninas*100, pct_abaixo_sim*100, fb_pct_abaixo_media*100, tw_pct_abaixo*100]
cores3 = ['#c9a14a', '#6b6a64', '#c0392b', '#c0392b']

bars3 = ax3.bar(eras, pcts, color=cores3, alpha=0.88, width=0.55)
ax3.axhline(50, color='#333', linestyle='--', linewidth=1.2)
ax3.text(-0.42, 51.5, 'linha do "seria justo": 50%', fontsize=10, color='#333')
ax3.set_ylabel('Pessoas abaixo da média dos amigos (%)')
ax3.set_ylim(0, 108)
ax3.set_title('Quanto maior a variância, mais quebrada a régua\n'
              'O paradoxo escalou junto com as plataformas',
              fontsize=12.5, pad=15)

for bar, v in zip(bars3, pcts):
    ax3.text(bar.get_x() + bar.get_width()/2, v + 2,
             f'{v:.0f}%', ha='center', fontweight='bold', fontsize=14)

plt.figtext(0.5, 0.01,
            'Fontes: Feld (1991); Ugander et al. (2011); Hodas et al. (2013) | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-089-grafico-03-escalada.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**Na média, os seus amigos têm mais amigos que você. E isso não
diz nada sobre você.**

Em 1961, numa escola sem internet, 80 de 146 meninas tinham menos
amigos que a média das próprias amigas. Quase duas para uma.
No Facebook, com a variância de popularidade multiplicada por mil,
**92,7% das pessoas estão abaixo da régua dos próprios amigos.**

A equação é curta e não tem escapatória:

**média dos amigos = a sua média + variância/média**

Enquanto existir alguém mais popular que alguém, a régua dos seus
amigos aponta para cima. Você compara a sua vida social inteira
com uma amostra que superamostra os populares. O feed é essa
amostra em looping infinito, com filtro.

E aqui está o consolo, escrito pelo próprio Feld em 1991, décadas
antes de ser tão necessário: entender o paradoxo mostra que **a sua
posição é relativamente muito melhor do que a sua experiência
pessoal fez você acreditar.**

A sua vida social nunca foi pequena.

A régua é que nasceu quebrada.

A pergunta que fica:

*Que outras réguas quebradas você consulta todos os dias
como se fossem espelhos?*

### ⚠️ Limitações do Modelo

- O teorema vale para a média da rede; a proporção de indivíduos
  abaixo da régua depende do arranjo das amizades (60% a 75% na
  mesma distribuição, nos exemplos do próprio Feld), e existe
  contraexemplo arquitetado em que a maioria fica acima, raro e
  sem razão teórica para ocorrer
- Coleman (1961) operacionaliza amizade como nomeação mútua em
  escolas americanas dos anos 1950; a estrutura de amizades
  adolescentes não representa todas as redes
- Facebook e Twitter medem conexões de plataforma, não amizade
  íntima; a comparação entre eras é de mecanismo, não de construto
- A rede sintética (semente 42) é ilustração do teorema com modelo
  de configuração, suposição declarada; não é estimativa nova
- Comparação com a mediana dá resultado quase idêntico (Feld,
  nota 2), então a escolha da média não dirige a conclusão
- Par declarado com o Dia #083: lá a rede encolhia com a idade;
  aqui, a rede de qualquer tamanho parece menor que a dos outros.
  Claims independentes

*A ciência é honesta sobre o que não sabe. O modelo também.*
