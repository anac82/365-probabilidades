# 365 Probabilidades — Dia #047
## Qual a probabilidade de a novidade fazer o seu ano parecer mais longo?

**Tipo:** Comportamental
**Data de publicação:** 2026-07-30
**Ferramenta:** Python
**Decisão analisada:** Como faço meu tempo render mais, em vez de os anos evaporarem?
**Hashtag:** #365Probabilidades #Dia047

---

### 📖 A História

Essa semana eu joguei tênis alguns dias.

Faz pouco que voltei a pegar numa raquete, então cada partida ainda é meio desajeitada, com erro na bola e o corpo reaprendendo a coordenar o que a cabeça pede. E, mesmo assim, ou talvez justamente por isso, foram os dias que mais ficaram em mim.

Porque tem uma coisa que me assusta em silêncio, a velocidade com que os anos passam agora. As semanas se dissolvem, os meses se desfazem, e de vez em quando eu me pego pensando, como assim já passou da metade do ano?

Esses últimos dias, porém, não se desfizeram. Ficaram, e seguem bem vivos na minha memória. E quando fui entender por quê, caí exatamente no estudo de hoje, que fala sobre por que o tempo parece acelerar com a idade e por que uma raquete nova na mão pode ser um dos antídotos mais simples que existem.

---

### 📚 O Conceito: Por Que o Tempo Acelera

Duas coisas fazem o tempo parecer mais rápido conforme a gente envelhece.

A primeira é pura aritmética. Robert Lemlich (1975) propôs que a velocidade subjetiva do tempo cresce com a raiz quadrada da idade. Para uma criança de 10 anos, um ano é uma fatia enorme da vida vivida. Para quem tem 40, é uma fatia bem menor, e o cérebro a percebe como mais curta. Contra essa parte não dá pra lutar.

A segunda parte é novidade, e essa você controla. Mas ela tem um truque, o paradoxo dos dois relógios. No relógio do momento, a novidade faz o tempo voar, porque você está absorto e nem vê passar. No relógio da memória, inverte. Um período cheio de primeiras vezes fica denso de lembranças e, ao olhar pra trás, parece longo. Um período de pura rotina quase não deixa marcador e evapora, como se não tivesse acontecido. É o paradoxo das férias, a semana nova voa enquanto acontece, mas parece enorme depois.

---

### 🧮 O Modelo

Uso o modelo de Lemlich para calcular quão mais rápido um ano passa hoje em comparação com a infância, mostro a fatia da vida que cada ano representa, e comparo o efeito da rotina versus o da novidade sobre a duração lembrada.

**Fontes:**
- Lemlich, R. (1975). *Subjective Acceleration of Time with Aging*. **Perceptual and Motor Skills**, 41, 235-238. Modelo: velocidade subjetiva ∝ √idade.
- Wittmann, M. & Lehnhoff, S. (2005). *Age Effects in Perception of Time*. Survey com N=500, idades 14-94, replicado em 6 países.
- Mecanismo de memória/novidade (paradoxo das férias): mais experiências novas → maior duração percebida em retrospecto.

**Nota metodológica:** o número central é um modelo matemático (Lemlich), não uma proporção de survey autorrelatado. Por isso o fator ×0.80 não se aplica.


````python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

DOURADO = '#c8a84b'; VERMELHO = '#c0392b'; VERDE = '#1a5f5a'; CINZA = '#6b6a64'
np.random.seed(47)  # Dia #047
print("✅ Bibliotecas carregadas")
````

````python
# --- DADOS DA LITERATURA ---
idade_ref = 10     # infância, referência
idade_ana = 44     # idade atual

# Lemlich (1975): velocidade subjetiva ∝ raiz quadrada da idade
def aceleracao_lemlich(idade, ref=idade_ref):
    return np.sqrt(idade / ref)
# Janet (1877): proporção 1/idade
def aceleracao_janet(idade, ref=idade_ref):
    return idade / ref

N_wittmann = 500

print("=" * 66)
print("  DADOS DA LITERATURA — PERCEPÇÃO DO TEMPO")
print("=" * 66)
print(f"  Lemlich 1975: velocidade subjetiva do tempo ∝ √idade")
print(f"  Wittmann & Lehnhoff 2005: N={N_wittmann}, idades 14-94, 6 países")
print(f"    → confirmam: o tempo parece acelerar com a idade")
print()
print(f"  Comparando {idade_ana} anos com {idade_ref} anos:")
print(f"  → Modelo de Lemlich (√idade):  {aceleracao_lemlich(idade_ana):.2f}× mais rápido")
print(f"  → Modelo de Janet (1/idade):   {aceleracao_janet(idade_ana):.2f}× mais rápido")
print("=" * 66)
````

````python
# --- O MODELO ---
acel = aceleracao_lemlich(idade_ana)
fracao_ref = 1/idade_ref * 100
fracao_ana = 1/idade_ana * 100

# Paradoxo das férias (duração lembrada relativa) — valores ilustrativos declarados
dur_rotina = 1.0
dur_novidade = 2.0

print("=" * 66)
print("  MODELO — O TEMPO QUE ACELERA E O QUE DÁ PRA ESTICAR")
print("=" * 66)
print(f"  A PARTE QUE É ARITMÉTICA (idade):")
print(f"  → Aos {idade_ana}, um ano passa ~{acel:.1f}× mais rápido que aos {idade_ref}.")
print(f"  → Um ano era {fracao_ref:.0f}% da sua vida aos {idade_ref};")
print(f"    hoje é só {fracao_ana:.1f}%. Contra isso não dá pra brigar.")
print()
print(f"  A PARTE QUE VOCÊ CONTROLA (novidade):")
print(f"  → No momento, a novidade faz o tempo voar.")
print(f"  → Na memória, ela estica: o mesmo período parece ~{dur_novidade/dur_rotina:.0f}× mais longo")
print(f"    em retrospecto. Rotina evapora; novidade fica. (paradoxo das férias)")
print("=" * 66)
````

````python
# --- VISUALIZAÇÃO — 3 GRÁFICOS SEPARADOS ---

# ── GRÁFICO 1 — ASSINATURA MATEMÁTICA: curva de aceleração ──
fig1, ax1 = plt.subplots(figsize=(10, 6))
idades = np.linspace(5, 80, 400)
ax1.plot(idades, aceleracao_lemlich(idades), color=VERDE, lw=2.5, label='Modelo de Lemlich  (velocidade ∝ √idade)')
ax1.plot(idades, aceleracao_janet(idades), color=CINZA, lw=1.5, ls='--', alpha=0.7, label='Modelo de Janet  (∝ 1/idade)')
ax1.axvline(idade_ana, color=DOURADO, ls='--', alpha=0.8)
ax1.scatter([idade_ana], [aceleracao_lemlich(idade_ana)], color=DOURADO, s=130, zorder=5, edgecolor='#333')
ax1.annotate(f'aos {idade_ana}:\n~{aceleracao_lemlich(idade_ana):.1f}× mais rápido\nque aos {idade_ref}',
             xy=(idade_ana, aceleracao_lemlich(idade_ana)),
             xytext=(idade_ana+4, aceleracao_lemlich(idade_ana)+0.6),
             fontsize=11, color='#8a6d10', fontweight='bold')
ax1.text(0.97, 0.08, r'velocidade subjetiva  $\propto \sqrt{idade}$',
         transform=ax1.transAxes, fontsize=13, color=VERDE, fontweight='bold', ha='right')
ax1.set_xlabel('Idade (anos)')
ax1.set_ylabel('Velocidade percebida do tempo (relativa aos 10 anos)')
ax1.set_title('A ASSINATURA MATEMÁTICA\nPor que um ano hoje passa mais rápido que na infância', fontsize=13, pad=15)
ax1.legend(loc='upper left', fontsize=9); ax1.set_ylim(0, 6)
plt.figtext(0.5, 0.005, 'Modelo de Lemlich 1975 · apoio empírico em Wittmann & Lehnhoff 2005 (N=500) | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout(); plt.savefig('dia-047-grafico-01-assinatura.png', dpi=150, bbox_inches='tight'); plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 — A fração da vida que cada ano representa ──
fig2, ax2 = plt.subplots(figsize=(10, 6))
idades_barra = [5, 10, 20, idade_ana, 60, 80]
fracoes = [100/i for i in idades_barra]
cores2 = [DOURADO if i == idade_ana else CINZA for i in idades_barra]
bars = ax2.bar([str(i) for i in idades_barra], fracoes, color=cores2, alpha=0.9, width=0.6)
ax2.set_xlabel('Idade (anos)'); ax2.set_ylabel('Quanto um ano representa da vida vivida (%)')
ax2.set_title('A fração que encolhe\nAos 44, um ano é uma fatia bem menor da sua vida do que aos 10', fontsize=13, pad=15)
for bar, f, i in zip(bars, fracoes, idades_barra):
    ax2.text(bar.get_x()+bar.get_width()/2, f+0.4, f'{f:.1f}%', ha='center', fontweight='bold', fontsize=13,
             color='#8a6d10' if i == idade_ana else '#333')
plt.figtext(0.5, 0.005, 'Intuição proporcional (Janet) · cada ano novo pesa menos no total | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout(); plt.savefig('dia-047-grafico-02-fracao.png', dpi=150, bbox_inches='tight'); plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 — A saída: rotina encurta, novidade estica ──
fig3, ax3 = plt.subplots(figsize=(10, 6))
cats3 = ['Um ano de\nrotina', 'Um ano cheio\nde novidades']
vals3 = [dur_rotina, dur_novidade]
bars3 = ax3.bar(cats3, vals3, color=[CINZA, VERDE], alpha=0.9, width=0.45)
ax3.set_ylabel('Duração lembrada, em retrospecto (relativa)'); ax3.set_ylim(0, 2.5); ax3.set_yticks([])
ax3.set_title('A saída: novidade estica o tempo lembrado\nO paradoxo das férias — mais primeiras vezes, ano mais longo na memória', fontsize=13, pad=15)
ax3.annotate('', xy=(1, dur_novidade), xytext=(0, dur_rotina), arrowprops=dict(arrowstyle='->', color=DOURADO, lw=2.5))
ax3.text(0.5, 0.62, 'um esporte novo,\numa raquete na mão,\num lugar novo', ha='center', fontsize=11,
         color='#8a6d10', fontweight='bold', transform=ax3.transAxes)
plt.figtext(0.5, 0.005, 'Ilustração da direção do efeito (memória/novidade) · valores relativos declarados | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout(); plt.savefig('dia-047-grafico-03-novidade.png', dpi=150, bbox_inches='tight'); plt.close()
print("✅ Gráfico 3 salvo!")
````

### 💡 O Insight

**Aos 44, um ano passa cerca de duas vezes mais rápido do que aos 10.**

Parte disso é aritmética pura, e não adianta lutar. Um ano é uma fração cada vez menor de tudo o que você já viveu, e o cérebro percebe essa fatia como mais curta. Essa metade da conta é inevitável.

Mas a outra metade não é, e é aqui que mora o pulo do gato, o paradoxo dos dois relógios. No relógio do momento, a novidade faz o tempo voar, você mergulha e nem vê passar. No relógio da memória, tudo se inverte. O período cheio de primeiras vezes fica denso de lembranças e, em retrospecto, parece longo. O período de pura rotina evapora, como se o ano tivesse sumido.

Foi nisso que eu tropecei sem querer com o tênis dessa semana. Enquanto eu jogava, o tempo voava. Mas quando a semana acabou, foram justamente esses dias atrapalhados que mais ficaram. Sem querer, eu tinha dado ao meu cérebro coisas que valiam a pena guardar.

Você não consegue frear o relógio. Mas consegue encher os dias de novidade suficiente pra que o ano volte a durar um ano na memória.

*Quantas primeiras vezes você viveu no último mês, ou ele já se desfez, igual a todos os outros?*

### ⚠️ Limitações do Modelo

- O modelo de Lemlich tem apoio empírico inicial, mas foi questionado (Friedman & Janssen, 2010) por depender da memória de experiências distantes.
- Wittmann & Lehnhoff (2005) é um survey de percepção subjetiva; confirma a direção, não um valor individual exato.
- A idade de referência (10 anos) e a idade atual (44) são escolhas do modelo; a razão muda conforme as idades comparadas.
- O Gráfico 3 usa valores relativos declarados para ilustrar a direção do efeito de novidade, não medidas absolutas.
- O fator ×0.80 não foi aplicado: o número central é um modelo matemático, não survey autorrelatado.
