### 📖 A História

Já demorei mais tempo escolhendo o que assistir do que efetivamente assistindo.

Abri o catálogo, rolei, voltei, abri de novo. Vinte minutos depois, fechei tudo sem escolher nada. E fui dormir irritada — não com o conteúdo, mas comigo mesma, por não ter conseguido tomar uma decisão tão simples.

Sempre achei que isso era falta de foco. Indecisão minha.

Não é. Tem nome, tem décadas de pesquisa, e começa num lugar inesperado: uma mesa de geleias num supermercado da Califórnia.

---

### 📚 O Conceito: Paradoxo da Escolha e Sobrecarga de Opções

O **paradoxo da escolha** (Schwartz, 2004) propõe que, a partir de um certo ponto, aumentar o número de opções não aumenta a satisfação — ela diminui. O mecanismo: quanto mais opções existem, maior o custo cognitivo de comparar, maior o risco percebido de fazer a escolha errada, e maior o arrependimento potencial após decidir.

Iyengar & Lepper (2000) testaram isso empiricamente com três estudos: geleias num supermercado, chocolates, e redações opcionais de faculdade. Em todos os contextos, o grupo com menos opções agiu mais e relatou mais satisfação com sua escolha.

A meta-análise de Scheibehenne, Greifeneder & Todd (2010) jogou água fria: 50 estudos, efeito médio perto de zero. Mas Chernev, Böckenholt & Goodman (2015) identificaram por quê: o efeito existe, mas só sob condições específicas — complexidade das opções, dificuldade da decisão, incerteza de preferência.

O paradoxo da escolha não é falso. É condicional.

E ele não ficou só no supermercado. Um levantamento da Nielsen (2019) encontrou que apenas um terço dos usuários de streaming navega pelo catálogo até encontrar algo — e **21% simplesmente desistem** de assistir quando não conseguem decidir. Na TV tradicional, com menos opções, 58% das pessoas voltam para o canal favorito em vez de desistir. Pesquisadores batizaram esse fenômeno específico de \"Netflix syndrome\" (Kim & Choi, 2024): gastar mais tempo escolhendo o que assistir do que efetivamente assistindo — associado a estresse e irritação, não a satisfação.

Nos restaurantes, a mesma coisa: um levantamento da Prezzo (2023, N=2.000) encontrou que **86% da Geração Z** experimenta \"menu anxiety\" ao pedir comida — ansiedade gerada em parte pelo excesso de opções no cardápio. Uma pesquisa da OnePoll nos EUA (2023, N=2.000) confirmou: **30% dos americanos** sentem ansiedade diante de um cardápio, subindo para **mais de 40%** entre jovens adultos.

A mesa de geleias de 2000, décadas depois, numa tela e num cardápio diferentes.

### 🧮 O Modelo
**Fontes:**
- Iyengar, S.S. & Lepper, M.R. (2000) — *When Choice Is Demotivating: Can One Desire Too Much of a Good Thing?* — Journal of Personality and Social Psychology, 79(6), 995-1006 — N=754 (compradores observados), Estudo 1 (geleias); 3 estudos no total (geleias, chocolates, redações)
- Scheibehenne, B., Greifeneder, R. & Todd, P.M. (2010) — *Can There Ever Be Too Many Options? A Meta-Analytic Review of Choice Overload* — Journal of Consumer Research, 37(3), 409-425 — 50 estudos, N=5.036
- Chernev, A., Böckenholt, U. & Goodman, J. (2015) — *Choice Overload: A Conceptual Review and Meta-Analysis* — Journal of Consumer Psychology, 25(2), 333-358
- Nielsen (2019) — Total Audience Report — levantamento de mercado sobre comportamento de usuários de streaming (não é estudo peer-reviewed)
- Prezzo / UK (2023) — levantamento com N=2.000 sobre \"menu anxiety\" em restaurantes (não é estudo peer-reviewed)
- OnePoll / USA (2023) — levantamento com N=2.000 sobre ansiedade diante de cardápios (não é estudo peer-reviewed)
- Kim, H. & Choi, J. (2024) — conceito de \"Netflix syndrome\" — Asian Journal for Public Opinion Research


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print('✅ Bibliotecas carregadas')
```

```python
# --- DADOS DA LITERATURA ---
# Iyengar & Lepper (2000) — Estudo 1 — Loja Draeger's, Menlo Park, California
n_total_loja = 754

# Condicao com MUITAS opcoes (24 geleias)
n_passaram_24 = 242
n_pararam_24 = 145
p_parou_24 = n_pararam_24 / n_passaram_24
n_comprou_24 = 4
p_comprou_24 = n_comprou_24 / n_pararam_24

# Condicao com POUCAS opcoes (6 geleias)
n_passaram_6 = 260
n_pararam_6 = 104
p_parou_6 = n_pararam_6 / n_passaram_6
n_comprou_6 = 31
p_comprou_6 = n_comprou_6 / n_pararam_6

razao_conversao = p_comprou_6 / p_comprou_24

# Meta-analise — Scheibehenne, Greifeneder & Todd (2010)
n_estudos_meta = 50
n_condicoes_meta = 63
n_total_meta = 5036

# Dados de mercado contemporaneos (NAO peer-reviewed)
p_desiste_streaming = 0.21
p_menu_anxiety_genz = 0.86
p_menu_anxiety_geral = 0.67
p_menu_anxiety_usa = 0.30
p_menu_anxiety_usa_jovens = 0.40

fator_correcao = 0.80

print('=' * 65)
print('  DADOS DA LITERATURA — PARADOXO DA ESCOLHA')
print('=' * 65)
print(f'\n  Iyengar & Lepper (2000) — Estudo 1 (geleias), N={n_total_loja}:')
print(f'\n  24 geleias:')
print(f'  → Pararam para provar : {p_parou_24*100:.0f}% ({n_pararam_24} de {n_passaram_24})')
print(f'  → Compraram          : {p_comprou_24*100:.1f}% ({n_comprou_24} de {n_pararam_24})')
print(f'\n  6 geleias:')
print(f'  → Pararam para provar : {p_parou_6*100:.0f}% ({n_pararam_6} de {n_passaram_6})')
print(f'  → Compraram          : {p_comprou_6*100:.1f}% ({n_comprou_6} de {n_pararam_6})')
print(f'\n  Razao de conversao: {razao_conversao:.1f}x mais compras com menos opcoes')
print(f'\n  Scheibehenne et al. (2010) — Meta-analise:')
print(f'  → {n_estudos_meta} estudos . {n_condicoes_meta} condicoes . N={n_total_meta:,}')
print(f'  → Efeito medio: d ≈ 0 (virtualmente zero)')
print(f'\n  Dados de mercado contemporaneos (nao peer-reviewed):')
print(f'  → Nielsen 2019: {p_desiste_streaming*100:.0f}% desistem de assistir no streaming')
print(f'  → Prezzo UK 2023: {p_menu_anxiety_genz*100:.0f}% da Gen Z tem menu anxiety')
print(f'  → OnePoll USA 2023: {p_menu_anxiety_usa*100:.0f}% dos americanos sentem ansiedade com cardapios')
print(f'\n  Fator de correcao (conversao Iyengar): x{fator_correcao}')
print('=' * 65)
print()
print('  ⚠️  NOTA METODOLOGICA:')
print('  O estudo das geleias tem N pequeno para os grupos de conversao')
print('  (n=145 e n=104 quem parou; n=4 e n=31 quem comprou). O mesmo')
print('  padrao foi replicado nos Estudos 2 (chocolates) e 3 (redacoes)')
print('  do mesmo paper. Os dados de Nielsen, Prezzo e OnePoll sao')
print('  levantamentos de mercado, nao estudos peer-reviewed.')
```

```python
# --- O MODELO ---

p_comprou_6_corrigido = p_comprou_6 * fator_correcao
p_comprou_24_corrigido = p_comprou_24 * fator_correcao
razao_conversao_corrigida = p_comprou_6_corrigido / p_comprou_24_corrigido

# IC 95% via distribuicao Beta, prior de Jeffreys
alpha_6 = n_comprou_6
beta_6 = n_pararam_6 - n_comprou_6
dist_6 = stats.beta(alpha_6 + 0.5, beta_6 + 0.5)
ic_6 = dist_6.interval(0.95)

alpha_24 = n_comprou_24
beta_24 = n_pararam_24 - n_comprou_24
dist_24 = stats.beta(alpha_24 + 0.5, beta_24 + 0.5)
ic_24 = dist_24.interval(0.95)

print('=' * 65)
print('  MODELO — RESULTADO')
print('=' * 65)
print(f'\n  Taxa de conversao com 6 geleias:')
print(f'  → Estimativa : {p_comprou_6*100:.1f}%')
print(f'  → IC 95%     : [{ic_6[0]*100:.1f}%, {ic_6[1]*100:.1f}%]')
print(f'  → Corrigida  : {p_comprou_6_corrigido*100:.1f}%')
print(f'\n  Taxa de conversao com 24 geleias:')
print(f'  → Estimativa : {p_comprou_24*100:.1f}%')
print(f'  → IC 95%     : [{ic_24[0]*100:.1f}%, {ic_24[1]*100:.1f}%]')
print(f'  → Corrigida  : {p_comprou_24_corrigido*100:.1f}%')
print(f'\n  Razao de conversao: {razao_conversao:.1f}x (mesmo com correcao: {razao_conversao_corrigida:.1f}x)')
print(f'\n  CONTEXTO CRITICO:')
print(f'  → Meta-analise 2010 (N=5.036, 50 estudos): efeito medio d≈0')
print(f'  → O fenomeno existe, mas e condicional — nao universal')
print(f'  → Moderadores: complexidade, dificuldade, incerteza de preferencia')
print('=' * 65)
```

```python
# --- VISUALIZACAO ---

# GRAFICO 1 — Atracao vs conversao
fig1, axes = plt.subplots(1, 2, figsize=(14, 8))

ax_atr = axes[0]
categorias = ['6 geleias', '24 geleias']
atracao = [p_parou_6 * 100, p_parou_24 * 100]
cores_atr = ['#2a8a82', '#c8a84b']
bars_atr = ax_atr.bar(categorias, atracao, color=cores_atr, alpha=0.85, width=0.45)
ax_atr.set_ylim(0, 80)
ax_atr.set_ylabel('% que parou para provar', fontsize=12)
ax_atr.set_title('Atracao\n(quem parou na mesa)', fontsize=13)
for bar, v in zip(bars_atr, atracao):
    ax_atr.text(bar.get_x() + bar.get_width() / 2, v + 1.5,
                f'{v:.0f}%', ha='center', fontweight='bold', fontsize=16)

ax_conv = axes[1]
conversao = [p_comprou_6 * 100, p_comprou_24 * 100]
cores_conv = ['#2a8a82', '#c0392b']
bars_conv = ax_conv.bar(categorias, conversao, color=cores_conv, alpha=0.85, width=0.45)
ax_conv.set_ylim(0, 40)
ax_conv.set_ylabel('% que comprou (entre quem parou)', fontsize=12)
ax_conv.set_title('Conversao\n(quem de fato comprou)', fontsize=13)
for bar, v in zip(bars_conv, conversao):
    ax_conv.text(bar.get_x() + bar.get_width() / 2, v + 0.8,
                 f'{v:.0f}%', ha='center', fontweight='bold', fontsize=16)

fig1.suptitle('O Paradoxo da Escolha — Iyengar & Lepper (2000)\nDraeger\'s, Menlo Park, CA . N=754 compradores observados',
              fontsize=13, y=1.02)
plt.tight_layout()
plt.savefig('dia-021-grafico-01-atracao-vs-conversao.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Grafico 1 salvo!')


# GRAFICO 2 — IC 95% das taxas de conversao
fig2, ax2 = plt.subplots(figsize=(12, 8))

x_vals = [p_comprou_6 * 100, p_comprou_24 * 100]
x_labels = ['6 geleias', '24 geleias']
ic_low = [ic_6[0] * 100, ic_24[0] * 100]
ic_high = [ic_6[1] * 100, ic_24[1] * 100]
erros_low = [x - l for x, l in zip(x_vals, ic_low)]
erros_high = [h - x for x, h in zip(x_vals, ic_high)]
cores2 = ['#2a8a82', '#c0392b']

bars2 = ax2.bar(x_labels, x_vals, color=cores2, alpha=0.85, width=0.45,
                yerr=[erros_low, erros_high], capsize=14,
                error_kw={'elinewidth': 2.5, 'capthick': 2.5})
ax2.set_ylim(0, 50)
ax2.set_ylabel('Taxa de conversao (%) — com IC 95%', fontsize=12)
ax2.set_title('Incerteza real em torno dos numeros\nIyengar & Lepper (2000) — Distribuicao Beta com prior de Jeffreys',
              fontsize=13, pad=15)

for bar, v, il, ih in zip(bars2, x_vals, ic_low, ic_high):
    ax2.text(bar.get_x() + bar.get_width() / 2, ih + 2.5,
             f'{v:.0f}%\n[{il:.0f}%–{ih:.0f}%]', ha='center', fontsize=12)

plt.tight_layout()
plt.savefig('dia-021-grafico-02-ic-conversao.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Grafico 2 salvo!')


# GRAFICO 3 — Efeito medio da meta-analise (ilustrativo)
fig3, ax3 = plt.subplots(figsize=(12, 8))

x = np.linspace(-1.5, 1.5, 1000)
y = stats.norm.pdf(x, loc=0, scale=0.45)

ax3.plot(x, y, color='#888888', linewidth=3)
ax3.fill_between(x, y, alpha=0.15, color='#888888')
ax3.axvline(x=0, color='#333333', linestyle='--', linewidth=2,
            label='Efeito medio ≈ 0 (Scheibehenne et al. 2010)')
ax3.axvline(x=1.1, color='#c0392b', linewidth=2.5,
            label='Estudo das geleias (efeito grande, contexto especifico)')

ax3.set_xlabel('Tamanho do efeito (d de Cohen)', fontsize=12)
ax3.set_ylabel('Densidade de estudos', fontsize=12)
ax3.set_title('O paradoxo da escolha nao e falso — e condicional\nScheibehenne et al. (2010) — 50 estudos, N=5.036 . ilustracao da distribuicao',
              fontsize=13, pad=15)
ax3.legend(fontsize=11)
ax3.set_xlim(-1.5, 1.5)

plt.tight_layout()
plt.savefig('dia-021-grafico-03-metaanalise.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Grafico 3 salvo!')
```

### 💡 O Insight

Num supermercado da Califórnia, em dois sábados consecutivos, pesquisadores montaram uma mesa de geleias — ora com 24 sabores, ora com 6. Com 24, **60% das pessoas pararam para provar**. Com 6, apenas 40%.

Mas na hora da compra: da mesa das 24, apenas **3% compraram** (4 pessoas). Da mesa das 6, **30% compraram** (31 pessoas). Uma razão de quase 11 para 1. O mesmo padrão se repetiu com chocolates e redações opcionais — em todos os contextos, menos opções geraram mais ação e mais satisfação.

Uma meta-análise de 50 estudos (N=5.036) encontrou efeito médio praticamente zero — mas isso não significa que o fenômeno não existe. Significa que ele é **condicional**: aparece quando as opções são complexas, difíceis de comparar, e quando a pessoa ainda não sabe o que quer.

Fora do laboratório, o padrão se confirma em escala: **21% dos usuários de streaming desistem** de assistir quando não conseguem decidir (Nielsen, 2019). **86% da Geração Z** experimenta "menu anxiety" em restaurantes (Prezzo, 2023). **30% dos americanos** sentem ansiedade diante de um cardápio (OnePoll, 2023).

Mais opções não é sinônimo de mais liberdade. Às vezes é só mais ruído entre você e a decisão.

*Quando você está paralisada diante de muitas opções, a pergunta que ajuda não é "qual é melhor?" — é "eu já sei o que quero, ou ainda estou descobrindo?" Porque se você já sabe, mais opções só atrapalham.*

---

### ⚠️ Limitações do Modelo
- O estudo das geleias tem N pequeno nos grupos de conversão (n=4 e n=31 compradores) — os percentuais são reais, mas os intervalos de confiança são amplíssimos, como mostra o Gráfico 2
- O padrão foi replicado em dois outros estudos do mesmo paper (chocolates e redações) e em dezenas de estudos subsequentes com itens diferentes (balas, canetas, café), o que aumenta a confiança no fenômeno
- A meta-análise de 2010 encontrou variância muito alta entre estudos — o efeito médio zero não significa que o fenômeno não existe, mas que ele não é universal
- Chernev et al. (2015) identificaram moderadores específicos: complexidade das opções, dificuldade da decisão, incerteza de preferência e objetivo da decisão
- O Gráfico 3 é uma ilustração da distribuição de efeitos, não os dados brutos da meta-análise
- Os dados de Nielsen (2019), Prezzo (2023) e OnePoll (2023) são levantamentos de mercado, não estudos peer-reviewed — servem como ilustração do fenômeno em escala, não como evidência com o mesmo peso metodológico
- O fator de correção (×0,80) foi aplicado às estimativas centrais de conversão

*A ciência é honesta sobre o que não sabe. O modelo também.*
