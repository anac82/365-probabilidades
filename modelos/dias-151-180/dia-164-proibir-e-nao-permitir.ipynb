# 365 Probabilidades - Dia #164
## Qual a probabilidade de 'proibir' e 'não permitir' darem respostas diferentes?

**Tipo:** Experimental · Linguagem  
**Data de publicação:** 2026-11-24  
**Ferramenta:** Python  
**Decisão analisada:** Quanto confiar no resultado de uma pesquisa quando não se sabe como a pergunta foi escrita?  
**Hashtag:** #365Probabilidades #Dia164

---

### 📖 A História

Pense em duas perguntas.

A prefeitura deveria permitir carros no centro da cidade?

A prefeitura deveria proibir carros no centro da cidade?

Parecem a mesma pergunta com o sinal trocado. Quem responde que não deveria permitir também deveria responder que sim, deveria proibir.

Em 1941, um pesquisador americano fez um teste parecido, e as respostas não bateram.

Fui ver se isso ainda acontece, e de que tamanho.

---

### 📚 O Conceito

A pesquisa de opinião parte de uma ideia simples. Cada pessoa tem uma opinião, e a pergunta só serve para registrá-la.

O experimento de **pergunta dividida** testa essa ideia. A amostra é sorteada em duas metades, e cada metade recebe uma versão da mesma pergunta. Se as respostas mudam, foi a palavra que mudou, e não a opinião.

A diferença entre "proibir" e "não permitir" é o caso mais estudado. A ciência chama esse fenômeno de **assimetria entre proibir e permitir**. A explicação mais aceita, de Bregje Holleman, é que as pessoas têm a mesma opinião nas duas versões, mas usam as opções de resposta de outro jeito. "Proibir" soa forte, e muita gente que não quer permitir também não quer proibir.

---

### 🧮 O Modelo

Quatro fontes e quatro contas.

1. **O clássico.** A diferença de Rugg (1941) entre as duas versões.
2. **O efeito ao longo do tempo.** O clássico, a média de 52 perguntas reunidas por Holleman e o experimento norueguês de 2015, na mesma unidade (pontos percentuais).
3. **O experimento grande.** O tamanho do efeito em 31.112 eleitores de Utrecht, transformado na chance de uma resposta à versão negativa ser mais favorável ao tema do que uma resposta à versão positiva (200 mil sorteios, semente 42).
4. **Quem muda mais.** O mesmo cálculo para os menos e os mais informados sobre política.

**Fontes:**
- Rugg, D. (1941). Experiments in wording questions II. *Public Opinion Quarterly*, 5(1), 91-92. **N não localizado.** Números conferidos em fonte secundária.
- Holleman, B. C. (1999). Wording effects in survey research: Using meta-analysis to explain the forbid/allow asymmetry. *Journal of Quantitative Linguistics*, 6(1), 29-40. **52 perguntas** de experimentos feitos desde 1940. Lido no resumo e na descrição de Kamoen, Holleman e Van den Bergh (2007).
- Holleman, B., Kamoen, N., Krouwel, A., van de Pol, J. & de Vreese, C. (2016). Positive vs. negative: The impact of question polarity in voting advice applications. *PLOS ONE*, 11(10), e0164184. **31.112 usuários** de um aplicativo de orientação de voto nas eleições municipais de Utrecht, em 2014. Lido no texto completo.
- Estudo norueguês (2015). Does it matter how you ask? The forbid-allow asymmetry in the measurement of attitudes towards drug policies. *International Journal of Drug Policy*. **2.182 adultos** noruegueses. Lido no resumo. **Os autores não foram conferidos.**

**Nota metodológica sobre o fator ×0,80.** Não se aplica. O que se mede é a diferença entre duas versões da pergunta numa amostra sorteada, e não a proporção de uma pesquisa.

**Cordas declaradas:** #081 (a mesma decisão muda quando muda a frase) e #072 (o burnout que depende da régua). No #072, a proporção mudava com a definição. Aqui a pesquisa, a amostra e o assunto são os mesmos, e só muda uma palavra.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

VERMELHO, DOURADO, VERDE, CINZA = '#c0392b', '#c8a84b', '#1a5f5a', '#6b6a64'
SEMENTE, N_SIM = 42, 200_000
rng = np.random.default_rng(SEMENTE)

def jeffreys(k, n, nivel=0.95):
    """Intervalo de Jeffreys para uma proporção."""
    a = (1 - nivel) / 2
    return stats.beta.ppf([a, 1 - a], k + 0.5, n - k + 0.5)

def sorteia(k, n):
    return rng.beta(k + 0.5, n - k + 0.5, N_SIM)

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---

# 1) Rugg (1941), Public Opinion Quarterly 5(1). Pergunta sobre discursos públicos contra a democracia.
rugg = {'não permitiriam': 0.75, 'proibiriam': 0.54}

# 2) Holleman (1999), Journal of Quantitative Linguistics 6(1): meta-análise de 52 perguntas desde 1940.
#    "Não" à versão com "proibir" foi 14,2 pontos percentuais mais frequente que "sim" à versão com "permitir".
meta = dict(perguntas=52, diferenca_pp=14.2)

# 3) Holleman et al. (2016), PLOS ONE 11(10): experimento sorteado, eleições municipais de Utrecht, 2014.
utrecht = dict(n=31_112, escala=5,
               implicitas=dict(itens=6, d=0.13, z=11.6),     # proibir/permitir e pares parecidos
               explicitas=dict(itens=10, d=0.05, z=5.25),    # pares com "não"
               d_menos_informados=0.39, d_mais_informados=0.06)

# 4) Estudo norueguês (2015), International Journal of Drug Policy (autores não conferidos): 2.182 adultos noruegueses, política de drogas.
noruega = dict(n=2182, diferenca_pp=2.0, ic=(0.0, 5.0))

aplica_fator_080 = False

fmt = lambda x, c=1: f"{x:.{c}f}".replace('.', ',')
print("=" * 72)
print("  DADOS DA LITERATURA - PROIBIR E NÃO PERMITIR")
print("=" * 72)
print(f"\n  Rugg (1941): {fmt(rugg['não permitiriam']*100, 0)}% não permitiriam discursos contra a democracia, {fmt(rugg['proibiriam']*100, 0)}% proibiriam")
print(f"\n  Holleman (1999): {meta['perguntas']} perguntas  ·  diferença média de {fmt(meta['diferenca_pp'])} pontos percentuais")
print(f"\n  Holleman et al. (2016): {utrecht['n']:,} eleitores de Utrecht".replace(',', '.'))
print(f"  → Pares como proibir/permitir ({utrecht['implicitas']['itens']} afirmações): tamanho do efeito d = {fmt(utrecht['implicitas']['d'], 2)}")
print(f"  → Pares com 'não' ({utrecht['explicitas']['itens']} afirmações): d = {fmt(utrecht['explicitas']['d'], 2)}")
print(f"  → Menos informados sobre política: d = {fmt(utrecht['d_menos_informados'], 2)}  ·  mais informados: d = {fmt(utrecht['d_mais_informados'], 2)}")
print(f"\n  Noruega (2015): {noruega['n']:,} adultos  ·  diferença de {fmt(noruega['diferenca_pp'], 0)} pp [{fmt(noruega['ic'][0], 0)}; {fmt(noruega['ic'][1], 0)}]".replace(',', '.', 1))
print(f"\n  Fator ×0,80 aplicado: {'sim' if aplica_fator_080 else 'não'}")
print("=" * 72)
```

```python
# --- O MODELO ---

# Conta 1 · O clássico
rugg_dif = (rugg['não permitiriam'] - rugg['proibiriam']) * 100
rugg_nao_proibir = (1 - rugg['proibiriam']) * 100           # não proibiriam
rugg_permitir = (1 - rugg['não permitiriam']) * 100         # permitiriam

# Conta 2 · O efeito ao longo do tempo (pontos percentuais)
ep_nor = (noruega['ic'][1] - noruega['ic'][0]) / (2 * 1.96)
sim_nor = rng.normal(noruega['diferenca_pp'], ep_nor, N_SIM)
p_nor = (sim_nor > 0).mean()
encolhimento = {'Rugg (1941)': rugg_dif, 'Média de 52 perguntas (Holleman 1999)': meta['diferenca_pp'],
                'Noruega (2015)': noruega['diferenca_pp']}

# Conta 3 · O experimento grande: d virado em duelo
# Chance de uma resposta à versão negativa ser mais favorável ao tema do que uma resposta à versão positiva.
def duelo(d): return stats.norm.cdf(d / np.sqrt(2)) * 100
ep_d = utrecht['implicitas']['d'] / utrecht['implicitas']['z']   # erro padrão recuperado do z publicado
sim_d = rng.normal(utrecht['implicitas']['d'], ep_d, N_SIM)
duelo_sim = duelo(sim_d)
duelo_ic = np.percentile(duelo_sim, [2.5, 50, 97.5])
duelo_explicitas = duelo(utrecht['explicitas']['d'])

# Conta 4 · Quem muda mais
duelo_menos = duelo(utrecht['d_menos_informados'])
duelo_mais = duelo(utrecht['d_mais_informados'])
razao_d = utrecht['d_menos_informados'] / utrecht['d_mais_informados']

print("=" * 72)
print("  MODELO - RESULTADO")
print("=" * 72)
print("\n  CONTA 1 · O clássico")
print(f"  → Diferença: {fmt(rugg_dif, 0)} pontos percentuais")
print(f"  → {fmt(rugg_nao_proibir, 0)}% não proibiriam, mas só {fmt(rugg_permitir, 0)}% permitiriam")
print("\n  CONTA 2 · O efeito ao longo do tempo")
for k, v in encolhimento.items():
    print(f"  → {k:<40} {fmt(v)} pp")
print(f"  → Noruega: P(diferença na direção do clássico) = {fmt(p_nor*100, 0)}%")
print("\n  CONTA 3 · O experimento grande (Utrecht, 2014)")
print(f"  → Pares como proibir/permitir: {fmt(duelo_ic[1], 1)} de 100 duelos [{fmt(duelo_ic[0], 1)}; {fmt(duelo_ic[2], 1)}]  ·  empate seria 50")
print(f"  → Pares com 'não': {fmt(duelo_explicitas, 1)} de 100")
print("\n  CONTA 4 · Quem muda mais")
print(f"  → Menos informados sobre política: {fmt(duelo_menos, 0)} de 100  ·  mais informados: {fmt(duelo_mais, 0)} de 100")
print(f"  → O efeito é {fmt(razao_d, 1)} vezes maior entre os menos informados")
print("=" * 72)
```

```python
# --- VISUALIZAÇÃO ---

# GRÁFICO 1 · O clássico de 1941
fig1, ax1 = plt.subplots(figsize=(12, 8))
rot = ['Deveria permitir?\nresponderam "não"', 'Deveria proibir?\nresponderam "sim"']
b1 = ax1.bar(rot, [rugg['não permitiriam'] * 100, rugg['proibiriam'] * 100], color=[VERDE, DOURADO], width=0.45)
for b in b1:
    ax1.text(b.get_x() + b.get_width()/2, b.get_height() + 1.5, f"{fmt(b.get_height(), 0)}%", ha='center', fontsize=18, fontweight='bold')
ax1.annotate('', xy=(1, rugg['proibiriam'] * 100 + 6), xytext=(0, rugg['não permitiriam'] * 100 + 6), arrowprops=dict(arrowstyle='->', color=VERMELHO, lw=2))
ax1.text(0.5, 84, f"{fmt(rugg_dif, 0)} pontos", ha='center', fontsize=14, color=VERMELHO, fontweight='bold')
ax1.set_ylim(0, 90); ax1.set_ylabel('Contra os discursos (%)'); ax1.tick_params(axis='x', labelsize=13)
ax1.set_title('A mesma opinião, duas palavras\nDiscursos públicos contra a democracia, Estados Unidos, 1941', fontsize=14, pad=15)
plt.figtext(0.5, 0.005, 'Fonte: Rugg (1941), Public Opinion Quarterly 5(1) | #365Probabilidades', ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=[0, 0.03, 1, 1])
plt.savefig('dia-164-grafico-01-o-classico.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# GRÁFICO 2 · O efeito ao longo do tempo
fig2, ax2 = plt.subplots(figsize=(12, 8))
nomes = list(encolhimento); ys = np.arange(len(nomes))[::-1]
for y, n in zip(ys, nomes):
    v = encolhimento[n]
    if n.startswith('Noruega'):
        ax2.errorbar(v, y, xerr=[[v - noruega['ic'][0]], [noruega['ic'][1] - v]], fmt='o', color=VERDE, markersize=14, capsize=7, linewidth=2.5)
        ax2.text(noruega['ic'][1] + 0.6, y, f"{fmt(v, 0)} pp [{fmt(noruega['ic'][0], 0)}; {fmt(noruega['ic'][1], 0)}]", va='center', fontsize=13)
    else:
        ax2.plot(v, y, 'o', color=DOURADO, markersize=14)
        ax2.text(v + 0.6, y, f"{fmt(v)} pp", va='center', fontsize=13)
ax2.axvline(0, color=CINZA, linestyle='--', linewidth=1.2)
ax2.set_yticks(ys); ax2.set_yticklabels(['Rugg\n1941', 'Média de 52 perguntas\n(Holleman, 1999)', 'Noruega\n2015, 2.182 adultos'], fontsize=13)
ax2.set_xlim(-1, 27); ax2.set_ylim(-0.6, 2.6)
ax2.set_xlabel('Diferença entre "não permitir" e "proibir" (pontos percentuais)')
ax2.set_title('O efeito existe, e é menor do que no clássico\nA diferença entre as duas versões, em três momentos', fontsize=14, pad=15)
plt.figtext(0.5, 0.005, 'Fontes: Rugg (1941); Holleman (1999); estudo norueguês (2015). Conta própria | #365Probabilidades', ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=[0, 0.03, 1, 1])
plt.savefig('dia-164-grafico-02-ao-longo-do-tempo.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# GRÁFICO 3 · Quem muda mais (Utrecht)
fig3, ax3 = plt.subplots(figsize=(12, 8))
rot3 = ['Mais informados\nsobre política', 'Todos os\n31.112 eleitores', 'Menos informados\nsobre política']
vals3 = [duelo_mais, duelo_ic[1], duelo_menos]
b3 = ax3.barh(rot3, [v - 50 for v in vals3], left=50, color=[CINZA, DOURADO, VERDE], height=0.5)
for i, v in enumerate(vals3):
    ax3.text(v + 0.3, i, f"{fmt(v, 0)} de 100", va='center', fontsize=15, fontweight='bold')
ax3.axvline(50, color=CINZA, linestyle='--', linewidth=1.2); ax3.text(50.1, 2.4, 'empate', fontsize=11, color=CINZA)
ax3.set_xlim(48, 66); ax3.tick_params(axis='y', labelsize=13)
ax3.set_xlabel('Vezes em 100 que a resposta à versão negativa é mais favorável ao tema')
ax3.set_title('A palavra pesa mais para quem conhece menos o assunto\nEleições municipais de Utrecht, Holanda, 2014', fontsize=14, pad=15)
plt.figtext(0.5, 0.005, 'Fonte: Holleman, Kamoen, Krouwel, van de Pol & de Vreese (2016), PLOS ONE 11(10). Conta própria | #365Probabilidades', ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=[0, 0.03, 1, 1])
plt.savefig('dia-164-grafico-03-quem-muda-mais.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**Em 1941, "não permitir" e "proibir" separaram 21 pontos percentuais da mesma opinião. No maior experimento recente, com 31.112 eleitores, a diferença existiu, mas foi pequena.**

A troca de palavra muda a resposta, e isso aparece há mais de 80 anos. A média de 52 perguntas reunidas por Holleman é de 14 pontos. Mas os estudos grandes de hoje encontram um efeito bem menor. Na Noruega, foram 2 pontos. Em Utrecht, a resposta à versão negativa foi mais favorável ao tema em 54 de cada 100 comparações, quando um empate daria 50.

O efeito não é igual para todo mundo. Entre os menos informados sobre política, a conta sobe para 61 em 100. Entre os mais informados, fica em 52, quase um empate.

O que os dados permitem dizer é que a palavra pesa mais quando a opinião é mais fraca. Quem tem uma posição firme responde parecido às duas versões. Quem não pensou muito no assunto se deixa levar mais pela forma da pergunta.

A pergunta que fica:

*Da próxima vez que ler o resultado de uma pesquisa, você vai procurar como a pergunta foi escrita?*

### ⚠️ Limitações do Modelo

- O N de Rugg (1941) não foi localizado, e os números do clássico foram conferidos em fonte secundária. Por isso a Conta 1 não tem intervalo.
- A meta-análise de Holleman (1999) foi lida no resumo e na descrição feita por Kamoen, Holleman e Van den Bergh (2007). O desvio-padrão entre as 52 perguntas não foi conferido, e por isso a variação entre estudos não entrou na conta. O resumo diz que essa variação é grande, e que em alguns estudos o efeito vai no sentido contrário.
- Os quatro estudos medem coisas diferentes. Rugg, Holleman e o estudo norueguês dão a diferença em pontos percentuais. O estudo de Utrecht dá o tamanho do efeito (d) numa escala de cinco pontos. Os números não devem ser comparados diretamente, e a ideia de que o efeito "encolheu" vale como direção, e não como medida.
- O duelo transforma o d em uma chance, supondo respostas com distribuição normal. É uma aproximação nossa. O erro padrão do d de Utrecht foi recuperado do valor z publicado.
- Os números por nível de informação política vêm do conjunto de seis afirmações do tipo "proibir e permitir", e o artigo não dá o intervalo de cada um.
- O estudo norueguês foi lido no resumo e seus autores não foram conferidos. Ele trata só de política de drogas, numa pesquisa online.
- O estudo de Utrecht cobre uma cidade e uma eleição, num aplicativo de orientação de voto, e um dos autores dirige a empresa que construiu o aplicativo.
- Nenhum dos estudos foi feito no Brasil ou em português.

*A ciência é honesta sobre o que não sabe. O modelo também.*
