### 📖 A História

Passei quase vinte anos liderando equipes.

E numa carreira em mercado financeiro, "dar feedback direto" é celebrado como cultura de alta performance. Você é treinada pra ser clara. Pra não enrolar. Pra dizer o que precisa ser dito.

Eu era boa nisso. Achava até que estava fazendo um favor. "Melhor ouvir de mim do que descobrir depois." "Não vou deixar meu time no escuro." "Feedback é presente."

Frases bonitas. Que eu repeti muitas vezes.

E que, honestamente, nunca testei.

Nunca voltei pra medir o que aconteceu depois. As pessoas melhoraram? Ficaram iguais? Ou pioraram? Eu não sabia. Assumia que estavam melhorando, porque a lógica parecia óbvia: aponto o problema, a pessoa corrige.

Só que a lógica era minha suposição, não o dado.

E o pivô que me atropelou foi perceber que eu fazia a mesma coisa em casa. Com o Edgar. "Está errado, é assim." "Você não prestou atenção." "Presta atenção agora." Eu, que exigia rigor de mim mesma no trabalho, nunca tinha rigorosamente medido se o meu feedback estava ajudando as pessoas que eu mais amava.

Fui atrás dos dados. E o que encontrei é desconfortável.

### 📚 O Conceito: O Feedback Que Não Ajuda

Em 1996, Avraham Kluger e Angelo DeNisi publicaram uma meta-análise no Psychological Bulletin que reorganizou o entendimento sobre feedback.

O ponto de partida era simples: será que feedback sempre melhora o desempenho, como a gente assume?

Eles agregaram 607 tamanhos de efeito, cobrindo mais de 23 mil observações, sobre intervenções de feedback em contextos variados. E o achado central quebra o mito.

Na média, feedback melhora o desempenho (efeito d de 0,41, moderado). Mas essa média esconde algo importante: em mais de um terço dos casos analisados, o feedback piorou o desempenho de quem o recebeu.

Trinta e oito por cento. Quase quatro em cada dez intervenções de feedback estudadas tornaram a pessoa pior do que ela era antes.

Isso não é ruído estatístico. É o próprio padrão dos dados.

E os autores foram além. Propuseram uma teoria pra explicar quando o feedback ajuda e quando ele destrói: a Feedback Intervention Theory.

A ideia central é sobre para onde a atenção da pessoa vai depois que ela recebe o feedback.

Se a atenção fica **na tarefa** ("como eu faço isso melhor?"), o feedback tende a melhorar o desempenho.

Se a atenção sobe pra **si mesma** ("eu sou incompetente", "estou sendo julgada", "não sirvo pra isso"), o desempenho despenca. A energia mental que devia estar na tarefa vai pra defesa da autoimagem.

E feedback negativo, especialmente quando é sobre a pessoa e não sobre o comportamento, é justamente o tipo mais eficaz em fazer essa atenção subir.

### 🧮 O Modelo

**Fontes:**
- Kluger, A. N. & DeNisi, A. (1996) — "The Effects of Feedback Interventions on Performance: A Historical Review, a Meta-Analysis, and a Preliminary Feedback Intervention Theory", Psychological Bulletin, 119(2), 254-284 — **607 tamanhos de efeito, 23.663 observações**
- Fong, C. J., Patall, E. A., Vasquez, A. C. & Stautberg, S. (2019) — "A Meta-Analysis of Negative Feedback on Intrinsic Motivation", Educational Psychology Review, 31(1), 121-162 — refinamento com foco em feedback negativo e motivação
- Villeval, M. C. (2020) — revisão sobre efeitos comportamentais de feedback positivo e negativo em economia experimental

**Nota metodológica:** neste dia **não aplico o fator de correção ×0.80**. Os números são tamanhos de efeito (d de Cohen) e proporções sobre estudos analisados numa meta-análise, não proporções de survey populacional. Não há viés de amostragem populacional a corrigir com um multiplicador.

A limitação relevante é outra: a meta-análise é de 1996 e usa base majoritariamente organizacional e laboratorial. O achado central (feedback pode piorar) foi confirmado em revisões posteriores, mas magnitudes específicas variam. Trato os números como ordem de grandeza robusta, não como precisão cirúrgica.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Kluger, A. N. & DeNisi, A. (1996), Psychological Bulletin 119(2):254-284
# Meta-análise sobre intervenções de feedback e desempenho

n_effect_sizes = 607           # tamanhos de efeito agregados
n_observations = 23663         # observações totais na meta-análise
d_medio = 0.41                 # efeito médio (Cohen's d, moderado positivo)
p_efeito_negativo = 0.38       # proporção de casos em que feedback PIOROU o desempenho

# Complementar: dos casos com feedback, a divisão aproximada
p_melhorou = 0.50              # ~metade melhorou
p_neutro = 0.12                # ~12% neutro
p_piorou = 0.38                # ~38% piorou

# NOTA: não se aplica o fator de correção ×0.80 neste dia.
# São tamanhos de efeito de meta-análise (d) e proporções sobre estudos analisados,
# não proporções de survey populacional. A limitação a declarar é a idade da
# meta-análise (1996) e a base majoritariamente organizacional/laboratorial.

print("=" * 68)
print("  DADOS — INTERVENÇÕES DE FEEDBACK (Kluger & DeNisi, 1996)")
print("=" * 68)
print(f"\n  Meta-análise:            {n_effect_sizes} tamanhos de efeito")
print(f"                           {n_observations:,} observações totais")
print(f"\n  Efeito MÉDIO:")
print(f"  → d de Cohen:            {d_medio}  (efeito moderado, positivo)")
print(f"\n  MAS a média esconde a distribuição:")
print(f"  → Feedback MELHOROU:     ~{p_melhorou*100:.0f}% dos casos")
print(f"  → Feedback NEUTRO:       ~{p_neutro*100:.0f}% dos casos")
print(f"  → Feedback PIOROU:       {p_piorou*100:.0f}% dos casos")
print(f"\n  Quase 4 em cada 10 intervenções de feedback ANALISADAS")
print(f"  fizeram a pessoa que as recebeu ficar PIOR do que antes.")
print("=" * 68)
```

```python
# --- O MODELO ---
# O mecanismo: para onde a atenção da pessoa vai depois do feedback

# Feedback Intervention Theory (FIT) — Kluger & DeNisi (1996)
# O feedback ajuda ou destrói dependendo do NÍVEL para onde ele desloca a atenção

niveis = {
    'TAREFA':
        'Atenção fica em COMO fazer isso melhor. Feedback tende a AJUDAR.',
    'META-TAREFA':
        'Atenção sobe pro processo. Efeito misto, depende do contexto.',
    'SELF':
        'Atenção sobe pra AUTOIMAGEM ("eu sou incompetente"). Feedback DESTRÓI.',
}

print("=" * 68)
print("  MODELO — FEEDBACK INTERVENTION THEORY")
print("=" * 68)
print(f"\n  Feedback direciona a atenção da pessoa para algum nível.")
print(f"  O efeito no desempenho depende de qual nível:\n")

for nivel, desc in niveis.items():
    print(f"  → {nivel}")
    print(f"    {desc}\n")

print(f"  A REGRA PRÁTICA:")
print(f"  → 'Nesse ponto específico, o cálculo pode ser refeito assim'")
print(f"     → mantém atenção NA TAREFA → tende a ajudar")
print(f"\n  → 'Você não presta atenção' / 'Você é assim'")
print(f"     → sobe pro SELF → tende a piorar")

print(f"\n  A LEITURA DIFÍCIL:")
print(f"  → Feedback pensado pra ajudar, feito por gente que quer ajudar,")
print(f"     em {p_piorou*100:.0f}% dos casos PIORA a pessoa que o recebe.")
print(f"  → A intenção não protege do efeito.")
print("=" * 68)
```

```python
# --- VISUALIZAÇÃO ---

cor_melhor = '#2a8a82'
cor_neutro = '#9c9b94'
cor_pior = '#c0392b'
cor_ouro = '#c8a84b'

# ── GRÁFICO 1 — A média esconde a distribuição ──
fig1, ax1 = plt.subplots(figsize=(11, 6.5))

categorias = ['Feedback\nMELHOROU\ndesempenho',
              'Efeito\nNEUTRO',
              'Feedback\nPIOROU\ndesempenho']
proporcoes = [p_melhorou*100, p_neutro*100, p_piorou*100]
cores = [cor_melhor, cor_neutro, cor_pior]

bars = ax1.bar(categorias, proporcoes, color=cores, alpha=0.88, width=0.55)
ax1.set_ylabel('Proporção dos estudos analisados (%)')
ax1.set_ylim(0, 60)
ax1.set_title('A média esconde o que importa\nFeedback melhora em média — mas em 38% dos casos, piora',
              fontsize=13, pad=15)

for bar, v in zip(bars, proporcoes):
    ax1.text(bar.get_x() + bar.get_width()/2, v + 1.5,
             f'{v:.0f}%', ha='center', fontweight='bold', fontsize=16)

plt.figtext(0.5, 0.005,
            'Fonte: Kluger & DeNisi (1996), Psychological Bulletin — 607 tamanhos de efeito, 23.663 observações | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-032-grafico-01-distribuicao.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 — Para onde a atenção vai (o mecanismo) ──
fig2, ax2 = plt.subplots(figsize=(11, 6.5))

niveis_labels = ['NA TAREFA\n"como faço\nmelhor?"',
                 'NO PROCESSO\n"que abordagem\nusar?"',
                 'NO SELF\n"eu sou\nincompetente"']
efeitos = [1.0, 0.2, -0.9]   # ilustrativo, direção do efeito
cores_n = [cor_melhor, cor_neutro, cor_pior]

bars = ax2.bar(niveis_labels, efeitos, color=cores_n, alpha=0.88, width=0.5)
ax2.axhline(y=0, color='#333', linewidth=1)
ax2.set_ylabel('Direção do efeito no desempenho (ilustrativa)')
ax2.set_ylim(-1.3, 1.4)
ax2.set_yticks([])
ax2.set_title('Para onde a atenção vai depois do feedback\nA mesma frase pode ajudar ou destruir, dependendo de onde a atenção sobe',
              fontsize=13, pad=15)

for bar, e in zip(bars, efeitos):
    if e > 0:
        ax2.text(bar.get_x() + bar.get_width()/2, e + 0.06, 'melhora',
                 ha='center', fontsize=12, color=cor_melhor, fontweight='bold', style='italic')
    elif e < 0:
        ax2.text(bar.get_x() + bar.get_width()/2, e - 0.15, 'piora',
                 ha='center', fontsize=12, color=cor_pior, fontweight='bold', style='italic')
    else:
        ax2.text(bar.get_x() + bar.get_width()/2, 0.1, 'misto',
                 ha='center', fontsize=11, color='#666', style='italic')

plt.figtext(0.5, 0.005,
            'Ilustração conceitual da Feedback Intervention Theory (Kluger & DeNisi, 1996) | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-032-grafico-02-mecanismo.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 — A distribuição do d nos estudos ──
fig3, ax3 = plt.subplots(figsize=(11, 6.5))

# Distribuição ilustrativa dos tamanhos de efeito, centrada em d=0.41
# com cauda esquerda mostrando que ~38% ficam abaixo de zero
np.random.seed(42)
efeitos_dist = np.random.normal(0.41, 0.7, 5000)

# Histograma
counts, bins, _ = ax3.hist(efeitos_dist, bins=45, color=cor_neutro, alpha=0.4, edgecolor='none')

# Colorir por região
for i in range(len(bins)-1):
    if bins[i] < 0:
        ax3.bar((bins[i]+bins[i+1])/2, counts[i], width=bins[i+1]-bins[i],
                color=cor_pior, alpha=0.75)
    elif bins[i+1] > 0:
        ax3.bar((bins[i]+bins[i+1])/2, counts[i], width=bins[i+1]-bins[i],
                color=cor_melhor, alpha=0.6)

ax3.axvline(x=0.41, color=cor_ouro, linewidth=2.5, linestyle='--')
ax3.axvline(x=0, color='#333', linewidth=1.5)

ax3.text(0.41, ax3.get_ylim()[1]*0.88, f'  média\n  d = 0,41',
         fontsize=11, color='#8a7020', fontweight='bold')
ax3.text(-1.4, ax3.get_ylim()[1]*0.70, '38% dos estudos\nficaram AQUI\n(feedback PIOROU)',
         fontsize=11, color=cor_pior, fontweight='bold', ha='left')

ax3.set_xlabel('Tamanho de efeito do feedback no desempenho (d de Cohen)')
ax3.set_ylabel('Frequência (ilustrativa)')
ax3.set_title('A distribuição dos efeitos\nA média positiva convive com uma cauda negativa grande',
              fontsize=13, pad=15)
ax3.set_xlim(-2.5, 3)

plt.figtext(0.5, 0.005,
            'Distribuição ilustrativa consistente com Kluger & DeNisi (1996) | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-032-grafico-03-distribuicao-d.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**Em 38% dos casos analisados, o feedback piorou o desempenho de quem o recebeu.**

Quase quatro em cada dez.

Não é sobre a pessoa que dá o feedback ser má intencionada. Kluger e DeNisi olharam intervenções pensadas pra ajudar, feitas por gente que queria ajudar. E ainda assim, quatro em cada dez vezes, o resultado foi o oposto do pretendido.

O mecanismo é claro: quando o feedback desloca a atenção da tarefa pra si mesmo, o desempenho cai. E feedback negativo, especialmente quando é dado sobre a pessoa e não sobre o comportamento, é o tipo que mais causa esse deslocamento.

"Você não presta atenção" desloca. "Você é assim." desloca. "Eu esperava mais de você" desloca.

"Nesse ponto específico, o cálculo pode ser refeito assim" não desloca. Mantém na tarefa.

A diferença parece pequena. Nos dados, ela decide se a pessoa vai melhorar ou piorar depois de te ouvir.

Eu passei anos entregando feedback direto no trabalho, orgulhosa da minha franqueza. E a mesma pessoa que exigia dados pra tudo nunca tinha medido se essa franqueza estava ajudando ou destruindo. Assumi. E assumir é o oposto do rigor que eu defendia em todo o resto.

O que me pegou não foi descobrir que feedback pode fazer mal. Foi perceber que a probabilidade de fazer mal não é pequena. É quatro em dez.

E que essa mesma probabilidade se aplica no escritório e em casa. Na conversa com o time e na conversa com o filho.

*Da última vez que você apontou o que estava errado em alguém, você tem alguma forma real de saber se a pessoa ficou melhor ou pior depois?*

### ⚠️ Limitações do Modelo

- A meta-análise original é de 1996. O campo teve refinamentos importantes desde então (Fong et al. 2019, entre outros), e a magnitude exata do efeito varia por contexto. O achado central (feedback pode piorar em proporção significativa dos casos) é robusto; os 38% são a estimativa dessa meta-análise específica.
- A base dos estudos é majoritariamente organizacional e laboratorial. Generalizar diretamente para relações íntimas (família, casal) requer cautela; o mecanismo parece se aplicar, mas a magnitude pode ser diferente.
- "Piorar o desempenho" no estudo significa piorar a tarefa medida no experimento. Não capta efeitos emocionais ou de longo prazo que também importam.
- Feedback bem dado (focado em tarefa, específico, com caminho de melhoria) tem efeitos positivos consistentes. Este texto não é contra feedback, é contra dar feedback sem medir se ele funciona.

*A ciência é honesta sobre o que não sabe. O modelo também.*
