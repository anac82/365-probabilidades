### 📖 A História

Todo mundo tem um exame do ano esperando para ser marcado. Ele não tem data, ninguém cobra,
ninguém liga para lembrar. Enquanto isso, o e-mail que chegou com "urgente" no assunto é
respondido em quatro minutos.

A explicação de sempre é que o urgente é mais importante. Muitas vezes é mesmo. A pergunta
deste dia é outra. O que acontece quando o urgente vale menos, e a gente sabe que vale menos?

Este é o sétimo dia da trilha do tempo. O #137 mediu a crença de que mês que vem vai sobrar
tempo. Este mede o que acontece com o tempo de hoje quando um prazo curto aparece na frente.

---

### 📚 O Conceito: o efeito da mera urgência

Em 2018, Meng Zhu, Yang Yang e Christopher Hsee publicaram no *Journal of Consumer Research*
cinco experimentos com o mesmo desenho. A pessoa escolhe entre duas versões da mesma tarefa de
digitação, de 3 minutos. Uma paga US$ 0,12 e a outra US$ 0,16. Sem prazo curto, quase ninguém
escolhe a que paga menos. Quando a que paga menos passa a expirar em 5 minutos, e a outra em 50,
muita gente passa a escolher a pior. Ela paga menos e ainda dá menos tempo para ser feita. Não
existe motivo racional para escolhê-la.

Os autores chamaram isso de efeito da mera urgência. A explicação deles é de atenção. O prazo
curto chama o olho, e o valor de cada opção sai de foco na hora da escolha. Em outros
experimentos do mesmo artigo, o efeito foi mais forte em quem se considera ocupado e diminuiu
quando o valor foi lembrado no momento de escolher.

Um efeito descoberto por um só grupo pede réplica. Ela veio em 2023, numa tese de doutorado da
Universidade de Nottingham. Richard Mills, orientado por Chris Starmer e Robin Cubitt, pediu os
dados e o questionário à própria Meng Zhu, refez o experimento quatro vezes com 2.003 pessoas e,
a cada rodada, tirou uma possível falha do desenho original. A opção pior aparecia sempre em
primeiro lugar na tela. O valor de cada tarefa sumia da tela na hora de escolher. Ninguém
avisava que dava para ir embora depois dos 3 minutos. O efeito encolheu a cada correção. E não
sumiu em nenhuma.

### 🧮 O Modelo

**Fontes:**
- Zhu, M., Yang, Y. & Hsee, C. K. (2018). The mere urgency effect. *Journal of Consumer
  Research*, 45(3), 673-690. doi 10.1093/jcr/ucy008. Cinco experimentos; o Exp. 2B (Amazon
  Mechanical Turk, 204 observações na regressão da réplica) é o que foi replicado. Dados do
  Exp. 2B reanalisados por Mills com os mesmos números do artigo (7,3% e 48,1%).
- Mills, R. (2023). *Essays on time allocation*, capítulo 1, "The Mere Urgency Effect: an
  artefact of experimental design?". Tese de doutorado, Universidade de Nottingham, licença
  CC BY 4.0. Quatro réplicas no Mechanical Turk (EUA), 2019-2020: N = 400, 808, 647 e 315
  (total 2.003). Proporções por condição nas Tabelas A1 a A4; efeitos marginais na Tabela A8.

**Desenho e família do teste:** experimentos entre sujeitos com sorteio da condição; desfecho
binário (escolheu ou não a opção pior). A tese testa com qui-quadrado e regressão probit. Aqui
cada proporção recebe uma posterior Beta de Jeffreys, e as comparações são feitas por Monte
Carlo sobre as posteriores.

**Camada brasileira (o exame do ano):** IBGE, Pesquisa Nacional de Saúde 2019, microdados
(moradores selecionados, pesos V00291), já baixados por Ana para o #149 e recortados em
`dados151/dia-151-pns2019-exames.csv`. Variáveis R014 e R015 (mamografia pedida por médico e
feita), R01701 (última mamografia), R00101 (último preventivo) e R002010 (principal motivo de
não ter feito o preventivo nos últimos três anos). Só mulheres, porque o módulo R é feminino.
Proporções ponderadas pelo peso da pesquisa, com intervalo por reamostragem das unidades
primárias de amostragem (UPA).

**×0,80 só na camada brasileira,** e só nas proporções de quem diz ter feito o exame em dia,
porque é autorrelato de comportamento desejável. As escolhas do experimento ficam sem o fator.
Os motivos ficam sem o fator, porque são a distribuição de uma resposta e não uma prevalência.

**Cordas declaradas:** #137 (Zauberman & Lynch 2005, a folga de tempo que a gente espera no
futuro), #107 (falácia do planejamento) e #044 (procrastinação). Objetos distintos. Christopher
Hsee volta como fonte no #179 (aversão à ociosidade), a declarar lá. O sobrenome Zhu aparece no
#060 em outro artigo, de outra autora; não há reuso.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

SEMENTE = 149
rng = np.random.default_rng(SEMENTE)
SORTEIOS = 200_000

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

VERMELHO, DOURADO, VERDE, CINZA = '#c0392b', '#c8a84b', '#1a5f5a', '#6b6a64'
RODAPE = ('Fonte: Zhu, Yang & Hsee (2018), Journal of Consumer Research 45(3) · Mills (2023), tese, Univ. de Nottingham, '
          'quatro réplicas, N = 2.003 · cálculo próprio | #365Probabilidades')

def vir(x, casas=1):
    return f"{x:.{casas}f}".replace(".", ",")

def posterior(p, n):
    # Beta de Jeffreys a partir da proporção publicada e do N do braço
    k = p * n
    return stats.beta(k + 0.5, n - k + 0.5)

print("✅ Bibliotecas carregadas · semente", SEMENTE)
```

```python
import pandas as pd
# --- DADOS DA LITERATURA ---
# N de cada rodada e número de braços sorteados com probabilidade igual (Mills 2023, Tab. 1.5 e A8).
# As tabelas A1 a A4 trazem a proporção de cada braço, não o seu N. O N por braço abaixo é a divisão
# por igual, aproximação declarada nas limitações.
RODADAS = {
    'Original': {'N': 204, 'bracos': 2, 'nome': 'Zhu et al. 2018, Exp. 2B'},
    'Réplica 1': {'N': 400, 'bracos': 2, 'nome': 'mesmas instruções'},
    'Réplica 2': {'N': 808, 'bracos': 5, 'nome': 'ordem sorteada e três correções'},
    'Réplica 3': {'N': 647, 'bracos': 4, 'nome': 'ordem sorteada e correções combinadas'},
    'Réplica 4': {'N': 315, 'bracos': 2, 'nome': 'instruções reescritas do zero'},
}

# Proporção que escolheu a opção pior (paga 12 centavos em vez de 16)
# tipo: controle (sem prazo curto) · urgencia (prazo curto, desenho original) · aviso (só avisa que dá para sair)
#       justo (pagamento e prazo na mesma tela, ordem sorteada) · antidoto (prazo primeiro, pagamento por último)
CONDICOES = [
    ('Original',  'Sem prazo curto',                       'controle', 0.073),
    ('Original',  'Com prazo curto',                       'urgencia', 0.481),
    ('Réplica 1', 'Sem prazo curto',                       'controle', 0.116),
    ('Réplica 1', 'Com prazo curto',                       'urgencia', 0.436),
    ('Réplica 2', 'Sem prazo curto',                       'controle', 0.087),
    ('Réplica 2', 'Com prazo curto',                       'urgencia', 0.453),
    ('Réplica 2', 'Prazo curto + aviso de saída',          'aviso',    0.399),
    ('Réplica 2', 'Prazo curto, tudo na mesma tela',       'justo',    0.256),
    ('Réplica 2', 'Prazo curto, valor visto por último',   'antidoto', 0.09),
    ('Réplica 3', 'Sem prazo curto',                       'controle', 0.085),
    ('Réplica 3', 'Com prazo curto',                       'urgencia', 0.355),
    ('Réplica 3', 'Prazo curto, mesma tela, valor antes',  'justo',    0.21),
    ('Réplica 3', 'Prazo curto, mesma tela, prazo antes',  'justo',    0.25),
    ('Réplica 4', 'Sem prazo curto',                       'controle', 0.113),
    ('Réplica 4', 'Com prazo curto, instruções novas',     'justo',    0.359),
]

for i, (r, nome, tipo, p) in enumerate(CONDICOES):
    n = RODADAS[r]['N'] / RODADAS[r]['bracos']
    CONDICOES[i] = {'rodada': r, 'nome': nome, 'tipo': tipo, 'p': p, 'n': n}

# Apoios da tese (texto da seção 1.5 e Figura A3)
ATENCAO_PRAZO = 0.291    # quem prestou mais atenção ao prazo que ao valor: +29,1 pontos de chance de escolher pior
ATENCAO_VALOR = -0.235   # quem prestou mais atenção ao valor: -23,5 pontos
ORDEM_R4 = {'opção pior em primeiro': 0.51, 'opção pior em segundo': 0.21}   # Réplica 4, com prazo curto
COMECOU_DEPOIS = (13, 0.006)   # só 13 pessoas (0,6%) usaram o prazo longo; o curto não foi usado como compromisso

print("=" * 72)
print("  DADOS · QUEM ESCOLHEU A OPÇÃO PIOR (paga menos e tem menos tempo)")
print("=" * 72)
atual = None
for c in CONDICOES:
    if c['rodada'] != atual:
        atual = c['rodada']
        info = RODADAS[atual]
        print(f"\n  {atual} · {info['nome']} · N = {info['N']} em {info['bracos']} braços")
    print(f"    {c['nome']:<40} {vir(c['p'] * 100):>5}%   (N do braço ≈ {round(c['n'])})")
print(f"\n  Atenção ao prazo maior que ao valor: {vir(ATENCAO_PRAZO * 100)} pontos a mais de escolher pior")
print(f"  Atenção ao valor maior que ao prazo: {vir(ATENCAO_VALOR * 100)} pontos")
print(f"  Ordem na tela (Réplica 4, prazo curto): pior em primeiro {vir(ORDEM_R4['opção pior em primeiro'] * 100, 0)}% · "
      f"pior em segundo {vir(ORDEM_R4['opção pior em segundo'] * 100, 0)}%")
print("=" * 72)


# --- CAMADA BRASILEIRA · PNS 2019 ---
PNS = pd.read_csv('dados151/dia-151-pns2019-exames.csv')
PNS = PNS[PNS['C006'] == 2]   # módulo R: mulheres
MOTIVOS = {2: 'não acha necessário', 3: 'tem vergonha', 7: 'espera longa no serviço',
           4: 'não foi orientada', 9: 'horário incompatível', 6: 'dificuldade financeira',
           8: 'serviço distante', 11: 'está marcado, mas não fez'}
print("\n  PNS 2019 · mulheres entrevistadas no módulo de saúde da mulher:", f"{len(PNS):,}".replace(",", "."))
```

```python
# --- O MODELO ---

def sortear(c, k=SORTEIOS):
    return posterior(c['p'], c['n']).rvs(size=k, random_state=rng)

def agrupar(lista):
    # média ponderada pelo N do braço, sorteio a sorteio
    pesos = np.array([c['n'] for c in lista])
    amostras = np.vstack([sortear(c) for c in lista])
    return (pesos[:, None] * amostras).sum(axis=0) / pesos.sum()

def ic(x):
    return np.percentile(x, [2.5, 97.5])

# CAMADA 1 · A escada
print("=" * 72)
print("  CAMADA 1 · A ESCADA (posterior de Jeffreys por condição)")
print("=" * 72)
for c in CONDICOES:
    lo, hi = posterior(c['p'], c['n']).interval(0.95)
    c['ic'] = (lo, hi)
    print(f"  {c['rodada']:<10} {c['nome']:<40} {vir(c['p'] * 100):>5}%  IC 95% [{vir(lo * 100)}; {vir(hi * 100)}]")

# CAMADA 2 · O teste justo contra o sem prazo curto
controles = [c for c in CONDICOES if c['tipo'] == 'controle' and c['rodada'] != 'Original']
justos = [c for c in CONDICOES if c['tipo'] == 'justo']
original_u = [c for c in CONDICOES if c['rodada'] == 'Original' and c['tipo'] == 'urgencia'][0]
original_c = [c for c in CONDICOES if c['rodada'] == 'Original' and c['tipo'] == 'controle'][0]

P_JUSTO = agrupar(justos)
P_CONTROLE = agrupar(controles)
P_ORIG_U = sortear(original_u)
P_ORIG_C = sortear(original_c)
DIF = P_JUSTO - P_CONTROLE
RAZAO = P_JUSTO / P_CONTROLE
SOBRA = DIF / (P_ORIG_U - P_ORIG_C)

print("\n" + "=" * 72)
print("  CAMADA 2 · O TESTE JUSTO (as quatro condições em que valor e prazo aparecem juntos)")
print("=" * 72)
print(f"  Com prazo curto, teste justo:  {vir(P_JUSTO.mean() * 100)}%  IC 95% [{vir(ic(P_JUSTO)[0] * 100)}; {vir(ic(P_JUSTO)[1] * 100)}]")
print(f"  Sem prazo curto (réplicas):    {vir(P_CONTROLE.mean() * 100)}%  IC 95% [{vir(ic(P_CONTROLE)[0] * 100)}; {vir(ic(P_CONTROLE)[1] * 100)}]")
print(f"  Diferença:                     {vir(DIF.mean() * 100)} pontos  IC 95% [{vir(ic(DIF)[0] * 100)}; {vir(ic(DIF)[1] * 100)}]")
print(f"  Razão:                         {vir(RAZAO.mean())} vezes  IC 95% [{vir(ic(RAZAO)[0])}; {vir(ic(RAZAO)[1])}]")
p_pos = (DIF > 0).mean()
print(f"  P(diferença > 0):              " + ("acima de 99,99%" if p_pos > 0.9999 else f"{vir(p_pos * 100, 2)}%"))
print(f"  Em 100 pessoas: cerca de {round(P_JUSTO.mean() * 100)} escolhem a pior com prazo curto, "
      f"cerca de {round(P_CONTROLE.mean() * 100)} sem ele")
print(f"\n  O quanto encolheu: no original a urgência somava {vir((P_ORIG_U - P_ORIG_C).mean() * 100)} pontos;")
print(f"  no teste justo sobram {vir(DIF.mean() * 100)}. Fica {vir(SOBRA.mean() * 100, 0)}% do efeito original "
      f"(IC 95% [{vir(ic(SOBRA)[0] * 100, 0)}; {vir(ic(SOBRA)[1] * 100, 0)}]).")
print(f"  Heterogeneidade: as quatro condições justas vão de {vir(min(c['p'] for c in justos) * 100, 0)}% "
      f"a {vir(max(c['p'] for c in justos) * 100, 0)}%; o agrupamento ponderado é uma média, não uma lei.")

# CAMADA 3 · A semana
ESCOLHAS_SEMANA = 5     # premissa: cinco escolhas por semana entre um urgente que vale menos e um importante sem prazo
SEMANAS = 48
k_justo = rng.binomial(ESCOLHAS_SEMANA, P_JUSTO)
k_controle = rng.binomial(ESCOLHAS_SEMANA, P_CONTROLE)
DIST_JUSTO = np.bincount(k_justo, minlength=ESCOLHAS_SEMANA + 1) / SORTEIOS
DIST_CONTROLE = np.bincount(k_controle, minlength=ESCOLHAS_SEMANA + 1) / SORTEIOS
ao_menos_uma_j = 1 - DIST_JUSTO[0]
ao_menos_uma_c = 1 - DIST_CONTROLE[0]
ano_j = P_JUSTO * ESCOLHAS_SEMANA * SEMANAS
ano_c = P_CONTROLE * ESCOLHAS_SEMANA * SEMANAS

print("\n" + "=" * 72)
print(f"  CAMADA 3 · A SEMANA ({ESCOLHAS_SEMANA} escolhas por semana, Monte Carlo com " + f"{SORTEIOS:,}".replace(",", ".") + " semanas)")
print("=" * 72)
print(f"  Pelo menos uma escolha pior na semana:  com prazo curto {vir(ao_menos_uma_j * 100, 0)}% · sem ele {vir(ao_menos_uma_c * 100, 0)}%")
for k in range(ESCOLHAS_SEMANA + 1):
    print(f"    {k} escolha(s) pior(es): {vir(DIST_JUSTO[k] * 100):>5}%  contra {vir(DIST_CONTROLE[k] * 100):>5}%")
print(f"  Em {SEMANAS} semanas: cerca de {round(ano_j.mean())} escolhas piores com prazo curto, "
      f"{round(ano_c.mean())} sem ele; a urgência responde por cerca de {round((ano_j - ano_c).mean())}")
print("  Conta de bolso: supõe escolhas independentes e o efeito do laboratório levado para a vida.")

# CAMADA 4 · O antídoto
r2 = {c['nome']: c for c in CONDICOES if c['rodada'] == 'Réplica 2'}
A = sortear(r2['Prazo curto, valor visto por último'])
C = sortear(r2['Sem prazo curto'])
U = sortear(r2['Com prazo curto'])
DIF_A = A - C
print("\n" + "=" * 72)
print("  CAMADA 4 · O ANTÍDOTO (Réplica 2: mesmo prazo curto, valor visto por último)")
print("=" * 72)
print(f"  Com prazo curto, desenho original:   {vir(U.mean() * 100)}%")
print(f"  Com prazo curto, valor por último:   {vir(A.mean() * 100)}%")
print(f"  Sem prazo curto:                     {vir(C.mean() * 100)}%")
print(f"  Valor por último menos sem prazo:    {vir(DIF_A.mean() * 100)} pontos  IC 95% [{vir(ic(DIF_A)[0] * 100)}; {vir(ic(DIF_A)[1] * 100)}]")
print(f"  P(diferença dentro de ±5 pontos):    {vir((np.abs(DIF_A) < 0.05).mean() * 100, 0)}%")
print("  Com o valor por último diante dos olhos, a urgência some do resultado.")
print("=" * 72)


# CAMADA 5 · O exame do ano no Brasil (PNS 2019)
def prop_pond(df, cond, reps=500):
    w = df['V00291'].values
    p = (w * cond.values).sum() / w.sum()
    upas = df['UPA_PNS'].values
    unicos = np.unique(upas)
    idx = {u: np.where(upas == u)[0] for u in unicos}
    boot = []
    for _ in range(reps):
        sel = np.concatenate([idx[u] for u in rng.choice(unicos, len(unicos))])
        boot.append((w[sel] * cond.values[sel]).sum() / w[sel].sum())
    return p, np.percentile(boot, [2.5, 97.5])

print("\n" + "=" * 72)
print("  CAMADA 5 · O EXAME DO ANO NO BRASIL (PNS 2019, mulheres)")
print("=" * 72)
m50 = PNS[PNS['C008'].between(50, 69) & PNS['R014'].notna()]
MAMO, MAMO_IC = prop_pond(m50, m50['R01701'].isin([1, 2]))
print(f"  50 a 69 anos, mamografia nos últimos 2 anos: {vir(MAMO * 100)}%  IC 95% [{vir(MAMO_IC[0] * 100)}; {vir(MAMO_IC[1] * 100)}]  (N = " + f"{len(m50):,}".replace(",", ".") + ")")
print(f"     com ×0,80 (autorrelato): {vir(MAMO * 0.8 * 100)}% em dia, {vir((1 - MAMO * 0.8) * 100)}% atrasadas")
mp = PNS[PNS['R014'] == 1]
NAO_FEZ, NAO_FEZ_IC = prop_pond(mp, mp['R015'] == 2)
print(f"  Médico pediu mamografia e ela não fez: {vir(NAO_FEZ * 100)}%  IC 95% [{vir(NAO_FEZ_IC[0] * 100)}; {vir(NAO_FEZ_IC[1] * 100)}]  (N = " + f"{len(mp):,}".replace(",", ".") + ")")
m25 = PNS[PNS['C008'].between(25, 64) & PNS['R00101'].notna()]
PREV, PREV_IC = prop_pond(m25, m25['R00101'].isin([1, 2, 3]))
print(f"  25 a 64 anos, preventivo nos últimos 3 anos: {vir(PREV * 100)}%  IC 95% [{vir(PREV_IC[0] * 100)}; {vir(PREV_IC[1] * 100)}]")
print(f"     com ×0,80 (autorrelato): {vir(PREV * 0.8 * 100)}% em dia")
mot = PNS[PNS['C008'].between(25, 64) & PNS['R002010'].notna() & ~PNS['R002010'].isin([1, 12, 99])]
print(f"\n  Por que não fez o preventivo nos últimos 3 anos (25 a 64, N = " + f"{len(mot):,}".replace(",", ".") + "):")
MOT = {}
for k, nome in MOTIVOS.items():
    MOT[nome] = prop_pond(mot, mot['R002010'] == k, reps=200)
    print(f"   {nome:<28} {vir(MOT[nome][0] * 100)}%  [{vir(MOT[nome][1][0] * 100)}; {vir(MOT[nome][1][1] * 100)}]")
print("  O motivo que mais aparece não é falta de tempo. É achar que não precisa.")
print("=" * 72)
```

### 💡 O Insight

**Com prazo curto, cerca de 1 em cada 4 pessoas escolhe a tarefa que paga menos e ainda dá menos
tempo para ser feita. Sem o prazo curto, cerca de 1 em cada 10. No teste mais justo das quatro
réplicas, a diferença é de 17 pontos, e a chance fica 2,6 vezes maior.**

O número famoso do artigo de 2018 era maior, quase metade das pessoas. As réplicas mostraram que
parte dele vinha do jeito como a tela era montada. A opção pior aparecia sempre em primeiro lugar,
e o valor sumia da tela na hora de escolher. Corrigido tudo isso, sobram cerca de 4 em cada 10
pontos do efeito original. E esse resto não some com nenhuma correção.

Levado para uma semana com cinco escolhas desse tipo, a conta de bolso dá ao menos uma escolha
pior em cerca de 8 de cada 10 semanas, contra 4 de cada 10 sem o prazo curto na frente.

O achado mais útil está na quinta tela da segunda réplica. O mesmo prazo curto, as mesmas duas
tarefas, só que o valor de cada uma foi a última coisa que a pessoa viu antes de escolher. A
escolha pior caiu para 9%, o mesmo nível de quem nem tinha prazo curto. A urgência não muda o que
a gente valoriza. Ela muda para onde a gente olha.

### ⚠️ Limitações do Modelo

- O experimento é de laboratório online. A escolha é uma só, entre centavos, numa tarefa de
  digitação de 3 minutos. Levar o número para o e-mail urgente e o exame do ano é extrapolação,
  e a camada da semana é conta de bolso com cinco decisões por semana como premissa.
- Todas as réplicas usam trabalhadores do Mechanical Turk nos Estados Unidos, entre 2019 e 2020.
  Não há réplica com brasileiros nem fora da internet.
- As tabelas da tese dão a proporção de cada condição, não o seu N. Usamos o N da rodada
  dividido igualmente pelos braços, porque o sorteio foi com probabilidade igual. Os intervalos
  podem estar um pouco mais estreitos ou mais largos do que os reais.
- O "teste justo" agrupa quatro condições que vão de 21% a 36%. A média ponderada resume, mas a
  variação entre elas é real. A Réplica 4, com as instruções mais limpas, deu o número mais alto,
  e o próprio autor não sabe dizer por quê.
- A ordem na tela pesa sozinha. Na Réplica 4, com prazo curto, a opção pior foi escolhida por
  51% quando aparecia em primeiro e por 21% quando aparecia em segundo.
- A explicação pela atenção usa atenção autorrelatada depois da escolha. Medir com rastreamento
  do olhar ainda está por fazer.
- O antídoto (valor visto por último) vem de uma única condição da Réplica 2, com proporção
  publicada com duas casas (9%). É resultado promissor, não regra testada em vários contextos.
- A camada brasileira é só de mulheres, porque o módulo de saúde da mulher da PNS é o único que
  pergunta se um exame pedido foi feito e por quê. É autorrelato, de 2019, e mostra o atraso,
  não o efeito da urgência. Ninguém mediu na PNS o que as pessoas fizeram no lugar do exame.
- Os intervalos da PNS vêm de reamostragem das unidades primárias, sem usar os estratos. É uma
  aproximação do desenho da pesquisa.
- ×0,80 só nas proporções da PNS de quem diz estar em dia. As escolhas do experimento ficam sem
  o fator, porque são escolhas observadas.

*A ciência é honesta sobre o que não sabe. O modelo também.*
