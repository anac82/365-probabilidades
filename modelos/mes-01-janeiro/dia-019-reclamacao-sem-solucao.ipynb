### 📖 A História

Tem uma cena que se repete: alguém chega contando o mesmo problema de sempre. Você escuta, se importa, e depois de um tempo oferece uma ideia concreta — um caminho, um contato, uma sugestão prática. A resposta, quase sempre, é "sim, mas...". E na semana seguinte, a mesma pessoa volta com a mesma reclamação, intacta.

Isso pode parecer ingratidão, ou que sua ideia não era boa o suficiente. Mas a psicologia social tem uma explicação mais simples e, de certa forma, mais aliviadora: a maioria das reclamações nunca foi, desde o início, um pedido de solução.

Pesquisadores pediram para estudantes universitários registrarem, em diário, cada reclamação que faziam a outras pessoas durante alguns dias. O resultado: a grande maioria dessas reclamações não tinha como objetivo mudar a situação reclamada. Elas existiam por outro motivo completamente diferente.

---

### 📚 O Conceito: Reclamação Instrumental vs. Não Instrumental

A pesquisa de Alicke e colegas (1992), publicada na *Personality and Social Psychology Bulletin*, foi a primeira a examinar empiricamente o papel da reclamação na interação social cotidiana. Os pesquisadores distinguem dois tipos de reclamação:

- **Reclamação instrumental**: expressa com o objetivo real de mudar a situação. A pessoa quer uma solução, um conselho, uma ação.
- **Reclamação não instrumental**: expressa por outros motivos — desabafar frustração, buscar simpatia, validar a própria experiência, ou simplesmente se conectar socialmente através de uma queixa compartilhada.

Esse segundo tipo se conecta com um conceito clínico mais antigo, descrito por Berger e Rosenbaum já em 1967: o **"help-rejecting complainer"** (reclamante que rejeita ajuda) — alguém que pede ajuda repetidamente, mas recusa toda sugestão concreta oferecida. Eric Berne, criador da Análise Transacional, descreveu um padrão parecido: o jogo psicológico do "Sim, mas...", em que cada solução oferecida é rejeitada, mantendo a pessoa na posição de quem sofre sem alívio — e sem precisar agir.

### 🧮 O Modelo
**Fontes:**
- Alicke, M.D., Braun, J.C., Glor, J.E., Klotz, M.L., Magee, J., Sederholm, H. & Siegel, R. (1992) — *Complaining Behavior in Social Interaction* — Personality and Social Psychology Bulletin, 18(3), 286-295


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print('✅ Bibliotecas carregadas')
```

```python
# --- DADOS DA LITERATURA ---
# Alicke et al. (1992) — Personality and Social Psychology Bulletin

# Metodologia: estudantes universitarios registraram em diario as reclamacoes
# feitas a outras pessoas, por 3 dias consecutivos, duas vezes no semestre

p_reclamacao_nao_instrumental = 0.75  # mais de 75% das reclamacoes
p_reclamacao_instrumental      = 1 - p_reclamacao_nao_instrumental  # restante (estimativa por complemento)

# IMPORTANTE: o paper original (resumo disponivel) reporta a proporcao mas
# nao especifica o N exato de participantes nem o N total de reclamacoes
# registradas. Por isso este modelo NAO calcula IC 95% sobre um N inventado —
# trabalha apenas com a proporcao central reportada, sendo transparente sobre essa limitacao.

fator_correcao = 0.80

print('=' * 65)
print('  DADOS DA LITERATURA — RECLAMAÇÃO SEM BUSCA DE SOLUÇÃO')
print('=' * 65)
print(f'\n  Alicke et al. 1992 (diário, 3 dias x 2 no semestre):')
print(f'  → Reclamações NÃO instrumentais (sem buscar solução): {p_reclamacao_nao_instrumental*100:.0f}%+')
print(f'  → Reclamações instrumentais (buscam mudar a situação): {p_reclamacao_instrumental*100:.0f}% (estimativa por complemento)')
print(f'\n  Fator de correção aplicado: x{fator_correcao}')
print('=' * 65)
print()
print('  ⚠️  NOTA METODOLÓGICA:')
print('  O resumo disponível do paper original reporta a proporção (>75%)')
print('  mas não especifica o N exato de participantes nem de reclamações')
print('  registradas. Por isso, este modelo não calcula IC 95% artificial —')
print('  trabalha com a proporção central reportada, sendo transparente')
print('  sobre essa limitação de dados.')
```

```python
# --- O MODELO ---
# Como nao ha N confiavel para Beta+IC, o modelo foca em razao e proporcao direta

p_corrigido = p_reclamacao_nao_instrumental * fator_correcao
razao_nao_instrumental_vs_instrumental = p_reclamacao_nao_instrumental / p_reclamacao_instrumental

print('=' * 65)
print('  MODELO — RESULTADO')
print('=' * 65)
print(f'  Proporção de reclamações sem busca de solução:')
print(f'  → Estimativa central : {p_reclamacao_nao_instrumental*100:.0f}%+')
print(f'  → Corrigido (×0.80)  : {p_corrigido*100:.1f}%')
print()
print(f'  Razão reclamação não instrumental / instrumental: {razao_nao_instrumental_vs_instrumental:.1f}x')
print(f'  → Para cada reclamação que busca solução real,')
print(f'    existem aproximadamente {razao_nao_instrumental_vs_instrumental:.0f} que não buscam')
print('=' * 65)
```

```python
# --- VISUALIZACAO ---

# GRAFICO 1 — Reclamacao instrumental vs nao instrumental
fig1, ax1 = plt.subplots(figsize=(12, 8))

categorias = ['Não buscam\nmudar a situação\n(desabafo, simpatia)', 'Buscam mudar\na situação\n(solução real)']
valores = [p_reclamacao_nao_instrumental * 100, p_reclamacao_instrumental * 100]
cores1 = ['#c0392b', '#2a8a82']

bars = ax1.bar(categorias, valores, color=cores1, alpha=0.85, width=0.45)
ax1.set_ylim(0, 90)
ax1.set_ylabel('Proporção de reclamações registradas (%)', fontsize=12)
ax1.set_title('A maioria das reclamações não busca solução\nAlicke et al. (1992) — método de diário, estudantes universitários',
              fontsize=13, pad=15)

for bar, v in zip(bars, valores):
    label = f'{v:.0f}%+' if v > 70 else f'<{v:.0f}%'
    ax1.text(bar.get_x() + bar.get_width() / 2, v + 2,
             label, ha='center', fontweight='bold', fontsize=16)

plt.tight_layout()
plt.savefig('dia-019-grafico-01-instrumental.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Grafico 1 salvo!')


# GRAFICO 2 — Os motivos da reclamacao nao instrumental (ilustrativo, baseado na literatura qualitativa)
fig2, ax2 = plt.subplots(figsize=(12, 8))

motivos = ['Desabafar\nfrustração', 'Buscar\nsimpatia', 'Validar a\nprópria experiência', 'Conectar-se\nsocialmente']
relevancia = [0.85, 0.70, 0.55, 0.45]  # ilustrativo, ordenado por frequencia tipica na literatura qualitativa
cores2 = ['#c0392b', '#c8a84b', '#c8a84b', '#888888']

bars2 = ax2.barh(motivos, [r * 100 for r in relevancia], color=cores2, alpha=0.85, height=0.55)
ax2.set_xlabel('Relevância relativa como motivo (ilustrativo)', fontsize=12)
ax2.set_xlim(0, 100)
ax2.set_title('Por que as pessoas reclamam quando não querem solução\nBaseado na literatura de "help-rejecting complainers"',
              fontsize=13, pad=15)

plt.tight_layout()
plt.savefig('dia-019-grafico-02-motivos.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Grafico 2 salvo!')


# GRAFICO 3 — Razao visual nao-instrumental vs instrumental
fig3, ax3 = plt.subplots(figsize=(10, 10))

labels3 = ['Sem busca\nde solução', 'Com busca\nde solução']
sizes3 = [p_reclamacao_nao_instrumental * 100, p_reclamacao_instrumental * 100]
cores3 = ['#c0392b', '#2a8a82']

wedges, texts, autotexts = ax3.pie(sizes3, labels=labels3, colors=cores3,
                                     autopct='%1.0f%%', startangle=90,
                                     textprops={'fontsize': 15, 'fontweight': 'bold'},
                                     wedgeprops={'alpha': 0.85})

ax3.set_title('De cada 4 reclamações que você escuta,\n3 não pedem solução — pedem testemunha',
              fontsize=14, pad=20)

plt.tight_layout()
plt.savefig('dia-019-grafico-03-proporcao.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Grafico 3 salvo!')
```

### 💡 O Insight

Em estudo no qual estudantes universitários registraram em diário cada reclamação que faziam a outras pessoas, durante dias consecutivos, **mais de 75% dessas reclamações não tinham como objetivo mudar a situação reclamada**. Elas serviam para desabafar frustração, buscar simpatia, ou simplesmente se conectar com outra pessoa através de uma queixa compartilhada.

Isso ajuda a explicar uma frustração comum: quando alguém reclama repetidamente do mesmo problema e você oferece uma solução concreta — um contato, um caminho prático — e a pessoa não segue adiante. Não é, necessariamente, que sua ideia era ruim. É que, estatisticamente, é mais provável que aquela reclamação nunca tenha sido um pedido de solução. Era um pedido de testemunha.

Saber disso não significa parar de se importar. Significa ajustar a expectativa: oferecer presença e escuta tem valor próprio, separado de "resolver o problema da pessoa". E, quando alguém genuinamente quer ajuda prática, ela costuma pedir de um jeito diferente — mais específico, mais aberto a perguntas, menos repetitivo.

*Da próxima vez que alguém vier reclamar com você, vale perguntar antes de sugerir: "você quer que eu te ajude a resolver isso, ou você só precisa desabafar agora?"*

### ⚠️ Limitações do Modelo
- O resumo disponível do estudo de Alicke et al. (1992) reporta a proporção central (>75%), mas não especifica o N exato de participantes nem o número total de reclamações registradas no diário — por isso, este modelo não calcula um intervalo de confiança, para não simular precisão estatística que a fonte não oferece
- A amostra original era de estudantes universitários americanos dos anos 1990 — o padrão pode variar em outros contextos culturais, etários e na era da comunicação digital
- "Buscar solução" e "desabafar" não são categorias mutuamente excludentes na prática — uma mesma reclamação pode conter elementos de ambas
- O modelo não captura a possibilidade de que a mesma pessoa reclame de formas diferentes em momentos diferentes — não é necessariamente um traço fixo de personalidade
- O fator de correção (×0,80) é uma aproximação padrão do projeto

*A ciência é honesta sobre o que não sabe. O modelo também.*
