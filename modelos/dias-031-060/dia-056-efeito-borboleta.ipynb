### 📚 O Conceito: O Efeito Borboleta

Em 1963, o meteorologista Edward Lorenz descobriu, quase por acidente, que certos
sistemas são extremamente sensíveis às condições iniciais. Ele arredondou um
número numa simulação de clima, de 0,506127 para 0,506, e a previsão inteira
mudou. Uma diferença invisível na largada virou um mundo diferente na chegada.

Anos depois, ele deu a isso a imagem que ficou famosa: o bater de asas de uma
borboleta no Brasil pode desencadear um tornado no Texas. Não porque a borboleta
seja poderosa, mas porque, num sistema sensível, o pequeno se amplifica.

A sua vida não é o clima. Mas a lógica se repete: às vezes uma decisão do
tamanho de um passo reorganiza tudo o que vem depois.

---

### 🧮 O Modelo

Simulo o próprio sistema de Lorenz, o berço do efeito borboleta. Duas trajetórias
começam exatamente iguais. Num certo instante, o "salto", uma delas recebe um
empurrãozinho minúsculo, do tamanho de 0,0001. Depois é só deixar o tempo correr
e medir a distância entre as duas vidas.

**Fontes:**
- Lorenz, E. N. (1963). *Deterministic Nonperiodic Flow* · Journal of the
  Atmospheric Sciences, 20, 130-141 · a origem matemática do caos determinístico
- Lorenz, E. N. (1972). *Predictability: Does the Flap of a Butterfly's Wings in
  Brazil Set Off a Tornado in Texas?* · palestra na AAAS · a imagem da borboleta

**Nota metodológica:** o efeito borboleta é um fenômeno matemático de sistemas
determinísticos, não um estudo empírico com N. Aplicá-lo a uma vida é metáfora
mais simulação ilustrativa do mecanismo, não uma medição. O ×0.80 não se aplica.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

DOURADO='#c8a84b'; VERMELHO='#c0392b'; VERDE='#1a5f5a'; CINZA='#6b6a64'
print("✅ Bibliotecas carregadas")
```

```python
# --- O SISTEMA DE LORENZ (parâmetros clássicos) ---
sigma, rho, beta = 10.0, 28.0, 8.0/3.0

def lorenz(t, s):
    x, y, z = s
    return [sigma*(y - x), x*(rho - z) - y, x*y - beta*z]

s0      = [1.0, 1.0, 1.0]   # as duas vidas começam idênticas
t_leap  = 10.0              # o instante do salto
t_end   = 45.0             # até onde acompanhamos
EPS     = 1e-4             # o tamanho do "empurrão" no salto (0,0001)

print("=" * 66)
print("  O SISTEMA DE LORENZ — O BERÇO DO EFEITO BORBOLETA")
print("=" * 66)
print(f"\n  Duas vidas partem do MESMO ponto: {s0}")
print(f"  No instante do salto (t={t_leap:.0f}), uma delas muda por apenas {EPS}")
print(f"  Depois, o tempo corre e a gente mede a distância entre elas.")
print("=" * 66)
```

```python
# --- O MODELO — a divergência exponencial ---
opt = dict(rtol=1e-9, atol=1e-9)

# fase comum (as duas vidas são idênticas até o salto)
t_pre = np.linspace(0, t_leap, 1500)
pre = solve_ivp(lorenz, [0, t_leap], s0, t_eval=t_pre, **opt)
s_leap = pre.y[:, -1]

# depois do salto: A não pulou; B pulou (empurrãozinho EPS em x)
t_pos = np.linspace(t_leap, t_end, 6000)
A = solve_ivp(lorenz, [t_leap, t_end], s_leap, t_eval=t_pos, **opt)                     # não pulou
B = solve_ivp(lorenz, [t_leap, t_end], [s_leap[0]+EPS, s_leap[1], s_leap[2]], t_eval=t_pos, **opt)  # pulou

# distância entre as duas trajetórias ao longo do tempo
dist = np.sqrt((A.y[0]-B.y[0])**2 + (A.y[1]-B.y[1])**2 + (A.y[2]-B.y[2])**2)

print("=" * 66)
print("  MODELO — O SALTO PEQUENO, O DESTINO ENORME")
print("=" * 66)
print(f"\n  Diferença no instante do salto:   {EPS}")
print(f"  Distância entre as vidas no fim:  {dist[-1]:.1f}")
print(f"  → um empurrão de {EPS} virou uma distância de {dist[-1]:.0f}")
print(f"\n  E o crescimento é EXPONENCIAL: na escala log, a distância sobe")
print(f"  praticamente em linha reta, dobrando e dobrando de novo, até")
print(f"  saturar no tamanho do próprio atrator.")
print(f"\n  Tradução: o tamanho do salto não estava na origem (um passo).")
print(f"  Estava no destino, que o tempo foi abrindo em leque.")
print("=" * 66)
```

```python
# --- VISUALIZAÇÃO — 3 GRÁFICOS SEPARADOS (fundo branco, dpi=150) ---

# ── GRÁFICO 1 — Duas vidas que começam iguais ──
fig1, ax1 = plt.subplots(figsize=(10, 6))
ax1.plot(t_pre, pre.y[0], color=CINZA, linewidth=1.8)                 # fase comum
ax1.plot(t_pos, A.y[0], color=CINZA, linewidth=1.6, alpha=0.9, label='A vida que não pulou')
ax1.plot(t_pos, B.y[0], color=DOURADO, linewidth=1.8, label='A vida que pulou')
ax1.axvline(t_leap, color=VERMELHO, linestyle='--', linewidth=1.8)
ax1.text(t_leap+0.3, ax1.get_ylim()[1]*0.86, 'o salto\n(diferença de 0,0001)', color=VERMELHO, fontsize=10)
ax1.set_xlabel('Tempo'); ax1.set_ylabel('Trajetória (coordenada x)')
ax1.set_title('Duas Vidas Que Começam Exatamente Iguais\nAté o salto, idênticas. Depois, dois mundos diferentes.',
              fontsize=13, pad=15)
ax1.legend(loc='lower left', frameon=False, fontsize=10)
plt.figtext(0.5, 0.005, 'Simulação do sistema de Lorenz (1963) | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout(); plt.savefig('dia-056-grafico-01-duas-vidas.png', dpi=150, bbox_inches='tight'); plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 — A ASSINATURA: a divergência exponencial ──
fig2, ax2 = plt.subplots(figsize=(10, 6))
ax2.semilogy(t_pos - t_leap, dist, color=VERMELHO, linewidth=2.5)
ax2.axhline(EPS, color=CINZA, linestyle=':', linewidth=1.4)
ax2.text(0.4, EPS*1.4, 'diferença inicial: 0,0001', color=CINZA, fontsize=10)
ax2.set_xlabel('Tempo desde o salto'); ax2.set_ylabel('Distância entre as duas vidas (log)')
ax2.set_title('A Assinatura — O Pequeno Vira Enorme\n'
              'na escala log, a divergência é quase uma reta: crescimento exponencial',
              fontsize=13, pad=15)
ax2.annotate('reta em escala log =\ncrescimento exponencial', xy=(6, dist[np.argmin(np.abs((t_pos-t_leap)-6))]),
             xytext=(9, 1e-2), fontsize=10, color=VERDE,
             arrowprops=dict(arrowstyle='->', color=VERDE, lw=1.4))
plt.figtext(0.5, 0.005, 'Sensibilidade às condições iniciais · Lorenz (1963) | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout(); plt.savefig('dia-056-grafico-02-divergencia.png', dpi=150, bbox_inches='tight'); plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 — A borboleta (o atrator de Lorenz) ──
fig3, ax3 = plt.subplots(figsize=(9, 7))
xb = np.concatenate([pre.y[0], B.y[0]]); zb = np.concatenate([pre.y[2], B.y[2]])
ax3.plot(xb, zb, color=DOURADO, linewidth=0.5, alpha=0.8)
ax3.set_xlabel('x'); ax3.set_ylabel('z'); ax3.grid(alpha=0.15)
ax3.set_title('E o Mesmo Sistema Desenha uma Borboleta\nO atrator de Lorenz: caos com uma forma linda',
              fontsize=13, pad=15)
plt.figtext(0.5, 0.02, 'Atrator de Lorenz | #365Probabilidades', ha='center', fontsize=9, color='gray')
plt.tight_layout(); plt.savefig('dia-056-grafico-03-borboleta.png', dpi=150, bbox_inches='tight'); plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**A decisão foi um só gesto: um sim, uma porta, um único ponto. E, num sistema
sensível, um ponto minúsculo já basta: aqui, um empurrão de 0,0001 virou uma
distância inteira.**

É por isso que, na beira, ninguém consegue medir o próprio salto. O tamanho dele
não está no gesto de decidir, que é um instante. Está no futuro que esse gesto
destrava, e esse futuro só se revela com o tempo, abrindo em leque.

Tem um lado que assusta nisso: se o pequeno decide tanto, a gente controla menos
do que gosta de pensar. Mas tem um lado que liberta, e é o que fica: você não
precisa de um gesto gigante para mudar tudo. Uma decisão do tamanho certo, na
hora certa, já é uma borboleta.

Um ano depois, é isso que os números me dizem sobre o salto que eu dei: na hora
ele pareceu grande, e hoje eu entendo que foi imenso, porque um sistema sensível
transforma um único gesto em destino.

A pergunta que fica:

*Qual foi a decisão pequena que, olhando para trás, reescreveu o resto?*

---

### ⚠️ Limitações do Modelo

- O efeito borboleta é um fenômeno matemático de sistemas determinísticos (Lorenz). Uma vida não é o sistema de Lorenz; usar a simulação aqui é ilustrar o mecanismo (pequeno vira grande), não medir a sua biografia.
- Sensibilidade às condições iniciais não quer dizer que "tudo é aleatório". O sistema é determinístico; ele é imprevisível na prática, não sem regra.
- A divergência é aproximadamente exponencial no início e depois satura no tamanho do atrator. A taxa exata (expoente de Lyapunov, cerca de 0,9 por unidade de tempo no sistema de Lorenz) varia ao longo da trajetória, então preferi mostrar o crescimento pela reta na escala log a cravar um número.
- A metáfora tem um limite honesto: na vida, nem todo pequeno vira grande. O caos amplifica algumas diferenças, não todas.
- ×0.80 não se aplica: não há proporção de survey, é simulação de um sistema dinâmico.

*A ciência é honesta sobre o que não sabe. O modelo também.*
