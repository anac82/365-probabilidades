# 365 Probabilidades — Dia #048
## Qual a probabilidade de mostrar um defeito te aproximar das pessoas, em vez de afastar?

**Tipo:** Comportamental
**Data de publicação:** 2026-07-31
**Ferramenta:** Python
**Decisão analisada:** Escondo minhas imperfeições para ser levada a sério, ou mostro?
**Hashtag:** #365Probabilidades #Dia048

---

### 📖 A História

Eu sempre fui bem transparente com as pessoas próximas.

No dia a dia, isso quase sempre significa deixar minha vulnerabilidade à mostra, e às vezes situações francamente cômicas. Nunca tive muito talento pra disfarçar quando algo dá errado, e faz tempo que parei de tentar.

Essa semana eu postei um vídeo bem desajeitado, uma tentativa de rebater a bolinha num jogo de tênis que terminou num vacilo digno de bloopers. Nada elegante, nada montado. Só eu, atrapalhada, rindo do próprio erro.

E me surpreendi com o tanto de reação e visualização que aquilo teve. Bem mais do que quando eu apareço impecável.

O que eu não sabia é que isso tem nome, e um experimento de 1966 pra sustentar. Com uma condição curiosa no meio, que muda tudo.

---

### 📚 O Conceito: O Efeito Pratfall

Elliot Aronson gravou entrevistas com candidatos a um torneio de perguntas e respostas. Uns pareciam brilhantes, acertando quase tudo. Outros, medianos. E, em metade das gravações, no fim, a pessoa derrubava uma xícara de café em si mesma e se atrapalhava, um vacilo bobo, humano.

Depois, mediu quão simpática cada uma parecia. O resultado ficou conhecido como **efeito pratfall**. O candidato brilhante que derrubou o café foi o mais querido de todos, mais até do que o brilhante impecável. O vacilo, nele, não tirou pontos. Adicionou.

Mas tem o outro lado. No candidato mediano, o mesmo vacilo fez o oposto, derrubou a simpatia de vez. A imperfeição aproxima, sim, mas com uma condição: precisa haver competência embaixo. A falha humaniza quem já se admira. No resto, só confirma a dúvida.

---

### 🧮 O Modelo

Uso as médias de simpatia das quatro condições do experimento (competência alta ou média, com ou sem o vacilo) para medir o tamanho e a direção do efeito, que é uma interação clássica.

**Fonte:**
- Aronson, E., Willerman, B. & Floyd, J. (1966). *The Effect of a Pratfall on Increasing Interpersonal Attractiveness*. **Psychonomic Science**, 4(6), 227-228. N=48. Escore de simpatia somado (itens de −7 a +7).

**Nota metodológica:** são escores de simpatia de um experimento com medida direta, não proporções de survey autorrelatado. Por isso o fator ×0.80 não se aplica.


````python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

DOURADO = '#c8a84b'; VERMELHO = '#c0392b'; VERDE = '#1a5f5a'; CINZA = '#6b6a64'
np.random.seed(48)  # Dia #048
print("✅ Bibliotecas carregadas")
````

````python
# --- DADOS DA LITERATURA ---
# Aronson, Willerman & Floyd (1966) — N=48 — média de simpatia por condição
sup_sem   = 20.8   # competente, sem vacilo
sup_vacilo = 30.2  # competente, com vacilo  -> o mais querido
med_sem   = 17.8   # mediano, sem vacilo
med_vacilo = -2.5  # mediano, com vacilo     -> o menos querido
N = 48

print("=" * 64)
print("  DADOS DA LITERATURA — EFEITO PRATFALL (Aronson 1966, N=48)")
print("=" * 64)
print(f"  Média de simpatia por condição:")
print(f"  → Competente + vacilo:   {sup_vacilo:+.1f}   (o mais querido)")
print(f"  → Competente sem vacilo: {sup_sem:+.1f}")
print(f"  → Mediano sem vacilo:    {med_sem:+.1f}")
print(f"  → Mediano + vacilo:      {med_vacilo:+.1f}   (o menos querido)")
print("=" * 64)
````

````python
# --- O MODELO: quanto o vacilo muda a simpatia, e pra quem ---
ganho_competente = sup_vacilo - sup_sem
pct_competente = ganho_competente / sup_sem * 100
queda_mediano = med_vacilo - med_sem

print("=" * 64)
print("  MODELO — A IMPERFEIÇÃO APROXIMA (COM UMA CONDIÇÃO)")
print("=" * 64)
print(f"  Se você é vista como COMPETENTE:")
print(f"  → o vacilo AUMENTOU a simpatia em {ganho_competente:+.1f} pontos")
print(f"    ({pct_competente:+.0f}%). O mais querido de todos foi quem")
print(f"    era brilhante E derramou o café.")
print()
print(f"  Se você é vista como MEDIANA:")
print(f"  → o mesmo vacilo DERRUBOU a simpatia em {queda_mediano:+.1f} pontos.")
print(f"    Nele, a falha não humanizou, confirmou a dúvida.")
print()
print(f"  RESPOSTA:")
print(f"  → Mostrar um defeito aproxima, sim, quando existe competência")
print(f"    embaixo. A perfeição cria distância; a fissura é a ponte.")
print("=" * 64)
````

````python
# --- VISUALIZAÇÃO — 3 GRÁFICOS SEPARADOS ---

# ── GRÁFICO 1 — As quatro condições ──
fig1, ax1 = plt.subplots(figsize=(10, 6))
cats = ['Competente\nsem vacilo', 'Competente\ncom vacilo', 'Mediano\nsem vacilo', 'Mediano\ncom vacilo']
vals = [sup_sem, sup_vacilo, med_sem, med_vacilo]
cores = [CINZA, DOURADO, CINZA, VERMELHO]
bars = ax1.bar(cats, vals, color=cores, alpha=0.9, width=0.6)
ax1.axhline(0, color=CINZA, lw=1)
ax1.set_ylabel('Simpatia (escore do estudo)')
ax1.set_ylim(-8, 35)
ax1.set_title('O mais querido foi o brilhante que derramou o café\nEfeito Pratfall — Aronson, Willerman & Floyd 1966 (N=48)', fontsize=13, pad=15)
for bar, v in zip(bars, vals):
    off = 0.8 if v >= 0 else -2.2
    ax1.text(bar.get_x()+bar.get_width()/2, v+off, f'{v:+.1f}', ha='center', fontweight='bold', fontsize=15)
plt.figtext(0.5, 0.005, 'O vacilo eleva quem é competente e afunda quem é mediano | Aronson 1966 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout(); plt.savefig('dia-048-grafico-01-quatro.png', dpi=150, bbox_inches='tight'); plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 — ASSINATURA: a interação ──
fig2, ax2 = plt.subplots(figsize=(10, 6))
x = [0, 1]
ax2.plot(x, [sup_sem, sup_vacilo], color=VERDE, lw=3, marker='o', markersize=13, label='Vista como COMPETENTE')
ax2.plot(x, [med_sem, med_vacilo], color=VERMELHO, lw=3, marker='o', markersize=13, label='Vista como MEDIANA')
ax2.axhline(0, color=CINZA, lw=1, alpha=0.5)
ax2.set_xticks([0, 1]); ax2.set_xticklabels(['Sem vacilo', 'Com vacilo (derrubou o café)'])
ax2.set_ylabel('Simpatia (escore do estudo)')
ax2.set_title('A ASSINATURA ESTATÍSTICA\nO mesmo vacilo sobe ou desce, dependendo de quem erra', fontsize=13, pad=15)
ax2.annotate(f'{sup_vacilo-sup_sem:+.1f}', xy=(1, sup_vacilo), xytext=(1.02, sup_vacilo-1),
             fontsize=13, color=VERDE, fontweight='bold')
ax2.annotate(f'{med_vacilo-med_sem:+.1f}', xy=(1, med_vacilo), xytext=(1.02, med_vacilo-1),
             fontsize=13, color=VERMELHO, fontweight='bold')
ax2.legend(loc='center left', fontsize=11); ax2.set_xlim(-0.15, 1.35)
plt.figtext(0.5, 0.005, 'Interação competência × vacilo · o cruzamento é o efeito | Aronson 1966 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout(); plt.savefig('dia-048-grafico-02-interacao.png', dpi=150, bbox_inches='tight'); plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 — Pra quem é competente, o vacilo é ponte ──
fig3, ax3 = plt.subplots(figsize=(10, 6))
cats3 = ['Competente\nimpecável', 'Competente\nque vacilou']
vals3 = [sup_sem, sup_vacilo]
bars3 = ax3.bar(cats3, vals3, color=[CINZA, DOURADO], alpha=0.9, width=0.45)
ax3.set_ylabel('Simpatia (escore do estudo)'); ax3.set_ylim(0, 35)
ax3.set_title('A fissura é a ponte\nPara quem já é vista como competente, o defeito à mostra aproxima', fontsize=13, pad=15)
for bar, v in zip(bars3, vals3):
    ax3.text(bar.get_x()+bar.get_width()/2, v+0.7, f'{v:+.1f}', ha='center', fontweight='bold', fontsize=16)
ax3.annotate('', xy=(1, sup_vacilo), xytext=(0, sup_sem), arrowprops=dict(arrowstyle='->', color=VERMELHO, lw=2))
ax3.text(0.5, 0.78, f'+{pct_competente:.0f}% de simpatia', ha='center', fontsize=13, color=VERMELHO,
         fontweight='bold', transform=ax3.transAxes)
plt.figtext(0.5, 0.005, 'A perfeição cria distância; o defeito humaniza | Aronson 1966 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout(); plt.savefig('dia-048-grafico-03-ponte.png', dpi=150, bbox_inches='tight'); plt.close()
print("✅ Gráfico 3 salvo!")
````

### 💡 O Insight

**O mais querido de todos foi o brilhante que derrubou o café.**

Para quem já é visto como competente, mostrar um defeito não custou simpatia. Rendeu, cerca de 45% a mais. A perfeição impecável cria uma distância educada, admiração de longe. A fissura é por onde as pessoas entram.

Mas o experimento é honesto sobre a condição. No candidato mediano, o mesmo vacilo derrubou tudo. A imperfeição só aproxima quando existe competência embaixo dela, quando a falha é a exceção, e não a regra. Não é sobre errar de propósito. É sobre parar de esconder a costura quando você já entregou o suficiente pra ser levada a sério.

Foi mais ou menos isso que aconteceu com o meu vídeo desajeitado de tênis essa semana. A versão impecável de você é a menos convidativa. É o vacilo, mostrado sem vergonha, que faz alguém pensar "ah, ela é gente como a gente". A ciência só me deu o nome pra uma coisa que eu já fazia sem saber.

*Que imperfeição sua você esconde achando que afasta, quando talvez seja exatamente ela que aproxima?*

---

### ⚠️ Limitações do Modelo

- Aronson 1966 tem N=48, estudantes homens, e usou gravações, não interação real.
- O efeito é condicional à competência percebida e pode variar com gênero, cultura e contexto; replicações posteriores são mistas e apontam moderadores.
- Os escores são de simpatia percebida numa escala do estudo, não uma medida absoluta.
- "Competência" aqui é a impressão que os outros têm, não um valor objetivo seu.
- O fator ×0.80 não foi aplicado: escores de experimento, não survey autorrelatado.

*A ciência é honesta sobre o que não sabe. O modelo também.*
