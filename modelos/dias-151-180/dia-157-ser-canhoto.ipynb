### 📖 A História

Passei um tempo com meus sobrinhos recentemente. O mais velho quebrou o punho ao jogar bola e ficou com o braço engessado…

Durante uma brincadeira em que ele precisava escrever, percebi que ele é canhoto. E percebi outra coisa. Depois de umas duas semanas com o braço esquerdo engessado, o direito já dava sinais de que poderia quebrar um galho, pelo menos para escrever! Fiquei pensando muito nisso, porque escuto histórias de canhotos que aprenderam a escrever com a direita. Principalmente antigamente, quando o desconhecido poderia causar mais estranheza.

Todo mundo já reparou num canhoto. Em qualquer sala, quando alguém pega a caneta com a esquerda, a gente nota. Ou quando vai abrir uma lata, ou usar uma tesoura… E quase todo mundo tem a mesma impressão, a de que são poucos.

Fui ver quantos são, e quanto disso vem de família.

---

### 📚 O Conceito

Antes de contar, é preciso decidir quem conta como canhoto.

Há pesquisas que perguntam só com que mão a pessoa escreve. Outras aplicam um questionário com várias tarefas (escrever, arremessar, usar a tesoura, escovar os dentes) e separam três grupos, os destros, os canhotos e os que usam uma mão para cada coisa. E há as que dividem o mundo em destros e não destros, e aí qualquer pessoa que não faça tudo com a direita entra na conta.

O número muda conforme o critério. Por isso este dia trabalha com uma faixa e com uma melhor estimativa, e não com um número só.

Existe ainda uma segunda dificuldade. A mão que a pessoa usa hoje nem sempre é a mão com que ela começou. Durante boa parte do século XX, em muitos países, crianças canhotas foram ensinadas a escrever com a direita. Os estudos medem a mão que ficou.

E há a pergunta da família. Ser canhoto tem um componente genético, e ele é menor do que se imagina. Não existe um gene do canhoto. Estudos com gêmeos atribuem aos genes cerca de um quarto da variação entre as pessoas, espalhado por dezenas de pontos do genoma, cada um com efeito muito pequeno. O resto vem de fatores que não são herdados.

---

### 🧮 O Modelo

A fonte central é a maior meta-análise já feita sobre o assunto. Em cima dela entram cinco contas próprias.

1. **Os três critérios.** A proporção de canhotos pelo critério rígido, pela melhor estimativa e pelo critério mais frouxo, cada uma com o seu intervalo.
2. **A sala de 30 pessoas.** Quantos canhotos esperar numa sala, e a chance de não haver nenhum. A conta exata usa a distribuição binomial, P(nenhum) = (1 − p)^n, e a simulação (200 mil salas, semente 42) carrega a incerteza do próprio p.
3. **Homens e mulheres.** A diferença publicada, traduzida em pessoas a cada mil e na chance de um canhoto sorteado ser homem.
4. **As gerações.** A proporção de canhotos nos estudos publicados antes de 1976 contra os estudos posteriores, e quanto dessa diferença caberia em canhotos que não foram contados.
5. **A família.** A chance de um filho canhoto conforme a mão dos pais, e a conta inversa, que pergunta de que tipo de casal vêm os canhotos. Essa conta supõe casais formados sem relação com a mão de cada um.

**Fontes:**

Lido no manuscrito aceito, depositado no repositório da Universidade de St Andrews.

- Medland, S. E., Duffy, D. L., Wright, M. J., Geffen, G. M., Hay, D. A., Levy, F., ... Boomsma, D. I. (2009). Efeitos genéticos aditivos explicam 23,64% da variação (IC 95% de 20,17 a 27,09); o restante é ambiente não compartilhado. Lido no resumo.

- Annett, M. (2002). *Handedness and Brain Asymmetry: The Right Shift Theory*. Psychology Press. Proporção de filhos canhotos conforme a mão dos pais (9,5%, 19,5% e pouco mais de 26%), como citada por Sebastian Ocklenburg, coautor da meta-análise de 2020. O tamanho da amostra por trás desses três números não foi localizado.
- Gilbert, A. N. & Wysocki, C. J. (1992). Hand preference and age in the United States. *Neuropsychologia*, 30(7), 601-608. N = 1.177.507, de 10 a 86 anos. Os percentuais por geração entram como citados por Papadatou-Pastou et al. (2020); o artigo original foi lido só no resumo.

**Nota metodológica sobre o fator ×0,80.** Não se aplica. A fonte é uma meta-análise, e a mão com que a pessoa escreve é um fato observável, não uma opinião que o autorrelato costuma inflar. Na própria meta-análise, os estudos com autorrelato (10,50%) e os estudos sem autorrelato (11,90%) não diferem de forma estatisticamente detectável (p = 0,458). A incerteza real está na definição de canhoto, e é ela que o modelo mostra.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
from math import log, ceil

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

SEMENTE = 42
N_SIM = 200_000
rng = np.random.default_rng(SEMENTE)

VERMELHO = '#c0392b'
DOURADO = '#c8a84b'
VERDE = '#1a5f5a'
CINZA = '#7f8c8d'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Papadatou-Pastou et al. (2020), Psychological Bulletin 146(6), 481-524
# Todos os valores em %, com IC 95% publicado: (estimativa, limite inferior, limite superior)

n_total = 2_396_170
n_homens = 1_112_365
n_mulheres = 1_283_805
k_estudos = 200
k_conjuntos = 262

# As cinco meta-análises
canhoto_total = (10.60, 9.71, 11.50)      # todos os conjuntos (k=262)
canhoto_rigido = (9.34, 7.92, 10.80)      # classificação direita/mista/esquerda (k=72, n=789.090)
canhoto_escolha = (10.20, 9.14, 11.20)    # escolha forçada direita/esquerda (k=173, n=413.560)
nao_destro = (18.10, 13.90, 22.30)        # destro contra não destro (k=26, n=1.203.403)
mao_mista = (9.33, 6.67, 12.00)           # categoria do meio (k=72, n=789.090)
sem_atletas = 10.40                       # melhor estimativa sem atletas de elite
i2_total = 97.42                          # heterogeneidade entre estudos (%)

# Sexo (meta-análise 1, todos os conjuntos)
mulheres = (9.53, 8.75, 10.30)
homens = (11.62, 10.66, 12.60)
dif_sexo = (2.09, 1.61, 2.57)             # homens menos mulheres, em pontos percentuais
ep_dif_sexo = 0.24                        # erro padrão publicado da diferença

# Época de publicação do estudo: (rótulo, estimativa, inferior, superior, conjuntos)
epocas = [
    ('antes de 1976', 7.19, 5.62, 8.76, 18),
    ('1976-1985', 10.73, 8.81, 12.65, 69),
    ('1986-1995', 10.64, 8.90, 12.38, 70),
    ('1996-2007', 11.70, 9.57, 13.82, 46),
    ('2008-2019', 10.77, 8.62, 12.92, 57),
]

# Ancestralidade inferida pelo local do estudo: (rótulo, estimativa, inferior, superior, conjuntos)
ancestralidade = [
    ('Europeia', 11.12, 10.09, 12.15, 184),
    ('África subsaariana', 7.71, 6.25, 9.18, 9),
    ('Leste da Ásia', 5.69, 3.49, 7.88, 23),
]

# Instrumento de medida
mao_de_escrever = (9.29, 8.45, 10.10)
observacao = (15.11, 9.15, 21.10)
autorrelato = 10.50
sem_autorrelato = 11.90
p_autorrelato = 0.458

# Gilbert & Wysocki (1992), como citados por Papadatou-Pastou et al. (2020)
gw_nascidos_1900 = 3.0     # "apenas 3%"
gw_nascidos_1945 = 12.0    # "quase 12%" de 1945 em diante

# Papadatou-Pastou et al. (2008): razão de chances homens/mulheres
or_sexo_2008 = (1.23, 1.19, 1.27)

# Genética
# Medland et al. (2009), Neuropsychologia 47(2): gêmeos australianos e holandeses
n_medland_pessoas = 54_270
n_medland_familias = 25_732
herdabilidade = (23.64, 20.17, 27.09)     # % da variação explicada por efeitos genéticos aditivos
# Cuellar-Partida et al. (2021), Nature Human Behaviour 5(1)
n_gwas = 1_766_671
loci_canhoto = 41
loci_duas_maos = 7
# Annett (2002), como citada por Ocklenburg: % de filhos canhotos conforme a mão dos pais
# (N por trás destes três números não localizado)
filho_canhoto = {
    'Pai e mãe destros': 9.5,
    'Um dos dois canhoto': 19.5,
    'Pai e mãe canhotos': 26.0,
}

aplica_fator_080 = False

print("=" * 68)
print("  DADOS DA LITERATURA - CANHOTOS")
print("=" * 68)
print(f"\n  Papadatou-Pastou et al. (2020), Psychological Bulletin")
print(f"  → {k_estudos} estudos, {k_conjuntos} conjuntos, N = " + f"{n_total:,}".replace(',', '.'))
print(f"\n  Proporção de canhotos, por critério (IC 95%):")
print(f"  → Critério rígido (três grupos):   {canhoto_rigido[0]:.2f}% [{canhoto_rigido[1]:.2f}; {canhoto_rigido[2]:.2f}]")
print(f"  → Escolha direita ou esquerda:     {canhoto_escolha[0]:.2f}% [{canhoto_escolha[1]:.2f}; {canhoto_escolha[2]:.2f}]")
print(f"  → MELHOR ESTIMATIVA (todos):       {canhoto_total[0]:.2f}% [{canhoto_total[1]:.2f}; {canhoto_total[2]:.2f}]")
print(f"  → Qualquer um que não é destro:    {nao_destro[0]:.2f}% [{nao_destro[1]:.2f}; {nao_destro[2]:.2f}]")
print(f"  → Usa as duas mãos (grupo do meio): {mao_mista[0]:.2f}% [{mao_mista[1]:.2f}; {mao_mista[2]:.2f}]")
print(f"  → Heterogeneidade entre estudos:   I² = {i2_total:.1f}%")
print(f"\n  Homens e mulheres:")
print(f"  → Mulheres: {mulheres[0]:.2f}% [{mulheres[1]:.2f}; {mulheres[2]:.2f}]")
print(f"  → Homens:   {homens[0]:.2f}% [{homens[1]:.2f}; {homens[2]:.2f}]")
print(f"  → Diferença: +{dif_sexo[0]:.2f} pp [{dif_sexo[1]:.2f}; {dif_sexo[2]:.2f}]")
print(f"\n  Época de publicação do estudo:")
for rot, est, lo, hi, k in epocas:
    print(f"  → {rot:<14} {est:5.2f}% [{lo:5.2f}; {hi:5.2f}]  ({k} conjuntos)")
print(f"\n  Ancestralidade (pelo local do estudo):")
for rot, est, lo, hi, k in ancestralidade:
    print(f"  → {rot:<20} {est:5.2f}% [{lo:5.2f}; {hi:5.2f}]  ({k} conjuntos)")
print(f"\n  Gilbert & Wysocki (1992), citados pela meta-análise:")
print(f"  → Nascidos em 1900: {gw_nascidos_1900:.0f}%  ·  nascidos de 1945 em diante: quase {gw_nascidos_1945:.0f}%")
print(f"\n  Genética:")
print(f"  → Medland et al. (2009): " + f"{n_medland_pessoas:,}".replace(',', '.') + " pessoas em " + f"{n_medland_familias:,}".replace(',', '.') + " famílias de gêmeos")
print(f"  → Variação explicada pelos genes: {herdabilidade[0]:.1f}% [{herdabilidade[1]:.1f}; {herdabilidade[2]:.1f}]")
print(f"  → Cuellar-Partida et al. (2021): N = " + f"{n_gwas:,}".replace(',', '.') + f", {loci_canhoto} pontos do genoma")
print(f"  Filhos canhotos conforme os pais (Annett 2002, N não localizado):")
for casal, taxa in filho_canhoto.items():
    print(f"  → {casal:<22} {taxa:.1f}%")
print(f"\n  Fator ×0,80 aplicado: {'sim' if aplica_fator_080 else 'não'}")
print("=" * 68)
```

```python
# --- O MODELO ---
# Quatro contas próprias sobre os números publicados. Semente 42, 200 mil sorteios.

def erro_padrao(ic):
    # erro padrão implícito no IC 95% publicado
    return (ic[2] - ic[1]) / (2 * 1.96)

p = canhoto_total[0] / 100
ep_p = erro_padrao(canhoto_total) / 100

# ---------- Camada 1: os três critérios ----------
razao_criterios = nao_destro[0] / canhoto_rigido[0]

# ---------- Camada 2: a sala de 30 pessoas ----------
SALA = 30
esperado = SALA * p
p_nenhum = (1 - p) ** SALA                      # conta exata, binomial
p_ao_menos_um = 1 - p_nenhum
pmf = stats.binom.pmf(np.arange(0, SALA + 1), SALA, p)
mais_provavel = int(np.argmax(pmf))
faixa_95_exata = stats.binom.interval(0.95, SALA, p)

# simulação com a incerteza do próprio p
p_sim = np.clip(rng.normal(p, ep_p, N_SIM), 0.001, 0.999)
canhotos_na_sala = rng.binomial(SALA, p_sim)
p_nenhum_sim = (canhotos_na_sala == 0).mean()
faixa_95_sim = np.percentile(canhotos_na_sala, [2.5, 97.5])
p_dois_a_cinco = ((canhotos_na_sala >= 2) & (canhotos_na_sala <= 5)).mean()

# tamanho do grupo para ter ao menos um canhoto
def grupo_para(prob, taxa):
    return ceil(log(1 - prob) / log(1 - taxa))

grupo_50 = grupo_para(0.50, p)
grupo_95 = grupo_para(0.95, p)

# sensibilidade: a mesma sala com outros critérios e outras regiões
cenarios = [
    ('Critério rígido', canhoto_rigido[0]),
    ('Melhor estimativa', canhoto_total[0]),
    ('Não destros', nao_destro[0]),
    ('Estudos na Europa e descendentes', ancestralidade[0][1]),
    ('Estudos no Leste da Ásia', ancestralidade[2][1]),
]
tabela_sala = [(nome, taxa, SALA * taxa / 100, (1 - taxa / 100) ** SALA) for nome, taxa in cenarios]

# ---------- Camada 3: homens e mulheres ----------
por_mil_h = homens[0] * 10
por_mil_m = mulheres[0] * 10
pm_sim = rng.normal(mulheres[0], erro_padrao(mulheres), N_SIM)
dif_sim = rng.normal(dif_sexo[0], ep_dif_sexo, N_SIM)
ph_sim = pm_sim + dif_sim
# chance de um canhoto sorteado ser homem, supondo metade homens e metade mulheres
homem_dado_canhoto = homens[0] / (homens[0] + mulheres[0])
homem_dado_canhoto_sim = ph_sim / (ph_sim + pm_sim)
ic_homem_dado_canhoto = np.percentile(homem_dado_canhoto_sim, [2.5, 97.5])
p_dif_positiva = (dif_sim > 0).mean()
razao_h_m = homens[0] / mulheres[0]

# ---------- Camada 4: as gerações ----------
antiga = epocas[0]
ep_antiga = erro_padrao((antiga[1], antiga[2], antiga[3]))
antiga_sim = rng.normal(antiga[1], ep_antiga, N_SIM)
total_sim = p_sim * 100
faltam_sim = total_sim - antiga_sim                 # canhotos "que faltam" a cada 100 pessoas
fracao_sim = faltam_sim / total_sim                 # fração dos canhotos que não aparece
faltam = canhoto_total[0] - antiga[1]
fracao = faltam / canhoto_total[0]
ic_faltam = np.percentile(faltam_sim, [2.5, 97.5])
ic_fracao = np.percentile(fracao_sim, [2.5, 97.5])
p_antiga_menor = (faltam_sim > 0).mean()
fracao_gw = 1 - gw_nascidos_1900 / gw_nascidos_1945

# ---------- Camada 5: a família ----------
nao_genetico = 100 - herdabilidade[0]
taxas_casal = np.array(list(filho_canhoto.values())) / 100
# casais formados sem relação com a mão de cada um, com a mesma proporção p nas duas gerações
peso_casal = np.array([(1 - p) ** 2, 2 * p * (1 - p), p ** 2])
canhotos_por_casal = peso_casal * taxas_casal
p_filho_implicita = canhotos_por_casal.sum()          # checagem: deve ficar perto da proporção geral
origem_canhotos = canhotos_por_casal / p_filho_implicita
razao_um_canhoto = taxas_casal[1] / taxas_casal[0]
razao_dois_canhotos = taxas_casal[2] / taxas_casal[0]
destro_com_pais_canhotos = 1 - taxas_casal[2]

print("=" * 68)
print("  MODELO - RESULTADO")
print("=" * 68)
print(f"\n  CAMADA 1 · Os três critérios")
print(f"  → Do critério rígido ao mais frouxo: {canhoto_rigido[0]:.1f}% a {nao_destro[0]:.1f}%")
print(f"  → O critério frouxo conta {razao_criterios:.1f} vezes mais gente")
print(f"  → Melhor estimativa: {canhoto_total[0]:.1f}%, ou 1 em cada {1/p:.1f} pessoas")
print(f"\n  CAMADA 2 · A sala de {SALA} pessoas (p = {p*100:.1f}%)")
print(f"  → Canhotos esperados:                 {esperado:.1f}")
print(f"  → Número mais provável:               {mais_provavel}")
print(f"  → Faixa de 95% (conta exata):         {int(faixa_95_exata[0])} a {int(faixa_95_exata[1])}")
print(f"  → Faixa de 95% (com incerteza de p):  {int(faixa_95_sim[0])} a {int(faixa_95_sim[1])}")
print(f"  → P(entre 2 e 5 canhotos):            {p_dois_a_cinco*100:.1f}%")
print(f"  → P(nenhum canhoto), conta exata:     {p_nenhum*100:.1f}%")
print(f"  → P(nenhum canhoto), simulação:       {p_nenhum_sim*100:.1f}%")
print(f"  → P(ao menos um canhoto):             {p_ao_menos_um*100:.1f}%")
print(f"  → Grupo para 50% de chance de ter um: {grupo_50} pessoas")
print(f"  → Grupo para 95% de chance de ter um: {grupo_95} pessoas")
print(f"\n  A mesma sala com outras réguas:")
for nome, taxa, esp, pn in tabela_sala:
    print(f"  → {nome:<34} {taxa:5.2f}%  esperados {esp:4.1f}  P(nenhum) {pn*100:5.1f}%")
print(f"\n  CAMADA 3 · Homens e mulheres")
print(f"  → A cada 1.000 homens:   {por_mil_h:.0f} canhotos")
print(f"  → A cada 1.000 mulheres: {por_mil_m:.0f} canhotas")
print(f"  → Razão homens/mulheres: {razao_h_m:.2f}")
print(f"  → P(a diferença ser positiva): {p_dif_positiva*100:.1f}%")
print(f"  → P(homem | canhoto), com metade de cada sexo: {homem_dado_canhoto*100:.1f}% "
      f"[{ic_homem_dado_canhoto[0]*100:.1f}; {ic_homem_dado_canhoto[1]*100:.1f}]")
print(f"\n  CAMADA 4 · As gerações")
print(f"  → Estudos antes de 1976: {antiga[1]:.2f}%  ·  todos os estudos: {canhoto_total[0]:.2f}%")
print(f"  → Diferença: {faltam:.1f} canhotos a cada 100 pessoas [{ic_faltam[0]:.1f}; {ic_faltam[1]:.1f}]")
print(f"  → Fração dos canhotos que não aparece nos estudos antigos: {fracao*100:.0f}% "
      f"[{ic_fracao[0]*100:.0f}; {ic_fracao[1]*100:.0f}]")
print(f"  → P(estudos antigos abaixo da estimativa geral): {p_antiga_menor*100:.1f}%")
print(f"  → Mesma conta com Gilbert & Wysocki (3% contra quase 12%): {fracao_gw*100:.0f}%")
print(f"\n  CAMADA 5 · A família")
print(f"  → Variação explicada pelos genes: {herdabilidade[0]:.0f}%  ·  por fatores não herdados: {nao_genetico:.0f}%")
print(f"  → Um dos pais canhoto: chance {razao_um_canhoto:.1f} vezes a de dois pais destros")
print(f"  → Os dois pais canhotos: chance {razao_dois_canhotos:.1f} vezes a de dois pais destros")
print(f"  → Mesmo com pai e mãe canhotos, {destro_com_pais_canhotos*100:.0f}% dos filhos são destros")
print(f"  → Tipos de casal (pais sorteados, p = {p*100:.1f}%): "
      f"dois destros {peso_casal[0]*100:.1f}%, um canhoto {peso_casal[1]*100:.1f}%, dois canhotos {peso_casal[2]*100:.1f}%")
print(f"  → De cada 100 canhotos, {origem_canhotos[0]*100:.0f} têm pai e mãe destros, "
      f"{origem_canhotos[1]*100:.0f} têm um dos pais canhoto e {origem_canhotos[2]*100:.0f} têm os dois")
print(f"  → Checagem: proporção de filhos canhotos implícita na conta = {p_filho_implicita*100:.1f}% "
      f"(estimativa geral {canhoto_total[0]:.1f}%)")
print("=" * 68)
```

### 💡 O Insight

**De cada dez pessoas, uma é canhota.**

A melhor estimativa é de 10,6%, com intervalo de 9,7% a 11,5%, tirada de 200 estudos e quase 2,4 milhões de pessoas. O número fica em 9,3% no critério mais rígido e sobe para 18,1% quando se conta todo mundo que não é destro.

Numa sala de 30 pessoas, o esperado são três canhotos, e a chance de não haver nenhum é de 3,5%. Com sete pessoas reunidas já é mais provável ter um canhoto do que não ter.

Entre os homens há um pouco mais. São 116 canhotos a cada mil homens e 95 canhotas a cada mil mulheres.

E há os que não aparecem na conta. Os estudos publicados antes de 1976 encontravam 7,2% de canhotos, e os de depois encontram perto de 11%. A diferença equivale a um em cada três canhotos. A explicação que os autores consideram mais provável é a pressão para escrever com a direita, que diminuiu ao longo do século XX.

Nos lugares e nas épocas em que essa pressão diminuiu, a proporção se aproxima de um em dez.

Sobre a família, os genes contam e não decidem. Eles explicam cerca de um quarto da variação entre as pessoas. Com pai ou mãe canhotos a chance de um filho canhoto dobra, de 9,5% para 19,5%, e mesmo com os dois canhotos três em cada quatro filhos são destros. Feita a conta inversa, dois em cada três canhotos têm pai e mãe destros.

### ⚠️ Limitações do Modelo

- A meta-análise foi lida no manuscrito aceito pelos autores, não na versão final diagramada pela revista. Os números podem diferir em detalhe da cópia publicada.
- A heterogeneidade entre os estudos é alta (I² = 97,4%). O intervalo de 9,71% a 11,50% descreve a média dos estudos. O artigo não publica intervalo de predição, então a faixa esperada para uma população específica é mais larga do que essa.
- Os autores encontraram sinal de viés de estudos pequenos (teste de Egger, p < 0,001). A correção que eles simularam deslocaria a estimativa para cima, não para baixo.
- A resposta depende da definição. Entre o critério rígido e o mais frouxo, a proporção vai de 9,3% a 18,1%.
- Dos 216 conjuntos de dados com região registrada, 184 vêm de populações de ancestralidade europeia. Leste da Ásia (5,7%) e África subsaariana (7,7%) têm poucos estudos, e não há estimativa separada para o Brasil.
- A sala de 30 pessoas supõe pessoas independentes entre si. Numa família ou num time de esporte de raquete a conta não vale, porque a preferência de mão tem componente familiar e alguns esportes atraem canhotos.
- A chance de um canhoto sorteado ser homem supõe um grupo com metade de homens e metade de mulheres.
- A camada das gerações compara estudos pelo ano de publicação, não pessoas pelo ano de nascimento. Estudos antigos diferem dos recentes também em método e em população. O "um em cada três" é leitura nossa da diferença, não uma contagem de canhotos que trocaram de mão.
- A relação entre pressão cultural e proporção de canhotos é associação. Os próprios autores dizem que não conseguem separar com segurança cultura de genética, embora favoreçam a explicação cultural. A literatura traz outras hipóteses, entre elas pessoas mais velhas que escondiam a mão esquerda ao responder.
- Os percentuais de Gilbert & Wysocki (3% e quase 12%) entram como citados pela meta-análise. O artigo original foi lido só no resumo.
- Os três percentuais de filhos canhotos conforme a mão dos pais vêm de Annett (2002), citados em segunda mão, e o tamanho da amostra não foi localizado. A camada da família não muda o número do dia e deve ser lida como ordem de grandeza.
- A conta de onde vêm os canhotos supõe casais formados sem relação com a mão de cada um e a mesma proporção de canhotos na geração dos pais. A checagem interna dá 11,6% de filhos canhotos, perto da estimativa geral de 10,6%, sem coincidir com ela.
- Os 24% de Medland et al. (2009) descrevem a variação entre pessoas numa população de gêmeos australianos e holandeses. Não significam que 24% da mão de cada pessoa seja herdada.
- A história do sobrinho na abertura é uma observação pessoal. Usar a mão direita com o braço esquerdo engessado é adaptação, e o modelo não mede isso.
- A fonte mede a direção da preferência (qual mão), não o grau (quanto a pessoa depende dela) nem a habilidade de cada mão.

*A ciência é honesta sobre o que não sabe. O modelo também.*
