## Qual a probabilidade de a sua dor de cabeça ser falta de café?

**Tipo:** Saúde
**Data de publicação:** 2026-11-10
**Ferramenta:** Python
**Decisão analisada:** Essa dor de cabeça é falta de café, e o café ajuda o remédio a funcionar?
**Hashtag:** #365Probabilidades #Dia150

---

### 📖 A História

Durante a semana, tenho o hábito de tomar pelo menos três xícaras de café… Nos fins de semana, essa quantidade cai muito e, às vezes, chega a zero. Demorei para entender que a falta dele podia ser o que estragava o meu dia. Justo o dia mais esperado da semana…

Quase todo mundo que toma café todo dia conhece essa dor de cabeça. O domingo que começa mais tarde, a viagem longa, o exame de sangue em jejum. Lá pelo meio da tarde
a cabeça começa a pesar, e fica a dúvida entre cansaço, fome e o café que não veio.

Essa dúvida tem resposta medida. E o assunto tem um segundo lado, porque o mesmo café que faz
falta também entra na fórmula de vários remédios para dor de cabeça.

---

### 📚 O Conceito: a abstinência de cafeína

Quem toma cafeína todo dia se acostuma a ela, e o corpo reage quando ela falta. Em 2004, Laura
Juliano e Roland Griffiths, da Johns Hopkins, revisaram 57 estudos experimentais e 9 pesquisas
sobre o que acontece quando a pessoa para. A dor de cabeça foi o sintoma mais comum, presente em
50% dos participantes dos estudos experimentais. Em 13%, os sintomas foram fortes a ponto de
atrapalhar o dia. Eles costumam começar de 12 a 24 horas depois do último café, ficam mais fortes
entre o primeiro e o segundo dia e duram de 2 a 9 dias. Quanto maior a dose habitual, mais
frequentes os sintomas, mas parar uma dose de apenas 100 mg por dia, perto de uma xícara, já foi
suficiente para produzi-los. A revisão ajudou a incluir a abstinência de cafeína no manual de
diagnósticos da psiquiatria americana, o DSM-5.

O estudo que melhor isola o efeito é de 1992, do mesmo grupo. Kenneth Silverman e colegas
acompanharam 62 adultos que tomavam pouco ou moderadamente café, em média 235 mg por dia, cerca
de 2,5 xícaras. Cada pessoa passou dois períodos de dois dias sem café na dieta, tomando cápsulas
sem saber o que havia dentro. Num período, as cápsulas tinham a cafeína habitual. No outro, eram
placebo. Com placebo, 52% tiveram dor de cabeça moderada ou forte. Com a cafeína habitual, 6%. No
período normal, antes do teste, 2%.

O outro lado vem de uma revisão Cochrane de 2014, de Christopher Derry, Sheena Derry e Andrew
Moore. Ela reuniu 20 ensaios duplo-cegos que compararam o mesmo analgésico com e sem cafeína, em
dores agudas como dor de cabeça, dor de dente depois de cirurgia e dor pós-parto. Com cafeína
junto, 48% tiveram alívio bom, contra 41% só com o analgésico. O ganho aparece a partir de cerca
de 100 mg de cafeína, perto de uma caneca de café.

### 🧮 O Modelo

**Fontes:**
- Silverman, K., Evans, S. M., Strain, E. C. & Griffiths, R. R. (1992). Withdrawal syndrome after
  the double-blind cessation of caffeine consumption. *New England Journal of Medicine*, 327(16),
  1109-1114. N = 62, desenho dentro do mesmo indivíduo, duplo-cego, placebo contra cafeína habitual.
- Juliano, L. M. & Griffiths, R. R. (2004). A critical review of caffeine withdrawal. *Psychopharmacology*,
  176(1), 1-29. 57 estudos experimentais e 9 pesquisas.
- Derry, C. J., Derry, S. & Moore, R. A. (2014). Caffeine as an analgesic adjuvant for acute pain in
  adults. *Cochrane Database of Systematic Reviews*, Issue 12, CD009281. 20 ensaios duplo-cegos,
  4.262 pessoas nas análises de eficácia, 27 comparações.

**Corda declarada:** Griffiths assina as duas primeiras fontes, e a revisão de 2004 inclui o estudo
de 1992. São estudos diferentes do mesmo laboratório.

**Fator ×0,80:** não se aplica. Os números vêm de experimento duplo-cego e de ensaios randomizados,
não de pesquisa populacional de autorrelato.

**Três camadas:**
1. Posterior de Jeffreys para a dor de cabeça em cada período de Silverman 1992.
2. A resposta à pergunta do dia. Entre quem toma café todo dia, ficou dois dias sem e teve dor de
   cabeça moderada ou forte, que parte dessa dor se deve à falta da cafeína. É a fração atribuível,
   1 menos a razão entre a dor com cafeína e a dor com placebo, com 200 mil sorteios.
3. O outro lado. A diferença do Cochrane traduzida em pessoas, com o número necessário para tratar.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

SEMENTE = 150
rng = np.random.default_rng(SEMENTE)
SORTEIOS = 200_000

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

VERMELHO, DOURADO, VERDE, CINZA = '#c0392b', '#c8a84b', '#1a5f5a', '#6b6a64'
RODAPE = ('Fonte: Silverman et al. (1992), NEJM 327(16) · Juliano & Griffiths (2004), Psychopharmacology 176(1) · '
          'Derry, Derry & Moore (2014), Cochrane CD009281 · cálculo próprio | #365Probabilidades')

def vir(x, casas=1):
    return f"{x:.{casas}f}".replace(".", ",")

def posterior(p, n):
    # Beta de Jeffreys a partir da proporção publicada e do N
    k = p * n
    return stats.beta(k + 0.5, n - k + 0.5)

def ic(x):
    return np.percentile(x, [2.5, 97.5])

print("✅ Bibliotecas carregadas · semente", SEMENTE)
```

```python
# --- DADOS DA LITERATURA ---

# Silverman, Evans, Strain & Griffiths (1992), NEJM 327(16) · N = 62, cada pessoa nos três períodos
N_SILVERMAN = 62
DOSE_MEDIA_MG = 235
DOR_MOD_FORTE = {                      # dor de cabeça moderada ou forte
    'Período normal': 0.02,
    'Dois dias com a cafeína habitual': 0.06,
    'Dois dias com placebo': 0.52,
}
OUTROS_PLACEBO = {                     # proporção com escore anormal no período de placebo (cafeína entre parênteses)
    'Depressão (Beck) alta': (0.11, 0.03),
    'Pouco vigor (POMS)': (0.11, 0.02),
    'Ansiedade-traço alta': (0.08, 0.02),
    'Fadiga alta (POMS)': (0.08, 0.00),
    'Remédio não autorizado': (0.13, 0.02),
}

# Juliano & Griffiths (2004), Psychopharmacology 176(1) · 57 estudos experimentais e 9 pesquisas
JG_DOR = 0.50                          # incidência de dor de cabeça nos estudos experimentais
JG_PREJUIZO = 0.13                     # sofrimento ou prejuízo clinicamente relevante
JG_INICIO_H = (12, 24)                 # horas até os sintomas começarem
JG_PICO_DIAS = (1, 2)                  # pico de intensidade
JG_DURACAO_DIAS = (2, 9)
JG_DOSE_MINIMA_MG = 100

# Derry, Derry & Moore (2014), Cochrane CD009281
CO_SO_ANALGESICO = 0.41                # alívio bom (50% da dor máxima aliviada em 4 a 6 h)
CO_COM_CAFEINA = 0.48
CO_N = 4262                            # pessoas nas análises de eficácia, 27 comparações
CO_RR_PUB = (1.2, 1.1, 1.3)
CO_NNT_PUB = (14, 9.9, 24)

print("=" * 72)
print("  DADOS DA LITERATURA")
print("=" * 72)
print(f"\n  Silverman et al. 1992 · N = {N_SILVERMAN} · dose média {DOSE_MEDIA_MG} mg/dia")
for k, v in DOR_MOD_FORTE.items():
    print(f"    {k:<36} dor de cabeça moderada ou forte: {vir(v * 100, 0)}%")
print("    No placebo, outros sinais (com cafeína entre parênteses):")
for k, (pl, ca) in OUTROS_PLACEBO.items():
    print(f"      {k:<28} {vir(pl * 100, 0)}%  ({vir(ca * 100, 0)}%)")
print(f"\n  Juliano & Griffiths 2004 · dor de cabeça {vir(JG_DOR * 100, 0)}% · prejuízo {vir(JG_PREJUIZO * 100, 0)}%")
print(f"    início em {JG_INICIO_H[0]} a {JG_INICIO_H[1]} h · pico em {JG_PICO_DIAS[0]} a {JG_PICO_DIAS[1]} dias · "
      f"duração de {JG_DURACAO_DIAS[0]} a {JG_DURACAO_DIAS[1]} dias · a partir de {JG_DOSE_MINIMA_MG} mg/dia")
print(f"\n  Cochrane 2014 · N = {CO_N:,}".replace(",", "."))
print(f"    alívio bom só com analgésico: {vir(CO_SO_ANALGESICO * 100, 0)}% · com cafeína: {vir(CO_COM_CAFEINA * 100, 0)}%")
print(f"    publicado: RR {vir(CO_RR_PUB[0])} [{vir(CO_RR_PUB[1])}; {vir(CO_RR_PUB[2])}] · "
      f"NNT {CO_NNT_PUB[0]} [{vir(CO_NNT_PUB[1])}; {CO_NNT_PUB[2]}]")
print("=" * 72)
```

```python
# --- O MODELO ---

# CAMADA 1 · Dor de cabeça em cada período
print("=" * 72)
print("  CAMADA 1 · DOR DE CABEÇA MODERADA OU FORTE (posterior de Jeffreys, N = 62)")
print("=" * 72)
POST = {}
for k, p in DOR_MOD_FORTE.items():
    d = posterior(p, N_SILVERMAN)
    POST[k] = d.rvs(size=SORTEIOS, random_state=rng)
    lo, hi = d.interval(0.95)
    print(f"  {k:<36} {vir(p * 100, 0):>3}%  IC 95% [{vir(lo * 100)}; {vir(hi * 100)}]")

P_PLACEBO = POST['Dois dias com placebo']
P_CAFEINA = POST['Dois dias com a cafeína habitual']

# CAMADA 2 · A pergunta do dia: fração atribuível à falta de cafeína
FA = 1 - P_CAFEINA / P_PLACEBO
EXCESSO = P_PLACEBO - P_CAFEINA
print("\n" + "=" * 72)
print("  CAMADA 2 · QUANTO DA DOR É FALTA DE CAFÉ (fração atribuível, 200 mil sorteios)")
print("=" * 72)
print(f"  Dor de cabeça a mais por ficar sem a cafeína: {vir(EXCESSO.mean() * 100)} pontos  "
      f"IC 95% [{vir(ic(EXCESSO)[0] * 100)}; {vir(ic(EXCESSO)[1] * 100)}]")
print(f"  Entre quem teve dor sem cafeína, parte que vem da falta: {vir(FA.mean() * 100)}%  "
      f"IC 95% [{vir(ic(FA)[0] * 100)}; {vir(ic(FA)[1] * 100)}]")
p50 = (FA > 0.5).mean()
print("  P(fração atribuível > 50%): " + ("acima de 99,9%" if p50 > 0.999 else f"{vir(p50 * 100)}%"))
print(f"  P(fração atribuível > 75%): {vir((FA > 0.75).mean() * 100)}%")
print(f"  Em 10 pessoas que tomam café todo dia, ficam dois dias sem e têm dor de cabeça moderada ou forte,")
print(f"  cerca de {round(FA.mean() * 10)} teriam essa dor por causa da falta da cafeína.")

# CAMADA 3 · O outro lado: café junto do analgésico
n_braco = CO_N / 2                     # divisão por igual entre os braços, aproximação declarada
CO_A = posterior(CO_SO_ANALGESICO, n_braco).rvs(size=SORTEIOS, random_state=rng)
CO_B = posterior(CO_COM_CAFEINA, n_braco).rvs(size=SORTEIOS, random_state=rng)
CO_DIF = CO_B - CO_A
CO_RR = CO_B / CO_A
CO_NNT = 1 / CO_DIF
print("\n" + "=" * 72)
print("  CAMADA 3 · O OUTRO LADO (Cochrane 2014, braços divididos por igual)")
print("=" * 72)
print(f"  Diferença de alívio bom: {vir(CO_DIF.mean() * 100)} pontos  IC 95% [{vir(ic(CO_DIF)[0] * 100)}; {vir(ic(CO_DIF)[1] * 100)}]")
print(f"  Razão: {vir(CO_RR.mean(), 2)}  IC 95% [{vir(ic(CO_RR)[0], 2)}; {vir(ic(CO_RR)[1], 2)}]  · publicado: {vir(CO_RR_PUB[0])} [{vir(CO_RR_PUB[1])}; {vir(CO_RR_PUB[2])}]")
print(f"  NNT (mediana): {vir(np.median(CO_NNT), 0)}  IC 95% [{vir(ic(CO_NNT)[0], 0)}; {vir(ic(CO_NNT)[1], 0)}]  · publicado: 14 [9,9; 24]")
print(f"  Em 100 pessoas com dor: {round(CO_SO_ANALGESICO * 100)} melhoram bem só com o analgésico, "
      f"{round(CO_COM_CAFEINA * 100)} com a cafeína junto.")
print("  A conferência com o publicado vale como checagem: o intervalo do Cochrane combina as 27 comparações")
print("  separadamente, e o nosso trata os dois braços como um grupo único.")
print("=" * 72)
```

```python
# --- VISUALIZAÇÃO ---

# GRÁFICO 1 · Os três períodos
fig1, ax1 = plt.subplots(figsize=(12, 7))
nomes = list(DOR_MOD_FORTE.keys())
cores = [CINZA, VERDE, VERMELHO]
for i, (k, cor) in enumerate(zip(nomes, cores)):
    p = DOR_MOD_FORTE[k]
    lo, hi = posterior(p, N_SILVERMAN).interval(0.95)
    ax1.bar(i, p * 100, color=cor, alpha=0.85, width=0.55)
    ax1.plot([i, i], [lo * 100, hi * 100], color='#333', linewidth=2)
    ax1.text(i, hi * 100 + 2, f"{vir(p * 100, 0)}%", ha='center', fontsize=18, fontweight='bold', color=cor)
ax1.set_xticks(range(3))
ax1.set_xticklabels(['Período normal', 'Dois dias com\na cafeína habitual', 'Dois dias com\nplacebo'], fontsize=12)
ax1.set_ylim(0, 75)
ax1.set_ylabel('Dor de cabeça moderada ou forte (%) · barra e IC 95%', fontsize=12)
ax1.set_title('As mesmas 62 pessoas, sem saber o que havia na cápsula\n'
              'sem a cafeína, mais da metade teve dor de cabeça moderada ou forte', fontsize=14, pad=16)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=8, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-150-grafico-01-tres-periodos.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# GRÁFICO 2 · Quanto da dor é falta de café
fig2, ax2 = plt.subplots(figsize=(12, 7))
ax2.hist(FA * 100, bins=np.linspace(50, 100, 201), density=True, color=DOURADO, alpha=0.85)
m = FA.mean() * 100
lo, hi = ic(FA) * 100
ax2.axvline(m, color=VERMELHO, linewidth=2.5)
ax2.axvline(lo, color=CINZA, linestyle='--', linewidth=1.5)
ax2.axvline(hi, color=CINZA, linestyle='--', linewidth=1.5)
ax2.text(m - 0.5, ax2.get_ylim()[1] * 0.9, f"{vir(m, 0)}% ", ha='right', fontsize=20, fontweight='bold', color=VERMELHO)
ax2.text(lo, ax2.get_ylim()[1] * 0.75, f" IC 95%\n [{vir(lo, 0)}; {vir(hi, 0)}]", fontsize=11, color=CINZA)
ax2.set_xlabel('Parte da dor de cabeça, entre quem ficou sem café, que vem da falta da cafeína (%)', fontsize=12)
ax2.set_ylabel('Densidade da posterior', fontsize=12)
ax2.set_title('Sem café e com dor de cabeça, cerca de 9 em 10 dores são da falta dele\n'
              'fração atribuível = 1 − (dor com cafeína ÷ dor com placebo) · 200 mil sorteios', fontsize=13, pad=16)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=8, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-150-grafico-02-falta-de-cafe.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# GRÁFICO 3 · O outro lado, em 100 pessoas
fig3, ax3 = plt.subplots(figsize=(12, 5.2))
for linha, (rot, p, cor) in enumerate([('Só o analgésico', CO_SO_ANALGESICO, CINZA),
                                       ('Analgésico com cafeína', CO_COM_CAFEINA, VERDE)]):
    k = round(p * 100)
    for j in range(100):
        x, y = j % 50, (j // 50) * 0.9 + linha * 2.6
        ax3.scatter(x, -y, s=110, color=cor if j < k else '#e6e4de', edgecolors='none')
    ax3.text(-1.5, -linha * 2.6 - 0.45, f"{rot}\n{k} em 100", ha='right', va='center', fontsize=12, color=cor)
ax3.set_xlim(-14, 50)
ax3.set_ylim(-4.0, 0.7)
ax3.axis('off')
ax3.set_title('O outro lado: o café junto do remédio\n'
              f'7 pessoas a mais em cada 100 com alívio bom · NNT cerca de 14 · Cochrane, N = 4.262', fontsize=14, pad=16)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=8, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-150-grafico-03-o-outro-lado.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**Se você toma café todo dia, ficou sem ele e teve dor de cabeça forte, a chance de essa dor ser
falta de café é de cerca de 87%, numa faixa de 72% a 96%. Em 62 pessoas que não sabiam o que havia na cápsula, a dor de
cabeça moderada ou forte foi de 6% com a cafeína habitual para 52% com placebo, em dois dias.**

A abstinência de cafeína não é coisa da cabeça de quem toma café. Ela aparece de forma consistente
em 57 estudos experimentais, começa de 12 a 24 horas depois do último café e pode durar até 9
dias. E vem com dose baixa. Parar só 100 mg por dia, perto de uma xícara, já basta.

O outro lado também é medido. Junto do analgésico, a cafeína faz 7 pessoas a mais em cada 100
terem alívio bom. É um efeito pequeno, de qualidade alta, e por isso ela aparece na fórmula de
tantos remédios para dor de cabeça.

### ⚠️ Limitações do Modelo

- Silverman 1992 tem 62 pessoas, todas com consumo baixo ou moderado, e mede só dois dias sem
  cafeína. A fração atribuível vale para esse cenário e para dor de cabeça moderada ou forte.
- O desenho é dentro do mesmo indivíduo, mas o artigo publica só as proporções de cada período.
  O modelo trata os períodos como independentes, o que deixa os intervalos um pouco mais largos do
  que seriam com os dados pareados.
- A fração atribuível supõe que a dor com cafeína (6%) é a dor de fundo que existiria de qualquer
  jeito. Ela não diz, para uma pessoa específica, se aquela dor é ou não falta de café.
- Juliano & Griffiths e Silverman são do mesmo laboratório, e a revisão inclui o estudo de 1992.
  As fontes não são independentes, o que está declarado.
- No Cochrane, o N de cada braço não é publicado de forma direta. Dividimos as 4.262 pessoas por
  igual entre os dois braços, e o intervalo do modelo serve só como conferência com o publicado
  (RR 1,2 [1,1; 1,3], NNT 14 [9,9; 24]). A revisão cobre várias dores agudas, não só dor de cabeça.
- Ninguém replicou o estudo de 1992 com brasileiros. O café do Brasil costuma ser mais forte que
  o americano, e o efeito tende a crescer com a dose habitual.

*A ciência é honesta sobre o que não sabe. O modelo também.*
