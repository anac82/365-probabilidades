### 📖 A História

Você se sente espontâneo. Sério, você sente.

Hoje você quase foi almoçar em outro lugar. Semana passada mudou
o caminho de volta. Você hesita, escolhe, muda de ideia; a sua
vida, vista de dentro, é uma sequência de decisões em aberto.

Aí um físico chamado Barabási pegou os registros de antena de
**50 mil celulares**, três meses de rastros anonimizados, e mediu
a coisa mais desconfortável que dá para medir sobre uma pessoa:
o tamanho da sua surpresa.

O nome técnico é entropia. A tradução de mesa: **quantas
perguntas de sim ou não alguém precisaria fazer para acertar o
seu próximo lugar**. Um dado honesto de 50 faces custa quase 6
perguntas. Você custa **0,8**.

Menos de uma pergunta de sim ou não. Esse é o tamanho da sua
espontaneidade, medida.

E quando essa entropia passa pela matemática que transforma
surpresa em previsibilidade, sai o número do dia: **um teto de
93%**. O melhor algoritmo possível acertaria o seu próximo
paradeiro 93 vezes em 100.

A parte que dói: tanto faz se você roda a cidade ou fica no
bairro. A previsibilidade é quase a mesma.

---

### 📚 O Conceito: Da Surpresa ao Teto

O caminho do estudo tem três degraus, e vale a pena subir um por
um.

**Degrau 1, a entropia:** olhando a sequência de lugares de cada
pessoa, com ordem e horário, mede-se quanta surpresa existe nela.
Rotina forte, surpresa pequena. O valor típico: 0,8 bit.

**Degrau 2, a desigualdade de Fano:** um teorema da teoria da
informação que responde: dado esse tamanho de surpresa, qual é o
MÁXIMO que qualquer previsor poderia acertar? É uma ponte exata
entre a entropia e um teto de acerto. A assinatura matemática do
dia é resolver essa ponte, literalmente, no código abaixo.

**Degrau 3, a honestidade do teto:** 93% é o limite superior, não
uma taxa medida. Mas o teto não é ficção: na Costa do Marfim, com
500 mil pessoas, o teto deu ~88% e modelos simples de Markov
**realizaram ~87%**. E no Haiti, medido depois do terremoto de
2010, o teto seguia em 85%. Nem um terremoto desmonta a rotina
humana.

### 🧮 O Modelo

**Fontes:**
- Song, C., Qu, Z., Blumm, N. & Barabási, A.-L. (2010) - Limits
  of Predictability in Human Mobility - Science, 327(5968),
  1018-1021 - trajetórias anonimizadas de **50.000 usuários** de
  celular (base de 10 milhões), 3 meses. Entropia real típica
  **S ≈ 0,8 bit**; teto de previsibilidade **Π = 93%** via
  desigualdade de Fano; previsibilidade quase independente da
  distância percorrida. **Fonte central.**
- Lu, X., Wetter, E., Bharti, N., Tatem, A. J. & Bengtsson, L.
  (2013) - Approaching the limit of predictability in human
  mobility - Scientific Reports, 3, 2923 - **N=500.000** (Costa
  do Marfim): teto ~88%, cadeias de Markov realizam ~87%. Apoio:
  o teto é alcançável.
- Lu, X. et al. (PNAS) - Haiti pós-terremoto de 2010: teto de
  85% mesmo após o desastre. Apoio.
- Smith, G. et al. - com restrições espaciais de alcance, teto
  mais apertado: 81 a 85%. Contraponto declarado.

**Fator ×0,80: não se aplica.** Registros de antena, medida
direta de comportamento; nenhum autorrelato.

**A conta nossa (assinatura matemática):** resolver a
desigualdade de Fano numericamente: S = H(Π) + (1-Π)·log2(N-1),
onde H é a entropia binária e N o número de lugares. Três
entregas: a curva completa Π(S), a sensibilidade do teto ao
número de lugares (10, 50, 100), e o contraste com um dado
honesto de 50 faces.

**Fronteira declarada do dia:** o estudo mede LUGARES, não
decisões de vida. A extrapolação de "onde você estará" para
"o que você fará" é analogia deste texto, não resultado do paper.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
from scipy.optimize import brentq

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---

# Song et al. (2010), Science 327
n_usuarios = 50_000
S_real = 0.8          # bits: entropia real típica (ordem + horário)
teto_song = 0.93      # Π_max médio via Fano
# achado: previsibilidade quase independente da distância percorrida

# Lu et al. (2013), Scientific Reports: Costa do Marfim
n_lu = 500_000
teto_lu, realizado_lu = 0.88, 0.87   # Markov simples chega a ~87%

# Lu et al. (PNAS): Haiti pós-terremoto de 2010
teto_haiti = 0.85

# Smith et al.: com restrições espaciais
teto_smith = (0.81, 0.85)

print("=" * 65)
print("  DADOS DA LITERATURA - A PREVISIBILIDADE HUMANA")
print("=" * 65)
print(f"\n  Song et al. 2010 (N={n_usuarios:,}, 3 meses):".replace(',','.'))
print(f"  → Entropia real típica:      S ≈ {S_real} bit")
print(f"     (menos de uma pergunta de sim/não por passo)")
print(f"  → Teto de previsibilidade:   {teto_song*100:.0f}%  (via Fano)")
print(f"  → Quase independente da distância percorrida")
print(f"\n  Lu et al. 2013 (N={n_lu:,}, Costa do Marfim):".replace(',','.'))
print(f"  → Teto: {teto_lu*100:.0f}%  ·  Markov simples realiza: {realizado_lu*100:.0f}%")
print(f"  → O teto é alcançável, não ficção")
print(f"\n  Haiti, depois do terremoto de 2010: teto ainda em {teto_haiti*100:.0f}%")
print(f"  Smith et al., com restrições: teto de {teto_smith[0]*100:.0f} a {teto_smith[1]*100:.0f}%")
print(f"\n  Fator ×0,80: NÃO aplicado (registros de antena)")
print("=" * 65)
```

```python
# --- O MODELO ---
# A conta nossa: resolver a desigualdade de Fano, literalmente
# S = H(Π) + (1-Π)·log2(N-1), com H(Π) a entropia binária

def H2(p):
    p = np.clip(p, 1e-12, 1 - 1e-12)
    return -p*np.log2(p) - (1-p)*np.log2(1-p)

def fano_S(pi, N):
    return H2(pi) + (1 - pi) * np.log2(N - 1)

def teto(S, N):
    # maior Π com fano_S(Π, N) = S (ramo decrescente, Π > 1/N)
    lo, hi = 1.0/N + 1e-9, 1 - 1e-9
    if S >= np.log2(N):  # entropia máxima: uniforme
        return 1.0/N
    return brentq(lambda p: fano_S(p, N) - S, lo, hi)

# 1) O teto humano, com sensibilidade ao número de lugares
Ns = [10, 50, 100]
tetos_humanos = {N: teto(S_real, N) for N in Ns}

# 2) O contraste: um dado honesto de 50 faces
S_dado = np.log2(50)
teto_dado = teto(S_dado, 50)

print("=" * 65)
print("  MODELO - FANO RESOLVIDA (conta própria)")
print("=" * 65)
print(f"\n  S = H(Π) + (1-Π)·log2(N-1)   →   resolvida para Π")
print(f"\n  Com a entropia humana (S = {S_real} bit):")
for N in Ns:
    print(f"  → {N:>3} lugares frequentados:  teto = {tetos_humanos[N]*100:.1f}%")
print(f"  → O teto QUASE NÃO DEPENDE de quantos lugares você tem.")
print(f"     A rotina, não o mapa, decide a sua previsibilidade.")
print(f"\n  O contraste do dado honesto:")
print(f"  → 50 faces, entropia máxima (S = {S_dado:.2f} bits):")
print(f"     teto = {teto_dado*100:.0f}%")
print(f"  → Você: ~93%. Um dado: {teto_dado*100:.0f}%. A distância entre")
print(f"     esses dois números é o tamanho da sua rotina.")
print(f"\n  TETO CONTRA REALIZADO (a honestidade do dia):")
print(f"  → Song: teto de {teto_song*100:.0f}% (máximo teórico, não taxa)")
print(f"  → Lu: teto {teto_lu*100:.0f}%, realizado {realizado_lu*100:.0f}%:")
print(f"     modelos simples chegam a ~{realizado_lu/teto_lu*100:.0f}% do teto")
print(f"\n  A LEITURA:")
print(f"  → A sua espontaneidade, medida, cabe em 0,8 bit.")
print(f"  → E o que sobra (7 passos em 100) não é ruído: é onde")
print(f"     moram as exceções que você escolhe de verdade.")
print("=" * 65)
```

```python
# --- VISUALIZAÇÃO - GRÁFICOS SEPARADOS ---

# ── GRÁFICO 1 - A curva de Fano: da surpresa ao teto ──
fig1, ax1 = plt.subplots(figsize=(10, 6))

S_grid = np.linspace(0.01, np.log2(50), 300)
curva = [teto(s, 50) for s in S_grid]

ax1.plot(S_grid, np.array(curva)*100, color='#2a8a82', linewidth=3)
ax1.scatter([S_real], [tetos_humanos[50]*100], s=150, color='#c9a14a', zorder=3)
ax1.annotate(f'a humanidade\nS=0,8 → {tetos_humanos[50]*100:.0f}%',
             xy=(S_real, tetos_humanos[50]*100), xytext=(1.7, 88),
             fontsize=11.5, color='#8a6d2f', fontweight='bold',
             arrowprops=dict(arrowstyle='->', color='#8a6d2f', lw=1.5))
ax1.scatter([S_dado], [teto_dado*100], s=150, color='#c0392b', zorder=3)
ax1.annotate(f'um dado de 50 faces\nS=5,6 → {teto_dado*100:.0f}%',
             xy=(S_dado, teto_dado*100), xytext=(3.6, 25),
             fontsize=11.5, color='#c0392b', fontweight='bold',
             arrowprops=dict(arrowstyle='->', color='#c0392b', lw=1.5))
ax1.set_xlabel('Entropia S (bits): o tamanho da surpresa')
ax1.set_ylabel('Teto de previsibilidade (%)')
ax1.set_title('A desigualdade de Fano, resolvida: S = H(Π) + (1-Π)·log2(N-1)\n'
              'Quanto menor a surpresa, mais alto o teto (N=50 lugares)',
              fontsize=12, pad=15)

plt.figtext(0.5, 0.01,
            'Conta própria sobre Song et al. (2010) | equação resolvida numericamente | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-095-grafico-01-fano.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 - Teto contra realizado ──
fig2, ax2 = plt.subplots(figsize=(10, 6))

labels2 = ['Song 2010\nteto\n(N=50 mil)', 'Lu 2013\nteto\n(N=500 mil)',
           'Lu 2013\nREALIZADO\n(Markov simples)', 'Haiti\nteto\n(pós-terremoto)']
vals2 = [teto_song*100, teto_lu*100, realizado_lu*100, teto_haiti*100]
cores2 = ['#c9a14a', '#c9a14a', '#2a8a82', '#6b6a64']

bars2 = ax2.bar(labels2, vals2, color=cores2, alpha=0.88, width=0.5)
ax2.set_ylim(0, 105)
ax2.set_ylabel('Previsibilidade (%)')
ax2.set_title('Teto contra realizado: o limite não é ficção\n'
              'Modelos simples chegam a ~99% do teto na Costa do Marfim',
              fontsize=12.5, pad=15)

for bar, v in zip(bars2, vals2):
    ax2.text(bar.get_x() + bar.get_width()/2, v + 1.8,
             f'{v:.0f}%', ha='center', fontweight='bold', fontsize=14)

plt.figtext(0.5, 0.01,
            'Fontes: Song et al. (2010); Lu et al. (2013, Sci Rep); Lu et al. (PNAS) | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-095-grafico-02-teto-realizado.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 - O teto não depende do mapa ──
fig3, ax3 = plt.subplots(figsize=(10, 6))

labels3 = [f'{N} lugares' for N in Ns]
vals3 = [tetos_humanos[N]*100 for N in Ns]
bars3 = ax3.bar(labels3, vals3, color='#2a8a82', alpha=0.88, width=0.45)
ax3.set_ylim(0, 105)
ax3.set_ylabel('Teto de previsibilidade (%)')
ax3.set_title('Com S = 0,8 bit, o teto quase não depende de quantos lugares você tem\n'
              'A rotina, não o mapa, decide a sua previsibilidade',
              fontsize=12, pad=15)

for bar, v in zip(bars3, vals3):
    ax3.text(bar.get_x() + bar.get_width()/2, v + 1.8,
             f'{v:.1f}%', ha='center', fontweight='bold', fontsize=14)

ax3.text(0.5, 0.28, 'ecoa o achado do paper: previsibilidade\nquase independente da distância percorrida',
         transform=ax3.transAxes, ha='center', fontsize=11.5, color='#555', style='italic')

plt.figtext(0.5, 0.01,
            'Conta própria: sensibilidade da Fano ao N | Song et al. (2010) | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-095-grafico-03-sensibilidade.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**A sua espontaneidade, medida, cabe em 0,8 bit. E o teto da sua
previsibilidade é de 93%.**

Menos de uma pergunta de sim ou não por passo. Um dado honesto de
50 faces tem teto de 2%; você, de 93%. A distância entre esses
dois números é o tamanho da sua rotina, e ela não diminui se você
viaja muito: a previsibilidade é quase independente da distância.
A rotina, não o mapa, decide.

A honestidade em primeiro plano: 93% é teto, não taxa. Mas o teto
não é ficção: onde testaram de verdade, modelos simples
realizaram 87% de um teto de 88%. E nem um terremoto derruba o
padrão: no Haiti de 2010, o teto seguia em 85%.

E aqui está a leitura que este projeto escolhe: os 7 passos em
100 que nenhum algoritmo alcança não são ruído. São a margem onde
moram as exceções que você escolhe de verdade: o desvio
deliberado, o sim inesperado, o caminho novo por vontade. A
espontaneidade existe. Ela só é mais rara, e portanto mais
valiosa, do que a sensação sugere.

A pergunta que fica:

*Se 93 dos seus próximos 100 passos são anunciáveis,
o que você vai fazer com os 7 que não são?*

### ⚠️ Limitações do Modelo

- 93% é TETO teórico via desigualdade de Fano, não taxa medida;
  a distinção estrutura o dia e fica repetida aqui
- O estudo mede LUGARES (torres de celular), não decisões de
  vida; a extrapolação de "onde você estará" para "o que você
  fará" é analogia deste texto, declarada, não resultado do paper
- Resolução de antena: lugares são células de torre, não
  endereços; granularidade afeta a entropia
- Dados de 2010, era pré-smartphone ubíquo; padrões podem ter
  mudado (o rastreamento certamente aumentou)
- Com restrições espaciais de alcance, o teto aperta para 81 a
  85% (Smith et al.), declarado
- A conta própria usa S=0,8 como valor típico da base; a
  distribuição individual varia
- O "dado de 50 faces" é contraste didático: comportamento
  aleatório uniforme, não um modelo de pessoa

*A ciência é honesta sobre o que não sabe. O modelo também.*
