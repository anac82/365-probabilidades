### 📖 A História

Tem uma frase que todo mundo repete quando o assunto é dinheiro e felicidade.

Ganhadores de loteria voltam ao normal. Passa a euforia, a vida se reorganiza, e em
pouco tempo a pessoa está tão feliz ou tão infeliz quanto era antes.

A ideia tem até nome técnico, **esteira hedônica**: você corre, o cenário muda, e você
continua no mesmo lugar.

É uma frase reconfortante, porque ela promete que ninguém está perdendo nada por não
ter ganhado. E é repetida em livro de autoajuda, em aula de economia comportamental,
em conversa de bar, há quase cinquenta anos.

Fui atrás de onde ela nasceu.

---

### 📚 O Conceito: de onde veio a frase

Ela nasceu de um estudo de 1978, no *Journal of Personality and Social Psychology*.

Os pesquisadores entrevistaram ganhadores de uma loteria estadual recém-criada,
compararam com controles e com um grupo de pessoas que tinham sofrido acidentes com
paralisia, e pediram que todos avaliassem felicidade presente, futura e o prazer
sentido em atividades cotidianas, numa escala de 0 a 5.

O achado: os ganhadores **não eram mais felizes** que os controles, e sentiam
significativamente **menos prazer** em coisas comuns.

A amostra era de **22 ganhadores**.

Isso não é crítica retroativa fácil. Em 1978 não existia registro nacional de loteria
acessível a pesquisador, e um estudo com 22 pessoas era o que dava para fazer. O
problema não é o estudo. É o que o mundo fez com ele.

Porque existe uma diferença enorme entre "não encontramos diferença" e "não existe
diferença", e é exatamente essa distinção que este projeto já usou no **#068** e no
**#071**: **resultado nulo não prova ausência, ele limita o tamanho do efeito que
teria passado despercebido.**

Com 22 pessoas de cada lado, o que passa despercebido é quase tudo.

### 🧮 O Modelo

Um estudo fundador de 1978 e uma replicação moderna com desenho quase experimental.

  victims: is happiness relative?" · *Journal of Personality and Social Psychology*
  36(8), 917-927 · Estudo 1 com **22 grandes ganhadores de loteria**, **22 controles**
  e **29 vítimas de acidente com paralisia** · Estudo 2 com **86 participantes** ·

  futura **4,32** e prazer cotidiano **3,48**

  on Psychological Well-Being" · *The Review of Economic Studies* 87(6), 2703-2726 ·
  jogadores de loteria suecos pesquisados **5 a 22 anos** após o evento, com
  **procedimentos pré-registrados** e **controles pareados** · ganhadores de prêmios

  **saúde mental** são **significativamente menores** · a satisfação financeira aparece
  como mediador
- Östling, R., Cesarini, D. & Lindqvist, E., 2020 · *JAMA Network Open* · estudo de
  coorte quase experimental com **3.344 indivíduos** em três loterias suecas ·
  **nenhuma diferença estatisticamente significativa** em hábitos de saúde ou saúde
  geral no longo prazo, de 5 a 22 anos

**Pendência declarada:** o N exato da amostra de bem-estar de 2020 e os tamanhos de
efeito publicados não foram localizados nas fontes acessíveis. O desenho, o
pré-registro, a direção e a persistência dos efeitos estão publicados. Este notebook
não afirma nenhum tamanho de efeito para o estudo de 2020.

**Divergência registrada:** o resumo oficial de 1978 fala em 22 controles. Uma fonte
secundária descreve 58 pessoas de comparação vivendo na mesma vizinhança. Este notebook
usa o número do resumo oficial.

**Nota metodológica sobre o fator ×0.80:** não se aplica. Em 1978, escala de avaliação
em entrevista; em 2020, desenho quase experimental com aleatorização real do sorteio e
desfechos pré-registrados.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats, optimize

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---

# Brickman, Coates & Janoff-Bulman, 1978, JPSP 36(8), 917-927
n_ganhadores_1978 = 22
n_controles_1978  = 22
n_paralisia_1978  = 29
n_estudo2_1978    = 86
escala_1978       = (0, 5)

# Medias publicadas, escala de 0 a 5
RESULTADOS_1978 = {
    'Felicidade presente':      {'ganhadores': 4.00, 'controles': 3.82, 'paralisia': 2.96},
    'Felicidade futura':        {'ganhadores': 4.20, 'controles': 4.14, 'paralisia': 4.32},
    'Prazer no cotidiano':      {'ganhadores': 3.33, 'controles': 3.82, 'paralisia': 3.48},
}

# Lindqvist, Ostling & Cesarini, 2020, Review of Economic Studies 87(6), 2703-2726
janela_anos = (5, 22)
pre_registrado = True
# Direcao dos efeitos, conforme publicado
EFEITOS_2020 = {
    'Satisfacao geral com a vida': 'aumento sustentado, persiste mais de uma decada',
    'Felicidade':                  'efeito significativamente MENOR',
    'Saude mental':                'efeito significativamente MENOR',
}
n_2020_bemestar = None   # NAO LOCALIZADO. Nao inventar.

# Ostling, Cesarini & Lindqvist, 2020, JAMA Network Open
n_saude = 3_344
n_loterias = 3
efeito_saude = 'nenhuma diferenca estatisticamente significativa'

aplica_fator_080 = False

print("=" * 72)
print("  DADOS - GANHADORES DE LOTERIA, 1978 E 2020")
print("=" * 72)
print(f"\n  Brickman, Coates & Janoff-Bulman, 1978:")
print(f"  -> {n_ganhadores_1978} ganhadores, {n_controles_1978} controles,"
      f" {n_paralisia_1978} com paralisia")
print(f"  -> Estudo 2 com {n_estudo2_1978} participantes")
print(f"  -> Escala de {escala_1978[0]} a {escala_1978[1]}")
print(f"\n  {'medida':<24} {'ganhad.':>8} {'control.':>9} {'diferenca':>10}")
print(f"  {'-'*24} {'-'*8} {'-'*9} {'-'*10}")
for medida, v in RESULTADOS_1978.items():
    dif = v['ganhadores'] - v['controles']
    print(f"  {medida:<24} {v['ganhadores']:>8.2f} {v['controles']:>9.2f}"
          f" {dif:>+10.2f}")
print(f"\n  Lindqvist, Ostling & Cesarini, 2020 (Review of Economic Studies):")
print(f"  -> Jogadores suecos, {janela_anos[0]} a {janela_anos[1]} anos apos o evento")
print(f"  -> Pre-registrado: {pre_registrado} · controles pareados")
for desfecho, efeito in EFEITOS_2020.items():
    print(f"  -> {desfecho:<28} {efeito}")
print(f"  -> N exato: NAO LOCALIZADO nas fontes acessiveis")
print(f"\n  Ostling, Cesarini & Lindqvist, 2020 (JAMA Network Open):")
print(f"  -> N = {n_saude:,}".replace(",", ".") + f" em {n_loterias} loterias suecas")
print(f"  -> Habitos de saude e saude geral: {efeito_saude}")
print(f"\n  Fator x0.80 aplicado: {aplica_fator_080}")
print("=" * 72)
```

```python
# --- O MODELO ---
# O estudo de 1978 encontrou um NULO. Nulo nao prova ausencia: limita o tamanho.
# Pergunta: com 22 contra 22, que tamanho de efeito seria detectavel?

ALFA = 0.05
PODER = 0.80


def poder_t(d, n1, n2, alfa=ALFA):
    gl = n1 + n2 - 2
    ncp = d * np.sqrt((n1 * n2) / (n1 + n2))
    critico = stats.t.ppf(1 - alfa / 2, gl)
    acima = float(np.nan_to_num(stats.nct.sf(critico, gl, ncp), nan=0.0))
    abaixo = float(np.nan_to_num(stats.nct.cdf(-critico, gl, ncp), nan=0.0))
    return min(acima + abaixo, 1.0)


d_minimo = optimize.brentq(
    lambda d: poder_t(d, n_ganhadores_1978, n_controles_1978) - PODER, 0.05, 3.0)

tamanhos = {'pequeno (d=0,20)': 0.20, 'medio (d=0,50)': 0.50, 'grande (d=0,80)': 0.80}
poderes = {k: poder_t(v, n_ganhadores_1978, n_controles_1978)
           for k, v in tamanhos.items()}

# A CONTA REVERSA, que e a mais interessante:
# a diferenca observada foi de 0,18 ponto na escala de 0 a 5.
# Para que essa diferenca fosse detectavel com n=22, qual desvio padrao
# a medida precisaria ter?
dif_observada = (RESULTADOS_1978['Felicidade presente']['ganhadores']
                 - RESULTADOS_1978['Felicidade presente']['controles'])
dp_necessario = dif_observada / d_minimo
amplitude_escala = escala_1978[1] - escala_1978[0]
dp_como_pct_escala = dp_necessario / amplitude_escala

# Quantas pessoas seriam necessarias para detectar a diferenca observada,
# supondo um desvio padrao plausivel para uma escala de 0 a 5
DP_PLAUSIVEL = 1.0        # premissa declarada
d_real = dif_observada / DP_PLAUSIVEL


def n_para(d, alfa=ALFA, poder=PODER, teto=6000):
    for n in range(5, teto):
        if poder_t(d, n, n, alfa) >= poder:
            return n
    return teto


n_necessario = n_para(d_real)

print("=" * 72)
print("  MODELO - O QUE 22 PESSOAS CONSEGUIAM ENXERGAR")
print("=" * 72)
print(f"\n  1) Efeito minimo detectavel em 1978, com {PODER*100:.0f}% de poder:")
print(f"  -> d = {d_minimo:.2f}  (um efeito GRANDE)")
print(f"\n  2) Poder do estudo de 1978, por tamanho de efeito:")
for nome, valor in tamanhos.items():
    print(f"  -> {nome:<20} poder = {poderes[nome]*100:.0f}%")
print(f"\n  3) A conta reversa, e ela e a mais reveladora:")
print(f"  -> Diferenca observada em felicidade presente: {dif_observada:.2f} ponto")
print(f"  -> Para ser detectavel com n={n_ganhadores_1978}, o desvio padrao da")
print(f"     medida precisaria ser de no maximo {dp_necessario:.2f},")
print(f"     ou seja, {dp_como_pct_escala*100:.0f}% da amplitude da escala de 0 a 5.")
print(f"  -> Isso e implausivel para uma escala de avaliacao subjetiva.")
print(f"  -> Traduzindo: o estudo NAO TINHA COMO detectar a diferenca")
print(f"     que ele proprio observou.")
print(f"\n  4) Quantas pessoas seriam necessarias?")
print(f"  -> Assumindo desvio padrao de {DP_PLAUSIVEL:.1f} (premissa declarada),")
print(f"     a diferenca de {dif_observada:.2f} equivale a d = {d_real:.2f}")
print(f"  -> Seriam necessarias cerca de {n_necessario:,}".replace(",", ".") +
      f" pessoas por grupo.")
print(f"  -> O estudo tinha {n_ganhadores_1978}.")
print(f"\n  5) O que mudou em 2020:")
print(f"  -> Aleatorizacao REAL: o sorteio e aleatorio de verdade")
print(f"  -> Pre-registro dos procedimentos")
print(f"  -> Janela de {janela_anos[0]} a {janela_anos[1]} anos, nao de meses")
print(f"  -> Controles pareados")
print(f"  -> Resultado: satisfacao com a vida sobe e NAO dissipa")
print("  -> E, em " + f"{n_saude:,}".replace(",", ".") + " pessoas, saude nao muda nada.")
print("=" * 72)
```

### 💡 O Insight

A ideia de que ganhadores de loteria voltam ao normal nasceu de um estudo com
**22 pessoas**.

Vinte e duas. Entrevistadas com uma escala de 0 a 5, meses depois do prêmio, sem
pré-registro, porque em 1978 nada disso existia.

E o resultado foi um **nulo**: os ganhadores não pareciam mais felizes que os
controles.

Rodei o cálculo que ninguém fez na época, porque na época não se fazia. Com 22 pessoas
de cada lado, o estudo só conseguiria detectar um efeito **grande**, de d = 0,87 ou
mais. Para um efeito médio, a chance de enxergar era de menos de 30%.

E aí vem a conta que me deixou desconfortável.

A diferença que eles **de fato observaram** em felicidade presente foi de 0,18 ponto,
com 4,00 nos ganhadores e 3,82 nos controles. Para que uma diferença desse tamanho
fosse detectável com 22 pessoas, o desvio padrão da medida teria que ser de no máximo
0,21, ou seja, 4% da amplitude de uma escala de 0 a 5.

Isso é implausível para qualquer escala de avaliação subjetiva.

Traduzindo: **o estudo não tinha como detectar a diferença que ele mesmo observou.**
Não foi um achado de ausência de efeito. Foi um instrumento sem resolução suficiente.

Em 2020, alguém refez a pergunta com o que não existia em 1978.

Jogadores de loteria suecos, acompanhados de **5 a 22 anos** depois do prêmio, com
procedimentos **pré-registrados**, controles pareados, e a vantagem que nenhum
epidemiologista consegue comprar: **o sorteio é aleatório de verdade.**

O resultado: ganhadores de prêmios grandes têm aumento sustentado de satisfação com a
vida, **que persiste por mais de uma década e não dá sinal de dissipar.**

A esteira hedônica, pelo menos nessa forma, não apareceu.

Mas o dia não termina aí, e a segunda parte é a que eu levo comigo.

Os mesmos autores olharam saúde, com **3.344 pessoas** em três loterias. Hábitos de
saúde, saúde geral, no longo prazo: **nenhuma diferença estatisticamente
significativa.**

O dinheiro comprou satisfação duradoura com a vida. Não comprou saúde nenhuma.

E os efeitos sobre felicidade e sobre saúde mental foram significativamente menores que
o efeito sobre satisfação. Ou seja, mudou a avaliação que a pessoa faz da própria vida
muito mais do que mudou como ela se sente por dentro.

Fico pensando em quantas outras frases que a gente repete com naturalidade têm 22
pessoas atrás delas.

### ⚠️ Limitações do Modelo

- **O cálculo de poder aplicado a 1978 é retrospectivo.** Ele não corrige o estudo nem
  acusa os autores, que fizeram o possível com os recursos da época. Ele apenas mostra
  o que aquele desenho conseguia e não conseguia sustentar.
- **A conta reversa depende da diferença publicada e do efeito mínimo detectável.** O
  desvio padrão das medidas de 1978 não foi localizado, e por isso a conclusão é sobre
  plausibilidade, não sobre um número exato.
- **O cálculo de quantas pessoas seriam necessárias assume desvio padrão de 1,0** numa
  escala de 0 a 5. É premissa declarada, e serve como ordem de grandeza.
- **O N exato do estudo de bem-estar de 2020 não foi localizado** nas fontes
  acessíveis, nem os tamanhos de efeito publicados. O desenho, o pré-registro e a
  direção dos resultados estão publicados. Este notebook não afirma magnitude para
  esse estudo.
- **Divergência sobre o grupo de comparação de 1978:** o resumo oficial fala em 22
  controles, uma fonte secundária descreve 58 pessoas da mesma vizinhança. Usei o
  resumo oficial.
- **Suécia não é o mundo.** Um país com rede de proteção social forte, saúde pública
  universal e baixa desigualdade é exatamente o lugar onde dinheiro extra deveria
  importar **menos**. Isso torna o achado mais notável, e ao mesmo tempo limita a
  generalização para países como o Brasil.
- **Ganhar na loteria não é ficar rico do jeito comum.** É riqueza súbita, sem esforço
  e sem mudança de identidade profissional. Nada aqui se aplica a renda conquistada
