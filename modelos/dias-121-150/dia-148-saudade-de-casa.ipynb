# 365 Probabilidades - Dia #148
## Qual a probabilidade de quem mora em outro país sentir saudade de casa este mês?

**Tipo:** Emocional
**Data de publicação:** 2026-11-08
**Ferramenta:** Python
**Decisão analisada:** A saudade de casa é coisa das primeiras semanas, ou acompanha quem mora fora?
**Hashtag:** #365Probabilidades #Dia148

---

### 📖 A História

É comum pensar que saudade de casa é coisa de quem acabou de sair. Das primeiras
semanas num lugar novo, antes de aprender o caminho do mercado e o nome dos vizinhos.
Depois disso, a vida se acomoda e a saudade some.

Fui ver se alguém mediu isso em adultos que moram fora há anos.

---

### 📚 O Conceito: saudade de casa não é nostalgia

A psicologia separa duas coisas que em português cabem na mesma palavra. A nostalgia é
a lembrança afetiva de um tempo, e costuma fazer bem (foi o tema do #042). A saudade de
casa, que a literatura chama de *homesickness*, é a falta de um lugar e das pessoas que
ficaram nele, sentida por quem está longe. Ela pode pesar.

Quase tudo o que se sabe sobre ela vem de estudantes no primeiro ano fora de casa e de
crianças em acampamento. O estudo deste dia é raro porque mediu adultos, com idade
média de 42 anos, que trabalhavam em outro país havia anos. A pergunta foi simples:
nas últimas quatro semanas, você sentiu saudade de casa?

---

### 🧮 O Modelo

Quatro camadas. A fonte dá as proporções. As contas são nossas.

1. **A chance no mês.** A proporção que sentiu saudade de casa nas últimas quatro
   semanas, com intervalo de 95%, o fator ×0,80 e a segunda medida do próprio estudo.
2. **Por idade.** As proporções por faixa de idade, como publicadas, e o quanto o
   intervalo se alarga porque o tamanho de cada faixa não foi lido.
3. **Pelo tempo fora.** As proporções por anos no país. É a variável que mais pesou no
   estudo.
4. **Uma mesa de dez.** Numa mesa com dez pessoas que moram fora, quantas sentiram
   saudade de casa este mês.

**Fonte central:**
- Eurelings-Bontekoe, E. H. M., Brouwers, E. P. M. & Verschuur, M. J. (2000).
  "Homesickness among foreign employees of a multinational high-tech company in the
  Netherlands". *Environment and Behavior* 32(3), 443-456. DOI
  10.1177/00139160021972612. **N = 171** funcionários estrangeiros de uma multinacional
  na Holanda (71 mulheres e 100 homens), de 24 a 63 anos, idade média de 42,1. Franceses,
  britânicos, italianos, alemães, escandinavos e outros europeus. Autorrelato: **29,6%**
  sentiram saudade de casa nas últimas quatro semanas; 7 pessoas disseram "muita". Uma
  lista de nove sintomas, aplicada em paralelo, apontou 18,9%. Por idade: até 34 anos
  35,7%, de 34 a 39 anos 45,0%, de 40 a 51 anos 25,0%, de 52 a 63 anos 15,9%. Por anos
  no país: de 0,5 a 5 anos 42,1%, de 6 a 8 anos 44,2%, de 9 a 18 anos 21,2%, de 19 a 38
  anos 17,5%. Na regressão, o tempo de estadia foi o único preditor independente. **Lido
  no texto do artigo. O tamanho de cada faixa de idade e de tempo não foi lido.**

**Fonte de apoio:**
- Hamid, A. A. R. M. (2022). "Psychological distress and homesickness among Sudanese
  migrants in the United Arab Emirates". *Frontiers in Psychology* 12, 710115.
  N = 1.444 migrantes, de 18 a 72 anos, idade média de 40. A saudade de casa, medida por
  escala, diminui com a idade (r = −0,11) e com o tempo de estadia. Não publica proporção.

**Nota metodológica.** O fator ×0,80 do projeto se aplica. A medida central é a resposta
a uma pergunta de pesquisa ("sentiu saudade de casa nas últimas quatro semanas?"), que
é autorrelato de proporção. O fator entra em todas as proporções do estudo, e não só na
central. A lista de nove sintomas, que deu 18,9%, serve de contraprova: ela fica entre
o número publicado e o corrigido.

**Ponte interna:** #042 (nostalgia), com outra literatura e outra pergunta.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

SEMENTE = 42
SORTEIOS = 200_000
rng = np.random.default_rng(SEMENTE)

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

VERMELHO, DOURADO, VERDE, CINZA = '#c0392b', '#c8a84b', '#1a5f5a', '#6b6a64'
RODAPE = ('Fonte: Eurelings-Bontekoe, Brouwers & Verschuur (2000), Environment and Behavior 32(3) · '
          'N = 171 · cálculo próprio | #365Probabilidades')

def mil(n):
    return f"{int(round(n)):,}".replace(",", ".")

def vir(x, casas=1):
    return f"{x:.{casas}f}".replace(".", ",")

def jeffreys(k, n):
    return stats.beta(k + 0.5, n - k + 0.5)

print("✅ Bibliotecas carregadas · semente", SEMENTE)
```

```python
# --- DADOS DA LITERATURA ---
# Eurelings-Bontekoe, Brouwers & Verschuur (2000). Proporções lidas no texto.

N = 171
P_MES = 0.296                     # sentiram saudade de casa nas últimas 4 semanas (autorrelato)
MUITA = 7                         # disseram "muita" saudade
P_LISTA = 0.189                   # lista de nove sintomas, aplicada em paralelo
FATOR = 0.80

POR_IDADE = {'até 34': 0.357, '34 a 39': 0.450, '40 a 51': 0.250, '52 a 63': 0.159}
POR_TEMPO = {'0,5 a 5': 0.421, '6 a 8': 0.442, '9 a 18': 0.212, '19 a 38': 0.175}   # anos no país

K_MES = round(P_MES * N)          # contagem reconstruída da proporção publicada

print("=" * 70)
print("  DADOS · EURELINGS-BONTEKOE, BROUWERS & VERSCHUUR (2000)")
print("=" * 70)
print(f"\n  {N} funcionários estrangeiros, de 24 a 63 anos, idade média de 42")
print(f"  Sentiram saudade de casa nas últimas 4 semanas: {vir(P_MES * 100)}% (≈ {K_MES} pessoas)")
print(f"  Disseram 'muita' saudade: {MUITA} pessoas ({vir(MUITA / N * 100)}%)")
print(f"  Lista de nove sintomas: {vir(P_LISTA * 100)}%")
print("\n  Por idade:")
for k, v in POR_IDADE.items():
    print(f"    {k:>8} anos  {vir(v * 100):>6}%")
print("\n  Por anos no país:")
for k, v in POR_TEMPO.items():
    print(f"    {k:>8} anos  {vir(v * 100):>6}%")
print("\n  O tamanho de cada faixa não foi lido. As contagens abaixo são reconstruídas.")
print("=" * 70)
```

```python
# --- O MODELO ---

# CAMADA 1 · A chance no mês
dist = jeffreys(K_MES, N)
ic = dist.interval(0.95)
p_corr = P_MES * FATOR
ic_corr = (ic[0] * FATOR, ic[1] * FATOR)
print("=" * 70)
print("  CAMADA 1 · A CHANCE DE SENTIR SAUDADE DE CASA NO MÊS")
print("=" * 70)
print(f"\n  Publicado (autorrelato):     {vir(P_MES * 100)}%  IC 95% [{vir(ic[0] * 100)}; {vir(ic[1] * 100)}]")
print(f"  Com o fator ×0,80:           {vir(p_corr * 100)}%  [{vir(ic_corr[0] * 100)}; {vir(ic_corr[1] * 100)}]")
print(f"  Lista de nove sintomas:      {vir(P_LISTA * 100)}%")
print(f"  Saudade forte ('muita'):     {vir(MUITA / N * 100)}%  "
      f"IC 95% [{vir(jeffreys(MUITA, N).interval(0.95)[0] * 100)}; {vir(jeffreys(MUITA, N).interval(0.95)[1] * 100)}]")
print(f"\n  De cada quatro pessoas que moram fora, uma sentiu saudade de casa no último mês.")
print(f"  De cada {vir(N / MUITA, 0)}, uma sentiu muita.")

# CAMADA 2 · Por idade, com o tamanho das faixas desconhecido
print("\n" + "=" * 70)
print("  CAMADA 2 · POR IDADE")
print("=" * 70)
print(f"\n  {'faixa':>10} {'publicado':>10} {'×0,80':>8}   {'IC 95% se a faixa tiver 25 pessoas':>36}   {'se tiver 60':>14}")
idade_corr = {}
for k, v in POR_IDADE.items():
    idade_corr[k] = v * FATOR
    i25 = jeffreys(round(v * 25), 25).interval(0.95)
    i60 = jeffreys(round(v * 60), 60).interval(0.95)
    print(f"  {k:>10} {vir(v * 100):>9}% {vir(v * FATOR * 100):>7}%   "
          f"[{vir(i25[0] * 100)}; {vir(i25[1] * 100)}]{'':>22} [{vir(i60[0] * 100)}; {vir(i60[1] * 100)}]")
razao_40 = POR_IDADE['34 a 39'] / POR_IDADE['40 a 51']
print(f"\n  O pico é dos 34 aos 39 anos. Dos 40 aos 51 a proporção é {vir(razao_40)} vezes menor.")
print("  Com 171 pessoas em quatro faixas, os intervalos se cruzam. A forma da curva é sugestão, e não medida.")

# CAMADA 3 · Pelo tempo fora
print("\n" + "=" * 70)
print("  CAMADA 3 · PELO TEMPO FORA")
print("=" * 70)
ate_8 = np.mean([POR_TEMPO['0,5 a 5'], POR_TEMPO['6 a 8']])
de_9 = np.mean([POR_TEMPO['9 a 18'], POR_TEMPO['19 a 38']])
print(f"\n  Até 8 anos no país:      cerca de {vir(ate_8 * 100, 0)}% (×0,80: {vir(ate_8 * FATOR * 100, 0)}%)")
print(f"  De 9 anos em diante:     cerca de {vir(de_9 * 100, 0)}% (×0,80: {vir(de_9 * FATOR * 100, 0)}%)")
print(f"  Razão: {vir(ate_8 / de_9)} vezes")
print("  O estudo diz que o tempo de estadia foi o único preditor que sobrou na regressão.")
print("  A média simples das duas faixas supõe faixas do mesmo tamanho, o que não foi lido.")

# CAMADA 4 · Uma mesa de dez
MESA = 10
print("\n" + "=" * 70)
print(f"  CAMADA 4 · UMA MESA DE {MESA} PESSOAS QUE MORAM FORA")
print("=" * 70)
p_mesa = rng.beta(K_MES + 0.5, N - K_MES + 0.5, SORTEIOS) * FATOR     # incerteza da amostra + fator
saudosos = rng.binomial(MESA, p_mesa)
print(f"\n  Sentiram saudade de casa este mês: média {vir(saudosos.mean())} de {MESA} · mediana {int(np.median(saudosos))}")
print(f"  Nenhuma: {vir((saudosos == 0).mean() * 100)}% das mesas")
print(f"  Pelo menos uma: {vir((saudosos >= 1).mean() * 100)}%")
print(f"  Metade ou mais: {vir((saudosos >= MESA / 2).mean() * 100)}%")
print("  Premissa: as dez pessoas são independentes e parecidas com a amostra do estudo.")
print("=" * 70)
```

```python
# --- VISUALIZAÇÃO ---

# GRÁFICO 1 · A chance no mês
fig1, ax1 = plt.subplots(figsize=(12, 8))
rot = ['sentiram saudade\n(autorrelato)', 'com o fator\n×0,80', 'lista de nove\nsintomas', 'sentiram\n"muita"']
val = [P_MES * 100, p_corr * 100, P_LISTA * 100, MUITA / N * 100]
cores = [VERMELHO, DOURADO, CINZA, VERDE]
bars = ax1.bar(rot, val, color=cores, alpha=0.85, width=0.6)
ax1.errorbar([0, 1], [P_MES * 100, p_corr * 100],
             yerr=[[(P_MES - ic[0]) * 100, (p_corr - ic_corr[0]) * 100], [(ic[1] - P_MES) * 100, (ic_corr[1] - p_corr) * 100]],
             fmt='none', ecolor='#333', capsize=8, lw=1.5)
tops = [ic[1] * 100, ic_corr[1] * 100, P_LISTA * 100, MUITA / N * 100]
for b, v, t in zip(bars, val, tops):
    ax1.text(b.get_x() + b.get_width() / 2, t + 1.2, f"{vir(v)}%", ha='center', fontsize=17, fontweight='bold')
ax1.set_ylim(0, 48)
ax1.set_ylabel('Pessoas que moram fora (%)', fontsize=12)
ax1.set_title('Saudade de casa nas últimas quatro semanas\nadultos que trabalham em outro país, idade média de 42',
              fontsize=14, pad=16)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-148-grafico-01-no-mes.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# GRÁFICO 2 · Por idade
fig2, ax2 = plt.subplots(figsize=(12, 8))
x = np.arange(len(POR_IDADE))
pub = np.array(list(POR_IDADE.values())) * 100
ax2.bar(x - 0.2, pub, width=0.4, color=VERMELHO, alpha=0.85, label='publicado')
ax2.bar(x + 0.2, pub * FATOR, width=0.4, color=DOURADO, alpha=0.85, label='com o fator ×0,80')
for i, v in enumerate(pub):
    ax2.text(i - 0.2, v + 1, f"{vir(v)}%", ha='center', fontsize=13, fontweight='bold')
    ax2.text(i + 0.2, v * FATOR + 1, f"{vir(v * FATOR)}%", ha='center', fontsize=12, color='#6b5a1e')
ax2.set_xticks(x)
ax2.set_xticklabels([f"{k} anos" for k in POR_IDADE], fontsize=12)
ax2.set_ylim(0, 55)
ax2.set_ylabel('Sentiram saudade de casa no mês (%)', fontsize=12)
ax2.set_title('O pico é no fim dos 30, e depois dos 40 cai\n'
              'o tamanho de cada faixa não foi publicado, então os intervalos se cruzam', fontsize=14, pad=16)
ax2.legend(frameon=False, fontsize=12)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-148-grafico-02-por-idade.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# GRÁFICO 3 · Pelo tempo fora
fig3, ax3 = plt.subplots(figsize=(12, 8))
x = np.arange(len(POR_TEMPO))
pub_t = np.array(list(POR_TEMPO.values())) * 100
ax3.plot(x, pub_t, color=VERDE, linewidth=3.5, marker='o', markersize=10, label='publicado')
ax3.plot(x, pub_t * FATOR, color=DOURADO, linewidth=3, marker='o', markersize=8, linestyle='--', label='com o fator ×0,80')
for i, v in enumerate(pub_t):
    ax3.annotate(f"{vir(v)}%", (i, v), xytext=(0, 12), textcoords='offset points', ha='center', fontsize=13, fontweight='bold')
ax3.set_xticks(x)
ax3.set_xticklabels([f"{k} anos\nno país" for k in POR_TEMPO], fontsize=12)
ax3.set_ylim(0, 55)
ax3.set_ylabel('Sentiram saudade de casa no mês (%)', fontsize=12)
ax3.set_title('O tempo fora pesa mais do que a idade\n'
              'até 8 anos, cerca de 43%; de 9 anos em diante, cerca de 19%', fontsize=14, pad=16)
ax3.legend(frameon=False, fontsize=12)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-148-grafico-03-tempo-fora.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")

# GRÁFICO 4 · Uma mesa de dez
fig4, ax4 = plt.subplots(figsize=(12, 8))
ks = np.arange(0, MESA + 1)
freq = np.array([(saudosos == k).mean() for k in ks]) * 100
ax4.bar(ks, freq, color=VERMELHO, alpha=0.85, width=0.7)
for k in ks:
    if freq[k] >= 1:
        ax4.text(k, freq[k] + 0.6, f"{vir(freq[k], 0)}", ha='center', fontsize=12)
ax4.set_xticks(ks)
ax4.set_xlabel(f'Pessoas com saudade de casa este mês, de {MESA} que moram fora', fontsize=12)
ax4.set_ylabel('Mesas simuladas (%)', fontsize=12)
ax4.set_title(f'Numa mesa de {MESA} pessoas que moram fora\n'
              f'{mil(SORTEIOS)} mesas simuladas, com o fator ×0,80 e a incerteza da amostra', fontsize=14, pad=16)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-148-grafico-04-mesa-de-dez.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 4 salvo!")
```

### 💡 O Insight

**A probabilidade de quem mora em outro país sentir saudade de casa este mês é de 24%.**
No estudo, 29,6% disseram que sim nas últimas quatro semanas. Com o fator ×0,80 do
projeto, 23,7%, com intervalo de 19% a 30%. A lista de sintomas aplicada em paralelo deu
18,9%, dentro dessa faixa.

Uma em cada quatro pessoas que moram fora. E uma em cada 24 sentiu muita.

A saudade de casa não é coisa das primeiras semanas. As pessoas do estudo tinham, em
média, 42 anos e viviam fora havia anos.

O que mais pesa é o tempo. Até oito anos no país, cerca de 43% sentiram saudade no mês.
De nove anos em diante, cerca de 19%. A saudade cai pela metade, e não some.

A idade também aparece. O pico é dos 34 aos 39 anos, com 45%. Dos 40 aos 51, 25%. Dos
52 aos 63, 16%. Mas o estudo não publica quantas pessoas há em cada faixa, e com 171
pessoas divididas em quatro os intervalos se cruzam. A forma da curva é sugestão.

Numa mesa com dez pessoas que moram fora, em média duas ou três sentiram saudade de
casa este mês. Em 93% das mesas há pelo menos uma.

A pergunta que fica:

*De que lugar você sente saudade, e há quanto tempo saiu de lá?*

### ⚠️ Limitações do Modelo

- **Uma empresa, 171 pessoas, ano 2000.** Europeus que trabalhavam numa multinacional
  na Holanda. Não são migrantes em geral, nem brasileiros no exterior.
- **É autorrelato.** "Sentiu saudade de casa nas últimas quatro semanas?" é uma pergunta
  de pesquisa, e por isso o fator ×0,80 foi aplicado. A lista de nove sintomas do mesmo
  estudo serve de contraprova e fica dentro do intervalo corrigido.
- **O tamanho das faixas de idade e de tempo não foi lido.** Os percentuais por faixa
  são os publicados. Os intervalos da camada 2 são exercícios com tamanhos supostos.
- As médias "até 8 anos" e "de 9 em diante" tratam as faixas como se tivessem o mesmo
  tamanho, o que não foi conferido.
- A contagem de 51 pessoas é reconstruída do percentual publicado.
- O estudo é transversal. Ele compara pessoas diferentes com tempos diferentes de
  estadia, e não acompanha a mesma pessoa ao longo dos anos.
- A mesa de dez supõe pessoas independentes e parecidas com a amostra.
- A fonte de apoio (Hamid, 2022) mede saudade por escala e não publica proporção. Entra
  só para dizer que a direção, menos saudade com mais idade e mais tempo, se repete numa
  amostra maior e de outra região.

*A ciência é honesta sobre o que não sabe. O modelo também.*
