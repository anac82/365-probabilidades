### 📖 A História

A síndrome do impostor tem um problema que quase ninguém comenta: ninguém nunca
provou que ela é uma síndrome. E, mesmo assim, eu jurava ter uma.

Sabe a sensação? Você está numa sala que, no papel, é sua por mérito. Diploma,
cargos, anos de estrada. E lá dentro, uma voz baixinha e muito bem-educada
insiste: eles vão descobrir que você entrou aqui por engano.

Passei tempo demais tratando isso como um defeito de fábrica meu, algo a
consertar em silêncio. Até que a estatística em mim fez a pergunta que eu devia
ter feito antes: real como o quê, exatamente? Real como um sentimento que dá
para medir? Ou real como uma doença que se diagnostica?

Porque as duas coisas são muito diferentes. E a resposta muda tudo o que a
gente faz com esse desconforto.

---

### 📚 O Conceito: É real? Depende do que você chama de real

Primeira camada: como experiência, existe e se mede. Clance e Imes descreveram
o fenômeno em 1978, e a escala que criaram foi validada com boa confiabilidade
(Chrisman et al., 1995). Em psicologia, é isso que faz um construto existir:
você consegue medir de forma estável, e ele se correlaciona com coisas reais,
como ansiedade, depressão e burnout. Como vivência, o impostor é real e comum.

Segunda camada: como doença, não existe. Não está no DSM, não é diagnóstico
clínico. O nome correto na literatura é fenômeno do impostor, não síndrome.
"Síndrome" é apelido popular que dá ar de patologia e sugere que há algo
quebrado em você. Não há.

Terceira camada: parte não é você, é o ambiente. Tulshyan e Burey (Harvard
Business Review, 2021) argumentam que rotular isso como síndrome individualiza
um problema muitas vezes estrutural. Sentir-se impostora pode ser uma resposta
razoável a ser a única mulher, ou a única de fora, numa sala que nunca foi
desenhada para você.

---

### 🧮 O Modelo

Não modelo aqui "quanta" síndrome existe, porque o próprio número se recusa a
ser um número. Modelo por que ele escorrega: mostro como a mesma população
vira 82% ou 9% de "impostores" só mudando onde se corta a régua. Essa é a
prova estatística de que a parte "síndrome" é frágil.

**Fontes:**
- Bravata et al. (2019). *Prevalence, Predictors, and Treatment of Impostor
  Syndrome: a Systematic Review* · Journal of General Internal Medicine ·
  62 estudos, 14.161 participantes · prevalência de 9% a 82%
- Clance & Imes (1978), origem do conceito · Chrisman et al. (1995), validação
  da Clance IP Scale
- Tulshyan & Burey (2021), *Stop Telling Women They Have Imposter Syndrome* ·
  Harvard Business Review · a camada do ambiente

**Nota metodológica:** o fator ×0.80 (correção de autorrelato) é aplicado
apenas à prevalência, nunca à relação conceitual. Mesmo corrigida, a faixa
continua enorme, e é esse o ponto.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

DOURADO='#c8a84b'; VERMELHO='#c0392b'; VERDE='#1a5f5a'; CINZA='#6b6a64'
print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Bravata et al. (2019), Journal of General Internal Medicine — revisão sistemática
n_estudos       = 62
n_participantes = 14161
prev_min, prev_max = 0.09, 0.82   # prevalência varia com o instrumento e o corte

# Fator de correção cultural (autorrelato) — aplicado SÓ à prevalência
fator = 0.80
prev_min_corr = prev_min * fator
prev_max_corr = prev_max * fator

print("=" * 66)
print("  DADOS DA LITERATURA — O FENÔMENO DO IMPOSTOR")
print("=" * 66)
print(f"\n  Revisão sistemática (Bravata et al., 2019):")
print(f"  → Estudos:                 {n_estudos}")
print(f"  → Participantes:           {n_participantes:,}".replace(',', '.'))
print(f"  → Prevalência relatada:    {prev_min*100:.0f}% a {prev_max*100:.0f}%")
print(f"     (varia conforme o questionário e o ponto de corte)")
print(f"\n  Prevalência com fator ×0.80 (autorrelato):")
print(f"  → Faixa corrigida:         {prev_min_corr*100:.0f}% a {prev_max_corr*100:.0f}%")
print(f"     Mesmo corrigida, a faixa segue enorme.")
print("=" * 66)
```

```python
# --- O MODELO: por que a prevalência não é um número ---
# Ideia: escores de impostor numa população se distribuem mais ou menos como
# uma curva (aqui, Normal ilustrativa na faixa da Clance IP Scale, ~20 a 100).
# A "prevalência" é só a fração ACIMA de um ponto de corte escolhido.
# Prevalência(corte) = P(escore > corte) = 1 - Phi((corte - mu)/sigma)

mu, sigma = 62.0, 14.0   # média e desvio ilustrativos na escala da Clance IPS

# Onde teria que ficar o corte para gerar cada extremo relatado?
corte_82 = stats.norm.ppf(1 - prev_max, mu, sigma)   # corte que produz 82%
corte_09 = stats.norm.ppf(1 - prev_min, mu, sigma)   # corte que produz 9%

print("=" * 66)
print("  MODELO — A MESMA GENTE, DOIS CORTES, DUAS 'VERDADES'")
print("=" * 66)
print(f"\n  Prevalência = P(escore > corte)")
print(f"  → Para dar 82%, o corte cai em ~{corte_82:.0f} pontos.")
print(f"  → Para dar  9%, o corte sobe para ~{corte_09:.0f} pontos.")
print(f"  A mesma população. Só mudou a régua.")
print(f"\n  VEREDITO — 'a síndrome do impostor é real?'")
print(f"  → Como experiência (sente e se mede):   SIM")
print(f"  → Como diagnóstico clínico (DSM):        NÃO")
print(f"  → Como causa só sua:                     EM PARTE (o ambiente pesa)")
print("=" * 66)
```

```python
# --- VISUALIZAÇÃO — 3 GRÁFICOS SEPARADOS (fundo branco, dpi=150) ---

# ── GRÁFICO 1 — A ASSINATURA: a prevalência que escorrega ──
fig1, ax1 = plt.subplots(figsize=(10, 6))
cortes = np.linspace(40, 92, 400)
prev = (1 - stats.norm.cdf(cortes, mu, sigma)) * 100
ax1.plot(cortes, prev, color=VERDE, linewidth=3)
ax1.axhline(prev_max*100, color=CINZA, linestyle=':', linewidth=1.4)
ax1.axhline(prev_min*100, color=CINZA, linestyle=':', linewidth=1.4)
ax1.scatter([corte_82, corte_09], [prev_max*100, prev_min*100],
            color=DOURADO, s=80, zorder=5)
ax1.annotate(f'corte baixo\n{prev_max*100:.0f}% "têm"', xy=(corte_82, prev_max*100),
             xytext=(corte_82-1, prev_max*100-22), fontsize=10, color=VERMELHO, ha='center')
ax1.annotate(f'corte alto\n{prev_min*100:.0f}% "têm"', xy=(corte_09, prev_min*100),
             xytext=(corte_09+1, prev_min*100+16), fontsize=10, color=VERMELHO, ha='center')
ax1.set_xlabel('Onde você corta a régua (pontos na escala)')
ax1.set_ylabel('Prevalência de "impostor" (%)')
ax1.set_ylim(0, 100)
ax1.set_title('A Assinatura — A Prevalência Que Escorrega\n'
              r'prevalência(corte) = P(escore > corte)   ·   a mesma gente, de 9% a 82%',
              fontsize=13, pad=15)
plt.figtext(0.5, 0.005, 'Modelo ilustrativo sobre a faixa relatada por Bravata et al., 2019 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout(); plt.savefig('dia-055-grafico-01-corte.png', dpi=150, bbox_inches='tight'); plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 — A faixa, com e sem o fator ×0.80 ──
fig2, ax2 = plt.subplots(figsize=(10, 6))
ax2.barh(1, (prev_max-prev_min)*100, left=prev_min*100, height=0.32,
         color=CINZA, alpha=0.55)
ax2.barh(0, (prev_max_corr-prev_min_corr)*100, left=prev_min_corr*100, height=0.32,
         color=DOURADO, alpha=0.75)
ax2.set_yticks([0, 1])
ax2.set_yticklabels(['Com correção\n×0.80', 'Como relatado'])
ax2.set_xlim(0, 100)
ax2.set_xlabel('Prevalência de "síndrome do impostor" (%)')
for y, a, b in [(1, prev_min*100, prev_max*100), (0, prev_min_corr*100, prev_max_corr*100)]:
    ax2.text(a, y, f' {a:.0f}%', va='center', ha='right', fontsize=12, fontweight='bold')
    ax2.text(b, y, f'{b:.0f}% ', va='center', ha='left', fontsize=12, fontweight='bold')
ax2.set_title('Um Número Que Não Fecha\nDe 9% a 82%, e mesmo corrigido segue largo demais para ser "uma coisa"',
              fontsize=13, pad=15)
plt.figtext(0.5, 0.005, 'Faixa de prevalência: Bravata et al., 2019 · ×0.80 aplicado só ao autorrelato | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout(); plt.savefig('dia-055-grafico-02-correcao.png', dpi=150, bbox_inches='tight'); plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 — O VEREDITO em três camadas ──
fig3, ax3 = plt.subplots(figsize=(10, 6))
ax3.axis('off')
linhas = [
    ("Como experiência que se sente e se mede", "SIM", VERDE,
     "Clance IP Scale validada · liga a ansiedade, depressão, burnout"),
    ("Como diagnóstico clínico (DSM)", "NÃO", VERMELHO,
     "não é doença · é fenômeno, não síndrome"),
    ("Como causa que é só sua", "EM PARTE", DOURADO,
     "o ambiente pesa · ser a única na sala não é defeito seu"),
]
ax3.text(0.5, 0.93, 'A síndrome do impostor é real?', transform=ax3.transAxes,
         ha='center', fontsize=17, fontweight='bold', color='#222')
y = 0.72
for titulo, veredito, cor, nota in linhas:
    ax3.text(0.03, y, titulo, transform=ax3.transAxes, fontsize=13, color='#222', va='center')
    ax3.text(0.97, y, veredito, transform=ax3.transAxes, fontsize=15, color=cor,
             fontweight='bold', ha='right', va='center')
    ax3.text(0.03, y-0.075, nota, transform=ax3.transAxes, fontsize=10, color=CINZA, va='center')
    ax3.axhline(y-0.13, xmin=0.03, xmax=0.97, color='#e7e4dc', linewidth=1)
    y -= 0.235
plt.figtext(0.5, 0.02, 'Clance & Imes 1978 · Bravata et al. 2019 · Tulshyan & Burey 2021 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout(); plt.savefig('dia-055-grafico-03-veredito.png', dpi=150, bbox_inches='tight'); plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**A prevalência da "síndrome" do impostor vai de 9% a 82% dependendo de onde se
corta a régua. Um número que balança tanto não é a medida de uma doença, é o
retrato de um conceito frouxo.**
**O sentimento, esse sim, é real e se mede. O rótulo "síndrome" é que não se
sustenta.**

Existe como experiência: dá para medir, e vem acompanhado de ansiedade,
depressão e desgaste de verdade. Não existe como diagnóstico: não está no DSM,
não é doença. E boa parte dele nem é sobre você. Sentir-se fora do lugar quando
você é a única mulher, ou a única de fora, na sala não é um defeito de fábrica.
É uma leitura precisa do ambiente.

O que muda com isso é a decisão do dia. Se é doença sua, você tenta se
consertar, sozinha, em silêncio. Se é um sentimento comum diante de salas que
ainda não te veem como igual, você para de se corrigir e começa a olhar para a
sala. A primeira leitura te encolhe. A segunda te devolve o chão.

Não é negar o desconforto. É recusar o diagnóstico que ninguém te deu.

A pergunta que fica:

*E se o problema nunca tiver sido você, mas a sala?*

---

### ⚠️ Limitações do Modelo

- A curva do gráfico 1 é ilustrativa: serve para mostrar por que a prevalência depende do corte, não é a distribuição empírica exata dos escores.
- A faixa de 9% a 82% (Bravata et al., 2019) reúne estudos muito diferentes; não existe uma "prevalência verdadeira" única a extrair daí, e por isso não forcei uma Beta sobre um valor pontual.
- O ×0.80 é uma correção cultural de autorrelato aplicada só à prevalência; não muda a conclusão, apenas desloca a faixa.
- A camada do ambiente (Tulshyan & Burey, 2021) é um argumento conceitual bem fundamentado, não a medição de um efeito causal.
- "Não é diagnóstico clínico" é um fato sobre classificação (DSM), não um juízo sobre o sofrimento, que pode ser intenso e merece cuidado.

*A ciência é honesta sobre o que não sabe. O modelo também.*
