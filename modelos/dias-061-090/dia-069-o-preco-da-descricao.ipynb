### 📖 A História

Existe um momento, em toda entrevista e em toda avaliação de desempenho, em que
alguém pede para você falar do próprio trabalho.

Não é uma pergunta sobre fato. O fato já aconteceu. É uma pergunta sobre narrativa:
como você resume, em uma frase, uma coisa que só você viu inteira.

E todo mundo trata isso como detalhe. O que conta é o resultado, dizem. O número
está lá, a entrega está lá, a nota está lá.

Só que quem decide raramente vê o número. Vê a frase.

Dá para medir isso de um jeito bem direto: pegue pessoas que tiveram exatamente o
mesmo desempenho, peça que descrevam esse desempenho, e depois entregue **apenas a
descrição** a alguém que vai decidir com dinheiro real se contrata e por quanto.

---

### 📚 O Conceito: quando a régua é a frase

O desenho experimental aqui resolve o problema que arruína quase toda discussão
sobre esse assunto.

Fora do laboratório, é impossível separar desempenho de descrição: quem se descreve
melhor pode simplesmente ter feito melhor. Dentro do experimento, o desempenho é
**medido objetivamente e igualado na comparação**, e a única coisa que varia é a
frase.

Isso muda a pergunta. Não é mais "quem é melhor". É: **duas pessoas com a mesma nota
recebem a mesma oferta?**

E existe uma segunda camada, que é o que separa este dia do #063. Lá, o assunto era
o erro que a pessoa comete ao prever a si mesma. Aqui, o assunto é o que um terceiro
faz com essa descrição. Não é sobre o espelho. É sobre o preço.

---

### 🧮 O Modelo

Um experimento com desempenho objetivo medido, autoavaliação registrada e decisões
de contratação incentivadas com dinheiro real.

**Fontes:**
- Exley, C. L. & Kessler, J. B., 2022 · "The Gender Gap in Self-Promotion" ·
  *Quarterly Journal of Economics* 137(3), 1345-1381 · **mais de 4.000 participantes
  online** (3.892 recrutados no Mechanical Turk e no Prolific, em sete versões e cinco
  ondas de coleta) e **mais de 10.000 estudantes** de ensino fundamental e médio
- Autoavaliação de desempenho numa tarefa de matemática e ciências, escala de 0 a
  100 · o participante médio se descreve como **53** · o gap entre gêneros é de
  **13 pontos**, o que representa **24% da média**
- Robustez: gap estatisticamente significativo em **64 de 64 especificações**
- Persistência: o gap aparece **já na sexta série** e permanece mesmo quando todo
  incentivo de se promover é removido
- Consequência econômica: empregadores tomam **21 decisões de contratação** vendo
  **apenas** a autoavaliação · a disposição de contratar sobe **1 ponto percentual
  por ponto** na escala de 0 a 100 e **18 pontos percentuais por degrau** na escala
  Likert de seis pontos

**Referência interna:** o **#027** tratou de confiança em quem sabe pouco (Kruger &
Dunning) e o **#063** tratou do erro de prever a si mesmo (Epley & Dunning). Este dia
não é sobre estar errado sobre si. É sobre o que o mercado cobra por como você se
descreve.

**Nota metodológica sobre o fator ×0.80:** não se aplica. Experimento com desempenho
objetivo medido e decisões de contratação incentivadas com pagamento real.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Exley & Kessler, 2022, Quarterly Journal of Economics 137(3), 1345-1381

n_online       = 3_892       # MTurk e Prolific, 7 versoes, 5 ondas
n_estudantes   = 10_000      # "mais de 10.000" estudantes de escola
n_decisoes_emp = 21          # decisoes de contratacao por empregador

# Autoavaliacao numa escala de 0 a 100
autoavaliacao_media = 53
gap_pontos          = 13     # diferenca entre generos, mesma performance
gap_como_pct_media  = gap_pontos / autoavaliacao_media   # 24,5%, arredonda para 25
gap_pct_publicado   = 0.24    # o artigo publica 24% da media. Usar SEMPRE este.

# Robustez
espec_totais        = 64
espec_significativas = 64

# Resposta do empregador, que so ve a autoavaliacao
pp_por_ponto_escala = 1.0    # pontos percentuais de disposicao de contratar
pp_por_degrau_likert = 18.0  # escala Likert de 6 pontos

aplica_fator_080 = False

print("=" * 70)
print("  DADOS - A DESCRICAO E O PRECO DELA")
print("=" * 70)
print(f"\n  Exley & Kessler, 2022 (QJE 137(3), 1345-1381):")
print(f"  -> {n_online:,}".replace(",", ".") + " participantes online"
      f" e mais de {n_estudantes:,}".replace(",", ".") + " estudantes")
print(f"\n  Autoavaliacao (escala 0 a 100, mesma performance):")
print(f"  -> Media geral:                {autoavaliacao_media}")
print(f"  -> Gap entre generos:          {gap_pontos} pontos")
print(f"  -> Isso e {gap_pct_publicado*100:.0f}% da media (valor publicado)")
print(f"     conta direta: {gap_pontos}/{autoavaliacao_media}"
      f" = {gap_como_pct_media*100:.1f}%, arredondado pelos autores")
print(f"\n  Robustez:")
print(f"  -> Gap significativo em {espec_significativas} de {espec_totais}"
      f" especificacoes")
print(f"  -> Aparece ja na sexta serie")
print(f"  -> Persiste sem nenhum incentivo para se promover")
print(f"\n  O que o empregador faz com isso (ve SO a autoavaliacao):")
print(f"  -> +{pp_por_ponto_escala:.0f} ponto percentual de disposicao de contratar")
print(f"     para cada ponto na escala de 0 a 100")
print(f"  -> +{pp_por_degrau_likert:.0f} pontos percentuais por degrau na escala"
      f" Likert de 6 pontos")
print(f"\n  Fator x0.80 aplicado: {aplica_fator_080}")
print("=" * 70)
```

```python
# --- O MODELO ---
# A traducao central: quanto custa, em chance de contratacao, descrever o
# mesmo desempenho com 13 pontos a menos.

# 1) Custo direto, usando o coeficiente publicado do empregador
custo_pp = gap_pontos * pp_por_ponto_escala

# 2) Em quantas contratacoes isso aparece, nas 21 decisoes do experimento
efeito_em_21 = n_decisoes_emp * (custo_pp / 100)

# 3) Traducao para escala humana: a cada 100 processos seletivos
por_100 = custo_pp

# 4) O mesmo gap na escala Likert de 6 pontos.
#    Um degrau vale 18 pp. Os 13 pontos da escala de 100 equivalem a
#    quantos degraus? A escala Likert tem 6 niveis cobrindo 100 pontos,
#    entao cada degrau cobre cerca de 100/5 = 20 pontos.
pontos_por_degrau = 100 / (6 - 1)
degraus_equivalentes = gap_pontos / pontos_por_degrau
custo_likert_pp = degraus_equivalentes * pp_por_degrau_likert

# 5) Probabilidade de 64 de 64 especificacoes darem significativas
#    SE nao houvesse efeito nenhum. Nao e teste formal: as especificacoes sao
#    correlacionadas entre si. E um limite superior ilustrativo, declarado.
p_falso_positivo = 0.05
p_64_de_64_por_acaso = p_falso_positivo ** espec_totais

print("=" * 70)
print("  MODELO - O PRECO DA FRASE")
print("=" * 70)
print(f"\n  1) O custo direto, com desempenho IGUAL:")
print(f"  -> {gap_pontos} pontos a menos de autoavaliacao")
print(f"  -> x {pp_por_ponto_escala:.0f} ponto percentual por ponto")
print(f"  -> = {custo_pp:.0f} pontos percentuais a menos de chance de contratacao")
print(f"\n  2) Em escala humana:")
print(f"  -> A cada 100 processos com o mesmo desempenho,")
print(f"     cerca de {por_100:.0f} contratacoes a menos.")
print(f"  -> Nas {n_decisoes_emp} decisoes de cada empregador do experimento,")
print(f"     isso equivale a {efeito_em_21:.1f} contratacao a menos.")
print(f"\n  3) Pela escala Likert de 6 pontos (checagem de ordem de grandeza):")
print(f"  -> Cada degrau cobre cerca de {pontos_por_degrau:.0f} pontos da escala de 100")
print(f"  -> {gap_pontos} pontos = {degraus_equivalentes:.2f} degrau")
print(f"  -> x {pp_por_degrau_likert:.0f} pp por degrau"
      f" = {custo_likert_pp:.0f} pontos percentuais")
print(f"  -> As duas rotas dao a mesma ordem de grandeza. Nao e artefato de escala.")
print(f"\n  4) Sobre a robustez do achado:")
print(f"  -> {espec_significativas} de {espec_totais} especificacoes significativas.")
print(f"  -> Se nao houvesse efeito algum e as especificacoes fossem independentes,")
print(f"     isso teria probabilidade de {p_64_de_64_por_acaso:.1e}.")
print(f"     ELAS NAO SAO INDEPENDENTES, entao esse numero e apenas um limite")
print(f"     ilustrativo, nao um teste. Mas a ordem de grandeza diz o bastante.")
print(f"\n  O que o empregador ve nao e a performance. E a frase.")
print("=" * 70)
```

```python
# --- VISUALIZACAO ---

DOURADO = '#c8a84b'
VERMELHO = '#c0392b'
VERDE = '#2a8a82'
CINZA = '#6b6a64'

# GRAFICO 1 - Mesma performance, descricao diferente
fig1, (ax1a, ax1b) = plt.subplots(1, 2, figsize=(14, 7))

ax1a.bar(['Grupo A', 'Grupo B'], [100, 100], color=VERDE, alpha=0.75, width=0.5)
ax1a.set_ylim(0, 130)
ax1a.set_yticks([])
ax1a.set_title('O desempenho medido\n(igualado por construção)', fontsize=15, pad=14)
ax1a.text(0.5, 112, 'IGUAL', ha='center', fontsize=26, fontweight='bold', color=VERDE)
ax1a.grid(False)

alto = autoavaliacao_media + gap_pontos / 2
baixo = autoavaliacao_media - gap_pontos / 2
barras = ax1b.bar(['Grupo A', 'Grupo B'], [alto, baixo],
                  color=[DOURADO, VERMELHO], alpha=0.9, width=0.5)
for barra, valor in zip(barras, [alto, baixo]):
    ax1b.text(barra.get_x() + barra.get_width() / 2, valor + 1.5,
              f'{valor:.1f}'.replace('.', ','), ha='center', fontsize=20,
              fontweight='bold')
ax1b.annotate('', xy=(1, baixo), xytext=(0, alto),
              arrowprops=dict(arrowstyle='<->', color='#333', lw=2))
ax1b.text(0.5, autoavaliacao_media + 1, f'{gap_pontos} pontos', ha='center',
          fontsize=16, fontweight='bold', color='#333')
ax1b.set_ylim(0, 78)
ax1b.set_ylabel('Como descreveram o próprio desempenho (0 a 100)')
ax1b.set_title('A descrição do desempenho\n'
               f'Gap de {gap_pontos} pontos, ou {gap_pct_publicado*100:.0f}% da média',
               fontsize=15, pad=14)

fig1.suptitle('Mesma nota. Frases diferentes.\n'
              'Exley & Kessler, 2022, Quarterly Journal of Economics 137(3)',
              fontsize=16, y=1.04)
plt.figtext(0.5, 0.005,
            'Fonte: Exley & Kessler, 2022  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-069-grafico-01-mesma-nota.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 1 salvo")

# GRAFICO 2 - A regua do empregador, e o preco dos 13 pontos
fig2, ax2 = plt.subplots(figsize=(12, 8))

x = np.linspace(0, 100, 400)
base = 20                                   # ponto de partida ilustrativo da reta
y = base + x * pp_por_ponto_escala
ax2.plot(x, y, color=CINZA, linewidth=3)

y_alto = base + alto * pp_por_ponto_escala
y_baixo = base + baixo * pp_por_ponto_escala
ax2.scatter([alto], [y_alto], s=260, color=DOURADO, zorder=3)
ax2.scatter([baixo], [y_baixo], s=260, color=VERMELHO, zorder=3)

ax2.plot([baixo, baixo], [0, y_baixo], color=VERMELHO, linestyle=':', linewidth=1.6)
ax2.plot([alto, alto], [0, y_alto], color=DOURADO, linestyle=':', linewidth=1.6)
ax2.plot([0, alto], [y_alto, y_alto], color=DOURADO, linestyle=':', linewidth=1.6)
ax2.plot([0, baixo], [y_baixo, y_baixo], color=VERMELHO, linestyle=':', linewidth=1.6)

ax2.annotate('', xy=(2, y_baixo), xytext=(2, y_alto),
             arrowprops=dict(arrowstyle='<->', color='#333', lw=2.4))
ax2.text(4, (y_alto + y_baixo) / 2, f'{custo_pp:.0f} pontos percentuais\nde diferença',
         fontsize=15, fontweight='bold', color='#333', va='center')

ax2.set_xlim(0, 100)
ax2.set_ylim(0, 130)
ax2.set_xlabel('Autoavaliação entregue ao empregador (0 a 100)')
ax2.set_ylabel('Disposição de contratar (pontos percentuais)')
ax2.set_title('A régua do empregador: 1 ponto percentual por ponto de autoavaliação\n'
              'O empregador vê apenas a autoavaliação, nada mais',
              fontsize=15, pad=18)
ax2.text(0.98, 0.06,
         'A inclinação de 1 pp por ponto é o coeficiente publicado.\n'
         'O nível da reta é ilustrativo: o artigo publica a inclinação, não o intercepto.',
         transform=ax2.transAxes, ha='right', fontsize=10, color=CINZA, style='italic',
         bbox=dict(boxstyle='round,pad=0.5', facecolor='white', edgecolor=CINZA,
                   alpha=0.9))

plt.figtext(0.5, 0.005,
            'Fonte: Exley & Kessler, 2022, QJE 137(3)  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-069-grafico-02-preco.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 2 salvo")

# GRAFICO 3 - Assinatura: 64 de 64
fig3, ax3 = plt.subplots(figsize=(12, 8))

lado = 8
grade = np.ones((lado, lado))
ax3.imshow(grade, cmap='Greys', vmin=0, vmax=3)
for i in range(lado):
    for j in range(lado):
        ax3.add_patch(plt.Rectangle((j - 0.46, i - 0.46), 0.92, 0.92,
                                    facecolor=VERDE, alpha=0.85, edgecolor='white',
                                    linewidth=2))
ax3.set_xticks([]); ax3.set_yticks([])
ax3.grid(False)
for spine in ax3.spines.values():
    spine.set_visible(False)
ax3.set_title('A assinatura estatística: 64 de 64 especificações\n'
              'Cada quadrado é uma forma diferente de rodar a análise. '
              'Todas deram o mesmo sinal.',
              fontsize=15, pad=18)
ax3.text(3.5, 8.75,
         'Especificações definidas pela pergunta de autoavaliação, pela informação dada\n'
         'ao participante e pela versão do estudo. Elas são correlacionadas entre si,\n'
         'portanto isto não é um teste formal. É uma leitura de consistência.',
         ha='center', fontsize=11, color=CINZA, style='italic')

plt.figtext(0.5, 0.005,
            'Fonte: Exley & Kessler, 2022, QJE 137(3)  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-069-grafico-03-especificacoes.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 3 salvo")
```

### 💡 O Insight

Mesma tarefa. Mesma nota, medida por quem aplicou o teste.

Na hora de descrever esse desempenho numa escala de 0 a 100, aparece uma diferença
de **13 pontos** entre homens e mulheres. Como a média geral é 53, esses 13 pontos
são quase um quarto da escala usada na prática.

Sozinho, isso seria só um dado sobre autoimagem. O que faz este dia existir é o que
vem depois.

O empregador do experimento não vê a nota. Vê a frase. E responde a ela numa taxa
que dá para medir: **cada ponto a mais na autoavaliação vale 1 ponto percentual a
mais de disposição de contratar.**

Multiplique. **Treze pontos de descrição valem cerca de treze pontos percentuais de
chance de ser contratada.** Com desempenho idêntico.

A cada cem processos seletivos, isso é cerca de treze contratações que mudam de mão
sem que ninguém tenha feito nada melhor.

E o gap não é tática. Ele continua lá quando o incentivo de se promover é removido,
o que sugere que a pessoa não está negociando pior. Ela está descrevendo o que
acredita.

Dois detalhes que eu não consigo esquecer.

O primeiro: o gap aparece **já na sexta série**. Quando alguém chega na primeira
entrevista de emprego, isso já foi ensaiado por uma década.

O segundo: em **64 de 64** formas diferentes de rodar a análise, o resultado apareceu.
Esse tipo de consistência é raro em ciência do comportamento.

Uma coisa que este dia não diz: por que o gap existe. Os autores apontam para normas
e cultura, e a tarefa usada é de um tipo culturalmente associado ao masculino.
Existem leituras concorrentes sobre a origem, e nenhuma delas está resolvida por
este experimento.

O que está resolvido é o preço. Quem decide quase nunca vê o que você fez.

### ⚠️ Limitações do Modelo

- **Tarefa específica.** A tarefa é de matemática e ciências, de um tipo
  culturalmente associado ao masculino. O artigo indica que o padrão muda em outros
  tipos de tarefa, e este notebook não generaliza para qualquer contexto de trabalho.
- **Laboratório, não mercado de trabalho real.** Empregadores do experimento tomam 21
  decisões vendo apenas uma resposta de autoavaliação. Processos seletivos reais têm
  currículo, entrevista e referências, e o peso da autodescrição ali é desconhecido.
- **A multiplicação dos 13 pontos por 1 ponto percentual é minha, não do artigo.** O
  artigo publica a inclinação. A extrapolação linear ao longo dos 13 pontos é uma
  leitura de ordem de grandeza, e o notebook confere pela escala Likert como segunda
  rota. As duas batem, o que dá confiança, mas não substitui o número original.
- **O nível da reta do gráfico 2 é ilustrativo.** A inclinação é o coeficiente
  publicado; o intercepto não é.
- **As 64 especificações não são independentes.** A probabilidade calculada no
  notebook é limite ilustrativo, não teste formal.
- **O experimento não explica a origem do gap.** Os autores apontam para normas e
  cultura. Isto é interpretação dos autores, não resultado do experimento, e há
  leituras concorrentes.
- Participantes recrutados em mercados de trabalho online e estudantes americanos.
  Generalização para o Brasil não está testada.
- Fator ×0.80 não aplicado: desempenho objetivo medido e decisões incentivadas com
  pagamento real.
