### 📖 A História

Todo mundo acha que quer a verdade.

Mas repare no que você faz quando alguém discorda de você. Você escuta com curiosidade — ou já está montando a resposta antes da pessoa terminar a frase?

Passei quase vinte anos construindo modelos estatísticos para o mercado financeiro. O trabalho era, no fundo, dizer para pessoas poderosas coisas que elas não queriam ouvir. E aprendi cedo que o problema quase nunca era o número. Era o desconforto de quem recebia o número.

Nós não procuramos informação para descobrir o que é verdade. Procuramos informação para confirmar o que já sentimos.

E existe um nome científico para isso — com direito a tamanho de efeito medido.

---

### 📚 O Conceito: Exposição Seletiva

Em 1957, Leon Festinger descreveu a **dissonância cognitiva**: o desconforto de segurar, ao mesmo tempo, uma crença e uma evidência que a contradiz.

O cérebro trata esse desconforto como uma ameaça. E o resolve da forma mais barata possível: evitando a evidência.

É o que a psicologia chama de **exposição seletiva** — a tendência de buscar a informação que nos dá razão e desviar da que nos desafia.

Não é burrice. É economia emocional. Sentir-se certo é imediato. Estar certo dá trabalho.

---

### 🧮 O Modelo

Quão forte é essa tendência? A maior medição já feita responde.

Hart e colegas (2009) reuniram **91 estudos**, com quase **8.000 participantes**, para medir o tamanho exato da nossa preferência por informação que nos valida. O resultado foi um efeito **moderado** e consistente: d = 0,36.

Traduzindo para o dia a dia (via *common language effect size*): num confronto direto entre a informação que conforta e a que confronta, a maioria de nós escolhe o conforto por volta de **60% das vezes**.

Não é uma bolha total. É um dedo na balança. Mas um dedo que, repetido milhares de vezes ao longo da vida, decide em que mundo você acredita viver.

**Fontes:**
- Hart, W. et al. (2009) — *Feeling Validated Versus Being Correct: A Meta-Analysis of Selective Exposure to Information* — Psychological Bulletin, 135(4) — 91 estudos, N ≈ 7.930, d = 0,36
- Festinger, L. (1957) — *A Theory of Cognitive Dissonance* — Stanford University Press

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
from scipy.stats import norm

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Hart, W. et al. (2009) — Psychological Bulletin
# Meta-análise de exposição seletiva a informação

k_estudos        = 91      # estudos incluídos (67 relatos, 300 grupos independentes)
n_participantes  = 7930    # participantes (~8.000)
d_hart           = 0.36    # tamanho de efeito: preferência por info que confirma (congenial)

# Mecanismo teórico: dissonância cognitiva (Festinger, 1957) — referência, sem número

# NOTA METODOLÓGICA:
# O fator de correção x0.80 NÃO se aplica aqui.
# Ele vale apenas para proporções de survey populacional autorrelatado.
# d = 0,36 é tamanho de efeito de meta-análise — correção não se aplica a d.

print("=" * 65)
print("  DADOS DA LITERATURA — EXPOSIÇÃO SELETIVA")
print("=" * 65)
print(f"\n  Hart et al., 2009 (Psychological Bulletin):")
print(f"  → Estudos na meta-análise:         {k_estudos}")
print(f"  → Participantes:                   ~{n_participantes:,}")
print(f"  → Tamanho de efeito (d):           {d_hart}")
print(f"  → Interpretação:                    preferência MODERADA por info congenial")
print(f"\n  Festinger, 1957:")
print(f"  → Mecanismo:                       dissonância cognitiva")
print("=" * 65)
```

```python
# --- O MODELO ---
# De um tamanho de efeito (d) para uma probabilidade do dia a dia

# Passo 1 — Common Language Effect Size (CLES)
# Probabilidade de, num confronto direto, preferir a informação que valida:
#   P = Φ(d / √2)
p_conforto  = norm.cdf(d_hart / np.sqrt(2))   # sentir-se certo
p_confronto = 1 - p_conforto                  # estar certo

# Passo 2 — IC 95% publicado (efeito fixo), Hart et al. 2009
#   Valor exato do paper: d = 0,36, IC 95% [0,34, 0,39].
#   O desvio implícito vem do próprio IC publicado (não uma aproximação pelo N).
ic_d = (0.34, 0.39)
se_d = (ic_d[1] - ic_d[0]) / (2 * 1.96)

print("=" * 65)
print("  MODELO — SENTIR-SE CERTO x ESTAR CERTO")
print("=" * 65)
print(f"\n  Tamanho de efeito medido:          d = {d_hart}")
print(f"  IC 95% publicado (efeito fixo):    [{ic_d[0]:.2f}, {ic_d[1]:.2f}]")
print(f"\n  Traduzindo (CLES):")
print(f"  → P(escolher o conforto/validar):  {p_conforto*100:.1f}%")
print(f"  → P(escolher o confronto/verdade): {p_confronto*100:.1f}%")
print(f"\n  Leitura do dia:")
print(f"  → ~{p_conforto*100:.0f}% das vezes, preferimos nos SENTIR certos")
print(f"     a de fato ESTAR certos.")
print(f"  → Mas {p_confronto*100:.0f}% não é zero: a margem existe.")
print("=" * 65)
```

```python
# --- VISUALIZAÇÃO — GRÁFICOS SEPARADOS ---

COR_CONFORTO  = '#c0392b'   # vermelho — sentir-se certo
COR_CONFRONTO = '#2a8a82'   # verde-água — estar certo
COR_DOURADO   = '#c9a14a'

# ── GRÁFICO 1 — A escolha: conforto x confronto ──
fig1, ax1 = plt.subplots(figsize=(10, 6))
categorias = ['Sentir-se certo\n(buscar o que valida)', 'Estar certo\n(buscar o que confronta)']
valores = [p_conforto * 100, p_confronto * 100]
cores = [COR_CONFORTO, COR_CONFRONTO]

bars = ax1.bar(categorias, valores, color=cores, alpha=0.88, width=0.45)
ax1.set_ylim(0, 75)
ax1.set_ylabel('Probabilidade de escolha (%)')
ax1.set_title('Sentir-se certo x estar certo\nTradução de d = 0,36 (Hart et al., 2009) via common language effect size',
              fontsize=13, pad=15)
for bar, valor in zip(bars, valores):
    ax1.text(bar.get_x() + bar.get_width()/2, valor + 1.2,
             f'{valor:.0f}%', ha='center', fontweight='bold', fontsize=18)
plt.figtext(0.5, 0.01,
            'Fonte: Hart et al., 2009 — Psychological Bulletin (91 estudos, N≈7.930) | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-061-grafico-01-escolha.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 — Quão forte é o efeito? d na régua de Cohen ──
fig2, ax2 = plt.subplots(figsize=(10, 6))
benchmarks = {'pequeno\n(0,20)': 0.20, 'médio\n(0,50)': 0.50, 'grande\n(0,80)': 0.80}
ax2.hlines(1, 0, 0.95, color='#cccccc', linewidth=3, zorder=1)
for nome, val in benchmarks.items():
    ax2.plot(val, 1, 'o', color='#999999', markersize=10, zorder=2)
    ax2.text(val, 1.06, nome, ha='center', fontsize=10, color='#666666')
# Hart
ax2.plot(d_hart, 1, 'o', color=COR_CONFORTO, markersize=20, zorder=3)
ax2.errorbar(d_hart, 1, xerr=1.96*se_d, fmt='none', ecolor=COR_DOURADO, elinewidth=3, capsize=6, zorder=2)
ax2.text(d_hart, 0.90, f'Hart 2009\nd = {d_hart}', ha='center', fontsize=12,
         color=COR_CONFORTO, fontweight='bold')
ax2.set_xlim(0, 0.95)
ax2.set_ylim(0.8, 1.15)
ax2.set_yticks([])
ax2.set_xlabel('Tamanho de efeito (d de Cohen)')
ax2.set_title('Quão forte é a nossa preferência por se sentir certo?\nUm efeito moderado — real, mas longe de uma bolha total',
              fontsize=13, pad=15)
ax2.grid(axis='x', alpha=0.3)
ax2.spines['left'].set_visible(False)
plt.figtext(0.5, 0.01,
            'Barra dourada: IC 95% publicado (efeito fixo) | Fonte: Hart et al., 2009 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-061-grafico-02-forca-efeito.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 (ASSINATURA) — Distribuição do efeito com IC 95% ──
fig3, ax3 = plt.subplots(figsize=(10, 6))
x = np.linspace(d_hart - 4*se_d, d_hart + 4*se_d, 1000)
y = norm.pdf(x, loc=d_hart, scale=se_d)
ax3.plot(x, y, color=COR_CONFRONTO, linewidth=2.5)
ax3.fill_between(x, y, alpha=0.2, color=COR_CONFRONTO)
ax3.axvline(ic_d[0], color=COR_DOURADO, linestyle='--', alpha=0.9, linewidth=2)
ax3.axvline(ic_d[1], color=COR_DOURADO, linestyle='--', alpha=0.9, linewidth=2)
ax3.axvline(d_hart, color='#333333', linewidth=2)
ax3.text(d_hart, max(y)*1.02, f'd = {d_hart}', ha='center', fontsize=12, fontweight='bold', color='#333333')
ax3.text(ic_d[0], max(y)*0.5,
         f'IC 95%\n[{ic_d[0]:.2f}, {ic_d[1]:.2f}]',
         ha='right', fontsize=10, color=COR_DOURADO)
ax3.set_xlabel('Tamanho de efeito d (preferência por informação congenial)')
ax3.set_ylabel('Densidade')
ax3.set_title('Assinatura estatística — Distribuição do efeito\nHart et al., 2009 (91 estudos, N≈7.930)',
              fontsize=13, pad=15)
plt.figtext(0.5, 0.01,
            'IC 95% publicado (efeito fixo); um modelo de efeitos aleatórios seria um pouco mais largo | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-061-grafico-03-distribuicao-d.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**Cerca de 60% das vezes, escolhemos nos sentir certos em vez de estar certos.**

Parece pouco. Não é.

Porque essa escolha não acontece uma vez. Acontece toda vez que você abre uma notícia, decide quem seguir, julga qual estudo é "sério" e qual é "furado".

Sentir-se certo é gratuito e imediato. Estar certo custa desconforto — e às vezes custa admitir que você estava errado.

O dado não diz que somos irracionais. Diz que somos econômicos: o cérebro prefere a rota que dói menos.

A boa notícia é que 60% não é 100%. A margem existe. E ela se abre exatamente no instante em que você percebe o desconforto e, em vez de fugir dele, se aproxima.

*Da última vez que você mudou de ideia de verdade, foi porque procurou a informação que te contrariava — ou porque ela te encurralou?*

---

### ⚠️ Limitações do Modelo

- d = 0,36 é um efeito **médio**: varia com o contexto. Hart et al. encontraram a preferência mais forte quando a pessoa já tinha se comprometido com uma decisão ou crença.
- A tradução de d para "60%" usa o *common language effect size*, que assume distribuições normais. É uma aproximação didática, não uma proporção medida diretamente.
- O IC 95% mostrado é o publicado por Hart et al. (efeito fixo): [0,34, 0,39]. Um modelo de efeitos aleatórios, que acomoda a heterogeneidade entre os 91 estudos, tenderia a ser um pouco mais largo.
- Exposição seletiva é associação comportamental medida em contextos variados, não uma lei determinística.

*A ciência é honesta sobre o que não sabe. O modelo também.*
