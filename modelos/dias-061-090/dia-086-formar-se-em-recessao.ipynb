## Qual a probabilidade de o ano em que você se formou ainda estar no seu salário?

**Tipo:** Preditivo
**Data de publicação:** 2026-09-07
**Ferramenta:** Python
**Decisão analisada:** Quanto do que eu ganho hoje foi decidido por coisas que eu não escolhi?
**Hashtag:** #365Probabilidades #Dia086

---

### 📖 A História

Existe uma pergunta que aparece em toda avaliação de desempenho, em toda entrevista, em
toda conversa de fim de ano sobre carreira.

Por que fulano está ganhando mais.

E a resposta que se dá, quase sempre, é sobre a pessoa: trabalhou mais, negociou
melhor, escolheu certo, se dedicou.

Existe uma variável que quase ninguém coloca na conta, e ela não depende de esforço
nenhum. É a **data**. O ano em que a pessoa terminou os estudos e entrou no mercado.

Duas turmas formadas com um ano de diferença podem encontrar mercados completamente
diferentes. Uma sai num ano bom, outra sai numa recessão. Ninguém escolheu isso, ninguém
poderia ter previsto, e não é mérito nem culpa de ninguém.

A pergunta é quanto isso custa, e por quanto tempo.

---

### 📚 O Conceito: por que isso é quase um experimento

Esse tipo de pergunta costuma ser impossível de responder, porque quem entra no mercado
em ano ruim pode ser diferente de quem entra em ano bom.

Só que aqui existe uma vantagem rara. **O ano em que você se forma é, para efeitos
práticos, aleatório.** Ninguém escolhe nascer em 1985 em vez de 1988, e ninguém
consegue prever, quatro anos antes, se o mercado vai estar bom quando o diploma sair.

É o que os economistas chamam de experimento natural: uma variação que aconteceu por
motivos alheios às pessoas envolvidas.

O que se compara, então, são turmas parecidas que se formaram em anos diferentes, e se
acompanha o salário de cada uma ao longo de décadas, com dados administrativos e não
com autorrelato.

Isso permite responder duas coisas distintas, e a segunda importa mais que a primeira:
**quanto custa** e **quanto tempo dura**.

---

### 🧮 O Modelo

Dados administrativos de longo prazo, acompanhando coortes de formandos.

**Fontes:**
- Oreopoulos, P., von Wachter, T. & Heisz, A., 2012 · "The Short- and Long-Term Career
  Effects of Graduating in a Recession" · *American Economic Journal: Applied Economics*
  4(1), 1-29 · dados administrativos de formandos canadenses acompanhados ao longo da
  carreira
- **Magnitude:** uma recessão típica, definida como uma alta de **5 pontos percentuais**
  na taxa de desemprego, implica uma **perda inicial de cerca de 9% nos rendimentos
  anuais**
- **Duração:** a perda **cai pela metade em 5 anos** e **desaparece por volta de 10
  anos**, com os autores registrando que as perdas iniciais são significativas e
  eventualmente somem, mas só depois de **8 a 10 anos**
- **Heterogeneidade:** as perdas são **mais fortes para quem está entrando no mercado**
  e pequenas para quem já tem cerca de dois anos de experiência
- **Mecanismo proposto:** não é mercado neutro nem cicatriz permanente, e sim **busca
  demorada por empregadores melhores**, combinada com contratos salariais de longo
  prazo · há efeitos importantes sobre **mobilidade** e sobre as **características do
  empregador**, e efeitos pequenos sobre horas trabalhadas
- Kondo, A., 2015 · *IZA Journal of Labor Economics* 4(1) · documenta efeitos
  diferentes por gênero e por raça
- Evidência em outro país, publicada em 2022 no *Empirical Economics*: **um ponto
  percentual a mais de desemprego na formatura reduz o salário mensal inicial em
  8,77%**, com perda persistente, e aumenta em **7,58 pontos percentuais** a chance de
  trabalhar abaixo do nível de escolaridade e em **8,54** a chance de ter vínculo
  temporário · mecanismo apontado: incompatibilidade de qualificação e instabilidade do
  emprego

**Nota metodológica sobre o fator ×0.80:** não se aplica. Dados administrativos de
rendimento, não autorrelato de salário.


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Oreopoulos, von Wachter & Heisz, 2012, AEJ: Applied 4(1), 1-29

recessao_tipica_pp = 5          # alta de 5 p.p. na taxa de desemprego
perda_inicial = 0.09            # ~9% dos rendimentos anuais
anos_metade = 5                 # perda cai pela metade
anos_some = 10                  # desaparece por volta de 10 anos
janela_some = (8, 10)           # "8 a 10 anos", conforme os autores

# Heterogeneidade
EFEITOS = {
    'Quem esta entrando no mercado': 'perdas MAIS FORTES',
    'Quem ja tem ~2 anos de experiencia': 'perdas pequenas',
}

# Mecanismo proposto
MECANISMO = [
    'busca demorada por empregadores melhores',
    'contratos salariais de longo prazo',
    'efeitos sobre mobilidade e sobre o tipo de empregador',
    'efeitos pequenos sobre horas trabalhadas',
]

# Evidencia de outro pais (Empirical Economics, 2022)
por_pp_desemprego = 0.0877      # queda no salario mensal inicial por 1 p.p.
pp_subqualificacao = 7.58       # aumento na chance de trabalhar abaixo da escolaridade
pp_temporario = 8.54            # aumento na chance de vinculo temporario

aplica_fator_080 = False

print("=" * 72)
print("  DADOS - O CUSTO DE SE FORMAR NA HORA ERRADA")
print("=" * 72)
print(f"\n  Oreopoulos, von Wachter & Heisz, 2012 (AEJ: Applied 4(1)):")
print(f"  -> Recessao tipica: alta de {recessao_tipica_pp} p.p. no desemprego")
print(f"  -> Perda INICIAL nos rendimentos anuais: {perda_inicial*100:.0f}%")
print(f"  -> Cai pela metade em {anos_metade} anos")
print(f"  -> Desaparece por volta de {anos_some} anos"
      f" (os autores dizem {janela_some[0]} a {janela_some[1]})")
print(f"\n  Heterogeneidade:")
for grupo, efeito in EFEITOS.items():
    print(f"  -> {grupo:<38} {efeito}")
print(f"\n  Mecanismo proposto:")
for m in MECANISMO:
    print(f"  -> {m}")
print(f"\n  Evidencia de outro pais (Empirical Economics, 2022):")
print(f"  -> Cada 1 p.p. de desemprego na formatura:"
      f" -{por_pp_desemprego*100:.2f}% no salario inicial")
print(f"  -> +{pp_subqualificacao:.2f} p.p. de chance de trabalhar abaixo da escolaridade")
print(f"  -> +{pp_temporario:.2f} p.p. de chance de vinculo temporario")
print(f"\n  Fator x0.80 aplicado: {aplica_fator_080}")
print("=" * 72)
```

```python
# --- O MODELO ---
# Assinatura: a trajetoria da perda ao longo do tempo, e o custo ACUMULADO.
# A perda cai pela metade em 5 anos e some em 10. Uma queda exponencial
# reproduz esses dois pontos exatamente, e serve como interpolacao declarada.

# Meia-vida de 5 anos: perda(t) = perda_inicial * 0.5 ** (t / 5)
MEIA_VIDA = anos_metade


def perda(t):
    return perda_inicial * 0.5 ** (t / MEIA_VIDA)


anos = np.arange(0, 11)

# Custo acumulado, em "anos de salario perdidos"
acumulado = sum(perda(t) for t in anos)

# Traducao para dinheiro, com salario ilustrativo declarado
SALARIO_ANUAL = 100_000          # unidade arbitraria declarada
custo_total = acumulado * SALARIO_ANUAL

# Quanto do custo total acontece nos primeiros anos
custo_5_primeiros = sum(perda(t) for t in range(0, 5))
prop_5_primeiros = custo_5_primeiros / acumulado

print("=" * 72)
print("  MODELO - A TRAJETORIA DA PERDA")
print("=" * 72)
print(f"\n  PREMISSA DECLARADA: interpolacao exponencial com meia-vida de"
      f" {MEIA_VIDA} anos,")
print(f"  que reproduz exatamente os dois pontos publicados")
print(f"  (metade em {anos_metade} anos, praticamente zero em {anos_some}).")
print(f"\n  {'ano':>5} | {'perda no ano':>13}")
print(f"  {'-'*5}-+-{'-'*13}")
for t in anos:
    print(f"  {t:>5} | {perda(t)*100:>12.2f}%")
print(f"\n  1) Custo acumulado ao longo dos {anos_some} anos:")
print(f"  -> Soma das perdas anuais: {acumulado*100:.0f}% de um salario anual.")
print(f"  -> Ou seja, cerca de {acumulado:.2f} ano de salario perdido no total.")
print("  -> Com salario ilustrativo de " + f"{SALARIO_ANUAL:,}".replace(",", ".") +
      " por ano, isso da " + f"{custo_total:,.0f}".replace(",", ".") + ".")
print(f"     (unidade arbitraria declarada, so para dar dimensao)")
print(f"\n  2) Onde o custo se concentra:")
print(f"  -> {prop_5_primeiros*100:.0f}% do prejuizo acontece nos primeiros"
      f" {anos_metade} anos.")
print(f"  -> Justamente quando a pessoa esta comecando e tem menos margem.")
print(f"\n  3) A leitura que muda o tom:")
print(f"  -> A literatura popular fala em 'geracao perdida' e 'cicatriz permanente'.")
print(f"  -> O dado e mais gentil que isso: a perda DESAPARECE.")
print(f"  -> Mas custa {anos_some} anos, e {anos_some} anos e muito tempo.")
print(f"\n  4) E o mecanismo importa:")
print(f"  -> Nao e que a pessoa se torne pior profissional.")
print(f"  -> E que ela leva quase uma decada para chegar ao empregador")
print(f"     onde teria chegado direto, se tivesse se formado dois anos antes.")
print("=" * 72)
```

```python
# --- VISUALIZACAO ---

DOURADO = '#c8a84b'
VERMELHO = '#c0392b'
VERDE = '#2a8a82'
CINZA = '#6b6a64'

# GRAFICO 1 - Assinatura: a trajetoria da perda
fig1, ax1 = plt.subplots(figsize=(12, 8))

t = np.linspace(0, 12, 400)
ax1.plot(t, perda(t) * 100, color=VERMELHO, linewidth=3.6)
ax1.fill_between(t, perda(t) * 100, alpha=0.14, color=VERMELHO)
ax1.axhline(y=0, color='#333', linewidth=1.4)

for marco, cor in ((0, VERMELHO), (anos_metade, DOURADO), (anos_some, VERDE)):
    v = perda(marco) * 100
    ax1.scatter([marco], [v], s=190, color=cor, zorder=3)
    ax1.text(marco, v + 0.42, f'{v:.1f}%'.replace('.', ','), ha='center', fontsize=17,
             fontweight='bold', color=cor)

ax1.text(anos_metade, -0.72, 'metade em\n5 anos', ha='center', fontsize=12,
         color=DOURADO)
ax1.text(anos_some, -0.72, 'praticamente zero\nem 10 anos', ha='center', fontsize=12,
         color=VERDE)

ax1.set_xlim(-0.4, 12)
ax1.set_ylim(-1.6, 10.5)
ax1.set_xlabel('Anos desde a formatura')
ax1.set_ylabel('Perda nos rendimentos anuais (%)')
ax1.set_title('A assinatura estatística: a perda encolhe, e leva uma década\n'
              'Oreopoulos, von Wachter & Heisz, 2012 · recessão típica de 5 p.p. '
              'no desemprego',
              fontsize=15, pad=18)
ax1.text(0.5, -0.135,
         'Interpolação exponencial declarada, com meia-vida de 5 anos, que reproduz '
         'exatamente os dois pontos publicados.',
         transform=ax1.transAxes, ha='center', fontsize=11, color=CINZA, style='italic')

plt.figtext(0.5, 0.005,
            'Fonte: Oreopoulos, von Wachter & Heisz, 2012, AEJ: Applied 4(1), 1-29'
            '  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-086-grafico-01-trajetoria.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 1 salvo")

# GRAFICO 2 - O custo acumulado, ano a ano
fig2, ax2 = plt.subplots(figsize=(12, 8))

perdas_ano = [perda(a) * 100 for a in anos]
cores = [VERMELHO if a < anos_metade else CINZA for a in anos]
barras = ax2.bar([str(a) for a in anos], perdas_ano, color=cores, alpha=0.88, width=0.6)

for barra, v in zip(barras, perdas_ano):
    ax2.text(barra.get_x() + barra.get_width()/2, v + 0.15, f'{v:.1f}'.replace('.', ','),
             ha='center', fontsize=12, fontweight='bold')

ax2.set_ylim(0, 10.5)
ax2.set_xlabel('Anos desde a formatura')
ax2.set_ylabel('Perda no ano (%)')
ax2.set_title('Onde o prejuízo se concentra\n'
              f'{prop_5_primeiros*100:.0f}% do custo total acontece nos primeiros '
              f'{anos_metade} anos',
              fontsize=15, pad=18)
ax2.text(0.5, -0.13,
         f'Somando os {anos_some} anos, o total equivale a cerca de '
         f'{acumulado:.2f}'.replace('.', ',') +
         ' ano inteiro de salário. E é justamente no começo, quando há menos margem.',
         transform=ax2.transAxes, ha='center', fontsize=12, color=DOURADO,
         fontweight='bold')

plt.figtext(0.5, 0.005,
            'Cálculo próprio sobre a trajetória publicada  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-086-grafico-02-acumulado.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 2 salvo")

# GRAFICO 3 - O que muda alem do salario
fig3, ax3 = plt.subplots(figsize=(12, 7))

nomes = ['Salário mensal inicial\n(por 1 p.p. de desemprego)',
         'Chance de trabalhar abaixo\nda própria escolaridade',
         'Chance de ter vínculo\ntemporário']
vals = [-por_pp_desemprego * 100, pp_subqualificacao, pp_temporario]
cores3 = [VERMELHO, DOURADO, DOURADO]

y = np.arange(3)[::-1]
for yi, nome, v, cor in zip(y, nomes, vals, cores3):
    ax3.barh([yi], [v], color=cor, alpha=0.88, height=0.5)
    desloc = v + (0.35 if v > 0 else -0.35)
    ax3.text(desloc, yi, f'{v:+.2f}'.replace('.', ',') +
             ('%' if abs(v) > 8.7 and v < 0 else ' p.p.'),
             va='center', ha='left' if v > 0 else 'right', fontsize=16,
             fontweight='bold', color=cor)

ax3.axvline(x=0, color='#333', linewidth=1.6)
ax3.set_yticks(y)
ax3.set_yticklabels(nomes, fontsize=12)
ax3.set_xlim(-13, 13)
ax3.set_xticks([])
ax3.grid(False)
ax3.set_title('Não é só salário: o tipo de emprego também muda\n'
              'Evidência de outro país, publicada em 2022',
              fontsize=15, pad=18)
ax3.text(0.5, -0.13,
         'O mecanismo apontado é incompatibilidade de qualificação e instabilidade do '
         'emprego, não menor capacidade de quem se formou.',
         transform=ax3.transAxes, ha='center', fontsize=11, color=CINZA, style='italic')

plt.figtext(0.5, 0.005,
            'Fonte: estudo publicado em Empirical Economics, 2022'
            '  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-086-grafico-03-tipo-de-emprego.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 3 salvo")
```

### 💡 O Insight

Uma recessão típica, definida como uma alta de cinco pontos percentuais na taxa de
desemprego, custa a quem se forma naquele ano cerca de **9% dos rendimentos anuais**.

Não por ser pior profissional. Não por ter estudado menos. Por ter terminado o curso no
ano errado, o que ninguém escolhe e ninguém prevê.

Mas o número que importa não é o 9%. É o **tempo**.

A perda **cai pela metade em cinco anos** e **desaparece por volta dos dez**.

Somando ano a ano, o prejuízo acumulado equivale a cerca de **0,54 ano inteiro de
salário**. E **64% dele acontece nos primeiros cinco anos**, justamente quando a pessoa
está começando, tem menos reserva e menos margem para negociar qualquer coisa.

O mecanismo é a parte mais interessante, e ele não é o que a gente imagina.

Não é que a pessoa fique marcada, nem que perca capacidade. É que ela leva quase uma
década para **chegar ao empregador onde teria chegado direto** se tivesse se formado
dois anos antes. A busca por um lugar melhor é lenta, os contratos salariais são de
longo prazo, e cada degrau leva tempo.

E os efeitos não são só de salário. Num estudo em outro país, cada ponto percentual a
mais de desemprego na formatura aumenta em **7,58 pontos** a chance de a pessoa
trabalhar **abaixo do próprio nível de escolaridade**, e em **8,54 pontos** a chance de
ter **vínculo temporário**.

Ou seja, o custo aparece na forma do emprego, não só no valor dele.

Agora, a leitura que muda o tom deste dia.

A conversa pública sobre isso costuma falar em geração perdida e cicatriz permanente.
**O dado é mais gentil que isso: a perda desaparece.**

Só que ela custa dez anos. E dez anos, na casa dos vinte para os trinta, não é um
detalhe. É o período em que se compra casa, se decide ter filho, se acumula a reserva
que permite arriscar depois.

O que eu levo daqui é uma coisa sobre atribuição.

A gente explica salário quase inteiramente por mérito, esforço e escolha. E existe uma
variável ali, com efeito de quase um ano inteiro de renda, que é pura data de
nascimento.

Isso vale nos dois sentidos, e é essa a parte que ninguém diz. Quem se formou num ano
bom carrega, no salário de hoje, uma vantagem que não conquistou e que provavelmente
nem sabe que tem.

*Que parte do que você tem hoje foi decidida por coisas que você não escolheu?*

### ⚠️ Limitações do Modelo

- **A curva de decaimento é interpolação minha, declarada.** O artigo publica dois
  pontos: a perda cai pela metade em cinco anos e desaparece por volta de dez. Usei uma
  exponencial com meia-vida de cinco anos, que reproduz exatamente esses dois pontos.
  Os valores intermediários não são publicados.
- **O custo acumulado e a tradução em "ano de salário" são cálculos próprios**, feitos
  sobre essa interpolação. Servem para dar dimensão, não são números do artigo.
- **O salário de referência do notebook é uma unidade arbitrária declarada**, e não uma
  estimativa de renda de ninguém.
- **Dados canadenses**, de coortes específicas. Mercados de trabalho com regras de
  contratação, rotatividade e proteção diferentes podem apresentar padrões distintos, e
  o estudo de outro país citado aqui já mostra magnitudes diferentes.
- **A média esconde variação grande.** Kondo, em 2015, documenta efeitos diferentes por
  gênero e por raça, e as perdas são muito mais fortes para quem está entrando no
  mercado do que para quem já tem dois anos de experiência.
- **O desfecho é rendimento, não bem-estar.** Nada aqui mede satisfação com a carreira,
  saúde ou o custo emocional de começar em ano ruim.
- **"Desaparece" significa em média.** Uma média que volta ao normal pode conter
  trajetórias muito diferentes, incluindo pessoas que nunca recuperam.
- Fator ×0.80 não aplicado: dados administrativos de rendimento.

*A ciência é honesta sobre o que não sabe. O modelo também.*
