# 365 Probabilidades · Dia #084
## Qual a probabilidade de o seu gosto musical ter sido decidido aos 14 anos?

**Tipo:** Descritivo
**Data de publicação:** 2026-09-05
**Ferramenta:** Python
**Decisão analisada:** Eu escolhi as músicas que amo, ou elas me pegaram numa idade específica?
**Hashtag:** #365Probabilidades #Dia084

---

### 📖 A História

Existe um fenômeno que qualquer pessoa acima dos trinta reconhece na hora.

Toca uma música antiga em algum lugar, num restaurante, num rádio de táxi, numa lista
automática, e alguma coisa acontece. Não é só reconhecer. É voltar. A pessoa lembra
exatamente onde estava, com quem, o que estava usando, o que estava sentindo.

E é curioso que isso quase nunca acontece com músicas dos últimos cinco anos, mesmo com
as que você gosta.

A explicação de sempre é que a música de antigamente era melhor. Toda geração diz isso,
sobre gerações diferentes, o que já é suspeito.

Existe uma explicação alternativa: **não é a música. É a idade em que ela te pegou.**

E essa é uma daquelas raras crenças que vêm com número exato.

---

### 📚 O Conceito: o pico de reminiscência

Existe um achado antigo e muito replicado em psicologia da memória: quando se pede a
uma pessoa mais velha que conte lembranças da própria vida, elas **não se distribuem
igualmente** ao longo dos anos.

Aparece um excesso de memórias vindas da adolescência e do início da vida adulta.
Chama-se **pico de reminiscência**.

As explicações propostas envolvem a formação da identidade, a densidade de eventos
novos naquele período, e o fato de que muita coisa acontece pela primeira vez ali.

A pergunta deste dia é outra: **a música tem um pico próprio, ou ela apenas ilumina o
mesmo pico?**

E isso é testável, porque dá para comparar o pico das memórias evocadas por música com
o pico das memórias evocadas por palavras comuns.

---

### 🧮 O Modelo

Um estudo transversal com adultos de várias idades, usando músicas populares como
pistas de memória.

**Fontes:**
- Jakubowski, K., Eerola, T., Tillmann, B., Perrin, F. & Heine, L., 2020 · "A
  Cross-Sectional Study of Reminiscence Bumps for Music-Related Memories in Adulthood" ·
  *Music & Science* · **N=470 participantes, de 18 a 82 anos** · viram título e artista
  de **111 canções populares** que estiveram nas paradas entre **1950 e 2015** e
  avaliaram, para cada uma, o quanto ela evocava memórias autobiográficas, o quanto
  conheciam e o quanto gostavam
- **Achado:** pico de memórias na adolescência, com máximo por volta dos **14 anos**,
  tanto para **saliência autobiográfica** quanto para **familiaridade**
- **Comparação:** o pico das memórias evocadas por música fica em **14,0 anos**, e o
  das memórias evocadas por palavras em **14,7 anos**
- Krumhansl, C. L. & Zupnick, J. A., 2013 · **pico em cascata** · além do pico da
  própria adolescência, aparece um segundo pico na época em que os **pais** do
  participante eram adolescentes
- Trabalho recente comparando tipos de pista encontra que pistas **musicais** produzem
  um pico significativamente **mais pronunciado** que pistas de filmes

**Nota metodológica sobre o fator ×0.80:** não se aplica. Avaliação de estímulos
padronizados, com 111 canções idênticas apresentadas a todos os participantes, e não
proporção de survey sobre comportamento.


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Jakubowski, Eerola, Tillmann, Perrin & Heine, 2020, Music & Science

n_participantes = 470
faixa_idade = (18, 82)
n_cancoes = 111
periodo_paradas = (1950, 2015)

# Picos de reminiscencia, em anos de idade
pico_musica = 14.0        # saliencia autobiografica e familiaridade
pico_palavra = 14.7       # memorias evocadas por palavras

# O que foi avaliado para cada cancao
MEDIDAS = ['saliencia autobiografica', 'familiaridade', 'quanto gosta']

# Krumhansl & Zupnick, 2013 - pico em cascata
cascata = True

aplica_fator_080 = False

print("=" * 70)
print("  DADOS - A IDADE EM QUE A MUSICA GRUDA")
print("=" * 70)
print(f"\n  Jakubowski et al., 2020 (Music & Science):")
print(f"  -> N = {n_participantes} participantes, de {faixa_idade[0]} a"
      f" {faixa_idade[1]} anos")
print(f"  -> {n_cancoes} cancoes populares das paradas de"
      f" {periodo_paradas[0]} a {periodo_paradas[1]}")
print(f"  -> Medidas por cancao: {', '.join(MEDIDAS)}")
print(f"\n  Picos de reminiscencia:")
print(f"  -> Memorias evocadas por MUSICA:  {pico_musica:.1f} anos")
print(f"  -> Memorias evocadas por PALAVRA: {pico_palavra:.1f} anos")
print(f"  -> Diferenca: {abs(pico_palavra - pico_musica):.1f} ano")
print(f"\n  Krumhansl & Zupnick, 2013:")
print(f"  -> Pico em cascata: {cascata}")
print(f"     Alem do pico da propria adolescencia, aparece um segundo pico")
print(f"     na epoca em que os PAIS do participante eram adolescentes.")
print(f"\n  Fator x0.80 aplicado: {aplica_fator_080}")
print("=" * 70)
```

```python
# --- O MODELO ---
# A assinatura deste dia e a ARITMETICA DA JANELA, aplicada a cada idade.
# Se o pico esta aos 14, quanto da vida de alguem acontece DEPOIS dele?

def proporcao_depois(idade_atual, pico=pico_musica, inicio=0):
    if idade_atual <= pico:
        return 0.0
    return (idade_atual - pico) / (idade_atual - inicio)


idades = [20, 30, 40, 50, 60, 70, 82]

# A distancia entre os dois picos, em proporcao
dif_picos = abs(pico_palavra - pico_musica)
dif_relativa = dif_picos / pico_palavra

print("=" * 70)
print("  MODELO - O QUANTO DA VIDA VEM DEPOIS DO PICO")
print("=" * 70)
print(f"\n  1) Musica e palavra apontam para o MESMO lugar:")
print(f"  -> Pico por musica:  {pico_musica:.1f} anos")
print(f"  -> Pico por palavra: {pico_palavra:.1f} anos")
print(f"  -> Distancia: {dif_picos:.1f} ano, ou {dif_relativa*100:.0f}% do valor.")
print(f"  -> A musica NAO cria um pico proprio. Ela ilumina o mesmo pico")
print(f"     com mais forca.")
print(f"\n  2) Quanto da vida acontece depois dos {pico_musica:.0f} anos:")
print(f"  {'idade':>8} | {'anos vividos depois do pico':>28} | {'% da vida':>10}")
print(f"  {'-'*8}-+-{'-'*28}-+-{'-'*10}")
for i in idades:
    print(f"  {i:>8} | {i - pico_musica:>28.0f} | "
          f"{proporcao_depois(i)*100:>9.0f}%")
print(f"\n  3) A leitura:")
print(f"  -> Aos 50 anos, {proporcao_depois(50)*100:.0f}% da sua vida aconteceu")
print(f"     DEPOIS da idade em que a musica mais grudou.")
print(f"  -> E ainda assim e aquela musica que para tudo quando toca.")
print(f"\n  4) O detalhe que quase ninguem conta:")
print(f"  -> Alem do proprio pico, aparece um SEGUNDO pico na epoca em que")
print(f"     os pais do participante eram adolescentes.")
print(f"  -> Ou seja: parte do seu gosto musical foi decidido antes de")
print(f"     voce nascer, pela musica que tocava na casa.")
print("=" * 70)
```

```python
# --- VISUALIZACAO ---

DOURADO = '#c8a84b'
VERMELHO = '#c0392b'
VERDE = '#2a8a82'
CINZA = '#6b6a64'

# GRAFICO 1 - O pico de reminiscencia, em forma esquematica
fig1, ax1 = plt.subplots(figsize=(12, 8))

idade = np.linspace(0, 60, 800)
# Forma esquematica: excesso de memorias em torno do pico
curva = np.exp(-((idade - pico_musica) ** 2) / (2 * 7.5 ** 2))
curva = curva + 0.12 * np.exp(-((idade - 3) ** 2) / (2 * 3.0 ** 2)) * 0
ax1.plot(idade, curva * 100, color=DOURADO, linewidth=3.6)
ax1.fill_between(idade, curva * 100, alpha=0.16, color=DOURADO)

ax1.axvline(x=pico_musica, color=VERMELHO, linestyle='--', linewidth=2.4)
ax1.text(pico_musica + 1.0, 92, f'{pico_musica:.1f} anos'.replace('.', ','),
         fontsize=20, fontweight='bold', color=VERMELHO)
ax1.text(pico_musica + 1.0, 84, 'pico das memórias\nevocadas por música',
         fontsize=12, color=VERMELHO)

ax1.set_xlim(0, 60)
ax1.set_ylim(0, 108)
ax1.set_yticks([])
ax1.set_xlabel('Idade em que a música era popular (anos de vida do participante)')
ax1.set_ylabel('Quantidade de memórias evocadas (escala relativa)')
ax1.set_title('O pico de reminiscência musical\n'
              'Jakubowski et al., 2020 · N=' + f'{n_participantes}' +
              f' · {n_cancoes} canções das paradas de {periodo_paradas[0]} a '
              f'{periodo_paradas[1]}',
              fontsize=15, pad=18)
ax1.text(0.5, -0.13,
         'Forma esquemática do pico descrito no estudo, com escala relativa. '
         'Não é a curva ajustada publicada.',
         transform=ax1.transAxes, ha='center', fontsize=11, color=CINZA, style='italic')

plt.figtext(0.5, 0.005,
            'Fonte: Jakubowski, Eerola, Tillmann, Perrin & Heine, 2020, Music & Science'
            '  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-084-grafico-01-pico.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 1 salvo")

# GRAFICO 2 - Musica e palavra apontam para o mesmo lugar
fig2, ax2 = plt.subplots(figsize=(12, 6.5))

ax2.scatter([pico_musica], [1], s=460, color=VERMELHO, zorder=3)
ax2.scatter([pico_palavra], [0], s=460, color=CINZA, zorder=3)

ax2.text(pico_musica, 1.34, f'{pico_musica:.1f}'.replace('.', ','), ha='center',
         fontsize=26, fontweight='bold', color=VERMELHO)
ax2.text(pico_musica, 1.14, 'pistas MUSICAIS', ha='center', fontsize=13,
         color=VERMELHO)
ax2.text(pico_palavra, -0.36, f'{pico_palavra:.1f}'.replace('.', ','), ha='center',
         fontsize=26, fontweight='bold', color=CINZA)
ax2.text(pico_palavra, -0.16, 'pistas de PALAVRA', ha='center', fontsize=13,
         color=CINZA)

ax2.annotate('', xy=(pico_palavra, 0.5), xytext=(pico_musica, 0.5),
             arrowprops=dict(arrowstyle='<->', color=DOURADO, lw=2.6))
ax2.text((pico_musica + pico_palavra) / 2, 0.62,
         f'{dif_picos:.1f} ano de distância'.replace('.', ','), ha='center',
         fontsize=14, fontweight='bold', color=DOURADO)

ax2.set_xlim(12, 17)
ax2.set_ylim(-0.75, 1.75)
ax2.set_yticks([])
ax2.set_xlabel('Idade do pico de reminiscência (anos)')
ax2.set_title('A música não cria um pico próprio. Ela ilumina o mesmo pico.\n'
              'Diferença de menos de um ano entre os dois tipos de pista',
              fontsize=15, pad=18)

plt.figtext(0.5, 0.005,
            'Fonte: Jakubowski et al., 2020  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-084-grafico-02-musica-e-palavra.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 2 salvo")

# GRAFICO 3 - Assinatura: quanto da vida vem depois do pico
fig3, ax3 = plt.subplots(figsize=(12, 8))

y = np.arange(len(idades))[::-1]
for yi, i in enumerate(idades):
    pos = len(idades) - 1 - yi
    prop = proporcao_depois(i) * 100
    ax3.barh([pos], [pico_musica], color=DOURADO, alpha=0.9, height=0.55)
    ax3.barh([pos], [i - pico_musica], left=pico_musica, color=CINZA, alpha=0.35,
             height=0.55)
    ax3.text(i + 1.2, pos, f'{prop:.0f}% da vida depois do pico', va='center',
             fontsize=13, color='#333', fontweight='bold')

ax3.set_yticks(range(len(idades)))
ax3.set_yticklabels([f'{i} anos' for i in idades][::-1], fontsize=13)
ax3.set_xlim(0, 108)
ax3.set_xlabel('Anos de vida')
ax3.grid(False)
ax3.set_title('A assinatura estatística: quanto da vida vem depois dos 14 anos\n'
              'A parte dourada é a janela em que a música grudou',
              fontsize=15, pad=18)
ax3.text(0.5, -0.1,
         'Aos 50 anos, 72% da sua vida aconteceu depois da idade em que a música mais '
         'grudou. E ainda assim é aquela música que para tudo.',
         transform=ax3.transAxes, ha='center', fontsize=12, color=DOURADO,
         fontweight='bold')

plt.figtext(0.5, 0.005,
            'Cálculo próprio sobre o pico publicado  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-084-grafico-03-depois-do-pico.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 3 salvo")
```

### 💡 O Insight

Quatrocentos e setenta pessoas, de 18 a 82 anos. Cento e onze canções que estiveram nas
paradas entre 1950 e 2015, as mesmas para todo mundo. E uma pergunta simples para cada
uma: essa música traz alguma lembrança sua?

**O pico está aos 14,0 anos.**

Não aos vinte, quando a pessoa tinha liberdade e dinheiro para escolher o que ouvir.
Não aos trinta, quando o gosto já estava formado. Aos catorze.

E o número que fecha o argumento é a comparação. Quando o mesmo tipo de estudo é feito
com **palavras** comuns como pista, em vez de música, o pico aparece aos **14,7 anos**.

Menos de um ano de distância.

Ou seja: **a música não cria um pico próprio. Ela ilumina o mesmo pico com muito mais
força.** O que acontece aos catorze não é sobre música. É sobre a idade. A música só é
o melhor holofote que existe para aquele período.

Fiz a conta do que isso implica.

Aos 50 anos, **72% da sua vida aconteceu depois** da idade em que a música mais grudou.
Aos 70, são 80%. E ainda assim é aquela música, de quatro décadas atrás, que faz alguém
parar no meio do supermercado.

E tem um detalhe que eu não esperava encontrar.

Além do pico da própria adolescência, aparece um **segundo pico**, na época em que os
**pais** do participante eram adolescentes. Chama-se pico em cascata.

Ou seja: parte do seu gosto musical foi decidida **antes de você nascer**, pela música
que tocava na casa em que você cresceu.

O que me fica não é sobre música.

É que a gente costuma tratar gosto como identidade, como se fosse uma escolha que
revela quem a pessoa é. "Eu sou de rock", "eu sou de MPB", como se tivesse havido uma
avaliação.

E a maior parte disso foi decidida numa janela estreita, numa idade em que ninguém
escolhe muita coisa, com o que estava tocando por perto.

Isso não torna o gosto menos verdadeiro. As músicas continuam sendo suas, e o que elas
provocam continua sendo real.

Só significa que elas te escolheram tanto quanto você as escolheu.

*Qual música te leva de volta na hora, e quantos anos você tinha?*

### ⚠️ Limitações do Modelo

- **A curva do gráfico 1 é esquemática**, com escala relativa, e representa a forma do
  pico descrito no estudo. Não é a curva ajustada publicada.
- **Estudo transversal**, comparando pessoas de idades diferentes num mesmo momento.
  Parte do que parece efeito de idade pode ser diferença entre gerações, ou diferença
  entre as próprias canções de cada época.
- **Canções populares das paradas**, o que privilegia música de grande circulação. O
  padrão pode ser diferente para quem construiu o gosto fora do mainstream.
- **A comparação com pistas de palavra vem de outra literatura**, com desenhos e
  amostras diferentes. A proximidade dos dois picos é notável, mas não é comparação
  direta dentro do mesmo estudo.
- **A conta de quanto da vida vem depois do pico é minha**, derivada do pico publicado.
  É aritmética simples e serve para dar dimensão, não é um achado do artigo.
- **Amostra de um único estudo**, com 470 pessoas. O fenômeno do pico de reminiscência
  é muito replicado; a idade exata do pico musical, menos.
- **Saliência autobiográfica não é o mesmo que preferência.** O estudo mede quanta
  memória a música evoca, e a conclusão sobre "gosto musical" é uma extensão razoável,
  mas é extensão.
- Fator ×0.80 não aplicado.

*A ciência é honesta sobre o que não sabe. O modelo também.*
