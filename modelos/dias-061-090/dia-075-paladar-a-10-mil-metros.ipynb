### 📖 A História

A bandeja chega, alguém levanta a tampa de papel-alumínio, e vem sempre a mesma
avaliação: comida de avião é ruim.

É provavelmente a opinião mais consensual da humanidade. Ninguém defende a bandeja.
Companhias aéreas contratam chefs famosos, gastam fortunas em desenvolvimento de
cardápio, e a fama não muda.

O que quase ninguém considera é a hipótese mais simples: e se o problema não estiver
na comida?

A 10 mil metros, várias coisas mudam ao mesmo tempo. A pressão cai, o ar fica
extremamente seco, e existe um barulho constante de fundo que ninguém nota depois de
dez minutos, mas que está lá o voo inteiro, em torno de 85 decibéis.

Alguém resolveu isolar só o barulho.

---

### 📚 O Conceito: como se testa uma coisa dessas

O desenho é engenhoso justamente por não envolver avião nenhum.

As pessoas provaram soluções puras dos cinco gostos básicos, em três concentrações
cada, e avaliaram a intensidade numa escala que ia de "mal detectável" a "o mais forte
imaginável".

Cada participante fez isso **duas vezes**: uma sessão com ruído simulado de cabine de
avião e outra com ruído ambiente normal de sala. Comparação dentro da mesma pessoa,
o que elimina diferença entre indivíduos.

E o ponto crucial: **a comida era idêntica nas duas sessões.** Solução padronizada em
laboratório. Se a avaliação mudar, a mudança não pode estar no alimento.

Falta ainda descartar a explicação preguiçosa, que seria "com barulho a pessoa fica
distraída e avalia tudo pior". Para isso servem os controles, e eles são o melhor
pedaço deste estudo.

---

### 🧮 O Modelo

Experimento de psicofísica com comparação intraindividual e bateria de controles.

**Fontes:**
- Yan, K. S. & Dando, R., 2015 · "A Crossmodal Role for Audition in Taste Perception" ·
  *Journal of Experimental Psychology: Human Perception and Performance* 41(3), 590 ·
  cinco gostos básicos, três concentrações cada, duas sessões por participante
- **Condição de ruído:** espectro amplo simulando cabine de avião, pico em torno de
  **290 Hz**, a **80 a 85 decibéis**, tocado por 30 minutos antes do teste e mantido
  durante
- **Resultado:** sem influência sobre **salgado, amargo e azedo** · intensidade
  **suprimida para o doce em todas as concentrações** · intensidade **aumentada para
  o umami nas concentrações mais altas**
- **Controles negativos:** sem diferença entre as condições em avaliação tátil, em
  saturação de cor, em intensidade sonora de um tom, e **sem diferença em teste de
  tempo de reação simples**, o que descarta diferença de foco ou atenção
- **Mecanismo proposto:** doce e umami **compartilham o mesmo receptor gustativo**, ao
  contrário dos outros três. Os autores especulam que mutações genéticas prevalentes
  nesse receptor modulem a magnitude do efeito, e a inspeção dos dados individuais
  sugeriu dois grupos distintos: quem responde fortemente a ambos os gostos e quem é
  praticamente insensível a ambos.
- Woods, A. T. et al., 2011 · achado convergente anterior: ruído alto reduz a
  sensibilidade a doce e a salgado

**Pendência declarada:** o tamanho da amostra não foi localizado nas fontes acessíveis
consultadas. O desenho, os controles e a direção dos efeitos estão publicados; o N
não. Este notebook não afirma nenhum tamanho de amostra.

**Nota metodológica sobre o fator ×0.80:** não se aplica. Psicofísica com solução
padronizada e comparação da mesma pessoa entre duas condições.


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Yan & Dando, 2015, J Exp Psychol Hum Percept Perform 41(3), 590

db_cabine     = (80, 85)     # decibeis da condicao de ruido
pico_hz       = 290          # pico do espectro
minutos_antes = 30           # exposicao ao ruido antes do teste
concentracoes = 3            # niveis testados por gosto

# Efeito do ruido sobre cada gosto basico
# 'suprimido', 'aumentado' ou 'sem efeito'
GOSTOS = {
    'Doce':    ('suprimido', 'em todas as concentracoes'),
    'Umami':   ('aumentado', 'nas concentracoes mais altas'),
    'Salgado': ('sem efeito', ''),
    'Amargo':  ('sem efeito', ''),
    'Azedo':   ('sem efeito', ''),
}

# Controles negativos: nenhum apresentou diferenca entre as condicoes
CONTROLES = [
    'Avaliação tátil',
    'Saturação de cor',
    'Intensidade de um tom',
    'Tempo de reação simples',
]

# Receptor: doce e umami compartilham o mesmo. Os outros tres, nao.
COMPARTILHAM_RECEPTOR = ['Doce', 'Umami']

n_amostra = None   # NAO LOCALIZADO nas fontes acessiveis. Nao inventar.

aplica_fator_080 = False

print("=" * 70)
print("  DADOS - RUIDO DE CABINE E OS CINCO GOSTOS BASICOS")
print("=" * 70)
print(f"\n  Yan & Dando, 2015 (JEP:HPP 41(3), 590):")
print(f"  -> Ruido de cabine simulado: {db_cabine[0]} a {db_cabine[1]} dB,"
      f" pico em ~{pico_hz} Hz")
print(f"  -> {minutos_antes} min de exposicao antes do teste, mantido durante")
print(f"  -> {concentracoes} concentracoes por gosto, duas sessoes por pessoa")
print(f"  -> Tamanho da amostra: NAO LOCALIZADO nas fontes acessiveis")
print(f"\n  Efeito do ruido sobre cada gosto:")
for gosto, (efeito, detalhe) in GOSTOS.items():
    receptor = " (mesmo receptor)" if gosto in COMPARTILHAM_RECEPTOR else ""
    extra = f" {detalhe}" if detalhe else ""
    print(f"  -> {gosto:<9} {efeito:<11}{extra}{receptor}")
print(f"\n  Controles negativos, todos SEM diferenca entre as condicoes:")
for c in CONTROLES:
    print(f"  -> {c}")
print(f"\n  Fator x0.80 aplicado: {aplica_fator_080}")
print("=" * 70)
```

```python
# --- O MODELO ---
# A assinatura deste dia nao e um tamanho de efeito. E a ESPECIFICIDADE.
# Um efeito geral de distracao afetaria tudo. Este afeta 2 de 9 medidas.

medidas_totais = len(GOSTOS) + len(CONTROLES)
afetadas = [g for g, (e, _) in GOSTOS.items() if e != 'sem efeito']
nao_afetadas = medidas_totais - len(afetadas)

# As duas medidas afetadas sao exatamente as duas que compartilham receptor.
acerto_previsao = set(afetadas) == set(COMPARTILHAM_RECEPTOR)

# Se o ruido afetasse medidas ao acaso, qual a chance de acertar
# exatamente as duas que compartilham receptor, entre as 5 do paladar?
# Combinacao de 5 tomados 2 a 2 = 10 pares possiveis.
from math import comb
pares_possiveis = comb(len(GOSTOS), 2)
p_por_acaso = 1 / pares_possiveis

print("=" * 70)
print("  MODELO - A FORCA DESTE ACHADO ESTA NA ESPECIFICIDADE")
print("=" * 70)
print(f"\n  1) O que mudou e o que nao mudou:")
print(f"  -> Medidas avaliadas no total: {medidas_totais}"
      f" ({len(GOSTOS)} gostos + {len(CONTROLES)} controles)")
print(f"  -> Afetadas pelo ruido: {len(afetadas)} ({', '.join(afetadas)})")
print(f"  -> Nao afetadas: {nao_afetadas}")
print(f"\n  2) Por que isso importa:")
print(f"  -> A explicacao preguicosa seria distracao: com barulho, avalia-se")
print(f"     tudo pior. Mas salgado, amargo e azedo ficaram intactos,")
print(f"     e o tempo de reacao simples nao mudou.")
print(f"  -> E o umami AUMENTOU. Distracao nao aumenta nada.")
print(f"\n  3) A previsao mecanistica bateu:")
print(f"  -> Doce e umami compartilham o MESMO receptor gustativo.")
print(f"  -> Salgado, amargo e azedo usam vias distintas.")
print(f"  -> Os dois gostos afetados sao exatamente os dois que")
print(f"     compartilham receptor: {acerto_previsao}")
print(f"  -> Se dois gostos entre {len(GOSTOS)} fossem afetados ao acaso,")
print(f"     existiriam {pares_possiveis} pares possiveis, e a chance de sair")
print(f"     justamente esse par seria de 1 em {pares_possiveis}"
      f" ({p_por_acaso*100:.0f}%).")
print(f"     Isso nao e teste estatistico. E leitura de plausibilidade.")
print(f"\n  4) E a variacao entre pessoas:")
print(f"  -> A inspecao dos dados individuais sugeriu dois grupos:")
print(f"     quem responde forte aos dois gostos e quem e quase insensivel.")
print(f"  -> Isso explica por que a comida de aviao divide opinioes")
print(f"     mais do que qualquer outra comida do mundo.")
print("=" * 70)
```

```python
# --- VISUALIZACAO ---

DOURADO = '#c8a84b'
VERMELHO = '#c0392b'
VERDE = '#2a8a82'
CINZA = '#6b6a64'

# GRAFICO 1 - O que o ruido mexeu e o que nao mexeu
fig1, ax1 = plt.subplots(figsize=(12, 8))

itens = list(GOSTOS.keys()) + CONTROLES
tipo = ['gosto'] * len(GOSTOS) + ['controle'] * len(CONTROLES)
efeito = [GOSTOS[g][0] if g in GOSTOS else 'sem efeito' for g in itens]

y = np.arange(len(itens))[::-1]
for yi, item, t, e in zip(y, itens, tipo, efeito):
    if e == 'suprimido':
        cor, largura, texto = VERMELHO, -1.0, 'suprimido'
    elif e == 'aumentado':
        cor, largura, texto = VERDE, 1.0, 'aumentado'
    else:
        cor, largura, texto = CINZA, 0.0, 'sem efeito'
    if largura == 0:
        ax1.plot([0], [yi], marker='o', markersize=13, color=cor)
    else:
        ax1.barh([yi], [largura], color=cor, alpha=0.85, height=0.5)
    desloc = largura + (0.08 if largura >= 0 else -0.08)
    ax1.text(desloc, yi, texto, va='center',
             ha='left' if largura >= 0 else 'right',
             fontsize=13, color=cor, fontweight='bold')

ax1.axvline(x=0, color='#333', linewidth=1.6)
ax1.axhline(y=len(CONTROLES) - 0.5, color=CINZA, linestyle=':', linewidth=1.6)
ax1.text(-1.42, len(CONTROLES) + 1.6, 'GOSTOS', fontsize=12, color=CINZA,
         fontweight='bold')
ax1.text(-1.42, len(CONTROLES) - 1.1, 'CONTROLES', fontsize=12, color=CINZA,
         fontweight='bold')

ax1.set_yticks(y)
ax1.set_yticklabels(itens, fontsize=13)
ax1.set_xlim(-1.5, 1.6)
ax1.set_xticks([])
ax1.grid(False)
ax1.set_title('Nove medidas. Duas mudaram.\n'
              'Yan & Dando, 2015 · ruído de cabine simulado a 80-85 dB',
              fontsize=15, pad=18)
ax1.text(0.5, -0.09,
         'Se o efeito fosse distração, tudo pioraria. Salgado, amargo e azedo ficaram '
         'intactos,\ne o tempo de reação simples não mudou.',
         transform=ax1.transAxes, ha='center', fontsize=11, color=CINZA, style='italic')

plt.figtext(0.5, 0.005,
            'Fonte: Yan & Dando, 2015, J Exp Psychol Hum Percept Perform 41(3), 590'
            '  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-075-grafico-01-o-que-mudou.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 1 salvo")

# GRAFICO 2 - Assinatura: a previsao pelo receptor
fig2, ax2 = plt.subplots(figsize=(12, 8))

ax2.axis('off')
ax2.set_xlim(0, 10)
ax2.set_ylim(0, 10)

ax2.text(5, 9.3, 'A previsão que o mecanismo permite fazer',
         ha='center', fontsize=17, fontweight='bold')

caixa_a = dict(boxstyle='round,pad=0.6', facecolor='#f6f1e2', edgecolor=DOURADO,
               linewidth=2)
caixa_b = dict(boxstyle='round,pad=0.6', facecolor='white', edgecolor=CINZA,
               linewidth=1.6)

ax2.text(3.0, 7.6, 'DOCE\ne\nUMAMI', ha='center', va='center', fontsize=15,
         fontweight='bold', bbox=caixa_a)
ax2.text(3.0, 5.6, 'compartilham o\nMESMO receptor', ha='center', va='center',
         fontsize=13, color=DOURADO)
ax2.text(3.0, 3.8, 'os dois\nmudaram', ha='center', va='center', fontsize=16,
         fontweight='bold', color=VERMELHO,
         bbox=dict(boxstyle='round,pad=0.7', facecolor='#fdf3f1',
                   edgecolor=VERMELHO, linewidth=2))

ax2.text(7.0, 7.6, 'SALGADO\nAMARGO\nAZEDO', ha='center', va='center', fontsize=15,
         fontweight='bold', bbox=caixa_b)
ax2.text(7.0, 5.6, 'usam vias\ndistintas', ha='center', va='center', fontsize=13,
         color=CINZA)
ax2.text(7.0, 3.8, 'nenhum\nmudou', ha='center', va='center', fontsize=16,
         fontweight='bold', color=VERDE,
         bbox=dict(boxstyle='round,pad=0.7', facecolor='#f0f7f6',
                   edgecolor=VERDE, linewidth=2))

for x in (3.0, 7.0):
    ax2.annotate('', xy=(x, 4.6), xytext=(x, 5.1),
                 arrowprops=dict(arrowstyle='->', color=CINZA, lw=2))
    ax2.annotate('', xy=(x, 6.3), xytext=(x, 6.9),
                 arrowprops=dict(arrowstyle='->', color=CINZA, lw=2))

ax2.text(5, 2.2,
         f'Entre os cinco gostos existem {pares_possiveis} pares possíveis.\n'
         f'O par afetado foi exatamente o que compartilha receptor.',
         ha='center', fontsize=14, color='#333', fontweight='bold')
ax2.text(5, 0.8,
         'Leitura de plausibilidade, não teste estatístico. '
         'O mecanismo previa quais gostos mudariam, e acertou.',
         ha='center', fontsize=11, color=CINZA, style='italic')

plt.figtext(0.5, 0.005,
            'Fonte: Yan & Dando, 2015  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-075-grafico-02-receptor.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 2 salvo")

# GRAFICO 3 - Os dois grupos de pessoas
fig3, ax3 = plt.subplots(figsize=(12, 7))

grupos = ['Respondem forte\nao ruído', 'Praticamente\ninsensíveis']
ax3.barh([1, 0], [1, 1], color=[VERMELHO, VERDE], alpha=0.55, height=0.45)
ax3.set_yticks([1, 0])
ax3.set_yticklabels(grupos, fontsize=14)
ax3.set_xticks([])
ax3.set_xlim(0, 1.9)
ax3.grid(False)

ax3.text(0.5, 1, 'para essas pessoas, o doce some\ne o umami salta',
         ha='center', va='center', fontsize=14, color='white', fontweight='bold')
ax3.text(0.5, 0, 'para essas, a bandeja tem\no mesmo gosto de sempre',
         ha='center', va='center', fontsize=14, color='white', fontweight='bold')

ax3.set_title('Dois grupos distintos apareceram nos dados individuais\n'
              'É por isso que comida de avião divide opiniões mais que qualquer '
              'outra comida',
              fontsize=15, pad=18)
ax3.text(0.5, -0.16,
         'Os autores não publicam a proporção de cada grupo, e este notebook não a '
         'estima.\nO receptor compartilhado por doce e umami tem mutações genéticas '
         'prevalentes.',
         transform=ax3.transAxes, ha='center', fontsize=11, color=CINZA, style='italic')

plt.figtext(0.5, 0.005,
            'Fonte: Yan & Dando, 2015  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-075-grafico-03-dois-grupos.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 3 salvo")
```

### 💡 O Insight

Cinco gostos básicos, três concentrações cada, duas sessões por pessoa. A mesma
solução, feita em laboratório, nas duas vezes. A única coisa que mudou foi o barulho
no fundo.

**O doce foi suprimido em todas as concentrações. O umami aumentou nas concentrações
mais altas. Salgado, amargo e azedo não se mexeram.**

A explicação fácil seria distração: com barulho, a pessoa presta menos atenção e
avalia tudo pior. Os autores fecharam essa porta com quatro controles. Não houve
diferença em avaliação tátil, em saturação de cor, em intensidade de um tom, nem em
tempo de reação simples.

E tem um detalhe que mata de vez a hipótese da distração: **o umami aumentou.**
Distração não aumenta nada.

Agora a parte que eu acho elegante.

Doce e umami **compartilham o mesmo receptor gustativo**. Salgado, amargo e azedo
usam vias distintas. Entre cinco gostos existem dez pares possíveis, e o par que o
ruído mexeu foi exatamente aquele que divide receptor.

Não é prova de mecanismo, é plausibilidade. Mas é o tipo de coincidência que raramente
é coincidência.

E aparece ainda um último achado, quase escondido no artigo: **os dados individuais
sugeriram dois grupos distintos.** Gente que responde fortemente ao ruído nos dois
gostos, e gente praticamente insensível aos dois. O receptor compartilhado tem mutações
genéticas prevalentes.

Se você já discutiu com alguém sobre comida de avião e as duas versões pareciam
descrever refeições diferentes, é possível que estivessem descrevendo mesmo.

O que me fica não é sobre bandeja.

É que a gente atribui a coisas objetivas o que na verdade depende de onde a coisa está
sendo experimentada. A comida do avião não é pior. Ela é comida numa sala com 85
decibéis, e uma parte do seu paladar está desligada ali dentro.

E se um ruído de fundo que ninguém nota consegue apagar um gosto inteiro, vale
perguntar o que mais o ambiente está mudando sem pedir licença.

### ⚠️ Limitações do Modelo

- **O tamanho da amostra não foi localizado** nas fontes acessíveis consultadas. O
  desenho, os controles e a direção dos efeitos estão publicados; o N não. Este
  notebook não afirma nenhum tamanho de amostra, e a magnitude dos efeitos deve ser
  lida com essa reserva.
- **Ruído simulado, não voo real.** Num avião mudam também a pressão da cabine e a
  umidade do ar, que é extremamente baixa e afeta o olfato, que por sua vez responde
  por boa parte do que a gente chama de sabor. Este estudo isola apenas o som.
- **Soluções puras, não comida.** Doce, salgado, azedo, amargo e umami foram testados
  isolados. Um prato é a combinação dos cinco mais aroma, textura e temperatura.
- **A conta dos dez pares é leitura de plausibilidade, não teste estatístico.** Ela
  serve para mostrar que a previsão mecanística tinha como falhar e não falhou.
- **A proporção de cada grupo de respondedores não é publicada**, e este notebook não
  a estima. A existência dos dois grupos vem de inspeção dos dados individuais, o que é
  uma observação exploratória.
- O achado convergente de Woods e colegas, em 2011, encontrou redução de doce **e de
  salgado**, o que não bate exatamente com este resultado. A literatura não está
  fechada.
- Fator ×0.80 não aplicado: psicofísica com solução padronizada e comparação
  intraindividual.

*A ciência é honesta sobre o que não sabe. O modelo também.*
