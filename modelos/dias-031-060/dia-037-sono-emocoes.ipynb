### 📖 A História

Tem um tipo de dia ruim que começa na véspera.

Você dormiu mal. Não necessariamente pouco — às vezes foram 6 horas, às vezes foram 8 horas fragmentadas. Mas quando acorda, alguma coisa está diferente. O trânsito incomoda mais. A reunião que seria neutra parece carregada. Uma mensagem ambígua parece agressiva. Você reage de um jeito que, horas depois, não reconhece como seu.

Por muito tempo assumi que isso era fraqueza. Que pessoas mais disciplinadas ou mais resilientes não deixavam o cansaço afetar suas reações.

Não é fraqueza. É neurologia. E tem um número para isso.

---

### 📚 O Conceito: A Desconexão Pré-frontal-Amígdala

A amígdala é a região do cérebro responsável por detectar e responder a estímulos emocionalmente relevantes — especialmente ameaças. Em condições normais, o córtex pré-frontal medial (mPFC) exerce um controle inibitório sobre a amígdala: ele avalia o contexto, calibra a resposta, e evita que o sistema de alarme dispare de forma desproporcional.

Quando você dorme, esse sistema é restaurado. Quando você não dorme, a conexão entre o pré-frontal e a amígdala enfraquece — e o sistema de alarme fica sem freio.

Yoo e colegas da Harvard Medical School e UC Berkeley testaram exatamente isso usando ressonância magnética funcional (fMRI). Expuseram participantes privados de sono a imagens progressivamente mais negativas — e mediram a resposta da amígdala.

O resultado: a amígdala dos participantes sem sono foi **60% mais reativa** do que a do grupo que dormiu normalmente. E o volume de amígdala ativado foi **3 vezes maior**.

### 🧮 O Modelo
**Fontes:**
- Yoo, S.S., Gujar, N., Hu, P., Jolesz, F.A. & Walker, M.P. (2007) — *The human emotional brain without sleep — a prefrontal amygdala disconnect* — Current Biology, 17(20), R877-R878 — N=26 (13 privados de sono, 13 controle)
- Walker, M.P. (2017) — *Why We Sleep: Unlocking the Power of Sleep and Dreams* — Scribner (síntese das implicações dos dados de fMRI)


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print('✅ Bibliotecas carregadas')
```

```python
# --- DADOS DA LITERATURA ---
# Yoo, Gujar, Hu, Jolesz & Walker (2007) — Current Biology
# doi: 10.1016/j.cub.2007.08.007
# N=26 (13 privados de sono, 13 controle com sono normal)
# Estatística: t(24)=3.2, p=0.004 para reatividade da amígdala

n_total = 26
n_por_grupo = 13

# Aumento de reatividade da amígdala — grupo sem sono vs grupo controle
aumento_reatividade_amigdala_pct = 0.60   # +60% de magnitude de ativação

# Aumento de volume de amígdala ativado
aumento_volume_amigdala = 3.0   # 3x mais volume ativado (t(24)=2.8, p=0.009)

# Estatísticas do paper
t_reatividade = 3.2    # t(24)=3.2, p=0.004
p_valor_reatividade = 0.004
t_volume = 2.8         # t(24)=2.8, p=0.009
p_valor_volume = 0.009

fator_correcao = 0.80

print('=' * 65)
print('  DADOS DA LITERATURA — SONO E REATIVIDADE EMOCIONAL')
print('=' * 65)
print(f'\n  Yoo et al. (2007) — Current Biology — N={n_total} ({n_por_grupo} por grupo):')
print(f'  → Aumento de reatividade da amígdala (sem sono vs controle):')
print(f'    +{aumento_reatividade_amigdala_pct*100:.0f}% de magnitude de ativação')
print(f'    t({n_total-2})={t_reatividade}, p={p_valor_reatividade}')
print(f'  → Aumento de volume de amígdala ativado:')
print(f'    {aumento_volume_amigdala:.0f}x mais volume ativado')
print(f'    t({n_total-2})={t_volume}, p={p_valor_volume}')
print(f'\n  Mecanismo:')
print(f'  → Sem sono: conexão amígdala-mPFC enfraquece')
print(f'  → mPFC perde capacidade de inibir a amígdala')
print(f'  → Cérebro reverte a padrão mais primitivo de processamento emocional')
print(f'\n  Fator de correção: x{fator_correcao}')
print('=' * 65)
print()
print('  ⚠️  NOTA METODOLÓGICA:')
print('  Estudo de fMRI com N=26 (típico de neuroimagem). Os participantes')
print('  foram privados de sono por uma noite inteira (privação total),')
print('  não parcial — o efeito de privação parcial pode ser menor em')
print('  magnitude, mas estudos subsequentes confirmaram a direção do efeito.')
print('  O grupo controle dormiu normalmente em laboratório na mesma noite.')
```

```python
# --- O MODELO ---

# Efeito de Cohen d estimado a partir do t e N
# d = t / sqrt(N/2) para dois grupos iguais
d_reatividade = t_reatividade / np.sqrt(n_por_grupo / 2)
d_volume = t_volume / np.sqrt(n_por_grupo / 2)

# Aumento de reatividade com fator de correção
aumento_corrigido_pct = aumento_reatividade_amigdala_pct * fator_correcao

# P(reagir de forma desproporcional a estímulo negativo) após noite sem sono
# Estimativa baseada no d de Cohen calculado
p_reacao_desproporcional = stats.norm.cdf(d_reatividade)
p_reacao_desproporcional_corrigido = stats.norm.cdf(d_reatividade * fator_correcao)

print('=' * 65)
print('  MODELO — RESULTADO')
print('=' * 65)
print(f'\n  Tamanho de efeito:')
print(f'  → d (reatividade da amígdala): {d_reatividade:.2f} (efeito grande)')
print(f'  → d (volume ativado):          {d_volume:.2f} (efeito grande)')
print(f'\n  Aumento de reatividade emocional após privação de sono:')
print(f'  → Original: +{aumento_reatividade_amigdala_pct*100:.0f}%')
print(f'  → Corrigido (×{fator_correcao}): +{aumento_corrigido_pct*100:.0f}%')
print(f'\n  P(reagir de forma desproporcional a estímulo negativo):')
print(f'  → Estimativa: {p_reacao_desproporcional*100:.0f}%')
print(f'  → Corrigida: {p_reacao_desproporcional_corrigido*100:.0f}%')
print(f'\n  INTERPRETAÇÃO:')
print(f'  → Sem sono, o cérebro trata estímulos neutros como ameaças')
print(f'  → O córtex pré-frontal perde capacidade de contextualizar')
print(f'  → Reações emocionais tornam-se menos calibradas e mais primitivas')
print('=' * 65)
```

```python
# --- VISUALIZACAO ---

# GRAFICO 1 — Reatividade da amígdala: com vs sem sono
fig1, ax1 = plt.subplots(figsize=(12, 8))

# Representacao do aumento relativo de reatividade
categorias = ['Grupo com sono normal', 'Grupo sem sono\n(privação total)']
valores = [1.0, 1.0 + aumento_reatividade_amigdala_pct]   # relativo ao controle
cores1 = ['#2a8a82', '#c0392b']

bars = ax1.bar(categorias, valores, color=cores1, alpha=0.85, width=0.45)
ax1.set_ylim(0, 2.0)
ax1.set_ylabel('Reatividade relativa da amígdala (controle = 1.0)', fontsize=12)
ax1.set_title('Uma noite sem sono aumenta a reatividade emocional em 60%\nYoo et al. (2007) — Current Biology — N=26 · t(24)=3.2, p=0.004',
              fontsize=13, pad=15)

ax1.text(bars[0].get_x() + bars[0].get_width()/2, valores[0] + 0.03,
         '1.0 (referência)', ha='center', fontweight='bold', fontsize=14)
ax1.text(bars[1].get_x() + bars[1].get_width()/2, valores[1] + 0.03,
         f'+{aumento_reatividade_amigdala_pct*100:.0f}%', ha='center', fontweight='bold', fontsize=18, color='#c0392b')

ax1.axhline(y=1.0, color='#333333', linestyle='--', linewidth=1.5, label='Linha de base (sono normal)')
ax1.legend(fontsize=11)

plt.tight_layout()
plt.savefig('dia-026-grafico-01-reatividade-amigdala.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Gráfico 1 salvo!')


# GRAFICO 2 — O mecanismo: a desconexão pré-frontal-amígdala (ilustrativo)
fig2, ax2 = plt.subplots(1, 2, figsize=(14, 8))

# Com sono: pré-frontal inibe amígdala (seta forte)
# Sem sono: pré-frontal fraco, amígdala descontrolada
for ax, titulo, cor_seta, forca_seta, cor_amigdala, alpha_pfrontal in [
    (ax2[0], 'Com sono normal', '#2a8a82', 3.0, '#2a8a82', 0.9),
    (ax2[1], 'Sem sono', '#c0392b', 0.5, '#c0392b', 0.3)
]:
    # Prefrontal
    pfrontal = plt.Circle((0.5, 0.75), 0.18, color='#c9a14a', alpha=alpha_pfrontal)
    ax.add_patch(pfrontal)
    ax.text(0.5, 0.75, 'Córtex\nPré-frontal', ha='center', va='center',
            fontsize=11, fontweight='bold', color='#0d0d0d')

    # Amígdala
    amigdala = plt.Circle((0.5, 0.25), 0.18, color=cor_amigdala, alpha=0.7)
    ax.add_patch(amigdala)
    ax.text(0.5, 0.25, 'Amígdala', ha='center', va='center',
            fontsize=12, fontweight='bold', color='white')

    # Seta de controle (pré-frontal → amígdala)
    ax.annotate('', xy=(0.5, 0.44), xytext=(0.5, 0.56),
                arrowprops=dict(arrowstyle='->', color=cor_seta, lw=forca_seta))

    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.axis('off')
    ax.set_title(titulo, fontsize=14, fontweight='bold')

ax2[0].text(0.5, 0.505, 'controle\ninibidor', ha='center', fontsize=10, color='#2a8a82')
ax2[1].text(0.5, 0.505, 'controle\nenfraquecido', ha='center', fontsize=10, color='#c0392b')

fig2.suptitle('O mecanismo: sem sono, o freio emocional enfraquece\nYoo et al. (2007) — ilustração do padrão descrito no paper',
              fontsize=13, y=0.02)
plt.tight_layout()
plt.savefig('dia-026-grafico-02-mecanismo.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Gráfico 2 salvo!')


# GRAFICO 3 — Distribuição do efeito: d de Cohen calculado do t
fig3, ax3 = plt.subplots(figsize=(12, 8))

# Distribuicao amostral de d dado t e N
se_d = np.sqrt(2/n_total + d_reatividade**2 / (2*(n_total-2)))
x = np.linspace(0, d_reatividade * 2.5, 1000)
y = stats.norm.pdf(x, loc=d_reatividade, scale=se_d)

ic_d_low = d_reatividade - 1.96 * se_d
ic_d_high = d_reatividade + 1.96 * se_d

ax3.plot(x, y, color='#c0392b', linewidth=3)
ax3.fill_between(x, y, alpha=0.2, color='#c0392b')
ax3.axvline(x=ic_d_low, color='#c9a14a', linestyle='--', linewidth=2,
            label=f'IC 95%: [{ic_d_low:.2f}, {ic_d_high:.2f}]')
ax3.axvline(x=ic_d_high, color='#c9a14a', linestyle='--', linewidth=2)
ax3.axvline(x=d_reatividade, color='#c0392b', linewidth=2,
            label=f'd = {d_reatividade:.2f} (efeito grande)')
ax3.axvline(x=0, color='#333333', linestyle=':', linewidth=1.5, label='d = 0 (sem efeito)')
ax3.axvline(x=0.8, color='#888888', linestyle=':', linewidth=1.5, label='d = 0.8 (limiar efeito grande)')

ax3.set_xlabel('Tamanho de efeito d de Cohen', fontsize=12)
ax3.set_ylabel('Densidade', fontsize=12)
ax3.set_title('Distribuição do tamanho de efeito — Privação de Sono e Amígdala\nYoo et al. (2007) — N=26 · d calculado a partir de t(24)=3.2',
              fontsize=13, pad=15)
ax3.legend(fontsize=11)

plt.tight_layout()
plt.savefig('dia-026-grafico-03-efeito.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Gráfico 3 salvo!')
```

### 💡 O Insight

Em um estudo de fMRI com 26 pessoas, a amígdala dos participantes privados de sono por uma noite foi **60% mais reativa** a estímulos negativos do que a dos participantes que dormiram normalmente. E o volume de amígdala ativado foi **3 vezes maior**.

O mecanismo é direto: sem sono, a conexão entre o córtex pré-frontal medial — responsável por avaliar contextos e calibrar respostas emocionais — e a amígdala enfraquece. O sistema de alarme do cérebro fica ativo sem o freio que normalmente o calibra. Como o próprio Walker descreveu: "sem sono, o cérebro reverte a um padrão mais primitivo de atividade, tornando-se incapaz de colocar experiências emocionais em contexto."

Isso não é fraqueza de caráter. É o resultado previsível de privar o sistema nervoso de um processo restaurador que ele executa toda noite.

A implicação prática: as decisões mais importantes da sua vida — em relacionamentos, no trabalho, com seus filhos — têm mais chance de acontecer num estado emocional alterado do que você imagina. E a variável que mais influencia esse estado pode ser algo que você tratou como opcional na noite anterior.

### ⚠️ Limitações do Modelo
- O estudo usa N=26, típico de neuroimagem mas pequeno para fazer generalizações amplas — o efeito foi estatisticamente significativo (p=0.004), mas estudos com N maior são necessários para confirmar a magnitude exata
- Os participantes foram submetidos a privação total de sono (uma noite inteira sem dormir), não privação parcial — dormir 5-6 horas em vez de 8 pode produzir efeitos de menor magnitude, embora na mesma direção
- Os estímulos usados foram imagens negativas padronizadas (IAPS) em laboratório — pode não reproduzir exatamente as situações reais do dia a dia onde o efeito importa
- O efeito de 60% de aumento refere-se à magnitude de ativação da amígdala medida por fMRI, não diretamente a comportamento ou tomada de decisão — o link entre ativação neural e comportamento real é uma inferência, não uma medida direta
- O fator de correção (×0,80) foi aplicado ao aumento percentual de reatividade para fins de estimativa conservadora

*A ciência é honesta sobre o que não sabe. O modelo também.*
