### 📖 A História

Existe uma frase que virou dogma. Você já ouviu, provavelmente já repetiu, e possivelmente já usou ela como régua pra medir se estava indo bem em algo.

"Precisa de 21 dias para formar um hábito."

Está em livros de autoajuda, apps de produtividade, cursos de coaching, palestras corporativas. Vira meme, vira desafio na internet, vira estrutura de programas caríssimos que prometem "transformar sua vida em 21 dias". A número redondo, memorável, otimista.

O único problema é que ele é falso.

A origem do mito é conhecida e vale contar. Nos anos 1950, um cirurgião plástico chamado Maxwell Maltz percebeu que seus pacientes precisavam, em média, de umas três semanas para se acostumar com o próprio novo rosto depois de uma cirurgia. Em 1960, ele publicou um livro chamado *Psycho-Cybernetics* em que registrou essa observação com a seguinte frase: **"exige um mínimo de cerca de 21 dias"** para que uma mudança de imagem mental se instale.

Note o "no mínimo". E note o "cerca de".

Ao longo das décadas seguintes, o "no mínimo" caiu. O "cerca de" caiu. O contexto (cirurgia plástica, imagem mental do próprio corpo) caiu. Sobrou o número. Sobrou o "21 dias formam um hábito".

Uma observação clínica de um cirurgião, feita em 1960, sem controle experimental, sobre um contexto muito específico, virou lei universal da psicologia comportamental. E ninguém testou de verdade. Até 2010.

---

### 📚 O Conceito: A Curva Real do Aprendizado

Em 2009, a psicóloga Phillippa Lally, do University College London, resolveu fazer o que ninguém tinha feito: medir na prática quanto tempo leva para um comportamento novo virar automático.

O desenho experimental foi elegante. Ela recrutou 96 voluntários. Cada um escolheu um comportamento simples pra fazer todos os dias, no mesmo contexto (por exemplo: beber um copo d'água depois do café da manhã, ou correr por dez minutos depois do almoço). E, todos os dias durante 12 semanas, cada voluntário respondia a um pequeno questionário chamado *Self-Report Habit Index* (SRHI), medindo o quanto aquele comportamento já se sentia automático.

O SRHI mede coisas como: você faz isso sem pensar? Faz mesmo se não decidiu conscientemente? Se sentiria estranho se não fizesse? Cada dia gerava uma nota de automaticidade.

Depois, para cada participante, Lally ajustou uma curva estatística chamada **função assintótica**. Ela captura a intuição de que aprendemos rápido no começo, depois cada vez mais devagar, até estabilizar num "teto pessoal" — o platô da automaticidade.

O momento em que uma pessoa chega a 95% do seu próprio platô é o ponto em que se pode dizer, com segurança científica, que o comportamento virou hábito.

E aqui o achado desmontou o mito.

A mediana entre os participantes foi de **66 dias**. Três vezes o número de Maltz. Mas mais desconcertante que a mediana é a variação: alguns participantes chegaram em **18 dias**. Outros levaram **254 dias** (mais de oito meses).

Um mesmo comportamento simples, em uma amostra homogênea, com uma faixa que vai de menos de três semanas a mais de oito meses. Não existe "o número". Existe uma distribuição.

E ainda há um achado terceiro, quase moral: **perder um dia ocasional não atrapalhou a formação do hábito** para os participantes que perseveraram. A curva permaneceu subindo. Isso derruba outro mito muito repetido, o do "quebrou uma vez, começou tudo de novo".

### 🧮 O Modelo

**Fonte principal:**
- Lally, P., van Jaarsveld, C. H. M., Potts, H. W. W. & Wardle, J. (2010) — "How are habits formed: Modelling habit formation in the real world." *European Journal of Social Psychology*, 40(6), 998-1009 — **N=96 voluntários (82 com dados suficientes)**, comportamentos diários, 12 semanas de acompanhamento, SRHI diário, regressão não-linear ajustada individualmente

**Fontes de contexto:**
- Maltz, M. (1960) — *Psycho-Cybernetics* — Prentice-Hall — a origem observacional do "21 dias" que virou dogma
- Gardner, B., Lally, P. & Wardle, J. (2012) — "Making health habitual: the psychology of habit-formation and general practice." *British Journal of General Practice*, 62(605), 664-666 — revisão de aplicações clínicas

**Nota metodológica:** aplico o fator de correção ×0.80 sobre a mediana de 66 dias, chegando em ~53 dias como estimativa conservadora. Motivo: dados são autorrelatados via SRHI diário, amostra modesta (N=82), população homogênea (voluntários do Reino Unido). O ajuste conservador reduz o risco de superestimar a facilidade de aprendizado. Mesmo com a correção, o número corrigido (~53 dias) permanece mais de duas vezes o "21" do mito.

A **variação de 18 a 254 dias** permanece como reportada no paper — não faz sentido corrigir os extremos, porque a mensagem central é justamente a **enormidade da variação individual**, não a magnitude do valor central.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Lally, van Jaarsveld, Potts & Wardle (2010)
# European Journal of Social Psychology 40(6):998-1009

n_voluntarios = 96
n_analisados = 82
duracao_estudo_dias = 84   # 12 semanas

# Achados centrais
mito_maltz = 21             # o numero mitico
mediana_lally = 66          # a mediana real
min_lally = 18              # tempo minimo observado
max_lally = 254             # tempo maximo (projetado pela curva assintotica)

# Aplicacao do fator de correcao x0.80 (autorrelato)
fator_correcao = 0.80
mediana_corrigida = mediana_lally * fator_correcao   # ~53 dias

# Razoes de comparacao
razao_real_vs_mito = mediana_lally / mito_maltz            # 3.14
razao_corrigido_vs_mito = mediana_corrigida / mito_maltz   # 2.51

print("=" * 68)
print("  DADOS — FORMACAO DE HABITO (Lally et al., 2010)")
print("=" * 68)
print(f"\n  Amostra:                 {n_voluntarios} voluntarios ({n_analisados} analisados)")
print(f"  Duracao do estudo:       {duracao_estudo_dias} dias (12 semanas)")
print(f"  Metodo:                  SRHI diario + regressao nao-linear")

print(f"\n  O MITO:")
print(f"  {mito_maltz} dias (Maltz, 1960 — observacao clinica)")

print(f"\n  A REALIDADE MEDIDA:")
print(f"  Mediana:                 {mediana_lally} dias")
print(f"  Minimo observado:        {min_lally} dias")
print(f"  Maximo projetado:        {max_lally} dias")

print(f"\n  COM FATOR DE CORRECAO x{fator_correcao}:")
print(f"  Mediana corrigida:       ~{mediana_corrigida:.0f} dias")

print(f"\n  RAZOES:")
print(f"  Real / mito:             {razao_real_vs_mito:.1f}x mais")
print(f"  Corrigido / mito:        {razao_corrigido_vs_mito:.1f}x mais")
print("=" * 68)
print()
print(f"  O mito subestima em pelo menos 2.5x o tempo real de formacao.")
print(f"  E a variacao individual e enorme: entre {min_lally} e {max_lally} dias.")
```

```python
# --- O MODELO ---
# A curva assintotica de Lally e a assinatura estatistica da variacao

# Modelo assintotico de Lally: automaticidade(t) = A * (1 - exp(-k*t))
# onde A e o plato individual, k e a velocidade de aprendizado
# O tempo para atingir 95% do plato: t_95 = -ln(0.05) / k

# Para uma pessoa mediana (t_95 = 66 dias):
k_mediana = -np.log(0.05) / mediana_lally
# Para uma pessoa rapida (t_95 = 18 dias):
k_rapida = -np.log(0.05) / min_lally
# Para uma pessoa lenta (t_95 = 254 dias):
k_lenta = -np.log(0.05) / max_lally

# Distribuicao aproximada dos tempos individuais (log-normal)
# Ajustando aos dados observados: mediana=66, min=18, max=254
# log(t_95) tem distribuicao aproximadamente normal
log_min = np.log(min_lally)
log_max = np.log(max_lally)
log_mediana = np.log(mediana_lally)
sigma_log = (log_max - log_min) / (2 * 1.96)  # ~IC 95% aproximado

# IC 95% pra mediana amostral (N=82, distribuicao log-normal)
se_log_mediana = sigma_log / np.sqrt(n_analisados)
mediana_ic_low = np.exp(log_mediana - 1.96 * se_log_mediana)
mediana_ic_high = np.exp(log_mediana + 1.96 * se_log_mediana)

print("=" * 68)
print("  MODELO — A CURVA ASSINTOTICA DA AUTOMATICIDADE")
print("=" * 68)

print(f"\n  MODELO DE LALLY:")
print(f"    automaticidade(t) = A * (1 - exp(-k*t))")
print(f"    A = plato individual (maximo pessoal)")
print(f"    k = velocidade de aprendizado (individual)")
print(f"    t_95 = tempo para atingir 95% do plato = -ln(0.05) / k")

print(f"\n  VELOCIDADE k (por perfil):")
print(f"    Rapida  (t_95 = {min_lally} dias):   k = {k_rapida:.4f} / dia")
print(f"    Mediana (t_95 = {mediana_lally} dias):   k = {k_mediana:.4f} / dia")
print(f"    Lenta   (t_95 = {max_lally} dias):  k = {k_lenta:.4f} / dia")

print(f"\n  ASSINATURA ESTATISTICA (log-normal, N={n_analisados}):")
print(f"    Mediana:              {mediana_lally} dias")
print(f"    IC 95% da mediana:    [{mediana_ic_low:.0f}, {mediana_ic_high:.0f}] dias")
print(f"    Sigma (log):          {sigma_log:.2f}")

print(f"\n  PROGRESSO NO DIA 21 (o mito):")
progresso_mito_rapido = (1 - np.exp(-k_rapida * mito_maltz)) * 100
progresso_mito_mediana = (1 - np.exp(-k_mediana * mito_maltz)) * 100
progresso_mito_lento = (1 - np.exp(-k_lenta * mito_maltz)) * 100
print(f"    Pessoa rapida:        {progresso_mito_rapido:.0f}% do plato")
print(f"    Pessoa mediana:       {progresso_mito_mediana:.0f}% do plato")
print(f"    Pessoa lenta:         {progresso_mito_lento:.0f}% do plato")

print(f"\n  Ou seja: aos 21 dias, uma pessoa mediana esta com apenas")
print(f"  {progresso_mito_mediana:.0f}% do seu proprio plato de automaticidade.")
print(f"  Menos da metade do caminho.")
print("=" * 68)
```

```python
# --- VISUALIZACAO ---
# Segundo dia com a ASSINATURA ESTATISTICA VISIVEL

cor_mito = '#c0392b'
cor_lally = '#2a8a82'
cor_corrigido = '#c9a14a'
cor_variacao = '#8a7020'
cor_neutra = '#9c9b94'

# GRAFICO 1 — A CURVA ASSINTOTICA REAL (o modelo de Lally)
fig1, ax1 = plt.subplots(figsize=(11, 6.5))

dias = np.linspace(0, 270, 500)
curva_rapida = (1 - np.exp(-k_rapida * dias)) * 100
curva_mediana = (1 - np.exp(-k_mediana * dias)) * 100
curva_lenta = (1 - np.exp(-k_lenta * dias)) * 100

ax1.plot(dias, curva_rapida, color=cor_lally, linewidth=2, alpha=0.5,
         label=f'Rapido (t95 = {min_lally} dias)')
ax1.plot(dias, curva_mediana, color=cor_lally, linewidth=3.5,
         label=f'Mediana (t95 = {mediana_lally} dias)')
ax1.plot(dias, curva_lenta, color=cor_lally, linewidth=2, alpha=0.5,
         label=f'Lento (t95 = {max_lally} dias)')

# Linha do 95% do plato
ax1.axhline(y=95, color='#666', linestyle=':', linewidth=1.5, alpha=0.7)
ax1.text(245, 96.5, '95% do plato\n(habito formado)',
         fontsize=9, color='#666', style='italic', ha='right')

# Marcar o mito de 21 dias
ax1.axvline(x=mito_maltz, color=cor_mito, linestyle='--', linewidth=2, alpha=0.9)
progresso_21 = (1 - np.exp(-k_mediana * mito_maltz)) * 100
ax1.plot(mito_maltz, progresso_21, 's', color=cor_mito, markersize=14, zorder=5)
ax1.annotate(f'aos 21 dias (mito de Maltz)\nvoce esta com apenas\n{progresso_21:.0f}% do plato',
             xy=(mito_maltz, progresso_21), xytext=(80, 40),
             fontsize=11, color=cor_mito, fontweight='bold', ha='center',
             arrowprops=dict(arrowstyle='->', color=cor_mito, lw=1.5))

# Marcar a mediana real
ax1.axvline(x=mediana_lally, color=cor_lally, linestyle='--', linewidth=2, alpha=0.6)
ax1.plot(mediana_lally, 95, 'o', color=cor_lally, markersize=14, zorder=5)
ax1.text(mediana_lally+3, 97, f'mediana real\n{mediana_lally} dias',
         fontsize=11, color='#1a5f5a', fontweight='bold')

ax1.set_xlabel('Dias desde o inicio do comportamento')
ax1.set_ylabel('% do plato individual de automaticidade')
ax1.set_xlim(0, 270)
ax1.set_ylim(0, 105)
ax1.set_title('A curva assintotica real do aprendizado\nModelo: automaticidade(t) = A * (1 - exp(-k*t))',
              fontsize=13, pad=15)
ax1.legend(frameon=False, loc='center right', fontsize=10)

plt.figtext(0.5, 0.005,
            'Modelo de Lally et al. (2010), European Journal of Social Psychology, N=82 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-036-grafico-01-curva-assintotica.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 1 salvo!")

# GRAFICO 2 — A ASSINATURA ESTATISTICA VISIVEL (distribuicao dos tempos)
fig2, ax2 = plt.subplots(figsize=(11, 6.5))

# Distribuicao log-normal dos tempos individuais
t_range = np.linspace(5, 300, 500)
mu_log = np.log(mediana_lally)
densidade = stats.lognorm.pdf(t_range, s=sigma_log, scale=np.exp(mu_log))
densidade_norm = densidade / densidade.max()

ax2.plot(t_range, densidade_norm, color=cor_lally, linewidth=3)
ax2.fill_between(t_range, densidade_norm, alpha=0.18, color=cor_lally)

# Marcadores
ax2.axvline(x=mito_maltz, color=cor_mito, linestyle='--', linewidth=2.5)
ax2.text(mito_maltz-3, 0.72, f'mito\n{mito_maltz} dias',
         ha='right', fontsize=11, color=cor_mito, fontweight='bold')

ax2.axvline(x=mediana_lally, color=cor_lally, linestyle='--', linewidth=2.5)
ax2.text(mediana_lally+3, 0.72, f'mediana real\n{mediana_lally} dias',
         ha='left', fontsize=11, color='#1a5f5a', fontweight='bold')

ax2.axvline(x=mediana_corrigida, color=cor_corrigido, linestyle=':', linewidth=2)
ax2.text(mediana_corrigida+2, 0.55, f'corrigido x0.80:\n~{mediana_corrigida:.0f} dias',
         ha='left', fontsize=10, color=cor_variacao, fontweight='bold')

# Faixa da variacao individual
ax2.annotate('', xy=(max_lally, 0.05), xytext=(min_lally, 0.05),
             arrowprops=dict(arrowstyle='<->', color=cor_variacao, lw=2))
ax2.text((min_lally+max_lally)/2, 0.10,
         f'variacao real: {min_lally} a {max_lally} dias',
         ha='center', fontsize=11, color=cor_variacao, fontweight='bold', style='italic')

ax2.set_xlabel('Tempo para atingir 95% do plato de automaticidade (dias)')
ax2.set_ylabel('Densidade (normalizada)')
ax2.set_xlim(0, 280)
ax2.set_ylim(0, 1.1)
ax2.set_title('A assinatura estatistica da formacao de habito\nDistribuicao log-normal ajustada aos dados de Lally (2010)',
              fontsize=13, pad=15)

plt.figtext(0.5, 0.005,
            f'IC 95% da mediana: [{mediana_ic_low:.0f}, {mediana_ic_high:.0f}] dias | Nao existe UM numero. Existe uma distribuicao. | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-036-grafico-02-assinatura-estatistica.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 2 salvo!")

# GRAFICO 3 — Onde voce esta aos 21 dias (progresso por perfil)
fig3, ax3 = plt.subplots(figsize=(11, 6.5))

perfis = ['Perfil rapido\n(t95 = 18 dias)',
          'Perfil mediano\n(t95 = 66 dias)',
          'Perfil lento\n(t95 = 254 dias)']
progressos_21 = [
    (1 - np.exp(-k_rapida * mito_maltz)) * 100,
    (1 - np.exp(-k_mediana * mito_maltz)) * 100,
    (1 - np.exp(-k_lenta * mito_maltz)) * 100,
]
cores = [cor_lally, cor_corrigido, cor_mito]

bars = ax3.barh(perfis, progressos_21, color=cores, alpha=0.88, height=0.5)
ax3.set_xlim(0, 105)
ax3.set_xlabel('% do plato individual atingido aos 21 dias')
ax3.axvline(x=95, color='#666', linestyle=':', linewidth=1.5)
ax3.text(96, 2.5, '95%\n(habito formado)', fontsize=9, color='#666',
         style='italic', ha='left')

for bar, v in zip(bars, progressos_21):
    ax3.text(v + 1.5, bar.get_y() + bar.get_height()/2,
             f'{v:.0f}%', va='center', fontweight='bold', fontsize=15)

ax3.set_title('Onde voce esta aos 21 dias, dependendo do seu perfil\nO mito assume que todo mundo esta em 100%. Nenhum esta.',
              fontsize=13, pad=15)

plt.figtext(0.5, 0.005,
            'Calculado a partir do modelo assintotico de Lally et al. (2010) | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-036-grafico-03-progresso-21-dias.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 3 salvo!")
```

### 💡 O Insight

**A mediana real é 66 dias. Aplicando o fator conservador do projeto, chega em ~53 dias. Mais de duas vezes o número mítico dos 21.**

Mas o número que mais me pegou não foi a mediana. Foi o intervalo. **Entre 18 e 254 dias**, dependendo da pessoa e do comportamento.

Isso significa que se você tem tentado formar um hábito há três meses e ainda não sente que ele "pegou", você não é anormal, não é fraco, e não está fazendo errado. Você está simplesmente dentro de uma distribuição que a ciência conhece e mede há mais de uma década.

Mas o mito dos 21 dias fez três estragos silenciosos que vale nomear.

O primeiro é o abandono precoce. Muita gente desistiu de um comportamento novo por volta do dia 30, 40, 50, sentindo que "não estava funcionando". Estava. Estava exatamente na fase esperada da curva, subindo devagar em direção ao platô. Mas a régua interna dizia que já era pra estar pronto. E a régua interna estava errada.

O segundo é a culpa desproporcional pela falha isolada. O paper de Lally é claro: perder um dia ocasional não atrapalhou o processo. O que atrapalha é abandonar. Mas a régua dos 21 dias criou uma mentalidade binária, "cumpri ou quebrei", que transforma qualquer falha num recomeço do zero. Isso não é psicológico. É prático: se cada tropeço zera a contagem, quase ninguém termina.

O terceiro é o efeito reverso, mais raro mas real: pessoas que **conseguem** algo em três semanas e concluem que "já está pronto", relaxam a repetição, e vêem o hábito se dissolver. Chegar a 21 dias não é chegar ao platô. É estar, no melhor cenário, no meio do caminho.

O que a ciência da formação de hábitos ensina, uma vez que a gente sai do reino dos números redondos, é mais generoso e mais realista ao mesmo tempo. **Um hábito se forma em curva, não em ponto. Persistir vale mais do que ser perfeito. E o tempo real é seu, não do relógio de Maltz.**

*Se você olhar honestamente pra sua vida, quantos hábitos você abandonou por volta do dia 30, achando que já deveria estar pronto — mas você estava, na verdade, no meio do caminho?*

### ⚠️ Limitações do Modelo

- Amostra de 82 participantes analisados, todos do Reino Unido, na maioria adultos jovens. Cultura, idade e contexto socioeconômico afetam a formação de hábitos, então a generalização exige cautela.
- Os comportamentos escolhidos eram simples e específicos (beber água, caminhar, comer uma fruta). Hábitos complexos (como uma rotina de trabalho profundo, ou uma prática artística) podem seguir dinâmica diferente e provavelmente levar mais tempo.
- Autorrelato via SRHI tem viés inerente. Pessoas podem superestimar ou subestimar quanto o comportamento já se sente automático. Aplico o fator conservador ×0.80 sobre a mediana pra reduzir esse risco.
- A distribuição individual (18 a 254 dias) foi estimada por extrapolação da curva assintótica ajustada, não observada diretamente para todos os casos (o estudo durou 84 dias). O extremo superior é uma projeção com incerteza real.
- Não confunda "hábito automático" com "hábito saudável". O paper mede formação, não benefício. Um hábito ruim pode se formar exatamente pelo mesmo mecanismo, e frequentemente com curva mais rápida (porque envolve recompensa imediata).

*A ciência é honesta sobre o que não sabe. O modelo também.*
