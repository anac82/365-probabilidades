### 📖 A História

Hoje é Dia das Bruxas. Fantasia de abóbora, criança pedindo doce na porta. A bruxa virou
enfeite de outubro.

Na minha infância, esse hype de Halloween quase não existia. Pelo menos não na minha
vida. O máximo que eu lembro é a minha mãe avisando, para mim e para os meus irmãos, que
era dia de dormir cedo. Muito cedo. Porque era dia das bruxas. E eu ia, tranquila, sem
grandes questionamentos.

Hoje eu penso nessa bruxa que fazia a gente dormir cedo. Não era a da fantasia, que não
existia por aqui. Era a outra, a crença de que certas pessoas conseguem fazer mal a
outras por meios que não se veem. Fui ver se essa ainda existe. Quantas pessoas
acreditam nisso hoje. E se o Brasil está perto ou longe do resto do mundo.

A resposta me surpreendeu mais do que qualquer fantasia.

### 📚 O Conceito: bruxaria como crença medida

A antropologia define bruxaria como a capacidade atribuída a certas pessoas de causar dano
por meios sobrenaturais. Não é religião nem superstição genérica. É uma crença sobre quem
pode causar o mal.

Por décadas essa crença foi estudada em aldeias, com etnografia. O que faltava era uma
medida comparável entre países. Ela apareceu quando um economista juntou seis rodadas de
pesquisas do Pew Research Center, de 2008 a 2017, em 95 países, e encontrou uma pergunta
que estava em todas elas, com a mesma redação.

A pergunta é esta: "Você acredita no mau-olhado, ou que certas pessoas podem lançar
maldições ou feitiços que fazem coisas ruins acontecerem com alguém?". O autor reconhece
que o mau-olhado embaralha um pouco, mas a segunda metade da frase é exatamente a
definição de bruxaria.

---

### 🧮 O Modelo

Quatro camadas. A fonte dá as proporções. As contas são nossas.

1. **O Brasil e o mundo.** A proporção brasileira, com o fator ×0,80 e intervalo, contra a
   média dos 95 países e os dois extremos.
2. **A América Latina.** Os países da pesquisa de 2014, lado a lado.
3. **A rua.** Numa rua com dez vizinhos, quantos acreditam, e a chance de nenhum acreditar.
4. **Quem acredita.** O que a fonte diz sobre idade, escolaridade, renda e religiosidade,
   sem número, porque os coeficientes não foram lidos.

**Fonte central:**
- Gershman, B. (2022). "Witchcraft beliefs around the world: an exploratory analysis".
  *PLOS ONE* 17(11), e0276872. DOI 10.1371/journal.pone.0276872. Seis rodadas de pesquisas
  do Pew Research Center (2008-2017), **95 países e territórios, mais de 140 mil pessoas**;
  84 das 95 pesquisas representam 95% ou mais da população adulta do país. Pergunta única
  presente em todas: "Do you believe in the evil eye, or that certain people can cast
  curses or spells that cause bad things to happen to someone?". **Mais de 40% dos
  respondentes disseram sim; média entre países de 43%, de 9% na Suécia a 90% na
  Tunísia.** Perto de um bilhão de adultos só nos 95 países. A crença cai com idade,
  escolaridade e bem-estar material, sobe com religiosidade e crença em deus, não difere
  entre cristãos e muçulmanos, e é mais alta onde as instituições são fracas e a confiança
  social é baixa. China e Índia não estão na amostra. **Lido no texto completo.**

**Fonte do dado brasileiro:**
- Pew Research Center (2014). *Religion in Latin America*, capítulo 3. Crença no
  mau-olhado ("evil eye", mesma pergunta): **Brasil 51%**; Panamá 64%, Venezuela 57%,
  República Dominicana 55%, Honduras 54%, Chile 53%; hispânicos nos EUA 39%. Em todos os
  países pesquisados, pelo menos um terço acredita. **O N da amostra brasileira não foi
  lido**; o intervalo usa N = 1.000 como piso conservador (as amostras nacionais do Pew
  costumam ser maiores).

**Nota metodológica.** O fator ×0,80 do projeto se aplica: a medida é resposta
autorrelatada a uma pergunta de pesquisa, e entra em todas as proporções. A pergunta
mistura mau-olhado e feitiço, e isso está declarado; o número é um teto para "bruxaria"
no sentido estrito e um piso para "alguma crença em dano sobrenatural".

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

SEMENTE = 42
SORTEIOS = 200_000
rng = np.random.default_rng(SEMENTE)

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

VERMELHO, DOURADO, VERDE, CINZA = '#c0392b', '#c8a84b', '#1a5f5a', '#6b6a64'
RODAPE = ('Fonte: Gershman (2022), PLOS ONE 17(11), 95 países, N > 140.000 · Pew Research Center (2014), Religion in Latin America · '
          'cálculo próprio | #365Probabilidades')

def mil(n):
    return f"{int(round(n)):,}".replace(",", ".")

def vir(x, casas=1):
    return f"{x:.{casas}f}".replace(".", ",")

def jeffreys(k, n):
    return stats.beta(k + 0.5, n - k + 0.5)

print("✅ Bibliotecas carregadas · semente", SEMENTE)
```

```python
# --- DADOS DA LITERATURA ---
FATOR = 0.80

# Gershman 2022 · 95 países, Pew 2008-2017
P_MUNDO = 0.43          # média entre países
P_MIN, PAIS_MIN = 0.09, 'Suécia'
P_MAX, PAIS_MAX = 0.90, 'Tunísia'
N_MUNDO = 140_000
PAISES = 95

# Pew 2014 · Religion in Latin America · "evil eye" (mesma pergunta)
LATAM = {'Panamá': 0.64, 'Venezuela': 0.57, 'Rep. Dominicana': 0.55, 'Honduras': 0.54, 'Chile': 0.53, 'Brasil': 0.51}
P_BRASIL = LATAM['Brasil']
P_HISPANICOS_EUA = 0.39
N_BRASIL = 1000          # piso conservador; o N real não foi lido

print("=" * 70)
print("  DADOS · QUEM ACREDITA EM BRUXARIA")
print("=" * 70)
print(f"\n  Gershman 2022: {PAISES} países, mais de {mil(N_MUNDO)} pessoas")
print(f"    média entre países {vir(P_MUNDO * 100, 0)}% · mínimo {vir(P_MIN * 100, 0)}% ({PAIS_MIN}) · máximo {vir(P_MAX * 100, 0)}% ({PAIS_MAX})")
print(f"\n  Pew 2014, América Latina (mesma pergunta):")
for p, v in LATAM.items():
    print(f"    {p:<16} {vir(v * 100, 0):>3}%")
print(f"    {'hispânicos nos EUA':<16} {vir(P_HISPANICOS_EUA * 100, 0):>3}%")
print("=" * 70)
```

```python
# --- O MODELO ---

# CAMADA 1 · O Brasil e o mundo
k_br = round(P_BRASIL * N_BRASIL)
d_br = jeffreys(k_br, N_BRASIL); ic_br = d_br.interval(0.95)
p_br_corr = P_BRASIL * FATOR; ic_corr = (ic_br[0] * FATOR, ic_br[1] * FATOR)
print("=" * 70)
print("  CAMADA 1 · O BRASIL E O MUNDO")
print("=" * 70)
print(f"\n  Brasil, publicado: {vir(P_BRASIL * 100, 0)}%  IC 95% [{vir(ic_br[0] * 100)}; {vir(ic_br[1] * 100)}] (com N = {mil(N_BRASIL)} como piso)")
print(f"  Brasil, com o fator ×0,80: {vir(p_br_corr * 100, 0)}%  [{vir(ic_corr[0] * 100)}; {vir(ic_corr[1] * 100)}]")
print(f"  Média do mundo: {vir(P_MUNDO * 100, 0)}% (×0,80: {vir(P_MUNDO * FATOR * 100, 0)}%) · {PAIS_MIN} {vir(P_MIN * 100, 0)}% · {PAIS_MAX} {vir(P_MAX * 100, 0)}%")
print(f"  O Brasil está {vir((P_BRASIL - P_MUNDO) * 100, 0)} pontos acima da média dos {PAISES} países.")
print(f"  Chance de a proporção brasileira ser maior que a média mundial: {vir((1 - d_br.cdf(P_MUNDO)) * 100)}%")

# CAMADA 2 · A América Latina
print("\n" + "=" * 70)
print("  CAMADA 2 · A AMÉRICA LATINA (Pew 2014)")
print("=" * 70)
print(f"\n  {'país':<18} {'publicado':>10} {'×0,80':>7}")
for p, v in LATAM.items():
    print(f"  {p:<18} {vir(v * 100, 0):>9}% {vir(v * FATOR * 100, 0):>6}%")
print(f"\n  O Brasil é o sexto dos que passam da metade na lista do Pew; em todo país pesquisado, pelo menos um terço acredita.")
print(f"  Hispânicos nos EUA: {vir(P_HISPANICOS_EUA * 100, 0)}%. A crença viaja, mas perde 12 pontos na mudança.")

# CAMADA 3 · A rua
RUA = 10
p_rua = rng.beta(k_br + 0.5, N_BRASIL - k_br + 0.5, SORTEIOS) * FATOR
vizinhos = rng.binomial(RUA, p_rua)
print("\n" + "=" * 70)
print(f"  CAMADA 3 · UMA RUA COM {RUA} VIZINHOS ({mil(SORTEIOS)} ruas sorteadas, ×0,80)")
print("=" * 70)
print(f"\n  Acreditam: média {vir(vizinhos.mean())} · mediana {int(np.median(vizinhos))}")
print(f"  Nenhum acredita: {vir((vizinhos == 0).mean() * 100, 2)}% das ruas")
print(f"  Pelo menos um: {vir((vizinhos >= 1).mean() * 100, 1)}%")
print(f"  Maioria acredita (6 ou mais): {vir((vizinhos >= 6).mean() * 100, 0)}%")
print(f"  Chance de o vizinho da direita acreditar: {vir(p_br_corr * 100, 0)}%. Da esquerda também. Os dois: {vir(p_br_corr ** 2 * 100, 0)}%.")

# CAMADA 4 · Quem acredita
print("\n" + "=" * 70)
print("  CAMADA 4 · QUEM ACREDITA (Gershman 2022, direção dos efeitos)")
print("=" * 70)
print("\n  Cai com a idade, com a escolaridade e com o bem-estar material.")
print("  Sobe com a religiosidade e com a crença em deus; não muda entre cristãos e muçulmanos.")
print("  Entre países: mais alta onde as instituições são fracas, a confiança é baixa e a cultura é conformista.")
print("  Entre países, renda, escolaridade média e IDH NÃO explicam a diferença; instituições e confiança, sim.")
print("  Os coeficientes não foram lidos; esta camada é direção, não número.")
print("=" * 70)
```

```python
# --- VISUALIZAÇÃO ---

# GRÁFICO 1 · O Brasil e o mundo
fig1, ax1 = plt.subplots(figsize=(12, 8))
rot = [f'{PAIS_MIN}\n(menor)', f'média de\n{PAISES} países', 'Brasil', f'{PAIS_MAX}\n(maior)']
val = np.array([P_MIN, P_MUNDO, P_BRASIL, P_MAX]) * 100
cores = [VERDE, CINZA, VERMELHO, DOURADO]
ax1.bar(range(4), val, color=cores, alpha=0.3, width=0.62)
ax1.bar(range(4), val * FATOR, color=cores, alpha=0.95, width=0.62)
for i, v in enumerate(val):
    ax1.text(i, v + 1.5, f"{vir(v, 0)}%", ha='center', fontsize=15, fontweight='bold')
    ax1.text(i, v * FATOR - 5, f"×0,80\n{vir(v * FATOR, 0)}%", ha='center', fontsize=10, color='white', fontweight='bold')
ax1.set_xticks(range(4)); ax1.set_xticklabels(rot, fontsize=12)
ax1.set_ylim(0, 100)
ax1.set_ylabel('Acreditam em mau-olhado, maldição ou feitiço (%)', fontsize=12)
ax1.set_title('O Brasil fica acima da média do mundo\n"certas pessoas podem lançar maldições ou feitiços que fazem coisas ruins acontecerem"',
              fontsize=14, pad=16)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-140-grafico-01-brasil-e-mundo.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# GRÁFICO 2 · A América Latina
fig2, ax2 = plt.subplots(figsize=(12, 8))
nomes = list(LATAM.keys()) + ['hispânicos\nnos EUA']
vals = np.array(list(LATAM.values()) + [P_HISPANICOS_EUA]) * 100
cores2 = [VERMELHO if n == 'Brasil' else CINZA for n in nomes]
ax2.barh(range(len(nomes)), vals, color=cores2, alpha=0.9, height=0.6)
for i, v in enumerate(vals):
    ax2.text(v + 1, i, f"{vir(v, 0)}%", va='center', fontsize=13, fontweight='bold')
ax2.set_yticks(range(len(nomes))); ax2.set_yticklabels(nomes, fontsize=12); ax2.invert_yaxis()
ax2.set_xlim(0, 75)
ax2.set_xlabel('Acreditam no mau-olhado ou em feitiços (%)', fontsize=12)
ax2.set_title('Na América Latina, o Brasil está entre os que passam da metade\nPew Research Center, 2014, valores publicados; em todo país pesquisado, ao menos um terço acredita', fontsize=14, pad=16)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-140-grafico-02-america-latina.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# GRÁFICO 3 · A rua
fig3, ax3 = plt.subplots(figsize=(12, 8))
ks = np.arange(0, RUA + 1)
freq = np.array([(vizinhos == k).mean() for k in ks]) * 100
ax3.bar(ks, freq, color=VERMELHO, alpha=0.85, width=0.7)
for k in ks:
    if freq[k] >= 0.5:
        ax3.text(k, freq[k] + 0.5, f"{vir(freq[k], 0)}", ha='center', fontsize=12)
ax3.set_xticks(ks)
ax3.set_xlabel(f'Vizinhos que acreditam, numa rua de {RUA}', fontsize=12)
ax3.set_ylabel('Ruas simuladas (%)', fontsize=12)
ax3.set_title(f'Numa rua com {RUA} vizinhos, {vir(vizinhos.mean(), 0)} acreditam\n'
              f'nenhum acredita em {vir((vizinhos == 0).mean() * 100, 1)}% das ruas · com o fator ×0,80 e a incerteza da amostra', fontsize=14, pad=16)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-140-grafico-03-a-rua.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")

# GRÁFICO 4 · A posterior do Brasil
fig4, ax4 = plt.subplots(figsize=(12, 8))
x = np.linspace(0.30, 0.65, 1000)
ax4.plot(x * 100, d_br.pdf(x), color=VERMELHO, lw=3, label=f'publicado · {vir(P_BRASIL * 100, 0)}%')
ax4.fill_between(x * 100, d_br.pdf(x), alpha=0.15, color=VERMELHO)
xc = x * FATOR
ax4.plot(xc * 100, d_br.pdf(x) / FATOR, color=DOURADO, lw=3, label=f'com o fator ×0,80 · {vir(p_br_corr * 100, 0)}%')
ax4.fill_between(xc * 100, d_br.pdf(x) / FATOR, alpha=0.2, color=DOURADO)
ax4.axvline(P_MUNDO * 100, color=CINZA, lw=2, ls='--')
ax4.text(P_MUNDO * 100 + 0.4, ax4.get_ylim()[1] * 0.9, f'média do mundo\n{vir(P_MUNDO * 100, 0)}%', fontsize=11, color=CINZA)
ax4.set_xlabel('Proporção de brasileiros que acreditam (%)', fontsize=12)
ax4.set_ylabel('Densidade', fontsize=12)
ax4.set_title(f'A proporção brasileira, com a incerteza de uma amostra de {mil(N_BRASIL)}\n'
              f'posterior de Jeffreys · o N real do Pew não foi lido, então o intervalo é conservador', fontsize=14, pad=16)
ax4.legend(frameon=False, fontsize=12)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-140-grafico-04-bayesiano.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 4 salvo!")
```

### 💡 O Insight

**51% dos brasileiros dizem acreditar que certas pessoas podem lançar maldições ou
feitiços que fazem mal a alguém. Com o fator ×0,80, 41%. A média de 95 países é 43%.**

A bruxa não é enfeite de outubro. É uma crença viva em quase metade do país, acima da
média do mundo e abaixo dos vizinhos latino-americanos. Numa rua com dez casas, quatro
acreditam. A chance de nenhuma acreditar é de meio por cento.

Dentro de cada país, a crença cai com a idade, a escolaridade e a renda, e sobe com a
religiosidade. Entre países, renda e escolaridade não explicam a diferença; instituições
e confiança, sim. Mas ela atravessa todos os grupos. É uma crença antiga sobre quem pode
causar o mal, e ainda está na mesa de muita gente.

*Você conhece alguém que acredita nisso? Ou acreditou um dia?*

---

### ⚠️ Limitações do Modelo

- A pergunta mistura mau-olhado e feitiço. O 51% é teto para "bruxaria" no sentido
  estrito e piso para "alguma crença em dano sobrenatural".
- O N da amostra brasileira do Pew não foi lido; o intervalo usa N = 1.000 como piso.
- O dado brasileiro é de 2014. A crença pode ter mudado em doze anos.
- A média de 43% é média entre países, não entre pessoas; China e Índia estão fora.
- A camada "quem acredita" é direção, sem coeficientes.

*A ciência é honesta sobre o que não sabe. O modelo também.*
