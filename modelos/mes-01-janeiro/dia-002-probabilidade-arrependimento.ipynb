# 365 Probabilidades — Dia #002
## Qual a probabilidade de você se arrepender de uma grande decisão?

**Tipo:** Preditivo

---

### 📖 A História

Depois de 20 anos no mesmo lugar, decidi sair.

Não foi uma decisão fácil. Foi calculada, adiada, revisada,
questionada — e tomada mesmo assim.

E a primeira pergunta que veio depois foi a mais humana de todas:
*E se eu me arrepender?*

A ciência estuda arrependimento há décadas. E o que ela descobriu
surpreende: não nos arrependemos do que fizemos.
Nos arrependemos do que não fizemos.

Este modelo calcula a probabilidade de arrependimento
por tipo de decisão — e o resultado muda tudo.

---

### 🧮 O Modelo

Baseado em três fontes:
- **Gilovich & Medvec, 1995** — The experience of regret: what, when and why
- **Roese & Summerville, 2005** — What we regret most and why
- **Landman, 1993** — Regret: The persistence of the possible

```python
# 365 Probabilidades — Dia #002
# Qual a probabilidade de você se arrepender de uma grande decisão?

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Configuração visual padrão do projeto
plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---

# Gilovich & Medvec, 1995
# Estudo sobre arrependimento de curto vs longo prazo
p_arrepend_acao_curto_prazo = 0.53    # arrependimento de FAZER no curto prazo
p_arrepend_omissao_longo_prazo = 0.84 # arrependimento de NÃO FAZER no longo prazo

# Roese & Summerville, 2005
# As 6 áreas de maior arrependimento na vida
areas_arrependimento = {
    'Educação':       0.32,
    'Carreira':       0.22,
    'Relacionamento': 0.15,
    'Criação dos filhos': 0.10,
    'Autocuidado':    0.11,
    'Finanças':       0.10
}

# Landman, 1993
# Arrependimento por tipo de decisão
p_arrepend_decisao_imposta = 0.42  # decisão forçada por circunstâncias
p_arrepend_decisao_propria = 0.31  # decisão tomada por escolha própria
p_arrepend_indecisao = 0.74        # arrependimento de não ter decidido

# Fator de correção conservador
fator_correcao = 0.80

dados = {
    'Fonte': [
        'Gilovich & Medvec, 1995',
        'Gilovich & Medvec, 1995',
        'Landman, 1993',
        'Landman, 1993',
        'Landman, 1993'
    ],
    'Indicador': [
        'Arrependimento de AGIR no curto prazo',
        'Arrependimento de NÃO AGIR no longo prazo',
        'Arrependimento de decisão imposta',
        'Arrependimento de decisão própria',
        'Arrependimento de indecisão'
    ],
    'Probabilidade': [
        p_arrepend_acao_curto_prazo,
        p_arrepend_omissao_longo_prazo,
        p_arrepend_decisao_imposta,
        p_arrepend_decisao_propria,
        p_arrepend_indecisao
    ]
}

df = pd.DataFrame(dados)
print("✅ Dados carregados")
print(f"\n{df.to_string(index=False)}")
```

```python
# --- O MODELO ---
# Cálculo principal
# --- O MODELO ---
# Probabilidade de arrependimento por tipo de decisão

# Cenário 1 — Você AGE (toma a decisão)
p_arrepend_agir_curto = p_arrepend_acao_curto_prazo
p_arrepend_agir_longo = 1 - p_arrepend_omissao_longo_prazo  # inverso

# Cenário 2 — Você NÃO AGE (adia ou evita)
p_arrepend_nao_agir_curto = 1 - p_arrepend_acao_curto_prazo  # inverso
p_arrepend_nao_agir_longo = p_arrepend_omissao_longo_prazo

# Aplicando fator de correção conservador
p_agir_curto_corrigido = p_arrepend_agir_curto * fator_correcao
p_agir_longo_corrigido = p_arrepend_agir_longo * fator_correcao
p_nao_agir_curto_corrigido = p_arrepend_nao_agir_curto * fator_correcao
p_nao_agir_longo_corrigido = p_arrepend_nao_agir_longo * fator_correcao

print("=" * 60)
print("  PROBABILIDADE DE ARREPENDIMENTO POR DECISÃO")
print("=" * 60)
print(f"\n  SE VOCÊ AGIR:")
print(f"  → Arrependimento no curto prazo:  {p_agir_curto_corrigido*100:.0f}%")
print(f"  → Arrependimento no longo prazo:  {p_agir_longo_corrigido*100:.0f}%")
print(f"\n  SE VOCÊ NÃO AGIR:")
print(f"  → Arrependimento no curto prazo:  {p_nao_agir_curto_corrigido*100:.0f}%")
print(f"  → Arrependimento no longo prazo:  {p_nao_agir_longo_corrigido*100:.0f}%")
print(f"\n  ─────────────────────────────────────────────────────")
print(f"  Decisão própria vs imposta:")
print(f"  → Arrependimento de decisão própria:  {p_arrepend_decisao_propria*fator_correcao*100:.0f}%")
print(f"  → Arrependimento de decisão imposta:  {p_arrepend_decisao_imposta*fator_correcao*100:.0f}%")
print(f"  → Arrependimento de indecisão:        {p_arrepend_indecisao*fator_correcao*100:.0f}%")
print(f"\n  CONCLUSÃO:")
print(f"  Agir tem {p_agir_curto_corrigido*100:.0f}% de chance de arrependimento agora.")
print(f"  Não agir tem {p_nao_agir_longo_corrigido*100:.0f}% de chance de arrependimento depois.")
print("=" * 60)
```

### 💡 O Insight

**59% das pessoas se arrependem de não ter decidido.**

Mais do que de qualquer decisão que tomaram.

A matemática é contraintuitiva — achamos que agir é arriscado.
Mas os dados mostram o oposto:

- Agir tem **42%** de chance de arrependimento agora
- Não agir tem **67%** de chance de arrependimento depois

E quando a decisão é sua — não imposta por circunstâncias —
a probabilidade de arrependimento cai para **25%**.

Autonomia reduz arrependimento. Indecisão o maximiza.

A pergunta não é *e se eu me arrepender?*
A pergunta é: **de qual arrependimento você prefere?**

```python
# --- CAMADA 2 — MODELO ESTATÍSTICO ---
# Distribuição Beta com tamanhos amostrais reais dos estudos
# P(arrependimento) modelada como distribuição Beta(α, β)
# onde α = casos de arrependimento, β = casos sem arrependimento

from scipy import stats
import numpy as np

# --- FONTE 1: Yeung & Feldman, 2022 — N = 988 ---
# Replicação do padrão ação/omissão
# Arrependimento de AÇÃO no curto prazo
n_yeung = 988
p_acao_curto = 0.53  # proporção original replicada (com efeito mais fraco)
alpha_acao = p_acao_curto * n_yeung
beta_acao = (1 - p_acao_curto) * n_yeung

# Arrependimento de OMISSÃO no longo prazo
p_omissao_longo = 0.75  # efeito mais fraco que o original (0.84)
alpha_omissao = p_omissao_longo * n_yeung
beta_omissao = (1 - p_omissao_longo) * n_yeung

# --- FONTE 2: Roese & Summerville, 2005 — N = 3.041 ---
# Meta-análise das 6 maiores áreas de arrependimento
n_roese = 3041
areas_roese = {
    'Educação':       0.32,
    'Carreira':       0.22,
    'Relacionamento': 0.15,
    'Criação dos filhos': 0.10,
    'Autocuidado':    0.11,
    'Finanças':       0.10
}

# Distribuições Beta
dist_acao = stats.beta(alpha_acao, beta_acao)
dist_omissao = stats.beta(alpha_omissao, beta_omissao)

# Intervalos de confiança 95%
ic_acao = dist_acao.interval(0.95)
ic_omissao = dist_omissao.interval(0.95)

# Distribuições Beta para áreas (Roese & Summerville)
dists_areas = {}
ics_areas = {}
for area, p in areas_roese.items():
    alpha = p * n_roese
    beta = (1 - p) * n_roese
    dist = stats.beta(alpha, beta)
    dists_areas[area] = dist
    ics_areas[area] = dist.interval(0.95)

# Aplicando fator de correção
print("=" * 65)
print("  MODELO BAYESIANO — INTERVALO DE CONFIANÇA 95%")
print("  Yeung & Feldman, 2022 (N=988) + Roese & Summerville, 2005 (N=3.041)")
print("=" * 65)
print(f"\n  Arrependimento de AGIR (curto prazo):")
print(f"  → Estimativa central: {p_acao_curto*fator_correcao*100:.0f}%")
print(f"  → IC 95%: [{ic_acao[0]*fator_correcao*100:.1f}%, {ic_acao[1]*fator_correcao*100:.1f}%]")
print(f"\n  Arrependimento de NÃO AGIR (longo prazo):")
print(f"  → Estimativa central: {p_omissao_longo*fator_correcao*100:.0f}%")
print(f"  → IC 95%: [{ic_omissao[0]*fator_correcao*100:.1f}%, {ic_omissao[1]*fator_correcao*100:.1f}%]")
print(f"\n  Áreas de maior arrependimento (Roese & Summerville, N=3.041):")
for area, ic in ics_areas.items():
    p = areas_roese[area]
    print(f"  → {area:<22} {p*100:.0f}%  IC 95%: [{ic[0]*100:.1f}%, {ic[1]*100:.1f}%]")
print(f"\n  CONCLUSÃO COM INCERTEZA:")
print(f"  → Agir:     [{ic_acao[0]*fator_correcao*100:.1f}%, {ic_acao[1]*fator_correcao*100:.1f}%] de arrependimento")
print(f"  → Não agir: [{ic_omissao[0]*fator_correcao*100:.1f}%, {ic_omissao[1]*fator_correcao*100:.1f}%] de arrependimento")
print(f"  → Com 95% de confiança")
print("=" * 65)
```

```python
# --- VISUALIZAÇÃO CAMADA 2 ---

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Gráfico 1 — Distribuições Beta sobrepostas
x = np.linspace(0.3, 0.9, 1000)

y_acao = dist_acao.pdf(x / fator_correcao) / fator_correcao
y_omissao = dist_omissao.pdf(x / fator_correcao) / fator_correcao

axes[0].plot(x * 100, y_acao, color='#1a5f5a', linewidth=2.5,
             label=f'Agir (IC 95%: 39.9%–44.9%)')
axes[0].fill_between(x * 100, y_acao, alpha=0.15, color='#1a5f5a')

axes[0].plot(x * 100, y_omissao, color='#c0392b', linewidth=2.5,
             label=f'Não agir (IC 95%: 57.8%–62.1%)')
axes[0].fill_between(x * 100, y_omissao, alpha=0.15, color='#c0392b')

# IC 95% — agir
axes[0].axvline(x=39.9, color='#1a5f5a', linestyle='--', alpha=0.5)
axes[0].axvline(x=44.9, color='#1a5f5a', linestyle='--', alpha=0.5)

# IC 95% — não agir
axes[0].axvline(x=57.8, color='#c0392b', linestyle='--', alpha=0.5)
axes[0].axvline(x=62.1, color='#c0392b', linestyle='--', alpha=0.5)

axes[0].set_xlabel('Probabilidade de Arrependimento (%)')
axes[0].set_ylabel('Densidade')
axes[0].set_title('Distribuição Beta\nAgir vs Não Agir', fontsize=12, pad=15)
axes[0].legend(fontsize=9)

# Gráfico 2 — Áreas com IC
areas = list(ics_areas.keys())
centrais = [areas_roese[a] * 100 for a in areas]
ic_min = [ics_areas[a][0] * 100 for a in areas]
ic_max = [ics_areas[a][1] * 100 for a in areas]
erros = [[c - mn for c, mn in zip(centrais, ic_min)],
         [mx - c for c, mx in zip(centrais, ic_max)]]

cores_areas = ['#c0392b', '#e67e22', '#c8a84b', '#1a5f5a', '#2980b9', '#8e44ad']
bars = axes[1].barh(areas, centrais, xerr=erros,
                    color=cores_areas, alpha=0.8,
                    error_kw=dict(ecolor='gray', capsize=4, linewidth=1.5))

axes[1].set_xlabel('Frequência de Arrependimento (%)')
axes[1].set_title('Áreas de Maior Arrependimento\ncom IC 95% (N=3.041)', fontsize=12, pad=15)

for bar, valor in zip(bars, centrais):
    axes[1].text(valor + 0.5, bar.get_y() + bar.get_height()/2,
                 f'{valor:.0f}%', va='center', fontsize=10, fontweight='bold')

plt.suptitle(
    '365 Probabilidades — Dia #002 | Yeung & Feldman 2022 (N=988) + Roese & Summerville 2005 (N=3.041) | Fator de correção: 0.80',
    fontsize=9, color='gray', y=0
)
plt.tight_layout()
plt.savefig('dia-002-grafico-bayesiano.png', dpi=150, bbox_inches='tight')
plt.show()
print("✅ Gráfico bayesiano salvo!")
```

### 💡 O Insight

**Não agir tem entre 58% e 62% de chance de arrependimento.**
**Agir tem entre 40% e 45%.**

A diferença é de quase 20 pontos percentuais — com 95% de confiança.

E o maior arrependimento da vida humana não é uma decisão errada.
É uma oportunidade de educação que passou.
É uma carreira que nunca foi tentada.

A pergunta não é *e se eu me arrepender de agir?*
A ciência responde: você vai se arrepender mais de não ter agido.
