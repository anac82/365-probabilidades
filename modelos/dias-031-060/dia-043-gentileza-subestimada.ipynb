### 📚 O Conceito: O Ponto Cego da Gentileza

A gente é péssimo em avaliar o efeito da própria gentileza. Não porque não se importe,
mas por um motivo específico: quem dá olha para o **objeto** ("é só um chocolate", "é só
uma mensagem"), enquanto quem recebe olha para o **calor** do gesto, para o fato de
alguém ter pensado nele.

Amit Kumar e Nicholas Epley chamam isso de miscalibração. Os dois lados avaliam o mesmo
ato por lentes diferentes, e por isso quem faz o bem quase sempre acha que fez menos do
que de fato fez. O gesto chega maior do outro lado.

E isso tem uma consequência prática: a gente deixa de fazer gentilezas pequenas porque
acha que elas não importam. Mas os dados dizem que importam, e mais do que imaginamos.

---

### 🧮 O Modelo

Uso o experimento de campo central do estudo: pessoas num parque foram convidadas a doar
um copo de chocolate quente a um estranho. Quem doou estimou quão "grande" o outro acharia
o gesto e quão positivo o outro se sentiria. Comparo essas estimativas com o que os
recebedores de fato relataram, e traduzo a diferença em tamanho de efeito (Cohen's d).

**Fonte:**
- Kumar, A. & Epley, N. (2022). *A Little Good Goes an Unexpectedly Long Way:
  Underestimating the Positive Impact of Kindness on Recipients*. **Journal of
  Experimental Psychology: General**.
  Experimento de campo (chocolate quente, Maggie Daley Park, Chicago): N=84 (75 doaram).

*Nota metodológica: os números centrais são tamanhos de efeito de experimentos com
medida direta de doadores e recebedores. Por isso o fator de correção ×0.80, reservado
a proporções de survey autorrelatado, não se aplica.*


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

# Paleta 365 Probabilidades
DOURADO = '#c8a84b'
VERMELHO = '#c0392b'
VERDE    = '#1a5f5a'
CINZA    = '#6b6a64'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Fonte: Kumar & Epley (2022), JEP: General — experimento do chocolate quente

N_doaram = 75   # recrutados N=84; 75 concordaram em doar

# "Quão GRANDE foi o gesto" (índice de percepção)
grande_doador   = 3.76   # SD 2.41 — o que quem DEU achou
grande_recebedor = 7.00  # SD 1.89 — o que quem RECEBEU sentiu
d_grande = 1.16          # paired t(74)=9.97, p<.0001

# "Quão POSITIVO o recebedor se sentiu"
pos_previsto_doador = 2.73  # SD 1.36 — o que o doador PREVIU
pos_real_recebedor  = 3.52  # SD 1.17 — o que o recebedor de fato SENTIU
d_pos = 0.48                # paired t(74)=4.10, p=.0001

# Efeito cascata (indireta reciprocidade) — quem recebe repassa mais
repassa_recebeu_gentileza = 48.0   # US$ doados a estranhos depois
repassa_controle          = 41.0   # US$ no grupo controle

print("=" * 65)
print("  DADOS DA LITERATURA — A GENTILEZA SUBESTIMADA")
print("=" * 65)
print(f"  Experimento do chocolate quente (N={N_doaram} doaram):")
print()
print(f"  'Quão grande foi o gesto':")
print(f"  → quem DEU achou:        {grande_doador:.2f}")
print(f"  → quem RECEBEU sentiu:   {grande_recebedor:.2f}   (d={d_grande})")
print()
print(f"  'Quão positivo o recebedor se sentiu':")
print(f"  → o doador PREVIU:       {pos_previsto_doador:.2f}")
print(f"  → o recebedor SENTIU:    {pos_real_recebedor:.2f}   (d={d_pos})")
print()
print(f"  Efeito cascata: quem recebe repassa mais")
print(f"  → recebeu gentileza:     ${repassa_recebeu_gentileza:.0f}")
print(f"  → controle:              ${repassa_controle:.0f}")
print("=" * 65)
```

```python
# --- O MODELO: quanto o gesto cresce do outro lado ---

# O quanto o gesto é MAIOR na percepção de quem recebe
razao_grande = grande_recebedor / grande_doador
subestima_grande = (grande_recebedor - grande_doador) / grande_recebedor

# O quanto a positividade real supera a prevista
subestima_pos = (pos_real_recebedor - pos_previsto_doador) / pos_real_recebedor

# Efeito cascata
ganho_cascata = (repassa_recebeu_gentileza - repassa_controle) / repassa_controle

print("=" * 65)
print("  MODELO — O TAMANHO REAL DE UM GESTO PEQUENO")
print("=" * 65)
print(f"  Percepção de 'grandeza' do gesto:")
print(f"  → quem recebe sente o gesto {razao_grande:.2f}x maior do que quem deu imagina")
print(f"  → o doador subestima em {subestima_grande*100:.0f}%  (efeito grande, d={d_grande})")
print()
print(f"  Positividade sentida:")
print(f"  → o recebedor se sente {subestima_pos*100:.0f}% mais positivo do que o doador previu")
print(f"  → (efeito médio, d={d_pos})")
print()
print(f"  E não para em quem recebe:")
print(f"  → quem recebeu gentileza repassa {ganho_cascata*100:.0f}% mais a estranhos")
print()
print(f"  Você acha que seu gesto é pequeno.")
print(f"  Do outro lado, ele chega quase o dobro do tamanho.")
print("=" * 65)
```

```python
# --- VISUALIZAÇÃO — 3 GRÁFICOS SEPARADOS ---

# ── GRÁFICO 1 — O gesto cresce do outro lado ──
fig1, ax1 = plt.subplots(figsize=(10, 6))
cats = ['O tamanho que\nQUEM DEU imaginou', 'O tamanho que\nQUEM RECEBEU sentiu']
vals = [grande_doador, grande_recebedor]
bars = ax1.bar(cats, vals, color=[CINZA, DOURADO], alpha=0.9, width=0.45)
ax1.set_ylabel("Quão 'grande' foi o gesto (índice do estudo)")
ax1.set_ylim(0, 8)
ax1.set_title('O mesmo gesto, dois tamanhos\nDoar um chocolate quente a um estranho — Kumar & Epley 2022',
              fontsize=13, pad=15)
for bar, v in zip(bars, vals):
    ax1.text(bar.get_x()+bar.get_width()/2, v+0.15, f'{v:.2f}',
             ha='center', fontweight='bold', fontsize=16)
ax1.annotate('', xy=(1, grande_recebedor), xytext=(0, grande_doador),
             arrowprops=dict(arrowstyle='->', color=VERMELHO, lw=2))
ax1.text(0.5, 5.6, f'{razao_grande:.1f}x maior', ha='center',
         fontsize=13, color=VERMELHO, fontweight='bold')
plt.figtext(0.5, 0.005,
            'N=75 · doadores vs. recebedores | Kumar & Epley 2022 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-043-grafico-01-tamanho.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 — ASSINATURA: Cohen's d das duas medidas ──
fig2, ax2 = plt.subplots(figsize=(10, 6))
labels = ["'Grandeza' do gesto", 'Positividade sentida']
ds = [d_grande, d_pos]
ypos = np.arange(len(labels))
ax2.barh(ypos, ds, color=[DOURADO, VERDE], alpha=0.9, height=0.5)
ax2.set_yticks(ypos); ax2.set_yticklabels(labels)
ax2.axvline(0.5, color=CINZA, ls='--', lw=1, alpha=0.7)
ax2.axvline(0.8, color=CINZA, ls='--', lw=1, alpha=0.7)
ax2.text(0.5, 1.42, 'médio', fontsize=9, color=CINZA, ha='center')
ax2.text(0.8, 1.42, 'grande', fontsize=9, color=CINZA, ha='center')
ax2.set_xlabel("Cohen's d  (o quanto o doador subestima)")
ax2.set_xlim(0, 1.35)
for i, dv in enumerate(ds):
    ax2.text(dv+0.02, i, f'd = {dv:.2f}', va='center', fontweight='bold', fontsize=13)
ax2.set_title("A ASSINATURA ESTATÍSTICA\nQuem faz o bem subestima o efeito, em dois níveis",
              fontsize=13, pad=15)
plt.figtext(0.5, 0.005,
            "Tamanhos de efeito de experimento de campo | Kumar & Epley 2022 | #365Probabilidades",
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-043-grafico-02-assinatura.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 — O efeito cascata ──
fig3, ax3 = plt.subplots(figsize=(10, 6))
cats3 = ['Não recebeu\ngentileza', 'Recebeu\num gesto gentil']
vals3 = [repassa_controle, repassa_recebeu_gentileza]
bars3 = ax3.bar(cats3, vals3, color=[CINZA, VERDE], alpha=0.9, width=0.45)
ax3.set_ylabel('Quanto repassou a estranhos depois (US$)')
ax3.set_ylim(0, 58)
ax3.set_title('A gentileza não para em quem recebe\nQuem recebeu um gesto foi mais generoso com o próximo — Kumar & Epley 2022',
              fontsize=13, pad=15)
for bar, v in zip(bars3, vals3):
    ax3.text(bar.get_x()+bar.get_width()/2, v+1, f'${v:.0f}',
             ha='center', fontweight='bold', fontsize=16)
plt.figtext(0.5, 0.005,
            f'+{ganho_cascata*100:.0f}% de generosidade repassada | Kumar & Epley 2022 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-043-grafico-03-cascata.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**Quem deu um chocolate quente a um estranho achou que o gesto valia 3,8. Quem recebeu
sentiu que valia 7,0. O mesmo ato, quase o dobro do tamanho do outro lado.**

A razão é simples e humana. Você olha para o que fez pela lente do custo: foi barato, foi
rápido, foi pouco. Quem recebe olha pela lente do significado: alguém parou, pensou, se
importou. E significado pesa mais que preço.

O problema é que essa miscalibração cobra um preço invisível. A gente engole o impulso de
mandar a mensagem, de elogiar, de oferecer ajuda, porque uma voz diz "não é nada demais,
vão achar bobo". Os dados mostram que essa voz está sistematicamente errada, e não por
pouco: é um efeito grande (d = 1,16).

E há o que vem depois. Quem recebe um gesto gentil fica mais generoso com o próximo,
repassando cerca de 17% mais a estranhos. O gesto que você acha pequeno não só chega
maior, como continua andando.

*Qual gentileza você tem engolido essa semana por achar que "não é nada demais"?*

---

### ⚠️ Limitações do Modelo

- Os efeitos vêm de experimentos com amostras modestas (o de campo teve N=75 doadores); o
  padrão se repetiu em vários experimentos do estudo, o que reforça a robustez.
- As medidas de "grandeza" e positividade são autorrelatadas em escalas do estudo; não são
  grandezas físicas absolutas.
- O experimento mede gestos entre estranhos num contexto específico; a intensidade pode
  variar entre culturas e tipos de relação.
- O efeito cascata foi medido num jogo econômico de laboratório, um proxy de generosidade,
  não a vida real inteira.

*A ciência é honesta sobre o que não sabe. O modelo também.*
