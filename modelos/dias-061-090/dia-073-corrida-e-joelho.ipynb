## Qual a probabilidade de correr destruir o seu joelho?

**Tipo:** Preditivo
**Data de publicação:** 2026-08-25
**Ferramenta:** Python
**Decisão analisada:** Devo parar de correr para poupar as articulações?
**Hashtag:** #365Probabilidades #Dia073

---

### 📖 A História

Todo corredor já ouviu essa frase, quase sempre de alguém que não corre.

"Isso vai acabar com o seu joelho."

Ela vem com uma lógica que parece física de ensino médio: impacto repetido desgasta
cartilagem, e cartilagem não regenera. Milhares de passos por semana, multiplicados
por anos, dariam num joelho gasto.

O argumento é bom o suficiente para ninguém conferir.

E é exatamente por isso que ele merece ser conferido. Porque se estiver certo, muita
gente deveria parar. E se estiver errado, muita gente parou por nada.

---

### 📚 O Conceito: a comparação que falta na frase

A frase esconde uma comparação incompleta. Ela compara correr com **não fazer nada**,
como se não fazer nada fosse neutro para a articulação.

Não é. O joelho de quem não se mexe também tem um destino, e ele precisa entrar na
conta.

Então a pergunta correta tem três grupos, não dois: quem corre por prazer, quem corre
profissionalmente, e quem não corre.

E existe uma armadilha embutida em qualquer estudo desse tipo, que precisa ser dita
antes dos números: **quem tem joelho ruim para de correr.** Então quem ainda corre
depois de muitos anos é, em parte, quem já tinha joelho bom. Isso se chama viés do
corredor saudável, e nenhum ajuste estatístico resolve em desenho observacional.

Não anula o achado. Explica parte dele, e a peça precisa dizer isso.

---

### 🧮 O Modelo

Revisão sistemática com meta-análise, comparando três grupos.

**Fontes:**
- Alentorn-Geli, E., Samuelsson, K., Musahl, V., Green, C. L., Bhandari, M. &
  Karlsson, J., 2017 · *Journal of Orthopaedic & Sports Physical Therapy* 47(6),
  373-390 · **25 estudos (n = 125.810) incluídos, 17 estudos (n = 114.829)
  meta-analisados**
- Prevalência de artrose de quadril e joelho: **competitivos 13,3%** [IC 95%:
  11,6% · 15,2%], **recreativos 3,5%** [3,4% · 3,6%], **controles sedentários 10,2%**
  [9,9% · 10,6%]
- Razão de chances, controles sedentários como referência: **competitivos 1,34**
  [0,97 · 1,86] e **recreativos 0,86** [0,69 · 1,07], com **p < 0,001 para a
  diferença** entre os dois grupos de corredores
- Subgrupo por tempo de exposição: correr há **menos de 15 anos** se associou a menos
  artrose que os controles, **OR = 0,6** [0,49 · 0,73]
- Definições do artigo: **competitivo** = profissional, de elite, ou que participa de
  competições internacionais. **Recreativo** = corre em contexto amador.

**Conclusão dos autores:** tanto um estilo de vida mais sedentário quanto longa
exposição a alto volume ou alta intensidade de corrida estão associados a artrose de
quadril e joelho. É uma curva em U, não uma reta.

**Referência interna:** o **#068** também tratou de corrida, mas ali o objeto era o
mecanismo da euforia. Aqui é articulação. Fontes e desenhos distintos.

**Nota metodológica sobre o fator ×0.80:** não se aplica. Prevalência de artrose por
diagnóstico, não proporção autorrelatada de survey.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Alentorn-Geli, Samuelsson, Musahl, Green, Bhandari & Karlsson, 2017
# JOSPT 47(6), 373-390

n_estudos_incluidos = 25
n_incluidos         = 125_810
n_estudos_meta      = 17
n_meta              = 114_829

# Prevalencia de artrose de quadril e/ou joelho, com IC 95%
PREVALENCIA = {
    'Corredores competitivos': (0.133, (0.116, 0.152)),
    'Corredores recreativos':  (0.035, (0.034, 0.036)),
    'Controles sedentarios':   (0.102, (0.099, 0.106)),
}

# Razao de chances, com os SEDENTARIOS como referencia
OR = {
    'Competitivos': (1.34, (0.97, 1.86)),
    'Recreativos':  (0.86, (0.69, 1.07)),
    'Menos de 15 anos correndo': (0.60, (0.49, 0.73)),
}
p_diferenca_grupos = 0.001    # p < 0,001 para a diferenca entre competitivos e recreativos

aplica_fator_080 = False

print("=" * 70)
print("  DADOS - CORRIDA E ARTROSE DE QUADRIL E JOELHO")
print("=" * 70)
print(f"\n  Alentorn-Geli et al., 2017 (JOSPT 47(6), 373-390):")
print(f"  -> {n_estudos_incluidos} estudos incluidos, n = {n_incluidos:,}"
      .replace(",", "."))
print(f"  -> {n_estudos_meta} estudos meta-analisados, n = {n_meta:,}".replace(",", "."))
print(f"\n  Prevalencia de artrose:")
for grupo, (p, ic) in PREVALENCIA.items():
    print(f"  -> {grupo:<26} {p*100:>5.1f}%"
          f"  IC 95% [{ic[0]*100:.1f}% · {ic[1]*100:.1f}%]")
print(f"\n  Razao de chances (sedentarios como referencia):")
for grupo, (orr, ic) in OR.items():
    cruza = "CRUZA o 1" if ic[0] < 1 < ic[1] else "nao cruza o 1"
    print(f"  -> {grupo:<28} {orr:.2f}  IC 95% [{ic[0]:.2f} · {ic[1]:.2f}]  {cruza}")
print(f"\n  -> Diferenca entre competitivos e recreativos: p < {p_diferenca_grupos}")
print(f"\n  Fator x0.80 aplicado: {aplica_fator_080}")
print("=" * 70)
```

```python
# --- O MODELO ---
# Duas leituras dos mesmos dados: a prevalencia bruta e a razao de chances ajustada.
# Elas nao contam a mesma historia, e o dia depende de mostrar as duas.

p_rec = PREVALENCIA['Corredores recreativos'][0]
p_sed = PREVALENCIA['Controles sedentarios'][0]
p_com = PREVALENCIA['Corredores competitivos'][0]

# 1) Leitura bruta, em escala humana
por_mil = {g: p * 1000 for g, (p, _) in PREVALENCIA.items()}
dif_rec_sed_pp = (p_sed - p_rec) * 100
razao_sed_rec = p_sed / p_rec

# 2) Leitura ajustada: quais intervalos cruzam o 1
significativos = {g: not (ic[0] < 1 < ic[1]) for g, (_, ic) in OR.items()}

# 3) Numero necessario para... nao adoecer.
#    Diferenca de risco entre sedentarios e recreativos, em pessoas.
dif_risco = p_sed - p_rec
nnt_bruto = 1 / dif_risco

# 4) O que a razao de chances de 0,60 do subgrupo significa em risco absoluto,
#    aplicada sobre a prevalencia dos sedentarios. Traducao propria, declarada.
def or_para_prob(orr, p_ref):
    odds_ref = p_ref / (1 - p_ref)
    odds = orr * odds_ref
    return odds / (1 + odds)


p_menos15 = or_para_prob(OR['Menos de 15 anos correndo'][0], p_sed)
ic_menos15 = tuple(or_para_prob(x, p_sed) for x in OR['Menos de 15 anos correndo'][1])

print("=" * 70)
print("  MODELO - AS DUAS LEITURAS DO MESMO ARTIGO")
print("=" * 70)
print(f"\n  1) A leitura bruta, em escala humana (a cada 1.000 pessoas):")
for grupo, valor in por_mil.items():
    print(f"  -> {grupo:<26} {valor:>5.0f} com artrose")
print(f"  -> Diferenca entre sedentarios e recreativos: {dif_rec_sed_pp:.1f} pp")
print(f"  -> O sedentario tem {razao_sed_rec:.1f}x mais artrose que o corredor amador")
print(f"  -> Uma a cada {nnt_bruto:.0f} pessoas separa os dois grupos")
print(f"\n  2) A leitura ajustada, e aqui a historia muda:")
for grupo, (orr, ic) in OR.items():
    marca = "SIGNIFICATIVO" if significativos[grupo] else "nao significativo"
    print(f"  -> {grupo:<28} OR {orr:.2f} [{ic[0]:.2f} · {ic[1]:.2f}]  {marca}")
print(f"\n  ATENCAO: os dois OR principais CRUZAM o 1.")
print(f"  -> Correr como amador nao reduziu significativamente a chance,")
print(f"     e correr como profissional nao aumentou significativamente,")
print(f"     quando cada um e comparado a ficar parado.")
print(f"  -> O que E significativo: a diferenca entre os dois tipos de corredor")
print(f"     (p < {p_diferenca_grupos}) e o subgrupo por tempo de exposicao.")
print(f"\n  3) O achado mais solido do artigo:")
print(f"  -> Correr ha menos de 15 anos: OR 0,60 [0,49 · 0,73], nao cruza o 1")
print(f"  -> Traduzido em risco absoluto sobre a prevalencia dos sedentarios:")
print(f"     {p_menos15*100:.1f}%  [{ic_menos15[0]*100:.1f}% · {ic_menos15[1]*100:.1f}%]")
print(f"     contra {p_sed*100:.1f}% dos sedentarios")
print(f"     (traducao propria, declarada: o artigo publica a razao, nao o absoluto)")
print(f"\n  A conclusao dos autores nao e 'correr protege'. E que os dois extremos,")
print(f"  o sofa e o alto volume por muito tempo, se associam a artrose.")
print("=" * 70)
```

```python
# --- VISUALIZACAO ---

def br(n):
    return f"{n:,}".replace(",", ".")


DOURADO = '#c8a84b'
VERMELHO = '#c0392b'
VERDE = '#2a8a82'
CINZA = '#6b6a64'

# GRAFICO 1 - A curva em U da prevalencia
fig1, ax1 = plt.subplots(figsize=(12, 8))

ordem = ['Controles sedentarios', 'Corredores recreativos', 'Corredores competitivos']
rotulos = ['Sedentários\n(não correm)', 'Corredores\nrecreativos',
           'Corredores\ncompetitivos']
valores = [PREVALENCIA[g][0] * 100 for g in ordem]
ics = [PREVALENCIA[g][1] for g in ordem]
cores = [VERMELHO, VERDE, VERMELHO]

barras = ax1.bar(rotulos, valores, color=cores, alpha=0.88, width=0.5)
for barra, valor, ic in zip(barras, valores, ics):
    x = barra.get_x() + barra.get_width() / 2
    ax1.plot([x, x], [ic[0] * 100, ic[1] * 100], color='#222', linewidth=2.5)
    ax1.text(x, valor + 0.9, f'{valor:.1f}%'.replace('.', ','), ha='center',
             fontsize=22, fontweight='bold')

ax1.set_ylim(0, 17.5)
ax1.set_ylabel('Prevalência de artrose de quadril e joelho (%)')
ax1.set_title('O sofá também tem um efeito no joelho\n'
              f'Alentorn-Geli et al., 2017 · {n_estudos_meta} estudos · '
              f'n = {br(n_meta)}',
              fontsize=15, pad=18)
ax1.text(1, 15.6, 'Curva em U: os dois extremos têm mais artrose que o meio',
         ha='center', fontsize=13, color='#333', style='italic')

plt.figtext(0.5, 0.005,
            'Fonte: Alentorn-Geli et al., 2017, JOSPT 47(6), 373-390'
            '  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-073-grafico-01-curva-u.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 1 salvo")

# GRAFICO 2 - Assinatura: as razoes de chances e o que cruza o 1
fig2, ax2 = plt.subplots(figsize=(12, 8))

nomes = list(OR.keys())
y = np.arange(len(nomes))[::-1]
for yi, nome in zip(y, nomes):
    orr, ic = OR[nome]
    cruza = ic[0] < 1 < ic[1]
    cor = CINZA if cruza else VERDE
    ax2.plot([ic[0], ic[1]], [yi, yi], color=cor, linewidth=3.4,
             solid_capstyle='round')
    for lim in ic:
        ax2.plot([lim, lim], [yi - 0.09, yi + 0.09], color=cor, linewidth=3.4)
    ax2.scatter([orr], [yi], s=260, color=cor, zorder=3)
    ax2.text(orr, yi + 0.25, f'{orr:.2f}'.replace('.', ','), ha='center',
             fontsize=17, fontweight='bold', color=cor)
    ax2.text(2.05, yi, 'cruza o 1' if cruza else 'não cruza o 1',
             va='center', fontsize=12, color=cor,
             fontweight='normal' if cruza else 'bold')

ax2.axvline(x=1.0, color='#333', linestyle='--', linewidth=1.8)
ax2.set_yticks(y)
ax2.set_yticklabels(['Corredores\ncompetitivos', 'Corredores\nrecreativos',
                     'Correndo há\nmenos de 15 anos'], fontsize=12)
ax2.set_xscale('log')
ax2.set_xlim(0.42, 3.2)
ax2.set_xticks([0.5, 0.75, 1, 1.5, 2])
ax2.get_xaxis().set_major_formatter(plt.FuncFormatter(
    lambda v, _: (f'{v:.2f}' if v < 1 else f'{v:g}').replace('.', ',')))
ax2.get_xaxis().set_minor_formatter(plt.NullFormatter())
ax2.tick_params(axis='x', which='minor', length=0)
ax2.set_ylim(-0.6, y.max() + 0.7)
ax2.set_xlabel('Razão de chances de artrose, sedentários como referência (escala log)')
ax2.set_title('A assinatura estatística: dois dos três intervalos cruzam o 1\n'
              'O achado sólido é o do tempo de exposição, não o da corrida em si',
              fontsize=15, pad=18)
ax2.text(0.5, -0.135,
         'Diferença entre corredores competitivos e recreativos: p < 0,001',
         transform=ax2.transAxes, ha='center', fontsize=12, color=DOURADO,
         fontweight='bold')

plt.figtext(0.5, 0.005,
            'Fonte: Alentorn-Geli et al., 2017, JOSPT 47(6), 373-390'
            '  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-073-grafico-02-razoes.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 2 salvo")

# GRAFICO 3 - A cada mil pessoas
fig3, ax3 = plt.subplots(figsize=(12, 7.5))

grupos3 = ['Corredores\nrecreativos', 'Controles\nsedentários',
           'Corredores\ncompetitivos']
valores3 = [por_mil['Corredores recreativos'], por_mil['Controles sedentarios'],
            por_mil['Corredores competitivos']]
cores3 = [VERDE, DOURADO, VERMELHO]

barras3 = ax3.barh(np.arange(3)[::-1], valores3, color=cores3, alpha=0.88, height=0.5)
for barra, valor in zip(barras3, valores3):
    ax3.text(valor + 3, barra.get_y() + barra.get_height() / 2,
             f'{valor:.0f} em 1.000', va='center', fontsize=18, fontweight='bold')

ax3.set_yticks(np.arange(3)[::-1])
ax3.set_yticklabels(grupos3, fontsize=13)
ax3.set_xlim(0, 175)
ax3.set_xlabel('Pessoas com artrose de quadril ou joelho, a cada 1.000')
ax3.set_title('A mesma coisa em escala humana\n'
              f'O sedentário tem {razao_sed_rec:.1f} vezes mais artrose que o corredor '
              f'amador'.replace('.', ','),
              fontsize=15, pad=18)
ax3.text(0.98, 0.94,
         'Prevalência bruta. Quando se ajusta e se olha o\n'
         'intervalo de confiança, a diferença deixa de ser\n'
         'estatisticamente significativa. Ver gráfico 2.',
         transform=ax3.transAxes, ha='right', va='top', fontsize=11, color=CINZA,
         style='italic',
         bbox=dict(boxstyle='round,pad=0.5', facecolor='white', edgecolor=CINZA,
                   alpha=0.92))

plt.figtext(0.5, 0.005,
            'Fonte: Alentorn-Geli et al., 2017, JOSPT 47(6), 373-390'
            '  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-073-grafico-03-escala-humana.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 3 salvo")
```

### 💡 O Insight

**3,5% dos corredores recreativos têm artrose de quadril ou joelho. Entre os
sedentários, são 10,2%.**

Trinta e cinco pessoas em mil contra cento e duas em mil. Quase três vezes mais no
grupo que não corre.

Se a história parasse aqui, a manchete seria fácil e enganosa. Ela não para.

O terceiro grupo é o que dá sentido ao conjunto: entre corredores **competitivos**,
profissionais e de elite, a prevalência sobe para 13,3%. Mais alta que a dos
sedentários.

Isso é uma curva em U. Os dois extremos, o sofá e o alto volume por muitos anos, se
associam a mais artrose. O meio é onde está o corredor amador.

E agora a parte que quase nenhuma cobertura desse estudo contou.

Quando os autores calculam a razão de chances, comparando cada grupo com os
sedentários, os intervalos de confiança são **1,34 [0,97 · 1,86]** para competitivos e
**0,86 [0,69 · 1,07]** para recreativos.

**Os dois cruzam o 1.**

Traduzindo: com esses dados, não dá para afirmar que correr como amador reduz
significativamente a chance de artrose, nem que correr profissionalmente aumenta. O
que é estatisticamente significativo é a **diferença entre os dois tipos de corredor**,
e um achado específico.

Esse achado é o mais sólido do artigo, e é sobre tempo: quem corre há **menos de 15
anos** tem razão de chances de **0,60** [0,49 · 0,73] em relação a quem não corre. Esse
não cruza o 1.

Falta ainda o problema que nenhum ajuste resolve: **quem tem joelho ruim para de
correr.** Quem continua correndo depois de muitos anos é, em parte, quem já tinha
joelho bom. Parte da vantagem dos corredores pode ser seleção, não proteção.

Então o que dá para dizer, sem exagerar em nenhuma direção?

**A frase "correr destrói o joelho" não tem apoio nos dados.** Se tivesse, os
corredores amadores apareceriam pior que os sedentários, e eles aparecem bem melhor.

E a frase oposta, "correr protege o joelho", também vai além do que o intervalo
permite.

O que sobra é mais chato e mais útil: **a dose importa mais que a atividade.** Correr
por prazer, em volume amador, não aparece como fator de risco em lugar nenhum desses
114 mil dados. Correr como profissional, por décadas, aparece. E ficar parado também.

### ⚠️ Limitações do Modelo

- **Observacional. Associação, não causa.** Ninguém sorteou quem correria e quem
  ficaria parado.
- **Viés do corredor saudável.** Quem tem dor articular tende a abandonar a corrida, o
  que enriquece o grupo de corredores com pessoas de articulação boa. Isso pode
  explicar parte da diferença e não é corrigível neste desenho.
- **Os dois intervalos principais cruzam o 1.** A leitura de prevalência bruta favorece
  o corredor amador com folga, mas a razão de chances ajustada não sustenta afirmação
  de proteção nem de dano.
- **Os intervalos de prevalência são suspeitosamente estreitos** (3,4% a 3,6% para os
  recreativos). Isso decorre do agrupamento de amostras muito grandes e **não reflete a
  heterogeneidade entre estudos**, que é alta. Ler esse intervalo como precisão real
  seria erro.
- **Classificação grosseira.** "Competitivo" e "recreativo" são categorias amplas, sem
  volume semanal, ritmo ou tipo de terreno. A dose exata não está medida.
- **A tradução da razão de chances em risco absoluto é minha, não do artigo.** Feita
  aplicando a razão sobre a prevalência dos sedentários, e declarada como tal.
- Artrose de quadril e joelho é o desfecho. Nada aqui fala sobre lesão aguda, tendão,
  fratura por estresse ou dor sem artrose.
- Quem já tem artrose diagnosticada está fora do escopo deste dia. A decisão de correr
  nesse caso é clínica e individual.
- Fator ×0.80 não aplicado: prevalência por diagnóstico, não autorrelato de survey.

*A ciência é honesta sobre o que não sabe. O modelo também.*
