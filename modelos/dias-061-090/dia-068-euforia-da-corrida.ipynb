### 📖 A História

Tem explicações que a gente aprende sem nunca ter checado. Elas circulam com tanta
naturalidade que perguntar de onde vieram parece falta de educação.

"É endorfina" é uma delas.

Você corre, termina cansado e estranhamente bem, e alguém explica que são as
endorfinas. A frase aparece em revista de saúde, em aula de educação física, em
conversa de rua. Há quatro décadas.

Só que existe um jeito direto de testar isso, e ele é quase grosseiro na
simplicidade: dá para bloquear o sistema opioide da pessoa com um remédio, mandar
ela correr, e perguntar se a sensação apareceu do mesmo jeito.

Se for endorfina, com o receptor bloqueado a euforia não deveria vir.

Alguém fez esse experimento.

---

### 📚 O Conceito: por que a explicação já era suspeita

Antes mesmo do experimento havia um problema de encanamento.

Correr de fato aumenta a beta-endorfina no sangue, e isso foi medido muitas vezes.
O detalhe é que **as endorfinas circulantes atravessam mal a barreira
hematoencefálica**. Elas estão no sangue, e o sentimento acontece no cérebro. Faltava
o caminho entre uma coisa e a outra.

Os endocanabinoides, produzidos pelo próprio corpo e quimicamente parecidos com os
princípios ativos da cannabis, são lipofílicos e atravessam essa barreira com
facilidade. Isso os torna candidatos melhores por razões de física, não de gosto.

Em 2015, no *PNAS*, o grupo de Fuss testou isso em camundongos bloqueando as duas
vias e mostrou que a redução de ansiedade e a analgesia induzida pelo exercício
dependem da sinalização endocanabinoide, não da opioide.

Faltava fazer em gente.

---

### 🧮 O Modelo

Um ensaio randomizado, duplo-cego, controlado por placebo, com bloqueio
farmacológico e medida bioquímica no plasma.

**Fontes:**
- Siebers, M., Biedermann, S. V., Bindila, L., Lutz, B. & Fuss, J., 2021 ·
  *Psychoneuroendocrinology* 126, 105173 · **N=63** praticantes regulares de
  exercício de resistência · duas condições, corrida e caminhada de 45 minutos em
  esteira · randomizados para **naltrexona (n=32)**, bloqueador opioide central, ou
  **placebo (n=31)** · a corrida aumentou euforia e reduziu ansiedade em relação à
  caminhada, elevou os endocanabinoides anandamida e 2-araquidonoilglicerol no
  plasma, e **o bloqueio opioide não impediu nada disso**
- Fuss, J. et al., 2015 · *PNAS* · modelo animal · ansiedade reduzida e analgesia
  induzida pelo exercício dependem da via endocanabinoide, não da opioide
- Siebers, M., Canales-Romero, D. & Fuss, J., 2026 · revisão narrativa · o campo
  abandona a dicotomia entre endorfina e endocanabinoide e propõe uma leitura
  neuromoduladora integrativa
- Registro de frequência: apenas **69% a 77%** dos corredores de resistência relatam
  já ter experimentado o barato do corredor alguma vez, faixa entre estudos

**Referência interna:** o **#034** tratou de duração de exercício e humor, com Reed &
Ones. Lá o objeto era o efeito. Aqui é o mecanismo. Fontes e desenhos distintos.

**Nota metodológica sobre o fator ×0.80:** não se aplica. Ensaio randomizado
duplo-cego, com desfecho experimental e medida bioquímica plasmática.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats, optimize

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---

# Siebers, Biedermann, Bindila, Lutz & Fuss, 2021
# Psychoneuroendocrinology 126, 105173
n_total     = 63
n_placebo   = 31
n_naltrexona = 32
minutos     = 45          # corrida e caminhada, intensidade moderada

# Desfechos, conforme publicados
# Euforia: quase o DOBRO depois de correr; praticamente inalterada ao caminhar
razao_euforia_corrida = 2.0     # "nearly twofold" - leitura qualitativa do artigo
# Bloqueio opioide NAO impediu:
bloqueio_impediu_euforia   = False
bloqueio_impediu_ansiolise = False
bloqueio_impediu_ecb       = False

# Frequencia do fenomeno entre corredores: FAIXA entre estudos, nao estimativa unica
freq_min = 0.69
freq_max = 0.77

aplica_fator_080 = False

print("=" * 70)
print("  DADOS - O BARATO DO CORREDOR SOB BLOQUEIO OPIOIDE")
print("=" * 70)
print(f"\n  Siebers et al., 2021 (Psychoneuroendocrinology 126, 105173):")
print(f"  -> N total: {n_total}  ({n_placebo} placebo, {n_naltrexona} naltrexona)")
print(f"  -> Duas condicoes: corrida e caminhada, {minutos} min em esteira")
print(f"  -> Duplo-cego, randomizado, controlado por placebo")
print(f"\n  O que a corrida produziu (contra a caminhada):")
print(f"  -> Euforia cerca de {razao_euforia_corrida:.0f}x maior")
print(f"  -> Ansiedade reduzida")
print(f"  -> Anandamida e 2-AG elevados no plasma")
print(f"\n  O que o bloqueio opioide impediu:")
print(f"  -> Euforia?              {'sim' if bloqueio_impediu_euforia else 'NAO'}")
print(f"  -> Reducao de ansiedade? {'sim' if bloqueio_impediu_ansiolise else 'NAO'}")
print(f"  -> Alta dos endocanabinoides? "
      f"{'sim' if bloqueio_impediu_ecb else 'NAO'}")
print(f"\n  Frequencia do fenomeno entre corredores de resistencia:")
print(f"  -> {freq_min*100:.0f}% a {freq_max*100:.0f}%"
      f"  (FAIXA entre estudos, nao estimativa pontual)")
print(f"\n  Fator x0.80 aplicado: {aplica_fator_080}")
print("=" * 70)
```

```python
# --- O MODELO ---
# Este e um resultado NULO: o bloqueio nao impediu o efeito.
# Resultado nulo nao prova ausencia de efeito. Ele LIMITA o tamanho do efeito
# que teria passado despercebido.
# A assinatura estatistica deste dia e, portanto, o EFEITO MINIMO DETECTAVEL.

ALFA = 0.05
PODER = 0.80


def poder_t_duas_amostras(d, n1, n2, alfa=ALFA):
    "Poder de um teste t bicaudal para diferenca de medias, efeito d de Cohen."
    gl = n1 + n2 - 2
    ncp = d * np.sqrt((n1 * n2) / (n1 + n2))     # parametro de nao centralidade
    critico = stats.t.ppf(1 - alfa / 2, gl)
    acima = stats.nct.sf(critico, gl, ncp)       # cauda que interessa
    abaixo = stats.nct.cdf(-critico, gl, ncp)    # cauda oposta, desprezivel
    # Em alguns ambientes a cauda oposta retorna NaN por underflow com ncp grande.
    acima = float(np.nan_to_num(acima, nan=0.0))
    abaixo = float(np.nan_to_num(abaixo, nan=0.0))
    return min(acima + abaixo, 1.0)

# Qual o menor efeito que este estudo detectaria com 80% de poder?
d_minimo = optimize.brentq(
    lambda d: poder_t_duas_amostras(d, n_placebo, n_naltrexona) - PODER, 0.05, 3.0)

# E qual seria o poder para efeitos de tamanhos convencionais?
tamanhos = {'pequeno (d=0,20)': 0.20, 'medio (d=0,50)': 0.50,
            'grande (d=0,80)': 0.80, 'muito grande (d=1,20)': 1.20}
poderes = {k: poder_t_duas_amostras(v, n_placebo, n_naltrexona)
           for k, v in tamanhos.items()}

# Quantas pessoas seriam necessarias para descartar um efeito medio?
def n_para(d, alfa=ALFA, poder=PODER, teto=3000):
    "Menor n por grupo que atinge o poder desejado. Busca direta, sem raiz."
    for n in range(5, teto):
        if poder_t_duas_amostras(d, n, n, alfa) >= poder:
            return n
    return teto


n_para_medio = n_para(0.50)
n_para_pequeno = n_para(0.20)

print("=" * 70)
print("  MODELO - O QUE UM RESULTADO NULO PODE E NAO PODE DIZER")
print("=" * 70)
print(f"\n  O desenho e uma falsificacao classica:")
print(f"  -> SE a euforia depende de endorfina,")
print(f"     ENTAO bloquear o receptor opioide deve reduzi-la.")
print(f"  -> Bloquearam. A euforia apareceu igual nos dois grupos.")
print(f"  -> A previsao da hipotese das endorfinas falhou.")
print(f"\n  Mas nulo nao e prova de ausencia. Com {n_placebo} contra"
      f" {n_naltrexona} participantes:")
print(f"  -> Efeito minimo detectavel com {PODER*100:.0f}% de poder:"
      f" d = {d_minimo:.2f}")
print(f"  -> Ou seja, o estudo descarta um efeito GRANDE do bloqueio opioide.")
print(f"     Um efeito pequeno ou medio poderia ter passado despercebido.")
print(f"\n  Poder do estudo para cada tamanho de efeito:")
for nome, valor in tamanhos.items():
    print(f"  -> {nome:<22} poder = {poderes[nome]*100:.0f}%")
print(f"\n  Para descartar com seguranca um efeito medio (d=0,50),")
print(f"  seriam necessarios cerca de {n_para_medio} por grupo.")
print(f"  Para um efeito pequeno (d=0,20), cerca de {n_para_pequeno} por grupo.")
print(f"\n  CONCLUSAO HONESTA:")
print(f"  -> A endorfina nao explica o essencial do fenomeno.")
print(f"  -> Nao esta provado que ela nao participa em nada.")
print(f"  -> E o campo ja abandonou o 'ou um ou outro' (Siebers et al., 2026).")
print("=" * 70)
```

### 💡 O Insight

**Sessenta e três pessoas. Metade com o sistema opioide bloqueado. A euforia apareceu
igual nos dois grupos.**

Esse é o tipo de experimento que eu gosto, porque ele podia ter dado errado. A
hipótese das endorfinas faz uma previsão clara e falsificável: bloqueie o receptor e
a sensação some. Bloquearam. Não sumiu.

E o que subiu no sangue, com e sem bloqueio, foram os endocanabinoides. Moléculas que
o próprio corpo fabrica, parentes químicas do princípio ativo da cannabis, e que
atravessam a barreira hematoencefálica com facilidade, coisa que as endorfinas
circulantes fazem mal.

Ou seja: a explicação que todo mundo repete há quarenta anos tinha um problema de
encanamento antes mesmo de ter um problema de dado.

Agora a parte em que eu seguro a manchete.

Um resultado nulo não prova ausência. Com 31 pessoas contra 32, este estudo descarta
um efeito **grande** do bloqueio opioide. Um efeito pequeno ou médio poderia ter
passado batido. Para descartar com segurança um efeito médio, seriam necessárias
cerca de 64 pessoas por grupo, o dobro do que o estudo teve. Para um efeito pequeno,
quase 400 por grupo.

E o próprio campo já saiu da briga. Numa revisão de 2026, os mesmos autores dizem
que a leitura de "ou é endorfina ou é endocanabinoide" precisa ser abandonada em
favor de uma explicação com vários sistemas agindo juntos.

Então a lição não é trocar de dogma. É outra:

**"É endorfina" nunca foi um achado. Era uma frase.** Ela sobreviveu quarenta anos
porque soava científica, explicava o que a gente sentia, e ninguém tinha motivo para
duvidar. E quando alguém finalmente montou o teste que podia derrubá-la, ela caiu.

A maior parte do que a gente repete sobre o próprio corpo nunca passou por um teste
desses.

*Que explicação você repete há anos sem nunca ter perguntado de onde ela veio?*

### ⚠️ Limitações do Modelo

- **Um único estudo humano controlado.** É a evidência humana mais direta que existe
  sobre isso, e é uma só. Não é consenso fechado.
- **Resultado nulo com N=63.** O cálculo de poder deste notebook mostra que o estudo
  descarta apenas efeitos grandes do bloqueio opioide. A conclusão correta é que a
  endorfina não explica o essencial do fenômeno, não que ela não participe de nada.
- O cálculo de poder assume teste t bicaudal para diferença de médias entre os dois
  grupos, com alfa de 0,05. O artigo usa análises mais elaboradas, então este número
  é uma referência de ordem de grandeza, não a reprodução da análise original.
- **O campo abandonou a dicotomia.** A revisão de 2026 dos mesmos autores propõe uma
  leitura integrativa, com vários sistemas neuromoduladores. Substituir "é endorfina"
  por "é endocanabinoide" repetiria o erro.
- A faixa de 69% a 77% é **variação entre estudos**, não estimativa pontual, e os
  próprios autores apontam que parte dela vem da má definição do fenômeno.
- Participantes saudáveis e já treinados. Não se sabe se o mesmo vale para quem está
  começando, para quem corre em outra intensidade, ou fora da esteira.
- Correlação entre endocanabinoides plasmáticos e sensação subjetiva não é prova de
  que a molécula medida no sangue seja a causa do que acontece no cérebro. O
  argumento da barreira hematoencefálica é plausível, não é demonstração direta.
- Fator ×0.80 não aplicado: ensaio randomizado duplo-cego com medida bioquímica.

*A ciência é honesta sobre o que não sabe. O modelo também.*
