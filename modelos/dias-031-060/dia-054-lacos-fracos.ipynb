### 📖 A História

*(gancho a ser desenvolvido no Substack)*

Você publica algo que levou meses para construir. E nota um padrão estranho:
gente distante curte, comenta e compartilha, enquanto algumas das pessoas mais
próximas passam, assistem e seguem sem um clique.

A leitura fácil é dolorida: eles não torcem por mim. Mas existe uma explicação
de rede que é mais precisa, e bem mais generosa. É dela que este dia trata.

---

### 📚 O Conceito: A Força dos Laços Fracos

Em 1973, Mark Granovetter mostrou algo contraintuitivo: os laços fortes (família,
amigos próximos) tendem a conhecer as mesmas pessoas e a mesma informação que
você. São redundantes para espalhar algo novo. Já os laços fracos (conhecidos,
gente distante) funcionam como pontes para outros grupos, e é por isso que eles
carregam o seu trabalho para fora da sua bolha.

Ou seja, o barulho de quem está longe não é sinal de que eles torcem mais. É
sinal de que estão estruturalmente melhor posicionados para propagar. E o
silêncio de quem está perto tem duas razões, nenhuma delas sobre o quanto te
querem bem: eles já estão dentro da sua bolha (não têm para onde levar a
novidade), e a comparação social torna o endosso público de um par mais custoso
do que o de um estranho.

---

### 🧮 O Modelo

Não meço aqui a torcida de ninguém. Modelo a mecânica da rede: como o alcance
novo e a chance de oportunidade mudam conforme a força do laço, e por que o
apoio dos próximos costuma ser real, mas invisível nas métricas públicas.

**Fontes:**
- Granovetter (1973). *The Strength of Weak Ties* · American Journal of
  Sociology, 78(6), 1360-1380 · um dos artigos mais citados das ciências sociais
- Rajkumar, Saint-Jacques, Bojinov, Brynjolfsson & Aral (2022). *A causal test
  of the strength of weak ties* · Science, 377, 1304-1310 · ~20 milhões de
  pessoas no LinkedIn, ~600 mil novos empregos, teste causal

**Nota metodológica:** o fator ×0.80 não se aplica; os dados são experimento
causal e teoria de rede, não proporção de survey autorrelatado.


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

DOURADO='#c8a84b'; VERMELHO='#c0392b'; VERDE='#1a5f5a'; CINZA='#6b6a64'
print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Granovetter (1973) — teoria da força dos laços fracos (conceitual)
# Rajkumar et al. (2022, Science) — teste causal no LinkedIn
n_pessoas   = 20_000_000     # ~20 milhões de usuários
n_empregos  = 600_000        # ~600 mil novos empregos observados
anos        = 5

print("=" * 66)
print("  DADOS DA LITERATURA — A FORÇA DOS LAÇOS FRACOS")
print("=" * 66)
print(f"\n  Granovetter (1973): laços fortes são redundantes;")
print(f"  laços fracos são pontes para redes novas.")
print(f"\n  Rajkumar et al. (2022, Science) — teste causal:")
print(f"  → Pessoas:        {n_pessoas:,}".replace(',', '.'))
print(f"  → Novos empregos: {n_empregos:,}".replace(',', '.'))
print(f"  → Duração:        {anos} anos")
print(f"  → Achado: os laços MODERADAMENTE FRACOS foram os que")
print(f"     mais aumentaram a mobilidade de trabalho.")
print("=" * 66)
```

```python
# --- O MODELO — a mecânica da rede ---
# Eixo comum: PROXIMIDADE do laço, de 0 (distante) a 1 (íntimo).

prox = np.linspace(0, 1, 200)

# 1) Alcance novo (Granovetter): quanto de gente FORA da sua bolha aquele laço
#    alcança. Cai com a proximidade (curva ilustrativa da lógica de redes).
alcance_novo = np.exp(-3.0 * prox)          # alto no distante, baixo no próximo

# 2) Benefício de oportunidade (Rajkumar): NÃO é monotônico. O melhor laço para
#    uma porta nova não é o mais forte nem o mais fraco, e sim o MODERADAMENTE
#    fraco (curva em U invertido, ilustrando o achado causal).
forca = 1 - prox                            # força do laço (0=fraco, 1=forte)
beneficio = np.exp(-((forca - 0.30) ** 2) / (2 * 0.18 ** 2))
pico = forca[np.argmax(beneficio)]

print("=" * 66)
print("  MODELO — POR QUE LONGE ESPALHA E PERTO SILENCIA")
print("=" * 66)
print(f"\n  Alcance para fora da bolha:")
print(f"  → o laço distante alcança muito mais gente nova que o")
print(f"     íntimo (a curva ilustra a lógica, sem número exato).")
print(f"\n  Melhor laço para uma oportunidade (Rajkumar):")
print(f"  → pico em força ≈ {pico:.2f}  (moderadamente fraco), não no mais forte.")
print(f"\n  O silêncio de perto = duas mecânicas de rede, nenhuma sobre afeto:")
print(f"  → redundância (já estão na sua bolha)")
print(f"  → custo social de endossar um par publicamente")
print("=" * 66)
```

```python
# --- VISUALIZAÇÃO — 3 GRÁFICOS SEPARADOS (fundo branco, dpi=150) ---

# ── GRÁFICO 1 — Alcance novo cai com a proximidade (Granovetter) ──
fig1, ax1 = plt.subplots(figsize=(10, 6))
ax1.plot(prox, alcance_novo, color=VERDE, linewidth=3)
ax1.fill_between(prox, alcance_novo, color=VERDE, alpha=0.12)
ax1.set_xlabel('Proximidade do laço  (0 = distante,  1 = íntimo)')
ax1.set_ylabel('Alcance para fora da sua bolha')
ax1.set_yticks([])
ax1.scatter([0.05, 0.95], [alcance_novo[10], alcance_novo[-6]], color=DOURADO, s=80, zorder=5)
ax1.text(0.06, alcance_novo[10], '  quem está longe\n  é ponte para redes novas', fontsize=10, color=VERMELHO, va='center')
ax1.text(0.62, alcance_novo[-30]+0.06, 'quem está perto já\nestá dentro da bolha', fontsize=10, color=CINZA)
ax1.set_title('Por Que Longe Espalha Mais\nO laço distante alcança gente que você nunca alcançaria sozinha',
              fontsize=13, pad=15)
plt.figtext(0.5, 0.005, 'Lógica da força dos laços fracos, Granovetter 1973 (curva ilustrativa) | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout(); plt.savefig('dia-054-grafico-01-alcance.png', dpi=150, bbox_inches='tight'); plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 — A ASSINATURA: o melhor laço é o moderadamente fraco (Rajkumar) ──
fig2, ax2 = plt.subplots(figsize=(10, 6))
ax2.plot(forca, beneficio, color=VERMELHO, linewidth=3)
ax2.fill_between(forca, beneficio, color=VERMELHO, alpha=0.10)
ax2.axvline(pico, color=DOURADO, linestyle='--', linewidth=2)
ax2.scatter([pico], [beneficio.max()], color=DOURADO, s=90, zorder=5)
ax2.text(0.55, beneficio.max()*0.82, 'laço moderadamente\nfraco: o mais útil', fontsize=10, color=DOURADO)
ax2.set_xlabel('Força do laço  (0 = fraco,  1 = forte)')
ax2.set_ylabel('Ganho de oportunidade (novas portas)')
ax2.set_yticks([])
ax2.set_title('O Melhor Laço Para Uma Porta Nova Não É o Mais Forte\n'
              'Curva em U invertido: pico nos laços moderadamente fracos',
              fontsize=13, pad=15)
plt.figtext(0.5, 0.005, 'Forma qualitativa do achado causal de Rajkumar et al., 2022, Science (~20 mi pessoas) | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout(); plt.savefig('dia-054-grafico-02-oportunidade.png', dpi=150, bbox_inches='tight'); plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 — O apoio de perto é real, mas invisível nas métricas ──
fig3, ax3 = plt.subplots(figsize=(10, 6))
grupos = ['Pessoas\npróximas', 'Pessoas\ndistantes']
visivel   = [25, 75]   # curtida/comentário público (ilustrativo)
invisivel = [75, 25]   # viu, mandou mensagem, comentou pessoalmente (ilustrativo)
x = np.arange(len(grupos))
ax3.bar(x, visivel, width=0.5, color=DOURADO, label='Apoio visível (curtida pública)')
ax3.bar(x, invisivel, width=0.5, bottom=visivel, color=VERDE, alpha=0.55,
        label='Apoio invisível (viu, mandou mensagem)')
ax3.set_xticks(x); ax3.set_xticklabels(grupos)
ax3.set_ylabel('Composição do apoio (%)')
ax3.set_ylim(0, 100)
ax3.legend(loc='upper center', frameon=False, fontsize=10, ncol=1)
ax3.set_title('O Apoio de Quem Está Perto É Real, Só Não Aparece\nAs métricas públicas contam curtidas, não quem te assistiu calado',
              fontsize=13, pad=15)
plt.figtext(0.5, 0.005, 'Decomposição ilustrativa · comparação social e apoio invisível | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout(); plt.savefig('dia-054-grafico-03-invisivel.png', dpi=150, bbox_inches='tight'); plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**O engajamento de quem está longe não é sinal de que eles torcem mais. É sinal
de que estão melhor posicionados para espalhar: laço distante é ponte, laço
próximo é bolha.**
**E o silêncio de quem está perto quase nunca é sobre o quanto te querem bem.**

Quem está longe carrega o seu trabalho para fora da sua rede, porque conhece
gente que você não conhece. Quem está perto já está dentro da sua bolha, então
não tem para onde levar a novidade, e ainda paga um custo social maior para
endossar publicamente um par. Duas mecânicas de rede, zero veredito sobre afeto.

E tem a parte que as métricas escondem: muita gente próxima apoia de um jeito
que não vira número. Assistiu tudo, mandou uma mensagem no privado, comentou com
você pessoalmente. As plataformas contam curtidas, não quem torceu em silêncio.

Não é que os de perto não torçam. É que a torcida deles raramente aparece no
lugar onde você foi procurar.

A pergunta que fica:

*Quanto do apoio que você não viu aconteceu num lugar que a métrica não conta?*

---

### ⚠️ Limitações do Modelo

- Granovetter (1973) e Rajkumar et al. (2022) tratam de difusão de informação e de empregos, não especificamente de curtidas de amigos nas redes. Aplicar a lógica ao seu caso é uma extrapolação razoável, não uma medição direta.
- As curvas dos gráficos 1 e 2 são ilustrativas da forma dos achados (queda com a proximidade; U invertido de Rajkumar), não ajustes a microdados.
- A decomposição do gráfico 3 (visível x invisível) é ilustrativa: mostra o mecanismo do apoio que não vira métrica, sem afirmar proporções exatas.
- A camada de comparação social é um mecanismo plausível e documentado, não um efeito medido neste dia.
- ×0.80 não aplicado: não há proporção de survey autorrelatado no núcleo do dia.

*A ciência é honesta sobre o que não sabe. O modelo também.*
