### 📖 A História

Sou a caçula, e acho que sou a mais criativa da família.

Não sei se cheguei a essa conclusão sozinha ou se me contaram isso tantas vezes que
virou verdade. E é exatamente essa a pergunta que dá para testar com dado.

A crença é uma das mais espalhadas do mundo. O primogênito é o responsável, o do meio é
o negociador, o caçula é o criativo e o rebelde. Aparece em livro de psicologia
popular, em conversa de família, em piada de irmão.

E ela tem uma qualidade rara: é específica o bastante para ser medida.

---

### 📚 O Conceito: o problema que torna isso difícil de medir

Comparar primogênitos com caçulas parece simples, e não é.

Se você pega mil pessoas ao acaso e compara os primogênitos com os caçulas, está
comparando **famílias diferentes**. E famílias com muitos filhos são sistematicamente
diferentes de famílias com poucos: renda, escolaridade dos pais, época, religião,
região.

Como todo caçula de três vem de uma família de pelo menos três filhos, e um filho único
nunca é caçula, a comparação entre famílias mistura ordem de nascimento com **tamanho
da família** e com tudo que vem junto.

A solução é comparar **irmãos dentro da mesma família**. Aí renda, pais, casa e época
são iguais, e o que sobra é a posição.

Fazer isso exige amostra gigante, porque é preciso ter várias famílias com vários
irmãos medidos. Foi isso que faltou por décadas.

---

### 🧮 O Modelo

Três painéis nacionais, com desenho dentro da família e entre famílias.

**Fontes:**
- Rohrer, J. M., Egloff, B. & Schmukle, S. C., 2015 · "Examining the effects of birth
  order on personality" · *PNAS* 112(46), 14224-14229 · três amostras representativas
  nacionais: **5.240 nos Estados Unidos**, **4.489 na Grã-Bretanha** e **10.457 na
  Alemanha**, total de **20.186** · análises **dentro da família** e **entre famílias**
- **Achado:** primogênitos pontuam **ligeiramente mais** em inteligência medida
  objetivamente e em **intelecto autorrelatado**
- **Ausência de efeito** em **extroversão**, **estabilidade emocional**,
  **amabilidade**, **conscienciosidade** e **imaginação**
- Damian, R. I. & Roberts, B. W., 2015 · estudo companheiro com cerca de **272.000
  estudantes** americanos do Project Talent · o efeito médio da ordem de nascimento
  sobre personalidade equivale a uma **correlação de 0,02**
- Rohrer, J. M. e colegas, 2017 · *Psychological Science* · análise de curva de
  especificação em traços estreitos, com pouco efeito encontrado
- Trabalho posterior com o modelo HEXACO confirma a ausência de efeito, com exceção do
  domínio ligado a intelecto

**Nota metodológica sobre o fator ×0.80:** não se aplica. Inteligência medida por teste
objetivo, e personalidade por escalas validadas em painéis nacionais representativos,
com comparação dentro da mesma família.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

RNG = np.random.default_rng(42)

print("Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Rohrer, Egloff & Schmukle, 2015, PNAS 112(46), 14224-14229

AMOSTRAS = {'Estados Unidos': 5_240, 'Grã-Bretanha': 4_489, 'Alemanha': 10_457}
n_total = sum(AMOSTRAS.values())

# Efeito encontrado por desfecho
DESFECHOS = {
    'Inteligência (teste objetivo)': 'efeito LIGEIRO a favor do primogênito',
    'Intelecto autorrelatado':       'efeito LIGEIRO a favor do primogênito',
    'Extroversão':                   'sem efeito',
    'Estabilidade emocional':        'sem efeito',
    'Amabilidade':                   'sem efeito',
    'Conscienciosidade':             'sem efeito',
    'Imaginação':                    'sem efeito',
}

# Damian & Roberts, 2015 - estudo companheiro
n_project_talent = 272_000
r_medio = 0.02       # efeito medio da ordem de nascimento sobre personalidade

aplica_fator_080 = False

print("=" * 70)
print("  DADOS - ORDEM DE NASCIMENTO E PERSONALIDADE")
print("=" * 70)
print(f"\n  Rohrer, Egloff & Schmukle, 2015 (PNAS 112(46)):")
for pais, n in AMOSTRAS.items():
    print(f"  -> {pais:<16} N = {n:,}".replace(",", "."))
print(f"  -> TOTAL:          N = {n_total:,}".replace(",", "."))
print(f"  -> Analises DENTRO da familia e ENTRE familias")
print(f"\n  {'desfecho':<32} resultado")
print(f"  {'-'*32} {'-'*40}")
for nome, res in DESFECHOS.items():
    print(f"  {nome:<32} {res}")
sem_efeito = [k for k, v in DESFECHOS.items() if v == 'sem efeito']
com_efeito = [k for k, v in DESFECHOS.items() if v != 'sem efeito']
print(f"\n  -> Desfechos COM efeito: {len(com_efeito)}")
print(f"  -> Desfechos SEM efeito: {len(sem_efeito)}")
print(f"\n  Damian & Roberts, 2015 (Project Talent):")
print(f"  -> N = {n_project_talent:,}".replace(",", ".") + " estudantes")
print(f"  -> Efeito medio sobre personalidade: r = {r_medio:.2f}")
print(f"\n  Fator x0.80 aplicado: {aplica_fator_080}")
print("=" * 70)
```

```python
# --- O MODELO ---
# Assinatura: traduzir r = 0,02 em coisas que dao para imaginar.

# 1) Variancia explicada
var_explicada = r_medio ** 2 * 100

# 2) Probabilidade de superioridade: sorteando um primogenito e um cacula,
#    qual a chance de o primogenito pontuar mais alto no traco?
#    Conversao padrao entre r e d: d = 2r / sqrt(1 - r^2)
d_equivalente = 2 * r_medio / np.sqrt(1 - r_medio ** 2)
p_superioridade = stats.norm.cdf(d_equivalente / np.sqrt(2))

# 3) Quantos pares seria preciso observar para notar a diferenca?
#    Ou seja, a cada 100 duplas de irmaos, quantas "acertam" a previsao?
acertos_em_100 = p_superioridade * 100

# 4) Quantas pessoas para detectar r = 0,02 com 80% de poder
ALFA, PODER = 0.05, 0.80


def poder_correlacao(r, n, alfa=ALFA):
    if r <= 0 or n <= 3:
        return alfa
    z = 0.5 * np.log((1 + r) / (1 - r))
    ep = 1 / np.sqrt(n - 3)
    critico = stats.norm.ppf(1 - alfa / 2)
    p = stats.norm.sf(critico - abs(z) / ep) + stats.norm.cdf(-critico - abs(z) / ep)
    return float(np.nan_to_num(p, nan=0.0))


n_necessario = None
for n in range(100, 200_000, 100):
    if poder_correlacao(r_medio, n) >= PODER:
        n_necessario = n
        break

print("=" * 70)
print("  MODELO - O QUE SIGNIFICA UM EFEITO DE r = 0,02")
print("=" * 70)
print(f"\n  1) Variancia explicada:")
print(f"  -> r = {r_medio:.2f} explica {var_explicada:.2f}% da variacao do traco.")
print(f"  -> Ou seja, {100 - var_explicada:.2f}% vem de outra coisa.")
print(f"\n  2) Probabilidade de superioridade:")
print(f"  -> Sorteando um primogenito e um cacula ao acaso,")
print(f"     a chance de o primogenito pontuar mais alto e de"
      f" {p_superioridade*100:.1f}%.")
print(f"  -> Se nao houvesse efeito nenhum, seriam 50,0%.")
print(f"  -> A cada 100 duplas de irmaos, cerca de {acertos_em_100:.0f}")
print(f"     seguiriam a previsao, contra 50 no puro acaso.")
print(f"\n  3) Quanta gente e necessaria para VER esse efeito:")
print(f"  -> Para detectar r = {r_medio:.2f} com {PODER*100:.0f}% de poder,")
print(f"     seriam necessarias cerca de {n_necessario:,}".replace(",", ".") +
      " pessoas.")
print(f"  -> E por isso que o efeito so aparece em estudos com centenas")
print(f"     de milhares de participantes, e nunca na sua familia.")
print(f"\n  4) A leitura honesta:")
print(f"  -> Nao e que o efeito seja ZERO. Em inteligencia ele existe.")
print(f"  -> E que ele e pequeno demais para ser percebido em qualquer")
print(f"     familia real, com dois, tres ou quatro filhos.")
print(f"  -> A cren\u00e7a e forte. O efeito e minusculo. Os dois sao verdade.")
print("=" * 70)
```

```python
# --- VISUALIZACAO ---

DOURADO = '#c8a84b'
VERMELHO = '#c0392b'
VERDE = '#2a8a82'
CINZA = '#6b6a64'

# GRAFICO 1 - Onde apareceu e onde nao apareceu
fig1, ax1 = plt.subplots(figsize=(12, 8))

nomes = list(DESFECHOS.keys())
y = np.arange(len(nomes))[::-1]
for yi, nome in zip(y, nomes):
    tem = DESFECHOS[nome] != 'sem efeito'
    cor = DOURADO if tem else CINZA
    largura = 1.0 if tem else 0.12
    ax1.barh([yi], [largura], color=cor, alpha=0.88, height=0.55)
    texto = 'efeito ligeiro' if tem else 'sem efeito'
    ax1.text(largura + 0.04, yi, texto, va='center', fontsize=13, color=cor,
             fontweight='bold')

ax1.set_yticks(y)
ax1.set_yticklabels(nomes, fontsize=12)
ax1.set_xlim(0, 1.7)
ax1.set_xticks([])
ax1.grid(False)
ax1.set_title('Sete desfechos medidos. Dois com efeito, cinco sem.\n'
              'Rohrer, Egloff & Schmukle, 2015 · N=' +
              f'{n_total:,}'.replace(",", ".") +
              ' em três países',
              fontsize=15, pad=18)
ax1.text(0.5, -0.09,
         'Nenhum dos traços de personalidade da crença popular apareceu. '
         'O que sobrou foi inteligência, e ligeiramente.',
         transform=ax1.transAxes, ha='center', fontsize=12, color=CINZA, style='italic')

plt.figtext(0.5, 0.005,
            'Fonte: Rohrer, Egloff & Schmukle, 2015, PNAS 112(46), 14224-14229'
            '  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-082-grafico-01-desfechos.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 1 salvo")

# GRAFICO 2 - Assinatura: o que r = 0,02 quer dizer, em 100 duplas de irmaos
fig2, ax2 = plt.subplots(figsize=(12, 8))

lado = 10
acertos = int(round(acertos_em_100))
for i in range(100):
    linha, col = divmod(i, lado)
    segue = i < acertos
    cor = DOURADO if segue else CINZA
    alpha = 0.85 if segue else 0.22
    ax2.add_patch(plt.Rectangle((col, -linha), 0.84, 0.84, facecolor=cor,
                                alpha=alpha, edgecolor='white', linewidth=2))

ax2.set_xlim(-0.4, lado + 0.4)
ax2.set_ylim(-lado - 1.6, 1.9)
ax2.set_xticks([]); ax2.set_yticks([])
ax2.grid(False)
for spine in ax2.spines.values():
    spine.set_visible(False)

ax2.text(lado / 2, 1.4,
         f'{acertos} em cada 100 duplas de irmãos seguem a previsão',
         ha='center', fontsize=17, fontweight='bold', color=DOURADO)
ax2.set_title('A assinatura estatística: o que significa r = 0,02\n'
              'Se não houvesse efeito nenhum, seriam 50 em 100',
              fontsize=15, pad=18)
ax2.text(lado / 2, -lado - 0.5,
         f'A diferença entre {acertos} e 50 é o tamanho real da crença mais repetida '
         f'do mundo sobre irmãos.\n'
         f'Ela explica {var_explicada:.2f}'.replace('.', ',') +
         '% da variação do traço.',
         ha='center', fontsize=12, color=CINZA, style='italic')

plt.figtext(0.5, 0.005,
            'Cálculo próprio sobre o efeito médio publicado por Damian & Roberts, 2015'
            '  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-082-grafico-02-cem-duplas.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 2 salvo")

# GRAFICO 3 - Por que nenhuma familia consegue ver isso
fig3, ax3 = plt.subplots(figsize=(12, 8))

ns = np.array([2, 5, 10, 50, 500, 5_000, 50_000, n_necessario])
poderes = [poder_correlacao(r_medio, n) * 100 for n in ns]

ax3.plot(ns, poderes, marker='o', markersize=11, linewidth=3, color=VERDE)
ax3.axhline(y=80, color=DOURADO, linestyle='--', linewidth=2)
ax3.set_xscale('log')

for n, p in zip(ns, poderes):
    if n in (2, 50, n_necessario):
        rot = 'sua família' if n == 2 else ('uma escola' if n == 50 else
                                            f'{n:,}'.replace(",", ".") + ' pessoas')
        ax3.text(n, p + 5, f'{rot}\n{p:.0f}%', ha='center', fontsize=12,
                 fontweight='bold', color=VERDE)

ax3.set_xlim(1.4, n_necessario * 2.4)
ax3.set_ylim(0, 105)
ax3.set_xlabel('Número de pessoas observadas (escala log)')
ax3.set_ylabel('Chance de enxergar um efeito de r = 0,02 (%)')
ax3.set_title('Por que ninguém consegue ver isso na própria família\n'
              'Seriam necessárias cerca de ' +
              f'{n_necessario:,}'.replace(",", ".") +
              ' pessoas para detectar o efeito',
              fontsize=15, pad=18)
ax3.text(0.5, -0.135,
         'O efeito existe e é pequeno demais para ser percebido em qualquer família '
         'real. A crença não veio da observação.',
         transform=ax3.transAxes, ha='center', fontsize=12, color=DOURADO,
         fontweight='bold')

plt.figtext(0.5, 0.005,
            'Cálculo de poder por transformação z de Fisher, alfa = 0,05'
            '  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-082-grafico-03-poder.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 3 salvo")
```

### 💡 O Insight

Vinte mil cento e oitenta e seis pessoas, em três países, com painéis nacionais
representativos. E o cuidado que faltava nos estudos antigos: comparar **irmãos dentro
da mesma família**, para que renda, pais, casa e época fossem iguais.

**Extroversão: nenhum efeito. Estabilidade emocional: nenhum. Amabilidade: nenhum.
Conscienciosidade: nenhum. Imaginação: nenhum.**

Sobrou um efeito, e ele é real: primogênitos pontuam **ligeiramente mais** em
inteligência medida por teste e em intelecto autorrelatado.

Repare que "imaginação" é o traço mais próximo do que a gente chama de criatividade. E
ele está na lista dos que não apareceram.

Num estudo companheiro, com cerca de **272 mil** estudantes americanos, o efeito médio
da ordem de nascimento sobre personalidade equivale a uma correlação de **0,02**.

Traduzi esse número, porque 0,02 não diz nada sozinho.

Ele explica **0,04% da variação** do traço. Sorteando um primogênito e um caçula ao
acaso, a chance de o primogênito pontuar mais alto é de **50,6%**, contra 50% do puro
acaso.

Ou seja: **a cada 100 duplas de irmãos, cerca de 51 seguem a previsão.** Cinquenta e
um contra cinquenta.

E aí fiz a conta que eu queria ter feito desde o começo: quanta gente seria necessária
para **enxergar** um efeito desse tamanho.

A resposta é cerca de **20 mil pessoas**.

Na sua família, com dois ou três irmãos, a chance de detectar isso é de aproximadamente
5%. Ou seja, nenhuma.

**A crença não veio da observação.** Não tem como ter vindo. Ninguém, em nenhuma
família da história, teve amostra suficiente para notar meio ponto percentual de
diferença.

Ela veio de outro lugar: de repetição, de narrativa, e do fato de que rótulo de família
é profecia que se cumpre sozinha. Se todo mundo trata o caçula como o criativo desde os
cinco anos, ele recebe tinta, elogio, permissão e desculpa. E aí vira.

O que me diverte é que eu abri este texto dizendo que sou a caçula e acho que sou a
mais criativa da família.

Continuo achando. Só parei de achar que isso tem a ver com a ordem em que eu nasci.

### ⚠️ Limitações do Modelo

- **O efeito não é zero.** Em inteligência medida objetivamente e em intelecto
  autorrelatado, ele existe e aparece de forma consistente. O que não existe é o efeito
  sobre os traços da crença popular.
- **As traduções do r são minhas, não do artigo.** A variância explicada, a
  probabilidade de superioridade e a conta das 100 duplas derivam do efeito médio
  publicado, usando a conversão padrão entre correlação e d de Cohen, sob normalidade.
  São leituras de ordem de grandeza.
- **O cálculo de quanta gente é necessária** usa transformação z de Fisher para uma
  correlação simples, enquanto os estudos usam modelos com covariáveis. Serve para
  mostrar a ordem de grandeza, não para reproduzir a análise original.
- **Personalidade medida por autorrelato** nas escalas de traço, ainda que
  inteligência tenha sido medida por teste objetivo.
- **Amostras de três países ocidentais de renda alta.** Efeitos de ordem de nascimento
  podem variar em contextos com estruturas familiares e expectativas diferentes.
- **Ordem de nascimento não é o mesmo que tamanho da família**, e a análise dentro da
  família resolve boa parte, mas não toda, a confusão entre as duas coisas.
- **Nada aqui explica de onde veio a crença.** A hipótese da profecia que se cumpre é
  interpretação minha, plausível e não testada por estes estudos.
- Fator ×0.80 não aplicado.

*A ciência é honesta sobre o que não sabe. O modelo também.*
