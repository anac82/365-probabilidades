# 365 Probabilidades · Dia #065
## Qual a probabilidade de uma lembrança sua nunca ter acontecido?

**Tipo:** Comportamental
**Data de publicação:** 2026-08-17
**Ferramenta:** Python
**Decisão analisada:** Posso tratar a minha lembrança como prova do que aconteceu?
**Hashtag:** #365Probabilidades #Dia065

---

### 📖 A História

Existe um tipo de discussão que toda família conhece. Duas pessoas contam a mesma
cena de infância de jeitos incompatíveis, cada uma com certeza absoluta, e não tem
como as duas estarem certas.

Normalmente isso termina em empate, porque não existe árbitro. A cena não foi
filmada e o que sobrou são duas versões igualmente vívidas.

O que quase ninguém considera é a terceira possibilidade: que a cena não tenha
acontecido de jeito nenhum, para nenhuma das duas.

Parece exagero, mas é uma pergunta com desenho experimental. Dá para pegar
voluntários, combinar com a família deles um evento de infância que comprovadamente
nunca ocorreu, sugerir esse evento em entrevistas, e depois contar quantos passam a
lembrar dele.

Esse experimento foi feito muitas vezes. E as taxas variavam tanto de estudo para
estudo que a própria literatura ficou desacreditada, até alguém resolver o problema
de um jeito elegante.

---

### 📚 O Conceito: por que os números não batiam

O problema não era a memória. Era a régua.

Cada laboratório definia "falsa memória" de um jeito. Uns exigiam que a pessoa
inventasse detalhes novos. Outros aceitavam que ela apenas concordasse com o evento
sugerido. Comparar essas taxas era comparar coisas diferentes com o mesmo nome, e
foi assim que surgiram estimativas que iam de quase zero a mais da metade.

Em 2017, sete pesquisadores fizeram o que quase ninguém faz: juntaram os
**relatos originais, caso a caso**, de oito estudos publicados, e recodificaram
tudo com um único sistema de sete critérios.

Isso não é meta-análise. Meta-análise combina resultados. Isto é mega-análise:
combina os dados brutos. É mais trabalhoso, mais raro, e responde a pergunta que a
meta-análise não conseguia responder.

---

### 🧮 O Modelo

Uma fonte central com contagem e denominador reais, mais dois contrapesos que
impedem a leitura sensacionalista.

**Fontes:**
- Scoboria, A., Wade, K. A., Lindsay, D. S., Azad, T., Strange, D., Ost, J. &
  Hyman, I. E., 2017 · *Memory* 25(2), 146-163 · mega-análise dos relatos de
  **oito estudos** de implantação, **N=423** · codificação por avaliadores
  independentes em sete critérios · **30,4% classificados como falsa memória** e
  outros **23% como aceitação do evento em algum grau** · na condição com
  informação autorrelevante, procedimento de imaginação e **sem** foto do evento,
  a taxa foi de **46,1%**
- Moderador com intervalo publicado, dentro da mesma mega-análise:
  **39,6% [IC 95%: 32 · 47]** com informação autorrelevante contra
  **24,0% [19 · 30]** sem
- Pezdek, K. et al., 1997 · citado e reanalisado dentro da mega-análise · taxa de
  **0%** quando o evento sugerido era de baixa plausibilidade e a entrevista foi
  conduzida por um irmão, e não por pesquisador treinado
- Brewin, C. R. & Andrews, B., 2017 · revisão da mesma literatura com estimativa
  bem mais cautelosa · a crítica de Scoboria é que a revisão agrupou estudos com
  definições distintas de falsa memória e calculou média simples

**Nota metodológica sobre o fator ×0.80:** não se aplica. São transcrições de
entrevista codificadas por avaliadores independentes segundo critérios declarados,
em laboratório. Não é proporção de survey populacional autorrelatado.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---

# Scoboria, Wade, Lindsay, Azad, Strange, Ost & Hyman, 2017, Memory 25(2), 146-163
# Mega-analise: dados caso a caso de oito estudos de implantacao, recodificados
n_total          = 423
k_estudos        = 8
criterios        = 7

p_falsa_memoria  = 0.304   # classificados como falsa memoria
p_aceitou        = 0.23    # aceitaram o evento em algum grau
p_condicao_max   = 0.461   # autorrelevante + imaginacao + SEM foto

# Moderador com IC 95% publicado dentro da propria mega-analise
p_com_autorrel   = 0.396; ic_com_autorrel = (0.32, 0.47)
p_sem_autorrel   = 0.240; ic_sem_autorrel = (0.19, 0.30)

# Contrapesos
p_pezdek_zero    = 0.00    # baixa plausibilidade + entrevista por irmao

aplica_fator_080 = False

# Aceitacao total: falsa memoria completa + aceitacao parcial
p_aceitacao_total = p_falsa_memoria + p_aceitou

print("=" * 68)
print("  DADOS DA LITERATURA - MEMORIA IMPLANTADA")
print("=" * 68)
print(f"\n  Scoboria et al., 2017 (Memory 25(2)):")
print(f"  -> {k_estudos} estudos recodificados · N={n_total} · {criterios} criterios")
print(f"  -> Falsa memoria completa:          {p_falsa_memoria*100:.1f}%")
print(f"  -> Aceitaram o evento em algum grau: {p_aceitou*100:.0f}%")
print(f"  -> Aceitacao total:                 {p_aceitacao_total*100:.1f}%")
print(f"  -> Condicao mais favoravel a sugestao: {p_condicao_max*100:.1f}%")
print(f"     (informacao autorrelevante + imaginacao + SEM foto)")
print(f"\n  Moderador com IC 95% publicado:")
print(f"  -> Com informacao autorrelevante: {p_com_autorrel*100:.1f}%"
      f"  IC 95% [{ic_com_autorrel[0]*100:.0f}% · {ic_com_autorrel[1]*100:.0f}%]")
print(f"  -> Sem informacao autorrelevante: {p_sem_autorrel*100:.1f}%"
      f"  IC 95% [{ic_sem_autorrel[0]*100:.0f}% · {ic_sem_autorrel[1]*100:.0f}%]")
print(f"\n  Contrapeso (Pezdek et al., 1997):")
print(f"  -> Taxa de {p_pezdek_zero*100:.0f}% com evento implausivel e entrevista por irmao")
print(f"\n  Fator x0.80 aplicado: {aplica_fator_080}")
print("=" * 68)
```

```python
# --- O MODELO ---
# Assinatura estatistica: distribuicao Beta sobre contagem e denominador reais.
#     Beta(alpha, beta)  com  alpha = p x N  e  beta = (1 - p) x N

casos_fm  = p_falsa_memoria * n_total          # casos classificados como falsa memoria
alpha_fm  = casos_fm
beta_fm   = n_total - casos_fm
dist_fm   = stats.beta(alpha_fm, beta_fm)
ic_fm     = dist_fm.interval(0.95)

# Aceitacao total (falsa memoria + aceitacao parcial) sobre o mesmo denominador
casos_ac  = p_aceitacao_total * n_total
dist_ac   = stats.beta(casos_ac, n_total - casos_ac)
ic_ac     = dist_ac.interval(0.95)

# Diferenca entre os dois bracos do moderador, com os IC publicados.
# Os dois IC nao se sobrepoem, o que ja indica diferenca; a distancia pontual e:
dif_moderador = p_com_autorrel - p_sem_autorrel

# Traducao em pessoas, que e como o numero se le fora da estatistica
em_dez = p_falsa_memoria * 10

print("=" * 68)
print("  MODELO - QUANTAS LEMBRANCAS NASCERAM DE UMA SUGESTAO")
print("=" * 68)
print(f"\n  Beta com IC 95% (N={n_total}, contagem e denominador reais):")
print(f"  -> Falsa memoria completa: {p_falsa_memoria*100:.1f}%"
      f"  IC 95% [{ic_fm[0]*100:.1f}% · {ic_fm[1]*100:.1f}%]")
print(f"     casos = {casos_fm:.0f} de {n_total}")
print(f"  -> Aceitacao total:        {p_aceitacao_total*100:.1f}%"
      f"  IC 95% [{ic_ac[0]*100:.1f}% · {ic_ac[1]*100:.1f}%]")
print(f"\n  Leitura em pessoas:")
print(f"  -> Cerca de {em_dez:.0f} em cada 10 participantes passaram a relatar")
print(f"     uma lembranca de um evento que nunca aconteceu")
print(f"\n  O que muda a taxa (moderador da propria mega-analise):")
print(f"  -> Com informacao autorrelevante: {p_com_autorrel*100:.1f}%")
print(f"  -> Sem informacao autorrelevante: {p_sem_autorrel*100:.1f}%")
print(f"  -> Distancia: {dif_moderador*100:.1f} pontos percentuais")
print(f"  -> Os dois IC 95% publicados nao se sobrepoem")
print(f"\n  O que derruba a taxa a zero:")
print(f"  -> Evento implausivel + entrevista por familiar, nao por pesquisador")
print("=" * 68)
```

```python
# --- VISUALIZACAO ---

DOURADO = '#c8a84b'
VERMELHO = '#c0392b'
VERDE = '#2a8a82'
CINZA = '#6b6a64'

# GRAFICO 1 - O que aconteceu com os 423 casos
fig1, ax1 = plt.subplots(figsize=(12, 6.5))

fatias = [p_falsa_memoria, p_aceitou, 1 - p_aceitacao_total]
rotulos = ['Falsa memória\ncompleta', 'Aceitou o evento\nem algum grau',
           'Rejeitou\na sugestão']
cores = [VERMELHO, DOURADO, VERDE]

esquerda = 0
for fatia, rot, cor in zip(fatias, rotulos, cores):
    ax1.barh([0], [fatia * 100], left=esquerda * 100, color=cor, alpha=0.9, height=0.5)
    ax1.text(esquerda * 100 + fatia * 50, 0,
             f'{fatia*100:.1f}%'.replace('.', ','),
             ha='center', va='center', fontsize=22, fontweight='bold', color='white')
    ax1.text(esquerda * 100 + fatia * 50, -0.42, rot, ha='center', va='top',
             fontsize=12, color=cor)
    esquerda += fatia

ax1.set_xlim(0, 100)
ax1.set_ylim(-1.1, 0.75)
ax1.set_yticks([])
ax1.set_xlabel('Proporção dos 423 casos recodificados (%)')
ax1.grid(False)
ax1.set_title('O que aconteceu com os 423 participantes\n'
              f'Scoboria et al., 2017 · {k_estudos} estudos recodificados com '
              f'{criterios} critérios',
              fontsize=14, pad=18)
ax1.text(50, 0.5,
         f'{p_aceitacao_total*100:.1f}% aceitaram, em algum grau, um evento que nunca '
         f'aconteceu'.replace('.', ','),
         ha='center', fontsize=13, color='#333', style='italic')

plt.figtext(0.5, 0.005,
            'Fonte: Scoboria et al., 2017, Memory 25(2), 146-163  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-065-grafico-01-o-que-aconteceu.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 1 salvo")

# GRAFICO 2 - O que faz a taxa subir e o que faz cair
fig2, ax2 = plt.subplots(figsize=(12, 8))

nomes = ['Evento implausível,\nentrevista por um irmão\n(Pezdek et al., 1997)',
         'Sem informação\nautorrelevante',
         'Com informação\nautorrelevante',
         'Autorrelevante + imaginação\ne SEM foto do evento']
valores = [p_pezdek_zero, p_sem_autorrel, p_com_autorrel, p_condicao_max]
ics = [None, ic_sem_autorrel, ic_com_autorrel, None]
cores2 = [VERDE, VERDE, DOURADO, VERMELHO]
y = np.arange(len(valores))

for yi, val, ic, cor in zip(y, valores, ics, cores2):
    ax2.barh([yi], [val * 100], color=cor, alpha=0.85, height=0.5)
    if ic is not None:
        ax2.plot([ic[0] * 100, ic[1] * 100], [yi, yi], color='#222', linewidth=2.5)
        ax2.plot([ic[0] * 100, ic[0] * 100], [yi - 0.09, yi + 0.09], color='#222', linewidth=2.5)
        ax2.plot([ic[1] * 100, ic[1] * 100], [yi - 0.09, yi + 0.09], color='#222', linewidth=2.5)
        ax2.text(ic[1] * 100 + 1.5, yi,
                 f'IC 95% [{ic[0]*100:.0f}% · {ic[1]*100:.0f}%]',
                 va='center', fontsize=11, color=CINZA)
    ax2.text(val * 100 + 1.2 if ic is None else 1.5, yi,
             f'{val*100:.1f}%'.replace('.', ','),
             va='center', ha='left', fontsize=17, fontweight='bold',
             color=cor if ic is None else 'white')

ax2.axvline(x=p_falsa_memoria * 100, color='#333', linestyle='--', linewidth=1.5)
ax2.text(p_falsa_memoria * 100 + 0.8, y.max() + 0.55,
         f'taxa geral: {p_falsa_memoria*100:.1f}%'.replace('.', ','),
         fontsize=12, color='#333')

ax2.set_yticks(y)
ax2.set_yticklabels(nomes, fontsize=11)
ax2.set_xlim(0, 62)
ax2.set_ylim(-0.6, y.max() + 0.8)
ax2.set_xlabel('Taxa de formação de falsa memória (%)')
ax2.set_title('A sugestão não é mágica: depende de plausibilidade, de quem pergunta\n'
              'e de haver ou não uma foto na mesa',
              fontsize=14, pad=18)

plt.figtext(0.5, 0.005,
            'Fontes: Scoboria et al., 2017 · Pezdek et al., 1997, conforme recodificados '
            'na mega-análise  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-065-grafico-02-moderadores.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 2 salvo")

# GRAFICO 3 - Assinatura estatistica: Beta com IC 95%
fig3, ax3 = plt.subplots(figsize=(12, 8))

x = np.linspace(0.20, 0.42, 3000)
y3 = dist_fm.pdf(x)

ax3.plot(x * 100, y3, color=VERMELHO, linewidth=3)
ax3.fill_between(x * 100, y3, alpha=0.18, color=VERMELHO)
dentro = (x >= ic_fm[0]) & (x <= ic_fm[1])
ax3.fill_between(x[dentro] * 100, y3[dentro], alpha=0.32, color=VERMELHO)

for limite in ic_fm:
    ax3.axvline(x=limite * 100, color=DOURADO, linestyle='--', linewidth=2)
ax3.axvline(x=p_falsa_memoria * 100, color='#333', linewidth=2)

topo = y3.max()
ax3.text(p_falsa_memoria * 100, topo * 1.05, '30,4%', ha='center',
         fontsize=17, fontweight='bold', color='#333')
ax3.text(p_falsa_memoria * 100, topo * 0.055,
         f'IC 95% [{ic_fm[0]*100:.1f}% · {ic_fm[1]*100:.1f}%]'.replace('.', ','),
         ha='center', fontsize=13, color=DOURADO, fontweight='bold')

ax3.set_ylim(0, topo * 1.24)
ax3.set_xlabel('Proporção classificada como falsa memória (%)')
ax3.set_ylabel('Densidade')
ax3.set_title('A assinatura estatística: Beta(p·N, (1−p)·N)\n'
              f'{casos_fm:.0f} casos em {n_total} · Scoboria et al., 2017',
              fontsize=14, pad=18)
ax3.text(0.5, -0.13,
         'Contagem e denominador reais: a Beta aqui não depende de nenhuma suposição extra',
         transform=ax3.transAxes, ha='center', fontsize=12, color=DOURADO,
         fontweight='bold')

plt.figtext(0.5, 0.005,
            'Fonte: Scoboria et al., 2017, Memory 25(2), 146-163  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-065-grafico-03-bayesiano.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 3 salvo")
```

### 💡 O Insight

**30,4% dos participantes passaram a lembrar de um evento que nunca aconteceu.**

Cerca de 3 em cada 10. E outros 23% aceitaram o evento em algum grau, sem chegar a
formar uma lembrança completa. Somando, **mais da metade dos 423 casos incorporou
alguma coisa que a família confirmou nunca ter existido.**

Mas o número que eu acho mais perturbador não é esse. É o que muda a taxa.

Quando a sugestão vem misturada com informação real sobre a sua vida, a taxa sobe
de 24% para 39,6%. E na condição mais favorável de todas, com detalhes seus, com um
pedido para você imaginar a cena, e **sem** foto do evento, chega a 46,1%.

A foto derruba a taxa. Você esperaria o contrário, porque foto parece prova. Só que
a foto verdadeira dá ao cérebro um material concreto para comparar, e a comparação
atrapalha a invenção. É a imaginação sem âncora que constrói a cena.

E tem o outro lado, que é o que impede este dia de virar sensacionalismo. Num dos
estudos recodificados, a taxa foi de **zero**: o evento sugerido era implausível e
quem conduzia a entrevista era um irmão, não um pesquisador. Sugestão não é mágica.
Ela precisa de plausibilidade e de alguém com autoridade para sustentá-la.

Nada disso diz que as suas lembranças são falsas. Diz outra coisa, mais
desconfortável e mais útil:

**a sensação de nitidez não é evidência.** Você não tem, por dentro, nenhum
mecanismo que separe a lembrança que aconteceu da que foi construída. As duas
chegam com a mesma cara.

Numa discussão de família isso é anedota. Num tribunal, num consultório, numa
conversa em que alguém insiste bastante e você começa a lembrar, é outra coisa.

*Qual das suas lembranças mais nítidas nunca foi confirmada por ninguém além de
você?*

### ⚠️ Limitações do Modelo

- Os 30,4% são a taxa **dentro de estudos de implantação**, com entrevistas
  repetidas, sugestão deliberada e corroboração falsa combinada com a família. Não
  é a prevalência de falsas memórias na vida cotidiana, e não deve ser lida assim.
- A mega-análise recodifica oito estudos com um único sistema, o que é a força do
  trabalho, mas os estudos originais diferiram em método de indução, plausibilidade
  do evento e perfil dos participantes.
- Existe leitura discordante da mesma literatura. Brewin e Andrews, em 2017,
  chegaram a uma estimativa bem mais baixa. A crítica de Scoboria é que a revisão
  agrupou estudos com definições diferentes de falsa memória. O desacordo é
  metodológico e continua aberto.
- Amostras majoritariamente de estudantes universitários, o que limita a
  generalização por idade e escolaridade.
- O braço do moderador entra com o intervalo publicado no artigo. Não recalculei
  intervalo sobre denominador que o artigo não publica.
- Falsa memória de evento de infância não é o mesmo objeto que memória de trauma
  vivido. Este dia não fala sobre isso, e a literatura clínica exige cuidado que
  uma peça de divulgação não comporta.
- Fator ×0.80 não aplicado: classificação de transcrições por avaliadores
  independentes em laboratório, não survey autorrelatado.

*A ciência é honesta sobre o que não sabe. O modelo também.*
