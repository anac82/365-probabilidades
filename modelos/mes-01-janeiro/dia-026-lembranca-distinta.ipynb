### 📖 A História

Há alguns anos, resolvi levar meus pais para viajar.

Queria dar pra eles experiências que a rotina nunca ia oferecer. Um lugar diferente. Coisas que nem passavam pela cabeça deles até ali.

Meu pai não tinha passaporte. Tirou o primeiro da vida pra essa viagem.

Levei ele pra ver neve pela primeira vez. Fomos pro sul do Chile, uma cidade chamada Puerto Varas.

Não foi uma viagem longa nem elaborada. Foi só neve. Uma coisa que boa parte do mundo vê todo ano e nem repara mais.

Só que aquilo ficou entalado na cabeça dele de um jeito que quase nada mais ficou. Ele contava a história de novo e de novo. Lembrava de detalhes que eu já tinha esquecido. Uma tarde, em meio a milhares de outras tardes da vida dele, virou uma das mais fortes que ele teve.

Eu não sabia, na hora, que estava fazendo exatamente o que a neurociência da memória já sabia fazer há décadas.

---

### 📚 O Conceito: Por Que o Novo Grava Mais Fundo

Em 1933, a psicóloga alemã Hedwig von Restorff fez um experimento simples: mostrou pra várias pessoas listas de itens parecidos entre si, com um item destacado no meio, diferente em cor, formato ou categoria.

O resultado se repetiu tantas vezes ao longo do século seguinte que virou um dos efeitos mais replicados da psicologia da memória: itens distintos são lembrados com quase o dobro da frequência de itens comuns. E o efeito não desaparece com a idade. Estudos mostram que ele permanece detectável até a velhice.

Só que o "por que isso acontece" só ficou claro décadas depois, com neuroimagem. Em 2006, pesquisadores do University College London mostraram que uma experiência genuinamente nova ativa uma região específica do cérebro, a substantia nigra/área tegmental ventral, que libera dopamina direto no hipocampo. Essa dopamina reduz o limiar necessário para uma memória virar permanente. É um circuito literal, ativado só por novidade, não por qualquer estímulo.

E o achado que mais me tocou: em 2012, o mesmo grupo testou justamente cérebros mais velhos, numa fase da vida em que a produção natural de dopamina já está em declínio, e mostrou que esse reforço ainda funciona. Um cérebro envelhecendo ainda grava com força total quando o estímulo é de verdade novo.

### 🧮 O Modelo

**Fontes:**
- Von Restorff, H. (1933) — "Über die Wirkung von Bereichsbildung im Spurenfeld", Psychologische Forschung, 18, 299-342 — efeito de distinção/isolamento, replicado inúmeras vezes desde então
- Bunzeck, N. & Düzel, E. (2006) — "Absolute coding of stimulus novelty in the human substantia nigra/VTA", Neuron, 51(3), 369-379
- Chowdhury, R., Guitart-Masip, M., Bunzeck, N., Dolan, R. J., & Düzel, E. (2012) — "Dopamine modulates episodic memory persistence in old age", Journal of Neuroscience, 32(41), 14193-14204


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---

# Von Restorff, H. (1933) e replicações posteriores
# Efeito de isolamento/distinção: itens distintos vs. itens comuns numa mesma lista
p_distintivo = 0.80   # taxa de recall de itens distintos/novos
p_comum = 0.40         # taxa de recall de itens comuns/rotineiros na mesma lista

# Fator de correção padrão do projeto
fator_correcao = 0.80

# Razão de força de memória (quantas vezes mais um evento distinto é lembrado)
razao_bruta = p_distintivo / p_comum

print("=" * 65)
print("  DADOS DA LITERATURA — POR QUE O NOVO GRAVA MAIS FUNDO")
print("=" * 65)
print(f"\n  Von Restorff (1933) e replicações:")
print(f"  → Recall de itens distintos/novos:     {p_distintivo*100:.0f}%")
print(f"  → Recall de itens comuns/rotineiros:    {p_comum*100:.0f}%")
print(f"  → Razão bruta de força de memória:      {razao_bruta:.1f}x")
print(f"\n  Bunzeck & Düzel (2006), Neuron 51:369-379:")
print(f"  → Novidade ativa dopamina (SN/VTA) → hipocampo")
print(f"  → Reduz o limiar de consolidação de longo prazo")
print(f"\n  Chowdhury et al. (2012), J Neurosci 32:14193-14204:")
print(f"  → Efeito replicado em população idosa")
print(f"  → Reforço dopaminérgico melhora persistência da memória")
print(f"     mesmo com ativação hipocampal fraca na codificação")
print("=" * 65)
```

```python
# --- O MODELO ---
# Probabilidade corrigida de uma experiência distinta virar memória duradoura

p_corrigido = p_distintivo * fator_correcao
razao_corrigida = p_corrigido / p_comum

# Ganho absoluto (pontos percentuais) entre comum e distinto, já corrigido
ganho_pp = (p_corrigido - p_comum) * 100

print("=" * 65)
print("  MODELO — A LEMBRANÇA QUE VOCÊ CRIA SEM SABER")
print("=" * 65)
print(f"\n  P(lembrar de uma experiência comum):              {p_comum*100:.0f}%")
print(f"  P(lembrar de uma experiência distinta/nova):      {p_distintivo*100:.0f}%")
print(f"\n  Com fator de correção 0.80:")
print(f"  → P(lembrar de uma experiência distinta), corrigido: {p_corrigido*100:.1f}%")
print(f"  → Razão corrigida de força de memória:                {razao_corrigida:.2f}x")
print(f"  → Ganho absoluto sobre o comum:                       +{ganho_pp:.1f}pp")
print(f"\n  RESULTADO CONCRETO:")
print(f"  → Mesmo depois de uma correção conservadora, uma experiência")
print(f"     genuinamente nova tem cerca de 60% mais chance de virar")
print(f"     lembrança duradoura do que um dia comum.")
print(f"  → Esse efeito persiste até a velhice (Bireta et al., 2008;")
print(f"     Chowdhury et al., 2012) — não é coisa de cérebro jovem.")
print("=" * 65)
```

```python
# --- VISUALIZAÇÃO ---

# GRÁFICO 1 — Recall: distinto vs comum (dado bruto da literatura)
fig1, ax1 = plt.subplots(figsize=(10, 6))
categorias = ['Experiência\ncomum/rotineira', 'Experiência\ndistinta/nova']
valores = [p_comum * 100, p_distintivo * 100]
cores = ['#c8a84b', '#2a8a82']

bars = ax1.bar(categorias, valores, color=cores, alpha=0.85, width=0.45)
ax1.set_ylim(0, 100)
ax1.set_ylabel('Taxa de recall / lembrança (%)')
ax1.set_title('Por que o novo grava mais fundo\nVon Restorff (1933) e replicações — efeito de distinção',
              fontsize=13, pad=15)

for bar, valor in zip(bars, valores):
    ax1.text(bar.get_x() + bar.get_width()/2, valor + 1.5,
             f'{valor:.0f}%', ha='center', fontweight='bold', fontsize=16)

ax1.annotate('', xy=(1, p_distintivo*100), xytext=(0, p_comum*100),
             arrowprops=dict(arrowstyle='->', color='#c0392b', lw=2))
ax1.text(0.5, 60, f'{razao_bruta:.1f}x mais lembrado',
         ha='center', fontsize=11, color='#c0392b', fontweight='bold')

plt.figtext(0.5, 0.01,
            'Fonte: Von Restorff (1933) e replicações posteriores | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-026-grafico-01-distincao.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# GRÁFICO 2 — Estimativa bruta vs corrigida (fator 0.80)
fig2, ax2 = plt.subplots(figsize=(10, 6))
categorias2 = ['Estimativa bruta\n(literatura)', 'Estimativa corrigida\n(×0.80)']
valores2 = [p_distintivo * 100, p_corrigido * 100]
cores2 = ['#2a8a82', '#1a5f5a']

bars2 = ax2.bar(categorias2, valores2, color=cores2, alpha=0.85, width=0.4)
ax2.axhline(y=p_comum*100, color='#c8a84b', linestyle='--', linewidth=2, alpha=0.8)
ax2.text(1.35, p_comum*100 + 1, 'linha de base\n(experiência comum)',
         fontsize=9, color='#8a7020', ha='left')
ax2.set_ylim(0, 100)
ax2.set_ylabel('P(virar lembrança duradoura) (%)')
ax2.set_title('Aplicando o fator de correção conservador\nEstimativa central ajustada em -20%',
              fontsize=13, pad=15)

for bar, valor in zip(bars2, valores2):
    ax2.text(bar.get_x() + bar.get_width()/2, valor + 1.5,
             f'{valor:.1f}%', ha='center', fontweight='bold', fontsize=16)

plt.figtext(0.5, 0.01,
            'Fator de correção padrão do projeto (×0.80) aplicado à estimativa central | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-026-grafico-02-correcao.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# GRÁFICO 3 — Curva de esquecimento ilustrativa (conceitual, baseada no modelo de Ebbinghaus)
# NOTA: curva ilustrativa/conceitual, não dado empírico replicado ponto a ponto.
fig3, ax3 = plt.subplots(figsize=(10, 6))

t = np.linspace(0, 10, 500)  # anos
S_comum = 0.8       # constante de retenção baixa (esquecimento rápido)
S_distintivo = 6.0  # constante de retenção alta (esquecimento lento)

R_comum = np.exp(-t / S_comum)
R_distintivo = 0.3 + 0.7 * np.exp(-t / S_distintivo)  # plateau alto, reflete consolidação forte

ax3.plot(t, R_comum * 100, color='#c8a84b', linewidth=3, label='Experiência comum/rotineira')
ax3.plot(t, R_distintivo * 100, color='#2a8a82', linewidth=3, label='Experiência distinta/nova')
ax3.fill_between(t, R_comum * 100, alpha=0.15, color='#c8a84b')
ax3.fill_between(t, R_distintivo * 100, alpha=0.15, color='#2a8a82')

ax3.set_xlabel('Anos desde a experiência')
ax3.set_ylabel('Força estimada da memória (%)')
ax3.set_title('Curva de esquecimento — ilustrativa\nModelo conceitual baseado em Ebbinghaus (1885) + efeito de distinção',
              fontsize=12, pad=15)
ax3.legend(loc='upper right', frameon=False)
ax3.set_ylim(0, 100)

plt.figtext(0.5, 0.01,
            'Curva ilustrativa/conceitual — não é dado medido ponto a ponto | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-026-grafico-03-esquecimento.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**Uma experiência genuinamente nova tem cerca de 60% mais chance de virar lembrança duradoura do que um dia comum**, mesmo depois de uma correção conservadora.

Isso não é sobre gastar mais. Não é sobre viagens caras ou datas especiais no calendário. É sobre quebrar a rotina de um jeito que o cérebro reconheça como diferente de tudo o mais.

O meu pai não lembrou com tanta força da neve porque foi caro ou porque foi bem planejado. Lembrou porque, pra ele, aquilo nunca tinha acontecido antes. E o cérebro dele, mesmo mais velho, ainda sabia exatamente o que fazer com isso.

Você não precisa saber neurociência pra criar uma lembrança que dura a vida toda em alguém que você ama. Só precisa entender que o comum se apaga, e o novo fica.

E hoje, pensando que ele partiu, isso me dá uma certa paz. Porque de tudo que eu podia ter dado a ele, acabei dando uma das coisas que a gente menos controla e mais guarda: uma lembrança que não desbotou.

### ⚠️ Limitações do Modelo

- As porcentagens de recall (80% distinto / 40% comum) são um benchmark ilustrativo consolidado a partir do paradigma clássico de Von Restorff e suas réplicas, não um número único medido em um estudo com N reportado. A proporção exata varia conforme o desenho experimental.
- Os estudos de neuroimagem (Bunzeck & Düzel, 2006; Chowdhury et al., 2012) usam amostras típicas de fMRI, relativamente pequenas, e demonstram mecanismo cerebral, não uma garantia individual de que qualquer experiência nova vai virar memória forte.
- Memórias autobiográficas reais (como a da neve) são mais ricas, mais longas e mais carregadas emocionalmente do que os itens de lista usados nos experimentos de laboratório — a generalização direta tem limites.
- O gráfico 3 (curva de esquecimento) é um modelo ilustrativo/conceitual baseado no framework de Ebbinghaus (1885), não uma reprodução de dados medidos ponto a ponto.

*A ciência é honesta sobre o que não sabe. O modelo também.*
