## Qual a probabilidade de o que um pai deixa continuar agindo mesmo depois que ele se vai?

**Tipo:** Preditivo

**Ferramenta:** Python
**Decisão analisada:** Vale a pena investir presença como pai, se o efeito só aparece anos depois?
**Hashtag:** #365Probabilidades #Dia057

---

### 📖 A História

Todo mundo carrega uma herança que não está em inventário nenhum. Não é
dinheiro, nem casa, nem sobrenome. É um jeito de agir, uma frase que volta na
hora certa, um valor que a gente segue sem nem lembrar de onde veio.

Boa parte dessa herança tem a marca de um pai presente. Não do pai perfeito,
nem do pai que estava sempre por perto. Do pai que se envolveu: que participou,
que se responsabilizou, que apareceu de verdade.

O que a ciência consegue perguntar é bonito e mensurável: quanto do que um pai
deixa continua agindo na vida do filho anos depois, mesmo quando ele já não
está mais ali para lembrar de nada?

A resposta não é uma opinião de cartão de Dia dos Pais. São décadas de estudos
que acompanharam crianças ao longo do tempo. E o que eles encontram é
consistente demais para ser coincidência.

---

### 📚 O Conceito: o pai de antes prevê o filho de depois

A palavra-chave da pesquisa é envolvimento (*involvement*), e ela não é só
morar na mesma casa. Envolvimento é engajamento ativo e regular: brincar,
conversar, cuidar, se responsabilizar. Os estudos separam essas camadas de
propósito, porque elas não pesam igual.

O desenho mais forte dessa literatura é longitudinal: mede o envolvimento do pai
num momento e só vai olhar o desfecho da criança pelo menos um ano depois. Isso
inverte a seta do tempo a favor da honestidade. Não é uma foto do mesmo instante,
em que causa e efeito se confundem. É o pai de antes prevendo a criança de
depois.

E há um segundo cuidado importante: boa parte desses estudos controla o nível
socioeconômico da família. Ou seja, o efeito do pai presente sobra mesmo depois
de descontar renda. Não é só o dinheiro disfarçado de afeto.

---

### 🧮 O Modelo

Duas leituras honestas do mesmo achado, as duas assinaturas do dia. Primeiro, a
força do consenso: de quantos estudos, dos que acompanharam crianças ao longo
do tempo, o pai presente saiu como fator positivo. Depois, o tamanho do efeito:
quão grande é essa vantagem, traduzida em linguagem comum.

**Fontes:**
- Sarkadi, A., Kristiansson, R., Oberklaid, F. & Bremberg, S. (2008).
  *Fathers' involvement and children's developmental outcomes: a systematic
  review of longitudinal studies.* Acta Paediatrica, 97(2), 153-158.
  Revisão sistemática de 24 estudos longitudinais; 22 acharam efeitos positivos;
  16 controlaram nível socioeconômico.
- Jeynes, W. H. (2015). *A Meta-Analysis: The Relationship between Father
  Involvement and Student Academic Achievement.* Urban Education.
  Meta-análise de 66 estudos; efeito do envolvimento paterno em torno de 0,2
  desvio-padrão.

**Nota metodológica:** o ×0.80 não se aplica. Ele corrige proporção de survey
autorrelatado; aqui são estudos longitudinais com desfechos medidos ao longo do
tempo. Aplicá-lo seria errado. Além disso, apesar do desenho longitudinal e do
controle de renda, isto é associação forte, não prova de causa isolada: a
linguagem correta é "prevê" e "associa-se", não "causa".

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

DOURADO = '#c8a84b'; VERMELHO = '#c0392b'; VERDE = '#1a5f5a'; CINZA = '#6b6a64'
print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---

# Sarkadi et al., 2008 — Acta Paediatrica (revisão de estudos longitudinais)
n_estudos_sarkadi = 24        # estudos longitudinais incluídos
n_positivos       = 22        # acharam efeito positivo do pai presente
n_controlou_ses   = 16        # controlaram nível socioeconômico
prop_consenso     = n_positivos / n_estudos_sarkadi

# Jeynes, 2015 — Urban Education (meta-análise)
n_estudos_jeynes  = 66        # estudos na meta-análise
d_efeito          = 0.20      # tamanho de efeito ~0,2 desvio-padrão

# CLES — Common Language Effect Size: P de a criança com pai mais presente
# sair melhor numa comparação aleatória, dado um efeito d (Gauss).
cles = stats.norm.cdf(d_efeito / np.sqrt(2))

# Fator de correção
# ×0.80 NÃO se aplica: não é proporção de survey autorrelatado, e sim
# estudos longitudinais com desfechos medidos.

print("=" * 66)
print("  DADOS DA LITERATURA — O QUE UM PAI PRESENTE DEIXA")
print("=" * 66)
print(f"\n  Sarkadi et al., 2008 (estudos longitudinais):")
print(f"  → Estudos acompanhando crianças no tempo:  {n_estudos_sarkadi}")
print(f"  → Acharam efeito positivo do pai presente: {n_positivos}")
print(f"  → Controlaram nível socioeconômico:        {n_controlou_ses}")
print(f"  → Consenso da literatura:                  {prop_consenso*100:.1f}%")
print(f"\n  Jeynes, 2015 (meta-análise, N={n_estudos_jeynes} estudos):")
print(f"  → Tamanho de efeito do envolvimento paterno: d = {d_efeito:.2f} DP")
print(f"  → Em linguagem comum (CLES):                 {cles*100:.1f}%")
print("=" * 66)
```

```python
# --- O MODELO — as duas assinaturas ---

# ASSINATURA 1 — a força do consenso (distribuição Beta)
# 22 de 24 estudos positivos. Beta(sucessos, fracassos) resume a
# proporção E a incerteza em torno dela.
alpha = n_positivos            # 22
beta_ = n_estudos_sarkadi - n_positivos  # 2
dist_beta = stats.beta(alpha, beta_)
ic_beta = dist_beta.interval(0.95)
media_beta = dist_beta.mean()

# ASSINATURA 2 — o tamanho de efeito (Cohen's d -> linguagem comum)
# d = 0,2 (Jeynes). CLES já calculado acima.

print("=" * 66)
print("  MODELO — DUAS ASSINATURAS")
print("=" * 66)
print(f"\n  [1] FORÇA DO CONSENSO — distribuição Beta({alpha}, {beta_})")
print(f"      Estimativa central: {prop_consenso*100:.1f}%")
print(f"      IC 95%: [{ic_beta[0]*100:.1f}%, {ic_beta[1]*100:.1f}%]")
print(f"      Leitura: entre os estudos que seguiram crianças no tempo,")
print(f"      quase todos apontam o pai presente como fator positivo.")
print(f"\n  [2] TAMANHO DO EFEITO — d = {d_efeito:.2f} desvio-padrão")
print(f"      Em linguagem comum (CLES): {cles*100:.1f}%")
print(f"      Leitura: em cerca de {cles*100:.0f} de cada 100 comparações")
print(f"      ao acaso, a criança com pai mais presente sai melhor.")
print(f"      Efeito pequeno por comparação, mas consistente em {n_estudos_jeynes} estudos.")
print("=" * 66)
```

```python
# --- VISUALIZAÇÃO — 3 GRÁFICOS SEPARADOS (fundo branco, dpi=150) ---

# ── GRÁFICO 1 — O consenso da literatura (22 de 24) ──
fig1, ax1 = plt.subplots(figsize=(10, 6))
labels = ['Acharam efeito\npositivo do pai\npresente', 'Não acharam']
valores = [n_positivos, n_estudos_sarkadi - n_positivos]
cores = [VERDE, CINZA]
bars = ax1.bar(labels, valores, color=cores, alpha=0.9, width=0.5)
ax1.set_ylim(0, 26)
ax1.set_ylabel('Número de estudos longitudinais')
ax1.set_title('O Consenso da Literatura\n22 de 24 estudos que acompanharam crianças no tempo',
              fontsize=13, pad=15)
for bar, v in zip(bars, valores):
    ax1.text(bar.get_x()+bar.get_width()/2, v+0.4, str(v),
             ha='center', fontweight='bold', fontsize=17)
ax1.text(0.5, 0.90, f'{n_controlou_ses} deles ainda controlaram a renda da família',
         ha='center', transform=ax1.transAxes, fontsize=11, color=CINZA, style='italic')
plt.figtext(0.5, 0.01, 'Fonte: Sarkadi et al., 2008 — Acta Paediatrica | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-057-grafico-01-consenso.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 — A ASSINATURA: distribuição Beta do consenso ──
fig2, ax2 = plt.subplots(figsize=(10, 6))
x = np.linspace(0.60, 1.0, 1000)
y = dist_beta.pdf(x)
ax2.plot(x*100, y, color=VERDE, linewidth=2.8)
ax2.fill_between(x*100, y, alpha=0.18, color=VERDE)
ax2.axvline(ic_beta[0]*100, color=DOURADO, linestyle='--', linewidth=2)
ax2.axvline(ic_beta[1]*100, color=DOURADO, linestyle='--', linewidth=2)
ax2.axvline(media_beta*100, color='#333', linewidth=2)
ax2.text(ic_beta[0]*100-0.5, max(y)*0.55,
         f'IC 95%\n[{ic_beta[0]*100:.0f}%, {ic_beta[1]*100:.0f}%]',
         fontsize=10, color=DOURADO, ha='right')
ax2.set_xlabel('Proporção de estudos com efeito positivo do pai presente (%)')
ax2.set_ylabel('Densidade')
ax2.set_title('A Assinatura do Consenso — Distribuição Beta\nBeta(22, 2): a força e a incerteza do que a literatura diz',
              fontsize=12.5, pad=15)
plt.figtext(0.5, 0.01, 'Fonte: Sarkadi et al., 2008 (22 de 24 estudos) | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-057-grafico-02-beta.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 — A ASSINATURA: o tamanho de efeito (d = 0,2) ──
fig3, ax3 = plt.subplots(figsize=(10, 6))
xx = np.linspace(-4, 4, 1000)
sem = stats.norm.pdf(xx, 0.0, 1.0)          # pai menos presente
com = stats.norm.pdf(xx, d_efeito, 1.0)     # pai mais presente (deslocado por d)
ax3.plot(xx, sem, color=CINZA, linewidth=2.2, label='Pai menos presente')
ax3.fill_between(xx, sem, alpha=0.12, color=CINZA)
ax3.plot(xx, com, color=VERDE, linewidth=2.6, label='Pai mais presente')
ax3.fill_between(xx, com, alpha=0.16, color=VERDE)
ax3.axvline(0.0, color=CINZA, linestyle=':', linewidth=1.2)
ax3.axvline(d_efeito, color=VERDE, linestyle=':', linewidth=1.2)
ax3.set_ylim(0, 0.46)
ax3.annotate('', xy=(d_efeito, 0.30), xytext=(0.0, 0.30),
             arrowprops=dict(arrowstyle='<->', color=VERMELHO, lw=1.8))
ax3.text(d_efeito/2, 0.315, f'd = {d_efeito:.1f} DP', ha='center',
         color=VERMELHO, fontweight='bold', fontsize=11)
ax3.set_yticks([])
ax3.set_xlabel('Desfecho da criança (em desvios-padrão)')
ax3.set_title('O Tamanho do Efeito — Pequeno, mas que Desloca a Curva\n'
              f'd = 0,2: em ~{cles*100:.0f}% das comparações ao acaso, a criança com pai presente sai melhor',
              fontsize=12.5, pad=15)
ax3.legend(loc='upper right', frameon=False, fontsize=10)
plt.figtext(0.5, 0.01, 'Fonte: Jeynes, 2015 — Urban Education (66 estudos) | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-057-grafico-03-efeito.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**22 de 24 estudos que acompanharam crianças ao longo do tempo apontam o pai
presente como fator positivo.** E não é um efeito frágil: ele sobra mesmo depois
de descontar a renda da família, e aparece anos depois de medido.

O tamanho por comparação é modesto, cerca de 0,2 desvio-padrão. Traduzido: em
torno de **56% das vezes**, numa comparação ao acaso, a criança com o pai mais
presente sai melhor. Parece pouco, mas é um empurrão que se repete em dezenas de
estudos, em milhares de crianças, na mesma direção.

O que fica é o mais bonito da história: o que um pai deixa não é um evento, é uma
inclinação. Uma pequena diferença de rota, colocada cedo, que segue empurrando na
mesma direção muito depois. O envolvimento de ontem ainda está agindo hoje, mesmo
quando o pai já não está por perto para ver.

*O que o seu pai deixou em você que ainda age, mesmo sem ele por perto?*

---

### ⚠️ Limitações do Modelo

- Associação forte, não prova de causa isolada. O desenho longitudinal e o
  controle de renda deixam o achado robusto, mas estudos observacionais não
  fecham a porta para outros fatores.
- "Pai" aqui inclui figuras paternas e arranjos diversos. O que os estudos
  medem é o envolvimento, não o vínculo biológico.
- Tamanho de efeito pequeno (d≈0,2). Vale no agregado, não é destino individual:
  há crianças que florescem sem pai presente e o contrário.
- Não é um julgamento de pais ausentes por circunstância. É uma medida do valor
  da presença, não uma acusação da falta.

*A ciência é honesta sobre o que não sabe. O modelo também.*
