### 📖 A História

Uma criança que eu amo está aprendendo a viver em dois mundos.

Dois idiomas, dois climas, duas formas de cumprimentar, duas
prateleiras de mercado. E eu, que processo o mundo por
estatística, fui checar o que a ciência sabe sobre quem carrega
duas culturas.

Fui com medo do que ia encontrar. Porque a história que a gente
ouve é antiga e cruel: quem vive entre dois mundos não pertence
a nenhum. A sociologia dos anos 1930 deu até nome a isso, o
"homem marginal", dilacerado entre culturas, condenado ao
desajuste.

Aí encontrei a meta-análise: **83 estudos, 322 correlações,
23.197 pessoas**. E a frase dos próprios autores: os resultados
**invalidam** a velha narrativa. Quem se engaja nas duas culturas
tende a estar MELHOR ajustado, psicológica e socialmente, do que
quem vive só numa delas. Qualquer uma delas.

A criança que aprende o segundo mundo não está dividindo o
pertencimento. Está dobrando.

E aí vem a parte que este projeto faz questão de contar inteira:
o tamanho dessa vantagem tem uma história de encolhimento, e a
honestidade sobre ela é o modelo de hoje.

---

### 📚 O Conceito: O Fim do Homem Marginal

O mito tinha lógica de aritmética simples: identidade seria um
bolo fixo, e cada cultura levaria metade. Os dados dizem que
identidade funciona como idioma: aprender o segundo não rouba
espaço do primeiro.

Mas a magnitude da vantagem bicultural é uma disputa viva, e o
dia declara a escada inteira:

**A manchete original (2013):** r=0,51, uma associação forte.
**A análise alternativa do próprio artigo:** r=0,10.
**A atualização de 2021, com os próprios autores na equipe** e
90 estudos novos, incluindo não-publicados: **r=0,12**, pequeno.
**E o desafio longitudinal** (Bierwiaczonek & Kunst, 2021): em
dados seguidos no tempo, o efeito pode não diferir de zero, com
resposta dos defensores em 2023.

O que atravessa TODAS as réguas sem virar: **o sinal**. Em
nenhuma estimativa o bicultural sai perdendo. O pior cenário
moderno é o empate. O mito previa desvantagem; nenhum método a
encontra.

**Ponte declarada com o Dia #014:** lá, o vínculo à distância;
hoje, o que a segunda cultura faz com quem a recebe. A mesma
família de inquietações, construtos e fontes distintos.

### 🧮 O Modelo

**Fontes:**
- Nguyen, A.-M. D. & Benet-Martínez, V. (2013) - Biculturalism
  and adjustment: A meta-analysis - Journal of Cross-Cultural
  Psychology, 44(1), 122-159 - **83 estudos, 322 correlações,
  N=23.197**. Efeitos aleatórios: **r=0,51 [0,48, 0,54]**;
  análise alternativa de efeitos fixos: r=0,10 (1% da
  variância). Biculturalismo mais associado a ajustamento do que
  qualquer cultura sozinha. **Fonte central.**
- Stogianni, M., Bender, M., Sleegers, W., Benet-Martínez, V. &
  Nguyen, A.-M. (2021) - An updated meta-analysis - **90 estudos
  novos**, incluindo não-publicados: **r=0,12**, pequeno.
  A atualização honesta, com os autores originais na equipe.
- Yoon, E. et al. (2013) - Journal of Counseling Psychology -
  325 estudos: r=0,10 com saúde mental positiva, r=-0,10 com
  negativa. Apoio independente da direção.
- Bierwiaczonek, K. & Kunst, J. R. (2021) - re-análise e meta
  longitudinal: efeito pode não diferir de zero; resposta de
  Grigoryev et al. (2023). O desafio, declarado.
- Contexto histórico: a tese do "homem marginal" (Stonequist,
  1937), a previsão que os dados enterraram.

**Fator ×0,80: não se aplica.** Correlações de escalas
contínuas de ajustamento, não proporções de survey.

**As contas nossas:** (1) a escada da magnitude traduzida em
pares: de 100 duelos entre um bicultural e um monocultural,
quantos o bicultural "vence" em ajustamento, sob cada régua;
(2) o teste do piso: a curva completa da tradução mostrando que
qualquer sinal positivo põe o bicultural acima da moeda, e que
nenhuma estimativa moderna o põe abaixo.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---

n_estudos, n_rs, n_pessoas = 83, 322, 23_197

reguas = {
    'Manchete 2013 (efeitos aleatórios)': 0.51,   # IC [0,48, 0,54]
    'Alternativa do próprio artigo':      0.10,   # r² = 1% da variância
    'Atualização 2021 (+90 estudos)':     0.12,
    'Yoon et al. 2013 (325 estudos)':     0.10,
}
ic_manchete = (0.48, 0.54)

# Desafio longitudinal: Bierwiaczonek & Kunst 2021: pode ser zero
piso_desafio = 0.0

print("=" * 65)
print("  DADOS DA LITERATURA - AS RÉGUAS DA VANTAGEM BICULTURAL")
print("=" * 65)
print(f"\n  Base: Nguyen & Benet-Martínez 2013")
print(f"  {n_estudos} estudos, {n_rs} correlações, N = {n_pessoas:,}".replace(',','.'))
print(f"\n  A escada da magnitude:")
for nome, r in reguas.items():
    print(f"  → {nome:<38} r = {r:.2f}")
print(f"  → Desafio longitudinal (B&K 2021):        r pode ser {piso_desafio:.2f}")
print(f"\n  O que atravessa todas as réguas: o SINAL.")
print(f"  Em nenhuma estimativa o bicultural sai perdendo.")
print(f"\n  Fator ×0,80: NÃO aplicado (correlações de escalas)")
print("=" * 65)
```

```python
# --- O MODELO ---
# Conta 1: a escada traduzida em pares (duelos)
# Conta 2: o teste do piso

def cles_de_r(r):
    if r <= 0:
        return 50.0 if r == 0 else float(stats.norm.cdf((2*r/np.sqrt(1-r**2))/np.sqrt(2))*100)
    d = 2*r / np.sqrt(1 - r**2)
    return stats.norm.cdf(d / np.sqrt(2)) * 100

duelos = {nome: cles_de_r(r) for nome, r in reguas.items()}
duelo_manchete = duelos['Manchete 2013 (efeitos aleatórios)']
duelo_honesto = duelos['Atualização 2021 (+90 estudos)']

print("=" * 65)
print("  MODELO - A ESCADA EM DUELOS E O TESTE DO PISO")
print("=" * 65)
print(f"\n  CONTA 1 - de 100 duelos bicultural × monocultural,")
print(f"  quantos o bicultural 'vence' em ajustamento?")
for nome, v in duelos.items():
    print(f"  → {nome:<38} {v:.1f} de 100")
print(f"  → Desafio longitudinal (r=0):              50,0 de 100 (empate)")
print(f"\n  → A manchete prometia {duelo_manchete:.0f} de 100.")
print(f"     O número honesto entrega {duelo_honesto:.0f} de 100.")
print(f"     E o pior cenário moderno entrega o EMPATE, não a derrota.")
print(f"\n  CONTA 2 - o teste do piso:")
print(f"  → Menor estimativa em qualquer régua: r = {min(reguas.values()):.2f} > 0")
print(f"  → Pior cenário do desafio: r = 0 → 50 de 100")
print(f"  → O mito do 'homem marginal' previa < 50: DESVANTAGEM.")
print(f"     Nenhum método, de nenhum lado da disputa, a encontra.")
print(f"\n  A LEITURA:")
print(f"  → O presente não é um superpoder. É um pertencimento")
print(f"     dobrado, com uma vantagem pequena e honesta por cima:")
print(f"     e com um piso que nenhuma régua fura.")
print("=" * 65)
```

```python
# --- VISUALIZAÇÃO - GRÁFICOS SEPARADOS ---

# ── GRÁFICO 1 - A escada das estimativas ──
fig1, ax1 = plt.subplots(figsize=(10, 6))

nomes = ['Manchete\n2013', 'Alternativa\ndo artigo', 'Atualização\n2021', 'Yoon\n2013', 'Desafio\nB&K 2021']
vals = [0.51, 0.10, 0.12, 0.10, 0.0]
cores = ['#c9a14a', '#c9a14a', '#2a8a82', '#c9a14a', '#c0392b']

ax1.axhline(0, color='#333', linewidth=1.4)
ax1.text(4.35, 0.012, 'zero: empate', fontsize=10, color='#333', ha='right')
bars = ax1.bar(nomes, vals, color=cores, alpha=0.88, width=0.5)
ax1.plot([0, 0], ic_manchete, color='#8a6d2f', linewidth=2.5)
for bar, v in zip(bars, vals):
    ax1.text(bar.get_x() + bar.get_width()/2, v + 0.015,
             f'{v:.2f}', ha='center', fontweight='bold', fontsize=13)
ax1.set_ylabel('Correlação biculturalismo × ajustamento (r)')
ax1.set_ylim(-0.12, 0.62)
ax1.set_title('A escada da magnitude: encolhe, mas o sinal não vira\n'
              'Nenhuma régua, de nenhum lado da disputa, encontra desvantagem',
              fontsize=12.5, pad=15)

plt.figtext(0.5, 0.01,
            'Fontes: Nguyen & Benet-Martínez (2013); Stogianni et al. (2021); Yoon et al. (2013); Bierwiaczonek & Kunst (2021) | #365Probabilidades',
            ha='center', fontsize=8, color='gray')
plt.tight_layout()
plt.savefig('dia-097-grafico-01-escada.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 - A tradução em duelos ──
fig2, ax2 = plt.subplots(figsize=(10, 6))

nomes2 = ['O que a manchete\nde 2013 prometia', 'O número honesto\n(atualização 2021)', 'O pior cenário\nmoderno (desafio)']
vals2 = [duelo_manchete, duelo_honesto, 50.0]
cores2 = ['#c9a14a', '#2a8a82', '#6b6a64']

bars2 = ax2.bar(nomes2, vals2, color=cores2, alpha=0.88, width=0.45)
ax2.axhline(50, color='#333', linestyle='--', linewidth=1.5)
ax2.text(2.35, 50.9, 'moeda: 50', fontsize=10.5, color='#333', ha='right')
ax2.set_ylim(40, 90)
ax2.set_ylabel('Duelos vencidos pelo bicultural (de 100)')
ax2.set_title('A tradução em pares: a promessa, o honesto e o piso\n'
              'O mito previa abaixo da linha; ninguém o encontra lá',
              fontsize=12.5, pad=15)
for bar, v in zip(bars2, vals2):
    ax2.text(bar.get_x() + bar.get_width()/2, v + 0.9,
             f'{v:.0f}', ha='center', fontweight='bold', fontsize=15)

plt.figtext(0.5, 0.01,
            'Conta própria (probabilidade de superioridade a partir de r) | suposições declaradas | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-097-grafico-02-duelos.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 - O teste do piso: a curva completa ──
fig3, ax3 = plt.subplots(figsize=(10, 6))

rr = np.linspace(-0.3, 0.6, 300)
curva = [cles_de_r(r) for r in rr]
ax3.plot(rr, curva, color='#2a8a82', linewidth=3)
ax3.axhline(50, color='#333', linestyle='--', linewidth=1.4)
ax3.axvline(0, color='#333', linewidth=1.2)
ax3.fill_between(rr, 40, 50, where=(rr < 0), alpha=0.12, color='#c0392b')
ax3.text(-0.155, 44, 'a região do mito:\no "marginal"\nem desvantagem', fontsize=10.5,
         color='#c0392b', ha='center')
for r, cor, dy in [(0.12, '#2a8a82', 3), (0.51, '#c9a14a', 3)]:
    ax3.scatter([r], [cles_de_r(r)], s=140, color=cor, zorder=3)
ax3.text(0.135, cles_de_r(0.12)-4.5, 'honesto:\nr=0,12 → 57', fontsize=10.5, color='#1a5f5a')
ax3.text(0.42, cles_de_r(0.51)+2.5, 'manchete:\nr=0,51 → 80', fontsize=10.5, color='#8a6d2f')
ax3.set_xlabel('Correlação (r)')
ax3.set_ylabel('Duelos vencidos pelo bicultural (de 100)')
ax3.set_ylim(40, 88)
ax3.set_title('O teste do piso: nenhuma estimativa moderna entra na região do mito\n'
              'Qualquer sinal positivo põe o bicultural acima da moeda',
              fontsize=12, pad=15)

plt.figtext(0.5, 0.01,
            'Conta própria | a região vermelha é onde o mito de 1937 morava; está vazia | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-097-grafico-03-piso.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**A velha história dizia que quem vive entre dois mundos não
pertence a nenhum. Os dados dizem que pertence aos dois.**

E este projeto conta a história completa: a vantagem bicultural
já foi anunciada como enorme (80 de 100 duelos), e o número
honesto, depois de 90 estudos novos e da inclusão dos trabalhos
não-publicados, é modesto: **57 de 100**. Há até quem defenda,
com dados longitudinais, que seja um empate.

Mas repare no que NINGUÉM encontra: a desvantagem. A região onde
o mito do "homem marginal" morava desde 1937, a da criança
dilacerada que pertenceria a menos, está vazia em todas as
réguas, dos dois lados da disputa. O pior cenário moderno para
quem carrega duas culturas é ser exatamente como todo mundo. O
provável é uma vantagem pequena por cima.

Identidade não é um bolo que as culturas dividem. É um idioma:
aprender o segundo não rouba espaço do primeiro.

Eu fui checar com medo, e voltei com isso: a segunda cultura de
uma criança não é a conta de subtração que o medo faz. No pior
cenário, é neutra. No provável, é presente.

Era o que eu precisava ler.

A pergunta que fica:

é pertencer a menos?*

### ⚠️ Limitações do Modelo

- Correlacional e transversal: os próprios autores declaram a
  direcionalidade aberta (biculturalismo → ajuste, ajuste →
  biculturalismo, ou uma terceira variável)
- A disputa de magnitude está viva e declarada: Bierwiaczonek &
  Kunst (2021) desafiam com dados longitudinais; Grigoryev et
  al. (2023) respondem; o dia se apoia no que atravessa os dois
  lados (o sinal), não na magnitude exata
- A manchete de 0,51 é sensível ao método: a alternativa do
  próprio artigo dá 0,10, e instrumentos de medida bidimensionais
  inflam a associação (moderador reportado)
- Amostras majoritariamente de contextos migratórios específicos;
  generalização entre países limitada
- A tradução em duelos assume normalidade e converte r em d;
  é régua de leitura, não novo teste
- A tese do "homem marginal" (Stonequist, 1937) entra como
  contexto histórico, não como fonte de dados
- Ponte declarada com o Dia #014 (vínculo à distância): mesma
  família de inquietações, construtos e fontes distintos

*A ciência é honesta sobre o que não sabe. O modelo também.*
