Quando alguém me convida para alguma coisa daqui a um mês, eu digo sim. Um café, uma

---

### 📚 O Conceito

Dois pesquisadores de marketing, Gal Zauberman e John Lynch, deram nome a isso em 2005:
folga esperada de recurso. A pessoa olha para o mês que vem e vê tempo sobrando, porque
os compromissos de lá ainda não têm nome. Olha para a conta bancária do mês que vem e vê
quase o mesmo saldo de hoje, porque dinheiro tem extrato.

A diferença muda a decisão. Se o tempo de daqui a um mês parece abundante, gastar uma
hora lá custa menos do que gastar uma hora hoje, e a pessoa aceita o compromisso. É o
desconto do futuro, mais forte para o tempo do que para o dinheiro. Em sete surveys com
estudantes, os dois mediram quanto as pessoas esperam ter de folga, quanto aceitam gastar
hoje e daqui a um mês, e quanto erram ao prever a própria agenda.

---

### 🧮 O Modelo

Quatro camadas. A fonte dá médias, proporções e correlações. As contas são nossas.

1. **Hoje e daqui a um mês.** A mesma tarefa, aceita por quantos quando o custo é em
   tempo hoje e quando é em tempo daqui a um mês, com a incerteza das duas proporções.
2. **Tempo contra dinheiro.** As quatro células do experimento, e a chance de o efeito do
   prazo ser maior para o tempo do que para o dinheiro.
3. **O erro de previsão.** Quanto a folga esperada passa do "igual a hoje", e quanto a
   previsão de tempo erra em comparação com a de dinheiro.
4. **A agenda de um ano.** Monte Carlo de doze convites, um por mês, aceitos de longe e
   avaliados na hora: quantos "sim" viram "droga".

**Fonte central:**
- Zauberman, G. & Lynch, J. G., Jr. (2005). "Resource slack and propensity to discount
  delayed investments of time versus money". *Journal of Experimental Psychology:
  General* 134(1), 23-37. doi 10.1037/0096-3445.134.1.23. Sete surveys com estudantes.
  **Experimento 1 (N = 76 analisados de 95):** folga esperada daqui a um mês numa escala
  de 1 a 10 em que 5,5 é "igual a hoje"; tempo **8,2** (DP 2,9; η²p = 0,57), dinheiro
  **7,0** (η²p = 0,23); diferença p = 0,013. **Experimento 5 (N = 264, entre sujeitos):**
  um custo em tempo é aceito por **39,4%** quando é hoje e **56,1%** quando é daqui a um
  mês; um custo em dinheiro, por 69,7% hoje e 62,1% em um mês; interação p = 0,047.
  **Experimento 6 (N = 48):** erro de previsão da folga maior para tempo (1,85) do que
  para dinheiro (1,10), p = 0,015. **Experimento 7 (N = 130):** a folga de tempo
  esperada para amanhã não prevê a de duas semanas (r = 0,03); a de dinheiro prevê
  (r = 0,45); quem espera mais tempo adia mais a ajuda em tempo (interação p = 0,003).
  **Estudo natural (N = 178):** um aviso específico não aumentou a presença numa
  palestra (19% contra 18%), e mais da metade dos que disseram que iam não foram.
  **As contagens do Experimento 5 são reconstruídas dos percentuais, com 66 por célula
  (declarado).** Os desvios-padrão do Experimento 6 não foram lidos.

**Nota metodológica.** O fator ×0,80 não se aplica. As proporções são escolhas feitas
no experimento, não hábito lembrado, e as médias e correlações estão fora da convenção.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

SEMENTE = 42
rng = np.random.default_rng(SEMENTE)
SORTEIOS = 200_000

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

VERMELHO, DOURADO, VERDE, CINZA = '#c0392b', '#c8a84b', '#1a5f5a', '#6b6a64'
RODAPE = 'Fonte: Zauberman & Lynch 2005, JEP: General 134(1) · cálculo próprio | #365Probabilidades'

def vir(x, casas=1):
    return f"{x:.{casas}f}".replace(".", ",")

def pct(x, casas=1):
    return vir(100 * x, casas) + "%"

def mil(n):
    return f"{int(n):,}".replace(",", ".")

def jeffreys(k, n):
    d = stats.beta(k + 0.5, n - k + 0.5)
    return d, d.interval(0.95)

print("✅ Bibliotecas carregadas · semente", SEMENTE)
```

```python
# --- DADOS DA LITERATURA ---
# Zauberman & Lynch (2005), JEP: General 134(1), 23-37

# Experimento 1 (N = 76): folga esperada daqui a um mês, escala 1 a 10, 5,5 = igual a hoje
N_E1, IGUAL = 76, 5.5
FOLGA_TEMPO, DP_TEMPO, ETA_TEMPO = 8.2, 2.9, 0.57
FOLGA_DINHEIRO, ETA_DINHEIRO = 7.0, 0.23   # DP do dinheiro não lido

# Experimento 5 (N = 264, entre sujeitos, 2 × 2): aceita o custo?
N_E5, POR_CELULA = 264, 66
P = {('tempo', 'hoje'): 0.394, ('tempo', 'mês'): 0.561,
     ('dinheiro', 'hoje'): 0.697, ('dinheiro', 'mês'): 0.621}
K = {c: round(p * POR_CELULA) for c, p in P.items()}   # contagens reconstruídas
P_INTERACAO = 0.047

# Experimento 6 (N = 48): erro de previsão da folga
N_E6, ERRO_TEMPO, ERRO_DINHEIRO, P_E6 = 48, 1.85, 1.10, 0.015

# Experimento 7 (N = 130): folga esperada para amanhã × para duas semanas
N_E7, R_TEMPO, R_DINHEIRO = 130, 0.03, 0.45

# Estudo natural (N = 178): presença na palestra
N_NAT, PRESENCA_AVISO, PRESENCA_SEM = 178, 0.19, 0.18

print("=" * 72)
print("  DADOS · ZAUBERMAN & LYNCH (2005)")
print("=" * 72)
print(f"\n  Experimento 1 (N = {N_E1}): folga esperada daqui a um mês, 5,5 = igual a hoje")
print(f"    tempo {vir(FOLGA_TEMPO)} (DP {vir(DP_TEMPO)}) · dinheiro {vir(FOLGA_DINHEIRO)} · diferença p = 0,013")
print(f"\n  Experimento 5 (N = {N_E5}, {POR_CELULA} por célula): aceitam o custo")
for rec in ('tempo', 'dinheiro'):
    print(f"    {rec:>8}: hoje {pct(P[(rec, 'hoje')])} ({K[(rec, 'hoje')]}/{POR_CELULA}) · em um mês {pct(P[(rec, 'mês')])} ({K[(rec, 'mês')]}/{POR_CELULA})")
print(f"    interação recurso × prazo: p = {vir(P_INTERACAO, 3)}")
print(f"\n  Experimento 6 (N = {N_E6}): erro de previsão da folga, tempo {vir(ERRO_TEMPO, 2)} contra dinheiro {vir(ERRO_DINHEIRO, 2)}, p = {vir(P_E6, 3)}")
print(f"  Experimento 7 (N = {N_E7}): folga de amanhã prevê a de duas semanas? tempo r = {vir(R_TEMPO, 2)} · dinheiro r = {vir(R_DINHEIRO, 2)}")
print(f"  Estudo natural (N = {N_NAT}): presença na palestra {pct(PRESENCA_AVISO, 0)} com aviso específico, {pct(PRESENCA_SEM, 0)} sem")
print("\n  Contagens do Experimento 5 reconstruídas dos percentuais (66 por célula). Sem ×0,80: escolha no experimento.")
print("=" * 72)
```

```python
# --- O MODELO ---

# ── CAMADA 1 · hoje e daqui a um mês ──
d_hoje, ic_hoje = jeffreys(K[('tempo', 'hoje')], POR_CELULA)
d_mes, ic_mes = jeffreys(K[('tempo', 'mês')], POR_CELULA)
s_hoje, s_mes = d_hoje.rvs(SORTEIOS, random_state=rng), d_mes.rvs(SORTEIOS, random_state=rng)
dif_tempo = s_mes - s_hoje
print("=" * 72)
print("  CAMADA 1 · A MESMA TAREFA, HOJE E DAQUI A UM MÊS (custo em tempo)")
print("=" * 72)
print(f"\n  Aceitam hoje:          {pct(P[('tempo', 'hoje')])}  IC 95% [{pct(ic_hoje[0])}; {pct(ic_hoje[1])}]")
print(f"  Aceitam em um mês:     {pct(P[('tempo', 'mês')])}  IC 95% [{pct(ic_mes[0])}; {pct(ic_mes[1])}]")
print(f"  Diferença:             {vir(100 * dif_tempo.mean())} pontos  [{vir(100 * np.percentile(dif_tempo, 2.5))}; {vir(100 * np.percentile(dif_tempo, 97.5))}]")
print(f"  P(em um mês > hoje):   {pct((dif_tempo > 0).mean())}")
print(f"  Razão:                 {vir((s_mes / s_hoje).mean(), 2)} vezes mais gente diz sim quando é para daqui a um mês")

# ── CAMADA 2 · tempo contra dinheiro ──
d_dh, _ = jeffreys(K[('dinheiro', 'hoje')], POR_CELULA)
d_dm, _ = jeffreys(K[('dinheiro', 'mês')], POR_CELULA)
dif_dinheiro = d_dm.rvs(SORTEIOS, random_state=rng) - d_dh.rvs(SORTEIOS, random_state=rng)
inter = dif_tempo - dif_dinheiro
print("\n" + "=" * 72)
print("  CAMADA 2 · TEMPO CONTRA DINHEIRO")
print("=" * 72)
print(f"\n  Efeito do prazo no tempo:     {vir(100 * dif_tempo.mean()):>5} pontos (sobe)")
print(f"  Efeito do prazo no dinheiro:  {vir(100 * dif_dinheiro.mean()):>5} pontos (desce)")
print(f"  Diferença dos efeitos:        {vir(100 * inter.mean())} pontos  [{vir(100 * np.percentile(inter, 2.5))}; {vir(100 * np.percentile(inter, 97.5))}]")
print(f"  P(o prazo ajuda mais o tempo do que o dinheiro): {pct((inter > 0).mean())}  (no artigo, interação p = 0,047)")
print(f"  P(o dinheiro cai com o prazo): {pct((dif_dinheiro < 0).mean())}  (o artigo não afirma a queda; só a interação)")

# ── CAMADA 3 · o erro de previsão ──
# Experimento 1: quanto a folga esperada passa do "igual a hoje"
d_folga = (FOLGA_TEMPO - IGUAL) / DP_TEMPO
ep = DP_TEMPO / np.sqrt(N_E1)
ic_folga = (FOLGA_TEMPO - 1.96 * ep, FOLGA_TEMPO + 1.96 * ep)
p_acima = 1 - stats.norm.cdf((IGUAL - FOLGA_TEMPO) / DP_TEMPO)
# Experimento 7: intervalos das correlações por Fisher
def ic_r(r, n):
    z = np.arctanh(r); e = 1 / np.sqrt(n - 3)
    return np.tanh(z - 1.96 * e), np.tanh(z + 1.96 * e)
ic_rt, ic_rd = ic_r(R_TEMPO, N_E7), ic_r(R_DINHEIRO, N_E7)
print("\n" + "=" * 72)
print("  CAMADA 3 · O ERRO DE PREVISÃO")
print("=" * 72)
print(f"\n  Folga de tempo esperada: {vir(FOLGA_TEMPO)} numa escala em que 5,5 é 'igual a hoje'")
print(f"    IC 95% da média [{vir(ic_folga[0])}; {vir(ic_folga[1])}] · d = {vir(d_folga, 2)} acima do 'igual'")
print(f"    Se a escala for aproximadamente normal, {pct(p_acima, 0)} das pessoas esperam mais tempo do que hoje (premissa declarada)")
print(f"  Erro de previsão (Exp. 6): tempo {vir(ERRO_TEMPO, 2)} contra dinheiro {vir(ERRO_DINHEIRO, 2)}, razão {vir(ERRO_TEMPO / ERRO_DINHEIRO, 2)}")
print(f"  A folga de amanhã prevê a de duas semanas? (Exp. 7, N = {N_E7})")
print(f"    tempo    r = {vir(R_TEMPO, 2)}  IC 95% [{vir(ic_rt[0], 2)}; {vir(ic_rt[1], 2)}]  → não prevê")
print(f"    dinheiro r = {vir(R_DINHEIRO, 2)}  IC 95% [{vir(ic_rd[0], 2)}; {vir(ic_rd[1], 2)}]  → prevê")

# ── CAMADA 4 · a agenda de um ano ──
# Doze convites, um por mês. Cada um é aceito de longe com p_mes e seria aceito na hora com p_hoje.
# Premissa: quem diria sim na hora também diz sim de longe (aceitação monótona). Então a fração dos
# "sim de longe" que viram "droga" na hora é (p_mes − p_hoje) / p_mes.
CONVITES = 12
aceitos = rng.binomial(CONVITES, s_mes)
frac_droga = np.clip((s_mes - s_hoje) / s_mes, 0, 1)
droga = rng.binomial(aceitos, frac_droga)
print("\n" + "=" * 72)
print(f"  CAMADA 4 · A AGENDA DE UM ANO ({mil(SORTEIOS)} anos sorteados)")
print("=" * 72)
print(f"\n  Doze convites, um por mês, aceitos de longe:")
print(f"    aceitos no ano: média {vir(aceitos.mean())} de {CONVITES} · 95% entre {int(np.percentile(aceitos, 2.5))} e {int(np.percentile(aceitos, 97.5))}")
print(f"    dos aceitos, viram 'droga' quando o dia chega: {pct(frac_droga.mean(), 0)} [{pct(np.percentile(frac_droga, 2.5), 0)}; {pct(np.percentile(frac_droga, 97.5), 0)}]")
print(f"    'sim, droga' por ano: média {vir(droga.mean())} · P(pelo menos um) = {pct((droga >= 1).mean())} · P(três ou mais) = {pct((droga >= 3).mean())}")
print(f"    se cada convite fosse avaliado na hora: {vir(CONVITES * P[('tempo', 'hoje')])} aceitos em vez de {vir(CONVITES * P[('tempo', 'mês')])}")
print("  Premissa: aceitação monótona (quem diz sim na hora diz sim de longe). É um limite inferior do arrependimento.")
print("=" * 72)
```

```python
# --- VISUALIZAÇÃO ---

# ── GRÁFICO 1 · as quatro células ──
fig1, ax1 = plt.subplots(figsize=(11, 6))
x = np.arange(2); w = 0.36
for i, (prazo, cor) in enumerate((('hoje', CINZA), ('mês', DOURADO))):
    vals = [100 * P[('tempo', prazo)], 100 * P[('dinheiro', prazo)]]
    bars = ax1.bar(x + (i - 0.5) * w, vals, w, color=cor, alpha=0.9, label='hoje' if prazo == 'hoje' else 'daqui a um mês')
    for b, v, rec in zip(bars, vals, ('tempo', 'dinheiro')):
        _, ic = jeffreys(K[(rec, prazo)], POR_CELULA)
        ax1.plot([b.get_x() + w / 2] * 2, [100 * ic[0], 100 * ic[1]], color='black', lw=1.5)
        ax1.text(b.get_x() + w / 2, 100 * ic[1] + 1.5, pct(v / 100), ha='center', fontsize=12, fontweight='bold')
ax1.set_xticks(x); ax1.set_xticklabels(['custo em tempo', 'custo em dinheiro'], fontsize=12)
ax1.set_ylim(0, 95); ax1.set_ylabel('Aceitam o custo (%)')
ax1.legend(frameon=False, fontsize=11, loc='upper left')
ax1.set_title('A mesma tarefa, hoje e daqui a um mês\nExperimento 5, N = 264 · o prazo faz o tempo subir e o dinheiro descer', fontsize=13, pad=14)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-137-grafico-01-hoje-e-um-mes.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 · as duas posteriores e a diferença ──
fig2, (ax2a, ax2b) = plt.subplots(1, 2, figsize=(12, 5.5))
xx = np.linspace(0.15, 0.8, 800)
ax2a.plot(100 * xx, d_hoje.pdf(xx), color=CINZA, lw=3, label='hoje'); ax2a.fill_between(100 * xx, d_hoje.pdf(xx), alpha=0.15, color=CINZA)
ax2a.plot(100 * xx, d_mes.pdf(xx), color=DOURADO, lw=3, label='daqui a um mês'); ax2a.fill_between(100 * xx, d_mes.pdf(xx), alpha=0.2, color=DOURADO)
ax2a.set_xlabel('Aceitam o custo em tempo (%)'); ax2a.set_ylabel('Densidade'); ax2a.legend(frameon=False, fontsize=11)
ax2a.set_title('Posterior de Jeffreys, 66 por célula', fontsize=12)
ax2b.hist(100 * dif_tempo, bins=80, color=VERMELHO, alpha=0.85, density=True)
ax2b.axvline(0, color='black', lw=1.2, ls='--')
lo, hi = np.percentile(100 * dif_tempo, [2.5, 97.5])
ax2b.axvline(lo, color=DOURADO, ls='--', lw=2); ax2b.axvline(hi, color=DOURADO, ls='--', lw=2)
ax2b.text(0.97, 0.9, f'diferença {vir(100 * dif_tempo.mean())} pontos\nIC 95% [{vir(lo)}; {vir(hi)}]\nP(> 0) = {pct((dif_tempo > 0).mean())}', transform=ax2b.transAxes, ha='right', va='top', fontsize=11)
ax2b.set_xlabel('Em um mês menos hoje (pontos)'); ax2b.set_title('A diferença', fontsize=12)
ax2b.yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: vir(v, 2)))
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.04, 1, 1))
plt.savefig('dia-137-grafico-02-bayesiano.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 · o erro de previsão ──
fig3, (ax3a, ax3b) = plt.subplots(1, 2, figsize=(12, 5.5))
esc = np.linspace(1, 10, 500)
ax3a.plot(esc, stats.norm.pdf(esc, FOLGA_TEMPO, DP_TEMPO), color=DOURADO, lw=3, label=f'tempo, média {vir(FOLGA_TEMPO)}')
ax3a.fill_between(esc, stats.norm.pdf(esc, FOLGA_TEMPO, DP_TEMPO), where=esc > IGUAL, alpha=0.2, color=DOURADO)
ax3a.axvline(IGUAL, color='black', lw=1.5, ls='--'); ax3a.text(IGUAL - 0.1, ax3a.get_ylim()[1] * 0.25, '5,5 =\nigual a hoje', ha='right', fontsize=10)
ax3a.axvline(FOLGA_DINHEIRO, color=VERDE, lw=2); ax3a.text(FOLGA_DINHEIRO + 0.1, ax3a.get_ylim()[1] * 0.6, f'dinheiro\n{vir(FOLGA_DINHEIRO)}', color=VERDE, fontsize=10)
ax3a.set_xlabel('Folga esperada daqui a um mês (1 a 10)'); ax3a.set_ylabel('Densidade'); ax3a.legend(frameon=False, fontsize=10, loc='upper left')
ax3a.set_title(f'Experimento 1, N = 76 · curva normal ilustrativa, DP {vir(DP_TEMPO)}', fontsize=11)
ax3a.yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: vir(v, 2)))
for i, (nome, r, ic, cor) in enumerate((('tempo', R_TEMPO, ic_rt, DOURADO), ('dinheiro', R_DINHEIRO, ic_rd, VERDE))):
    ax3b.plot([ic[0], ic[1]], [i, i], color=cor, lw=4); ax3b.scatter([r], [i], color=cor, s=120, zorder=5, edgecolor='white')
    ax3b.text(ic[1] + 0.03, i, f'r = {vir(r, 2)}  [{vir(ic[0], 2)}; {vir(ic[1], 2)}]', va='center', fontsize=11)
ax3b.axvline(0, color='black', lw=1, ls='--')
ax3b.set_yticks([0, 1]); ax3b.set_yticklabels(['tempo', 'dinheiro'], fontsize=12); ax3b.set_ylim(-0.7, 1.7); ax3b.set_xlim(-0.3, 1.0)
ax3b.set_xlabel('Correlação entre a folga esperada para amanhã e para duas semanas')
ax3b.set_title('Experimento 7, N = 130 · IC pela transformação de Fisher', fontsize=11)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.04, 1, 1))
plt.savefig('dia-137-grafico-03-o-erro-de-previsao.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")

# ── GRÁFICO 4 · a agenda de um ano ──
fig4, ax4 = plt.subplots(figsize=(11, 6))
vals, counts = np.unique(droga, return_counts=True)
cores4 = [CINZA if v == 0 else VERMELHO for v in vals]
ax4.bar(vals, 100 * counts / SORTEIOS, color=cores4, alpha=0.9)
for v, c in zip(vals, counts):
    if c / SORTEIOS > 0.005:
        ax4.text(v, 100 * c / SORTEIOS + 0.8, pct(c / SORTEIOS, 0), ha='center', fontsize=11)
ax4.set_xticks(range(0, int(vals.max()) + 1)); ax4.set_xlabel('Convites aceitos de longe que viram "droga" quando o dia chega')
ax4.set_ylabel('Anos sorteados (%)')
ax4.set_title(f'Doze convites por ano: {vir(aceitos.mean())} aceitos, {vir(droga.mean())} arrependidos\nP(pelo menos um) = {pct((droga >= 1).mean())} · premissa de aceitação monótona, limite inferior', fontsize=13, pad=14)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-137-grafico-04-a-agenda.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 4 salvo!")
```

### 💡 O Insight

**A mesma tarefa é aceita por 39% das pessoas quando o custo em tempo é hoje, e por 56%
quando é daqui a um mês.** A diferença é de 17 pontos, e em 97% dos sorteios ela é
positiva. Quase um em cada três "sim" dados de longe viraria "não" na hora.

Com dinheiro acontece o contrário. 70% aceitam pagar hoje, 62% daqui a um mês. O prazo
faz o tempo parecer abundante e o dinheiro, não. Perguntadas quanto tempo vão ter mês que
vem, as pessoas respondem 8,2 numa escala em que 5,5 é "igual a hoje". E a folga de tempo
que esperam para amanhã não tem relação com a que esperam para daqui a duas semanas: a
previsão de tempo é chute, a de dinheiro não.

Numa agenda com um convite por mês, isso dá cerca de sete aceitos por ano e dois
arrependidos. A chance de passar o ano sem nenhum "sim, droga" é de uma em cinco.

### ⚠️ Limitações do Modelo

- **Estudantes, cenários hipotéticos.** Os sete surveys são com universitários americanos
  respondendo a situações descritas no papel. O estudo natural, com 178 pessoas, é o único
  com comportamento real, e mede presença numa palestra.
- **As contagens do Experimento 5 são reconstruídas** dos percentuais, com 66 por célula.
  O artigo informa N = 264 e o desenho 2 × 2; a divisão igual é premissa, e os intervalos
  herdam o arredondamento.
- **A curva do Experimento 1 é ilustrativa.** A escala vai de 1 a 10 e a média 8,2 com DP
  2,9 não é normal; a fração "acima de 5,5" é aproximação declarada.
- **O desvio-padrão do dinheiro no Experimento 1 e os do Experimento 6 não foram lidos.**
- **A agenda de um ano supõe aceitação monótona** (quem diria sim na hora diz sim de
  longe) e convites independentes. É um limite inferior do arrependimento, não previsão.
- O artigo mede a folga esperada, não a folga real. Se o mês que vem for de fato mais
  folgado, o "sim" de longe estava certo. O Experimento 7 sugere que não é.

*A ciência é honesta sobre o que não sabe. O modelo também.*
