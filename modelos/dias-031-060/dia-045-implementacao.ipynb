### 📖 A História

Ontem eu admiti aqui que não é preguiça que me mantém longe da corrida. É emoção. Entender isso alivia, mas não resolve. Porque eu continuo com o mesmo problema prático: como é que eu saio da cama e vou?

O que eu não sabia é que a ciência já tinha medido por que isso não funciona. E que existe uma coisa quase ridícula de simples que funciona muito melhor do que qualquer discurso motivacional: uma frase.

---

### 📚 O Conceito: Intenção de Meta vs Intenção de Implementação

Peter Gollwitzer separou dois tipos de intenção que a gente costuma confundir.

A **intenção de meta** diz o que você quer: "quero voltar a correr". Ela aponta o destino, mas não diz como chegar. É aí que a maioria dos planos morre.

A **intenção de implementação** é um plano no formato **se-então**: "se for terça às 7h, então eu corro no parque". Ela amarra um gatilho concreto (dia, hora, lugar) a uma ação específica. Quando o momento chega, você não precisa decidir nem negociar com a vontade. O ambiente puxa o gatilho, e a ação já está pré-decidida.

A diferença não é de esforço. É de arquitetura. Uma intenção diz o quê. A outra diz quando, onde e como, e é isso que vira comportamento.

### 🧮 O Modelo

Uso o experimento de exercício de Milne, Orbell & Sheeran para medir quanto uma frase se-então muda a probabilidade real de agir, e o tamanho de efeito agregado da meta-análise de Gollwitzer & Sheeran.

**Fontes:**
- Gollwitzer, P. & Sheeran, P. (2006). *Implementation Intentions and Goal Achievement: A Meta-Analysis of Effects and Processes*. **Meta-análise: 94 estudos · d = 0,65** (médio-a-grande).
- Milne, S., Orbell, S. & Sheeran, P. (2002). *Combining motivational and volitional interventions to promote exercise participation*. **British Journal of Health Psychology** · N=248.
  Exercício ao menos 1×/semana: controle e só-motivação 35-38%; grupo com plano se-então **91%**.

**Nota metodológica:** são experimentos com medida de comportamento e tamanho de efeito de meta-análise, não proporções de survey autorrelatado. Por isso o fator ×0.80 não se aplica.


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

# Paleta 365 Probabilidades
DOURADO = '#c8a84b'
VERMELHO = '#c0392b'
VERDE    = '#1a5f5a'
CINZA    = '#6b6a64'

np.random.seed(45)  # Dia #045

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---

# Milne, Orbell & Sheeran (2002) — British Journal of Health Psychology, N=248
# % que se exercitou ao menos 1x/semana (a fonte reporta 35-38% para os dois
# grupos sem plano; uso valores dentro da faixa relatada)
p_controle   = 0.35   # só acompanhou a frequência
p_motivacao  = 0.38   # motivação (leu sobre benefícios) — sem plano
p_plano      = 0.91   # motivação + plano se-então
N_milne      = 248

# Gollwitzer & Sheeran (2006) — meta-análise
d_meta   = 0.65   # tamanho de efeito médio-a-grande
k_estudos = 94

print("=" * 66)
print("  DADOS DA LITERATURA — INTENÇÕES DE IMPLEMENTAÇÃO")
print("=" * 66)
print(f"  Milne, Orbell & Sheeran 2002 (N={N_milne}) — exercício 1x/semana:")
print(f"  → Controle:               {p_controle*100:.0f}%")
print(f"  → Só motivação:           {p_motivacao*100:.0f}%")
print(f"  → Plano se-então:         {p_plano*100:.0f}%")
print()
print(f"  Gollwitzer & Sheeran 2006 — meta-análise:")
print(f"  → {k_estudos} estudos · d = {d_meta} (médio-a-grande)")
print("=" * 66)
```

```python
# --- O MODELO: quanto uma frase muda a probabilidade de agir ---

# O salto na probabilidade de agir
salto_pp = (p_plano - p_motivacao) * 100          # pontos percentuais
razao = p_plano / p_motivacao                     # quantas vezes mais provável
ganho_vs_controle = (p_plano - p_controle) * 100

# Motivação sozinha: o quanto adiantou vs não fazer nada
efeito_motivacao = (p_motivacao - p_controle) * 100

print("=" * 66)
print("  MODELO — O PODER DE UMA FRASE")
print("=" * 66)
print(f"  Probabilidade de fazer o que planejou:")
print(f"  → Só com motivação:       {p_motivacao*100:.0f}%")
print(f"  → Com plano se-então:     {p_plano*100:.0f}%")
print()
print(f"  → Salto:                  +{salto_pp:.0f} pontos percentuais")
print(f"  → {razao:.1f}× mais provável agir com a frase do que só querendo")
print()
print(f"  A SURPRESA:")
print(f"  → Motivação sozinha adiantou só +{efeito_motivacao:.0f}pp vs não fazer nada.")
print(f"    Ler sobre benefícios quase não mudou o comportamento.")
print(f"    A frase mudou.")
print()
print(f"  E não é sorte de um estudo: meta-análise de {k_estudos} estudos,")
print(f"  d = {d_meta} (efeito médio-a-grande).")
print()
print(f"  RESPOSTA:")
print(f"  → Uma única frase se-então leva a sua chance de agir")
print(f"    de ~{p_motivacao*100:.0f}% para {p_plano*100:.0f}%.")
print("=" * 66)
```

```python
# --- VISUALIZAÇÃO — 3 GRÁFICOS SEPARADOS ---

# ── GRÁFICO 1 — A vontade não move; a frase move ──
fig1, ax1 = plt.subplots(figsize=(10, 6))
cats = ['Controle', 'Só motivação\n(ler sobre benefícios)', 'Plano se-então\n(uma frase)']
vals = [p_controle*100, p_motivacao*100, p_plano*100]
cores = [CINZA, VERMELHO, VERDE]
bars = ax1.bar(cats, vals, color=cores, alpha=0.9, width=0.5)
ax1.set_ylabel('Fizeram exercício ao menos 1×/semana (%)')
ax1.set_ylim(0, 105)
ax1.set_title('A vontade não move. A frase move.\nExercício — Milne, Orbell & Sheeran 2002 (N=248)',
              fontsize=13, pad=15)
for bar, v in zip(bars, vals):
    ax1.text(bar.get_x()+bar.get_width()/2, v+2, f'{v:.0f}%',
             ha='center', fontweight='bold', fontsize=16)
plt.figtext(0.5, 0.005,
            'Motivação sozinha ≈ controle · o plano se-então mais que dobrou a ação | Milne et al. 2002 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-045-grafico-01-tres-grupos.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 — ASSINATURA ESTATÍSTICA: Cohen's d da meta-análise ──
fig2, ax2 = plt.subplots(figsize=(10, 5.5))
for x, lab in [(0.2, 'pequeno'), (0.5, 'médio'), (0.8, 'grande')]:
    ax2.axvline(x, color=CINZA, ls='--', lw=1, alpha=0.6)
    ax2.text(x, 1.28, lab, ha='center', fontsize=10, color=CINZA)
ax2.scatter([d_meta], [1], color=DOURADO, s=320, zorder=5, edgecolor='#333')
ax2.text(d_meta, 0.72, f'd = {d_meta}', ha='center', fontsize=15, color='#8a6d10', fontweight='bold')
ax2.set_yticks([])
ax2.set_ylim(0.5, 1.45)
ax2.set_xlim(0, 1.0)
ax2.set_xlabel("Cohen's d — efeito das intenções de implementação sobre atingir metas")
ax2.set_title(f"A ASSINATURA ESTATÍSTICA\nMeta-análise de {k_estudos} estudos: efeito médio-a-grande",
              fontsize=13, pad=32)
plt.figtext(0.5, 0.01,
            'Gollwitzer & Sheeran 2006 · 94 estudos · d=0,65 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-045-grafico-02-assinatura.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 — A anatomia da frase que funciona ──
fig3, ax3 = plt.subplots(figsize=(10, 6))
ax3.axis('off')
ax3.set_xlim(0, 10); ax3.set_ylim(0, 10)

# SE
ax3.add_patch(plt.Rectangle((0.4, 5.2), 4.2, 3.2, facecolor=VERDE, alpha=0.12,
                            edgecolor=VERDE, lw=2))
ax3.text(2.5, 7.7, 'SE', ha='center', fontsize=22, color=VERDE, fontweight='bold', family='monospace')
ax3.text(2.5, 6.7, 'o gatilho concreto', ha='center', fontsize=13, color='#333')
ax3.text(2.5, 6.0, 'dia · hora · lugar', ha='center', fontsize=13, color=VERDE, style='italic')

# seta
ax3.annotate('', xy=(6.3, 6.8), xytext=(4.7, 6.8),
             arrowprops=dict(arrowstyle='->', color=DOURADO, lw=3))

# ENTÃO
ax3.add_patch(plt.Rectangle((6.4, 5.2), 3.2, 3.2, facecolor=DOURADO, alpha=0.15,
                            edgecolor=DOURADO, lw=2))
ax3.text(8.0, 7.7, 'ENTÃO', ha='center', fontsize=22, color='#8a6d10', fontweight='bold', family='monospace')
ax3.text(8.0, 6.7, 'a ação', ha='center', fontsize=13, color='#333')
ax3.text(8.0, 6.0, 'pré-decidida', ha='center', fontsize=13, color='#8a6d10', style='italic')

# exemplo
ax3.text(5.0, 3.6, 'Se for terça às 7h,', ha='center', fontsize=17, color='#333',
         style='italic', family='serif')
ax3.text(5.0, 2.7, 'então eu corro no parque.', ha='center', fontsize=17, color='#333',
         style='italic', family='serif', fontweight='bold')

ax3.text(5.0, 9.2, 'A ANATOMIA DA FRASE QUE FUNCIONA', ha='center',
         fontsize=13, color=CINZA, family='monospace')
plt.figtext(0.5, 0.02,
            'Estrutura de uma intenção de implementação · Gollwitzer & Sheeran 2006 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-045-grafico-03-anatomia.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**Só querendo, 38% agiram. Com uma frase se-então, 91%.**

O detalhe que me desarmou: a motivação sozinha quase não mudou nada. O grupo que leu sobre os benefícios do exercício ficou praticamente igual ao grupo que não fez nada. Discurso motivacional é barulho bonito que não vira passo.

O que virou passo foi escrever quando, onde e como. "Se for terça às 7h, então eu corro no parque." Uma frase. Ela funciona porque tira a decisão do momento em que a vontade está fraca e a deixa pronta lá atrás, quando você ainda estava lúcida. Na hora, o gatilho aparece e a ação já está combinada.

E não é achismo de um estudo só. Numa meta-análise de 94 estudos, o efeito é médio-a-grande (d = 0,65). É das intervenções mais confiáveis da psicologia comportamental, e cabe numa linha.

Ontem eu entendi que adio por emoção. Hoje eu escrevo a minha frase: *se amanhã for dia de treino, às 7h, então eu calço o tênis e vou, antes de negociar.*

*Que promessa você vem adiando que viraria ação se você dissesse, agora, o dia, a hora e o lugar?*

### ⚠️ Limitações do Modelo

- Milne et al. 2002 tem N=248, população específica; a fonte reporta 35-38% para os grupos sem plano (usei valores dentro dessa faixa).
- Intenções de implementação funcionam melhor para metas que dependem de iniciar uma ação; efeitos podem ser menores em metas muito complexas ou de longo prazo.
- A medida de exercício foi coletada no estudo original; o d=0,65 é agregado de 94 estudos com desfechos variados.
- O Gráfico 3 é a estrutura conceitual da técnica, não um dado.
- O fator ×0.80 não foi aplicado: experimentos e tamanho de efeito, não survey autorrelatado.
