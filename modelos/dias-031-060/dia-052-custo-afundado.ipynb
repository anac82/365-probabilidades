### 📖 A História

Eu passo os meus dias fazendo modelos sobre decisão racional. Calculo probabilidades, peso custo e benefício, escrevo sobre escolher com dados.

A parte de mim que entende de estatística sabe muito bem que o que já foi gasto não volta. A outra parte finge que não ouviu.

Custo afundado é o nome elegante pra isso. E o mais irônico é que conhecer o conceito não me livra dele. Eu caio igual, com gráfico e tudo.

O estudo de hoje mostra por que isso acontece, e ainda deixa uma pista de quando esse efeito finalmente afrouxa.

---

### 📚 O Conceito: O Custo Afundado

Um custo afundado é o que você já gastou e não recupera, tempo, dinheiro, energia. A lógica racional é simples e implacável, na hora de decidir o que fazer agora, o que já foi não deveria pesar. Só importa o que vem daqui pra frente.

Só que a cabeça humana não funciona assim. Arkes e Blumer mostraram que quanto mais a gente investe em algo, mais difícil fica abandonar, mesmo quando abandonar seria a melhor escolha. Não é ganância nem burrice. É a nossa aversão a encarar uma perda como definitiva. Continuar mantém viva a ilusão de que o gasto ainda pode valer a pena.

O detalhe mais interessante é que esse efeito não é eterno. Ele é mais forte logo depois do gasto e vai perdendo força com o tempo, à medida que a decisão antiga vira passado distante.

---

### 🧮 O Modelo

Uso o experimento clássico dos bilhetes de teatro. Pessoas que compravam a temporada de um teatro universitário foram designadas por sorteio a pagar o preço cheio ou um desconto. Depois, mediram quantas peças cada uma foi assistir. A única diferença entre os grupos era quanto tinham pago. Comparo o comparecimento e separo a primeira da segunda metade da temporada.

**Fonte:**
- Arkes, H. R. & Blumer, C. (1985). *The Psychology of Sunk Cost*. **Organizational Behavior and Human Decision Processes**, 35(1), 124-140. Experimento de campo com compradores de temporada de teatro (N≈60, três grupos de preço).

**Nota metodológica:** os desfechos são comportamentais (comparecimento real e escolha em experimento), não proporções de survey autorrelatado. Por isso o fator ×0.80 não se aplica.


````python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

DOURADO = '#c8a84b'; VERMELHO = '#c0392b'; VERDE = '#1a5f5a'; CINZA = '#6b6a64'
np.random.seed(52)  # Dia #052
print("✅ Bibliotecas carregadas")
````

````python
# --- DADOS DA LITERATURA ---
# Arkes & Blumer (1985) — bilhetes de teatro — N≈60
# Peças assistidas na PRIMEIRA metade da temporada, por grupo de preço:
cheio_1 = 4.11    # pagou preço cheio
descP_1 = 3.32    # desconto pequeno
descG_1 = 3.29    # desconto grande
media_desc_1 = (descP_1 + descG_1) / 2

# SEGUNDA metade: a diferença entre os grupos desaparece (não significativa)
efeito_1 = cheio_1 - media_desc_1   # diferença na 1a metade
efeito_2 = 0.0                       # diferença na 2a metade (some)

print("=" * 66)
print("  DADOS DA LITERATURA — CUSTO AFUNDADO (Arkes & Blumer 1985)")
print("=" * 66)
print(f"  Peças assistidas na 1a metade da temporada:")
print(f"  → Preço cheio:      {cheio_1:.2f}")
print(f"  → Desconto pequeno: {descP_1:.2f}")
print(f"  → Desconto grande:  {descG_1:.2f}")
print(f"  A única diferença entre os grupos era quanto tinham pago.")
print()
print(f"  Na 2a metade da temporada:")
print(f"  → a diferença entre os grupos desaparece (não significativa)")
print("=" * 66)
````

````python
# --- O MODELO: o quanto o já-gasto prende, e por quanto tempo ---
pct_a_mais = efeito_1 / media_desc_1 * 100

print("=" * 66)
print("  MODELO — PRESO PELO QUE JÁ GASTOU")
print("=" * 66)
print(f"  1a metade da temporada:")
print(f"  → quem pagou o preço cheio foi a {efeito_1:.2f} peça a mais,")
print(f"    cerca de {pct_a_mais:.0f}% mais comparecimento que os descontos.")
print(f"  → mesma peça, mesmo teatro. A diferença era só o valor pago.")
print()
print(f"  2a metade da temporada:")
print(f"  → o efeito some. Todos passam a ir mais ou menos igual.")
print(f"  → o custo afundado prende, mas com prazo de validade.")
print()
print(f"  RESPOSTA:")
print(f"  → Sim, o que você já gastou te prende. Ele te faz insistir em algo")
print(f"    só pra não encarar o gasto como perda. Mas esse laço enfraquece")
print(f"    com o tempo, e a decisão de agora só deveria olhar pra frente.")
print("=" * 66)
````

````python
# --- VISUALIZAÇÃO — 3 GRÁFICOS SEPARADOS ---

# ── GRÁFICO 1 — Comparecimento na 1a metade, por preço pago ──
fig1, ax1 = plt.subplots(figsize=(10, 6))
cats = ['Pagou\npreço cheio', 'Desconto\npequeno', 'Desconto\ngrande']
vals = [cheio_1, descP_1, descG_1]
cores = [VERMELHO, CINZA, CINZA]
bars = ax1.bar(cats, vals, color=cores, alpha=0.9, width=0.55)
ax1.set_ylabel('Peças assistidas (1a metade da temporada)')
ax1.set_ylim(0, 4.8)
ax1.set_title('Quem pagou mais foi a mais peças\nA única diferença entre os grupos era o valor pago — Arkes & Blumer 1985',
              fontsize=13, pad=15)
for bar, v in zip(bars, vals):
    ax1.text(bar.get_x()+bar.get_width()/2, v+0.08, f'{v:.2f}', ha='center', fontweight='bold', fontsize=16)
plt.figtext(0.5, 0.005, 'Mesmo teatro, mesmas peças, só muda quanto se pagou | Arkes & Blumer 1985 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout(); plt.savefig('dia-052-grafico-01-comparecimento.png', dpi=150, bbox_inches='tight'); plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 — ASSINATURA: o efeito tem prazo de validade ──
fig2, ax2 = plt.subplots(figsize=(10, 6))
fases = ['1a metade\nda temporada', '2a metade\nda temporada']
efeitos = [efeito_1, efeito_2]
bars2 = ax2.bar(fases, efeitos, color=[DOURADO, VERDE], alpha=0.9, width=0.5)
ax2.set_ylabel('Efeito do custo afundado\n(peças a mais de quem pagou cheio)')
ax2.set_ylim(0, 1.05)
ax2.set_title('A ASSINATURA ESTATÍSTICA\nO custo afundado é real, mas tem prazo de validade',
              fontsize=13, pad=15)
ax2.text(0, efeito_1+0.03, f'+{efeito_1:.2f} peça', ha='center', fontweight='bold', fontsize=15, color='#8a6d1e')
ax2.text(1, 0.06, 'diferença\ndesaparece', ha='center', fontweight='bold', fontsize=13, color=VERDE)
ax2.annotate('', xy=(1, 0.04), xytext=(0, efeito_1),
             arrowprops=dict(arrowstyle='->', color=VERMELHO, lw=2, connectionstyle='arc3,rad=-0.2'))
plt.figtext(0.5, 0.005, 'Mais forte logo após o gasto, some com o tempo | Arkes & Blumer 1985 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout(); plt.savefig('dia-052-grafico-02-assinatura.png', dpi=150, bbox_inches='tight'); plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 — A armadilha, e a saída ──
fig3, ax3 = plt.subplots(figsize=(10, 6))
ax3.axis('off')
ax3.set_title('A armadilha do custo afundado\nE a única pergunta que importa na hora de decidir', fontsize=13, pad=20)
# armadilha (topo)
ax3.add_patch(plt.Rectangle((0.06, 0.55), 0.40, 0.30, transform=ax3.transAxes,
              facecolor='#fbeeed', edgecolor=VERMELHO, lw=2))
ax3.text(0.26, 0.76, 'A ARMADILHA', ha='center', fontsize=13, color=VERMELHO, fontweight='bold', transform=ax3.transAxes)
ax3.text(0.26, 0.635, '"Já investi tanto...\nnão posso desistir agora."', ha='center', va='center',
         fontsize=12.5, color='#7a2f28', transform=ax3.transAxes, style='italic')
# saída (topo direita)
ax3.add_patch(plt.Rectangle((0.54, 0.55), 0.40, 0.30, transform=ax3.transAxes,
              facecolor='#eef3f2', edgecolor=VERDE, lw=2))
ax3.text(0.74, 0.76, 'A SAÍDA', ha='center', fontsize=13, color=VERDE, fontweight='bold', transform=ax3.transAxes)
ax3.text(0.74, 0.635, '"Daqui pra frente,\nisso ainda vale a pena?"', ha='center', va='center',
         fontsize=12.5, color='#134b46', transform=ax3.transAxes, style='italic')
# base
ax3.add_patch(plt.Rectangle((0.20, 0.16), 0.60, 0.16, transform=ax3.transAxes,
              facecolor='#faf5e6', edgecolor=DOURADO, lw=2.5))
ax3.text(0.5, 0.24, 'O que já foi gasto não volta. Só conta o que vem a seguir.',
         ha='center', va='center', fontsize=13, color='#8a6d1e', fontweight='bold', transform=ax3.transAxes)
ax3.annotate('', xy=(0.42, 0.33), xytext=(0.30, 0.54), transform=ax3.transAxes,
             arrowprops=dict(arrowstyle='->', color=CINZA, lw=1.5))
ax3.annotate('', xy=(0.58, 0.33), xytext=(0.70, 0.54), transform=ax3.transAxes,
             arrowprops=dict(arrowstyle='->', color=CINZA, lw=1.5))
plt.figtext(0.5, 0.03, 'Trocar a pergunta do passado pela do futuro é como se sai da armadilha | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout(); plt.savefig('dia-052-grafico-03-armadilha.png', dpi=150, bbox_inches='tight'); plt.close()
print("✅ Gráfico 3 salvo!")
````

### 💡 O Insight

**Quem pagou o preço cheio foi a cerca de 24% mais peças do que quem pagou barato. Mesma peça, mesmo teatro. A única diferença era o valor que já tinha saído do bolso.**

Repare no absurdo. O dinheiro já tinha sido pago pelos dois grupos, o valor não voltava pra ninguém. A decisão racional de ir ou não a uma peça deveria depender só de uma coisa, se você quer ver aquela peça hoje. Mas quem gastou mais se sentia mais obrigado a comparecer, pra que o gasto "não fosse desperdiçado". Só que ir a uma peça que você não quer ver não recupera o dinheiro. Só soma uma noite chata à conta.

É esse o laço do custo afundado. Ele transforma um gasto que já era passado em uma corrente no presente. E funciona com tudo, o livro pela metade, o projeto que já cansou, o caminho em que você pôs anos.

Mas o estudo traz um alívio. Na segunda metade da temporada, a diferença entre os grupos sumiu. O custo afundado é mais forte logo depois do gasto e vai perdendo força com o tempo. Ou seja, quanto mais você olha aquele investimento como passado distante, e não como ferida aberta, mais livre fica pra decidir com clareza.

Eu, que faço modelos de decisão, caio nessa igual a todo mundo. A diferença é que agora sei qual pergunta trocar. Não "quanto eu já gastei nisso", mas "daqui pra frente, isso ainda vale a pena".

*O que você continua carregando só porque já pagou caro por ele, mesmo que ele não te sirva mais?*

### ⚠️ Limitações do Modelo

- É um estudo clássico, mas antigo (1985) e de amostra modesta (N≈60 no experimento de teatro); a magnitude pode variar por contexto e cultura.
- Os valores de comparecimento da primeira metade estão bem documentados; o N exato, os valores de desconto e os números da segunda metade vêm do artigo original de 1985.
- O efeito é comportamental e médio, não uma regra individual; nem todo mundo cai, e nem sempre com a mesma força.
- Custo afundado interage com outros fatores (compromisso público, identidade, esperança realista de melhora); o experimento isola o valor pago, não a vida toda.
- O fator ×0.80 não foi aplicado: medidas comportamentais de experimento, não survey autorrelatado.
