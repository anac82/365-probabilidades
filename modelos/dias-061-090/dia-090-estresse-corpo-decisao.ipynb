---

### 📖 A História

Você conhece esse dia.

O coração acelerado desde cedo, o e-mail que não devia ter sido
lido antes do café, a agenda desabando. E, no meio do caos, a
decisão importante esperando: a proposta, a conversa, o sim ou não.

E você adia. Porque todo mundo sabe que não se decide nada
estressado. O corpo está gritando; imagine o estrago que ele deve
estar fazendo no julgamento.

A ciência foi medir as duas coisas separadamente. O grito do corpo
e a mudança na decisão.

E descobriu que elas quase não moram na mesma casa.

A reação do corpo ao estresse agudo é um dos maiores efeitos que a
psicofisiologia conhece: **g=1,25**. A mudança nas decisões sob
risco, medida em 32 conjuntos de dados: **d=0,19**. O corpo reage
com um efeito **seis vezes e meia maior** do que a decisão.

O corpo grita. A decisão quase não ouve.

---

### 📚 O Conceito: A Fenda Entre o Que Grita e o Que Decide

O estresse agudo dispara dois sistemas: o autônomo, rápido
(coração, adrenalina), e o eixo do cortisol, lento. Os dois sobem
com efeitos enormes e intervalos apertados. Até aí, o senso comum
acerta.

O que o senso comum erra é a transmissão. Três fendas aparecem
nos dados:

**Fenda 1, corpo contra decisão:** g=1,25 na fisiologia contra
d=0,19 na decisão sob risco. E no desconto temporal (sacrificar o
futuro pelo agora, o que todo mundo jura que o estresse causa), o
efeito é **nulo**: SMD=-0,18 com intervalo [-0,57, +0,20].

**Fenda 2, sentir contra medir:** cortisol e frequência cardíaca
correlacionam ρ=-0,22, não significativo, com a sensação subjetiva.
Sentir-se estressado e estar com cortisol alto são coisas quase
descorrelacionadas. O termômetro interno é ruim.

**Fenda 3, o mecanismo:** cortisol administrado em laboratório não
reproduz os efeitos cognitivos do estresse (p=0,009). O vilão da
história nem é quem a história diz.

Nos Dias #051 e #058 o cortisol já apareceu: lá, como resposta à
incerteza e à segunda-feira. Hoje o objeto é outro: a fenda entre
a reação e a consequência.

---

### 🧮 O Modelo

Duas camadas, ambas com semente fixa (`seed=42`):

1. **Bayesiana, sobre contagem real.** A diferença na taxa de escolhas
   seguras entre pessoas sob estresse e sem estresse (Parslow & Rose,
   2022), com posterior por pessoa, como o artigo testou. Inclui, declarada,
   a armadilha de tratar as 1.940 escolhas como independentes.
2. **Simulação do mecanismo.** Monte Carlo com 100.000 pares "estressado
   contra calmo", usando os tamanhos de efeito medidos por outros: quantas
   vezes o corpo denuncia o estresse, e quantas vezes a decisão denuncia.
   É um "e se" com parâmetros medidos, não uma auditoria de ninguém.

**Fontes:**
- Gu, Y. et al. (2025) - Psychoneuroendocrinology - meta-análise de
  reatividade ao estresse agudo, k=171 estudos, **N=8.452**.
  Fisiológica geral **g=1,25 [1,13, 1,38]**; sistema autônomo
  g=1,38 [1,23, 1,54]; eixo HPA/cortisol g=1,04 [0,86, 1,21];
  psicológica g=0,93 [0,79, 1,09]. Correlações entre domínios
  fracas; cortisol × frequência cardíaca ρ=-0,22, n.s.
  **Fonte central.**
- Starcke, K. & Brand, M. (2016) - Effects of Stress on Decisions
  Under Uncertainty - Psychological Bulletin, 142(9), 909-933 -
  32 conjuntos de dados, **N=1.829**. Decisão sob risco: d=0,19
  (estresse psicossocial) e d=0,09 (sistêmico). **Erratum de 2016
  checado em 02/09/2026:** corrige apenas a descrição da relação
  em U invertido (observada só em mulheres num estudo citado);
  os d não são afetados. ICs dos d não reportados aqui (artigo
  pago), limitação declarada.
- Forbes, S. et al. (2024) - Neurobiology of Stress - desconto
  temporal sob estresse, k=11, N=1.097: **SMD=-0,18 [-0,57, +0,20]**,
  nulo.
- Reforços independentes da fenda: Sokol-Hessner et al. (2016),
  N=120: cortisol subiu com p=8,9×10⁻¹⁶ e a atitude ao risco ficou
  em IC [-0,05, +0,06]; Parslow & Rose (2022), *Judgment and Decision
  Making* 17(4), 883-936, **pré-registrado**, N=194 (97 por grupo, 20
  escolhas cada, desenho entre sujeitos), poder de 90% para d=0,47:
  12,67 contra 13,16 escolhas seguras em 20 (63,4% contra 65,8%),
  p=0,302; fator de Bayes do próprio artigo BF01=3,89 e BF0+=12,2;
  só 52,7% dos estressados tiveram resposta de cortisol; Fricke et al. (2025), **RCT
  farmacológico duplo-cego**, N=94: marcadores subiram, o
  comportamento não.
- Shields, G. S., Sazma, M. A. & Yonelinas, A. P. (2016) -
  Neuroscience and Biobehavioral Reviews - 51 estudos, N=2.486:
  a nuance de que o estresse não é inócuo (ver modelo).

**Fator ×0,80: não se aplica.** Meta-análises de tamanhos de
efeito com medidas fisiológicas e comportamentais. Sem correção.

**Suposição declarada:** as traduções de tamanho de efeito
(probabilidade de superioridade) são contas próprias assumindo
normalidade.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# A fenda entre a reação do corpo e a mudança na decisão

# Gu et al. (2025), Psychoneuroendocrinology, k=171, N=8.452
g_fisio,  ic_fisio  = 1.25, (1.13, 1.38)   # reatividade fisiológica geral
g_auto,   ic_auto   = 1.38, (1.23, 1.54)   # sistema nervoso autônomo
g_hpa,    ic_hpa    = 1.04, (0.86, 1.21)   # eixo HPA (cortisol)
g_psico,  ic_psico  = 0.93, (0.79, 1.09)   # reação psicológica (sentir)
rho_cortisol_fc = -0.22                     # cortisol × sensação: n.s.

# Starcke & Brand (2016), Psych Bulletin, 32 datasets, N=1.829
d_decisao_psicossocial = 0.19   # decisão sob risco, estresse psicossocial
d_decisao_sistemico    = 0.09   # estresse sistêmico
# (erratum checado: d intactos; ICs não reportados no acesso disponível)

# Forbes et al. (2024), k=11, N=1.097
smd_desconto, ic_desconto = -0.18, (-0.57, 0.20)   # desconto temporal: NULO

# Reforços independentes
sokol_p_cortisol = 8.9e-16          # cortisol subiu (N=120)
sokol_ic_risco = (-0.05, 0.06)      # atitude ao risco: nada
parslow_sem, parslow_com = 63.4, 65.8   # % escolhas seguras (N=194, pré-registrado)
parslow_p = 0.302
media_seguras_sem, media_seguras_com = 12.67, 13.16   # escolhas seguras em 20, por pessoa
bf01_parslow, bf0mais_parslow = 3.890, 12.209          # fatores de Bayes do próprio artigo
p_resposta_cortisol = 52.69                            # % dos estressados com resposta de cortisol
reverse_sem, reverse_com = 1.99, 1.63                  # trocas inconsistentes (ruído), de 10

# Shields et al. (2016), 51 estudos, N=2.486: a nuance
g_memoria,   ic_memoria   = -0.197, (-0.330, -0.064)  # memória de trabalho piora
g_inib_cog,  ic_inib_cog  = -0.208, (-0.379, -0.035)  # inibição cognitiva piora
g_inib_resp, ic_inib_resp =  0.296, ( 0.018,  0.573)  # inibição de resposta MELHORA

print("=" * 65)
print("  DADOS DA LITERATURA - ESTRESSE, CORPO E DECISÃO")
print("=" * 65)
print(f"\n  O GRITO DO CORPO (Gu 2025, k=171, N=8.452):")
print(f"  → Autônomo (coração, adrenalina):  g = {g_auto:.2f}  [{ic_auto[0]:.2f}, {ic_auto[1]:.2f}]")
print(f"  → Fisiológico geral:               g = {g_fisio:.2f}  [{ic_fisio[0]:.2f}, {ic_fisio[1]:.2f}]")
print(f"  → Cortisol (eixo HPA):             g = {g_hpa:.2f}  [{ic_hpa[0]:.2f}, {ic_hpa[1]:.2f}]")
print(f"  → Sensação de estresse:            g = {g_psico:.2f}  [{ic_psico[0]:.2f}, {ic_psico[1]:.2f}]")
print(f"\n  A RESPOSTA DA DECISÃO:")
print(f"  → Decisão sob risco (psicossocial): d = {d_decisao_psicossocial:.2f}")
print(f"  → Decisão sob risco (sistêmico):    d = {d_decisao_sistemico:.2f}")
print(f"  → Desconto temporal:                SMD = {smd_desconto:.2f} [{ic_desconto[0]:.2f}, {ic_desconto[1]:.2f}]  ← NULO")
print(f"\n  O TERMÔMETRO QUEBRADO:")
print(f"  → Cortisol × sensação: ρ = {rho_cortisol_fc:.2f}, não significativo")
print(f"\n  REFORÇOS INDEPENDENTES:")
print(f"  → Sokol-Hessner 2016 (N=120): cortisol p={sokol_p_cortisol:.1e};")
print(f"     atitude ao risco IC [{sokol_ic_risco[0]:.2f}, {sokol_ic_risco[1]:.2f}]")
print(f"  → Parslow & Rose 2022 (pré-registrado, N=194):")
print(f"     {parslow_sem}% vs {parslow_com}% de escolhas seguras, p={parslow_p}")
print(f"     97 pessoas por grupo, 20 escolhas cada · BF01={bf01_parslow} · só {p_resposta_cortisol}% com resposta de cortisol")
print(f"  → Fricke 2025 (RCT duplo-cego, N=94): marcadores sobem, comportamento não")
print(f"\n  Fator ×0,80: NÃO aplicado (meta-análises de efeito)")
print("=" * 65)
```

```python
# --- O MODELO ---
# Duas camadas: (1) bayesiana, sobre contagem real; (2) simulação, sobre o mecanismo

rng = np.random.default_rng(42)   # semente fixa: o dia roda igual em qualquer máquina
N_SIM = 100_000

# ────────────────────────────────────────────────────────────────
# CAMADA 1 - BAYESIANA: o estresse muda a taxa de escolhas seguras?
# Parslow & Rose (2022): 97 pessoas por grupo, 20 escolhas cada
# ────────────────────────────────────────────────────────────────
n_pessoas = 97
n_escolhas = 20
n_total = n_pessoas * n_escolhas                       # 1.940 escolhas por grupo
k_seguras_sem = round(media_seguras_sem * n_pessoas)   # 1.229 escolhas seguras
k_seguras_com = round(media_seguras_com * n_pessoas)   # 1.277 escolhas seguras

# 1a. A posterior CORRETA: por pessoa, como o artigo testou
# O artigo reporta a diferença de médias (0,49 escolha em 20 = 2,45 p.p.) e p = 0,302
# num teste t bicaudal com 192 graus de liberdade. Disso sai o erro-padrão.
dif_obs_pp = (media_seguras_com - media_seguras_sem) / n_escolhas * 100     # +2,45 p.p.
gl = 2 * n_pessoas - 2
t_obs = stats.t.ppf(1 - parslow_p / 2, gl)
se_pp = dif_obs_pp / t_obs                                                   # ≈ 2,4 p.p.
# Prior plana sobre a diferença: a posterior é a própria verossimilhança
dif_pp = rng.normal(dif_obs_pp, se_pp, N_SIM)

ic_dif = np.percentile(dif_pp, [2.5, 97.5])
p_mais_seguro = np.mean(dif_pp > 0)
p_mais_arriscado = np.mean(dif_pp < 0)
p_dif_maior_5pp = np.mean(np.abs(dif_pp) > 5)
p_dif_maior_10pp = np.mean(np.abs(dif_pp) > 10)

# 1b. A ARMADILHA, declarada: tratar as 1.940 escolhas como independentes
# Beta com prior de Jeffreys sobre a contagem bruta. Sai precisão falsa,
# porque as 20 escolhas de uma mesma pessoa são correlacionadas.
post_sem = stats.beta(0.5 + k_seguras_sem, 0.5 + n_total - k_seguras_sem)
post_com = stats.beta(0.5 + k_seguras_com, 0.5 + n_total - k_seguras_com)
dif_pp_ingenua = (post_com.rvs(N_SIM, random_state=rng) - post_sem.rvs(N_SIM, random_state=rng)) * 100
ic_dif_ingenua = np.percentile(dif_pp_ingenua, [2.5, 97.5])
p_mais_seguro_ingenua = np.mean(dif_pp_ingenua > 0)

# ────────────────────────────────────────────────────────────────
# CAMADA 2 - SIMULAÇÃO: o duelo corpo contra decisão
# Parâmetros medidos por outros (meta-análises); o sorteio é nosso
# ────────────────────────────────────────────────────────────────
# Sorteio de 100.000 pares "estressado contra calmo"
corpo_calmo      = rng.normal(0,       1, N_SIM)
corpo_estressado = rng.normal(g_fisio, 1, N_SIM)
dec_calmo        = rng.normal(0,       1, N_SIM)
dec_estressado   = rng.normal(d_decisao_psicossocial, 1, N_SIM)

p_corpo_denuncia = np.mean(corpo_estressado > corpo_calmo)      # esperado: Φ(g/√2)
p_decisao_denuncia = np.mean(dec_estressado > dec_calmo)        # esperado: Φ(d/√2)
razao_fenda = g_fisio / d_decisao_psicossocial

# E as escolhas do dia a dia: 100.000 pessoas fazem 20 escolhas calmas e 20 sob estresse
esc_calmo = rng.binomial(n_escolhas, media_seguras_sem / n_escolhas, N_SIM)
esc_estresse = rng.binomial(n_escolhas, media_seguras_com / n_escolhas, N_SIM)
mudou_seguras = esc_estresse - esc_calmo
media_escolhas_mudadas = np.mean(np.abs(mudou_seguras))
p_zero_mudanca = np.mean(mudou_seguras == 0)
uma_em_cada = n_escolhas / abs(media_seguras_com - media_seguras_sem)   # 20 / 0,49 ≈ 41

print("=" * 65)
print("  MODELO - CAMADA 1: BAYES SOBRE O EXPERIMENTO PRÉ-REGISTRADO")
print("=" * 65)
print(f"\n  Parslow & Rose 2022: {n_pessoas} pessoas por grupo, {n_escolhas} escolhas cada")
print(f"  → Escolhas seguras sem estresse:  {k_seguras_sem:,} de {n_total:,}  ({k_seguras_sem/n_total*100:.1f}%)")
print(f"  → Escolhas seguras sob estresse:  {k_seguras_com:,} de {n_total:,}  ({k_seguras_com/n_total*100:.1f}%)")
print(f"\n  Posterior da diferença (sob estresse menos sem), por pessoa, prior plana:")
print(f"  → Mediana: {np.median(dif_pp):+.1f} p.p.   IC 95%: [{ic_dif[0]:+.1f}, {ic_dif[1]:+.1f}] p.p.")
print(f"  → P(estresse deixa MAIS cauteloso):   {p_mais_seguro*100:.0f}%")
print(f"  → P(estresse deixa MAIS arriscado):   {p_mais_arriscado*100:.0f}%")
print(f"  → P(diferença maior que 5 p.p.):      {p_dif_maior_5pp*100:.0f}%")
print(f"  → P(diferença maior que 10 p.p.):     {p_dif_maior_10pp*100:.1f}%")
print(f"\n  A armadilha, declarada: Beta sobre as {n_total:,} escolhas como se fossem independentes")
print(f"  → IC 95%: [{ic_dif_ingenua[0]:+.1f}, {ic_dif_ingenua[1]:+.1f}] p.p. · P(mais cauteloso) = {p_mais_seguro_ingenua*100:.0f}%")
print(f"  → Precisão falsa: 20 escolhas da mesma pessoa não são 20 pessoas.")
print(f"     O artigo testa por pessoa. O modelo também.")
print(f"\n  O artigo confirma com fator de Bayes: BF01 = {bf01_parslow}, BF0+ = {bf0mais_parslow}")
print(f"  → Os dados são {bf0mais_parslow:.0f}x mais prováveis sob 'nada muda'")
print(f"     do que sob 'o estresse aumenta o risco'.")

print("\n" + "=" * 65)
print("  MODELO - CAMADA 2: SIMULAÇÃO DO DUELO (seed 42, N=100.000)")
print("=" * 65)
print(f"\n  Sorteie uma pessoa estressada e uma calma:")
print(f"  → Pelo CORPO (g={g_fisio:.2f}), a estressada se destaca em {p_corpo_denuncia*100:.0f}% dos duelos")
print(f"  → Pela DECISÃO (d={d_decisao_psicossocial:.2f}), em {p_decisao_denuncia*100:.0f}%. A moeda dá 50.")
print(f"  → Razão da fenda: {razao_fenda:.1f}x")
print(f"\n  A mesma pessoa, {n_escolhas} escolhas calma e {n_escolhas} sob estresse:")
print(f"  → Escolhas que mudam, em média:  {media_escolhas_mudadas:.2f} de {n_escolhas}")
print(f"     (a maior parte é ruído binomial, não estresse)")
print(f"  → Diferença sistemática esperada: 1 escolha a cada {uma_em_cada:.0f}")
print(f"  → E a direção mais provável é cautela, não impulso.")
print(f"\n  A NUANCE HONESTA (Shields 2016, N=2.486):")
print(f"  → Memória de trabalho:   g = {g_memoria:.3f}  [{ic_memoria[0]:.3f}, {ic_memoria[1]:.3f}]  piora um pouco")
print(f"  → Inibição cognitiva:    g = {g_inib_cog:.3f}  [{ic_inib_cog[0]:.3f}, {ic_inib_cog[1]:.3f}]  piora um pouco")
print(f"  → Inibição de resposta:  g = +{g_inib_resp:.3f}  [+{ic_inib_resp[0]:.3f}, +{ic_inib_resp[1]:.3f}]  MELHORA")
print(f"  → O estresse não é inócuo. É específico e pequeno.")
print(f"\n  A LEITURA:")
print(f"  → 'Nunca decida estressado' superestima o inimigo em {razao_fenda:.1f}x.")
print(f"  → O que o estresse muda de verdade é como você SE SENTE decidindo,")
print(f"     não, na mesma medida, o que você decide.")
print("=" * 65)
```

### 💡 O Insight

**O corpo reage ao estresse com g=1,25. A decisão muda com d=0,19.**

Seis vezes e meia de fenda entre o grito e a consequência.

Sorteie uma pessoa estressada e uma calma, 100.000 vezes. Pelo
corpo, a estressada se destaca em 81% dos duelos. Pela decisão, em
56%: quase uma moeda. E no desconto temporal, aquele que todo mundo
jura que o estresse causa, o efeito é estatisticamente nulo.

No experimento pré-registrado com poder de sobra, a posterior da
diferença entre estressados e calmos fica em **+2,4 pontos
percentuais, com intervalo de -2,2 a +7,1**. A chance de a diferença
passar de 5 pontos é de 14%. De passar de 10, uma em mil. E a
direção mais provável nem é a que o senso comum diz: sob estresse,
as pessoas ficam um pouco mais cautelosas, não mais impulsivas. Em
20 escolhas, o estresse muda sistematicamente uma a cada 41. O
próprio artigo fecha com fator de Bayes: os dados são 12 vezes mais
prováveis sob "nada muda" do que sob "o estresse aumenta o risco".

No RCT farmacológico, os marcadores subiram e o comportamento nem
se mexeu. E o termômetro interno é ruim: sentir-se estressado e
estar com cortisol alto correlacionam ρ=-0,22, não significativo.

A honestidade com o mesmo peso: o estresse não é inócuo. Ele é
**específico e pequeno**: a memória de trabalho piora um pouco, a
inibição cognitiva também, e a inibição de resposta até melhora.
E o mecanismo nem é o cortisol: administrado sozinho, ele não
reproduz os efeitos.

"Nunca decida estressado" superestima o inimigo em seis vezes.
O que o estresse muda de verdade é como você **se sente** decidindo.
Não, na mesma medida, o que você decide.

A pergunta que fica:

*Quantas decisões você adiou esperando uma calma
que o seu julgamento nunca precisou?*

### ⚠️ Limitações do Modelo

- Estresse AGUDO de laboratório (falar em público, água gelada),
  não estresse crônico da vida real; decisões medidas em tarefas
  de risco padronizadas, não em decisões de vida com consequências
- d=0,19 é pequeno, não zero: em decisões repetidas em alta
  frequência, efeitos pequenos acumulam; o claim do dia é a fenda,
  não a imunidade
- ICs dos d de Starcke & Brand não reportados aqui (artigo pago);
  o erratum de 2016 foi checado e não afeta os d, apenas a
  descrição de uma relação em U invertido restrita a mulheres em
  um estudo citado
- A simulação dos duelos (81% contra 56%) é um "e se" com parâmetros
  de meta-análise e normalidade assumida; mede uma propriedade do
  modelo, não uma pessoa real
- A posterior por pessoa foi derivada da diferença de médias e do
  p-valor reportados (o artigo não publica o desvio-padrão no texto);
  prior plana. A Beta sobre as 1.940 escolhas aparece só para mostrar
  a precisão falsa de ignorar que as escolhas de uma mesma pessoa são
  correlacionadas
- Parslow & Rose: amostra de estudantes de economia e direito,
  idade média de 21 anos; e só 52,7% dos estressados responderam com
  cortisol
- Heterogeneidade esperada entre tarefas, estressores e momentos
  de medida; Shields mostra que o efeito muda de sinal conforme a
  função medida
- Pontes declaradas: #051 e #058 (cortisol como resposta à
  incerteza e à segunda-feira; aqui, a fenda entre reação e
  consequência) e #042 (privação de sono é outro construto)

*A ciência é honesta sobre o que não sabe. O modelo também.*
