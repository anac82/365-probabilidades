## Qual a probabilidade de você ganhar se trocar de porta?

**Tipo:** Curiosidade · Probabilidade condicional
**Data de publicação:** 2026-11-07
**Ferramenta:** Python
**Decisão analisada:** Quando alguém que sabe a resposta me dá uma pista, devo mudar a escolha?
**Hashtag:** #365Probabilidades #Dia147

---

### 📖 A História

Quem cresceu no Brasil no fim dos anos 1980 e no começo dos 1990 deve lembrar da Porta dos Desesperados, do Sérgio Mallandro. Três portas no palco. Atrás de uma, um prêmio, uma bicicleta, um brinquedo. Atrás das outras, um "monstro", que quase sempre era alguém vestido de gorila. A criança escolhia, a plateia gritava, e a porta abria. Esses tempos li algo sobre o quadro e resolvi relembrar.

Só que hoje eu olho para aquele palco de outro jeito. Ele é a versão brasileira de um dos problemas de probabilidade mais famosos do mundo, e um dos que mais engana quem tenta resolver de cabeça.

O problema tem o nome de um apresentador de televisão americano, Monty Hall. Há três
portas. Atrás de uma está um carro, atrás das outras duas, nada. Você escolhe uma. O
apresentador, que sabe onde está o carro, abre uma das outras duas e mostra que ali não
tem nada. Então pergunta se você quer ficar com a sua porta ou trocar pela que sobrou.

Quase todo mundo responde que tanto faz. Sobraram duas portas, então seria meio a meio.
Não é. Trocar ganha em 2 de cada 3 jogos.

---

### 📚 O Conceito: a informação que o apresentador entrega

Quando você escolheu, tinha 1 chance em 3 de acertar. Nada do que acontece depois muda
isso, porque o apresentador sempre vai abrir uma porta vazia, qualquer que seja a sua. As
outras 2 chances em 3 estavam nas duas portas que você não escolheu. Ele elimina uma delas
e as 2 chances ficam inteiras na que sobrou.

O detalhe que decide tudo é que ele sabe onde está o carro e nunca abre a porta dele. Se
ele abrisse uma porta ao acaso e, por sorte, ela estivesse vazia, aí sim daria meio a meio.

---

### 🧮 O Modelo

Quatro camadas. A primeira e a última são contas nossas. As do meio vêm das fontes.

1. **A conta.** Bayes exato e 100 mil jogos sorteados, trocando e ficando. Mais a versão
   com N portas e a versão em que o apresentador abre uma porta ao acaso.
2. **O que as pessoas fazem.** Granberg & Brown 1995 e Mueser & Granberg 1999: quantas
   trocam na primeira vez e depois de jogar muitas vezes. Intervalo de Jeffreys.
3. **Os pombos.** Herbranson & Schroeder 2010: seis pombos e treze estudantes jogando
   o mesmo jogo por um mês.
4. **O aprendiz.** Um jogador simulado que só aprende pelos resultados (amostragem de
   Thompson): em quantos jogos ele passa a trocar quase sempre.

**Fonte central:**
- Granberg, D. & Brown, T. A. (1995). "The Monty Hall dilemma". *Personality and Social
  Psychology Bulletin* 21(7), 711-723. Na primeira vez, **12%** trocam; com o jogo
  repetido, a troca chega a **55%**.

**Segunda fonte (réplica e contexto):**
- Mueser, P. R. & Granberg, D. (1999). "The Monty Hall dilemma revisited: understanding
  the interaction of problem definition and decision making". Working paper, University
  of Missouri. **328 universitários**. Primeira resposta, sem dinheiro: 19% (N = 54),
  11% (N = 53) e 18% (N = 56) trocam, conforme a versão. Com dinheiro em jogo e o
  apresentador do lado do jogador, 67% a 79%.

**Terceira fonte (a curiosidade):**
- Herbranson, W. T. & Schroeder, J. (2010). "Are birds smarter than mathematicians?
  Pigeons (Columba livia) perform optimally on a version of the Monty Hall dilemma".
  *Journal of Comparative Psychology* 124(1), 1-13. Seis pombos passaram a trocar quase
  sempre em cerca de um mês. Treze estudantes, depois de 200 jogos, trocavam em cerca de
  dois terços.

**Sem fator ×0,80:** são escolhas feitas em experimento, não proporção de pesquisa.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

SEMENTE = 42
rng = np.random.default_rng(SEMENTE)

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

VERMELHO, DOURADO, VERDE, CINZA = '#c0392b', '#c8a84b', '#1a5f5a', '#6b6a64'
RODAPE = 'Fonte: cálculo próprio · Granberg & Brown (1995) · Mueser & Granberg (1999) · Herbranson & Schroeder (2010) | #365Probabilidades'

def vir(x, casas=1):
    return f"{x:.{casas}f}".replace(".", ",")

def mil(n):
    return f"{int(round(n)):,}".replace(",", ".")

def jeffreys(k, n):
    return stats.beta(k + 0.5, n - k + 0.5)

print("✅ Bibliotecas carregadas · semente", SEMENTE)
```

```python
# --- DADOS DA LITERATURA ---
GRANBERG = {'primeira_vez': 0.12, 'repetido': 0.55}
MUESER = {  # primeira resposta, sem dinheiro: (proporção que troca, N)
    'versão padrão':          (0.19, 54),
    'versão ambígua':         (0.11, 53),
    'apresentador ao acaso':  (0.18, 56),
}
MUESER_DINHEIRO = (0.67, 0.79)   # faixa com dinheiro e apresentador do lado do jogador
HERBRANSON = {'pombos': 6, 'humanos': 13, 'jogos_humanos': 200, 'humanos_trocam_fim': 2 / 3}

print("=" * 70)
print("  DADOS")
print("=" * 70)
print(f"  Granberg & Brown 1995 · trocam na primeira vez {vir(GRANBERG['primeira_vez'] * 100, 0)}% · com o jogo repetido {vir(GRANBERG['repetido'] * 100, 0)}%")
for k, (p, n) in MUESER.items():
    print(f"  Mueser & Granberg 1999 · {k:<22} {vir(p * 100, 0)}% (N = {n})")
print(f"  Mueser & Granberg 1999 · com dinheiro e apresentador do lado do jogador: {vir(MUESER_DINHEIRO[0] * 100, 0)}% a {vir(MUESER_DINHEIRO[1] * 100, 0)}%")
print(f"  Herbranson & Schroeder 2010 · {HERBRANSON['pombos']} pombos, {HERBRANSON['humanos']} estudantes, {HERBRANSON['jogos_humanos']} jogos cada")
print("=" * 70)
```

```python
# --- O MODELO ---

# CAMADA 1 · A conta
# Bayes: P(carro atrás da sua porta | ele abriu uma vazia) continua 1/3, porque ele sempre pode abrir uma vazia
print("=" * 70)
print("  CAMADA 1 · A CONTA")
print("=" * 70)
p_ficar_exato, p_trocar_exato = 1 / 3, 2 / 3
print(f"  Exato · ficar {vir(p_ficar_exato * 100)}% · trocar {vir(p_trocar_exato * 100)}%")

N_JOGOS = 100_000
carro = rng.integers(0, 3, N_JOGOS)
escolha = rng.integers(0, 3, N_JOGOS)
aberta = np.empty(N_JOGOS, int)
for i in range(N_JOGOS):
    opcoes = [p for p in range(3) if p != carro[i] and p != escolha[i]]
    aberta[i] = opcoes[rng.integers(len(opcoes))]
troca = 3 - escolha - aberta
ganha_ficar = (escolha == carro); ganha_trocar = (troca == carro)
print(f"  {mil(N_JOGOS)} jogos sorteados · ficar {vir(ganha_ficar.mean() * 100)}% · trocar {vir(ganha_trocar.mean() * 100)}%")
curva_ficar = np.cumsum(ganha_ficar) / np.arange(1, N_JOGOS + 1)
curva_trocar = np.cumsum(ganha_trocar) / np.arange(1, N_JOGOS + 1)

# N portas, apresentador abre todas menos uma
portas = np.arange(3, 101)
p_trocar_n = (portas - 1) / portas
print(f"  Com 10 portas, trocar ganha {vir(p_trocar_n[portas == 10][0] * 100)}% · com 100 portas, {vir(p_trocar_n[-1] * 100)}%")

# Apresentador que abre ao acaso (descarta os jogos em que ele revela o carro)
aberta_acaso = np.array([[p for p in range(3) if p != e][rng.integers(2)] for e in escolha])
valido = aberta_acaso != carro
troca_acaso = 3 - escolha - aberta_acaso
p_trocar_acaso = (troca_acaso[valido] == carro[valido]).mean()
print(f"  Se ele abrisse ao acaso e por sorte mostrasse uma porta vazia: trocar {vir(p_trocar_acaso * 100)}% ({mil(valido.sum())} jogos válidos)")

# CAMADA 2 · O que as pessoas fazem
print("\n" + "=" * 70)
print("  CAMADA 2 · O QUE AS PESSOAS FAZEM")
print("=" * 70)
k_tot = sum(round(p * n) for p, n in MUESER.values()); n_tot = sum(n for _, n in MUESER.values())
post = jeffreys(k_tot, n_tot); ic = post.interval(0.95)
print(f"  Mueser & Granberg, as três versões juntas: {k_tot} de {n_tot} trocam = {vir(k_tot / n_tot * 100)}% [{vir(ic[0] * 100)}; {vir(ic[1] * 100)}]")
p_troca = k_tot / n_tot
ganho_medio = p_troca * p_trocar_exato + (1 - p_troca) * p_ficar_exato
print(f"  Com {vir(p_troca * 100, 0)}% trocando, o grupo ganha o carro em {vir(ganho_medio * 100)}% dos jogos. Trocando sempre, ganharia {vir(p_trocar_exato * 100)}%.")
print(f"  Granberg & Brown: de {vir(GRANBERG['primeira_vez'] * 100, 0)}% na primeira vez para {vir(GRANBERG['repetido'] * 100, 0)}% com o jogo repetido.")

# CAMADA 3 · Os pombos
print("\n" + "=" * 70)
print("  CAMADA 3 · OS POMBOS")
print("=" * 70)
print(f"  {HERBRANSON['pombos']} pombos: quase sempre trocam depois de cerca de um mês.")
print(f"  {HERBRANSON['humanos']} estudantes: depois de {HERBRANSON['jogos_humanos']} jogos, trocam em cerca de {vir(HERBRANSON['humanos_trocam_fim'] * 100, 0)}% das vezes.")

# CAMADA 4 · O aprendiz (amostragem de Thompson com duas Betas, uma para ficar e outra para trocar)
N_APRENDIZES, N_RODADAS = 2000, 200
trocou = np.zeros((N_APRENDIZES, N_RODADAS), bool)
for a in range(N_APRENDIZES):
    v = np.ones(2); d = np.ones(2)          # [ficar, trocar]
    for t in range(N_RODADAS):
        acao = int(rng.beta(v[1], d[1]) > rng.beta(v[0], d[0]))
        ganhou = rng.random() < (p_trocar_exato if acao else p_ficar_exato)
        v[acao] += ganhou; d[acao] += 1 - ganhou
        trocou[a, t] = acao
janela = 10
taxa = np.array([trocou[:, max(0, t - janela + 1):t + 1].mean() for t in range(N_RODADAS)])
primeira_90 = int(np.argmax(taxa >= 0.90)) + 1 if (taxa >= 0.90).any() else None
print("\n" + "=" * 70)
print("  CAMADA 4 · O APRENDIZ")
print("=" * 70)
for t in [1, 10, 25, 50, 100, 200]:
    print(f"  jogo {t:>3}: troca em {vir(trocou[:, t - 1].mean() * 100, 0)}% dos aprendizes")
print(f"  O aprendiz passa de 90% de trocas por volta do jogo {primeira_90}.")
print(f"  Os estudantes de Herbranson, no jogo 200, ainda estavam perto de {vir(HERBRANSON['humanos_trocam_fim'] * 100, 0)}%.")
```

```python
# --- VISUALIZAÇÃO ---

# GRÁFICO 1 · A conta, jogo a jogo
fig1, ax1 = plt.subplots(figsize=(12, 8))
x = np.arange(1, N_JOGOS + 1)
ax1.plot(x, curva_trocar * 100, color=VERDE, lw=2.5, label='trocando')
ax1.plot(x, curva_ficar * 100, color=VERMELHO, lw=2.5, label='ficando')
ax1.axhline(200 / 3, color=VERDE, ls='--', lw=1); ax1.axhline(100 / 3, color=VERMELHO, ls='--', lw=1)
ax1.axhline(50, color=CINZA, ls=':', lw=1.5); ax1.text(1500, 51.5, 'meio a meio, o que quase todo mundo acha', color=CINZA, fontsize=11)
ax1.text(N_JOGOS * 0.6, 69, f'{vir(ganha_trocar.mean() * 100)}%', color=VERDE, fontsize=15, fontweight='bold')
ax1.text(N_JOGOS * 0.6, 28, f'{vir(ganha_ficar.mean() * 100)}%', color=VERMELHO, fontsize=15, fontweight='bold')
ax1.set_xscale('log'); ax1.set_ylim(0, 100)
ax1.set_xlabel('Jogos sorteados (escala logarítmica)', fontsize=12); ax1.set_ylabel('Jogos ganhos até ali (%)', fontsize=12)
ax1.set_title('Trocar ganha o dobro de ficar\n100 mil jogos sorteados', fontsize=14, pad=16)
ax1.legend(frameon=False, fontsize=12, loc='upper right')
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-147-grafico-01-a-conta.png', dpi=150, bbox_inches='tight'); plt.close()
print("✅ Gráfico 1 salvo!")

# GRÁFICO 2 · As portas
fig2, ax2 = plt.subplots(figsize=(12, 8))
ax2.plot(portas, p_trocar_n * 100, color=VERDE, lw=3, label='trocar, com o apresentador que sabe')
ax2.plot(portas, 100 / portas, color=VERMELHO, lw=3, label='ficar')
ax2.axhline(50, color=DOURADO, ls='--', lw=2, label='trocar, se ele abrisse as portas ao acaso')
for n in [3, 10, 100]:
    v = (n - 1) / n * 100
    ax2.plot(n, v, 'o', ms=11, color=VERDE); ax2.text(n, v - 6, f'{n} portas\n{vir(v)}%', ha='center', fontsize=11, color=VERDE)
ax2.set_xscale('log'); ax2.set_ylim(0, 105)
ax2.set_xticks([3, 5, 10, 20, 50, 100]); ax2.set_xticklabels(['3', '5', '10', '20', '50', '100'])
ax2.set_xlabel('Número de portas (o apresentador abre todas menos uma)', fontsize=12); ax2.set_ylabel('Chance de ganhar (%)', fontsize=12)
ax2.set_title('Com cem portas, ninguém ficaria com a primeira\no que decide é ele saber onde está o carro', fontsize=14, pad=16)
ax2.legend(frameon=False, fontsize=11, loc='center right')
plt.figtext(0.5, 0.01, 'Fonte: cálculo próprio | #365Probabilidades', ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-147-grafico-02-as-portas.png', dpi=150, bbox_inches='tight'); plt.close()
print("✅ Gráfico 2 salvo!")

# GRÁFICO 3 · O que as pessoas fazem
fig3, ax3 = plt.subplots(figsize=(12, 8))
itens = [('Granberg & Brown\nprimeira vez', GRANBERG['primeira_vez'], CINZA),
         ('Mueser & Granberg\nprimeira resposta', k_tot / n_tot, CINZA),
         ('Granberg & Brown\njogo repetido', GRANBERG['repetido'], DOURADO),
         ('Herbranson\nestudantes, jogo 200', HERBRANSON['humanos_trocam_fim'], DOURADO),
         ('Aprendiz simulado\njogo 200', trocou[:, -1].mean(), VERDE)]
xx = np.arange(len(itens))
ax3.bar(xx, [v * 100 for _, v, _ in itens], color=[c for *_, c in itens], width=0.6)
for i, (_, v, _) in enumerate(itens):
    topo = ic[1] * 100 if i == 1 else v * 100
    ax3.text(i, topo + 1.5, f"{vir(v * 100, 0)}%", ha='center', fontsize=14, fontweight='bold')
ax3.errorbar(1, k_tot / n_tot * 100, yerr=[[(k_tot / n_tot - ic[0]) * 100], [(ic[1] - k_tot / n_tot) * 100]], fmt='none', ecolor='#333', capsize=6)
ax3.set_xticks(xx); ax3.set_xticklabels([r for r, *_ in itens], fontsize=11)
ax3.set_ylim(0, 110); ax3.set_ylabel('Trocam de porta (%)', fontsize=12)
ax3.set_title('Quase ninguém troca na primeira vez\ne mesmo depois de muito jogo, as pessoas não chegam lá', fontsize=14, pad=16)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-147-grafico-03-as-pessoas.png', dpi=150, bbox_inches='tight'); plt.close()
print("✅ Gráfico 3 salvo!")

# GRÁFICO 4 · O aprendiz
fig4, ax4 = plt.subplots(figsize=(12, 8))
t = np.arange(1, N_RODADAS + 1)
ax4.plot(t, taxa * 100, color=VERDE, lw=3, label='aprendiz que só olha os resultados')
ax4.plot(N_RODADAS, HERBRANSON['humanos_trocam_fim'] * 100, 'o', ms=14, color=DOURADO, label='estudantes de Herbranson, jogo 200')
ax4.axhline(100, color=CINZA, ls=':', lw=1.5); ax4.text(4, 96.5, 'os pombos chegaram perto daqui em cerca de um mês', fontsize=11, color=CINZA)
ax4.set_ylim(0, 105); ax4.set_xlim(0, N_RODADAS + 5)
ax4.set_xlabel('Jogo', fontsize=12); ax4.set_ylabel('Trocam de porta (%), média dos últimos 10 jogos', fontsize=12)
ax4.set_title('Quem só aprende com o resultado passa a trocar\nas pessoas ficam no meio do caminho', fontsize=14, pad=16)
ax4.legend(frameon=False, fontsize=11, loc='lower right')
plt.figtext(0.5, 0.01, 'Fonte: cálculo próprio (amostragem de Thompson, 2.000 aprendizes) · Herbranson & Schroeder (2010) | #365Probabilidades', ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-147-grafico-04-o-aprendiz.png', dpi=150, bbox_inches='tight'); plt.close()
print("✅ Gráfico 4 salvo!")
```

### 💡 O Insight

**Trocar de porta ganha em 2 de cada 3 jogos. Ficar ganha em 1. Mesmo assim, só 12% a
16% das pessoas trocam na primeira vez, e um grupo que joga desse jeito leva o carro em
39% das vezes, quando poderia levar em 67%.**

O motivo é que o apresentador sabe onde está o carro. Quando ele abre uma porta vazia,
ele não muda a chance da sua porta, que continua 1 em 3. Ele só junta as outras 2 chances
na porta que sobrou. Com cem portas, isso fica óbvio. Ninguém ficaria com a primeira
depois de ver 98 portas vazias se abrirem.

E tem os pombos. Seis pombos aprenderam a trocar quase sempre em um mês. Treze
estudantes, depois de 200 jogos, ainda trocavam em dois terços das vezes. Um jogador
simulado que só olha os resultados passa de 90% de trocas por volta do jogo 44. O pombo
não tenta entender o problema. Ele só repara no que deu certo.

*Quando alguém que sabe mais do que você te mostra uma porta vazia, você ainda fica com a
sua primeira escolha?*

---

### ⚠️ Limitações do Modelo

- O resultado de 2 em 3 depende de o apresentador saber onde está o carro e sempre abrir
  uma porta vazia. Se ele abrisse ao acaso, trocar daria meio a meio (camada 1).
- Os números de Granberg & Brown 1995 vêm do resumo do artigo; o N de cada estudo não foi
  lido. Os de Mueser & Granberg 1999 vêm das tabelas do working paper, não de revista.
- Os percentuais dos pombos e dos estudantes de Herbranson & Schroeder 2010 foram lidos
  em divulgação do artigo ("quase sempre" e "cerca de dois terços"), não nas tabelas.
- O aprendiz simulado é um modelo, não uma pessoa nem um pombo. Ele mostra quanto
  resultado é preciso para a vantagem aparecer sozinha, não como um pombo aprende.
- Amostras pequenas (6 pombos, 13 estudantes, grupos de 53 a 56 universitários).
- Corda com o #138 (taxa-base e probabilidade condicional), objeto diferente.

*A ciência é honesta sobre o que não sabe. O modelo também.*
