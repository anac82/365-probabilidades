### 📖 A História

Toda estimativa de prazo é uma previsão sobre o futuro feita por alguém que tem acesso a um histórico e escolhe não olhar para ele.

A pessoa sabe que o último projeto atrasou. Sabe que o anterior também. E mesmo assim estima o próximo como se fosse a primeira vez.

Em 1994, três pesquisadores canadenses fizeram o teste mais limpo possível dessa distância entre o que a gente promete e o que a gente entrega. E, de quebra, deixaram dados que permitem transformar o achado numa regra de bolso.

---

### 📚 O Conceito: a visão de dentro

Kahneman e Tversky separaram dois jeitos de prever.

A **visão de dentro** olha para o plano: as etapas, o que precisa ser feito, quanto tempo cada parte deveria levar se as coisas andarem. É a visão natural, e é a que produz números bonitos.

A **visão de fora** ignora o plano e olha para a estatística: quanto tempo tarefas parecidas levaram, quantas atrasaram, por quanto. É a visão desconfortável, e é a que acerta.

A falácia do planejamento é o que acontece quando só existe a primeira. E o detalhe cruel é que ela sobrevive à experiência: as pessoas sabem que costumam atrasar e continuam achando que dessa vez o número está certo.

---

### 🧮 O Modelo

O estudo mediu previsão e realidade. Para transformar isso em ferramenta, o notebook faz três coisas.

1. **Calibra a distribuição do estouro.** Duas estatísticas publicadas amarram a distribuição da razão entre tempo real e tempo previsto: o estouro médio, de 64%, e a proporção que entregou no prazo, de 30%. São duas equações e dois parâmetros. Ajusto uma lognormal e, como teste de sensibilidade, repito com uma gama sob as mesmas duas restrições.
2. **Converte confiança em multiplicador.** Com a distribuição calibrada, cada nível de confiança vira um fator: por quanto multiplicar a própria estimativa para ter 50%, 80% ou 90% de chance de cumprir.
3. **Testa a régua fora do laboratório.** O remédio que a literatura recomenda contra a falácia do planejamento é trocar a visão de dentro pela visão de fora, olhando quanto tempo trabalhos parecidos levaram. A camada 4 faz isso com o próprio multiplicador, recalculando-o em quatro classes de referência, de estudantes a milhares de projetos de TI.
4. **Simula projetos com várias etapas.** 200.000 projetos, cada um com etapas de mesmo tamanho e estouro independente, para ver se a gordura necessária muda quando o trabalho é dividido.

**Fontes:**
- Buehler, R., Griffin, D. & Ross, M. (1994). "Exploring the 'planning fallacy': Why people underestimate their task completion times". *Journal of Personality and Social Psychology*, 67(3), 366-381. **N = 37 estudantes de psicologia**, previsão da própria monografia e data real de entrega registrada. Previsão realista de 33,9 dias, melhor cenário de 27,4, pior cenário de 48,6, tempo real de 55,5 dias, com 30% entregando dentro do prazo previsto.
- Segundo estudo da mesma equipe, sobre projetos acadêmicos pessoais, relatado no mesmo artigo e retomado em Buehler, Griffin & Ross (1995), *European Review of Social Psychology*, 6(1), 1-32: os alunos indicaram as datas em que tinham 50%, 75% e 99% de certeza de já ter terminado, e as proporções que realmente tinham terminado foram 13%, 19% e 45%. **N não identificado nas fontes consultadas.**
- Halkjelsvik, T. & Jørgensen, M. (2012). "From origami to software development: A review of studies on judgment-based predictions of performance time". *Psychological Bulletin*, 138(2), 238-271. Revisão integrativa do campo, que mostra que o achado de 1994 não é um estudo isolado. Distinção importante registrada por eles: o excesso de otimismo aparece na previsão de **quando a tarefa estará pronta**, e não na previsão de **quanto trabalho ela dá**. O multiplicador deste dia vale para a primeira.
- Halkjelsvik, T. & Jørgensen, M. (2018). *Time Predictions*. Simula SpringerBriefs on Computing 5, Springer, acesso aberto. Mediana de estouro de tempo em torno de **20%** em projetos grandes de software, com recomendação explícita de folga. **N não identificado.**
- Little, T. (2006). "Schedule estimation and uncertainty surrounding the cone of uncertainty". *IEEE Software*, 23(3), 48-54. **106 projetos de software** de uma empresa: prazo real de **2,0 vezes** o previsto em média, com a distribuição acumulada seguindo uma **lognormal**, e não uma normal. É o apoio empírico direto à família de distribuição escolhida aqui.
- Flyvbjerg, B., Budzier, A., Lee, J. S., Keil, M., Lunn, D. & Bester, D. W. (2022). "The Empirical Reality of IT Project Cost Overruns: Discovering A Power-Law Distribution". *Journal of Management Information Systems*, 39(3), 607-639. **5.392 projetos de TI** entre 2002 e 2014. Para o estouro de **prazo** (n=962), o melhor ajuste da cauda é uma **lei de potência** com alfa = 3,3 acima de 1,08. Para **custo** (n=4.677), a cauda é bem mais gorda, com alfa = 2,3. É o contraditório explícito a este modelo, e entra como camada 4.
- Base conceitual: Kahneman, D. & Tversky, A. (1979), "Intuitive prediction: biases and corrective procedures", sobre visão de dentro e visão de fora.

**Por que lognormal.** Três critérios, aplicados antes de olhar o resultado. Primeiro, a razão entre tempo real e previsto não pode ser negativa. Segundo, o atraso é multiplicativo e não aditivo: uma tarefa de dois dias não atrasa os mesmos quinze dias de uma de seis meses, ela atrasa uma proporção; quando o efeito é proporcional e vem do acúmulo de muitas causas, o logaritmo tende à normal. Terceiro, prazo tem piso e não tem teto, o que exige assimetria à direita. Little (2006) chegou empiricamente à mesma família para estouro de prazo.

**Por que duas restrições e não mais.** Existem exatamente duas estatísticas publicadas sobre essa razão: a média e a proporção dentro do prazo. Duas informações determinam dois parâmetros. Uma família de três parâmetros exigiria arbitrar um deles, e aí a conta viraria opinião com aparência de modelo.

**Por que a gama entra.** A escolha da família é a parte arbitrária. A gama satisfaz os mesmos três critérios, mas tem cauda mais leve. Se as duas discordassem muito, o dia não sustentaria uma regra de bolso.

**Onde a estimativa é forte e onde é fraca.** A mediana fica onde as restrições mandam, e as duas famílias concordam. O fator de 80% já é extrapolação. O de 90% está na cauda, que é exatamente onde nenhuma restrição tem força. Por isso a manchete do dia é 2,3 e não 3,0.

**Fator ×0,80:** não aplicado. A previsão é autorrelato, mas o desfecho é data real de entrega, registrada pelos pesquisadores.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats, optimize

SEED = 42
rng = np.random.default_rng(SEED)

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

DOURADO, VERMELHO, VERDE, CINZA = '#c9a14a', '#c0392b', '#2a8a82', '#6b6a64'

print("✅ Bibliotecas carregadas · semente", SEED)
```

```python
# --- DADOS DA LITERATURA ---
# Buehler, R., Griffin, D. & Ross, M. (1994). "Exploring the 'planning fallacy':
# Why people underestimate their task completion times."
# Journal of Personality and Social Psychology, 67(3), 366-381.

# Estudo da monografia · N = 37 estudantes de psicologia
N_TESE = 37
PREV_REALISTA = 33.9     # dias, previsão realista
PREV_MELHOR   = 27.4     # "se tudo correr o melhor possível"
PREV_PIOR     = 48.6     # "se tudo correr o pior possível"
REAL          = 55.5     # dias efetivamente levados
NO_PRAZO      = 0.30     # proporção que terminou dentro da própria previsão

# Segundo estudo da mesma equipe, sobre projetos acadêmicos pessoais:
# os alunos deram a data em que tinham 50%, 75% e 99% de certeza de já
# ter terminado. N não identificado nas fontes consultadas.
CONFIANCA_DECLARADA = np.array([0.50, 0.75, 0.99])
CONFIANCA_REAL      = np.array([0.13, 0.19, 0.45])

RAZAO_MEDIA = REAL / PREV_REALISTA

print("=" * 70)
print("  DADOS · A FALÁCIA DO PLANEJAMENTO")
print("=" * 70)
print(f"  Buehler, Griffin & Ross 1994 · N = {N_TESE} estudantes\n")
print(f"  {'Cenário':<34} {'Dias':>6}")
print(f"  {'Se tudo correr o melhor possível':<34} {PREV_MELHOR:>6.1f}")
print(f"  {'Previsão realista':<34} {PREV_REALISTA:>6.1f}")
print(f"  {'Se tudo correr o pior possível':<34} {PREV_PIOR:>6.1f}")
print(f"  {'REAL':<34} {REAL:>6.1f}")
print(f"\n  → O real superou a previsão realista em {REAL-PREV_REALISTA:.1f} dias ({RAZAO_MEDIA-1:+.0%})")
print(f"  → E superou o PIOR cenário imaginado em {REAL-PREV_PIOR:.1f} dias")
print(f"  → Terminaram dentro do prazo previsto: {NO_PRAZO:.0%}")
print(f"\n  Segundo estudo · datas por nível de certeza declarado:")
for d, r in zip(CONFIANCA_DECLARADA, CONFIANCA_REAL):
    print(f"  → diziam ter {d:.0%} de certeza · tinham terminado: {r:.0%}")
print(f"\n  Fator ×0,80: não se aplica. A previsão é autorrelato, mas o desfecho")
print(f"  é data real de entrega, registrada pelos pesquisadores.")
print("=" * 70)

# --- CLASSES DE REFERÊNCIA (camada 4) ---
# A falácia do planejamento não é um achado de 37 estudantes. Estas são as
# âncoras usadas para testar se o multiplicador sobrevive fora do laboratório.

# Halkjelsvik, T. & Jørgensen, M. (2012). "From origami to software development:
# A review of studies on judgment-based predictions of performance time."
# Psychological Bulletin, 138(2), 238-271. Revisão integrativa do campo.
# Distinção que importa: o excesso de otimismo aparece na previsão de QUANDO a
# tarefa estará pronta, e não na previsão de QUANTO TRABALHO ela dá.

# Halkjelsvik, T. & Jørgensen, M. (2018). Time Predictions. Simula SpringerBriefs
# on Computing 5, Springer (acesso aberto). Mediana de estouro de tempo em torno
# de 20% em projetos grandes de software. N não identificado.
MEDIANA_SOFTWARE = 1.20

# Little, T. (2006). "Schedule estimation and uncertainty surrounding the cone of
# uncertainty." IEEE Software, 23(3), 48-54. 106 projetos de software de uma
# empresa: razão prazo real/previsto de 2,0 em média, com a distribuição
# acumulada seguindo uma LOGNORMAL (e não uma normal).
RAZAO_LITTLE, N_LITTLE = 2.0, 106

# Flyvbjerg, B., Budzier, A., Lee, J. S., Keil, M., Lunn, D. & Bester, D. W.
# (2022). "The Empirical Reality of IT Project Cost Overruns: Discovering A
# Power-Law Distribution." Journal of Management Information Systems, 39(3),
# 607-639. 5.392 projetos de TI entre 2002 e 2014. Para o estouro de PRAZO
# (n=962), o melhor ajuste da cauda é uma LEI DE POTÊNCIA com alfa = 3,3 acima
# de x0 = 1,08. Para custo (n=4.677) a cauda é bem mais gorda: alfa = 2,3.
ALPHA_PRAZO, X0_PRAZO, N_PRAZO_TI = 3.3, 1.08, 962
ALPHA_CUSTO, X0_CUSTO = 2.3, 2.00
```

```python
# --- O MODELO ---
# Camada 1: calibrar a distribuição do estouro (real ÷ previsto)
# Camada 2: o multiplicador de prazo por nível de confiança
# Camada 3: o que acontece quando o projeto tem várias etapas
# Camada 4: a classe de referência (o multiplicador sobrevive fora do laboratório?)

# ---------------------------------------------------------------------------
# CAMADA 1
# Duas estatísticas publicadas amarram a distribuição da razão R = real/previsto:
#   E[R] = 55,5 / 33,9 = 1,637      e      P(R ≤ 1) = 0,30
# Duas equações, dois parâmetros. Ajusto uma lognormal e, como teste de
# sensibilidade, uma gama com as mesmas duas restrições.

def eqs_lognormal(par):
    mu, sigma = par
    return [np.exp(mu + sigma**2 / 2) - RAZAO_MEDIA,
            stats.lognorm(sigma, scale=np.exp(mu)).cdf(1.0) - NO_PRAZO]

mu, sigma = optimize.fsolve(eqs_lognormal, [0.3, 0.6])
LN = stats.lognorm(sigma, scale=np.exp(mu))

def eqs_gama(par):
    k, theta = par
    return [k * theta - RAZAO_MEDIA,
            stats.gamma(k, scale=theta).cdf(1.0) - NO_PRAZO]

k, theta = optimize.fsolve(eqs_gama, [3.0, 0.5])
GA = stats.gamma(k, scale=theta)

print("=" * 70)
print("  CAMADA 1 · A DISTRIBUIÇÃO DO ESTOURO")
print("=" * 70)
print(f"  Restrições usadas: média da razão = {RAZAO_MEDIA:.3f} e P(no prazo) = {NO_PRAZO:.0%}")
print(f"\n  Lognormal ajustada: mu = {mu:.3f}, sigma = {sigma:.3f}")
print(f"  → razão mediana: {LN.median():.2f}  (o prazo típico estoura em {LN.median()-1:+.0%})")
print(f"  → razão média:   {LN.mean():.2f}")
print(f"  → P(entregar no prazo prometido): {LN.cdf(1.0):.0%}")
print(f"  → P(levar o dobro ou mais):       {1-LN.cdf(2.0):.0%}")
print(f"\n  Gama ajustada (sensibilidade): k = {k:.2f}, escala = {theta:.3f}")
print(f"  → razão mediana: {GA.median():.2f} · P(dobro ou mais): {1-GA.cdf(2.0):.0%}")

# ---------------------------------------------------------------------------
# CAMADA 2 — o multiplicador
print("\n" + "=" * 70)
print("  CAMADA 2 · POR QUANTO MULTIPLICAR O SEU PRAZO")
print("=" * 70)
NIVEIS = [0.50, 0.70, 0.80, 0.90, 0.95]
print(f"  {'Para ter':<10} {'multiplique por':>16} {'(sensibilidade gama)':>22}")
MULT = {}
for c in NIVEIS:
    MULT[c] = LN.ppf(c)
    print(f"  {c:>8.0%}   {LN.ppf(c):>14.2f}x {GA.ppf(c):>20.2f}x")
print(f"\n  Exemplo prático: você estimou 10 dias.")
for c in (0.5, 0.8, 0.9):
    print(f"  → para {c:.0%} de chance de cumprir: prometa {10*LN.ppf(c):.0f} dias")

# ---------------------------------------------------------------------------
# CAMADA 3 — projetos com várias etapas
print("\n" + "=" * 70)
print("  CAMADA 3 · UM PROJETO É MAIS PREVISÍVEL QUE UMA TAREFA")
print("=" * 70)
R_SIM = 200_000
ETAPAS = [1, 4, 8, 16]
mult_etapas = {}
for n in ETAPAS:
    # n etapas de mesmo tamanho, cada uma com seu próprio estouro independente
    total = LN.rvs(size=(R_SIM, n), random_state=rng).mean(axis=1)
    mult_etapas[n] = {c: np.percentile(total, c * 100) for c in NIVEIS}
    print(f"  {n:>2} etapa(s): mediana {np.median(total):.2f}x · "
          f"para 80% de chance, multiplique por {np.percentile(total, 80):.2f} · "
          f"P(no prazo) = {(total <= 1).mean():.0%}")
print(f"\n  A mediana quase não muda, mas a gordura necessária encolhe:")
print(f"  {mult_etapas[1][0.8]:.2f}x para uma tarefa, {mult_etapas[16][0.8]:.2f}x para um projeto de 16 etapas.")
print(f"  Somar gordura tarefa por tarefa é exagero; a gordura pertence ao projeto.")

# ---------------------------------------------------------------------------
print("\n" + "=" * 70)
print("  A CALIBRAÇÃO DECLARADA (segundo estudo)")
print("=" * 70)
for d, r in zip(CONFIANCA_DECLARADA, CONFIANCA_REAL):
    print(f"  Certeza declarada {d:>4.0%} → terminaram {r:>4.0%}  (diferença de {(d-r)*100:.0f} pontos)")

print("\n" + "=" * 70)
print("  RESUMO PARA O INSIGHT")
print("=" * 70)
print(f"  Real {REAL:.1f} dias · previsto {PREV_REALISTA:.1f} · pior cenário {PREV_PIOR:.1f}")
print(f"  Estouro típico: {LN.median()-1:+.0%} · média {RAZAO_MEDIA-1:+.0%} · no prazo: {NO_PRAZO:.0%}")
print(f"  Multiplicador para 80%: {MULT[0.8]:.2f}x (gama: {GA.ppf(0.8):.2f}x) · para 90%: {MULT[0.9]:.2f}x")
print(f"  Projeto de 8 etapas, 80%: {mult_etapas[8][0.8]:.2f}x")
print(f"  Certeza de 99% declarada → 45% terminaram")

# ---------------------------------------------------------------------------
# CAMADA 4 — CLASSE DE REFERÊNCIA
# O remédio que a literatura recomenda contra a falácia do planejamento é trocar
# a visão de dentro pela visão de fora: olhar quanto tempo trabalhos parecidos
# levaram. Esta camada faz isso com o próprio multiplicador.
#
# Regra de calibração: quando a fonte publica UMA estatística só, mantenho a
# dispersão estimada com os dados de Buehler (sigma) e desloco apenas o centro.
# É suposição declarada, não dado.

print("\n" + "=" * 70)
print("  CAMADA 4 · A CLASSE DE REFERÊNCIA")
print("=" * 70)

classes = {}
classes['Monografia (Buehler, N=37)'] = LN
classes['Software, projetos grandes (mediana +20%)'] = stats.lognorm(sigma, scale=MEDIANA_SOFTWARE)
mediana_little = RAZAO_LITTLE / np.exp(sigma**2 / 2)          # média 2,0 -> mediana
classes['Software, 106 projetos (média 2,0x)'] = stats.lognorm(sigma, scale=mediana_little)

print(f"  {'Classe':<44} {'mediana':>8} {'80%':>7} {'90%':>7}")
for nome, d in classes.items():
    print(f"  {nome:<44} {d.median():>7.2f}x {d.ppf(0.80):>6.2f}x {d.ppf(0.90):>6.2f}x")

# Projetos de TI: a cauda do prazo é uma lei de potência, ajustada só acima de
# x0. O quantil abaixo é CONDICIONAL a já se estar nessa faixa.
b_pot = ALPHA_PRAZO - 1
q_ti = lambda c: X0_PRAZO * (1 - c) ** (-1 / b_pot)
print(f"  {'Projetos de TI (cauda de potência, n=962)':<44} {q_ti(0.50):>7.2f}x {q_ti(0.80):>6.2f}x {q_ti(0.90):>6.2f}x")
print(f"  (para TI, quantis CONDICIONAIS a estar na faixa de cauda, acima de {X0_PRAZO}x)")

print(f"\n  Multiplicador de 80% nas quatro âncoras: "
      f"{min([d.ppf(0.8) for d in classes.values()] + [q_ti(0.8)]):.1f}x a "
      f"{max([d.ppf(0.8) for d in classes.values()] + [q_ti(0.8)]):.1f}x")
print(f"  A régua de 2,3x fica dentro dessa faixa, e não depende de quem eram os 37.")

# ---------------------------------------------------------------------------
# Onde as famílias discordam: a cauda extrema
print("\n" + "=" * 70)
print("  A CAUDA EXTREMA · ONDE LOGNORMAL E LEI DE POTÊNCIA SE SEPARAM")
print("=" * 70)
print(f"  {'k':>5} {'lognormal':>12} {'lei de potência':>17}  razão")
for k in (1.5, 2, 3, 5, 10, 20):
    pl = (k / X0_PRAZO) ** -b_pot
    ln = 1 - LN.cdf(k)
    print(f"  {k:>5}x {ln:>11.4%} {pl:>16.4%}  {pl/ln:>6.1f}x")
print("\n  Até cerca de 3x as duas concordam, e é aí que mora a régua do dia.")
print("  Acima disso a lei de potência dá muito mais peso ao desastre:")
print(f"  em 10x, {(10/X0_PRAZO)**-b_pot / (1-LN.cdf(10)):.0f} vezes mais provável.")
print("  Mecanismo proposto por Flyvbjerg e colegas: interdependência entre")
print("  componentes, em que um atraso contamina os vizinhos em cascata.")
print("  Isso é o oposto da suposição de independência da Camada 3.")
print("\n  Leitura honesta: o multiplicador serve para o prazo do dia a dia.")
print("  Ele NÃO protege contra o desastre, que é raro e desproporcional.")
```

### 💡 O Insight

**Multiplique por 2,3.**

Essa é a regra que sai do modelo. Se você quer 80% de chance de cumprir o prazo que acabou de estimar, prometa 2,3 vezes esse número. Estimou dez dias, prometa 23. Para 90% de chance, o fator sobe para 3,0. Para simplesmente empatar, com 50% de chance, já é preciso 1,4.

O teste de sensibilidade não muda a recomendação: trocando a lognormal por uma gama sob as mesmas restrições, o fator de 80% vai a 2,4.

O número parece exagerado até você olhar de onde ele vem. Os estudantes previram 33,9 dias e levaram 55,5. Mas o dado que realmente assusta não é esse. É que, quando pediram para eles imaginarem o pior cenário possível, aquele em que tudo dá errado, eles disseram 48,6 dias.

**A realidade ficou sete dias além do pior caso que eles conseguiram imaginar.**

E quando outro grupo deu a data sobre a qual tinha 99% de certeza, quase certeza absoluta, **só 45% tinham terminado**.

Agora a parte que o modelo acrescenta ao estudo, e que muda como usar a regra.

Num projeto de oito etapas, o multiplicador de 80% cai de 2,3 para 1,9. Isso não é porque projetos grandes atrasam menos, eles atrasam na mesma proporção. É porque os estouros de cada etapa se compensam: algumas passam muito do previsto, outras nem tanto, e a soma fica mais previsível do que qualquer parte isolada.

Ou seja: quem coloca gordura em cada tarefa, uma a uma, está colocando gordura demais. **A gordura pertence ao projeto inteiro, não a cada pedaço.**

Falta responder a pergunta óbvia: 2,3 vezes é a régua do mundo ou é a régua de 37 estudantes de psicologia de 1994?

Refiz a conta em quatro classes de referência. Em projetos grandes de software, com mediana de estouro de 20%, o fator de 80% dá **2,0**. Nos 106 projetos que Little acompanhou, com prazo real de duas vezes o previsto em média, dá **2,8**. Na cauda de prazo dos 962 projetos de TI de Flyvbjerg e colegas, dá **2,2**. E nos estudantes, 2,3.

**Quatro mundos diferentes, quatro números entre 2,0 e 2,8.** A régua sobrevive fora do laboratório.

Mas a camada 4 traz uma má notícia junto, e ela é mais importante que a régua.

O estouro de prazo em projetos de TI não é lognormal: a cauda é uma lei de potência. Até três vezes o prazo prometido, as duas descrições dão quase o mesmo número, e é exatamente aí que o multiplicador trabalha. Acima disso elas se separam. Em dez vezes o prazo, a lei de potência diz que o desastre é **13 vezes mais provável** do que o meu modelo diz. Em vinte vezes, mais de trezentas vezes.

O mecanismo que eles propõem para isso é o mesmo que eu supus **não** existir na camada 3: interdependência. Quando um componente atrasa e contamina os vizinhos em cascata, a compensação entre etapas some e o estouro vira avalanche.

Então a leitura honesta do dia é esta. **O multiplicador serve para o prazo do dia a dia. Ele não protege contra o desastre.** Para o desastre não existe gordura suficiente, existe outra conversa: reduzir dependências, fatiar entrega, e aceitar que a cauda existe.

### ⚠️ Limitações do Modelo

- **A base é pequena e específica:** 37 estudantes de psicologia, uma tarefa acadêmica, num único estudo de 1994. O padrão se repete em outras pesquisas, mas os números exatos deste modelo vêm daqui.
- **A distribuição é suposição minha.** Não tenho os microdados, só duas estatísticas publicadas: o estouro médio e a proporção no prazo. Ajustei uma família de distribuições a esses dois números. Outra família daria fatores próximos, e é isso que o teste com a gama mostra, mas a forma exata da cauda não está medida, e é justamente a cauda que produz os desastres.
- **O N do segundo estudo não foi identificado** nas fontes que consultei, então os 13%, 19% e 45% entram como relato, sem intervalo de confiança.
- **A simulação de projeto supõe etapas independentes.** Na vida real os atrasos são correlacionados: a mesma pessoa cansada, o mesmo fornecedor atrasado, a mesma decisão pendente afetam tudo ao mesmo tempo. Com correlação, a gordura do projeto encolhe menos do que o modelo sugere. O 1,9 é um piso otimista.
- **Multiplicar prazo não corrige a causa.** O remédio testado na literatura é trocar a visão de dentro pela visão de fora, olhando quanto tempo tarefas parecidas levaram. Decompor a tarefa em partes também reduz o viés, segundo trabalhos posteriores.
- **As classes de referência da camada 4 são âncoras, não ajustes completos.** Quando a fonte publica uma estatística só, mantive a dispersão estimada com os dados de Buehler e desloquei apenas o centro. É suposição declarada: outra dispersão daria outro multiplicador para aquela classe.
- **Os quantis de projetos de TI são condicionais.** A lei de potência foi ajustada só à cauda, acima de 1,08 vez o prazo previsto, então o 2,2 vale dado que o projeto já está nessa faixa, e não para um projeto qualquer.
- **A mediana de 20% em projetos grandes de software vem de um livro de revisão**, sem N identificado nas fontes que consultei.
- **Prazo e custo não têm a mesma cauda.** No mesmo conjunto de dados, o custo é bem mais pesado (alfa 2,3) que o prazo (alfa 3,3). Este dia é sobre prazo.
- **E existe um efeito colateral conhecido:** prazo folgado tende a ser preenchido. O multiplicador serve para prometer, não necessariamente para planejar.

*A ciência é honesta sobre o que não sabe. O modelo também.*
