### 📖 A História

Existe uma crença confortável de que a felicidade é um projeto pessoal. Que depende de mindset, de escolhas, de trabalho interno. Em parte, é verdade. Mas há uma camada que essa crença esconde: o quanto da sua felicidade não vem de você.

Pesquisadores de Harvard e da Universidade da Califórnia em San Diego passaram mais de 20 anos rastreando uma rede social real — não online, mas de relações de vizinhança, amizade e parentesco — dentro do Framingham Heart Study, um dos estudos de saúde mais longos da história. Eles queriam responder uma pergunta simples: será que a felicidade se propaga de pessoa para pessoa, como uma espécie de contágio?

A resposta foi sim. E o alcance surpreendeu até os próprios pesquisadores: o efeito chega a se estender por três graus de separação — até o amigo do amigo do seu amigo, alguém que você provavelmente nunca vai conhecer.

---

### 📚 O Conceito: Contágio Emocional em Redes Sociais

O conceito de contágio emocional descreve a tendência humana de "capturar" emoções de outras pessoas através da observação e interação — um mecanismo estudado há décadas em laboratório, mas raramente medido em escala real e ao longo de tanto tempo.

O estudo de Fowler e Christakis (2008), publicado no *British Medical Journal*, acompanhou 4.739 pessoas de 1983 a 2003, medindo felicidade através de uma escala validada de quatro itens e mapeando minuciosamente as conexões sociais entre elas — amigos, vizinhos, irmãos, cônjuges, colegas de trabalho.

Um achado particularmente interessante quebra a intuição: a proximidade física importa mais que a proximidade formal do vínculo. Um vizinho de porta ao lado feliz teve mais impacto na felicidade da pessoa do que um cônjuge feliz que mora na mesma casa — provavelmente porque a interação cotidiana, ainda que mais superficial, acontece com mais frequência visível.

### 🧮 O Modelo
**Fontes:**
- Fowler, J.H. & Christakis, N.A. (2008) — *Dynamic Spread of Happiness in a Large Social Network: Longitudinal Analysis Over 20 Years in the Framingham Heart Study* — British Medical Journal, 337:a2338 — N=4.739


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print('✅ Bibliotecas carregadas')
```

```python
# --- DADOS DA LITERATURA ---
# Fowler & Christakis (2008) — BMJ — N=4.739, 20 anos de acompanhamento

n_estudo = 4739

# Aumento na probabilidade de felicidade por grau de separação social
aumento_grau_1 = 0.153   # conexão direta (amigo, familiar imediato)
aumento_grau_2 = 0.095   # uma vez removido (amigo de amigo)
aumento_grau_3 = 0.056   # duas vezes removido (amigo do amigo do amigo)

# Aumento por tipo de relação e proximidade física
aumento_amigo_1milha     = 0.25   # amigo que vive a até 1,6km, IC95% [1%, 57%]
aumento_amigo_meiamilha  = 0.42   # amigo que vive a até 0,8km
aumento_vizinho_porta    = 0.34   # vizinho de porta ao lado, IC95% [7%, 70%]
aumento_irmao_proximo    = 0.14   # irmão que vive a até 1,6km, IC95% [1%, 28%]
aumento_conjuge          = 0.08   # cônjuge coabitante, IC95% [0.2%, 16%]

fator_correcao = 0.80

print('=' * 65)
print('  DADOS DA LITERATURA — CONTÁGIO EMOCIONAL EM REDES SOCIAIS')
print('=' * 65)
print(f'\n  Fowler & Christakis 2008 (N={n_estudo:,}, 20 anos):')
print(f'  → Aumento de felicidade — conexão direta (1º grau) : +{aumento_grau_1*100:.1f}%')
print(f'  → Aumento de felicidade — 2º grau                  : +{aumento_grau_2*100:.1f}%')
print(f'  → Aumento de felicidade — 3º grau                  : +{aumento_grau_3*100:.1f}%')
print(f'\n  Por tipo de relação:')
print(f'  → Amigo a até 0,8km                : +{aumento_amigo_meiamilha*100:.0f}%')
print(f'  → Vizinho de porta ao lado          : +{aumento_vizinho_porta*100:.0f}%')
print(f'  → Amigo a até 1,6km                 : +{aumento_amigo_1milha*100:.0f}%')
print(f'  → Irmão que vive a até 1,6km        : +{aumento_irmao_proximo*100:.0f}%')
print(f'  → Cônjuge coabitante                : +{aumento_conjuge*100:.0f}%')
print(f'\n  Fator de correção aplicado: x{fator_correcao}')
print('=' * 65)
```

```python
# --- O MODELO ---
# O decaimento do efeito por grau de separacao segue padrao aproximadamente exponencial

graus = np.array([1, 2, 3])
aumentos = np.array([aumento_grau_1, aumento_grau_2, aumento_grau_3])

# Razao de decaimento entre graus consecutivos
razao_1_2 = aumento_grau_2 / aumento_grau_1
razao_2_3 = aumento_grau_3 / aumento_grau_2

# Efeito acumulado teorico ao longo da cadeia (efeito multiplicativo simplificado)
efeito_acumulado_3graus = (1 + aumento_grau_1) * (1 + aumento_grau_2) * (1 + aumento_grau_3) - 1

# Correcao aplicada ao efeito central (1o grau)
aumento_grau_1_corrigido = aumento_grau_1 * fator_correcao

print('=' * 65)
print('  MODELO — RESULTADO')
print('=' * 65)
print(f'  Decaimento do efeito por grau de separação:')
print(f'  → Razão 2º grau / 1º grau : {razao_1_2:.2f} (mantém {razao_1_2*100:.0f}% do efeito)')
print(f'  → Razão 3º grau / 2º grau : {razao_2_3:.2f} (mantém {razao_2_3*100:.0f}% do efeito)')
print()
print(f'  Efeito acumulado teórico ao longo de 3 graus: +{efeito_acumulado_3graus*100:.1f}%')
print()
print(f'  Aumento de felicidade (1º grau) corrigido (×0.80): +{aumento_grau_1_corrigido*100:.1f}%')
print('=' * 65)
```

```python
# --- VISUALIZACAO ---

# GRAFICO 1 — Decaimento do efeito por grau de separação
fig1, ax1 = plt.subplots(figsize=(12, 8))

ax1.plot(graus, aumentos * 100, color='#2a8a82', linewidth=3, marker='o', markersize=12)
ax1.fill_between(graus, aumentos * 100, alpha=0.15, color='#2a8a82')

for g, a in zip(graus, aumentos):
    ax1.annotate(f'+{a*100:.1f}%', xy=(g, a * 100), xytext=(0, 15),
                 textcoords='offset points', ha='center', fontsize=13, fontweight='bold', color='#2a8a82')

ax1.set_xticks(graus)
ax1.set_xticklabels(['1º grau\n(amigo direto)', '2º grau\n(amigo do amigo)', '3º grau\n(amigo do amigo\ndo amigo)'])
ax1.set_ylabel('Aumento na probabilidade de felicidade (%)', fontsize=12)
ax1.set_ylim(0, 20)
ax1.set_title('A felicidade se propaga até 3 graus de distância\nFowler & Christakis (2008) — N=4.739, 20 anos',
              fontsize=13, pad=15)

plt.tight_layout()
plt.savefig('dia-015-grafico-01-graus-separacao.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Grafico 1 salvo!')


# GRAFICO 2 — Efeito por tipo de relação e proximidade
fig2, ax2 = plt.subplots(figsize=(12, 8))

relacoes = ['Cônjuge\ncoabitante', 'Irmão\n(até 1,6km)', 'Amigo\n(até 1,6km)',
            'Vizinho de\nporta ao lado', 'Amigo\n(até 0,8km)']
valores2 = [aumento_conjuge * 100, aumento_irmao_proximo * 100, aumento_amigo_1milha * 100,
            aumento_vizinho_porta * 100, aumento_amigo_meiamilha * 100]
cores2 = ['#888888', '#c8a84b', '#c8a84b', '#2a8a82', '#2a8a82']

bars2 = ax2.barh(relacoes, valores2, color=cores2, alpha=0.85, height=0.55)
ax2.set_xlabel('Aumento na probabilidade de felicidade (%)', fontsize=12)
ax2.set_title('Proximidade física importa mais que o tipo formal do vínculo\nFowler & Christakis (2008) — N=4.739',
              fontsize=13, pad=15)

for bar, v in zip(bars2, valores2):
    ax2.text(bar.get_width() + 1, bar.get_y() + bar.get_height() / 2,
             f'+{v:.0f}%', va='center', fontweight='bold', fontsize=12)

plt.tight_layout()
plt.savefig('dia-015-grafico-02-tipo-relacao.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Grafico 2 salvo!')


# GRAFICO 3 — Distribuição do efeito (amigo a até 1,6km, IC 95% reportado)
fig3, ax3 = plt.subplots(figsize=(12, 8))

ic_amigo = (0.01, 0.57)  # IC 95% reportado no estudo
x = np.linspace(-0.05, 0.65, 1000)

# Aproximação da distribuição via normal, calibrada pelo IC reportado
media = aumento_amigo_1milha
se = (ic_amigo[1] - ic_amigo[0]) / (2 * 1.96)
y = stats.norm.pdf(x, loc=media, scale=se)

ax3.plot(x * 100, y, color='#2a8a82', linewidth=3)
ax3.fill_between(x * 100, y, alpha=0.2, color='#2a8a82')
ax3.axvline(x=ic_amigo[0] * 100, color='#c8a84b', linestyle='--', linewidth=2,
            label=f'IC 95%: [{ic_amigo[0]*100:.0f}%, {ic_amigo[1]*100:.0f}%]')
ax3.axvline(x=ic_amigo[1] * 100, color='#c8a84b', linestyle='--', linewidth=2)
ax3.axvline(x=media * 100, color='#2a8a82', linewidth=2,
            label=f'Estimativa central: +{media*100:.0f}%')
ax3.axvline(x=0, color='#333333', linestyle=':', linewidth=1.5, label='Sem efeito (0%)')

ax3.set_xlabel('Aumento na probabilidade de felicidade (%)', fontsize=12)
ax3.set_ylabel('Densidade', fontsize=12)
ax3.set_title('Distribuição do efeito de contágio — amigo próximo feliz\nFowler & Christakis (2008) — N=4.739',
              fontsize=13, pad=15)
ax3.legend(fontsize=11)

plt.tight_layout()
plt.savefig('dia-015-grafico-03-bayesiano.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Grafico 3 salvo!')
```

### 💡 O Insight

Um estudo de 20 anos com 4.739 pessoas mostrou que a felicidade se propaga por uma rede social real até **três graus de separação** — até o amigo do amigo do seu amigo, alguém que você provavelmente nunca vai conhecer pessoalmente. Ter um amigo próximo que fica feliz aumenta em **25% a 42%** a sua própria probabilidade de estar feliz, dependendo da proximidade física.

Curiosamente, a proximidade geográfica importa mais que o tipo formal do vínculo: um vizinho de porta ao lado feliz teve mais impacto (+34%) do que um cônjuge coabitante feliz (+8%) — provavelmente porque a interação cotidiana e visível acontece com mais frequência entre vizinhos do que se imagina.

E há uma assimetria importante: tristeza não se propaga pela rede com a mesma robustez que a felicidade. O efeito existe, mas é mais fraco e menos consistente — sugerindo que estar perto de pessoas felizes ajuda mais do que estar perto de pessoas tristes prejudica, ao menos nesse tipo específico de contágio social.

### ⚠️ Limitações do Modelo
- O estudo é observacional — não experimental — o que limita afirmações causais fortes, apesar dos controles estatísticos longitudinais aplicados pelos autores
- A amostra é de uma região específica dos EUA (Framingham, Massachusetts) ao longo de décadas — a generalização para outros contextos culturais e à era das redes sociais digitais requer cautela
- Os intervalos de confiança de alguns efeitos são amplos (ex: amigo a 1,6km: IC 95% [1%, 57%]), refletindo incerteza real na estimativa pontual
- "Felicidade" foi medida por uma escala de autorrelato de 4 itens — uma operacionalização específica que pode não capturar toda a complexidade do bem-estar subjetivo
- O fator de correção (×0,80) é uma aproximação padrão do projeto
