# 365 Probabilidades - Dia #163
## Qual a probabilidade de a sua melhor ideia vir na sua melhor hora do dia?

**Tipo:** Trabalho · Cognição  
**Data de publicação:** 2026-11-23  
**Ferramenta:** Python  
**Decisão analisada:** Em que hora do dia vale a pena reservar o trabalho que pede ideias novas?  
**Hashtag:** #365Probabilidades #Dia163

---

### 📖 A História

Existe um conselho que aparece em quase todo livro de produtividade. Acorde cedo, porque as melhores ideias vêm de manhã.

Há quem siga e funcione. E há quem passe a manhã inteira tentando e só comece a pensar direito no fim da tarde.

Fui ver se a hora das ideias é a mesma para todo mundo.

---

### 📚 O Conceito

Cada pessoa tem um relógio biológico com um horário próprio. A ciência chama esse horário de **cronotipo**. Os matutinos dormem e acordam cedo, os vespertinos dormem e acordam tarde, e a maioria fica no meio.

Muitas tarefas de atenção e memória rendem mais quando o horário do relógio de fora coincide com o do relógio de dentro. Esse encaixe se chama **efeito de sincronia**.

A pergunta de Jana Kühnel, Ronald Bledow e Markus Kiefer foi se a criatividade no trabalho segue a mesma regra.

---

### 🧮 O Modelo

Duas fontes e cinco contas.

1. **A vantagem do vespertino no fim da tarde.** O efeito do Estudo 1, em desvios-padrão, transformado na chance de um teste do fim da tarde superar um teste da manhã (200 mil sorteios, semente 42).
2. **Seis comparações, três estudos.** Para cada estudo e cada tipo (matutino e vespertino), a chance de o efeito ir na direção prevista pelos autores.
3. **E para quem fica no meio.** O efeito do horário calculado para o cronotipo médio, onde está a maioria.
4. **Quem estava nas amostras.** A parcela de matutinos, intermediários e vespertinos nos 817 participantes, com intervalo de Jeffreys.
5. **O contraponto de 2011.** A diferença do estudo de Wieth e Zacks, refeita com intervalo.

**Fontes:**
- Kühnel, J., Bledow, R. & Kiefer, M. (2022). There is a time to be creative: The alignment between chronotype and time of day. *Academy of Management Journal*, 65(1), 218-247. **Três estudos com 817 funcionários** (260, 238 e 319). Lido no texto completo, versão aceita no repositório da Singapore Management University.
- Wieth, M. B. & Zacks, R. T. (2011). Time of day effects on problem solving: When the non-optimal is optimal. *Thinking & Reasoning*, 17(4), 387-401. **223 estudantes analisados**. Lido no texto completo. Entra como contraponto.

**Nota metodológica sobre o fator ×0,80.** Não se aplica. O Estudo 1 usa um teste corrigido por avaliadores que não sabiam o horário. Os Estudos 2 e 3 usam escalas de criatividade, e não proporções de pesquisa.

**Cordas declaradas:** #013 (fazer nada para resolver um problema) e #075 (o relógio biológico). Aqui a pergunta é outra, a hora do dia em que cada pessoa é mais criativa.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

VERMELHO, DOURADO, VERDE, CINZA = '#c0392b', '#c8a84b', '#1a5f5a', '#6b6a64'
SEMENTE, N_SIM = 42, 200_000
rng = np.random.default_rng(SEMENTE)

def jeffreys(k, n, nivel=0.95):
    """Intervalo de Jeffreys para uma proporção."""
    a = (1 - nivel) / 2
    return stats.beta.ppf([a, 1 - a], k + 0.5, n - k + 0.5)

def sorteia(k, n):
    return rng.beta(k + 0.5, n - k + 0.5, N_SIM)

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---

# 1) Kühnel, Bledow & Kiefer (2022). Academy of Management Journal, 65(1), 218-247. Texto completo.
#    Cronotipo pelo meio do sono nos dias livres (MCTQ, Estudos 1 e 2) e pelo questionário rMEQ (Estudo 3).
#    Efeitos simples: mudança da criatividade da manhã para o fim da tarde, a 1 desvio-padrão do cronotipo médio.
#    Sinal positivo = mais criativo no fim da tarde. Sinal negativo = mais criativo de manhã.
estudos = {
    'Estudo 1': dict(n=260, quem='funcionários de várias áreas', medida='teste de usos alternativos, corrigido por dois avaliadores',
                     parcelas=(0.19, 0.66, 0.15), interacao=(0.12, 0.04),
                     vespertino=(0.17, 0.07), matutino=(-0.10, 0.07)),
    'Estudo 2': dict(n=238, quem='profissionais criativos (arquitetos, designers, artistas)', medida='escala de 7 pontos, autorrelato por período do dia',
                     parcelas=(0.24, 0.65, 0.11), interacao=(0.07, 0.03),
                     vespertino=(0.12, 0.04), matutino=(-0.02, 0.04)),
    'Estudo 3': dict(n=319, quem='funcionários recrutados online', medida='escala de 5 pontos, autorrelato às 9h e às 17h',
                     parcelas=(0.33, 0.50, 0.17), interacao=(0.03, 0.01),
                     vespertino=(0.07, None), matutino=(-0.18, 0.07)),
}
p_vesp_estudo3 = 0.350                   # o artigo dá p, e não o erro padrão, para este efeito
cronotipo_e1 = dict(media=3.91, dp=1.10)  # meio do sono nos dias livres, em horas depois da meia-noite (Estudo 1)
limite_matutino_e1 = 2.16                 # abaixo disso, a manhã foi significativamente melhor no Estudo 1
efeito_extremo_matutino = (-0.18, 0.09)

# 2) Wieth & Zacks (2011). Thinking & Reasoning, 17(4), 387-401. Texto completo. Contraponto.
wz = dict(n_fora=100, n_pico=123, insight_fora=0.42, insight_pico=0.33, qme=0.09, vespertinos=195, analisados=223)

aplica_fator_080 = False

fmt = lambda x, c=1: f"{x:.{c}f}".replace('.', ',')
pct = lambda p: 'acima de 99%' if p > 0.995 else '= ' + fmt(p*100, 0) + '%'
n_total = sum(e['n'] for e in estudos.values())
print("=" * 72)
print("  DADOS DA LITERATURA - A HORA DA IDEIA")
print("=" * 72)
print(f"\n  Kühnel, Bledow & Kiefer (2022): {n_total} funcionários em três estudos")
for nome, e in estudos.items():
    m, i, v = e['parcelas']
    print(f"\n  {nome}: {e['n']} {e['quem']}")
    print(f"  → Medida: {e['medida']}")
    print(f"  → Matutinos {fmt(m*100, 0)}%  ·  intermediários {fmt(i*100, 0)}%  ·  vespertinos {fmt(v*100, 0)}%")
    print(f"  → Interação cronotipo × horário: {fmt(e['interacao'][0], 2)} (erro padrão {fmt(e['interacao'][1], 2)})")
    print(f"  → Vespertinos, da manhã para a tarde: {fmt(e['vespertino'][0], 2):>6}  ·  matutinos: {fmt(e['matutino'][0], 2):>6}")
print(f"\n  Wieth & Zacks (2011): {wz['analisados']} estudantes, {wz['vespertinos']} vespertinos")
print(f"  → Problemas de insight: {fmt(wz['insight_fora']*100, 0)}% fora do melhor horário, {fmt(wz['insight_pico']*100, 0)}% no melhor horário")
print(f"\n  Fator ×0,80 aplicado: {'sim' if aplica_fator_080 else 'não'}")
print("=" * 72)
```

```python
# --- O MODELO ---

# O Estudo 3 publica p, e não o erro padrão, para o efeito nos vespertinos. Recuperamos o erro pelo p bilateral.
ep_v3 = abs(estudos['Estudo 3']['vespertino'][0]) / stats.norm.ppf(1 - p_vesp_estudo3 / 2)
estudos['Estudo 3']['vespertino'] = (estudos['Estudo 3']['vespertino'][0], ep_v3)

# Conta 1 · A vantagem do vespertino no fim da tarde (Estudo 1, criatividade em desvios-padrão)
ef, ep = estudos['Estudo 1']['vespertino']
sim_d = rng.normal(ef, ep, N_SIM)
duelo = stats.norm.cdf(sim_d / np.sqrt(2))       # chance de um teste da tarde superar um da manhã
duelo_ic = np.percentile(duelo, [2.5, 50, 97.5]) * 100
p_tarde_melhor = (sim_d > 0).mean()

# Conta 2 · Seis comparações, três estudos: chance de cada efeito ir na direção prevista
direcao = {}
for nome, e in estudos.items():
    for tipo, sinal in (('vespertino', +1), ('matutino', -1)):
        est, se = e[tipo]
        direcao[(nome, tipo)] = dict(est=est, se=se, ic=(est - 1.96 * se, est + 1.96 * se),
                                     p_prevista=stats.norm.cdf(sinal * est / se))
confirmadas = sum(1 for r in direcao.values() if r['ic'][0] > 0 or r['ic'][1] < 0)

# Conta 3 · O cronotipo médio: o efeito no meio é a média dos dois efeitos simples
meio = {nome: (e['vespertino'][0] + e['matutino'][0]) / 2 for nome, e in estudos.items()}
inclinacao_e1 = estudos['Estudo 1']['interacao'][0]
def efeito_e1(meio_do_sono):
    return meio['Estudo 1'] + inclinacao_e1 * (meio_do_sono - cronotipo_e1['media'])
duelo_meio = stats.norm.cdf(meio['Estudo 1'] / np.sqrt(2)) * 100

# Conta 4 · Quem estava nas amostras
contagens = {nome: [int(round(p * e['n'])) for p in e['parcelas']] for nome, e in estudos.items()}
soma = np.sum(list(contagens.values()), axis=0)
parcela_meio = soma[1] / soma.sum()
ic_meio = jeffreys(soma[1], soma.sum())
parcela_extremos = (soma[0] + soma[2]) / soma.sum()

# Conta 5 · O contraponto de 2011
ep_wz = np.sqrt(wz['qme'] * (1 / wz['n_fora'] + 1 / wz['n_pico']))
sim_wz = rng.normal(wz['insight_fora'] - wz['insight_pico'], ep_wz, N_SIM) * 100
ic_wz = np.percentile(sim_wz, [2.5, 50, 97.5])
p_wz = (sim_wz > 0).mean()

print("=" * 72)
print("  MODELO - RESULTADO")
print("=" * 72)
print("\n  CONTA 1 · O vespertino no fim da tarde (Estudo 1)")
print(f"  → Efeito: {fmt(ef, 2)} desvio-padrão  ·  P(mais criativo à tarde) {pct(p_tarde_melhor)}")
print(f"  → Num duelo entre um teste da tarde e um da manhã, a tarde vence {fmt(duelo_ic[1], 0)} de 100 vezes [{fmt(duelo_ic[0], 0)}; {fmt(duelo_ic[2], 0)}]")
print("\n  CONTA 2 · Seis comparações (direção prevista: vespertino melhor à tarde, matutino melhor de manhã)")
for (nome, tipo), r in direcao.items():
    marca = 'confirma' if (r['ic'][0] > 0 or r['ic'][1] < 0) else 'intervalo inclui o zero'
    print(f"  → {nome} · {tipo:<10} {fmt(r['est'], 2):>6} [{fmt(r['ic'][0], 2)}; {fmt(r['ic'][1], 2)}]  ·  P(direção prevista) {pct(r['p_prevista'])}  ·  {marca}")
print(f"  → Comparações em que o intervalo exclui o zero: {confirmadas} de 6")
print("\n  CONTA 3 · E para quem fica no meio")
for nome, v in meio.items():
    print(f"  → {nome}: efeito do horário no cronotipo médio = {fmt(v, 3)}")
print(f"  → No Estudo 1, num duelo, a tarde vence {fmt(duelo_meio, 0)} de 100 vezes para quem fica no meio")
print("\n  CONTA 4 · Quem estava nas amostras")
print(f"  → Matutinos {soma[0]}, intermediários {soma[1]}, vespertinos {soma[2]} (de {soma.sum()})")
print(f"  → Intermediários: {fmt(parcela_meio*100, 0)}% [{fmt(ic_meio[0]*100, 0)}; {fmt(ic_meio[1]*100, 0)}]  ·  extremos: {fmt(parcela_extremos*100, 0)}%")
print("\n  CONTA 5 · O contraponto de 2011")
print(f"  → Insight fora do melhor horário: +{fmt(ic_wz[1], 0)} pp [{fmt(ic_wz[0], 0)}; {fmt(ic_wz[2], 0)}]  ·  P(existe) = {fmt(p_wz*100, 0)}%")
print("=" * 72)
```

```python
# --- VISUALIZAÇÃO ---
RODAPE = 'Fonte: Kühnel, Bledow & Kiefer (2022), Academy of Management Journal 65(1). Conta própria | #365Probabilidades'

# GRÁFICO 1 · A curva do Estudo 1: o efeito do horário conforme o relógio biológico
fig1, ax1 = plt.subplots(figsize=(12, 8))
xs = np.linspace(1.8, 6.2, 200)
ax1.axvspan(1.8, cronotipo_e1['media'] - cronotipo_e1['dp'], color=DOURADO, alpha=0.10)
ax1.axvspan(cronotipo_e1['media'] + cronotipo_e1['dp'], 6.2, color=VERDE, alpha=0.10)
ax1.plot(xs, efeito_e1(xs), color=CINZA, linewidth=2.5)
for tipo, x, cor in (('matutino', cronotipo_e1['media'] - cronotipo_e1['dp'], DOURADO), ('vespertino', cronotipo_e1['media'] + cronotipo_e1['dp'], VERDE)):
    r = direcao[('Estudo 1', tipo)]
    ax1.errorbar(x, r['est'], yerr=1.96 * r['se'], fmt='o', color=cor, markersize=13, capsize=7, linewidth=2.5)
    ax1.text(x + 0.08, r['est'], f"{fmt(r['est'], 2)}", va='center', fontsize=13, fontweight='bold')
ax1.axhline(0, color=CINZA, linestyle='--', linewidth=1.2)
ax1.text(2.0, 0.33, 'mais matutinos', fontsize=12, color=DOURADO); ax1.text(5.2, 0.33, 'mais vespertinos', fontsize=12, color=VERDE)
ax1.text(6.15, 0.012, 'sem diferença', ha='right', fontsize=10, color=CINZA)
ax1.set_xlim(1.8, 6.2); ax1.set_ylim(-0.35, 0.38)
ax1.set_xlabel('Meio do sono nos dias livres (horas depois da meia-noite)')
ax1.set_ylabel('Criatividade no fim da tarde menos de manhã (desvios-padrão)')
ax1.set_title('Quanto mais tarde o relógio biológico, mais a tarde ajuda\nEstudo 1, 260 funcionários, teste corrigido por avaliadores', fontsize=14, pad=15)
plt.figtext(0.5, 0.005, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=[0, 0.03, 1, 1])
plt.savefig('dia-163-grafico-01-a-curva.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# GRÁFICO 2 · Seis comparações, três estudos
fig2, ax2 = plt.subplots(figsize=(12, 8))
nomes = list(estudos); ys = np.arange(len(nomes))[::-1] * 1.0
for y, nome in zip(ys, nomes):
    for tipo, desl, cor in (('vespertino', 0.15, VERDE), ('matutino', -0.15, DOURADO)):
        r = direcao[(nome, tipo)]
        ax2.errorbar(r['est'], y + desl, xerr=1.96 * r['se'], fmt='o', color=cor, markersize=12, capsize=5, linewidth=2.2,
                     label=('Vespertinos' if tipo == 'vespertino' else 'Matutinos') if y == ys[0] else None)
ax2.axvline(0, color=CINZA, linestyle='--', linewidth=1.2)
ax2.text(0.01, 2.45, 'melhor à tarde →', fontsize=11, color=CINZA); ax2.text(-0.01, 2.45, '← melhor de manhã', ha='right', fontsize=11, color=CINZA)
ax2.set_yticks(ys); ax2.set_yticklabels([f"{n}\n{estudos[n]['n']} pessoas" for n in nomes], fontsize=13)
ax2.set_ylim(-0.6, 2.65); ax2.set_xlim(-0.4, 0.4); ax2.legend(frameon=False, loc='lower right')
ax2.set_xlabel('Mudança da criatividade da manhã para o fim da tarde (unidades de cada estudo)')
ax2.set_title('Cada estudo confirma metade da previsão\nEfeito para matutinos e vespertinos, com intervalo de 95%', fontsize=14, pad=15)
plt.figtext(0.5, 0.005, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=[0, 0.03, 1, 1])
plt.savefig('dia-163-grafico-02-seis-comparacoes.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# GRÁFICO 3 · Quem estava nas amostras
fig3, ax3 = plt.subplots(figsize=(12, 8))
base = np.zeros(len(nomes))
for j, (rot, cor) in enumerate((('Matutinos', DOURADO), ('Intermediários', CINZA), ('Vespertinos', VERDE))):
    vals = np.array([estudos[n]['parcelas'][j] * 100 for n in nomes])
    ax3.barh(nomes, vals, left=base, color=cor, alpha=0.9 if j != 1 else 0.45, label=rot, height=0.55)
    for i, v in enumerate(vals):
        ax3.text(base[i] + v / 2, i, f"{fmt(v, 0)}%", ha='center', va='center', fontsize=14, fontweight='bold', color='white' if j != 1 else 'black')
    base += vals
ax3.invert_yaxis(); ax3.set_xlim(0, 100); ax3.set_xlabel('Participantes (%)'); ax3.tick_params(axis='y', labelsize=13)
ax3.legend(frameon=False, loc='upper center', bbox_to_anchor=(0.5, -0.1), ncol=3)
ax3.set_title(f'{fmt(parcela_meio*100, 0)}% dos participantes ficam no meio\nCronotipo dos {soma.sum()} funcionários dos três estudos', fontsize=14, pad=15)
plt.figtext(0.5, 0.005, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=[0, 0.03, 1, 1])
plt.savefig('dia-163-grafico-03-quem-estava.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")

# GRÁFICO 4 · O duelo do vespertino (Estudo 1)
fig4, ax4 = plt.subplots(figsize=(12, 8))
ax4.hist(duelo * 100, bins=120, color=VERDE, alpha=0.75, density=True)
for v, est in zip(duelo_ic, ('--', '-', '--')):
    ax4.axvline(v, color=DOURADO if est == '--' else 'black', linestyle=est, linewidth=2)
ax4.axvline(50, color=CINZA, linestyle=':', linewidth=1.5); ax4.text(49.8, ax4.get_ylim()[1] * 0.92, 'empate', ha='right', fontsize=11, color=CINZA)
ax4.text(duelo_ic[1] + 0.3, ax4.get_ylim()[1] * 0.92, f"{fmt(duelo_ic[1], 0)} de 100\n[{fmt(duelo_ic[0], 0)}; {fmt(duelo_ic[2], 0)}]", fontsize=13, fontweight='bold')
ax4.set_xlabel('Vezes em 100 que um teste do fim da tarde supera um teste da manhã')
ax4.set_ylabel('Densidade')
ax4.set_title('Para o vespertino, a tarde vence um pouco mais da metade das vezes\nEstudo 1, 200 mil sorteios', fontsize=14, pad=15)
plt.figtext(0.5, 0.005, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=[0, 0.03, 1, 1])
plt.savefig('dia-163-grafico-04-o-duelo.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 4 salvo!")
```

### 💡 O Insight

**Para quem é vespertino, o teste de criatividade do fim da tarde superou o da manhã em 55 de cada 100 vezes. Para quem fica no meio, como 6 em cada 10 participantes, a hora quase não mudou nada.**

A ligação entre o relógio biológico e a hora da criatividade apareceu nos três estudos, sempre pequena. E cada estudo confirmou metade da previsão. Nos dois primeiros, os vespertinos foram mais criativos à tarde e os matutinos não mudaram. No terceiro, os matutinos foram mais criativos de manhã e os vespertinos não mudaram.

O estudo de 2011 apontava o contrário, com mais problemas de ideia nova resolvidos fora do melhor horário. Ele usou quebra-cabeças com resposta certa e um grupo quase só de estudantes vespertinos. Os três estudos de 2022 mediram criatividade no trabalho, em 817 funcionários.

O que os dados permitem dizer é que não existe uma hora das ideias que valha para todo mundo. O conselho de acordar cedo para ter ideias funciona para uma parte das pessoas. Para quem rende mais tarde, a tarde ajuda um pouco. Para a maioria, que fica no meio, a hora pesa pouco.

A pergunta que fica:

*Você escolhe a hora do trabalho que pede ideias, ou ela é escolhida pela agenda?*

---

### ⚠️ Limitações do Modelo

- Só o Estudo 1 mede criatividade com um teste corrigido por avaliadores. Os Estudos 2 e 3 usam a avaliação da própria pessoa sobre a sua criatividade no trabalho.
- Os três estudos usam escalas diferentes, e por isso os efeitos não foram somados. A Conta 2 compara só a direção de cada efeito.
- Os efeitos são pequenos. No Estudo 1, a vantagem do vespertino é de 0,17 desvio-padrão. O duelo trata esse número como a diferença entre dois testes, o que é uma aproximação nossa.
- Os autores avisam que os cronotipos extremos ficaram pouco representados, em especial os vespertinos. O efeito para quem é muito matutino ou muito vespertino pode ser maior do que o calculado.
- No Estudo 1, a manhã só foi significativamente melhor para quem tem o meio do sono antes de cerca de 2h10 da manhã, um grupo pequeno.
- O Estudo 2 pede que a pessoa lembre o dia no fim do dia, o que pode distorcer a memória. No Estudo 3, os autores avisam que o horário de verão pode ter deslocado as medidas em relação ao relógio biológico.
- O texto lido não informa o país de cada estudo. Os participantes do Estudo 3 foram recrutados pela plataforma Amazon Mechanical Turk.
- O erro padrão do efeito dos vespertinos no Estudo 3 foi recuperado do valor de p publicado (0,35). As contagens por cronotipo foram reconstruídas das porcentagens publicadas.
- O título fala em "melhor ideia". Os estudos medem criatividade em um teste ou no trabalho, e não a melhor ideia de cada pessoa.
- O estudo de 2011 é um estudo só, com estudantes, e a maior parte da sua diferença vem de um único problema. Ele entra como contraponto, e não como resposta.

*A ciência é honesta sobre o que não sabe. O modelo também.*
