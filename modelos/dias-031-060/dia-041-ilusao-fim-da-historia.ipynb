### 📚 O Conceito: A Ilusão do Fim da História

Existe um erro que quase todo mundo comete, em qualquer idade. A gente olha para
trás e vê o quanto mudou. E olha para frente e imagina que, daqui pra frente, vai
ficar praticamente igual.

Jordi Quoidbach, Daniel Gilbert e Timothy Wilson chamaram isso de **ilusão do fim
da história**: a tendência de tratar o presente como o ponto final da própria
transformação, como se a pessoa que você é hoje fosse a pessoa que você será para
sempre. Adolescentes e avós cometem o mesmo erro. A história, como eles escreveram,
parece estar sempre terminando hoje.

E não é modéstia nem incerteza. É uma miopia sobre o próprio futuro.

---

### 🧮 O Modelo

O desenho é elegante. Você compara o quanto pessoas de uma certa idade **preveem**
que vão mudar nos próximos 10 anos com o quanto pessoas 10 anos mais velhas
**relatam** que mudaram na última década. Se a previsão for menor que o relato, existe
ilusão.

O ponto genial: para provar que o erro é de previsão e não de memória, os autores
compararam tudo com uma coorte real (MIDUS), cuja personalidade foi medida de verdade,
duas vezes, com 10 anos de intervalo. A memória bateu com a realidade. A previsão, não.

Uso a estabilidade medida (coeficiente de correlação intraclasse, ICC) como a
assinatura do modelo: quanto maior o ICC, menos mudança.

**Fonte:**
- Quoidbach, J., Gilbert, D. T. & Wilson, T. D. (2013). *The End of History Illusion*.
  **Science**, 339(6115), 96-98.
  6 estudos, N>19.000 (18-68 anos). Validação objetiva: coorte MIDUS, N=3.808.

*Nota metodológica: os números centrais são tamanhos de efeito e correlações validados
contra uma coorte longitudinal com medida objetiva. Por isso o fator de correção ×0.80,
reservado a proporções de survey autorrelatado, não se aplica.*


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

# Paleta 365 Probabilidades
DOURADO = '#c8a84b'
VERMELHO = '#c0392b'
VERDE    = '#1a5f5a'
CINZA    = '#6b6a64'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Fonte: Quoidbach, Gilbert & Wilson (2013), Science 339(6115)

# Estabilidade da personalidade (ICC): quanto MAIOR, MENOS mudança
icc_real     = 0.52   # mudança REAL (coorte MIDUS, medida 2x em 10 anos, N=3.808)
icc_memoria  = 0.51   # o que as pessoas LEMBRAM ter mudado (bate com a realidade)
icc_previsao = 0.65   # o que as pessoas PREVEEM mudar (esperam ser mais estáveis)

# Força da ilusão (efeito de condição, |beta|) por domínio — todos p<0.001
beta_personalidade = 0.14   # N=7.519
beta_valores       = 0.46   # N=2.717
beta_preferencias  = 0.19   # N=7.130
N_personalidade, N_valores, N_preferencias = 7519, 2717, 7130

# Consequência prática (estudo 4, N=170)
pago_futuro   = 129   # disposição a pagar por show da banda favorita ATUAL, daqui 10 anos
pago_presente = 80    # disposição a pagar por show da banda favorita de 10 anos ATRÁS, agora
sobrepreco    = (pago_futuro - pago_presente) / pago_presente  # 61%

# Quanto as pessoas se acham mais "prontas" do que serão
excesso_estabilidade = (icc_previsao - icc_real) / icc_real   # ~25%

print("=" * 65)
print("  DADOS DA LITERATURA — A ILUSÃO DO FIM DA HISTÓRIA")
print("=" * 65)
print(f"  Estabilidade da personalidade (ICC):")
print(f"  → Mudança REAL (coorte MIDUS):         {icc_real:.2f}")
print(f"  → O que a MEMÓRIA relata:              {icc_memoria:.2f}  (bate com a real)")
print(f"  → O que a PREVISÃO estima:             {icc_previsao:.2f}  (superestima estabilidade)")
print(f"  → As pessoas se acham {excesso_estabilidade*100:.0f}% mais 'prontas' do que serão")
print()
print(f"  Força da ilusão (|beta|), sempre na mesma direção:")
print(f"  → Personalidade (N={N_personalidade}):        {beta_personalidade}")
print(f"  → Valores (N={N_valores}):              {beta_valores}")
print(f"  → Preferências (N={N_preferencias}):        {beta_preferencias}")
print()
print(f"  Custo concreto: pagariam {sobrepreco*100:.0f}% a mais (${pago_futuro} vs ${pago_presente})")
print("=" * 65)
```

```python
# --- O MODELO: memória honesta, previsão míope ---

# Proxy de "quantidade de mudança" = 1 - estabilidade (ICC)
mud_real     = 1 - icc_real
mud_memoria  = 1 - icc_memoria
mud_previsao = 1 - icc_previsao

# O quanto a previsão subestima a mudança real
subestima = (mud_real - mud_previsao) / mud_real

print("=" * 65)
print("  MODELO — A ILUSÃO DO FIM DA HISTÓRIA")
print("=" * 65)
print(f"  Mudança real de personalidade em 10 anos (1-ICC):   {mud_real:.2f}")
print(f"  Mudança que a memória reconhece:                    {mud_memoria:.2f}")
print(f"  Mudança que a previsão admite:                      {mud_previsao:.2f}")
print()
print(f"  → A previsão enxerga só {mud_previsao/mud_real*100:.0f}% da mudança que de fato acontece.")
print(f"  → Subestima a própria transformação em {subestima*100:.0f}%.")
print()
print(f"  A memória acerta. A previsão erra.")
print(f"  Você sabe o quanto mudou. Só não acredita que vai mudar de novo.")
print()
print(f"  E o preço disso é real: {sobrepreco*100:.0f}% de sobrepreço por apostar")
print(f"  que um gosto de hoje ainda vai valer o mesmo daqui a 10 anos.")
print("=" * 65)
```

```python
# --- VISUALIZAÇÃO — 3 GRÁFICOS SEPARADOS ---

# ── GRÁFICO 1 — Memória honesta, previsão míope (os 3 ICCs) ──
fig1, ax1 = plt.subplots(figsize=(10, 6))
labels = ['Mudança REAL\n(coorte medida\n2x em 10 anos)',
          'O que a MEMÓRIA\nreconhece\n(passado)',
          'O que a PREVISÃO\nadmite\n(futuro)']
vals = [mud_real, mud_memoria, mud_previsao]
cores = [VERDE, VERDE, VERMELHO]
bars = ax1.bar(labels, vals, color=cores, alpha=0.85, width=0.55)
ax1.set_ylabel('Mudança de personalidade em 10 anos\n(1 − ICC · maior = mais mudança)')
ax1.set_ylim(0, 0.6)
ax1.set_title('A memória acerta. A previsão erra.\nAs pessoas sabem o quanto mudaram, mas não acreditam que vão mudar de novo',
              fontsize=12.5, pad=15)
for bar, v in zip(bars, vals):
    ax1.text(bar.get_x()+bar.get_width()/2, v+0.012, f'{v:.2f}',
             ha='center', fontweight='bold', fontsize=14)
ax1.annotate('', xy=(2, mud_previsao+0.02), xytext=(0, mud_real+0.02),
             arrowprops=dict(arrowstyle='->', color=CINZA, lw=1.5, ls='--'))
ax1.text(1, 0.52, f'a previsão enxerga só\n{mud_previsao/mud_real*100:.0f}% da mudança real',
         ha='center', fontsize=10, color=VERMELHO, fontweight='bold')
plt.figtext(0.5, 0.005,
            'Coorte MIDUS N=3.808 · Quoidbach, Gilbert & Wilson 2013 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-041-grafico-01-memoria-previsao.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 — ASSINATURA: a ilusão é robusta em 3 domínios ──
fig2, ax2 = plt.subplots(figsize=(10, 6))
dom = ['Preferências\n(N=7.130)', 'Personalidade\n(N=7.519)', 'Valores\n(N=2.717)']
betas = [beta_preferencias, beta_personalidade, beta_valores]
ypos = np.arange(len(dom))
ax2.barh(ypos, betas, color=DOURADO, alpha=0.9, height=0.5)
ax2.set_yticks(ypos)
ax2.set_yticklabels(dom)
ax2.set_xlabel('Força da ilusão  |β| do efeito de condição  (todos p < 0.001)')
ax2.set_xlim(0, 0.55)
for i, b in enumerate(betas):
    ax2.text(b+0.008, i, f'{b:.2f}', va='center', fontweight='bold', fontsize=13)
ax2.set_title('A ASSINATURA ESTATÍSTICA\nA ilusão aparece em tudo: no que você gosta, em quem você é, no que você valoriza',
              fontsize=12, pad=15)
plt.figtext(0.5, 0.005,
            'Predictors preveem menos mudança que reporters relatam, nos 3 domínios | Quoidbach et al. 2013 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-041-grafico-02-assinatura.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 — O custo concreto da ilusão ──
fig3, ax3 = plt.subplots(figsize=(10, 6))
cats = ['Pagar HOJE por um show\nda banda favorita de\n10 anos atrás',
        'Pagar hoje por um show\nda banda favorita ATUAL\ndaqui a 10 anos']
valores = [pago_presente, pago_futuro]
cores3 = [CINZA, VERMELHO]
bars3 = ax3.bar(cats, valores, color=cores3, alpha=0.85, width=0.45)
ax3.set_ylabel('Disposição a pagar (US$)')
ax3.set_ylim(0, 150)
ax3.set_title(f'O custo concreto: {sobrepreco*100:.0f}% de sobrepreço\nApostamos dinheiro na ideia de que nossos gostos de hoje vão durar',
              fontsize=13, pad=15)
for bar, v in zip(bars3, valores):
    ax3.text(bar.get_x()+bar.get_width()/2, v+3, f'${v}',
             ha='center', fontweight='bold', fontsize=16)
plt.figtext(0.5, 0.005,
            'Estudo 4, N=170 | Quoidbach, Gilbert & Wilson 2013 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-041-grafico-03-custo.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**A mudança real na personalidade em 10 anos é grande. A memória reconhece isso com
precisão. Mas a previsão enxerga só três quartos dela: a gente se acha 25% mais
"pronta" do que de fato vai ser.**

O detalhe que dói é que não é falta de memória. Você sabe o quanto mudou. Olha para
a pessoa de 10 anos atrás e mal a reconhece. O erro é acreditar que, a partir de agora,
a régua parou de andar. Que essa versão, finalmente, é a definitiva.

Não é. Nunca é. Aos 20, aos 40, aos 60, todo mundo tem a mesma sensação de ter
chegado, e todo mundo continua mudando na mesma velocidade de antes.

E isso custa caro. As pessoas pagam 61% a mais por experiências ligadas a gostos que
assumem ser permanentes. A gente toma decisões grandes hoje para agradar uma pessoa
que não vai mais existir amanhã.

Talvez a leitura mais leve seja essa: se você não gosta de quem é hoje, a estatística
está do seu lado. Você vai mudar mais do que imagina. E se gosta, aproveite, porque
essa também é só mais uma versão de passagem.

*Que decisão você está tomando hoje para agradar a pessoa que você tem certeza de que
vai continuar sendo?*

### ⚠️ Limitações do Modelo

- O "1 − ICC" é usado como proxy de magnitude de mudança para facilitar a leitura; o
  dado original é a correlação intraclasse (ICC-A1), não uma contagem direta de mudança.
- Boa parte das amostras veio do site de um programa de TV francês, com 80% de mulheres;
  os autores replicaram os efeitos separadamente por gênero, mas a amostra não é
  perfeitamente representativa.
- A coorte objetiva (MIDUS) mede personalidade; valores e preferências dependem de
  autorrelato, ainda que validados de forma cruzada.
- A ilusão diminui um pouco com a idade em alguns domínios, mas permaneceu presente até
  no grupo mais velho analisado (50+).

*A ciência é honesta sobre o que não sabe. O modelo também.*
