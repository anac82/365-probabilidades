### 📚 O Conceito: o riso como coisa de grupo

A pesquisa sobre o riso costuma tratá-lo como um comportamento social. A frase mais
repetida da área diz que as pessoas riem 30 vezes mais acompanhadas do que sozinhas.

Dois psicólogos canadenses, Rod Martin e Nicholas Kuiper, fizeram a conta de outro
jeito. Pediram a 80 pessoas que anotassem todas as risadas durante três dias, com a hora,
o motivo e quem estava junto. Com esse diário dá para responder duas perguntas
diferentes. De cada cem risadas, quantas acontecem a sós? E qual a chance de uma pessoa
rir sozinha pelo menos uma vez num dia?

A segunda pergunta é a do título. Ela não está respondida no artigo. Sai de uma conta
sobre os números dele, e a conta precisa de uma premissa sobre o quanto as pessoas
diferem entre si.

---

### 🧮 O Modelo

Cinco camadas. A fonte dá as contagens. As contas são nossas.

1. **De cada cem risadas.** A fatia das risadas que acontece a sós, com o intervalo
   contado de dois jeitos (por risada e por pessoa).
2. **Risadas a sós por dia.** O ritmo diário, e a chance de pelo menos uma num dia se
   todas as pessoas rissem no mesmo ritmo (processo de Poisson).
3. **As pessoas não são iguais.** O estudo registra dias com zero risada e dias com 89.
   Uso essa faixa para calibrar o quanto o ritmo varia de pessoa para pessoa, e refaço a
   chance de rir sozinho num dia.
4. **Uma semana.** Monte Carlo de sete dias, com a variação calibrada.
5. **E as 30 vezes?** Quanto tempo a pessoa precisaria passar sozinha para os 12% e as
   30 vezes serem o mesmo fato.

**Fonte central:**
- Martin, R. A. & Kuiper, N. A. (1999). "Daily occurrence of laughter: relationships
  with age, gender, and Type A personality". *Humor: International Journal of Humor
  Research* 12(4), 355-384. **N = 80** adultos da comunidade (50 mulheres e 30 homens),
  de 17 a 79 anos, com diário de risadas de três dias. Média de 17 a 18 risadas por dia
  (mulheres 18,5 e homens 16,1, sem diferença significativa), variando de 0 a 89.
  **Cerca de 88% das risadas com outras pessoas e 12% a sós.** Origem: 56% situações do
  momento, 18% mídia, 15% lembranças, 11% piadas. **Números lidos na descrição da tabela
  na página do artigo, arredondados. A tabela e o desvio padrão não foram lidos.**

**Fonte de apoio, lida em fonte secundária:**
- Provine, R. R. & Fischer, K. R. (1989). "Laughing, smiling, and talking: relation to
  sleeping and social context in humans". *Ethology* 83(4), 295-305. Origem da frase das
  30 vezes, lida em Wood, Templeton, Morrel, Schubert & Wheatley (2022), *Philosophical
  Transactions of the Royal Society B* 377(1863), 20210187, que a atribui a Provine.
  Amostra e método do estudo original não foram conferidos.

**Nota metodológica.** O fator ×0,80 do projeto não se aplica. Ele vale para proporções
de pesquisa de opinião autorrelatada. Aqui a medida é um diário de comportamento, em que
o registro é a própria medida, como no #111.

**Ponte interna:** #018 (senso de humor e longevidade), com outra fonte.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

SEMENTE = 42
SORTEIOS = 200_000

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

VERMELHO, DOURADO, VERDE, CINZA = '#c0392b', '#c8a84b', '#1a5f5a', '#6b6a64'

def mil(n):
    return f"{int(round(n)):,}".replace(",", ".")

def vir(x, casas=1):
    return f"{x:.{casas}f}".replace(".", ",")

def jeffreys(k, n):
    return stats.beta(k + 0.5, n - k + 0.5)

print("✅ Bibliotecas carregadas · semente", SEMENTE)
```

```python
# --- DADOS DA LITERATURA ---
# Martin & Kuiper (1999), Humor 12(4). Valores lidos na descrição da tabela, arredondados.

N_PESSOAS = 80
N_MULHERES, N_HOMENS = 50, 30
DIAS_DIARIO = 3
RISADAS_MULHERES, RISADAS_HOMENS = 18.5, 16.1      # média por dia
RISADAS_DIA = (N_MULHERES * RISADAS_MULHERES + N_HOMENS * RISADAS_HOMENS) / N_PESSOAS
MINIMO_DIA, MAXIMO_DIA = 0, 89                      # faixa observada de risadas num dia

P_SOZINHO = 0.12                                    # fatia das risadas a sós
ORIGEM = {'situações do momento': 0.56, 'TV, rádio e leitura': 0.18,
          'lembranças': 0.15, 'piadas': 0.11}

# Provine & Fischer (1989), lido em fonte secundária.
RAZAO_PROVINE = 30                                  # risadas por hora, acompanhado contra sozinho

PESSOA_DIAS = N_PESSOAS * DIAS_DIARIO
RISADAS_TOTAL = PESSOA_DIAS * RISADAS_DIA           # ordem de grandeza do número de risadas anotadas

print("=" * 70)
print("  DADOS · MARTIN & KUIPER (1999)")
print("=" * 70)
print(f"\n  {N_PESSOAS} pessoas ({N_MULHERES} mulheres e {N_HOMENS} homens), diário de {DIAS_DIARIO} dias")
print(f"  Risadas por dia: mulheres {vir(RISADAS_MULHERES)} · homens {vir(RISADAS_HOMENS)} · "
      f"média ponderada {vir(RISADAS_DIA)}")
print(f"  Faixa observada num dia: de {MINIMO_DIA} a {MAXIMO_DIA} risadas")
print(f"  Dias de diário: {PESSOA_DIAS} · risadas anotadas (ordem de grandeza): {mil(RISADAS_TOTAL)}")
print(f"\n  A sós: {vir(P_SOZINHO * 100, 0)}% das risadas · com outras pessoas: {vir((1 - P_SOZINHO) * 100, 0)}%")
print("  Origem da risada:")
for nome, p in ORIGEM.items():
    print(f"    {nome:<22} {vir(p * 100, 0)}%")
print(f"    soma: {vir(sum(ORIGEM.values()) * 100, 0)}%")
print("\n  Valores arredondados na fonte. O desvio padrão não foi lido.")
print("=" * 70)
```

```python
# --- O MODELO ---
rng = np.random.default_rng(SEMENTE)

# ── CAMADA 1 · de cada cem risadas ──
# O artigo publica o percentual. Dois jeitos de contar a incerteza:
#   a) cada risada como observação independente (ingênuo, intervalo estreito demais);
#   b) a pessoa como unidade (80 pessoas), que é o limite conservador.
k_risadas = P_SOZINHO * RISADAS_TOTAL
ic_risada = jeffreys(k_risadas, RISADAS_TOTAL).interval(0.95)
ic_pessoa = jeffreys(P_SOZINHO * N_PESSOAS, N_PESSOAS).interval(0.95)
print("=" * 70)
print("  CAMADA 1 · DE CADA CEM RISADAS")
print("=" * 70)
print(f"\n  A sós: {vir(P_SOZINHO * 100, 0)} de cada 100 · acompanhado: {vir((1 - P_SOZINHO) * 100, 0)}")
print(f"  Intervalo contando cada risada (~{mil(RISADAS_TOTAL)}): "
      f"[{vir(ic_risada[0] * 100)}; {vir(ic_risada[1] * 100)}]")
print(f"  Intervalo contando cada pessoa ({N_PESSOAS}): "
      f"[{vir(ic_pessoa[0] * 100)}; {vir(ic_pessoa[1] * 100)}]")
print("  O primeiro é estreito demais, porque as risadas de uma mesma pessoa se parecem.")
print("  O segundo é o limite conservador. O valor real fica entre os dois.")

# ── CAMADA 2 · risadas a sós por dia, se todos rissem no mesmo ritmo ──
LAMBDA_SOZINHO = RISADAS_DIA * P_SOZINHO
p_poisson = 1 - np.exp(-LAMBDA_SOZINHO)
print("\n" + "=" * 70)
print("  CAMADA 2 · RISADAS A SÓS POR DIA (todos no mesmo ritmo)")
print("=" * 70)
print(f"\n  {vir(RISADAS_DIA)} risadas por dia × {vir(P_SOZINHO * 100, 0)}% = "
      f"{vir(LAMBDA_SOZINHO, 2)} risadas a sós por dia, em média")
print(f"  P(pelo menos uma risada a sós no dia) = 1 − e^(−{vir(LAMBDA_SOZINHO, 2)}) = {vir(p_poisson * 100)}%")
print("  Essa conta supõe que todas as pessoas riem no mesmo ritmo. Não é o caso.")

# ── CAMADA 3 · as pessoas não são iguais ──
# Ritmo diário de cada pessoa-dia ~ Gama(forma k, média m). Contagem ~ Poisson(ritmo).
# O coeficiente de variação do ritmo é 1/√k. Para cada k, simulo 240 dias de diário
# e olho o maior e o menor número de risadas. O estudo observou 0 e 89.
def simula_diario(k, n_estudos=4000):
    ritmo = rng.gamma(k, RISADAS_DIA / k, size=(n_estudos, PESSOA_DIAS))
    contagem = rng.poisson(ritmo)
    return contagem.max(axis=1), contagem.min(axis=1)

GRADE_CV = np.round(np.arange(0.30, 1.21, 0.05), 2)
calib = {}
for cv in GRADE_CV:
    k = 1 / cv ** 2
    maximos, minimos = simula_diario(k)
    calib[cv] = (np.median(maximos), np.percentile(maximos, 10), np.percentile(maximos, 90),
                 (minimos == 0).mean())
medianas = np.array([calib[cv][0] for cv in GRADE_CV])
CV_CENTRAL = float(GRADE_CV[np.argmin(np.abs(medianas - MAXIMO_DIA))])
# faixa plausível: CVs em que 89 fica entre o percentil 10 e o 90 do máximo simulado
plaus = [cv for cv in GRADE_CV if calib[cv][1] <= MAXIMO_DIA <= calib[cv][2]]
CV_LO, CV_HI = float(min(plaus)), float(max(plaus))

def p_ao_menos_uma(cv, lam=None):
    lam = LAMBDA_SOZINHO if lam is None else lam
    if cv == 0:
        return 1 - np.exp(-lam)
    k = 1 / cv ** 2
    return 1 - (1 + lam / k) ** (-k)

P_CENTRAL = p_ao_menos_uma(CV_CENTRAL)
P_LO, P_HI = p_ao_menos_uma(CV_HI), p_ao_menos_uma(CV_LO)

print("\n" + "=" * 70)
print("  CAMADA 3 · AS PESSOAS NÃO SÃO IGUAIS")
print("=" * 70)
print(f"\n  Em {PESSOA_DIAS} dias de diário o estudo viu de {MINIMO_DIA} a {MAXIMO_DIA} risadas num dia.")
print("  Maior número de risadas num dia, simulado, conforme a variação entre pessoas:\n")
for cv in (0.30, 0.50, 0.70, CV_CENTRAL, 1.00, 1.20):
    med, p10, p90, zero = calib[round(cv, 2)]
    marca = "  ← mais perto de 89" if cv == CV_CENTRAL else ""
    print(f"    variação de {vir(cv * 100, 0):>3}%: máximo mediano {vir(med, 0):>3} "
          f"(entre {vir(p10, 0)} e {vir(p90, 0)}) · tem dia com zero em {vir(zero * 100, 0)}% das simulações{marca}")
print(f"\n  Variação calibrada: {vir(CV_CENTRAL * 100, 0)}% (plausível de {vir(CV_LO * 100, 0)}% a {vir(CV_HI * 100, 0)}%)")
print(f"\n  P(rir sozinho pelo menos uma vez no dia):")
print(f"    todos no mesmo ritmo:        {vir(p_poisson * 100)}%")
print(f"    com a variação calibrada:    {vir(P_CENTRAL * 100)}%  (de {vir(P_LO * 100)}% a {vir(P_HI * 100)}%)")
print(f"  Num dia qualquer, {vir((1 - P_CENTRAL) * 100, 0)} em cada 100 pessoas não riem sozinhas nenhuma vez.")
print("  Premissa nossa: a variação das risadas a sós é a mesma das risadas em geral.")

# ── CAMADA 4 · uma semana ──
k_c = 1 / CV_CENTRAL ** 2
# cada pessoa tem um ritmo próprio, que vale para os sete dias
ritmo_pessoa = rng.gamma(k_c, LAMBDA_SOZINHO / k_c, size=SORTEIOS)
semana = rng.poisson(ritmo_pessoa[:, None], size=(SORTEIOS, 7))
dias_rindo = (semana > 0).sum(axis=1)
p_semana_sem = (dias_rindo == 0).mean()
p_todo_dia = (dias_rindo == 7).mean()
print("\n" + "=" * 70)
print(f"  CAMADA 4 · UMA SEMANA ({mil(SORTEIOS)} pessoas simuladas)")
print("=" * 70)
print(f"\n  Dias da semana com pelo menos uma risada a sós: média {vir(dias_rindo.mean())} de 7")
print(f"  Riem sozinhas todos os sete dias: {vir(p_todo_dia * 100)}%")
print(f"  Passam a semana inteira sem rir sozinhas: {vir(p_semana_sem * 100)}%")
print(f"  Risadas a sós na semana: mediana {vir(np.median(semana.sum(axis=1)), 0)} · "
      f"95% entre {vir(np.percentile(semana.sum(axis=1), 2.5), 0)} e {vir(np.percentile(semana.sum(axis=1), 97.5), 0)}")
print("  Premissa: o ritmo de cada pessoa é o mesmo nos sete dias.")

# ── CAMADA 5 · e as 30 vezes? ──
# Se uma fração f do tempo acordado é passada a sós, a razão entre os ritmos por hora é
#   razão = [(1 − p) / (1 − f)] / [p / f],   com p = fatia das risadas a sós.
def razao(f, p=P_SOZINHO):
    return ((1 - p) / (1 - f)) / (p / f)

f_30 = RAZAO_PROVINE * P_SOZINHO / ((1 - P_SOZINHO) + RAZAO_PROVINE * P_SOZINHO)
print("\n" + "=" * 70)
print("  CAMADA 5 · OS 12% E AS 30 VEZES")
print("=" * 70)
print("\n  Razão entre o ritmo de riso acompanhado e a sós, conforme o tempo passado sozinho:")
for f in (0.20, 1 / 3, 0.50, 0.65, f_30):
    print(f"    {vir(f * 100, 0):>3}% do tempo acordado a sós: {vir(razao(f))} vezes")
print(f"\n  Para os 12% darem 30 vezes, a pessoa teria de passar {vir(f_30 * 100, 0)}% do tempo acordada sozinha.")
print("  Com um terço do tempo a sós, a razão é de cerca de 4 vezes.")
print("  Os dois estudos não medem a mesma coisa, ou as amostras vivem de modos diferentes.")
print("  O tempo a sós das 80 pessoas não foi lido. A diferença fica declarada, sem resolução.")
print("=" * 70)
```

```python
# --- VISUALIZAÇÃO ---
RODAPE = ('Fonte: Martin & Kuiper (1999), Humor 12(4) · N = 80, diário de três dias · '
          'cálculo próprio | #365Probabilidades')

# ══ GRÁFICO 1 · de cada cem risadas ══
fig1, (ax1a, ax1b) = plt.subplots(1, 2, figsize=(13, 7), gridspec_kw={'width_ratios': [1, 1.3]})
b = ax1a.bar(['com outras\npessoas', 'a sós'], [(1 - P_SOZINHO) * 100, P_SOZINHO * 100],
             color=[CINZA, VERMELHO], alpha=0.85, width=0.55)
for barra, v in zip(b, [(1 - P_SOZINHO) * 100, P_SOZINHO * 100]):
    ax1a.text(barra.get_x() + barra.get_width() / 2, v + 2, f'{vir(v, 0)}', ha='center',
              fontsize=20, fontweight='bold')
ax1a.set_ylim(0, 105)
ax1a.set_ylabel('De cada 100 risadas', fontsize=12)
ax1a.set_title('Com quem', fontsize=13)
nomes = list(ORIGEM.keys())[::-1]
vals = [ORIGEM[n] * 100 for n in nomes]
hb = ax1b.barh(nomes, vals, color=DOURADO, alpha=0.9)
for barra, v in zip(hb, vals):
    ax1b.text(v + 1, barra.get_y() + barra.get_height() / 2, f'{vir(v, 0)}', va='center',
              fontsize=14, fontweight='bold')
ax1b.set_xlim(0, 70)
ax1b.set_xlabel('De cada 100 risadas', fontsize=12)
ax1b.set_title('De onde vem', fontsize=13)
fig1.suptitle('De cada cem risadas, doze acontecem a sós', fontsize=15)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 0.96))
plt.savefig('dia-124-grafico-01-de-cada-cem.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ══ GRÁFICO 2 · rir sozinho hoje ══
fig2, ax2 = plt.subplots(figsize=(12, 8))
cvs = np.linspace(0, 1.3, 200)
ax2.plot(cvs * 100, [p_ao_menos_uma(c) * 100 for c in cvs], color=VERMELHO, linewidth=3)
ax2.axvspan(CV_LO * 100, CV_HI * 100, color=DOURADO, alpha=0.18, label='variação plausível pela faixa de 0 a 89')
ax2.scatter([0, CV_CENTRAL * 100], [p_poisson * 100, P_CENTRAL * 100], color=VERDE, s=110, zorder=5)
ax2.annotate(f'todos no mesmo ritmo\n{vir(p_poisson * 100)}%', (0, p_poisson * 100), xytext=(8, 93),
             fontsize=12, arrowprops=dict(arrowstyle='-', color=CINZA))
ax2.annotate(f'variação calibrada\n{vir(P_CENTRAL * 100)}%', (CV_CENTRAL * 100, P_CENTRAL * 100),
             xytext=(CV_CENTRAL * 100 + 8, P_CENTRAL * 100 + 8), fontsize=12,
             arrowprops=dict(arrowstyle='-', color=CINZA))
ax2.set_ylim(50, 100)
ax2.set_xlabel('Variação do ritmo de riso entre pessoas (coeficiente de variação, %)', fontsize=12)
ax2.set_ylabel('Chance de rir sozinho pelo menos uma vez no dia (%)', fontsize=12)
ax2.set_title('A chance de rir sozinho hoje\n'
              f'{vir(LAMBDA_SOZINHO, 1)} risadas a sós por dia, em média', fontsize=14, pad=16)
ax2.legend(frameon=False, fontsize=11, loc='lower left')
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-124-grafico-02-rir-sozinho-hoje.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ══ GRÁFICO 3 · a calibração pela faixa ══
fig3, ax3 = plt.subplots(figsize=(12, 8))
med = np.array([calib[cv][0] for cv in GRADE_CV])
p10 = np.array([calib[cv][1] for cv in GRADE_CV])
p90 = np.array([calib[cv][2] for cv in GRADE_CV])
ax3.fill_between(GRADE_CV * 100, p10, p90, color=VERDE, alpha=0.2, label='80% das simulações')
ax3.plot(GRADE_CV * 100, med, color=VERDE, linewidth=3, label='maior dia simulado (mediana)')
ax3.axhline(MAXIMO_DIA, color=VERMELHO, linewidth=2.5)
ax3.text(31, MAXIMO_DIA + 3, f'maior dia observado no estudo: {MAXIMO_DIA} risadas', color=VERMELHO, fontsize=12)
ax3.axvline(CV_CENTRAL * 100, color=DOURADO, linestyle='--', linewidth=2)
ax3.set_xlabel('Variação do ritmo de riso entre pessoas (coeficiente de variação, %)', fontsize=12)
ax3.set_ylabel(f'Maior número de risadas num dia, em {PESSOA_DIAS} dias de diário', fontsize=12)
ax3.set_title('Quanto as pessoas diferem\n'
              'a variação que reproduz um dia de 89 risadas', fontsize=14, pad=16)
ax3.legend(frameon=False, fontsize=11, loc='upper left')
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-124-grafico-03-a-calibracao.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")

# ══ GRÁFICO 4 · os 12% e as 30 vezes ══
fig4, ax4 = plt.subplots(figsize=(12, 8))
fs = np.linspace(0.05, 0.9, 300)
ax4.plot(fs * 100, razao(fs), color=VERMELHO, linewidth=3)
ax4.axhline(RAZAO_PROVINE, color=CINZA, linewidth=2, linestyle='--')
ax4.text(6, RAZAO_PROVINE + 1.2, 'as "30 vezes" da frase famosa', color=CINZA, fontsize=12)
ax4.scatter([100 / 3, f_30 * 100], [razao(1 / 3), RAZAO_PROVINE], color=DOURADO, s=110, zorder=5)
ax4.annotate(f'um terço do tempo a sós\n{vir(razao(1 / 3))} vezes', (100 / 3, razao(1 / 3)),
             xytext=(36, 12), fontsize=12, arrowprops=dict(arrowstyle='-', color=CINZA))
ax4.annotate(f'{vir(f_30 * 100, 0)}% do tempo a sós', (f_30 * 100, RAZAO_PROVINE),
             xytext=(f_30 * 100 - 26, 38), fontsize=12, arrowprops=dict(arrowstyle='-', color=CINZA))
ax4.set_ylim(0, 60)
ax4.set_xlabel('Parte do tempo acordado passada a sós (%)', fontsize=12)
ax4.set_ylabel('Ritmo de riso acompanhado ÷ ritmo a sós', fontsize=12)
ax4.set_title('Os 12% e as 30 vezes só se encontram se a pessoa vive quase sempre sozinha\n'
              'razão = [(1 − p) / (1 − f)] ÷ [p / f], com p = 12%', fontsize=14, pad=16)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-124-grafico-04-trinta-vezes.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 4 salvo!")
```

### 💡 O Insight

**A chance de uma pessoa rir sozinha pelo menos uma vez num dia é de 72%.** Com a
incerteza da calibração, entre 66% e 77%.

Em cada cem risadas, só doze acontecem a sós. Parece pouco. Mas uma pessoa ri 17 ou 18
vezes por dia, e doze em cem disso dá cerca de duas risadas a sós por dia.

Duas por dia é uma média. Se todas as pessoas rissem no mesmo ritmo, a chance de rir
sozinho num dia seria de 88%. Elas não riem. No diário há dias com zero risada e dias
com 89. Quando a conta leva essa diferença em consideração, a chance cai para 72%. De
cada cem pessoas, 28 passam o dia sem rir sozinhas.

Numa semana, o riso a sós aparece em 5 dos 7 dias, em média. Em 35% das semanas ele
aparece nos sete. Uma semana inteira sem rir sozinho acontece em 3% dos casos.

Sobra uma conta que não fecha. A frase famosa diz que se ri 30 vezes mais acompanhado.
Para os 12% e as 30 vezes serem o mesmo fato, a pessoa teria de passar 80% do tempo
acordada sozinha. Com um terço do tempo a sós, os 12% correspondem a 3,7 vezes. Os dois
números vêm de estudos diferentes, com métodos diferentes, e este notebook não decide
entre eles.

### ⚠️ Limitações do Modelo

- **Os números da fonte foram lidos na descrição da tabela na página do artigo, e estão
  arredondados.** A tabela completa e o desvio padrão não foram lidos.
- **A variação entre pessoas é premissa nossa.** Ela foi calibrada pelo maior dia
  registrado (89 risadas). Um único valor extremo é uma base frágil, e por isso a
  resposta sai com uma faixa (66% a 77%).
- O modelo supõe que o riso a sós varia entre pessoas do mesmo jeito que o riso total. O
  artigo, na parte lida, não informa isso.
- O modelo supõe que o ritmo de cada pessoa é o mesmo todos os dias. Dia de trabalho e
  fim de semana ficam iguais.
- A média de 17,6 risadas por dia é uma média ponderada nossa, das médias de mulheres e
  homens.
- O diário é um registro feito pela própria pessoa. Risadas curtas podem ficar sem
  anotação, e as que acontecem a sós podem ser mais fáceis de esquecer.
- São 80 adultos canadenses, em diário de três dias, num estudo de 1999. O celular ainda
  não era fonte de riso a sós.
- A frase das 30 vezes foi lida em fonte secundária. Amostra e método do estudo de
  Provine e Fischer não foram conferidos.
- A parte do tempo passada a sós, usada na camada 5, é uma premissa de ilustração. Ela
