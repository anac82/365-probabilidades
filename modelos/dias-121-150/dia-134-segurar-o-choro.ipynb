### 📚 O Conceito: supressão expressiva

A psicologia tem um nome para "quase nada". Chama-se supressão expressiva: sentir e não
deixar aparecer. É uma das duas estratégias que James Gross descreveu para lidar com a
emoção. A outra é a reavaliação, que muda o jeito de olhar a situação antes de a emoção
crescer. A supressão age depois. A emoção já está lá; o que se segura é a cara, a voz, o
choro.

A pergunta científica não é se segurar é certo ou errado. É se quem segura mais, como
hábito, vive pior ou morre antes do que quem segura menos. Isso foi medido em dezenas de
estudos com uma escala de poucas perguntas, do tipo "eu guardo minhas emoções para mim",
e cruzado com depressão, ansiedade, satisfação com a vida e, numa coorte, com a morte.

A correlação é pequena. É o que a psicologia costuma encontrar. O que importa é o sinal,
a consistência e o que o número vira quando é traduzido em pessoas.

### 🧮 O Modelo

Quatro camadas. A fonte dá as correlações e a razão de risco. As contas são nossas.

1. **A moeda.** A correlação entre supressão e sofrimento, traduzida em chance: de duas
   pessoas, uma que segura mais e outra que segura menos, qual a probabilidade de a que
   segura mais estar na metade pior em depressão, ansiedade e afeto negativo.
2. **As duas estratégias.** Supressão contra reavaliação, nos indicadores positivos e
   negativos, lado a lado.
3. **A conta da vida.** A razão de risco de mortalidade em 12 anos, traduzida em
   probabilidade, com o intervalo inteiro, porque ele encosta no zero.
4. **A sala de 30.** Quinze pessoas que seguram e quinze que não seguram, 200 mil salas
   sorteadas: quantas de cada lado estão na metade pior.

**Fonte central:**
- Hu, T., Zhang, D., Wang, J., Mistry, R., Ran, G. & Wang, X. (2014). "Relation between
  emotion regulation and mental health: a meta-analysis review". *Psychological Reports*
  114(2), 341-362. DOI 10.2466/03.20.PR0.114k22w4. **48 estudos, 51 amostras
  independentes, 157 tamanhos de efeito, N = 21.150.** Supressão expressiva: **r = 0,15**
  com indicadores negativos (depressão, ansiedade, afeto negativo) e **r = −0,12** com
  indicadores positivos (satisfação com a vida, afeto positivo). Em amostras de cultura
  ocidental, r = 0,19 com os negativos; em amostras de cultura oriental, r = 0,06.
  Reavaliação cognitiva: r = 0,26 com positivos e r = −0,20 com negativos. **Lido no
  resumo; os intervalos de confiança de cada r não foram lidos.**

**Fonte de apoio, declarada como frágil:**
- Chapman, B. P., Fiscella, K., Kawachi, I., Duberstein, P. & Muennig, P. (2013).
  "Emotion suppression and mortality risk over a 12-year follow-up". *Journal of
  Psychosomatic Research* 75(4), 381-385. DOI 10.1016/j.jpsychores.2013.07.014. Coorte
  General Social Survey ligada ao National Death Index, **N = 729**, escala de supressão
  aplicada em 1996, 12 anos de seguimento, **111 mortes**. Supressão no percentil 75
  contra o 25: mortalidade geral **HR 1,35 [1,00; 1,82]**, p = 0,049; câncer HR 1,70
  [1,01; 2,88]; cardiovascular HR 1,47 [0,87; 2,47], não significativo. Ajuste por idade,
  sexo, escolaridade e raça. Amostra pequena, intervalo que encosta no 1, uma coorte só.

**Nota metodológica.** O fator ×0,80 do projeto não se aplica. As medidas são
correlações de meta-análise e razão de risco de coorte, não proporções de survey. A
tradução de r em "chance de estar na metade pior" usa a conversão de Rosenthal e Rubin
(BESD), que supõe divisão ao meio nos dois lados. Cordas declaradas: #010 (crescimento
depois da perda, Tedeschi) e #018 (humor e mortalidade, outra coorte, outra exposição).

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

SEMENTE = 42
SORTEIOS = 200_000
rng = np.random.default_rng(SEMENTE)

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

VERMELHO, DOURADO, VERDE, CINZA = '#c0392b', '#c8a84b', '#1a5f5a', '#6b6a64'
RODAPE = ('Fonte: Hu et al. (2014), Psychological Reports 114(2), 48 estudos, N = 21.150 · Chapman et al. (2013), '
          'J Psychosom Res 75(4), N = 729 · cálculo próprio | #365Probabilidades')

def mil(n):
    return f"{int(round(n)):,}".replace(",", ".")

def vir(x, casas=1):
    return f"{x:.{casas}f}".replace(".", ",")

def besd(r):
    # Rosenthal & Rubin: taxa de "sucesso" nos dois grupos quando a correlação é r
    return 0.5 + r / 2, 0.5 - r / 2

print("✅ Bibliotecas carregadas · semente", SEMENTE)
```

```python
# --- DADOS DA LITERATURA ---

# Hu et al. 2014 · meta-análise
K_ESTUDOS, K_AMOSTRAS, K_EFEITOS, N_META = 48, 51, 157, 21_150
R_SUP_NEG = 0.15        # supressão × depressão, ansiedade, afeto negativo
R_SUP_POS = -0.12       # supressão × satisfação com a vida, afeto positivo
R_SUP_NEG_OCID = 0.19   # amostras de cultura ocidental
R_SUP_NEG_ORIENT = 0.06 # amostras de cultura oriental
R_REAV_POS = 0.26       # reavaliação × positivos
R_REAV_NEG = -0.20      # reavaliação × negativos

# Chapman et al. 2013 · coorte, 12 anos
N_COORTE, MORTES, ANOS = 729, 111, 12
HR, HR_LO, HR_HI = 1.35, 1.00, 1.82          # mortalidade geral, percentil 75 contra 25
HR_CANCER, HR_CA_LO, HR_CA_HI = 1.70, 1.01, 2.88
HR_CARDIO, HR_CV_LO, HR_CV_HI = 1.47, 0.87, 2.47

print("=" * 70)
print("  DADOS · SEGURAR O QUE SE SENTE")
print("=" * 70)
print(f"\n  Hu et al. 2014: {K_ESTUDOS} estudos, {K_AMOSTRAS} amostras, {K_EFEITOS} efeitos, N = {mil(N_META)}")
print(f"    supressão × sofrimento (depressão, ansiedade, afeto negativo):  r = {vir(R_SUP_NEG, 2)}")
print(f"    supressão × bem-estar (satisfação, afeto positivo):             r = {vir(R_SUP_POS, 2)}")
print(f"    ocidente r = {vir(R_SUP_NEG_OCID, 2)} · oriente r = {vir(R_SUP_NEG_ORIENT, 2)}")
print(f"    reavaliação × bem-estar r = {vir(R_REAV_POS, 2)} · reavaliação × sofrimento r = {vir(R_REAV_NEG, 2)}")
print(f"\n  Chapman et al. 2013: N = {N_COORTE}, {MORTES} mortes em {ANOS} anos")
print(f"    mortalidade geral HR {vir(HR, 2)} [{vir(HR_LO, 2)}; {vir(HR_HI, 2)}] · câncer HR {vir(HR_CANCER, 2)} [{vir(HR_CA_LO, 2)}; {vir(HR_CA_HI, 2)}]")
print(f"    cardiovascular HR {vir(HR_CARDIO, 2)} [{vir(HR_CV_LO, 2)}; {vir(HR_CV_HI, 2)}], não significativo")
print("=" * 70)
```

```python
# --- O MODELO ---

# CAMADA 1 · A moeda
p_sup, p_nao = besd(R_SUP_NEG)
p_sup_oc, p_nao_oc = besd(R_SUP_NEG_OCID)
p_sup_or, p_nao_or = besd(R_SUP_NEG_ORIENT)
# intervalo aproximado para r pela transformação de Fisher com o N total (piso otimista: ignora a dependência entre efeitos)
z = np.arctanh(R_SUP_NEG); se = 1 / np.sqrt(N_META - 3)
r_lo, r_hi = np.tanh(z - 1.96 * se), np.tanh(z + 1.96 * se)
print("=" * 70)
print("  CAMADA 1 · A MOEDA")
print("=" * 70)
print(f"\n  r = {vir(R_SUP_NEG, 2)} vira, na conversão BESD:")
print(f"    quem segura mais: {vir(p_sup * 100)}% de chance de estar na metade pior em sofrimento")
print(f"    quem segura menos: {vir(p_nao * 100)}%")
print(f"    diferença: {vir((p_sup - p_nao) * 100)} pontos. Uma moeda honesta dá 50% para os dois.")
print(f"  No Ocidente (r = {vir(R_SUP_NEG_OCID, 2)}): {vir(p_sup_oc * 100)}% contra {vir(p_nao_oc * 100)}%")
print(f"  No Oriente (r = {vir(R_SUP_NEG_ORIENT, 2)}): {vir(p_sup_or * 100)}% contra {vir(p_nao_or * 100)}%")
print(f"  Intervalo aproximado de r com o N total: [{vir(r_lo, 3)}; {vir(r_hi, 3)}] (piso otimista, o IC publicado não foi lido)")
print(f"  Variância explicada: {vir(R_SUP_NEG ** 2 * 100)}%. É pequeno, e é consistente em {K_AMOSTRAS} amostras.")

# CAMADA 2 · As duas estratégias
print("\n" + "=" * 70)
print("  CAMADA 2 · AS DUAS ESTRATÉGIAS")
print("=" * 70)
print(f"\n  {'':<22} {'bem-estar':>10} {'sofrimento':>11}")
print(f"  {'suprimir (segurar)':<22} {vir(R_SUP_POS, 2):>10} {vir(R_SUP_NEG, 2):>11}")
print(f"  {'reavaliar (reler)':<22} {vir(R_REAV_POS, 2):>10} {vir(R_REAV_NEG, 2):>11}")
pb_sup, _ = besd(-R_SUP_POS); pb_reav, _ = besd(R_REAV_POS)
print(f"\n  Chance de estar na metade melhor em bem-estar: quem reavalia {vir(pb_reav * 100)}%, quem suprime {vir((1 - pb_sup) * 100)}%.")
print("  As duas estratégias lidam com a mesma emoção. Uma antes de ela crescer, a outra depois. Os sinais são opostos.")

# CAMADA 3 · A conta da vida
p0 = MORTES / N_COORTE                       # mortalidade observada na coorte em 12 anos
def risco(hr, p_base):
    # P(morrer em 12 anos) para quem suprime mais, dado o HR e um risco de base de quem suprime menos
    return 1 - (1 - p_base) ** hr
p_base = 1 - (1 - p0) ** (1 / ((1 + HR) / 2))   # risco de quem está no percentil 25, tal que a média bata com a coorte
print("\n" + "=" * 70)
print(f"  CAMADA 3 · A CONTA DA VIDA (Chapman 2013, {ANOS} anos, apoio frágil)")
print("=" * 70)
print(f"\n  Mortalidade da coorte em {ANOS} anos: {vir(p0 * 100)}% ({MORTES} de {N_COORTE})")
print(f"  Quem segura menos (percentil 25): {vir(p_base * 100)}%")
print(f"  Quem segura mais (percentil 75): {vir(risco(HR, p_base) * 100)}%  [{vir(risco(HR_LO, p_base) * 100)}; {vir(risco(HR_HI, p_base) * 100)}]")
print(f"  Excesso: {vir((risco(HR, p_base) - p_base) * 100)} pontos em {ANOS} anos, com o limite inferior em zero")
print(f"  Chance de o efeito ser maior que zero, pela estatística do estudo (p = 0,049 bicaudal): cerca de 97,5%")
print(f"  Câncer: HR {vir(HR_CANCER, 2)} [{vir(HR_CA_LO, 2)}; {vir(HR_CA_HI, 2)}] · cardiovascular HR {vir(HR_CARDIO, 2)}, não significativo")
print("  Uma coorte de 729 pessoas com 111 mortes não fecha a questão. Fecha a direção.")

# CAMADA 4 · A sala de 30
SALA = 30; META = SALA // 2
seguram = rng.binomial(META, p_sup, SORTEIOS)
nao_seguram = rng.binomial(META, p_nao, SORTEIOS)
dif = seguram - nao_seguram
print("\n" + "=" * 70)
print(f"  CAMADA 4 · A SALA DE {SALA} ({mil(SORTEIOS)} salas sorteadas)")
print("=" * 70)
print(f"\n  {META} que seguram e {META} que não seguram. Na metade pior em sofrimento:")
print(f"    dos que seguram: média {vir(seguram.mean())} · dos que não seguram: média {vir(nao_seguram.mean())}")
print(f"    diferença média: {vir(dif.mean())} pessoas · P(mais gente sofrendo entre os que seguram): {vir(np.mean(dif > 0) * 100, 0)}%")
print(f"    P(diferença de 3 pessoas ou mais): {vir(np.mean(dif >= 3) * 100, 0)}% · P(empate ou invertido): {vir(np.mean(dif <= 0) * 100, 0)}%")
print("  O efeito é real e é pequeno. Numa sala de 30, são duas pessoas. Numa vida, é todo dia.")
print("=" * 70)
```

```python
# --- VISUALIZAÇÃO ---

# GRÁFICO 1 · A moeda
fig1, ax1 = plt.subplots(figsize=(12, 8))
grupos = ['todas as amostras\nr = 0,15', 'cultura ocidental\nr = 0,19', 'cultura oriental\nr = 0,06']
seg = [p_sup, p_sup_oc, p_sup_or]; nao = [p_nao, p_nao_oc, p_nao_or]
x = np.arange(3)
ax1.bar(x - 0.2, np.array(seg) * 100, width=0.38, color=VERMELHO, alpha=0.9, label='quem segura mais')
ax1.bar(x + 0.2, np.array(nao) * 100, width=0.38, color=VERDE, alpha=0.9, label='quem segura menos')
ax1.axhline(50, color='#333', lw=1.2, ls='--')
ax1.text(2.45, 50.8, 'moeda honesta', ha='right', fontsize=11, color='#333')
for i in range(3):
    ax1.text(i - 0.2, seg[i] * 100 + 1, f"{vir(seg[i] * 100)}%", ha='center', fontsize=14, fontweight='bold', color=VERMELHO)
    ax1.text(i + 0.2, nao[i] * 100 + 1, f"{vir(nao[i] * 100)}%", ha='center', fontsize=14, fontweight='bold', color=VERDE)
ax1.set_xticks(x); ax1.set_xticklabels(grupos, fontsize=12)
ax1.set_ylim(0, 75)
ax1.set_ylabel('Chance de estar na metade pior em sofrimento (%)', fontsize=12)
ax1.set_title('Segurar o que se sente, traduzido em moeda\n'
              'de duas pessoas, a que segura mais tem 57,5% de chance de estar na metade pior; a outra, 42,5%',
              fontsize=14, pad=16)
ax1.legend(frameon=False, fontsize=12)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-134-grafico-01-a-moeda.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# GRÁFICO 2 · As duas estratégias
fig2, ax2 = plt.subplots(figsize=(12, 8))
rot = ['bem-estar\n(satisfação, afeto positivo)', 'sofrimento\n(depressão, ansiedade, afeto negativo)']
x = np.arange(2)
ax2.bar(x - 0.2, [R_SUP_POS, R_SUP_NEG], width=0.38, color=VERMELHO, alpha=0.9, label='suprimir: segurar o que sente')
ax2.bar(x + 0.2, [R_REAV_POS, R_REAV_NEG], width=0.38, color=VERDE, alpha=0.9, label='reavaliar: mudar o jeito de olhar')
ax2.axhline(0, color='#333', lw=1.2)
for i, (a, b) in enumerate(zip([R_SUP_POS, R_SUP_NEG], [R_REAV_POS, R_REAV_NEG])):
    ax2.text(i - 0.2, a + (0.012 if a > 0 else -0.03), f"r = {vir(a, 2)}", ha='center', fontsize=13, fontweight='bold', color=VERMELHO)
    ax2.text(i + 0.2, b + (0.012 if b > 0 else -0.03), f"r = {vir(b, 2)}", ha='center', fontsize=13, fontweight='bold', color=VERDE)
ax2.set_xticks(x); ax2.set_xticklabels(rot, fontsize=12)
ax2.set_ylim(-0.3, 0.35)
ax2.set_ylabel('Correlação com a estratégia', fontsize=12)
ax2.set_title('Duas estratégias para a mesma emoção, sinais opostos\n48 estudos, 51 amostras, 21.150 pessoas', fontsize=14, pad=16)
ax2.legend(frameon=False, fontsize=12, loc='upper left')
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-134-grafico-02-duas-estrategias.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# GRÁFICO 3 · A conta da vida
fig3, ax3 = plt.subplots(figsize=(12, 8))
hrs = np.linspace(0.8, 2.0, 300)
ax3.plot(hrs, risco(hrs, p_base) * 100, color=CINZA, lw=2)
ax3.axvspan(HR_LO, HR_HI, color=DOURADO, alpha=0.18, lw=0)
ax3.axvline(HR, color=VERMELHO, lw=2.5)
ax3.axvline(1.0, color='#333', lw=1.2, ls='--')
ax3.scatter([HR], [risco(HR, p_base) * 100], s=160, color=VERMELHO, zorder=5, edgecolor='#333')
ax3.text(HR + 0.02, risco(HR, p_base) * 100 - 1.6, f'HR {vir(HR, 2)}\n{vir(risco(HR, p_base) * 100)}% em 12 anos', fontsize=12, color=VERMELHO, fontweight='bold')
ax3.text(1.0 - 0.02, risco(1.0, p_base) * 100 + 0.6, f'sem efeito\n{vir(p_base * 100)}%', ha='right', fontsize=11, color='#333')
ax3.text((HR_LO + HR_HI) / 2, risco(HR_HI, p_base) * 100 + 0.8, f'IC 95% [{vir(HR_LO, 2)}; {vir(HR_HI, 2)}]', ha='center', fontsize=11, color='#6b5a1e', fontweight='bold')
ax3.set_xlabel('Razão de risco de quem segura mais (percentil 75 contra 25)', fontsize=12)
ax3.set_ylabel('Chance de morrer em 12 anos (%)', fontsize=12)
ax3.set_title('A conta da vida, com o intervalo inteiro\n729 pessoas, 111 mortes; o limite inferior encosta no zero', fontsize=14, pad=16)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-134-grafico-03-a-conta-da-vida.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")

# GRÁFICO 4 · A sala de 30
fig4, ax4 = plt.subplots(figsize=(12, 8))
ks = np.arange(0, META + 1)
f_seg = np.array([(seguram == k).mean() for k in ks]) * 100
f_nao = np.array([(nao_seguram == k).mean() for k in ks]) * 100
ax4.bar(ks - 0.2, f_seg, width=0.38, color=VERMELHO, alpha=0.9, label=f'dos {META} que seguram · média {vir(seguram.mean())}')
ax4.bar(ks + 0.2, f_nao, width=0.38, color=VERDE, alpha=0.9, label=f'dos {META} que não seguram · média {vir(nao_seguram.mean())}')
ax4.set_xticks(ks)
ax4.set_xlabel('Pessoas na metade pior em sofrimento', fontsize=12)
ax4.set_ylabel('Salas simuladas (%)', fontsize=12)
ax4.set_title(f'A sala de {SALA}: quinze seguram, quinze não\n'
              f'em {vir(np.mean(dif > 0) * 100, 0)}% das salas há mais gente sofrendo do lado que segura; em {vir(np.mean(dif <= 0) * 100, 0)}%, empate ou o contrário',
              fontsize=14, pad=16)
ax4.legend(frameon=False, fontsize=12)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-134-grafico-04-a-sala.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 4 salvo!")
```

### 💡 O Insight

Eu não tinha um nome para o que escrevi naquele card. A ciência tem. Chama-se supressão,
e ela mediu quanto custa. **Quem costuma guardar o que sente tem 57,5% de chance de estar
na metade que sofre mais. Quem deixa sair, 42,5%.** Não é uma tragédia. É uma moeda
viciada, jogada todo dia, por anos, sem que a gente perceba que está jogando.

Eu joguei essa moeda por muito tempo achando que estava ganhando. O "quase nada" não era
ausência de dor. Era a dor guardada num lugar que o card não alcançava.

Perder meu pai foi o que rompeu a fortaleza. Não foi escolha, foi ruptura. Mas o que saiu
pela rachadura não foi só tristeza. Hoje eu me sinto muito mais conectada com o que
acontece ao meu redor, com as pessoas, com uma música, com uma lembrança. É por isso que
eu choro com mais facilidade. Não porque sinto mais dor. Porque sinto mais.

### ⚠️ Limitações do Modelo

- Correlação não é causa. Quem sofre mais pode passar a segurar mais, e não o contrário.
  A meta-análise é transversal na maior parte dos estudos.
- Os intervalos de confiança de cada r não foram lidos; o intervalo aproximado do
  notebook usa o N total e é otimista.
- A conversão em "metade pior" (BESD) é uma tradução didática de r, não uma medida
  direta; supõe divisão ao meio nos dois lados.
- A coorte de mortalidade tem 729 pessoas, 111 mortes, uma medida de supressão feita uma
  vez em 1996, e p = 0,049. Entra como apoio e direção, não como número de capa.
- O efeito é menor em amostras de cultura oriental (r = 0,06). Segurar o que se sente não
  custa o mesmo em todo lugar.
