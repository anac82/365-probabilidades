### 📚 O Conceito: As Quatro Trajetórias

Quando a ciência para de perguntar "você superou?" e passa a
medir a curva de cada pessoa no tempo, quatro desenhos aparecem,
nos mesmos formatos, estudo após estudo:

**Resiliente (65,7%):** sintomas baixos e estáveis desde o início.
Atenção ao que isso NÃO é: não é ausência de dor. É funcionar
enquanto dói. A pessoa sente, e a vida não desaba.

**Recuperação (20,8%):** afunda, e volta. O desenho que o senso
comum acha que é o único.

**Crônica (10,6%):** afunda, e fica. É para essa minoria que os
serviços de saúde mental deveriam mirar seus radares.

**Início tardio (8,9%):** parece bem, e piora depois. O desenho
mais traiçoeiro, e o argumento definitivo contra o "já passou,
já era".

E a réplica que atravessou a pandemia: quando o mundo inteiro
virou o estressor ao mesmo tempo, 28 estudos independentes
acharam a trajetória resiliente em **66,0%**. O número nem tremeu.

No Dia #010, o crescimento depois da perda. Hoje, o complemento
com fontes distintas: antes de crescer, a maioria simplesmente
não cai.

### 🧮 O Modelo

**Fontes:**
- Galatzer-Levy, I. R., Huang, S. H. & Bonanno, G. A. (2018) -
  Trajectories of resilience and dysfunction following potential
  trauma: A review and statistical evaluation - Clinical
  Psychology Review, 63, 41-55 - **54 estudos, 67 amostras**
  (2004-2016), todos longitudinais, após eventos potencialmente
  traumáticos. Resiliente **65,7% [61,6, 69,8]** (presente em 63
  das 67 amostras); recuperação 20,8% [16,2, 25,8]; crônica
  10,6% [8,6, 12,7]; início tardio 8,9% [5,3, 13,3].
  Heterogeneidade atribuída a diferenças reais de população.
  **Fonte central.**
- Schäfer, S. K., Kunzler, A. M., Kalisch, R., Tüscher, O. &
  Lieb, K. (2022) - Trajectories of resilience and mental
  distress to global major disruptions - Trends in Cognitive
  Sciences, 26(12), 1171-1189, acesso aberto - **28 estudos
  pandêmicos**: resiliente **66,0%** (variação entre estudos:
  29,0 a 88,9%); recuperação menor que fora da pandemia (13,0%);
  idosos mais resilientes, jovens mais afetados. Apoio.

**Fator ×0,80: não se aplica.** As proporções vêm de
classificação por trajetórias de medidas repetidas em desenhos
longitudinais, não de pergunta única de survey.

**Honestidades estruturais, declaradas de saída:** as proporções
são médias entre estudos e somam 106%, não 100 (cada estudo tem
seu conjunto; a conta abaixo normaliza e declara). Em crianças
(6 estudos), a resiliência cai para 52%. E existe debate
metodológico legítimo (Infurna & Luthar, 2016): com desfechos
múltiplos e especificações mais exigentes, as taxas caem. O
número exato depende das escolhas de modelagem; o que é robusto
entre métodos é a resiliência como trajetória modal.

**As contas nossas:** (1) posterior da proporção resiliente
(aproximação normal sobre o IC publicado); (2) **posterior
Dirichlet sobre as quatro trajetórias juntas**, com a
concentração calibrada para que a marginal da resiliente
reproduza o IC publicado, propagada para a sala de 30 com
intervalos de credibilidade e um intervalo preditivo (semente 42).


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Galatzer-Levy, Huang & Bonanno (2018), Clin. Psych. Review 63

n_estudos, n_amostras = 54, 67

trajetorias = ['Resiliente', 'Recuperação', 'Crônica', 'Início tardio']
props = np.array([65.7, 20.8, 10.6, 8.9])          # % médias entre estudos
ics = [(61.6, 69.8), (16.2, 25.8), (8.6, 12.7), (5.3, 13.3)]
presenca = [63, 49, 47, 22]                         # em quantas das 67 amostras

soma = props.sum()                                   # 106%: médias, não partição
props_norm = props / soma * 100                      # normalizadas, declarado

# Nuances declaradas
resiliencia_criancas = 52.0        # 6 estudos com crianças
# Debate Infurna & Luthar (2016): taxas caem com especificações duras

# Schäfer et al. (2022), Trends in Cognitive Sciences: pandemia
pandemia_resiliente = 66.0         # 28 estudos; variação 29,0 a 88,9
pandemia_recuperacao = 13.0

print("=" * 65)
print("  DADOS DA LITERATURA - AS QUATRO TRAJETÓRIAS")
print("=" * 65)
print(f"\n  Galatzer-Levy et al. 2018: {n_estudos} estudos, {n_amostras} amostras")
for tr, p, ic, pres in zip(trajetorias, props, ics, presenca):
    print(f"  → {tr:<14} {p:>5.1f}%  [{ic[0]}, {ic[1]}]   em {pres}/67 amostras")
print(f"\n  Soma: {soma:.0f}% (médias entre estudos, não partição; a conta normaliza)")
print(f"\n  Nuances declaradas:")
print(f"  → Crianças (6 estudos): resiliente {resiliencia_criancas:.0f}%")
print(f"  → Infurna & Luthar 2016: com critérios mais duros, as taxas caem")
print(f"\n  Réplica pandêmica (Schäfer et al. 2022, 28 estudos):")
print(f"  → Resiliente: {pandemia_resiliente:.1f}%  ·  recuperação menor: {pandemia_recuperacao:.1f}%")
print(f"\n  Fator ×0,80: NÃO aplicado (trajetórias longitudinais)")
print("=" * 65)
```

```python
# --- O MODELO ---
# Conta 1: posterior da proporção resiliente
# Conta 2: posterior DIRICHLET das quatro trajetórias + a sala de 30

# ═══ CONTA 1 - a posterior da resiliente ═══
p_hat = 65.7
se = (69.8 - 61.6) / (2 * 1.96)   # do IC publicado

p_acima_60   = 1 - stats.norm.cdf((60.0  - p_hat) / se)
p_acima_2em3 = 1 - stats.norm.cdf((200/3 - p_hat) / se)

# ═══ CONTA 2 - a Dirichlet ═══
# Médias: proporções normalizadas (somam 1).
# Concentração total calibrada para a marginal da resiliente
# reproduzir o IC publicado (SE na escala normalizada).
rng = np.random.default_rng(42)
medias = props / props.sum()
se_norm = se / props.sum()
alpha0 = medias[0]*(1 - medias[0]) / se_norm**2 - 1
alphas = alpha0 * medias

amostras = rng.dirichlet(alphas, size=100_000)
ic_cred = np.percentile(amostras, [2.5, 97.5], axis=0)

# A sala de 30, com incerteza propagada:
n_sala = 30
esperados = medias * n_sala
ic_esperados = ic_cred * n_sala                     # credibilidade dos esperados
salas = np.array([rng.multinomial(n_sala, p) for p in amostras[:20_000]])
ic_pred = np.percentile(salas, [2.5, 97.5], axis=0)  # preditivo da sala real
sala_exemplo = rng.multinomial(n_sala, medias)

print("=" * 65)
print("  MODELO - POSTERIOR DIRICHLET + A SALA DE 30 (semente 42)")
print("=" * 65)
print(f"\n  CONTA 1 - quão certa é a maioria?")
print(f"  → P(maioria robusta, >60%)      = {p_acima_60*100:.1f}%")
print(f"  → P(passa de 2 em 3, >66,7%)    = {p_acima_2em3*100:.0f}%")
print(f"  → Forte sem inflar: 3 em 5 com folga, não milagre.")
print(f"\n  CONTA 2 - a Dirichlet (α total = {alpha0:.0f}, calibrado no IC publicado):")
print(f"  {'trajetória':<16} | posterior 95%        | esperados na sala | preditivo")
for i, tr in enumerate(trajetorias):
    print(f"  {tr:<16} | [{ic_cred[0,i]*100:>4.1f}%, {ic_cred[1,i]*100:>4.1f}%]   |"
          f"  {esperados[i]:>4.1f}  [{ic_esperados[0,i]:>4.1f}, {ic_esperados[1,i]:>4.1f}] |"
          f"  [{ic_pred[0,i]:>2.0f}, {ic_pred[1,i]:>2.0f}]")
print(f"\n  → Na sala de 30: com 95% de credibilidade, entre")
print(f"     {ic_pred[0,0]:.0f} e {ic_pred[1,0]:.0f} pessoas seguem de pé (preditivo da sala real).")
print(f"  → Sorteio-exemplo: {sala_exemplo[0]}/{sala_exemplo[1]}/{sala_exemplo[2]}/{sala_exemplo[3]}")
print(f"\n  A LEITURA:")
print(f"  → Resiliente não é quem não sente. É quem funciona enquanto dói.")
print(f"  → E é a sala inteira que importa: mesmo no melhor sorteio,")
print(f"     ~1 em cada 5 daquela sala vai precisar de ajuda de verdade.")
print("=" * 65)
```

```python
# --- VISUALIZAÇÃO - GRÁFICOS SEPARADOS ---

# ── GRÁFICO 1 - As quatro trajetórias (estilizadas, declaradas) ──
fig1, ax1 = plt.subplots(figsize=(10, 6))

t = np.linspace(0, 24, 300)   # meses após o evento
resiliente = 15 + 5*np.exp(-t/2)
recuperacao = 70*np.exp(-t/7) + 15
cronica = 68 - 6*np.exp(-t/6)
tardia = 18 + 44/(1 + np.exp(-(t-12)/2.5))

ax1.plot(t, resiliente, color='#2a8a82', linewidth=3.2, label=f'Resiliente ({props[0]}%)')
ax1.plot(t, recuperacao, color='#c9a14a', linewidth=2.4, label=f'Recuperação ({props[1]}%)')
ax1.plot(t, cronica, color='#c0392b', linewidth=2.4, label=f'Crônica ({props[2]}%)')
ax1.plot(t, tardia, color='#6b6a64', linewidth=2.4, linestyle='--', label=f'Início tardio ({props[3]}%)')
ax1.set_xlabel('Meses após o evento')
ax1.set_ylabel('Sintomas')
ax1.set_yticks([])
ax1.legend(loc='center right', frameon=False, fontsize=11)
ax1.set_title('As quatro trajetórias depois do pior (formas ESTILIZADAS)\n'
              'A verde, embaixo o tempo todo, é a mais comum: 65,7%',
              fontsize=12.5, pad=15)

plt.figtext(0.5, 0.01,
            'Fonte: Galatzer-Levy, Huang & Bonanno (2018) | curvas ilustrativas dos formatos, proporções reais | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-094-grafico-01-trajetorias.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 - Forest das proporções ──
fig2, ax2 = plt.subplots(figsize=(10, 6))

ys = [3, 2, 1, 0]
cores = ['#2a8a82', '#c9a14a', '#c0392b', '#6b6a64']
for y, p, ic, cor in zip(ys, props, ics, cores):
    ax2.plot([ic[0], ic[1]], [y, y], color=cor, linewidth=3)
    ax2.plot([ic[0], ic[0]], [y-0.07, y+0.07], color=cor, linewidth=3)
    ax2.plot([ic[1], ic[1]], [y-0.07, y+0.07], color=cor, linewidth=3)
    ax2.scatter([p], [y], s=130, color=cor, zorder=3)
    ax2.text(74, y, f'{p:.1f}%  [{ic[0]}, {ic[1]}]',
             va='center', fontsize=11.5, fontfamily='monospace')

ax2.axvline(50, color='#333', linestyle='--', linewidth=1.2)
ax2.text(50.8, 3.42, 'linha da maioria', fontsize=10, color='#333')
ax2.scatter([66.0], [3], s=170, facecolors='none', edgecolors='#1a5f5a',
            linewidths=2.2, zorder=4)
ax2.text(59.2, 3.30, 'pandemia: 66,0%', fontsize=10, color='#1a5f5a')

ax2.set_yticks(ys)
ax2.set_yticklabels(trajetorias, fontsize=12)
ax2.set_xlim(0, 100)
ax2.set_ylim(-0.55, 3.65)
ax2.set_xlabel('Prevalência média entre estudos (%, IC 95%)')
ax2.set_title('Seguir de pé é a trajetória modal, com folga\n'
              '54 estudos (2018) + réplica pandêmica de 28 estudos (2022)',
              fontsize=12.5, pad=15)

plt.figtext(0.5, 0.01,
            'Fontes: Galatzer-Levy et al. (2018); Schäfer et al. (2022) | médias entre estudos somam 106%, declarado | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-094-grafico-02-forest.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 - A sala de 30, com a incerteza da Dirichlet ──
fig3, ax3 = plt.subplots(figsize=(10, 6))

cores_map = ['#2a8a82', '#c9a14a', '#c0392b', '#6b6a64']
x = np.arange(4)
bars3 = ax3.bar(x, esperados, color=cores_map, alpha=0.88, width=0.5)
ax3.errorbar(x, esperados,
             yerr=[esperados - ic_pred[0], ic_pred[1] - esperados],
             fmt='none', ecolor='#333', elinewidth=2, capsize=6, capthick=2)
ax3.set_xticks(x)
ax3.set_xticklabels(['Seguem\nde pé', 'Afundam\ne voltam', 'Afundam\ne ficam', 'Pioram\ndepois'], fontsize=11)
ax3.set_ylabel('Pessoas (numa sala de 30)')
ax3.set_ylim(0, 26)
ax3.set_title('A sala de 30, com a incerteza honesta (posterior Dirichlet)\n'
              f'Com 95% de credibilidade: entre {ic_pred[0,0]:.0f} e {ic_pred[1,0]:.0f} seguem de pé',
              fontsize=12.5, pad=15)

for xi, (e, lo, hi) in enumerate(zip(esperados, ic_pred[0], ic_pred[1])):
    ax3.text(xi, hi + 0.7, f'{e:.0f}\n[{lo:.0f}, {hi:.0f}]',
             ha='center', fontweight='bold', fontsize=11.5)

plt.figtext(0.5, 0.01,
            'Posterior Dirichlet calibrada no IC de Galatzer-Levy et al. (2018) | barras: intervalo preditivo da sala real | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-094-grafico-03-sala30.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**Depois do pior, 2 em cada 3 pessoas seguem de pé. Seguir de pé
é o padrão da espécie, não a exceção heroica.**

A posterior diz o quanto confiar: mais de 99,9% de probabilidade
de os resilientes serem maioria, 99,7% de passarem de 60%. E a
honestidade fina, porque este projeto não infla número: a chance
de passar de 2 em 3 é só 32%. É 3 em 5 com folga, não milagre.

Numa sala de 30 que atravessou o pior: cerca de 19 seguem de pé,
6 afundam e voltam, 3 afundam e ficam, 2 ou 3 pioram depois de
parecerem bem. E a posterior Dirichlet coloca a incerteza honesta
em cima: **com 95% de credibilidade, entre 13 e 24 pessoas da
sala real seguem de pé**. A sala inteira importa: o padrão não
apaga a minoria que vai precisar de ajuda de verdade.

E a definição importa mais que o número: **resiliente não é quem
não sente. É quem funciona enquanto dói.** Sintomas baixos não
são ausência de golpe; são a vida seguindo apesar dele.

### ⚠️ Limitações do Modelo

- As proporções são médias entre estudos e somam 106%, não 100:
  cada estudo encontra seu conjunto de trajetórias; a sala de 30
  usa proporções normalizadas, declarado
- O número exato depende das escolhas de modelagem (debate
  Infurna & Luthar, 2016): com desfechos múltiplos e critérios
  mais duros, as taxas de resiliência caem; o que é robusto entre
  métodos é a resiliência como trajetória modal, não o 65,7%
- Em crianças (6 estudos), a resiliência cai para 52%
- Trajetórias descrevem sintomas medidos, não sofrimento
  subjetivo integral; "seguir de pé" não significa não sofrer
- As curvas do gráfico 1 são estilizadas: os formatos são os do
  campo, os traçados exatos variam por estudo
- A posterior da resiliente usa aproximação normal sobre o IC
  publicado; a Dirichlet usa as proporções normalizadas como
  médias e concentração calibrada para a marginal da resiliente
  reproduzir esse mesmo IC. Suposições declaradas; os intervalos
  das outras três trajetórias são derivados, não publicados
- Ponte com o Dia #010 (crescimento pós-traumático): fontes e
  construtos distintos; crescer e não cair são claims diferentes

*A ciência é honesta sobre o que não sabe. O modelo também.*
