### 📖 A História

Reparei numa coisa que me incomodou.

Quando alguém me pergunta como foi o meu último ano, ou o que eu vivi nos dezoito anos de mercado financeiro, eu nunca respondo com a verdade completa. Não porque eu esteja mentindo. Porque a verdade completa não cabe.

O que sai da minha boca são duas ou três cenas. Um pico, geralmente positivo. Um momento difícil que ficou marcado. E o clima do fim, se o assunto tem fim. É isso. O resto (as centenas de dias intermediários, as reuniões esquecidas, as pequenas alegrias diárias, o cansaço acumulado que virou paisagem) evaporou.

Achei isso curioso primeiro. Depois preocupante. Se eu, que passei quase duas décadas obcecada por precisão de dados, resumo dezoito anos em três cenas quando alguém me pergunta, o que aconteceu com os outros cinco mil e quinhentos dias?

Fui atrás dos dados sobre isso. E o que encontrei é uma das descobertas mais desconcertantes da psicologia cognitiva das últimas décadas. Nossa memória de qualquer experiência, boa ou ruim, longa ou curta, não é uma média. Não é um resumo. É praticamente uma equação de dois termos: o pico da intensidade emocional, e o momento final.

Tudo o mais? Desprezado.

E não é falha. É funcionamento.

### 📚 O Conceito: A Regra do Pico e Fim

Em 1993, um grupo liderado por Daniel Kahneman (que anos depois ganharia o Nobel de Economia por trabalhos como esse) publicou um experimento que virou clássico. Batizado de *cold pressor*, ele funcionava assim: participantes mergulhavam a mão em água muito fria por um tempo, avaliavam o desconforto, e depois eram convidados a repetir a experiência.

Havia duas versões. A **versão curta**: 60 segundos de mão em água a 14°C. A **versão longa**: os mesmos 60 segundos a 14°C, mais 30 segundos adicionais em que a água era ligeiramente aquecida (para 15°C, ainda desconfortável, mas menos).

Em qualquer definição racional, a versão longa é pior. Ela tem 90 segundos totais de sofrimento, contra 60. Todo o desconforto da curta, mais 30 segundos a mais.

Quando perguntados qual versão gostariam de repetir, **69% dos participantes escolheram a versão longa**. A que dura mais tempo. A que tem mais sofrimento total.

Por quê? Porque a versão longa terminava melhor. E foi o final que ficou na memória.

Três anos depois, Redelmeier e Kahneman testaram isso num contexto real e delicado: pacientes fazendo colonoscopia. Compararam a avaliação em tempo real (a cada 60 segundos, durante o procedimento) com a lembrança retrospectiva depois. E descobriram algo que reorganizou o entendimento científico de como a memória funciona.

A lembrança do procedimento correlacionou fortemente com a média entre o pico da dor e o momento final. E correlacionou praticamente nada com a duração total. Uma colonoscopia de 4 minutos com pico de dor 8 e fim doloroso era lembrada como pior do que uma de 24 minutos com o mesmo pico mas fim tranquilo.

Kahneman deu nome a esse fenômeno: **peak-end rule** (regra do pico e fim), e um sub-fenômeno associado: *duration neglect* (desprezo pela duração).

Em 2003, Redelmeier fez um ensaio randomizado que confirmou causalmente o efeito. Prolongar a colonoscopia com três minutos adicionais de desconforto leve fez os pacientes avaliarem melhor a experiência inteira, e voltarem em maior número para o exame de acompanhamento.

Adicionar sofrimento tornou a memória melhor.

### 🧮 O Modelo

**Fontes:**

- Kahneman, D., Fredrickson, B. L., Schreiber, C. A. & Redelmeier, D. A. (1993) — "When More Pain Is Preferred to Less: Adding a Better End." *Psychological Science*, 4(6), 401-405 — o experimento *cold pressor* fundacional
- Redelmeier, D. A. & Kahneman, D. (1996) — "Patients' memories of painful medical treatments: real-time and retrospective evaluations of two minimally invasive procedures." *Pain*, 66(1), 3-8 — N=154 pacientes de colonoscopia, correlação retrospectiva com pico e fim ≈ 0,67
- Redelmeier, D. A., Katz, J. & Kahneman, D. (2003) — "Memories of colonoscopy: a randomized trial." *Pain*, 104(1-2), 187-194 — RCT causal, N=682

**Nota metodológica:** neste dia **não aplico o fator de correção ×0.80**. Os achados vêm de experimentos controlados e ensaios randomizados com medidas objetivas (avaliação em tempo real vs retrospectiva, escolha binária de repetir experiência), não de surveys populacionais com viés de autorrelato. Não há categoria de erro para o fator conservador corrigir.

A limitação relevante é outra e ela é importante: a peak-end rule foi validada com solidez em experiências curtas e coerentes (sessões de dor com fim claro, procedimentos médicos, séries de filmes). Miron-Shatz (2009) mostrou que a regra **não se aplica bem** a experiências longas e episódicas, como avaliar um dia inteiro composto de muitas sub-experiências independentes. Aplicar diretamente para "como foi o meu ano" ou "como foi minha carreira" é uma extensão razoável mas não completamente sustentada pela ciência primária.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Kahneman et al. (1993), Psychological Science 4(6):401-405
# Redelmeier & Kahneman (1996), Pain 66(1):3-8
# Redelmeier, Katz & Kahneman (2003), Pain 104:187-194

# Experimento cold pressor (Kahneman 1993)
n_cold_pressor = 32
p_preferiram_versao_longa = 0.69   # 69% preferiram a versao com MAIS dor total
                                    # porque tinha final menos doloroso

# Colonoscopia (Redelmeier & Kahneman 1996)
n_colonoscopia_1996 = 154
correlacao_peak_end = 0.67          # correlacao entre lembranca e (pico + fim) / 2
correlacao_duracao = 0.03           # correlacao entre lembranca e duracao total
                                    # (praticamente zero: duration neglect)

# RCT causal (Redelmeier, Katz & Kahneman 2003)
n_rct = 682
# Adicionar 3 min de desconforto leve no fim melhorou a avaliacao retrospectiva
# e aumentou a taxa de retorno para colonoscopia de acompanhamento
melhora_avaliacao_rct = 0.10        # ~10% de melhora na avaliacao subjetiva
aumento_retorno_rct = 0.05          # ~5pp de aumento na taxa de retorno

print("=" * 68)
print("  DADOS — PEAK-END RULE E DURATION NEGLECT")
print("=" * 68)

print(f"\n  EXPERIMENTO COLD PRESSOR (Kahneman et al., 1993):")
print(f"  N = {n_cold_pressor} participantes")
print(f"  Versao A: 60s de agua a 14C")
print(f"  Versao B: 60s a 14C + 30s a 15C (mais tempo, mais dor total)")
print(f"  {p_preferiram_versao_longa*100:.0f}% preferiram REPETIR a VERSAO LONGA")
print(f"    porque terminava com dor ligeiramente menor.")

print(f"\n  COLONOSCOPIA REAL (Redelmeier & Kahneman, 1996):")
print(f"  N = {n_colonoscopia_1996} pacientes")
print(f"  Correlacao entre lembranca e MEDIA (pico + fim):    r = {correlacao_peak_end}")
print(f"  Correlacao entre lembranca e DURACAO total:          r = {correlacao_duracao}")
print(f"  Duracao e praticamente ignorada pela memoria.")

print(f"\n  RCT CAUSAL (Redelmeier, Katz & Kahneman, 2003):")
print(f"  N = {n_rct} pacientes randomizados")
print(f"  Adicionar 3 min de desconforto LEVE no fim:")
print(f"  - Melhorou a avaliacao retrospectiva (~{melhora_avaliacao_rct*100:.0f}%)")
print(f"  - Aumentou o retorno para exame de acompanhamento (+{aumento_retorno_rct*100:.0f}pp)")
print(f"  Adicionar sofrimento tornou a memoria melhor.")
print("=" * 68)
print()
print(f"  NOTA: sem fator x0.80 neste dia.")
print(f"  Sao correlacoes e proporcoes de experimentos controlados,")
print(f"  nao surveys populacionais de autorrelato.")
```

```python
# --- O MODELO ---
# A equacao da memoria e a assinatura estatistica: Bayes sobre a correlacao

# A "formula" simplificada da lembranca segundo Kahneman:
#   Lembranca ~ (Pico + Fim) / 2
# ignorando a media dos momentos intermediarios e a duracao total.

# Modelo Bayesiano sobre a correlacao r = 0.67 do estudo de colonoscopia
# usando distribuicao Beta como conjugada aproximada para |r|
# (transformacao de Fisher r -> z para IC 95% da correlacao)

r_observado = correlacao_peak_end
n_amostra = n_colonoscopia_1996

# Transformacao z de Fisher
z_fisher = 0.5 * np.log((1 + r_observado) / (1 - r_observado))
se_z = 1 / np.sqrt(n_amostra - 3)

# Intervalo de confianca em z, depois transformando de volta para r
z_lower = z_fisher - 1.96 * se_z
z_upper = z_fisher + 1.96 * se_z
r_lower = (np.exp(2*z_lower) - 1) / (np.exp(2*z_lower) + 1)
r_upper = (np.exp(2*z_upper) - 1) / (np.exp(2*z_upper) + 1)

# Para comparacao: correlacao com duracao
r_duracao = correlacao_duracao
z_duracao = 0.5 * np.log((1 + r_duracao) / (1 - r_duracao))
z_duracao_lower = z_duracao - 1.96 * se_z
z_duracao_upper = z_duracao + 1.96 * se_z
r_duracao_lower = (np.exp(2*z_duracao_lower) - 1) / (np.exp(2*z_duracao_lower) + 1)
r_duracao_upper = (np.exp(2*z_duracao_upper) - 1) / (np.exp(2*z_duracao_upper) + 1)

print("=" * 68)
print("  MODELO — A ASSINATURA ESTATISTICA DA MEMORIA")
print("=" * 68)

print(f"\n  A EQUACAO SIMPLIFICADA DE KAHNEMAN:")
print(f"    Lembranca ~ (Pico + Fim) / 2")
print(f"    A media dos momentos intermediarios: DESPREZADA")
print(f"    A duracao total da experiencia:      DESPREZADA")

print(f"\n  INTERVALOS DE CONFIANCA 95% (Fisher z-transform, N={n_amostra}):")
print(f"\n  Correlacao lembranca ~ pico+fim:")
print(f"    r = {r_observado:.2f}   IC 95% = [{r_lower:.2f}, {r_upper:.2f}]")
print(f"    interpretacao: correlacao FORTE, robusta")
print(f"\n  Correlacao lembranca ~ duracao:")
print(f"    r = {r_duracao:.2f}   IC 95% = [{r_duracao_lower:.2f}, {r_duracao_upper:.2f}]")
print(f"    interpretacao: essencialmente ZERO (duration neglect)")

print(f"\n  OS DOIS EUS (Kahneman):")
print(f"  - EU QUE VIVE (experiencing self):")
print(f"    passa por tudo, sente cada momento, registra em tempo real")
print(f"  - EU QUE LEMBRA (remembering self):")
print(f"    constroi a historia depois, com apenas 2 pontos + fim")
print(f"    e QUEM decide o proximo passo da sua vida")
print("=" * 68)
```

```python
# --- VISUALIZACAO ---
# Neste dia trazemos de volta a ASSINATURA ESTATISTICA VISIVEL
# (distribuicoes, IC 95%, curva de credibilidade Bayesiana)

cor_pico = '#c0392b'
cor_lembra = '#c9a14a'
cor_vive = '#2a8a82'
cor_neutra = '#9c9b94'

# GRAFICO 1 — A anatomia da lembranca (linha temporal com pico e fim marcados)
fig1, ax1 = plt.subplots(figsize=(11, 6.5))

# Simulacao de uma experiencia com intensidade emocional ao longo do tempo
tempo = np.linspace(0, 100, 500)
np.random.seed(42)
# Uma curva com pico em ~35% do tempo e final moderado
intensidade = 3 + 4*np.exp(-((tempo-35)/12)**2) + 1.5*np.sin(tempo/8) + np.random.normal(0, 0.3, 500)
intensidade = np.clip(intensidade, 0, 10)

ax1.plot(tempo, intensidade, color=cor_vive, linewidth=2, alpha=0.7,
         label='O que voce viveu (momento a momento)')
ax1.fill_between(tempo, intensidade, alpha=0.10, color=cor_vive)

# Marcar o pico
idx_pico = np.argmax(intensidade)
pico_valor = intensidade[idx_pico]
ax1.plot(tempo[idx_pico], pico_valor, 'o', color=cor_pico, markersize=16, zorder=5)
ax1.annotate('PICO', xy=(tempo[idx_pico], pico_valor),
             xytext=(tempo[idx_pico], pico_valor + 1),
             fontsize=13, color=cor_pico, fontweight='bold', ha='center')

# Marcar o fim (media dos ultimos 5%)
fim_valor = np.mean(intensidade[-25:])
ax1.plot(100, fim_valor, 's', color=cor_pico, markersize=14, zorder=5)
ax1.annotate('FIM', xy=(100, fim_valor),
             xytext=(97, fim_valor + 1),
             fontsize=13, color=cor_pico, fontweight='bold', ha='center')

# Linha da lembranca (media entre pico e fim)
lembranca = (pico_valor + fim_valor) / 2
ax1.axhline(y=lembranca, color=cor_lembra, linestyle='--', linewidth=2.5, alpha=0.9)
ax1.text(50, lembranca + 0.25, f'O que voce vai lembrar: (pico + fim) / 2 = {lembranca:.1f}',
         ha='center', fontsize=11, color='#8a7020', fontweight='bold', style='italic')

# Media real vs media lembrada
media_real = np.mean(intensidade)
ax1.axhline(y=media_real, color=cor_neutra, linestyle=':', linewidth=1.5, alpha=0.7)
ax1.text(50, media_real - 0.5, f'Media real do que voce viveu: {media_real:.1f}',
         ha='center', fontsize=10, color='#555', style='italic')

ax1.set_xlabel('Tempo (% da experiencia)')
ax1.set_ylabel('Intensidade emocional')
ax1.set_ylim(0, 10)
ax1.set_xlim(0, 105)
ax1.set_title('A anatomia da lembranca\nA memoria descarta a media real e retem (pico + fim) / 2',
              fontsize=13, pad=15)
ax1.legend(frameon=False, loc='upper right', fontsize=10)

plt.figtext(0.5, 0.005,
            'Ilustracao conceitual da peak-end rule (Kahneman et al., 1993) | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-035-grafico-01-anatomia-lembranca.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 1 salvo!")

# GRAFICO 2 — A ASSINATURA ESTATISTICA VISIVEL
# Distribuicao amostral da correlacao com IC 95%
fig2, ax2 = plt.subplots(figsize=(11, 6.5))

# Distribuicao aproximada da correlacao (via transformacao de Fisher)
r_range = np.linspace(-0.2, 0.95, 500)
z_range = 0.5 * np.log((1 + r_range) / (1 - r_range))

# Densidade para r ~ 0.67 (pico+fim)
z_center_peakend = z_fisher
densidade_peakend = stats.norm.pdf(z_range, loc=z_center_peakend, scale=se_z)
# Ajuste para densidade sobre r (jacobiano)
densidade_peakend_r = densidade_peakend / (1 - r_range**2) * (1 - r_range**2)

# Densidade para r ~ 0.03 (duracao)
z_center_dur = z_duracao
densidade_dur = stats.norm.pdf(z_range, loc=z_center_dur, scale=se_z)

# Normalizar para plot
densidade_peakend_norm = densidade_peakend / np.max(densidade_peakend)
densidade_dur_norm = densidade_dur / np.max(densidade_dur)

ax2.plot(r_range, densidade_peakend_norm, color=cor_pico, linewidth=3,
         label=f'Correlacao com PICO + FIM  (r = {r_observado:.2f})')
ax2.fill_between(r_range, densidade_peakend_norm, alpha=0.20, color=cor_pico)

ax2.plot(r_range, densidade_dur_norm, color=cor_neutra, linewidth=3,
         label=f'Correlacao com DURACAO         (r = {r_duracao:.2f})')
ax2.fill_between(r_range, densidade_dur_norm, alpha=0.20, color=cor_neutra)

# IC 95% do peak-end
ax2.axvline(x=r_lower, color=cor_pico, linestyle='--', linewidth=1, alpha=0.7)
ax2.axvline(x=r_upper, color=cor_pico, linestyle='--', linewidth=1, alpha=0.7)
ax2.text(r_observado, 0.55, f'IC 95%\n[{r_lower:.2f}, {r_upper:.2f}]',
         ha='center', fontsize=10, color='#7a1810', fontweight='bold')

ax2.axvline(x=0, color='#333', linewidth=1, alpha=0.4)

ax2.set_xlabel('Coeficiente de correlacao (r) com a lembranca retrospectiva')
ax2.set_ylabel('Densidade (normalizada)')
ax2.set_xlim(-0.15, 0.95)
ax2.set_ylim(0, 1.15)
ax2.set_title('A assinatura estatistica da peak-end rule\nRedelmeier & Kahneman (1996), N=154 pacientes de colonoscopia',
              fontsize=13, pad=15)
ax2.legend(frameon=False, loc='upper left', fontsize=10)

plt.figtext(0.5, 0.005,
            'IC 95% via transformacao z de Fisher | Pico+fim: forte e robusto. Duracao: praticamente zero. | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-035-grafico-02-assinatura-estatistica.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 2 salvo!")

# GRAFICO 3 — O paradoxo da versao longa (Kahneman 1993)
fig3, ax3 = plt.subplots(figsize=(11, 6.5))

tempo_curto = np.linspace(0, 60, 100)
temp_curto = np.full_like(tempo_curto, 14)   # 14C durante toda a versao

tempo_longo = np.linspace(0, 90, 150)
temp_longo = np.where(tempo_longo <= 60, 14, 15)  # 14C ate 60s, 15C ate 90s

# Converter temperatura para "dor" (proxy inverso)
dor_curto = (17 - temp_curto)
dor_longo = (17 - temp_longo)

ax3.plot(tempo_curto, dor_curto, color=cor_pico, linewidth=3,
         label=f'Versao CURTA (60s): dor total menor')
ax3.fill_between(tempo_curto, dor_curto, alpha=0.15, color=cor_pico)

ax3.plot(tempo_longo, dor_longo, color=cor_vive, linewidth=3,
         label=f'Versao LONGA (90s): dor total MAIOR')
ax3.fill_between(tempo_longo, dor_longo, alpha=0.15, color=cor_vive)

ax3.axvline(x=60, color='#666', linestyle=':', linewidth=1, alpha=0.6)

ax3.text(75, 2.2, 'ligeira melhora\nno final',
         ha='center', fontsize=11, color=cor_vive, fontweight='bold', style='italic')

# Anotacao central
ax3.text(45, 4.5, f'{p_preferiram_versao_longa*100:.0f}% preferiram a VERSAO LONGA\n(mais dor total, final menos ruim)',
         ha='center', fontsize=13, color='#8a7020', fontweight='bold',
         bbox=dict(boxstyle='round,pad=0.6', facecolor='#fff5d6', edgecolor=cor_lembra, linewidth=1.5))

ax3.set_xlabel('Tempo (segundos)')
ax3.set_ylabel('Nivel de dor (proxy inverso da temperatura)')
ax3.set_ylim(0, 5.5)
ax3.set_xlim(0, 95)
ax3.set_title('O paradoxo do cold pressor\nAdicionar mais sofrimento tornou a memoria melhor',
              fontsize=13, pad=15)
ax3.legend(frameon=False, loc='lower right', fontsize=10)

plt.figtext(0.5, 0.005,
            'Kahneman, Fredrickson, Schreiber & Redelmeier (1993), Psychological Science | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-035-grafico-03-paradoxo-cold-pressor.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 3 salvo!")
```

### 💡 O Insight

**Sua lembrança de qualquer experiência é dominada por dois momentos: o pico da intensidade emocional e o momento final. A duração é praticamente ignorada.**

Isso significa que quando você conta sua vida, você não está mentindo, mas também não está descrevendo. Está reportando o resumo estatístico que sua própria memória calculou, sem sua permissão, usando dois pontos amostrais em vez de milhares.

Kahneman formalizou isso em três conceitos que vale registrar:

O **eu que vive** (*experiencing self*) é quem passa por todas as sensações momento a momento. Ele registra, mas não escolhe o que fica.

O **eu que lembra** (*remembering self*) é quem constrói a história depois. Ele opera com pico e fim, ignora duração, e é essa versão que decide o que você vai contar aos outros e a você mesma.

O problema é que **as decisões importantes da vida são tomadas pelo eu que lembra**, não pelo eu que vive. Você não decide voltar a uma viagem, um trabalho, um relacionamento, baseado na média das horas que passou lá. Decide baseado em como aquilo terminou e nos dois ou três momentos mais intensos que sobreviveram na memória.

Traduzido pra escolhas concretas: pessoas que tiveram um relacionamento longo, mas cujos últimos meses foram ruins, tendem a lembrar do relacionamento inteiro como pior do que ele foi. Trabalhos que terminaram bem são lembrados com uma nostalgia que a média dos dias não sustentaria. Férias com um dia final estragado por chuva ou briga contaminam a lembrança das outras seis semanas ensolaradas.

E o inverso também vale, e talvez seja mais útil de saber. Se você quer que uma experiência boa fique boa na memória, o que mais importa não é o meio. É o pico e o final. Aumentar o cuidado com esses dois momentos é uma alavanca de memória desproporcional em relação ao esforço.

O que me pegou não foi descobrir que a memória é seletiva. Isso todo mundo sabe. Foi descobrir que **a seleção segue uma fórmula previsível**. Não é ruído aleatório. É um viés estruturado, quase matemático, que reduz qualquer experiência da vida a duas variáveis mais uma constante.

*Se você olhar honestamente pra sua última grande decisão de vida (quem você amou, o que você deixou, o que você começou), essa decisão foi tomada com base no que você viveu, ou no que sua memória escolheu lembrar?*

### ⚠️ Limitações do Modelo

- Os experimentos originais foram feitos em contextos curtos e coerentes (dor de mão em água fria, procedimentos médicos, exposição a sons). A generalização para "como foi minha carreira" ou "como foi meu ano" é razoável mas não totalmente validada. Miron-Shatz (2009) documentou condições em que a peak-end rule falha em experiências mais longas.
- A força do efeito varia por tipo de experiência. Para dor aguda com fim claro, a correlação é altíssima. Para prazer prolongado, é mais fraca.
- O estudo de 2003 é o mais robusto (RCT com N=682) e confirma causalidade em contexto médico. Extrapolar para outros contextos requer cautela.
- Cultura pode moderar o efeito. A maior parte dos estudos foi feita em populações ocidentais.
- Este texto não é um manual pra manipular memória alheia. É um mapa para você reconhecer o próprio funcionamento cognitivo.
