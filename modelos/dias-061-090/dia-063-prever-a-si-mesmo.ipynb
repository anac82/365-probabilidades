## Qual a probabilidade de você ser menos ético do que acha que é?

**Tipo:** Comportamental

**Decisão analisada:** Posso confiar na minha própria previsão sobre o que eu vou fazer?

### 📖 A História

Pergunte a qualquer pessoa se ela vai votar. Quase todo mundo diz que sim.

Pergunte se ela vai doar, se vai ajudar, se vai cooperar quando ninguém estiver
olhando. A resposta é sim de novo, com uma segurança que não combina nada com o
que acontece depois.

Agora faça a mesma pergunta sobre os outros. A resposta cai. As pessoas acham
que os outros vão faltar, desistir, escolher o próprio bolso.

O detalhe interessante não é que a gente se ache melhor que a média. Isso já se
sabia. O detalhe é qual das duas previsões está errada.

Em 2000, dois psicólogos de Cornell mediram as duas ao mesmo tempo e depois
foram conferir o que as pessoas fizeram de verdade. A previsão sobre os outros
estava quase certa. A previsão sobre si mesmo estava muito errada.

Ou seja: o problema não é que você seja cínico demais com os outros. É que você
é péssimo em prever você.

---

### 📚 O Conceito: o erro está do lado de dentro

O fenômeno se chama **holier than thou**: as pessoas preveem que vão agir de
forma generosa, cívica ou honesta com frequência muito maior do que preveem para
os colegas.

Isso é logicamente suspeito, porque numa mesma sala todo mundo não pode estar
acima da média da sala. Mas a ambiguidade importante é outra: esse desvio nasce
de uma visão cínica demais sobre os outros ou de uma visão generosa demais sobre
si mesmo?

Epley e Dunning testaram e a resposta foi a segunda. As previsões sobre os
colegas ficaram próximas do comportamento observado. As previsões sobre si mesmo
foram sistematicamente infladas.

Há uma precisão de linguagem que este dia não pode perder: o que está medido é a
**superestimativa da própria conduta socialmente desejável**, e não o veredito de
que você seja uma pessoa pior que a média. São coisas diferentes, e a segunda os
dados não sustentam.

---

### 🧮 O Modelo

Três camadas: o mecanismo em amostra pequena, a escala em registro
administrativo, e o teste de se a sensação de superioridade moral prevê alguma
coisa.

**Fontes:**
- Epley, N. & Dunning, D., 2000 · *Journal of Personality and Social Psychology*
  79(6), 861-875 · Estudo 2, eleição presidencial · 90% previram que votariam,
  colegas previram 67%, taxa real de voto 61% · N do estudo não publicado nas
  fontes acessíveis
- Epley, N. & Dunning, D., 2006 · *Personality and Social Psychology Bulletin*
  32(5) · estudo suplementar, **N=141** eleitores elegíveis · 85% previram que
  votariam, 64% relataram ter votado
- Rogers, T. & Aida, M., 2014 · *American Politics Research* · sete pesquisas
  pré-eleitorais de três eleições com validação de voto por registro
  administrativo, **N=29.403** · flake-out e flake-in, este último entre 29% e 56%
- Rogers, T., ten Brinke, L. & Carney, D. R., 2016 · *PNAS* 113(23), 6449-6453 ·
  n=4.463 e n=3.064 · mais da metade de quem se prevê votante não vota
- Tappin, B. M. & McKay, R. T., 2017 · *Social Psychological and Personality
  Science* 8(6), 623-631 · **N=270** · praticamente todos inflam irracionalmente
  as próprias qualidades morais
- Tappin, B. M. & McKay, R. T., 2019 · *Social Psychological and Personality
  Science* 10(2), 135-143 · dois estudos pré-registrados, **N=827** e **N=825** ·
  superioridade moral autopercebida não prevê comportamento em jogos incentivados

**Nota metodológica sobre o fator ×0.80:** não se aplica. O fator corrige
proporções de survey populacional autorrelatado quando o autorrelato é ruído.
Aqui o autorrelato **é o objeto medido**: o dia inteiro é a distância entre o que
a pessoa diz que vai fazer e o que ela faz. Corrigir os 90% apagaria justamente o
número que a análise existe para expor.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

RNG = np.random.default_rng(42)

print("Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---

# Epley & Dunning, 2000, JPSP 79(6), 861-875 - Estudo 2 (eleicao presidencial)
# N do estudo nao publicado nas fontes acessiveis: ver limitacoes
p_previu_que_votaria      = 0.90   # previsao sobre si mesmo
p_colegas_previram        = 0.67   # previsao dos colegas sobre esses respondentes
p_votou_de_fato           = 0.61   # comportamento relatado apos a eleicao

erro_sobre_si     = p_previu_que_votaria - p_votou_de_fato
erro_sobre_outros = p_colegas_previram   - p_votou_de_fato

# Epley & Dunning, 2006, PSPB 32(5) - estudo suplementar com N publicado
# Unico ponto da linha de pesquisa com contagem e denominador reais
n_2006            = 141
p_previu_2006     = 0.85
p_votou_2006      = 0.64
gap_2006          = p_previu_2006 - p_votou_2006

# Rogers & Aida, 2014, American Politics Research
# Sete pesquisas, tres eleicoes, validacao por registro administrativo
n_rogers          = 29403
flake_in_min      = 0.29   # faixa ENTRE pesquisas, nao estimativa pontual
flake_in_max      = 0.56
flake_out_piso    = 0.50   # Rogers, ten Brinke & Carney, 2016, PNAS: mais da metade

# Tappin & McKay, 2017 e 2019
n_tappin_2017     = 270
n_tappin_2019_e1  = 827
n_tappin_2019_e2  = 825

# Fator de correcao: NAO se aplica neste dia (ver nota metodologica)
aplica_fator_080  = False

print("=" * 65)
print("  DADOS DA LITERATURA - PREVER A SI MESMO")
print("=" * 65)
print("\n  Epley & Dunning, 2000 (Estudo 2, eleicao):")
print(f"  -> Previram que votariam:            {p_previu_que_votaria*100:.0f}%")
print(f"  -> Colegas previram:                 {p_colegas_previram*100:.0f}%")
print(f"  -> Votaram de fato:                  {p_votou_de_fato*100:.0f}%")
print(f"  -> Erro da previsao sobre SI:        {erro_sobre_si*100:.0f} pp")
print(f"  -> Erro da previsao sobre OS OUTROS: {erro_sobre_outros*100:.0f} pp")
print(f"\n  Epley & Dunning, 2006 (N={n_2006}):")
print(f"  -> Previram que votariam:            {p_previu_2006*100:.0f}%")
print(f"  -> Relataram ter votado:             {p_votou_2006*100:.0f}%")
print(f"  -> Distancia:                        {gap_2006*100:.0f} pp")
print(f"\n  Rogers & Aida, 2014 (N={n_rogers:,}, registro administrativo):")
print(f"  -> Flake-out: mais de {flake_out_piso*100:.0f}% de quem se preve votante nao vota")
print(f"  -> Flake-in:  {flake_in_min*100:.0f}% a {flake_in_max*100:.0f}% de quem se preve"
      " nao votante vota")
print(f"\n  Tappin & McKay, 2017 (N={n_tappin_2017}):")
print("  -> Praticamente todos inflam irracionalmente as qualidades morais")
print(f"\n  Tappin & McKay, 2019 (N={n_tappin_2019_e1} e N={n_tappin_2019_e2}):")
print("  -> Superioridade moral autopercebida NAO preve comportamento")
print("=" * 65)
```

```python
# --- O MODELO ---
# Assinatura estatistica: distribuicao Beta com IC 95% sobre o unico ponto
# da literatura com contagem e denominador reais (Epley & Dunning, 2006, N=141).

# Beta(alpha, beta) com alpha = p*N e beta = (1-p)*N
alpha_previsto = p_previu_2006 * n_2006
beta_previsto  = (1 - p_previu_2006) * n_2006
alpha_real     = p_votou_2006 * n_2006
beta_real      = (1 - p_votou_2006) * n_2006

dist_previsto = stats.beta(alpha_previsto, beta_previsto)
dist_real     = stats.beta(alpha_real, beta_real)

ic_previsto = dist_previsto.interval(0.95)
ic_real     = dist_real.interval(0.95)

# Distancia entre previsao e comportamento, com incerteza.
# As duas medidas vem da MESMA amostra, logo sao pareadas. Tratar como
# independentes ALARGA o intervalo da diferenca: escolha conservadora, declarada.
amostras = 200_000
dif = dist_previsto.rvs(amostras, random_state=RNG) - dist_real.rvs(amostras, random_state=RNG)
ic_dif = np.percentile(dif, [2.5, 97.5])
p_dif_maior_que_zero = float((dif > 0).mean())
txt_p_dif = ">99,9%" if p_dif_maior_que_zero >= 0.999 else f"{p_dif_maior_que_zero*100:.1f}%"


print("=" * 65)
print("  MODELO - A DISTANCIA ENTRE DIZER E FAZER")
print("=" * 65)
print(f"\n  Epley & Dunning, 2000 (Estudo 2):")
print(f"  -> Previsao sobre si erra por          {erro_sobre_si*100:.0f} pp")
print(f"  -> Previsao sobre os outros erra por   {erro_sobre_outros*100:.0f} pp")
print(f"  -> Razao entre os dois erros:          {erro_sobre_si/erro_sobre_outros:.1f}x")
print(f"\n  Beta com IC 95% (Epley & Dunning, 2006, N={n_2006}):")
print(f"  -> Previram que votariam: {p_previu_2006*100:.1f}%"
      f"  IC 95% [{ic_previsto[0]*100:.1f}% - {ic_previsto[1]*100:.1f}%]")
print(f"  -> Votaram de fato:       {p_votou_2006*100:.1f}%"
      f"  IC 95% [{ic_real[0]*100:.1f}% - {ic_real[1]*100:.1f}%]")
print(f"  -> Distancia:             {gap_2006*100:.1f} pp"
      f"  IC 95% [{ic_dif[0]*100:.1f} pp - {ic_dif[1]*100:.1f} pp]")
print(f"  -> P(distancia > 0):      {txt_p_dif}")
print(f"\n  Escala (Rogers & Aida, 2014, N={n_rogers:,}):")
print(f"  -> O erro vai nos DOIS sentidos.")
print(f"     Flake-out: mais de {flake_out_piso*100:.0f}% dizem que votam e nao votam")
print(f"     Flake-in:  {flake_in_min*100:.0f}% a {flake_in_max*100:.0f}% dizem que"
      " nao votam e votam")
print(f"\n  Fator x0.80 aplicado: {aplica_fator_080}")
print("  Motivo: o autorrelato aqui e o objeto medido, nao o ruido a corrigir.")
print("=" * 65)
```

```python
# --- VISUALIZACAO ---

DOURADO = '#c8a84b'
VERMELHO = '#c0392b'
VERDE = '#2a8a82'
CINZA = '#6b6a64'

# GRAFICO 1 - Quem erra sobre quem
fig1, ax1 = plt.subplots(figsize=(12, 8))

rotulos = ['Previram que\nELES MESMOS votariam',
           'Colegas previram\nque eles votariam',
           'Votaram\nde fato']
valores = [p_previu_que_votaria*100, p_colegas_previram*100, p_votou_de_fato*100]
cores = [VERMELHO, VERDE, CINZA]

barras = ax1.bar(rotulos, valores, color=cores, alpha=0.9, width=0.55)
ax1.axhline(y=p_votou_de_fato*100, color=CINZA, linestyle=':', linewidth=1.5, zorder=0)
ax1.set_ylim(0, 108)
ax1.set_ylabel('Proporção (%)')
ax1.set_title('A previsão sobre os outros acerta. A previsão sobre si, não.\n'
              'Epley & Dunning, 2000, JPSP 79(6), Estudo 2',
              fontsize=14, pad=18)

for barra, valor in zip(barras, valores):
    ax1.text(barra.get_x() + barra.get_width()/2, valor + 1.5,
             f'{valor:.0f}%', ha='center', fontweight='bold', fontsize=18)

ax1.annotate('', xy=(0, p_votou_de_fato*100), xytext=(0, p_previu_que_votaria*100),
             arrowprops=dict(arrowstyle='<->', color=VERMELHO, lw=2))
ax1.text(0.31, (p_previu_que_votaria + p_votou_de_fato)/2*100,
         f'erra por {erro_sobre_si*100:.0f} pp', color=VERMELHO,
         fontsize=13, fontweight='bold', va='center')

ax1.annotate('', xy=(1, p_votou_de_fato*100), xytext=(1, p_colegas_previram*100),
             arrowprops=dict(arrowstyle='<->', color=VERDE, lw=2))
ax1.text(1.31, (p_colegas_previram + p_votou_de_fato)/2*100,
         f'erra por {erro_sobre_outros*100:.0f} pp', color=VERDE,
         fontsize=13, fontweight='bold', va='center')

plt.figtext(0.5, 0.005,
            'Fonte: Epley & Dunning, 2000, Journal of Personality and Social Psychology 79(6), 861-875'
            '  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-063-grafico-01-previsao-vs-realidade.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 1 salvo")

# GRAFICO 2 - O erro vai nos dois sentidos
fig2, ax2 = plt.subplots(figsize=(12, 8))

ax2.barh(1, flake_out_piso*100, left=flake_out_piso*100, height=0.30,
         color=VERMELHO, alpha=0.35, edgecolor=VERMELHO, linewidth=2)
ax2.barh(0, (flake_in_max - flake_in_min)*100, left=flake_in_min*100, height=0.30,
         color=VERDE, alpha=0.55, edgecolor=VERDE, linewidth=2)

ax2.set_yticks([0, 1])
ax2.set_yticklabels(['FLAKE-IN\nDisse que NÃO votaria\ne votou',
                     'FLAKE-OUT\nDisse que votaria\ne não votou'], fontsize=12)
ax2.set_xlim(0, 100)
ax2.set_ylim(-0.85, 1.55)
ax2.set_xlabel('Proporção dentro do próprio grupo de previsão (%)')
ax2.set_title('O erro de previsão sobre si mesmo vai nos dois sentidos\n'
              f'Rogers & Aida, 2014, N={n_rogers:,}, sete pesquisas com validação por registro administrativo',
              fontsize=14, pad=18)

ax2.text(flake_out_piso*100 + 2, 1, 'mais da metade',
         va='center', fontsize=13, fontweight='bold', color=VERMELHO)
ax2.text(flake_in_min*100 + 1.5, 0,
         f'{flake_in_min*100:.0f}% a {flake_in_max*100:.0f}%',
         va='center', fontsize=13, fontweight='bold', color=VERDE)

ax2.text(50, -0.63,
         'Faixa entre as sete pesquisas, não estimativa pontual.\n'
         'Piso do flake-out conforme Rogers, ten Brinke & Carney, 2016, PNAS 113(23).',
         ha='center', fontsize=10, color=CINZA, style='italic')

plt.figtext(0.5, 0.005,
            'Fonte: Rogers & Aida, 2014, American Politics Research  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-063-grafico-02-assimetria.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 2 salvo")

# GRAFICO 3 - Assinatura estatistica: Beta com IC 95%
fig3, ax3 = plt.subplots(figsize=(12, 8))

x = np.linspace(0.45, 1.0, 2000)
y_prev = dist_previsto.pdf(x)
y_real = dist_real.pdf(x)

ax3.plot(x*100, y_prev, color=VERMELHO, linewidth=3, label='O que disseram que fariam')
ax3.fill_between(x*100, y_prev, alpha=0.18, color=VERMELHO)
ax3.plot(x*100, y_real, color=CINZA, linewidth=3, label='O que fizeram')
ax3.fill_between(x*100, y_real, alpha=0.18, color=CINZA)

for limite in ic_previsto:
    ax3.axvline(x=limite*100, color=DOURADO, linestyle='--', linewidth=1.6, alpha=0.9)
for limite in ic_real:
    ax3.axvline(x=limite*100, color=DOURADO, linestyle='--', linewidth=1.6, alpha=0.9)

topo = max(y_prev.max(), y_real.max())
ax3.text(p_votou_2006*100, topo*1.02,
         f'IC 95%\n[{ic_real[0]*100:.1f}% - {ic_real[1]*100:.1f}%]',
         ha='center', fontsize=11, color=CINZA)
ax3.text(p_previu_2006*100, topo*1.02,
         f'IC 95%\n[{ic_previsto[0]*100:.1f}% - {ic_previsto[1]*100:.1f}%]',
         ha='center', fontsize=11, color=VERMELHO)

ax3.set_ylim(0, topo*1.28)
ax3.set_xlabel('Proporção (%)')
ax3.set_ylabel('Densidade')
ax3.set_title('A assinatura estatística: Beta(p·N, (1−p)·N) com IC 95%\n'
              f'Epley & Dunning, 2006, PSPB 32(5), N={n_2006}',
              fontsize=14, pad=18)
ax3.legend(frameon=False, fontsize=12, loc='upper left')

ax3.text(0.5, -0.155,
         f'Distância entre as duas: {gap_2006*100:.0f} pp'
         f'  ·  IC 95% [{ic_dif[0]*100:.1f} pp - {ic_dif[1]*100:.1f} pp]'
         f'  ·  P(distância > 0) = {txt_p_dif}',
         ha='center', transform=ax3.transAxes, fontsize=12, color=DOURADO,
         fontweight='bold')

plt.figtext(0.5, 0.005,
            'Fonte: Epley & Dunning, 2006, Personality and Social Psychology Bulletin 32(5)'
            '  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-063-grafico-03-bayesiano.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 3 salvo")
```

### 💡 O Insight

**90% disseram que votariam. 61% votaram.**

Os colegas, que mal conheciam essas pessoas, previram 67%. Erraram por 6 pontos.
Os próprios respondentes erraram por 29.

Um estranho olhando de fora acertou mais sobre você do que você.

E o erro não é só para cima. Quando sete pesquisas com 29.403 eleitores foram
cruzadas com o registro administrativo de votação, apareceu o outro lado: entre
29% e 56% das pessoas que disseram que não votariam foram votar. A previsão sobre
si mesmo não é otimista. É só ruim.

Agora a parte que quase ninguém conta. Dois estudos pré-registrados testaram se
essa sensação de ser mais ético que a média prevê alguma coisa no comportamento
real, com dinheiro em jogo. Não prevê. Nem confiança, nem justiça, nem
reciprocidade. A análise bayesiana ficou do lado da hipótese nula.

Então a manchete honesta não é que você seja pior do que pensa.

É que a sua sensação de ser uma pessoa correta não carrega informação nenhuma
sobre o que você vai fazer quando a conta chegar. Ela não é evidência. É uma
narrativa que você conta antes do fato, e o fato não é obrigado a obedecer.

O que prevê comportamento, nesses estudos, é o que a pessoa fez da última vez.

*Qual foi a última vez que você teve certeza do que faria e depois fez outra
coisa?*

### ⚠️ Limitações do Modelo

- O N do Estudo 2 de Epley & Dunning (2000) não está publicado nas fontes
  acessíveis. Por isso a Beta não foi calculada sobre os 90%, e sim sobre o
  estudo suplementar de 2006, único ponto com contagem e denominador reais.
- O comportamento "real" nos estudos de Epley e Dunning é autorrelato posterior.
  A validação por registro administrativo de votação vem do Rogers & Aida.
- O flake-in de 29% a 56% é **faixa entre sete pesquisas**, não estimativa
  pontual. Representado como intervalo, e não como barra única, de propósito.
- As duas proporções de 2006 vêm da mesma amostra, portanto são pareadas. O
  intervalo da diferença foi calculado tratando as distribuições como
  independentes, o que **alarga** o intervalo. É a escolha conservadora.
- Amostras americanas: universitários no Epley e Dunning, eleitores registrados
  nos EUA no Rogers e Aida. Generalização para outro país e outro sistema
  eleitoral não está testada.
- O que está medido é superestimativa da própria conduta socialmente desejável,
  não um veredito moral comparativo entre pessoas.
- Fator ×0.80 não aplicado. O autorrelato aqui é o objeto medido, não ruído.

*A ciência é honesta sobre o que não sabe. O modelo também.*
