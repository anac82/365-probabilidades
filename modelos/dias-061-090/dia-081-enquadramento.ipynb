### 📖 A História

Imagine que o país se prepara para o surto de uma doença incomum, que deve matar 600
pessoas. Dois programas foram propostos, e as estimativas científicas das consequências
são exatas.

Se o **Programa A** for adotado, **200 pessoas serão salvas**.

Se o **Programa B** for adotado, há um terço de chance de que **600 pessoas sejam
salvas** e dois terços de chance de que **ninguém seja salvo**.

Escolha um.

Agora a mesma situação, outra dupla de programas.

Se o **Programa C** for adotado, **400 pessoas morrerão**.

Se o **Programa D** for adotado, há um terço de chance de que **ninguém morra** e dois
terços de chance de que **600 pessoas morram**.

Escolha um.

Se você é como quase todo mundo, escolheu A na primeira e D na segunda.

E A e C são exatamente a mesma coisa. B e D também.

---

### 📚 O Conceito: o mesmo fato, dois pontos de partida

De 600 pessoas, salvar 200 e deixar morrer 400 são a mesma frase dita de dois jeitos.
A aritmética é idêntica, o resultado é idêntico, a incerteza é idêntica.

O que muda é o **ponto de referência**.

No primeiro par, a redação parte de 600 mortos como cenário inicial, e as opções
aparecem como **ganhos**: quantas pessoas eu consigo salvar. Diante de ganhos, as
pessoas evitam risco e preferem o certo.

No segundo par, a redação parte de 600 vivos, e as opções aparecem como **perdas**:
quantas pessoas eu vou perder. Diante de perdas, as pessoas aceitam risco para tentar
escapar delas.

A escolha não muda porque a informação mudou. Muda porque o **lugar de onde se olha**
mudou.

E, ao contrário de boa parte da psicologia dos anos 1970 e 1980, este efeito
**sobreviveu à replicação em massa**.

---

### 🧮 O Modelo

Um experimento clássico de 1981 e duas replicações coordenadas com dezenas de
laboratórios.

**Fontes:**
- Tversky, A. & Kahneman, D., 1981 · "The Framing of Decisions and the Psychology of
  Choice" · *Science* 211(4481), 453-458 · questionários em sala de aula, com
  estudantes de Stanford e da University of British Columbia · **n=152 no
  enquadramento de ganho** e **n=155 no de perda**, total de **307** ·
  **enquadramento de ganho: 72% escolheram o programa certo** (A) e 28% o arriscado ·
  **enquadramento de perda: 22% escolheram o certo** (C) e **78% o arriscado** (D)
- Klein, R. A. e colegas, 2014 · "Investigating Variation in Replicability: A 'Many
  Labs' Replication Project" · *Social Psychology* 45(3) · **13 efeitos, 36 amostras
  independentes, N=6.344** · o efeito de enquadramento foi um dos replicados, com a
  reversão de escolhas aparecendo em **31 dos 36 laboratórios**, e **nenhum dos 5
  restantes mostrou efeito na direção oposta**
- Klein, R. A. e colegas, 2018 · Many Labs 2 · efeito agrupado de **55 estimativas**:
  **g = 0,44, IC 95% [0,38 · 0,50]**
- **Contrapeso:** meta-análises anteriores (Kühberger, 1998; Piñon & Gambara, 2005)
  situam o efeito médio na faixa de pequeno a médio, com **variação ampla conforme
  domínio, numeracia e o quanto a decisão é de fato de alto risco**

**Nota metodológica sobre o fator ×0.80:** não se aplica. Experimento com escolha
binária entre opções descritas, e não proporção de survey autorrelatado.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Tversky & Kahneman, 1981, Science 211(4481), 453-458

n_ganho  = 152
n_perda  = 155
n_total_1981 = n_ganho + n_perda

# Proporcao que escolheu a opcao CERTA (nao arriscada) em cada enquadramento
p_certo_ganho = 0.72     # programa A: "200 serao salvas"
p_certo_perda = 0.22     # programa C: "400 morrerao"

# Many Labs 1 - Klein et al., 2014, Social Psychology 45(3)
n_efeitos_ml1   = 13
n_laboratorios  = 36
n_participantes_ml1 = 6_344
labs_com_reversao = 31
labs_efeito_oposto = 0

# Many Labs 2 - Klein et al., 2018
n_estimativas_ml2 = 55
g_ml2 = 0.44
ic_ml2 = (0.38, 0.50)

aplica_fator_080 = False

print("=" * 70)
print("  DADOS - O MESMO PROBLEMA, DUAS REDACOES")
print("=" * 70)
print(f"\n  Tversky & Kahneman, 1981 (N total = {n_total_1981}):")
print(f"  -> Enquadramento de GANHO (n={n_ganho}): '200 serao salvas'")
print(f"     {p_certo_ganho*100:.0f}% escolheram a opcao CERTA")
print(f"  -> Enquadramento de PERDA (n={n_perda}): '400 morrerao'")
print(f"     {p_certo_perda*100:.0f}% escolheram a opcao CERTA")
print(f"  -> As opcoes sao ARITMETICAMENTE IDENTICAS nos dois enquadramentos.")
print(f"\n  Many Labs 1 (Klein et al., 2014):")
print(f"  -> {n_efeitos_ml1} efeitos testados · {n_laboratorios} amostras"
      f" independentes · N={n_participantes_ml1:,}".replace(",", "."))
print(f"  -> Reversao encontrada em {labs_com_reversao} de {n_laboratorios} laboratorios")
print(f"  -> Laboratorios com efeito na direcao OPOSTA: {labs_efeito_oposto}")
print(f"\n  Many Labs 2 (Klein et al., 2018):")
print(f"  -> {n_estimativas_ml2} estimativas · efeito agrupado g = {g_ml2:.2f}"
      f"  IC 95% [{ic_ml2[0]:.2f} · {ic_ml2[1]:.2f}]")
print(f"\n  Fator x0.80 aplicado: {aplica_fator_080}")
print("=" * 70)
```

```python
# --- O MODELO ---

# 1) A inversao, em pessoas
escolheram_certo_ganho = round(p_certo_ganho * n_ganho)
escolheram_certo_perda = round(p_certo_perda * n_perda)
tabela = np.array([
    [escolheram_certo_ganho, n_ganho - escolheram_certo_ganho],
    [escolheram_certo_perda, n_perda - escolheram_certo_perda],
])
qui2, p_qui2, gl, _ = stats.chi2_contingency(tabela, correction=False)

# Razao de chances de escolher a opcao certa, ganho contra perda
a, b = tabela[0]
c, d = tabela[1]
odds = (a * d) / (b * c)
Z = stats.norm.ppf(0.975)
ep_log = np.sqrt(1/a + 1/b + 1/c + 1/d)
ic_odds = (np.exp(np.log(odds) - Z*ep_log), np.exp(np.log(odds) + Z*ep_log))

# Diferenca em pontos percentuais
dif_pp = (p_certo_ganho - p_certo_perda) * 100

# 2) ASSINATURA: se nao houvesse efeito, qual a chance de 31 de 36 laboratorios
#    encontrarem a reversao? Sob a hipotese nula, a direcao de cada laboratorio
#    seria uma moeda honesta.
p_moeda = 0.5
p_31_ou_mais = stats.binom.sf(labs_com_reversao - 1, n_laboratorios, p_moeda)
um_em = 1 / p_31_ou_mais

# 3) O que significa g = 0,44
def prob_superioridade(d):
    return stats.norm.cdf(d / np.sqrt(2))


print("=" * 70)
print("  MODELO - A INVERSAO E A SUA ROBUSTEZ")
print("=" * 70)
print(f"\n  1) A inversao no estudo original:")
print(f"  -> Ganho: {escolheram_certo_ganho} de {n_ganho} escolheram o certo"
      f" ({p_certo_ganho*100:.0f}%)")
print(f"  -> Perda: {escolheram_certo_perda} de {n_perda} escolheram o certo"
      f" ({p_certo_perda*100:.0f}%)")
print(f"  -> Diferenca: {dif_pp:.0f} pontos percentuais")
print(f"  -> Razao de chances: {odds:.1f}"
      f"  IC 95% [{ic_odds[0]:.1f} · {ic_odds[1]:.1f}]")
print(f"  -> Qui-quadrado = {qui2:.1f} (gl={gl}), p = {p_qui2:.2e}")
print(f"  -> Trocar a redacao multiplica por {odds:.0f} a chance de")
print(f"     alguem preferir a opcao segura. Com o mesmo fato.")
print(f"\n  2) ASSINATURA: a robustez entre laboratorios")
print(f"  -> {labs_com_reversao} de {n_laboratorios} laboratorios encontraram a reversao")
print(f"  -> Se nao houvesse efeito algum, a direcao de cada laboratorio")
print(f"     seria uma moeda honesta. A chance de sair {labs_com_reversao}")
print(f"     ou mais em {n_laboratorios} seria de {p_31_ou_mais:.2e},")
print(f"     ou 1 em {um_em:,.0f}".replace(",", ".") + ".")
print(f"  -> Isso NAO e um teste formal: e uma leitura de plausibilidade,")
print(f"     porque os laboratorios nao sao independentes do protocolo comum.")
print(f"  -> E nenhum dos {n_laboratorios - labs_com_reversao} restantes")
print(f"     encontrou efeito na direcao oposta.")
print(f"\n  3) O tamanho do efeito, em Many Labs 2:")
print(f"  -> g = {g_ml2:.2f}  IC 95% [{ic_ml2[0]:.2f} · {ic_ml2[1]:.2f}]")
print(f"  -> Isso equivale a {prob_superioridade(g_ml2)*100:.0f}% de chance de")
print(f"     uma pessoa do grupo de perda ser mais arriscada que uma do grupo")
print(f"     de ganho. Se nao houvesse efeito, seriam 50%.")
print(f"  -> E um efeito PEQUENO A MEDIO, nao gigante.")
print(f"     A robustez esta na CONSISTENCIA, nao na magnitude.")
print("=" * 70)
```

```python
# --- VISUALIZACAO ---

DOURADO = '#c8a84b'
VERMELHO = '#c0392b'
VERDE = '#2a8a82'
CINZA = '#6b6a64'

# GRAFICO 1 - A inversao
fig1, ax1 = plt.subplots(figsize=(12, 8))

x = np.arange(2)
larg = 0.38
certo = [p_certo_ganho * 100, p_certo_perda * 100]
arriscado = [100 - v for v in certo]

ax1.bar(x - larg/2, certo, larg, color=VERDE, alpha=0.9, label='escolheu a opção CERTA')
ax1.bar(x + larg/2, arriscado, larg, color=VERMELHO, alpha=0.9,
        label='escolheu a opção ARRISCADA')

for xi, (c, r) in enumerate(zip(certo, arriscado)):
    ax1.text(xi - larg/2, c + 1.5, f'{c:.0f}%', ha='center', fontsize=21,
             fontweight='bold', color=VERDE)
    ax1.text(xi + larg/2, r + 1.5, f'{r:.0f}%', ha='center', fontsize=21,
             fontweight='bold', color=VERMELHO)

ax1.set_xticks(x)
ax1.set_xticklabels([f'GANHO\n"200 serão salvas"\n(n={n_ganho})',
                     f'PERDA\n"400 morrerão"\n(n={n_perda})'], fontsize=13)
ax1.set_ylim(0, 100)
ax1.set_ylabel('Proporção de participantes (%)')
ax1.legend(frameon=False, fontsize=13, loc='upper center', ncol=2)
ax1.set_title('As mesmas opções, duas redações, escolhas invertidas\n'
              'Tversky & Kahneman, 1981, Science 211(4481), 453-458',
              fontsize=15, pad=18)
ax1.text(0.5, -0.16,
         f'De 600 pessoas, "200 serão salvas" e "400 morrerão" são a mesma frase. '
         f'A diferença é de {dif_pp:.0f} pontos percentuais.',
         transform=ax1.transAxes, ha='center', fontsize=12, color=DOURADO,
         fontweight='bold')

plt.figtext(0.5, 0.005,
            'Fonte: Tversky & Kahneman, 1981  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-081-grafico-01-inversao.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 1 salvo")

# GRAFICO 2 - Assinatura: 31 de 36 laboratorios
fig2, ax2 = plt.subplots(figsize=(12, 8))

lado = 6
for i in range(n_laboratorios):
    linha, col = divmod(i, lado)
    achou = i < labs_com_reversao
    cor = VERDE if achou else CINZA
    alpha = 0.9 if achou else 0.28
    ax2.add_patch(plt.Rectangle((col, -linha), 0.86, 0.86, facecolor=cor,
                                alpha=alpha, edgecolor='white', linewidth=2.5))

ax2.set_xlim(-0.3, lado + 0.3)
ax2.set_ylim(-lado - 1.4, 1.6)
ax2.set_xticks([]); ax2.set_yticks([])
ax2.grid(False)
for spine in ax2.spines.values():
    spine.set_visible(False)

ax2.text(lado/2, 1.25,
         f'{labs_com_reversao} de {n_laboratorios} laboratórios encontraram a inversão',
         ha='center', fontsize=17, fontweight='bold', color=VERDE)
ax2.set_title('A assinatura estatística: replicação em 36 amostras independentes\n'
              'Many Labs 1 · Klein et al., 2014 · N='
              + f'{n_participantes_ml1:,}'.replace(",", "."),
              fontsize=15, pad=18)
ax2.text(lado/2, -lado - 0.25,
         f'E nenhum dos {n_laboratorios - labs_com_reversao} restantes encontrou efeito '
         f'na direção oposta.\n'
         f'Se não houvesse efeito e cada laboratório fosse uma moeda honesta, '
         f'a chance de sair {labs_com_reversao} ou mais\nseria de 1 em '
         f'{um_em:,.0f}'.replace(",", ".") +
         '. Leitura de plausibilidade, não teste formal.',
         ha='center', fontsize=11, color=CINZA, style='italic')

plt.figtext(0.5, 0.005,
            'Fonte: Klein et al., 2014, Social Psychology 45(3)  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-081-grafico-02-laboratorios.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 2 salvo")

# GRAFICO 3 - O tamanho do efeito, e o que ele nao e
fig3, ax3 = plt.subplots(figsize=(12, 8))

ax3.plot([ic_ml2[0], ic_ml2[1]], [0, 0], color=DOURADO, linewidth=6,
         solid_capstyle='round')
for lim in ic_ml2:
    ax3.plot([lim, lim], [-0.06, 0.06], color=DOURADO, linewidth=6)
ax3.scatter([g_ml2], [0], s=420, color=DOURADO, zorder=3)
ax3.text(g_ml2, 0.16, f'g = {g_ml2:.2f}'.replace('.', ','), ha='center', fontsize=24,
         fontweight='bold', color=DOURADO)
ax3.text(g_ml2, -0.19, 'IC 95% [' + f'{ic_ml2[0]:.2f}'.replace('.', ',') + ' · '
         + f'{ic_ml2[1]:.2f}'.replace('.', ',') + ']',
         ha='center', fontsize=14, color=CINZA)

for valor, nome in ((0.20, 'pequeno'), (0.50, 'médio'), (0.80, 'grande')):
    ax3.axvline(x=valor, color=CINZA, linestyle=':', linewidth=1.6, alpha=0.8)
    ax3.text(valor, 0.42, nome, ha='center', fontsize=12, color=CINZA)

ax3.axvline(x=0, color=VERMELHO, linestyle='--', linewidth=2)
ax3.text(0.012, -0.42, 'efeito nenhum', fontsize=12, color=VERMELHO)

ax3.set_xlim(-0.05, 0.95)
ax3.set_ylim(-0.55, 0.55)
ax3.set_yticks([])
ax3.set_xlabel('Tamanho do efeito de enquadramento (g de Hedges)')
ax3.set_title('O efeito é pequeno a médio. A força dele está na consistência.\n'
              f'Many Labs 2 · {n_estimativas_ml2} estimativas',
              fontsize=15, pad=18)
ax3.text(0.5, -0.15,
         f'Um g de {g_ml2:.2f} equivale a '.replace('.', ',') +
         f'{prob_superioridade(g_ml2)*100:.0f}% de chance de alguém do grupo de perda '
         f'ser mais arriscado que alguém do grupo de ganho. Sem efeito, seriam 50%.',
         transform=ax3.transAxes, ha='center', fontsize=12, color=DOURADO,
         fontweight='bold')

plt.figtext(0.5, 0.005,
            'Fonte: Klein et al., 2018, Many Labs 2  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-081-grafico-03-tamanho.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 3 salvo")
```

### 💡 O Insight

**72% escolheram a opção segura quando ela foi descrita como "200 pessoas serão
salvas". 22% escolheram a mesma opção quando ela foi descrita como "400 pessoas
morrerão".**

Cinquenta pontos percentuais de diferença. A razão de chances é de cerca de 9: trocar
a redação multiplica por nove a chance de alguém preferir o caminho seguro.

E não há informação nova em lugar nenhum. De 600 pessoas, salvar 200 e perder 400 são
a mesma coisa dita de dois jeitos.

Isso é de 1981, e a essa altura você já deve estar desconfiada, porque este projeto
passou o mês inteiro derrubando clássicos dessa mesma época. A esteira hedônica caiu
ontem. O esgotamento do ego caiu no #067. A endorfina caiu no #068.

Esse não caiu.

Em 2014, um projeto coordenado testou treze efeitos clássicos em **36 amostras
independentes**, com **6.344 pessoas**. A inversão do enquadramento apareceu em
**31 dos 36 laboratórios**. E **nenhum dos cinco restantes encontrou efeito na direção
oposta.**

Se não houvesse efeito nenhum, e a direção de cada laboratório fosse uma moeda honesta,
a chance de sair 31 ou mais em 36 seria de uma em milhões.

Agora, a parte que quase nenhum texto sobre viés cognitivo conta.

**O efeito é pequeno a médio, não gigante.** No Many Labs 2, com 55 estimativas, o
efeito agrupado é **g = 0,44**, com intervalo de 0,38 a 0,50. Isso equivale a cerca de
**62% de chance** de uma pessoa do grupo de perda ser mais arriscada que uma do grupo
de ganho. Sem efeito nenhum, seriam 50%.

E meta-análises anteriores registram variação ampla conforme o domínio, a numeracia da
pessoa e o quanto a decisão é de fato de alto risco.

Ou seja: a força desse achado não está na magnitude. Está na **consistência**. Ele é
modesto e aparece em quase todo lugar onde se procura, o que é bem mais impressionante
que um efeito grande num laboratório só.

O que eu levo daqui não é a lista de vieses.

É que **quem escreve a pergunta participa da resposta**, e quase sempre sem intenção
nenhuma. A pessoa que redigiu o relatório escolheu falar em taxa de sucesso ou em taxa
de falha. Quem montou o slide escolheu mostrar o que se ganha ou o que se perde. Quem
te contou a notícia escolheu de onde começar a contar.

Nenhum deles precisa estar te manipulando. Basta que tenha escolhido um ponto de
partida, porque não existe redação sem ponto de partida.

A defesa possível é chata e funciona: antes de decidir, reescreva a opção do outro
lado. Se a sua escolha mudar, ela não era sobre o fato.

*Qual decisão sua deste ano você tomaria diferente se alguém tivesse escrito a mesma
coisa ao contrário?*

### ⚠️ Limitações do Modelo

- **O efeito é pequeno a médio.** g = 0,44 [0,38 · 0,50]. Textos de divulgação
  costumam apresentá-lo como se qualquer pessoa fosse virar a escolha com qualquer
  redação, e não é isso que os números dizem.
- **Variação ampla por contexto.** Meta-análises anteriores registram que o efeito
  muda bastante conforme domínio, numeracia e o quanto a decisão é de alto risco. O
  problema da doença é hipotético e sem consequência para quem responde.
- **O cálculo binomial dos 31 laboratórios é ilustrativo, não teste formal.** Os
  laboratórios seguem um protocolo comum e não são independentes nesse sentido.
- **A tabela do estudo de 1981 foi reconstruída** a partir das proporções e dos N
  publicados por condição. As contagens absolutas exatas não constam nas fontes
  acessíveis, e por isso o qui-quadrado e a razão de chances devem ser lidos como
  ordem de grandeza.
- **Amostra original de estudantes universitários** de duas universidades, em 1981,
  respondendo questionário em sala de aula.
- **Enquadramento não é sinônimo de manipulação.** Toda descrição tem um ponto de
  partida, inclusive esta. Não existe versão neutra de uma frase.
- Fator ×0.80 não aplicado: escolha binária entre opções descritas.

*A ciência é honesta sobre o que não sabe. O modelo também.*
