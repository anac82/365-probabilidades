## Qual a probabilidade de quem ganha mais sentir mais falta de tempo?

**Tipo:** Trabalho · Trilha do tempo, dia 3 de 24
**Data de publicação:** 2026-10-30
**Ferramenta:** Python
**Decisão analisada:** A sensação de falta de tempo vem das horas que eu trabalho ou de outra coisa?
**Hashtag:** #365Probabilidades #Dia139 #ProbabilidadesDoTempo

---

### 📖 A História

Tem um paradoxo que quase todo mundo conhece de perto. A pessoa ganha mais do que ganhava
dez anos atrás, trabalha mais ou menos as mesmas horas, e sente muito mais pressa.

A explicação comum é que quem ganha mais trabalha mais. Mas e se não for isso? E se, com as
mesmas horas de trabalho e de casa, o dinheiro sozinho aumentasse a pressa?

Este é o terceiro dia da trilha do tempo. O primeiro mediu a desculpa. O segundo mediu a
previsão. Este mede a pressa. Fui ver se alguém separou, em números, o peso das horas e o
peso da renda na sensação de falta de tempo.

---

### 📚 O Conceito: a pressa de quem tem o que fazer com o tempo

Dois economistas, Daniel Hamermesh e Jungmin Lee, pegaram pesquisas domiciliares de cinco
países que fazem a mesma pergunta: "com que frequência você se sente com pressa ou
pressionado pelo tempo?". E tinham, para as mesmas pessoas, as horas de trabalho fora e
dentro de casa e a renda da família.

A ideia deles é econômica. Tempo e dinheiro são os dois insumos de tudo o que a gente faz.
Quem tem mais dinheiro tem mais coisas que valem a pena fazer, e o tempo não cresce junto.
A pressa não é só falta de tempo. É excesso de coisas boas para fazer nele. Eles chamam a
hipótese, com ironia, de "yuppie kvetch", a lamúria do yuppie.

O teste é simples. Se a pressa vem só das horas, a renda não deveria importar depois de
descontar as horas. Se a renda importa mesmo assim, o dinheiro compra pressa.

---

### 🧮 O Modelo

Quatro camadas. A fonte dá as proporções e os coeficientes. As contas são nossas.

1. **Quem vive com pressa.** A proporção de pessoas que se sentem sempre ou quase sempre
   com pressa, em casais com dois salários, por país e por sexo. Com o fator ×0,80.
2. **O peso das horas.** Quanto cada hora semanal de trabalho a mais sobe a chance de
   pressa, pelos coeficientes publicados.
3. **O peso do dinheiro, com as mesmas horas.** Quanto a renda a mais sobe a chance de
   pressa depois de descontar as horas, nos oito grupos (quatro países, dois sexos). E a
   conversão: quantas horas de trabalho a mais produzem a mesma pressa que a renda a mais.
4. **A pressa dela.** A decomposição por sexo: quanta pressa a mulher teria com as horas
   do marido, e vice-versa.

**Fonte central:**
- Hamermesh, D. S. & Lee, J. (2007). "Stressed out on four continents: time crunch or
  yuppie kvetch?". *Review of Economics and Statistics* 89(2), 374-383 (lido na versão NBER
  Working Paper 10186, 2003). Cinco pesquisas domiciliares: **Austrália** (HILDA 2001,
  13.969 adultos em 7.682 domicílios), **Alemanha** (SOEP 2002, cerca de 12 mil
  domicílios), **Coreia** (Pesquisa de Uso do Tempo 1999, cerca de 17 mil domicílios, com
  diário), **Canadá** (GSS 1998, 1.268 homens e 1.176 mulheres na regressão) e **Estados
  Unidos** (Quality of American Life 1971 e 1978, 1.123 homens e 1.276 mulheres). Pergunta
  do tipo "com que frequência você se sente com pressa?". Em casais com dois salários,
  sempre ou quase sempre com pressa: Austrália 44,9% dos homens e 55,1% das mulheres;
  Alemanha 37,9% e 42,4%; Canadá ("sempre") 54,0% e 59,1%; Coreia ("sempre") 33,4% e
  41,7%; EUA ("sempre") 23,7% e 28,1%. Probits com horas de trabalho fora e em casa,
  renda familiar, idade, filhos, saúde. Coeficientes de renda positivos nos quatro
  países com dado recente; significativos em todos menos homens australianos de casais
  com dois salários. Decomposição por sexo (Tabela 12): na Austrália, a mulher com as
  características do marido teria 67,5% de pressa, contra 44,7% dele.

**Nota metodológica.** O fator ×0,80 do projeto se aplica às proporções de "sempre com
pressa", que são autorrelato de survey, e entra em todas elas. Não se aplica aos
coeficientes de horas e de renda, que são efeitos estimados em regressão. As amostras das
subpopulações de casais com dois salários não foram lidas para Austrália, Alemanha e
Coreia; os intervalos são calculados só onde o N foi lido. Cordas a declarar: #025
(comprar tempo, Whillans) e #153 (ocupado como status, Bellezza), com outra pergunta.
Os dados têm de 24 a 55 anos.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

SEMENTE = 42
rng = np.random.default_rng(SEMENTE)

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

VERMELHO, DOURADO, VERDE, CINZA = '#c0392b', '#c8a84b', '#1a5f5a', '#6b6a64'
RODAPE = ('Fonte: Hamermesh & Lee (2007), Review of Economics and Statistics 89(2) · cinco pesquisas domiciliares · '
          'cálculo próprio | #365Probabilidades')

def mil(n):
    return f"{int(round(n)):,}".replace(",", ".")

def vir(x, casas=1):
    return f"{x:.{casas}f}".replace(".", ",")

def jeffreys(k, n):
    return stats.beta(k + 0.5, n - k + 0.5)

print("✅ Bibliotecas carregadas · semente", SEMENTE)
```

```python
# --- DADOS DA LITERATURA ---
# Hamermesh & Lee (2007). Casais com dois salários, salvo indicação.
FATOR = 0.80

# Proporção "sempre ou quase sempre com pressa" (homens, mulheres), e N da regressão quando lido
PRESSA = {
    'Austrália':      {'h': 0.4493, 'm': 0.5508, 'n': None, 'medida': 'quase sempre ou muitas vezes'},
    'Alemanha':       {'h': 0.3785, 'm': 0.4239, 'n': None, 'medida': 'sempre ou muitas vezes'},
    'Canadá':         {'h': 0.5399, 'm': 0.5906, 'n': (654, 748), 'medida': 'todo dia'},
    'Coreia':         {'h': 0.3337, 'm': 0.4168, 'n': None, 'medida': 'sempre'},
    'Estados Unidos': {'h': 0.2370, 'm': 0.2812, 'n': (481, 601), 'medida': 'sempre'},
}

# Efeito de uma hora semanal de trabalho a mais (pontos de probabilidade), casais com dois salários
HORAS = {'Austrália': (0.00833, 0.00801), 'Alemanha': (0.0106, 0.0099),
         'Canadá': (0.0287 / 5, 0.0235 / 5), 'Coreia': (0.0368 / 5, 0.0544 / 5)}   # Canadá e Coreia: por hora diária, dividido por 5

# Efeito da renda familiar, por 1.000 unidades da moeda local por ano, com as mesmas horas
RENDA = {'Austrália': (-0.00004, 0.00078), 'Alemanha': (0.00053, 0.00057),
         'Canadá': (0.00107, 0.00070), 'Coreia': (0.00271, 0.00793)}   # Coreia: por milhão de won (renda imputada)
SIGNIF = {'Austrália': (False, True), 'Alemanha': (True, True), 'Canadá': (True, True), 'Coreia': (True, True)}
DEGRAU = {'Austrália': 'A$ 50 mil', 'Alemanha': '€ 50 mil', 'Canadá': 'C$ 50 mil', 'Coreia': '50 milhões de won'}
RENDA_A_MAIS = 50   # 50 unidades do coeficiente: 50 mil na moeda local, ou 50 milhões de won (cerca de US$ 40 mil em 1999)

# Decomposição por sexo (Tabela 12): fração prevista com pressa
DECOMP = {  # (coef homens, carac homens), (coef homens, carac mulheres), (coef mulheres, carac homens), (coef mulheres, carac mulheres)
    'Austrália': (0.447, 0.387, 0.675, 0.560),
    'Alemanha':  (0.371, 0.230, 0.528, 0.420),
    'Coreia':    (0.324, 0.368, 0.422, 0.408),
}

print("=" * 70)
print("  DADOS · HAMERMESH & LEE (2007), CASAIS COM DOIS SALÁRIOS")
print("=" * 70)
print(f"\n  {'país':<16} {'homens':>8} {'mulheres':>9}   medida")
for p, d in PRESSA.items():
    print(f"  {p:<16} {vir(d['h'] * 100):>7}% {vir(d['m'] * 100):>8}%   {d['medida']}")
print(f"\n  Efeito de 10 horas semanais a mais (pontos): " + " · ".join(f"{p} {vir(h * 1000, 0)}/{vir(m * 1000, 0)}" for p, (h, m) in HORAS.items()))
print(f"  Efeito de cerca de 50 mil a mais de renda anual, mesmas horas (pontos): " +
      " · ".join(f"{p} {vir(h * RENDA_A_MAIS * 100)}/{vir(m * RENDA_A_MAIS * 100)}" for p, (h, m) in RENDA.items()))
print("=" * 70)
```

```python
# --- O MODELO ---

# CAMADA 1 · Quem vive com pressa
print("=" * 70)
print("  CAMADA 1 · QUEM VIVE COM PRESSA (com o fator ×0,80)")
print("=" * 70)
print(f"\n  {'país':<16} {'homens':>8} {'×0,80':>7} {'mulheres':>9} {'×0,80':>7}   IC 95% (quando o N foi lido)")
for p, d in PRESSA.items():
    ic = ''
    if d['n']:
        nh, nm = d['n']
        ih = jeffreys(round(d['h'] * nh), nh).interval(0.95); im = jeffreys(round(d['m'] * nm), nm).interval(0.95)
        ic = f"H [{vir(ih[0] * FATOR * 100)}; {vir(ih[1] * FATOR * 100)}] · M [{vir(im[0] * FATOR * 100)}; {vir(im[1] * FATOR * 100)}]"
    print(f"  {p:<16} {vir(d['h'] * 100):>7}% {vir(d['h'] * FATOR * 100):>6}% {vir(d['m'] * 100):>8}% {vir(d['m'] * FATOR * 100):>6}%   {ic}")
media_h = np.mean([d['h'] for d in PRESSA.values()]); media_m = np.mean([d['m'] for d in PRESSA.values()])
print(f"\n  Média dos cinco países: homens {vir(media_h * 100, 0)}% (×0,80: {vir(media_h * FATOR * 100, 0)}%) · mulheres {vir(media_m * 100, 0)}% (×0,80: {vir(media_m * FATOR * 100, 0)}%)")
print("  Em cinco países e em cinco décadas de pesquisa, a mulher do casal relata mais pressa do que o homem.")

# CAMADA 2 · O peso das horas
print("\n" + "=" * 70)
print("  CAMADA 2 · O PESO DAS HORAS")
print("=" * 70)
for p, (h, m) in HORAS.items():
    print(f"  {p:<12} +10 h por semana = +{vir(h * 1000, 0)} pontos (homens) · +{vir(m * 1000, 0)} pontos (mulheres)")
print("  De 30 para 50 horas semanais, a chance de viver com pressa sobe de 16 a 22 pontos. As horas pesam, e muito.")

# CAMADA 3 · O peso do dinheiro, com as mesmas horas
print("\n" + "=" * 70)
print(f"  CAMADA 3 · O PESO DO DINHEIRO, COM AS MESMAS HORAS (cerca de 50 mil a mais por ano)")
print("=" * 70)
positivos = 0; signif = 0; conversao = {}
print(f"\n  {'grupo':<24} {'pontos':>7} {'equivale a':>12}")
for p in RENDA:
    for i, sexo in enumerate(('homens', 'mulheres')):
        pts = RENDA[p][i] * RENDA_A_MAIS * 100
        horas_eq = RENDA[p][i] * RENDA_A_MAIS / HORAS[p][i]
        conversao[(p, sexo)] = (pts, horas_eq)
        positivos += pts > 0; signif += SIGNIF[p][i]
        marca = '' if SIGNIF[p][i] else '  (não significativo)'
        print(f"  {p + ', ' + sexo:<24} {vir(pts):>6}  {vir(horas_eq):>5} h/semana{marca}")
print(f"\n  Coeficiente de renda positivo em {positivos} de 8 grupos; significativo em {signif} de 8.")
print(f"  Com as mesmas horas, cerca de 50 mil a mais de renda por ano produzem a pressa de 2 a 9 horas semanais a mais de trabalho (Austrália, Alemanha, Canadá).")
print("  Na Coreia, com renda imputada, o efeito é muito maior e menos confiável; fica declarado e fora da frase de capa.")
print("  Os Estados Unidos (dados de 1971 e 1978) ficam fora da conversão: só o coeficiente das mulheres é significativo lá.")

# CAMADA 4 · A pressa dela
print("\n" + "=" * 70)
print("  CAMADA 4 · A PRESSA DELA (decomposição por sexo)")
print("=" * 70)
for p, (hh, hm, mh, mm) in DECOMP.items():
    print(f"\n  {p}: homem com a vida dele {vir(hh * 100, 0)}% · mulher com a vida dela {vir(mm * 100, 0)}%")
    print(f"    mulher com as horas e a renda do marido: {vir(mh * 100, 0)}% · homem com as horas e a renda da mulher: {vir(hm * 100, 0)}%")
print("\n  Na Austrália e na Alemanha, a mulher com a vida do marido teria MAIS pressa do que tem, e o marido com a vida dela teria MENOS.")
print("  O que sobra não está nas horas medidas. Os autores apontam a gerência da casa, que o diário não captura.")
print("=" * 70)
```

```python
# --- VISUALIZAÇÃO ---

# GRÁFICO 1 · Quem vive com pressa
fig1, ax1 = plt.subplots(figsize=(12, 8))
paises = list(PRESSA.keys()); x = np.arange(len(paises))
h = np.array([PRESSA[p]['h'] for p in paises]) * 100; m = np.array([PRESSA[p]['m'] for p in paises]) * 100
ax1.bar(x - 0.2, h, width=0.38, color=CINZA, alpha=0.35)
ax1.bar(x - 0.2, h * FATOR, width=0.38, color=CINZA, alpha=0.95, label='homens (cheio: ×0,80; claro: publicado)')
ax1.bar(x + 0.2, m, width=0.38, color=VERMELHO, alpha=0.3)
ax1.bar(x + 0.2, m * FATOR, width=0.38, color=VERMELHO, alpha=0.95, label='mulheres (cheio: ×0,80; claro: publicado)')
for i in range(len(paises)):
    ax1.text(i - 0.2, h[i] + 1, f"{vir(h[i], 0)}%", ha='center', fontsize=12, color='#444')
    ax1.text(i + 0.2, m[i] + 1, f"{vir(m[i], 0)}%", ha='center', fontsize=12, fontweight='bold', color=VERMELHO)
ax1.set_xticks(x); ax1.set_xticklabels([f"{p}\n{PRESSA[p]['medida']}" for p in paises], fontsize=9.5)
ax1.set_ylim(0, 70)
ax1.set_ylabel('Sempre ou quase sempre com pressa (%)', fontsize=12)
ax1.set_title('Em casais com dois salários, a mulher relata mais pressa em todos os países\n'
              'cinco pesquisas domiciliares, de 1971 a 2002', fontsize=14, pad=16)
ax1.legend(frameon=False, fontsize=11, loc='upper right')
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-139-grafico-01-quem-vive-com-pressa.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# GRÁFICO 2 · Horas e dinheiro na mesma régua
fig2, ax2 = plt.subplots(figsize=(12, 8))
grupos = [(p, s) for p in ('Austrália', 'Alemanha', 'Canadá') for s in ('homens', 'mulheres')]   # Coreia fica fora dos gráficos: renda imputada
rot = [f"{p}\n{s}" for p, s in grupos]
pts_renda = [conversao[g][0] for g in grupos]
pts_horas = [HORAS[p][i] * 500 for p, s in grupos for i in [0 if s == 'homens' else 1]]
y = np.arange(len(grupos))
ax2.barh(y - 0.2, pts_horas, height=0.38, color=VERDE, alpha=0.9, label='+5 horas de trabalho por semana')
ax2.barh(y + 0.2, pts_renda, height=0.38, color=DOURADO, alpha=0.95, label='cerca de +50 mil de renda por ano, mesmas horas')
for i, g in enumerate(grupos):
    ax2.text(pts_horas[i] + 0.15, y[i] - 0.2, f"{vir(pts_horas[i])}", va='center', fontsize=11, color=VERDE)
    ax2.text(max(pts_renda[i], 0) + 0.15, y[i] + 0.2, f"{vir(pts_renda[i])}" + ('' if SIGNIF[g[0]][0 if g[1] == 'homens' else 1] else ' (n.s.)'),
             va='center', fontsize=11, color='#6b5a1e', fontweight='bold')
ax2.set_yticks(y); ax2.set_yticklabels(rot, fontsize=11); ax2.invert_yaxis()
ax2.set_xlabel('Pontos a mais na chance de viver com pressa', fontsize=12)
ax2.set_xlim(-0.8, 11)
ax2.set_title('Horas e dinheiro na mesma régua\n'
              'o que cinco horas a mais de trabalho fazem com a pressa, e o que o dinheiro faz sem mexer nas horas', fontsize=14, pad=16)
ax2.legend(frameon=False, fontsize=11, loc='lower right')
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-139-grafico-02-horas-e-dinheiro.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# GRÁFICO 3 · A conversão: renda em horas de pressa
fig3, ax3 = plt.subplots(figsize=(12, 8))
heq = [conversao[g][1] for g in grupos]
cores3 = [VERMELHO if s == 'mulheres' else CINZA for p, s in grupos]
bars = ax3.bar(range(len(grupos)), heq, color=cores3, alpha=0.9, width=0.65)
for i, v in enumerate(heq):
    ax3.text(i, max(v, 0) + 0.15, f"{vir(v)} h", ha='center', fontsize=12, fontweight='bold')
ax3.axhline(0, color='#333', lw=1)
ax3.set_xticks(range(len(grupos))); ax3.set_xticklabels(rot, fontsize=10.5)
ax3.set_ylabel('Horas semanais de trabalho que dão a mesma pressa que a renda a mais', fontsize=11)
ax3.set_ylim(-1, 11)
ax3.set_title('Quanto vale, em pressa, cerca de 50 mil a mais por ano\n'
              'de 2 a 9 horas de trabalho por semana, sem trabalhar uma hora a mais · cinza: homens, vermelho: mulheres', fontsize=14, pad=16)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-139-grafico-03-a-conversao.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")

# GRÁFICO 4 · A pressa dela
fig4, ax4 = plt.subplots(figsize=(12, 8))
pd_ = list(DECOMP.keys()); x = np.arange(len(pd_))
hh = [DECOMP[p][0] * 100 for p in pd_]; mh = [DECOMP[p][2] * 100 for p in pd_]; mm = [DECOMP[p][3] * 100 for p in pd_]
ax4.bar(x - 0.27, hh, width=0.26, color=CINZA, alpha=0.9, label='homem, com a vida dele')
ax4.bar(x, mm, width=0.26, color=VERMELHO, alpha=0.9, label='mulher, com a vida dela')
ax4.bar(x + 0.27, mh, width=0.26, color=DOURADO, alpha=0.95, label='mulher, com as horas e a renda do marido')
for i in range(len(pd_)):
    for dx, v in ((-0.27, hh[i]), (0, mm[i]), (0.27, mh[i])):
        ax4.text(i + dx, v + 1, f"{vir(v, 0)}%", ha='center', fontsize=12, fontweight='bold')
ax4.set_xticks(x); ax4.set_xticklabels(pd_, fontsize=12)
ax4.set_ylim(0, 80)
ax4.set_ylabel('Previsto com pressa (%)', fontsize=12)
ax4.set_title('A pressa dela não está nas horas\n'
              'com as horas e a renda do marido, a mulher teria mais pressa, não menos (Austrália e Alemanha)', fontsize=14, pad=16)
ax4.legend(frameon=False, fontsize=11, loc='upper right')
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-139-grafico-04-a-pressa-dela.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 4 salvo!")
```

### 💡 O Insight

**Com as mesmas horas de trabalho e de casa, mais renda é mais pressa. Em 7 dos 8 grupos
medidos em quatro países, o coeficiente da renda é positivo e significativo. Cerca de
50 mil a mais por ano produzem a pressa de 2 a 9 horas semanais a mais de trabalho, sem
trabalhar uma hora a mais.**

As horas pesam, e pesam muito: 10 horas semanais a mais sobem a chance de viver com pressa
em 8 a 11 pontos. Mas depois de descontar as horas, o dinheiro continua lá. Quem ganha mais
tem mais coisas que valem a pena fazer com o tempo, e o tempo não cresce junto.

E tem a pressa dela. Em casais com dois salários, a mulher relata mais pressa do que o
homem em todos os cinco países. Na Austrália, com as horas e a renda do marido, ela teria
68% de pressa, e não os 56% que relata. O que sobra não está nas horas que o diário mede.

*A sua falta de tempo aumentou quando as suas horas aumentaram, ou quando a sua renda
aumentou?*

---

### ⚠️ Limitações do Modelo

- Os dados têm de 24 a 55 anos (1971 a 2002). A relação pode ter mudado com o trabalho
  remoto e o celular; os autores não têm como saber, e nós também não.
- "Sempre com pressa" é autorrelato, e a pergunta muda de país para país (todo dia, sempre,
  quase sempre). O fator ×0,80 entra em todas as proporções; as comparações entre países
  são aproximadas.
- Os coeficientes de renda são pequenos por unidade; o degrau de 50 mil é escolha nossa
  para dar escala. Na Coreia a renda é imputada e o efeito é muito maior; ficou fora dos
  gráficos e da capa.
- Regressão não é causa. Pode haver algo que faz a pessoa ganhar mais e sentir mais pressa
  ao mesmo tempo, como o tipo de trabalho. A decomposição por sexo é contrafactual
  estatístico, não experimento.
- Os N das subamostras de casais com dois salários só foram lidos para Canadá e Estados
  Unidos; os intervalos aparecem só ali.

*A ciência é honesta sobre o que não sabe. O modelo também.*
