**Decisão analisada:** Posso confiar num diagnóstico que depende de onde alguém traçou a linha?

### 📖 A História

Burnout virou palavra de uso diário.

Aparece em conversa de segunda-feira, em post de LinkedIn, em justificativa de
atestado, em piada de fim de expediente. E vem quase sempre acompanhada de um número
com cara de precisão: tantos por cento dos profissionais de tal área estão em burnout.

Número com cara de precisão é o tipo de coisa que este projeto gosta de abrir.

Porque antes de saber quantas pessoas têm uma condição, é preciso saber o que conta
como ter essa condição. E aí o assunto muda de lugar: sai do cansaço e vai para a
régua.

Alguém foi conferir a régua. O que encontrou foi pior do que se imaginava.

---

### 📚 O Conceito: onde alguém traça a linha

O instrumento mais usado para medir burnout é o Maslach Burnout Inventory. Ele não
devolve um sim ou um não. Devolve **escores contínuos** em três dimensões: exaustão
emocional, despersonalização e realização pessoal.

Para transformar escore em diagnóstico, alguém precisa escolher um **ponto de corte**.
Acima da linha, tem. Abaixo, não tem.

E essa linha não é dada pela natureza. É uma decisão de quem analisa. Alguns estudos
usam valores fixos, outros usam quantis da própria amostra, o que significa que a
prevalência passa a depender de com quem a pessoa foi comparada.

Note a diferença em relação ao #066. Lá, o mesmo dado gerava duas conclusões conforme
se somava ou se separava. Aqui, o mesmo dado gera dezenas de prevalências conforme o
lugar onde a linha é traçada. **É o mesmo problema em outra roupa: o número não vem
sozinho, vem com a decisão que ninguém escreveu no gráfico.**

---

### 🧮 O Modelo

Uma revisão sistemática grande, com desfecho contado em número de definições, não em
tamanho de efeito.

**Fontes:**
- Rotenstein, L. S., Torre, M., Ramos, M. A., Rosales, R. C., Guille, C., Sen, S. &
  Mata, D. A., 2018 · "Prevalence of Burnout Among Physicians: A Systematic Review" ·
  *JAMA* 320(11), 1131-1150 · **182 estudos, 109.628 indivíduos, 45 países**,
  publicados entre 1991 e 2018
- **85,7%** dos estudos (156 de 182) usaram alguma versão do Maslach Burnout Inventory
- **Pelo menos 142 definições únicas** de critério para burnout geral ou para as
  subescalas
- Mesmo entre os que usaram o **mesmo instrumento**: pelo menos **47 definições
  distintas** de burnout geral, e **29, 26 e 26** definições para exaustão emocional,
  despersonalização e baixa realização pessoal
- Proporção de estudos que reportaram cada componente: **67,0%** burnout geral,
  **72,0%** exaustão emocional, **68,1%** despersonalização, **63,2%** baixa
  realização pessoal
- Faixa de prevalência resultante: **de 0% a 80,5%**

**Nota metodológica sobre o fator ×0.80:** não se aplica, e o motivo é interessante.
Trata-se de prevalência autorrelatada por questionário, que normalmente pediria o
fator. Mas o fator corrige um viés de **magnitude**, e o problema aqui é de
**definição**. Aplicar 0,80 sobre uma faixa que vai de 0% a 80,5% não corrigiria nada,
apenas encolheria a bagunça.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

RNG = np.random.default_rng(42)

print("Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Rotenstein, Torre, Ramos, Rosales, Guille, Sen & Mata, 2018
# JAMA 320(11), 1131-1150

n_estudos     = 182
n_individuos  = 109_628
n_paises      = 45
periodo       = (1991, 2018)

p_usaram_mbi  = 156 / n_estudos          # 85,7%

# Definicoes distintas encontradas
def_totais              = 142            # "pelo menos"
def_burnout_geral_mbi   = 47
def_exaustao            = 29
def_despersonalizacao   = 26
def_realizacao          = 26

# Proporcao de estudos que reportou cada componente
reporta = {
    'Burnout geral':          122 / n_estudos,
    'Exaustão emocional':     131 / n_estudos,
    'Despersonalização':      124 / n_estudos,
    'Baixa realização pessoal': 115 / n_estudos,
}

# Faixa de prevalencia resultante - FAIXA, nao estimativa pontual
prev_min, prev_max = 0.00, 0.805

aplica_fator_080 = False

print("=" * 70)
print("  DADOS - QUANTAS REGUAS EXISTEM PARA A MESMA COISA")
print("=" * 70)
print(f"\n  Rotenstein et al., 2018 (JAMA 320(11), 1131-1150):")
print(f"  -> {n_estudos} estudos · {n_individuos:,}".replace(",", ".") +
      f" individuos · {n_paises} paises · {periodo[0]} a {periodo[1]}")
print(f"  -> {p_usaram_mbi*100:.1f}% usaram alguma versao do Maslach Burnout Inventory")
print(f"\n  Definicoes distintas de criterio:")
print(f"  -> No total, pelo menos:                    {def_totais}")
print(f"  -> So para burnout geral, com o MESMO MBI:  {def_burnout_geral_mbi}")
print(f"  -> Exaustao emocional:                      {def_exaustao}")
print(f"  -> Despersonalizacao:                       {def_despersonalizacao}")
print(f"  -> Baixa realizacao pessoal:                {def_realizacao}")
print(f"\n  Proporcao de estudos que reportou cada componente:")
for nome, p in reporta.items():
    print(f"  -> {nome:<26} {p*100:.1f}%")
print(f"\n  Faixa de prevalencia relatada:")
print(f"  -> de {prev_min*100:.0f}% a {prev_max*100:.1f}%")
print(f"  -> FAIXA entre estudos. Nao existe 'a' prevalencia de burnout.")
print(f"\n  Fator x0.80 aplicado: {aplica_fator_080}")
print("=" * 70)
```

```python
# --- O MODELO ---
# Demonstracao do mecanismo: quanto a prevalencia muda SO por mover a linha.
# Populacao SIMULADA e declarada. Nao e estimativa de ninguem.

N_SIM = 100_000
MEDIA_SIM = 22.0      # escore medio de exaustao emocional, valor ilustrativo
DP_SIM = 9.0          # desvio padrao ilustrativo

escores = RNG.normal(MEDIA_SIM, DP_SIM, N_SIM)

# Uma unica populacao. Varios pontos de corte plausiveis na literatura.
cortes = [17, 20, 22, 25, 27, 30]
prevalencias = {c: float((escores >= c).mean()) for c in cortes}

# Amplitude produzida SOMENTE pela escolha do corte
amplitude = max(prevalencias.values()) - min(prevalencias.values())
razao = max(prevalencias.values()) / min(prevalencias.values())

# Quantas combinacoes de criterio sao possiveis com as definicoes encontradas
# (limite superior ilustrativo, so para dar ordem de grandeza)
combinacoes = def_exaustao * def_despersonalizacao * def_realizacao

# O outro caminho: corte por QUANTIL da propria amostra.
# Aqui a prevalencia deixa de depender da pessoa e passa a depender do grupo.
# 'top' = fracao da amostra classificada como tendo burnout, por construcao
tops = [0.25, 0.33, 0.50]

print("=" * 70)
print("  MODELO - A MESMA POPULACAO, VARIAS PREVALENCIAS")
print("=" * 70)
print(f"\n  Populacao SIMULADA e declarada: N = {N_SIM:,}".replace(",", ".") +
      f", media {MEDIA_SIM}, desvio padrao {DP_SIM}, semente 42.")
print(f"  Serve para mostrar o MECANISMO, nao para estimar nada.")
print(f"\n  A MESMA populacao, com pontos de corte diferentes:")
for c, p in prevalencias.items():
    barra = "#" * int(p * 50)
    print(f"  -> corte em {c:>2}: prevalencia de {p*100:>5.1f}%   {barra}")
print(f"\n  -> Amplitude produzida SO pela escolha da linha:"
      f" {amplitude*100:.1f} pontos percentuais")
print(f"  -> Do menor para o maior: {razao:.1f}x")
print(f"  -> Ninguem ficou mais cansado. So mudou onde a linha foi tracada.")
print(f"\n  E quando o corte e por quantil da propria amostra:")
for t in tops:
    print(f"  -> criterio 'quarto/terco/metade superior' com corte no top"
          f" {t*100:.0f}%: prevalencia de {t*100:.0f}% POR CONSTRUCAO")
print(f"     A prevalencia deixa de depender de voce e passa a depender")
print(f"     de com quem voce foi comparado.")
print(f"\n  Ordem de grandeza das combinacoes possiveis de criterio:")
print(f"  -> {def_exaustao} x {def_despersonalizacao} x {def_realizacao}"
      f" = {combinacoes:,}".replace(",", ".") + " combinacoes")
print(f"     (limite superior ilustrativo, as definicoes nao sao independentes)")
print("=" * 70)
```

```python
# --- VISUALIZACAO ---

DOURADO = '#c8a84b'
VERMELHO = '#c0392b'
VERDE = '#2a8a82'
CINZA = '#6b6a64'

# GRAFICO 1 - A faixa, desenhada como faixa
fig1, ax1 = plt.subplots(figsize=(12, 6.5))

ax1.barh([0], [(prev_max - prev_min) * 100], left=prev_min * 100, height=0.3,
         color=VERMELHO, alpha=0.55, edgecolor=VERMELHO, linewidth=2.5)
ax1.text((prev_min + prev_max) * 50, 0.28,
         f'de {prev_min*100:.0f}% a {prev_max*100:.1f}%'.replace('.', ','),
         ha='center', va='bottom', fontsize=24, fontweight='bold', color=VERMELHO)

ax1.set_xlim(0, 100)
ax1.set_ylim(-0.7, 0.85)
ax1.set_yticks([])
ax1.set_xlabel('Prevalência de burnout relatada (%)')
ax1.set_title('Não existe "a" prevalência de burnout\n'
              f'{n_estudos} estudos · {n_individuos:,}'.replace(",", ".") +
              f' pessoas · {n_paises} países',
              fontsize=15, pad=18)
ax1.text(50, -0.45,
         'Faixa entre estudos, não estimativa pontual. A variação vem principalmente\n'
         'da definição adotada, não da população estudada.',
         ha='center', fontsize=12, color=CINZA, style='italic')

plt.figtext(0.5, 0.005,
            'Fonte: Rotenstein et al., 2018, JAMA 320(11), 1131-1150'
            '  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-072-grafico-01-faixa.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 1 salvo")

# GRAFICO 2 - Assinatura: a mesma populacao, varias linhas
fig2, ax2 = plt.subplots(figsize=(12, 8))

x = np.linspace(MEDIA_SIM - 4 * DP_SIM, MEDIA_SIM + 4 * DP_SIM, 1000)
densidade = stats.norm(MEDIA_SIM, DP_SIM).pdf(x)
ax2.plot(x, densidade, color=CINZA, linewidth=2.5)
ax2.fill_between(x, densidade, alpha=0.10, color=CINZA)

paleta = [VERDE, VERDE, DOURADO, DOURADO, VERMELHO, VERMELHO]
for i, ((c, p), cor) in enumerate(zip(prevalencias.items(), paleta)):
    ax2.axvline(x=c, color=cor, linestyle='--', linewidth=2, alpha=0.9)
    altura = 1.16 if i % 2 == 0 else 1.05
    ax2.text(c, densidade.max() * altura, f'{p*100:.0f}%'.replace('.', ','),
             ha='center', fontsize=16, fontweight='bold', color=cor)

ax2.set_xticks(cortes)
ax2.set_xticklabels([str(c) for c in cortes], fontsize=12)
ax2.set_xlim(MEDIA_SIM - 2.6 * DP_SIM, MEDIA_SIM + 2.6 * DP_SIM)

ax2.set_ylim(0, densidade.max() * 1.28)
ax2.set_xlabel('Ponto de corte no escore de exaustão emocional (população simulada)')
ax2.set_ylabel('Densidade')
ax2.set_title('A assinatura estatística: a mesma população, seis prevalências\n'
              f'Amplitude de {amplitude*100:.0f} pontos percentuais produzida '
              f'apenas pela escolha da linha',
              fontsize=15, pad=18)
ax2.text(0.5, -0.135,
         f'População simulada e declarada: N = {N_SIM:,}'.replace(",", ".") +
         f', média {MEDIA_SIM:.0f}, desvio padrão {DP_SIM:.0f}, semente 42. '
         'Serve para mostrar o mecanismo, não para estimar nada.',
         transform=ax2.transAxes, ha='center', fontsize=11, color=CINZA, style='italic')

plt.figtext(0.5, 0.005,
            'Demonstração própria do mecanismo descrito em Rotenstein et al., 2018'
            '  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-072-grafico-02-cortes.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 2 salvo")

# GRAFICO 3 - Quantas definicoes existem
fig3, ax3 = plt.subplots(figsize=(12, 8))

nomes = ['Total de definições\nencontradas',
         'Só para burnout geral,\ncom o MESMO instrumento',
         'Exaustão\nemocional',
         'Despersonalização',
         'Baixa realização\npessoal']
valores = [def_totais, def_burnout_geral_mbi, def_exaustao,
           def_despersonalizacao, def_realizacao]
cores3 = [VERMELHO, DOURADO, CINZA, CINZA, CINZA]

barras = ax3.barh(np.arange(len(valores))[::-1], valores, color=cores3,
                  alpha=0.88, height=0.55)
for barra, valor in zip(barras, valores):
    ax3.text(valor + 2, barra.get_y() + barra.get_height() / 2, f'{valor}',
             va='center', fontsize=19, fontweight='bold')

ax3.set_yticks(np.arange(len(valores))[::-1])
ax3.set_yticklabels(nomes, fontsize=12)
ax3.set_xlim(0, 165)
ax3.set_xlabel('Número de definições distintas identificadas')
ax3.set_title('Quantas réguas existem para a mesma palavra\n'
              f'{n_estudos} estudos revisados, '
              + f'{p_usaram_mbi*100:.1f}'.replace('.', ',') +
              '% usando o mesmo instrumento',
              fontsize=15, pad=18)
ax3.text(0.98, 0.08,
         'Mesmo instrumento, mesma profissão,\n47 formas diferentes de decidir\n'
         'quem tem e quem não tem.',
         transform=ax3.transAxes, ha='right', fontsize=12, color=DOURADO,
         fontweight='bold')

plt.figtext(0.5, 0.005,
            'Fonte: Rotenstein et al., 2018, JAMA 320(11), 1131-1150'
            '  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-072-grafico-03-definicoes.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 3 salvo")
```

### 💡 O Insight

Cento e oitenta e dois estudos. Cento e nove mil seiscentas e vinte e oito pessoas.
Quarenta e cinco países.

**A prevalência de burnout encontrada vai de 0% a 80,5%.**

Não é variação entre profissões, nem entre países, nem entre épocas. É variação
entre **definições**.

Os estudos usaram pelo menos **142 definições distintas** do que conta como burnout. E
o número que eu acho pior: entre os que usaram o **mesmo instrumento**, o Maslach
Burnout Inventory, havia pelo menos **47 formas diferentes** de decidir quem tem e
quem não tem.

Mesmo questionário. Mesmas respostas. Quarenta e sete veredictos possíveis.

O mecanismo é simples de mostrar. Simulei uma população, com escores fixos e semente
declarada, e apliquei seis pontos de corte plausíveis. A prevalência foi de 18% a 71%.

Ninguém ficou mais cansado no meio do caminho. Só mudou onde a linha foi traçada.

E tem uma variante ainda mais desconcertante. Parte dos estudos define burnout pelo
**quantil da própria amostra**: quem está no quarto superior, tem. Nesse desenho, a
prevalência é 25% por construção, independentemente de quão exausto o grupo esteja. Um
hospital inteiro em colapso teria a mesma taxa de um hospital tranquilo, porque a
medida deixa de falar sobre você e passa a falar sobre com quem você foi comparado.

Isso não quer dizer que burnout não exista. As pessoas estão exaustas, e isso é
verificável de muitas outras formas.

Quer dizer outra coisa: **o número que circula não mede a exaustão. Mede a escolha de
quem contou.**

Da próxima vez que alguém te disser que tantos por cento de uma categoria estão em
burnout, a pergunta útil não é se o número é alto. É onde essa pessoa traçou a linha,
e por quê.

*Quantas vezes você já usou um número desses numa conversa sem saber de onde ele veio?*

### ⚠️ Limitações do Modelo

- **A revisão é qualitativa e descritiva, por decisão dos autores.** Justamente por
  causa da heterogeneidade de definições, eles não calcularam uma prevalência
  agregada. Este notebook também não calcula, e isso é o ponto.
- **A simulação do gráfico 2 é ilustrativa e declarada.** População normal com média
  22 e desvio padrão 9, semente 42. Não é estimativa da distribuição real de escores
  de ninguém, e serve apenas para demonstrar o mecanismo descrito no artigo.
- **As combinações possíveis de critério calculadas no notebook são limite superior.**
  As definições não são independentes entre si, e o número serve como ordem de
  grandeza, não como contagem real.
- A revisão cobre **médicos em atividade**, excluindo médicos em formação. Não é
  amostra da população trabalhadora em geral, e generalizar a faixa para outras
  profissões vai além do que o artigo mede.
- Prevalência aferida por questionário autorrelatado, com todos os limites que isso
  implica, embora o problema central do dia seja de definição e não de autorrelato.
- **A falta de consenso não é evidência de que a condição não exista.** É evidência de
  que a medida não está padronizada.
- Fator ×0.80 não aplicado: o fator corrige magnitude, e o problema aqui é definição.

*A ciência é honesta sobre o que não sabe. O modelo também.*
