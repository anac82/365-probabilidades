### 📖 A História

Ontem escrevi sobre a dor que engana sobre o futuro. Hoje quero falar da irmã dela: a preocupação.

Eu passei a vida me preocupando.

Não é figura de linguagem. É um traço meu, de sempre. A cabeça que vai lá na frente montar o cenário ruim antes mesmo de ele existir.

Pior cenário. Sempre o pior cenário. Eu construo ele inteiro, com detalhe, com desdobramento, com plano B e plano C. Passei quase vinte anos no mercado financeiro, onde isso é ofício: modelar o worst case, dimensionar a exposição, se preparar pro tail risk. Fui treinada pra isso. E fiquei boa nisso.

Só que no mercado o pior cenário vem sempre acompanhado de uma coisa: a probabilidade dele. Ninguém dimensiona risco sem calcular quão provável é o evento.

Na vida, eu esqueci de fazer essa segunda parte.

A cabeça apresentava o cenário em alta definição, às três da manhã, com riqueza de detalhes. Mas nunca apresentava a probabilidade. E eu comprava o cenário inteiro, todas as vezes.

Aí você olha pra trás, pra décadas disso, e faz a pergunta óbvia: de tudo que me tirou o sono ao longo da vida, quanto de fato aconteceu?

Muito pouco. E as coisas difíceis que vieram quase nunca foram as que eu tinha ensaiado.

E é sobre isso que a ciência tem algo bem específico a dizer.

### 📚 O Conceito: A Acurácia da Preocupação

Ontem o tema era o viés de durabilidade: a dor superestima **quanto tempo** vamos sofrer.

Hoje é o irmão dele, e o erro é em outro eixo: a preocupação superestima **se aquilo vai acontecer**.

Duração e probabilidade. Duas faces do mesmo problema — a emoção presente fazendo previsões ruins sobre o futuro.

A pergunta parece simples, mas quase ninguém a testa: das coisas específicas que você previu com medo, quantas de fato aconteceram?

Em 2020, Lucas LaFreniere e Michelle Newman decidiram medir isso de forma direta. Pegaram pessoas com Transtorno de Ansiedade Generalizada diagnosticado e pediram que registrassem, ao longo de dez dias, as preocupações que surgissem — quatro vezes por dia, no momento em que apareciam. Depois, acompanharam por trinta dias para verificar, uma a uma, se aquilo tinha realmente acontecido.

Não é um survey perguntando "você acha que se preocupa à toa?". É medição prospectiva: a previsão foi registrada antes, e o desfecho foi conferido depois.

E o mecanismo por trás disso aparece também em amostras muito maiores. Reiss e colegas, em 2019, mediram em 630 pessoas da população geral a relação entre tendência à preocupação e superestimação de risco. Quanto maior a propensão a se preocupar, maior a distorção na estimativa de probabilidade de eventos negativos.

Ou seja: o cenário que a cabeça monta é vívido. A probabilidade que ela atribui a ele é sistematicamente errada.

---

### 🧮 O Modelo

Neste dia, **não aplico o fator de correção ×0.80**. O 91,4% não é uma proporção de survey extrapolada para a população — é resultado direto de medição prospectiva num experimento controlado.

Mas há um limite muito mais importante que qualquer fator de correção, e ele precisa estar claro: a amostra do estudo principal é pequena (N=29) e clínica (pessoas com TAG diagnosticado). Esse número **não** deve ser lido como "91% das suas preocupações vão dar em nada". Ele mostra a direção e a magnitude do fenômeno num grupo específico — e essa direção é sustentada por uma amostra bem maior e não clínica (Reiss et al., N=630).

**Fontes:**
- LaFreniere, L. S. & Newman, M. G. (2020) — "Exposing Worry's Deceit: Percentage of Untrue Worries in Generalized Anxiety Disorder Treatment", Behavior Therapy, 51(3), 413-423 — N=29, registro prospectivo (EMA), 10 dias de coleta + 30 de acompanhamento
- Reiss, N., Warnecke, I., Tolgou, T. et al. (2019) — Frontiers in Psychology — N=630, população geral, relação entre propensão à preocupação e superestimação de risco negativo


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# LaFreniere & Newman (2020), Behavior Therapy 51(3):413-423
# N=29 (TAG diagnosticado) — registro prospectivo de preocupações (EMA)
# 10 dias de coleta, 4x/dia + 30 dias de acompanhamento dos desfechos

n_lafreniere = 29
p_nao_aconteceu = 0.914      # 91,4% das preocupações NÃO se confirmaram
p_aconteceu = 1 - p_nao_aconteceu

# Das que aconteceram: proporção que saiu MELHOR do que o previsto
p_melhor_que_previsto = 0.33  # ~1/3

# Participantes que não tiveram NENHUMA preocupação confirmada
p_zero_confirmadas = 0.25     # ~1 em cada 4

# Reiss et al. (2019), Frontiers in Psychology — N=630, população geral
n_reiss = 630

# NOTA: NÃO se aplica o fator de correção ×0.80 neste dia.
# O 91,4% é medição prospectiva direta, não proporção de survey extrapolada.
# A limitação relevante aqui é de GENERALIZAÇÃO (N pequeno, amostra clínica),
# e ela é declarada explicitamente — não "corrigida" com um multiplicador.

print("=" * 65)
print("  DADOS — A ACURÁCIA DA PREOCUPAÇÃO")
print("=" * 65)
print(f"\n  LaFreniere & Newman (2020) — N={n_lafreniere} (TAG diagnosticado)")
print(f"  Registro prospectivo: 10 dias · acompanhamento: 30 dias")
print(f"\n  → Preocupações que NÃO se confirmaram:  {p_nao_aconteceu*100:.1f}%")
print(f"  → Preocupações que aconteceram:         {p_aconteceu*100:.1f}%")
print(f"  → Das que aconteceram, saíram MELHOR")
print(f"     do que o previsto:                   ~{p_melhor_que_previsto*100:.0f}%")
print(f"  → Participantes com ZERO preocupações")
print(f"     confirmadas no estudo inteiro:       ~{p_zero_confirmadas*100:.0f}%")
print(f"\n  Reiss et al. (2019) — N={n_reiss} (população geral)")
print(f"  → Quanto maior a propensão à preocupação,")
print(f"     maior a superestimação de risco negativo")
print("=" * 65)
print(f"\n  ⚠️  N pequeno e amostra clínica. Este número descreve")
print(f"      ESSE grupo, não a população geral.")
print("=" * 65)
```

```python
# --- O MODELO ---
# O que a cabeça entrega vs. o que faltaria pra virar gestão de risco

# Decomposição do desfecho das preocupações registradas
nao_aconteceu = p_nao_aconteceu
aconteceu_pior_ou_igual = p_aconteceu * (1 - p_melhor_que_previsto)
aconteceu_melhor = p_aconteceu * p_melhor_que_previsto

# "Alarme falso": tudo que não se concretizou como temido
alarme_falso = nao_aconteceu + aconteceu_melhor

print("=" * 65)
print("  MODELO — O CENÁRIO SEM A PROBABILIDADE")
print("=" * 65)
print(f"\n  Decomposição das preocupações registradas:")
print(f"  → Não aconteceram:                       {nao_aconteceu*100:.1f}%")
print(f"  → Aconteceram, mas MELHOR que o previsto: {aconteceu_melhor*100:.1f}%")
print(f"  → Aconteceram como (ou pior que) previsto:{aconteceu_pior_ou_igual*100:.1f}%")
print(f"\n  → 'Alarme falso' total (não veio, ou veio")
print(f"     melhor do que a cabeça montou):        {alarme_falso*100:.1f}%")
print(f"\n  A LIÇÃO DE GESTÃO DE RISCO:")
print(f"  → No mercado: cenário + probabilidade = gestão de risco")
print(f"  → Na cabeça:  cenário sem probabilidade = história de terror")
print(f"\n  A mente entrega o cenário em alta definição.")
print(f"  Nunca entrega o denominador.")
print("=" * 65)
```

```python
# --- VISUALIZAÇÃO ---

cor_nao = '#2a8a82'      # não aconteceu
cor_melhor = '#c8a84b'   # aconteceu, mas melhor
cor_sim = '#c0392b'      # aconteceu como temido
cor_neutra = '#9c9b94'

# ── GRÁFICO 1 — O desfecho das preocupações ──
fig1, ax1 = plt.subplots(figsize=(11, 6.5))

categorias = ['Não aconteceram', 'Aconteceram,\nmas melhor\nque o previsto', 'Aconteceram\ncomo temido']
valores = [nao_aconteceu*100, aconteceu_melhor*100, aconteceu_pior_ou_igual*100]
cores = [cor_nao, cor_melhor, cor_sim]

bars = ax1.bar(categorias, valores, color=cores, alpha=0.88, width=0.5)
ax1.set_ylabel('Proporção das preocupações registradas (%)')
ax1.set_ylim(0, 100)
ax1.set_title('O que aconteceu com as preocupações que tiraram o sono\nLaFreniere & Newman (2020) — N=29, TAG diagnosticado',
              fontsize=13, pad=15)

for bar, v in zip(bars, valores):
    ax1.text(bar.get_x() + bar.get_width()/2, v + 1.5,
             f'{v:.1f}%', ha='center', fontweight='bold', fontsize=15)

plt.figtext(0.5, 0.005,
            'Amostra clínica pequena — descreve este grupo, não a população geral | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-030-grafico-01-desfecho.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 — O cenário sem o denominador ──
fig2, ax2 = plt.subplots(figsize=(11, 6.5))

etapas = ['O que a mente\nENTREGA', 'O que faltaria pra\nvirar gestão de risco']
# Representação conceitual: cenário (100% de vividez) vs probabilidade (ausente)
cenario = [100, 100]
probabilidade = [0, 100]

x = np.arange(len(etapas))
largura = 0.35
ax2.bar(x - largura/2, cenario, largura, label='Cenário (vividez, detalhe)', color=cor_sim, alpha=0.85)
ax2.bar(x + largura/2, probabilidade, largura, label='Probabilidade (o denominador)', color=cor_nao, alpha=0.85)

ax2.set_xticks(x)
ax2.set_xticklabels(etapas)
ax2.set_ylabel('Presença no raciocínio (ilustrativo)')
ax2.set_ylim(0, 115)
ax2.set_yticks([])
ax2.set_title('A metade que falta\nA mente monta o pior cenário — mas nunca calcula a chance dele',
              fontsize=13, pad=15)
ax2.legend(frameon=False, loc='upper left')

ax2.text(0 + largura/2, 4, 'ausente', ha='center', fontsize=11,
         color=cor_nao, fontweight='bold', style='italic')

plt.figtext(0.5, 0.005,
            'Ilustração conceitual — cenário sem probabilidade não é gestão de risco | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-030-grafico-02-denominador.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 — Distribuição Beta: incerteza do 91,4% com N=29 ──
fig3, ax3 = plt.subplots(figsize=(11, 6.5))

# Beta com o N REAL do estudo (29) — mostra honestamente a incerteza
alpha_p = p_nao_aconteceu * n_lafreniere
beta_p = (1 - p_nao_aconteceu) * n_lafreniere
dist = stats.beta(alpha_p, beta_p)
ic = dist.interval(0.95)

x = np.linspace(0.70, 1.0, 1000)
y = dist.pdf(x)

ax3.plot(x*100, y, color=cor_nao, linewidth=2.5)
ax3.fill_between(x*100, y, alpha=0.2, color=cor_nao)
ax3.axvline(x=ic[0]*100, color=cor_melhor, linestyle='--', linewidth=2, alpha=0.9)
ax3.axvline(x=ic[1]*100, color=cor_melhor, linestyle='--', linewidth=2, alpha=0.9)
ax3.axvline(x=p_nao_aconteceu*100, color='#333', linewidth=2)

ax3.text(ic[0]*100 - 0.5, max(y)*0.55,
         f'IC 95%\n[{ic[0]*100:.1f}%, {ic[1]*100:.1f}%]',
         fontsize=10, color='#8a7020', ha='right')

ax3.set_xlabel('Proporção de preocupações que NÃO se confirmaram (%)')
ax3.set_ylabel('Densidade')
ax3.set_title(f'A honestidade do N pequeno\nDistribuição Beta com N={n_lafreniere} — o intervalo é largo, e isso importa',
              fontsize=12, pad=15)

plt.figtext(0.5, 0.005,
            f'IC calculado com o N real do estudo (N={n_lafreniere}) — sem inventar amostra | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-030-grafico-03-bayesiano.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**91,4% das preocupações registradas no estudo não se confirmaram.**

E há um detalhe que torna o número ainda mais desconfortável: das poucas que aconteceram, cerca de um terço saiu melhor do que a pessoa tinha previsto. Ou seja, mesmo quando o evento vinha, o cenário catastrófico que a cabeça montou raramente se realizava por inteiro.

Cerca de um em cada quatro participantes não teve **nenhuma** preocupação confirmada ao longo de todo o estudo. Nenhuma.

Repito o limite, porque ele importa: isso foi medido em 29 pessoas com ansiedade generalizada diagnosticada. Não é uma lei sobre você. Mas a direção do fenômeno se sustenta numa amostra de 630 pessoas da população geral: quanto mais a pessoa se preocupa, mais ela superestima a probabilidade do desfecho ruim.

O que me pegou não foi o 91%. Foi perceber que eu fazia metade do trabalho.

No mercado, a gente monta o pior cenário **e** calcula a probabilidade dele. É isso que transforma medo em gestão de risco. Sem a probabilidade, o cenário é só uma história de terror bem produzida.

Na vida, a minha cabeça sempre entregou o cenário. Nunca entregou a probabilidade. E eu, que passei décadas cobrando esse rigor de planilhas, nunca cobrei da própria insônia.

Tem ainda um achado que vale mencionar: no estudo, quanto maior a proporção de preocupações que não se confirmavam, maior era a redução dos sintomas de ansiedade no tratamento. Perceber o próprio histórico de erro parece ter valor terapêutico. O dado, aqui, não é só curiosidade — é ferramenta.

*Da próxima vez que a sua cabeça te apresentar o pior cenário às três da manhã, você vai comprar a história inteira — ou vai perguntar qual é a probabilidade dela?*

---

### ⚠️ Limitações do Modelo

- O estudo principal tem N=29 e amostra clínica (TAG diagnosticado). O 91,4% descreve esse grupo, não a população geral. Generalizar diretamente seria exatamente o tipo de erro que este projeto combate.
- A verificação de "aconteceu ou não" depende de autorrelato do participante, com margem de interpretação sobre o que conta como a preocupação ter se concretizado.
- Preocupações vagas ou de longuíssimo prazo são difíceis de verificar em 30 dias — o desenho favorece preocupações concretas e de curto prazo.
- A literatura de preocupação e ansiedade tem números virais sem fonte real (o famoso "85% do que você teme nunca acontece, segundo Cornell" **não existe** como estudo). Os dados aqui vêm de papers revisados por pares, com metodologia declarada.
- Preocupação persistente e incapacitante é diferente de preocupação cotidiana, e tem tratamento eficaz. Saber que "a maioria não acontece" não substitui ajuda profissional quando ela é necessária.

*A ciência é honesta sobre o que não sabe. O modelo também.*

*Este é um tema sensível. Se a ansiedade estiver pesando de forma difícil de carregar, falar com um profissional pode ajudar — e não há nada de fraco nisso.*
