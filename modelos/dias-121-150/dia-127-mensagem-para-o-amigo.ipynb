# 365 Probabilidades - Dia #127
## Qual a probabilidade de você mandar a mensagem para aquele amigo?

**Tipo:** Comportamental
**Data de publicação:** 2026-10-18
**Ferramenta:** Python
**Decisão analisada:** Escrever hoje para o amigo com quem você perdeu contato, ou deixar para depois?
**Hashtag:** #365Probabilidades #Dia127

---

### 📖 A História

Tem um nome que aparece de vez em quando. Numa música, numa foto antiga, num lugar
por onde vocês passaram juntos. Você pensa "preciso escrever para ele". Abre a conversa,
que parou há anos. Lê a última mensagem. Fecha.

Não foi briga. Não foi nada. A vida andou, cada um para um lado, e a conversa ficou ali,
parada no meio.

Todo mundo tem um nome desses. A pergunta de hoje é quantas pessoas chegam a apertar
"enviar" quando a chance aparece, e o que muda essa conta.

---

### 📚 O Conceito

Duas pesquisadoras, uma no Canadá e outra na Inglaterra, deram a centenas de pessoas a
chance de escrever para um amigo antigo ali mesmo, na hora, com o texto já rascunhado.
Só uma em cada três mandou. A disposição para escrever ao amigo de quem se gosta é
igual à disposição para puxar conversa com um estranho.

O freio tem nome. As pessoas acham que o amigo não vai querer saber delas, e que vai
ser estranho aparecer depois de tanto tempo. Um segundo grupo de estudos mediu o outro
lado: quem recebe a mensagem gosta mais dela do que quem manda imagina, e o que explica
a diferença é a surpresa. Quem recebe está pensando "que bom que ele lembrou". Quem
manda está pensando no silêncio.

---

### 🧮 O Modelo

**Fontes:**
- Aknin, L. B. & Sandstrom, G. M. (2024). *People are surprisingly hesitant to reach
  out to old friends.* Communications Psychology 2, 34. Sete estudos, N total = 2.458.
  Estudo 1 (N = 401): **90,9%** perderam contato com um amigo de quem ainda gostam.
  Estudo 3 (N = 453): **27,8%** mandaram a mensagem quando tiveram a chance. Estudo 4
  (N = 604, controle mais realista): **36,8%**. Estudo 5 (N = 288): disposição para
  escrever ao amigo antigo igual à de falar com um estranho (médias 4,6 e 4,6 numa
  escala de 7; fator de Bayes a favor da igualdade de 19,5). Estudo 7 (N = 194):
  praticar antes, mandando mensagem para amigos atuais, elevou o envio de **31%** para
  **53%**.
- Liu, P. J., Rim, S., Min, L. & Min, K. E. (2023). *The surprise of reaching out:
  Appreciated more than we think.* Journal of Personality and Social Psychology
  124(4), 754-771. Treze estudos pré-registrados. Estudo 4a (N = 201): quem manda
  prevê apreciação de 4,36 e quem recebe relata 4,81 (escala de 7), d = 0,32.
  Estudo 7 (N = 1.602): a diferença é maior entre laços fracos (5,24 contra 5,92) do
  que entre laços fortes (6,35 contra 6,65). Replicação independente do Estudo 4a
  (Clearer Thinking, N = 742): d = 0,25, na mesma direção.

**Modelo próprio:**
- Posterior de Jeffreys sobre quem manda a mensagem, nos dois estudos e no conjunto.
  As contagens são reconstruídas dos percentuais publicados (declarado).
- Posterior da diferença entre treinar e não treinar (Estudo 7), por simulação.
- Cadeia da chance de hoje: ter um amigo perdido × mandar quando a chance aparece.
- Quatro domingos: a chance de ao menos uma mensagem sair em quatro oportunidades,
  com e sem o treino.
- Probabilidade de superioridade do d de Liu, traduzida em "de cada cem mensagens,
  quantas são mais apreciadas do que quem mandou imaginou".
- Sem ×0,80: o número central é comportamento observado no estudo (a mensagem foi ou
  não enviada), não autorrelato.

**Parentesco a declarar:** os dias #040 e #049 tratam de como a gente subestima o
outro; este dia mede a mensagem que não sai, e cita aqueles dois como antecedentes.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

VERMELHO, DOURADO, VERDE, CINZA = '#c0392b', '#c8a84b', '#1a5f5a', '#6b6a64'
SEMENTE = 42
ger = np.random.default_rng(SEMENTE)

def vir(x, casas=1):
    return f"{x:.{casas}f}".replace(".", ",")

def pct(x, casas=1):
    return vir(100 * x, casas) + "%"

def mil(n):
    return f"{int(round(n)):,}".replace(",", ".")

def beta_j(x, n):
    # posterior de Jeffreys para uma proporcao
    return stats.beta(x + 0.5, n - x + 0.5)

print("✅ Bibliotecas carregadas")
```

```python
# --- OS DADOS ---

# Aknin & Sandstrom 2024, Communications Psychology 2, 34
# contagens reconstruídas dos percentuais publicados (arredondadas ao inteiro)
N1, P1_PERDEU = 401, 0.909            # Estudo 1: perderam contato com amigo de quem gostam
N3, P3_MANDOU = 453, 0.278            # Estudo 3: mandaram a mensagem quando puderam
N4, P4_MANDOU = 604, 0.368            # Estudo 4: idem, controle mais realista
N7_TREINO, P7_TREINO = 101, 0.53      # Estudo 7: praticou antes com amigos atuais
N7_SEM, P7_SEM = 93, 0.31             # Estudo 7: navegou em rede social antes
M5_AMIGO, M5_ESTRANHO, BF01_5 = 4.6, 4.6, 19.52   # Estudo 5: disposição, escala de 7

x1 = round(N1 * P1_PERDEU); x3 = round(N3 * P3_MANDOU); x4 = round(N4 * P4_MANDOU)
x7t = round(N7_TREINO * P7_TREINO); x7s = round(N7_SEM * P7_SEM)

# Liu, Rim, Min & Min 2023, JPSP 124(4), 754-771
LIU_PREV, LIU_PREV_SD = 4.36, 1.31    # Estudo 4a, N = 201: o que quem manda prevê
LIU_RECEB, LIU_RECEB_SD = 4.81, 1.53  # o que quem recebe relata
LIU_D = 0.32
LIU_REPL_D, LIU_REPL_N = 0.25, 742    # replicação independente (Clearer Thinking)
LIU7_FRACO = (5.24, 5.92)             # Estudo 7, N = 1.602: laços fracos, previsto × recebido
LIU7_FORTE = (6.35, 6.65)             # laços fortes

print("=" * 72)
print("  OS DADOS")
print("=" * 72)
print("  Aknin & Sandstrom 2024 (Communications Psychology; 7 estudos, N = 2.458)")
print(f"    Perderam contato com um amigo de quem gostam:  {x1}/{N1} = {pct(P1_PERDEU)}")
print(f"    Mandaram a mensagem, Estudo 3:                 {x3}/{N3} = {pct(P3_MANDOU)}")
print(f"    Mandaram a mensagem, Estudo 4:                 {x4}/{N4} = {pct(P4_MANDOU)}")
print(f"    Disposição amigo antigo × estranho (0 a 7):    {vir(M5_AMIGO)} × {vir(M5_ESTRANHO)}, BF01 = {vir(BF01_5)}")
print(f"    Estudo 7, com treino:                          {x7t}/{N7_TREINO} = {pct(P7_TREINO, 0)}")
print(f"    Estudo 7, sem treino:                          {x7s}/{N7_SEM} = {pct(P7_SEM, 0)}")
print()
print("  Liu, Rim, Min & Min 2023 (JPSP 124(4); 13 estudos pré-registrados)")
print(f"    Estudo 4a (N = 201): previsto {vir(LIU_PREV, 2)} × recebido {vir(LIU_RECEB, 2)}, d = {vir(LIU_D, 2)}")
print(f"    Replicação independente (N = {LIU_REPL_N}): d = {vir(LIU_REPL_D, 2)}")
print(f"    Estudo 7 (N = 1.602): laços fracos {vir(LIU7_FRACO[0], 2)} × {vir(LIU7_FRACO[1], 2)}; fortes {vir(LIU7_FORTE[0], 2)} × {vir(LIU7_FORTE[1], 2)}")
print("=" * 72)
```

```python
# --- O MODELO ---
N_SIM = 200_000

# Camada 1: quem manda, posterior de Jeffreys
post3 = beta_j(x3, N3); post4 = beta_j(x4, N4)
post_junto = beta_j(x3 + x4, N3 + N4)
ic3 = post3.interval(0.95); ic4 = post4.interval(0.95); icj = post_junto.interval(0.95)
p_manda = (x3 + x4) / (N3 + N4)

# Camada 2: o treino (Estudo 7), posterior da diferença
post_t = beta_j(x7t, N7_TREINO); post_s = beta_j(x7s, N7_SEM)
amostra_t = post_t.rvs(N_SIM, random_state=ger); amostra_s = post_s.rvs(N_SIM, random_state=ger)
dif = amostra_t - amostra_s
dif_ic = np.percentile(dif, [2.5, 97.5]); p_treino_ajuda = np.mean(dif > 0)
razao = amostra_t / amostra_s; razao_ic = np.percentile(razao, [2.5, 97.5])

# Camada 3: a cadeia de hoje
post1 = beta_j(x1, N1)
amostra_1 = post1.rvs(N_SIM, random_state=ger)
amostra_j = post_junto.rvs(N_SIM, random_state=ger)
cadeia = amostra_1 * amostra_j            # tem amigo perdido E manda quando a chance aparece
cadeia_ic = np.percentile(cadeia, [2.5, 97.5])

# Camada 4: quatro domingos (a chance aparece uma vez por semana, independentes)
k = np.arange(1, 13)
p_um_sem = 1 - (1 - P3_MANDOU) ** k           # com a taxa do Estudo 3, a mais baixa
p_um_junto = 1 - (1 - p_manda) ** k
p_um_treino = 1 - (1 - P7_TREINO) ** k
p_um_semtreino = 1 - (1 - P7_SEM) ** k

# Camada 5: a surpresa de quem recebe (Liu), probabilidade de superioridade
def cles(d):
    return stats.norm.cdf(d / np.sqrt(2))
ps_liu = cles(LIU_D); ps_repl = cles(LIU_REPL_D)
# duelos: uma mensagem recebida contra a previsão de quem mandou, 200.000 vezes
receb = ger.normal(LIU_RECEB, LIU_RECEB_SD, N_SIM); prev = ger.normal(LIU_PREV, LIU_PREV_SD, N_SIM)
duelo = np.mean(receb > prev)

print("=" * 72)
print("  O MODELO")
print("=" * 72)
print(f"  P(manda a mensagem quando a chance aparece)")
print(f"    Estudo 3:  {pct(P3_MANDOU)}  IC 95% [{pct(ic3[0])}; {pct(ic3[1])}]")
print(f"    Estudo 4:  {pct(P4_MANDOU)}  IC 95% [{pct(ic4[0])}; {pct(ic4[1])}]")
print(f"    conjunto:  {pct(p_manda)}  IC 95% [{pct(icj[0])}; {pct(icj[1])}]  (n = {N3 + N4})")
print()
print(f"  O treino (Estudo 7): {pct(P7_TREINO, 0)} contra {pct(P7_SEM, 0)}")
print(f"    diferença: +{vir(np.mean(dif) * 100)} pontos, IC 95% [{vir(dif_ic[0] * 100)}; {vir(dif_ic[1] * 100)}]")
print(f"    P(treinar ajuda): {pct(p_treino_ajuda)}   razão {vir(np.mean(razao), 2)}× [{vir(razao_ic[0], 2)}; {vir(razao_ic[1], 2)}]")
print()
print(f"  A cadeia de hoje: tem amigo perdido ({pct(P1_PERDEU)}) × manda ({pct(p_manda)})")
print(f"    = {pct(np.mean(cadeia))}  IC 95% [{pct(cadeia_ic[0])}; {pct(cadeia_ic[1])}]")
print()
print("  Quatro domingos, ao menos uma mensagem enviada:")
for kk in [1, 2, 4, 8, 12]:
    print(f"    {kk:>2} chances  sem treino {pct(p_um_semtreino[kk - 1]):>6}   taxa do Estudo 3 {pct(p_um_sem[kk - 1]):>6}   com treino {pct(p_um_treino[kk - 1]):>6}")
print()
print(f"  A surpresa de quem recebe (Liu 4a, d = {vir(LIU_D, 2)}):")
print(f"    de cada cem mensagens, {vir(ps_liu * 100, 0)} são mais apreciadas do que quem mandou previu")
print(f"    com o d da replicação ({vir(LIU_REPL_D, 2)}): {vir(ps_repl * 100, 0)} em cem;  duelos simulados: {pct(duelo)}")
print("=" * 72)
```

```python
# --- VISUALIZAÇÃO ---

# ══ GRÁFICO 1 · as duas portas ══
fig1, ax1 = plt.subplots(figsize=(11, 6))
etapas = ["tem um amigo antigo\nde quem ainda gosta", "manda a mensagem\n(Estudo 3)", "manda a mensagem\n(Estudo 4)"]
vals = [P1_PERDEU, P3_MANDOU, P4_MANDOU]
ics = [post1.interval(0.95), ic3, ic4]
cores = [CINZA, VERMELHO, VERMELHO]
x = np.arange(3)
ax1.bar(x, np.array(vals) * 100, color=cores, alpha=0.9, width=0.55)
for i, (v, ic) in enumerate(zip(vals, ics)):
    ax1.errorbar(i, v * 100, yerr=[[(v - ic[0]) * 100], [(ic[1] - v) * 100]], fmt="none", ecolor="#333333", capsize=6, lw=1.5)
    ax1.text(i, ic[1] * 100 + 2, pct(v), ha="center", fontsize=14, fontweight="bold", color=cores[i])
ax1.set_xticks(x); ax1.set_xticklabels(etapas, fontsize=11.5)
ax1.set_ylim(0, 105); ax1.set_ylabel("pessoas (%)")
ax1.set_title("Nove em dez têm um amigo antigo de quem gostam; uma em três manda a mensagem quando pode\n"
              "a chance foi dada ali mesmo, com o texto rascunhado; intervalos de credibilidade de 95%", fontsize=12.5, pad=14)
plt.figtext(0.5, 0.005, "Fonte: Aknin & Sandstrom 2024, Communications Psychology 2, 34 · posteriores de Jeffreys  |  #365Probabilidades",
            ha="center", fontsize=8.5, color="gray")
plt.tight_layout(rect=[0, 0.03, 1, 1])
plt.savefig("dia-127-grafico-01-as-duas-portas.png", dpi=150, bbox_inches="tight"); plt.close()
print("✅ Gráfico 1 salvo!")

# ══ GRÁFICO 2 · o treino ══
fig2, ax2 = plt.subplots(figsize=(11, 6))
xx = np.linspace(0.1, 0.8, 800)
ax2.plot(xx * 100, post_s.pdf(xx), color=CINZA, lw=2.6, label=f"sem treino: {pct(P7_SEM, 0)} (n = {N7_SEM})")
ax2.fill_between(xx * 100, post_s.pdf(xx), color=CINZA, alpha=0.18)
ax2.plot(xx * 100, post_t.pdf(xx), color=VERDE, lw=2.6, label=f"treinou antes com amigos atuais: {pct(P7_TREINO, 0)} (n = {N7_TREINO})")
ax2.fill_between(xx * 100, post_t.pdf(xx), color=VERDE, alpha=0.18)
ax2.set_xlabel("quem manda a mensagem para o amigo antigo (%)"); ax2.set_ylabel("densidade")
ax2.legend(fontsize=10.5, loc="upper right")
ax2.text(0.98, 0.6, f"diferença +{vir(np.mean(dif) * 100, 0)} pontos\nIC 95% [{vir(dif_ic[0] * 100, 0)}; {vir(dif_ic[1] * 100, 0)}]\nP(treinar ajuda) = {pct(p_treino_ajuda)}",
         transform=ax2.transAxes, ha="right", fontsize=11, color=VERDE)
ax2.set_title("Mandar duas mensagens para amigos de agora, antes, sobe o envio de 31% para 53%\n"
              "Estudo 7, N = 194; o grupo de controle navegou em rede social antes", fontsize=12.5, pad=14)
plt.figtext(0.5, 0.005, "Fonte: Aknin & Sandstrom 2024 · posteriores de Jeffreys, diferença por simulação, semente 42  |  #365Probabilidades",
            ha="center", fontsize=8.5, color="gray")
plt.tight_layout(rect=[0, 0.03, 1, 1])
plt.savefig("dia-127-grafico-02-o-treino.png", dpi=150, bbox_inches="tight"); plt.close()
print("✅ Gráfico 2 salvo!")

# ══ GRÁFICO 3 · quatro domingos ══
fig3, ax3 = plt.subplots(figsize=(11, 6))
ax3.plot(k, p_um_semtreino * 100, marker="o", lw=2.6, color=CINZA, label=f"sem treino, {pct(P7_SEM, 0)} por chance (Estudo 7)")
ax3.plot(k, p_um_treino * 100, marker="o", lw=2.6, color=VERDE, label=f"com treino, {pct(P7_TREINO, 0)} por chance (Estudo 7)")
ax3.axvline(4, color="#333333", lw=1, ls="--")
for serie, cor, dy in [(p_um_semtreino, CINZA, -5.5), (p_um_treino, VERDE, 2.5)]:
    ax3.text(4.15, serie[3] * 100 + dy, pct(serie[3], 0), fontsize=12, color=cor, fontweight="bold")
ax3.text(4.15, 8, "quatro\ndomingos", fontsize=10.5, color="#333333")
ax3.set_xlabel("domingos em que você se dá a chance"); ax3.set_ylabel("chance de ao menos uma mensagem sair (%)")
ax3.set_xticks(k); ax3.set_ylim(0, 104)
ax3.legend(fontsize=10, loc="lower right")
ax3.set_title("Em quatro domingos, a chance de ao menos uma mensagem sair vai de 77% para 95%\n"
              "conta própria, 1 − (1 − p)ⁿ, chances tratadas como independentes", fontsize=12.5, pad=14)
plt.figtext(0.5, 0.005, "Modelo próprio sobre as taxas de Aknin & Sandstrom 2024; hipótese de independência declarada  |  #365Probabilidades",
            ha="center", fontsize=8.5, color="gray")
plt.tight_layout(rect=[0, 0.03, 1, 1])
plt.savefig("dia-127-grafico-03-quatro-domingos.png", dpi=150, bbox_inches="tight"); plt.close()
print("✅ Gráfico 3 salvo!")

# ══ GRÁFICO 4 · a surpresa de quem recebe ══
fig4, ax4 = plt.subplots(figsize=(11, 6))
xx = np.linspace(1, 7, 600)
ax4.plot(xx, stats.norm.pdf(xx, LIU_PREV, LIU_PREV_SD), color=CINZA, lw=2.6, label=f"o que quem manda prevê: {vir(LIU_PREV, 2)}")
ax4.fill_between(xx, stats.norm.pdf(xx, LIU_PREV, LIU_PREV_SD), color=CINZA, alpha=0.18)
ax4.plot(xx, stats.norm.pdf(xx, LIU_RECEB, LIU_RECEB_SD), color=DOURADO, lw=2.6, label=f"o que quem recebe sente: {vir(LIU_RECEB, 2)}")
ax4.fill_between(xx, stats.norm.pdf(xx, LIU_RECEB, LIU_RECEB_SD), color=DOURADO, alpha=0.22)
ax4.axvline(LIU_PREV, color=CINZA, lw=1.2, ls="--"); ax4.axvline(LIU_RECEB, color="#8a6f1f", lw=1.2, ls="--")
ax4.set_xlabel("apreciação da mensagem (1 a 7)"); ax4.set_ylabel("densidade")
ax4.set_xlim(1, 7); ax4.legend(fontsize=10.5, loc="upper left")
ax4.text(0.02, 0.62, f"d = {vir(LIU_D, 2)}\nde cada cem mensagens, {vir(ps_liu * 100, 0)}\nsão mais bem recebidas do que\nquem mandou imaginou",
         transform=ax4.transAxes, fontsize=11, color="#8a6f1f")
ax4.set_title("Quem recebe gosta mais da mensagem do que quem manda imagina\n"
              "Liu et al. 2023, Estudo 4a, N = 201; curvas normais com as médias e desvios publicados", fontsize=12.5, pad=14)
plt.figtext(0.5, 0.005, "Fonte: Liu, Rim, Min & Min 2023, JPSP 124(4) · replicação independente d = 0,25 · CLES = Φ(d/√2)  |  #365Probabilidades",
            ha="center", fontsize=8.5, color="gray")
plt.tight_layout(rect=[0, 0.03, 1, 1])
plt.savefig("dia-127-grafico-04-a-surpresa.png", dpi=150, bbox_inches="tight"); plt.close()
print("✅ Gráfico 4 salvo!")
```

### 💡 O Insight

**Quando a chance de escrever para o amigo antigo aparece pronta, com o texto
rascunhado, uma em cada três pessoas manda. 32,9%, somando os dois estudos.**

Não é falta de vontade. Nove em dez têm um amigo de quem ainda gostam e com quem
perderam contato. É o freio: a ideia de que o amigo não vai querer saber, e de que vai
ser estranho aparecer depois de tanto tempo. Escrever para ele dá a mesma vontade que
puxar conversa com um estranho.

O outro lado da mensagem diz o contrário. Quem recebe gosta mais dela do que quem manda
imagina, e o que explica a diferença é a surpresa. De cada cem mensagens, 59 são mais
bem recebidas do que quem mandou previu.

E existe um truque medido. Mandar antes duas mensagens para amigos de agora, para
aquecer, sobe o envio para o amigo antigo de 31% para 53%. Em quatro domingos, a chance
de ao menos uma mensagem sair passa de 77% para 95%.

### ⚠️ Limitações do Modelo

- Amostras online, de língua inglesa, recrutadas em plataformas de pesquisa. O envio é
  medido numa sessão de estudo, não na vida; na vida, a chance não vem com o texto
  rascunhado, o que pode puxar o número para baixo.
- As contagens dos Estudos 3, 4 e 7 são reconstruídas dos percentuais publicados; os
  intervalos de credibilidade herdam esse arredondamento (menos de meio ponto).
- A conta dos quatro domingos trata cada chance como independente. Quem não mandou no
  primeiro domingo pode estar menos disposto no segundo, ou mais. A curva é régua,
  não previsão.
- O treino do Estudo 7 foi testado uma vez, com 194 pessoas. A diferença de 22 pontos
  tem intervalo largo (8 a 35) e espera replicação.
- O d de Liu et al. é pequeno (0,32; 0,25 na replicação). A tradução "59 em cem" vale
  para a média das mensagens, não para uma mensagem específica.
- Nenhum dos dois estudos mede o que acontece depois: se o amigo respondeu, se a
  amizade voltou. O dia é sobre a mensagem sair.

*A ciência é honesta sobre o que não sabe. O modelo também.*
