### 📖 A História

Estou passando alguns dias com o meu filho, e por sorte caiu justo nesses dias um
encontro de famílias na escola dele. Deu para conhecer os professores, ver as salas,
entender um pouco do lugar onde ele passa o dia.

A primeira coisa que eu pensei foi: que sorte enorme.

A segunda foi: espera, qual é a chance disso?

E aí eu percebi que essa pergunta, do jeito que eu tinha feito, não tinha resposta. Ela
estava mal formulada.

E o mais interessante é que quase todas as perguntas sobre coincidência são mal
formuladas do mesmo jeito.

---

### 📚 O Conceito: a pergunta certa

Existe uma diferença entre duas perguntas que parecem a mesma.

A primeira é: **qual a chance de aquele evento específico cair naqueles dias
específicos?** Essa é pequena, e dá para calcular.

A segunda é: **qual a chance de alguma coisa boa coincidir com a minha viagem?** Essa é
grande, e é a que de fato aconteceu.

O erro está em viver a segunda e calcular a primeira. Você percebe a coincidência
depois que ela aconteceu, e aí compara com a probabilidade de exatamente aquilo, quando
o que deveria estar na conta é a probabilidade de **qualquer coisa** naquela categoria.

O exemplo clássico disso é o problema do aniversário. Pergunte a alguém quantas pessoas
são necessárias numa sala para que a chance de duas dividirem aniversário passe de 50%.
Quase todo mundo responde na casa das centenas.

A resposta é **23**.

O motivo é que ninguém está procurando alguém que divida aniversário com **você**.
Está procurando **qualquer par** entre todos os presentes. E o número de pares cresce
muito mais rápido que o número de pessoas.

### 🧮 O Modelo

Um artigo que formaliza como estudar coincidências, e as contas que ele permite fazer.

**Fontes:**
- Diaconis, P. & Mosteller, F., 1989 · "Methods for Studying Coincidences" · *Journal
  of the American Statistical Association* 84(408), 853-861 · desenvolvem uma versão do
  problema do aniversário geral o bastante para incluir **dependência, não
  homogeneidade e coincidências aproximadas ou múltiplas**, e revisam as técnicas de
  Fisher para dar crédito parcial a correspondências próximas
- **Os quatro princípios** que, descartadas as coincidências com causa aparente,
  explicam a maior parte das restantes: **causa oculta**; **psicologia**, incluindo
  memória e percepção; **multiplicidade de desfechos**, incluindo contar eventos
  próximos como se fossem idênticos; e a **lei dos números realmente grandes**,
  segundo a qual, quando números enormes de eventos, pessoas e suas interações se
  acumulam ao longo do tempo, **quase qualquer evento ultrajante está destinado a
  acontecer**
- **Problema do aniversário clássico:** bastam **23 pessoas** para que a chance de duas
  dividirem aniversário passe de 50%

**Referência interna:** o **#077** tratou do mesmo mecanismo pelo lado do medo, com a
mala que não chega. Aqui é pelo lado da sorte.

**Nota metodológica sobre o fator ×0.80:** não se aplica. Probabilidade combinatória
exata, mais modelo próprio com premissas declaradas.


```python
import numpy as np
import matplotlib.pyplot as plt
from math import comb

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Diaconis & Mosteller, 1989, JASA 84(408), 853-861

n_aniversario_50 = 23      # pessoas para passar de 50% no problema do aniversario
DIAS_ANO = 365

PRINCIPIOS = {
    'Causa oculta': 'existe um motivo que voce nao ve',
    'Psicologia': 'memoria, percepcao e atencao seletiva',
    'Multiplicidade de desfechos': 'contar o "quase igual" como identico',
    'Lei dos numeros realmente grandes': 'com eventos suficientes, o absurdo acontece',
}

aplica_fator_080 = False

print("=" * 74)
print("  DADOS - COMO SE ESTUDA UMA COINCIDENCIA")
print("=" * 74)
print(f"\n  Diaconis & Mosteller, 1989 (JASA 84(408), 853-861):")
print(f"  -> Versao geral do problema do aniversario, incluindo dependencia,")
print(f"     nao homogeneidade e coincidencias aproximadas")
print(f"  -> Problema classico: bastam {n_aniversario_50} pessoas para passar de 50%")
print(f"\n  Os quatro principios que explicam as coincidencias:")
for nome, desc in PRINCIPIOS.items():
    print(f"  -> {nome:<36} {desc}")
print(f"\n  Fator x0.80 aplicado: {aplica_fator_080}")
print("=" * 74)
```

```python
# --- O MODELO ---

# =========================================================
# PARTE 1 - O problema do aniversario, com a formula exata
# =========================================================
def p_aniversario(n, dias=DIAS_ANO):
    "Probabilidade de ao menos um par compartilhar aniversario."
    if n > dias:
        return 1.0
    p_nenhum = 1.0
    for i in range(n):
        p_nenhum *= (dias - i) / dias
    return 1 - p_nenhum


def pares(n):
    return comb(n, 2) if n >= 2 else 0


print("=" * 74)
print("  MODELO - PARTE 1: POR QUE 23 PESSOAS BASTAM")
print("=" * 74)
print(f"\n  {'pessoas':>8} | {'pares possiveis':>16} | {'P(algum par)':>13}")
print(f"  {'-'*8}-+-{'-'*16}-+-{'-'*13}")
for n in (5, 10, 20, 23, 30, 40, 50, 70):
    print(f"  {n:>8} | {pares(n):>16} | {p_aniversario(n)*100:>12.1f}%")

# O contraste que explica a intuicao errada
p_com_voce = 1 - ((DIAS_ANO - 1) / DIAS_ANO) ** (n_aniversario_50 - 1)
print(f"\n  E aqui esta a raiz do erro de intuicao:")
print(f"  -> Chance de alguem dividir aniversario COM VOCE, entre"
      f" {n_aniversario_50} pessoas: {p_com_voce*100:.1f}%")
print(f"  -> Chance de EXISTIR ALGUM PAR entre as mesmas"
      f" {n_aniversario_50} pessoas: {p_aniversario(n_aniversario_50)*100:.1f}%")
print(f"  -> A mesma sala, a mesma gente. Duas perguntas diferentes.")
print(f"  -> Sao {pares(n_aniversario_50)} pares possiveis, e nao"
      f" {n_aniversario_50 - 1}.")

# =========================================================
# PARTE 2 - A multiplicidade aplicada a uma viagem
# =========================================================
# PREMISSAS DECLARADAS. Nada disso vem do artigo.
DIAS_VIAGEM = 10           # duracao da visita
EVENTOS_ESCOLA_ANO = 8     # eventos abertos as familias por ano letivo
DIAS_LETIVOS = 200         # dias de escola no ano


def p_algum_evento(dias_viagem, eventos_ano, dias_letivos=DIAS_LETIVOS):
    "P(ao menos um dos eventos cair na janela da viagem)."
    p_um_evento_fora = 1 - dias_viagem / dias_letivos
    return 1 - p_um_evento_fora ** eventos_ano


print("\n" + "=" * 74)
print("  MODELO - PARTE 2: A CHANCE DAQUELE EVENTO")
print("=" * 74)
print(f"\n  PREMISSAS DECLARADAS, nao publicadas:")
print(f"  -> Viagem de {DIAS_VIAGEM} dias")
print(f"  -> {EVENTOS_ESCOLA_ANO} eventos abertos as familias por ano")
print(f"  -> {DIAS_LETIVOS} dias letivos no ano")

p_esse = p_algum_evento(DIAS_VIAGEM, EVENTOS_ESCOLA_ANO)
p_um_so = DIAS_VIAGEM / DIAS_LETIVOS

print(f"\n  -> Chance de UM evento especifico cair na viagem:"
      f" {p_um_so*100:.1f}%")
print(f"  -> Chance de ALGUM dos {EVENTOS_ESCOLA_ANO} eventos cair na viagem:"
      f" {p_esse*100:.1f}%")
print(f"  -> Ou seja, {p_esse/p_um_so:.1f}x mais provavel, so por haver mais de um.")

# =========================================================
# PARTE 3 - A multiplicidade completa: quantas coisas contariam
# =========================================================
# Se, em vez de UMA categoria (evento na escola), existirem varias
# categorias de coincidencia boa, a chance de ALGUMA acontecer sobe rapido.
CATEGORIAS = [1, 2, 3, 5, 8, 12]

print("\n" + "=" * 74)
print("  MODELO - PARTE 3: A LEI DOS NUMEROS REALMENTE GRANDES")
print("=" * 74)
print(f"\n  A pergunta certa nao e 'qual a chance daquilo'.")
print(f"  E 'qual a chance de ALGUMA coisa boa coincidir'.")
print(f"\n  Se cada categoria de coincidencia tem chance"
      f" {p_esse*100:.0f}% de ocorrer na viagem:")
print(f"\n  {'categorias':>11} | {'P(alguma acontecer)':>20}")
print(f"  {'-'*11}-+-{'-'*20}")
for k in CATEGORIAS:
    p_alguma = 1 - (1 - p_esse) ** k
    print(f"  {k:>11} | {p_alguma*100:>19.1f}%")

p_5 = 1 - (1 - p_esse) ** 5
print(f"\n  -> Com apenas 5 categorias possiveis, a chance de ALGUMA")
print(f"     coincidencia boa acontecer ja passa de {p_5*100:.0f}%.")
print(f"  -> E depois que ela acontece, voce calcula a probabilidade")
print(f"     de EXATAMENTE aquela, que era de {p_esse*100:.0f}%,")
print(f"     e conclui que foi sorte enorme.")
print(f"\n  -> Este e o principio da MULTIPLICIDADE DE DESFECHOS.")
print(f"     Voce nao escolheu o alvo antes. O alvo foi escolhido depois.")
print("=" * 74)
```

```python
# --- VISUALIZACAO ---

DOURADO = '#c8a84b'
VERMELHO = '#c0392b'
VERDE = '#2a8a82'
CINZA = '#6b6a64'

# GRAFICO 1 - O problema do aniversario
fig1, ax1 = plt.subplots(figsize=(12, 8))

ns = np.arange(2, 71)
curva = [p_aniversario(int(n)) * 100 for n in ns]
com_voce = [(1 - ((DIAS_ANO - 1) / DIAS_ANO) ** (int(n) - 1)) * 100 for n in ns]

ax1.plot(ns, curva, color=DOURADO, linewidth=3.6,
         label='chance de existir ALGUM par')
ax1.plot(ns, com_voce, color=CINZA, linewidth=3, linestyle='--',
         label='chance de alguém dividir COM VOCÊ')
ax1.fill_between(ns, curva, alpha=0.12, color=DOURADO)

ax1.axhline(y=50, color=VERMELHO, linestyle=':', linewidth=2)
ax1.axvline(x=n_aniversario_50, color=VERMELHO, linestyle=':', linewidth=2)
ax1.scatter([n_aniversario_50], [p_aniversario(n_aniversario_50) * 100], s=240,
            color=VERMELHO, zorder=4)
ax1.text(n_aniversario_50 + 1.5, 46,
         f'{n_aniversario_50} pessoas\n{p_aniversario(n_aniversario_50)*100:.1f}%'
         .replace('.', ','),
         fontsize=15, fontweight='bold', color=VERMELHO)

ax1.set_xlim(2, 70)
ax1.set_ylim(0, 105)
ax1.set_xlabel('Pessoas na sala')
ax1.set_ylabel('Probabilidade (%)')
ax1.legend(frameon=False, fontsize=13, loc='upper left')
ax1.set_title('A mesma sala, duas perguntas diferentes\n'
              'Bastam 23 pessoas para a chance de algum par passar de 50%',
              fontsize=15, pad=18)
ax1.text(0.5, -0.135,
         'Fórmula exata do problema do aniversário. Com 23 pessoas existem 253 pares '
         'possíveis, e não 22.',
         transform=ax1.transAxes, ha='center', fontsize=12, color=DOURADO,
         fontweight='bold')

plt.figtext(0.5, 0.005,
            'Problema do aniversário · Diaconis & Mosteller, 1989, JASA 84(408)'
            '  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia085_g1_aniversario.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 1 salvo")

# GRAFICO 2 - Assinatura: a lei dos numeros realmente grandes
fig2, ax2 = plt.subplots(figsize=(12, 8))

ks = np.arange(1, 21)
p_alguma_curva = [(1 - (1 - p_esse) ** int(k)) * 100 for k in ks]

ax2.plot(ks, p_alguma_curva, color=DOURADO, linewidth=3.6)
ax2.fill_between(ks, p_alguma_curva, alpha=0.14, color=DOURADO)

for k in (1, 5, 12):
    v = (1 - (1 - p_esse) ** k) * 100
    cor = VERMELHO if k == 1 else DOURADO
    ax2.scatter([k], [v], s=210, color=cor, zorder=4)
    rotulo = ('1 categoria' if k == 1 else f'{k} categorias')
    dy = -9 if k == 1 else 6
    ax2.text(k + 0.35, v + dy, f'{rotulo}\n{v:.0f}%', fontsize=13,
             fontweight='bold', color=cor)

ax2.set_xlim(0.5, 20.5)
ax2.set_ylim(0, 118)
ax2.set_xticks(range(1, 21, 2))
ax2.set_xlabel('Quantas categorias de coincidência boa seriam contadas')
ax2.set_ylabel('Chance de ALGUMA delas acontecer (%)')
ax2.set_title('A assinatura estatística: a lei dos números realmente grandes\n'
              'A chance daquilo é pequena. A chance de alguma coisa é enorme.',
              fontsize=15, pad=18)
ax2.text(0.5, -0.135,
         'Modelo próprio, com premissas declaradas. Cada categoria com a mesma chance '
         'de ocorrer, e categorias independentes.',
         transform=ax2.transAxes, ha='center', fontsize=11, color=CINZA, style='italic')

plt.figtext(0.5, 0.005,
            'Princípio da multiplicidade de desfechos · Diaconis & Mosteller, 1989'
            '  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia085_g2_multiplicidade.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 2 salvo")

# GRAFICO 3 - Os quatro principios
fig3, ax3 = plt.subplots(figsize=(12, 8))
ax3.axis('off'); ax3.set_xlim(0, 10); ax3.set_ylim(0, 10)

ax3.text(5, 9.4, 'Os quatro princípios das coincidências', ha='center', fontsize=17,
         fontweight='bold')
ax3.text(5, 8.7, 'depois de descartadas as que têm causa aparente', ha='center',
         fontsize=12, color=CINZA, style='italic')

blocos = [
    (7.0, 'Causa oculta', 'existe um motivo que você não vê', DOURADO),
    (5.2, 'Psicologia', 'memória, percepção e atenção seletiva', DOURADO),
    (3.4, 'Multiplicidade de desfechos',
     'contar o "quase igual" como se fosse idêntico', VERMELHO),
    (1.6, 'Lei dos números realmente grandes',
     'com eventos suficientes, o absurdo é inevitável', VERMELHO),
]
for y, titulo, desc, cor in blocos:
    ax3.text(0.6, y, titulo, ha='left', va='center', fontsize=15, fontweight='bold',
             color=cor)
    ax3.text(0.6, y - 0.55, desc, ha='left', va='center', fontsize=12, color='#333')
    ax3.plot([0.4, 9.6], [y - 0.95, y - 0.95], color='#e5e2da', linewidth=1)

ax3.text(5, 0.35,
         'Os dois últimos explicam a maior parte das coincidências da vida real.',
         ha='center', fontsize=13, fontweight='bold', color=VERMELHO)

plt.figtext(0.5, 0.005,
            'Fonte: Diaconis & Mosteller, 1989, JASA 84(408), 853-861'
            '  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia085_g3_principios.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 3 salvo")

# --- CONFIRMACAO: onde os arquivos foram salvos ---
import os
print("\n" + "=" * 74)
print("  ARQUIVOS GERADOS")
print("=" * 74)
print("  Pasta de trabalho:", os.getcwd())
for f in sorted(os.listdir()):
    if f.startswith("dia085_") and f.endswith(".png"):
        print(f"  -> {f}  ({os.path.getsize(f)/1024:.0f} KB)")
print("=" * 74)
```

### 💡 O Insight

Comecei este dia pensando que tive uma sorte enorme. E tive, mas não do tamanho que eu
achava, e por um motivo diferente do que eu imaginava.

O problema do aniversário resolve isso melhor que qualquer explicação.

Numa sala com 23 pessoas, a chance de **alguém dividir aniversário com você** é de
apenas **5,9%**. A chance de **existir algum par** entre as mesmas 23 pessoas é de
**50,7%**.

Mesma sala. Mesma gente. Duas perguntas diferentes, com respostas oito vezes distintas.

O motivo é que ninguém está procurando um par com você. Está procurando qualquer par. E
entre 23 pessoas existem **253 pares possíveis**, não 22.

Aplicado à minha viagem, a conta é a mesma.

A chance de **um evento específico** da escola cair nos dias em que eu estaria lá é de
**5%**. A chance de **algum** dos eventos do ano cair ali é de **34%**.

E se eu contar não só eventos da escola, mas qualquer coisa boa que eu chamaria de
coincidência, a chance de **alguma** delas acontecer passa de **87%** com cinco
categorias.

Ou seja: era quase certo que alguma coisa coincidisse. Só não havia como saber qual.

É isso que Diaconis e Mosteller chamam de **multiplicidade de desfechos**. Eu não
escolhi o alvo antes de atirar. O alvo foi escolhido depois que a flecha caiu.

E é isso que a **lei dos números realmente grandes** diz: quando números enormes de
eventos, pessoas e suas interações se acumulam ao longo do tempo, quase qualquer coisa
absurda está destinada a acontecer.

Este projeto já tratou do mesmo mecanismo pelo lado ruim, no **#077**. A mala que some
é rara e inesquecível, então ela ocupa mais espaço do que merece.

Aqui é o lado bom, e a assimetria me interessa mais.

Coincidência ruim a gente atribui a azar e guarda como história. Coincidência boa a
gente atribui a sinal, a destino, a mérito. As duas são a mesma aritmética.

E existe uma coisa que eu não quero perder de vista, mesmo tendo desmontado a própria
sorte.

Saber que era provável que **alguma coisa** coincidisse não faz o encontro na escola ser
menos bom. Continuo tendo conhecido os professores, visto as salas, entendido melhor o
lugar onde ele passa os dias.

A probabilidade explica por que aconteceu. Não explica por que valeu.

*Qual coincidência da sua vida você guarda como se fosse um sinal?*

### ⚠️ Limitações do Modelo

- **O problema do aniversário é exato. As contas da viagem são minhas.** Elas usam
  premissas declaradas que não vêm do artigo: dez dias de viagem, oito eventos abertos
  às famílias por ano, duzentos dias letivos.
- **As categorias de coincidência são construídas por mim**, e o modelo assume que cada
  uma tem a mesma chance de ocorrer e que são independentes. Nenhuma das duas coisas é
  exatamente verdade na vida real.
- **O número de categorias é arbitrário por natureza**, e isso não é um defeito do
  modelo, é o ponto dele: a conta depende de quantas coisas você aceitaria como
  coincidência, e essa lista só é definida depois do fato.
- **O problema do aniversário clássico supõe aniversários uniformes ao longo do ano.**
  Na prática, nascimentos não são uniformes, o que faz a probabilidade real ser
  ligeiramente maior. Diaconis e Mosteller desenvolvem justamente as versões com
  dependência e não homogeneidade.
- **Nada aqui prova que coincidências não têm causa.** O primeiro dos quatro princípios
  é a causa oculta, e ele existe justamente porque muitas coincidências têm explicação
  banal que a pessoa não viu.
- **Este dia não mede nada em ninguém.** É probabilidade combinatória e um modelo
  próprio, e não um estudo empírico com participantes.
- Fator ×0.80 não aplicado.

*A ciência é honesta sobre o que não sabe. O modelo também.*
