### 📖 A História

Tem um momento exato em que um hábito de saúde morre. Não é no dia em que você está mais cansada, mais ocupada, ou menos motivada. É no dia seguinte a uma falha.

Você quebrou a dieta no jantar de sábado. Pulou o treino de terça porque o dia ficou impossível. Tomou o remédio com três horas de atraso. E, em vez de simplesmente seguir adiante a partir daquele ponto, alguma coisa na cabeça decide: já que estragou, estraga tudo.

Psicólogos descreveram esse padrão há décadas e deram um nome a ele: o "what the hell effect" — o efeito "ah, que se dane". E ele não é um detalhe menor da motivação. Ele é, segundo a pesquisa, um dos principais motivos pelos quais hábitos de saúde — apps, dietas, rotinas de exercício — são abandonados não gradualmente, mas de uma vez, num único momento de desistência.

---

### 📚 O Conceito: O Efeito "What the Hell"

O conceito foi descrito por Janet Polivy e C. Peter Herman, da Universidade de Toronto, dentro do que chamaram de **Modelo de Fronteira** (*Boundary Model*) da alimentação restrita. A ideia central: pessoas que seguem regras rígidas sobre o próprio comportamento (uma dieta, um plano de exercícios, um horário de remédio) estabelecem uma fronteira cognitiva — um limite mental do que é "permitido".

Quando essa fronteira é cruzada — mesmo que minimamente — o efeito psicológico não é proporcional à falha. A pessoa não pensa "ok, hoje eu escapei um pouco, volto ao normal amanhã". Ela pensa, na prática, "já que quebrei a regra, não faz mais diferença continuar quebrando" — e abandona o comportamento por completo, pelo menos temporariamente, mergulhando no que os pesquisadores descrevem como um ciclo de indulgência, culpa e mais indulgência.

Esse padrão de tudo-ou-nada ajuda a explicar por que o abandono de comportamentos de saúde tende a ser abrupto, não gradual — e por que isso é tão comum em populações inteiras monitoradas digitalmente.

### 🧮 O Modelo
**Fontes:**
- Herman, C.P. & Polivy, J. (1984) — *A Boundary Model for the Regulation of Eating* — conceito original do "what the hell effect"
- Polivy, J. & Herman, C.P. (1985) — *Dieting and bingeing: A causal analysis* — American Psychologist, 40(2), 193-201
- Revisão sistemática (2024) — *When and Why Adults Abandon Lifestyle Behavior and Mental Health Mobile Apps* — Journal of Medical Internet Research — 18 estudos, N=525.824 participantes


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print('✅ Bibliotecas carregadas')
```

```python
# --- DADOS DA LITERATURA ---

# Revisao sistematica 2024 (JMIR) — 18 estudos, N=525.824 participantes
n_estudo_apps = 525824
n_estudos_consolidados = 18
p_abandono_100dias = 0.70  # mediana de abandono dentro dos primeiros 100 dias

# Padrao do abandono: curvilinear, mais abrupto no inicio
# (caracteristica central do "what the hell effect" — abandono nao-gradual)

fator_correcao = 0.80

print('=' * 65)
print('  DADOS DA LITERATURA — ABANDONO DE HÁBITOS DE SAÚDE')
print('=' * 65)
print(f'\n  Revisão sistemática 2024, JMIR ({n_estudos_consolidados} estudos, N={n_estudo_apps:,}):')
print(f'  → Abandono de apps de saúde em até 100 dias : {p_abandono_100dias*100:.0f}% (mediana)')
print(f'  → Padrão de abandono: curvilinear — mais abrupto logo no início')
print(f'\n  Conceito teórico (Polivy & Herman):')
print(f'  → "What the hell effect": uma única falha tende a gerar')
print(f'    abandono total, não apenas falha pontual')
print(f'\n  Fator de correção aplicado: x{fator_correcao}')
print('=' * 65)
print()
print('  ⚠️  NOTA METODOLÓGICA:')
print('  A revisão sistemática mede abandono de APPS de saúde — um proxy')
print('  observável e mensurável para abandono de hábito em geral. O')
print('  mecanismo psicológico do "what the hell effect" é bem documentado')
print('  qualitativamente (decadas de estudos experimentais), mas a fonte')
print('  consolidada usada aqui para o número (70%) mede o COMPORTAMENTO')
print('  de abandono, não isola estatisticamente a CAUSA psicológica.')
```

```python
# --- O MODELO ---
# Distribuicao Beta com IC 95% para a proporcao de abandono

alpha = p_abandono_100dias * n_estudo_apps
beta  = (1 - p_abandono_100dias) * n_estudo_apps
dist  = stats.beta(alpha, beta)
ic    = dist.interval(0.95)

p_corrigido = p_abandono_100dias * fator_correcao

# Proporcao que MANTEM o habito alem de 100 dias (complemento)
p_mantem = 1 - p_abandono_100dias

print('=' * 65)
print('  MODELO — RESULTADO')
print('=' * 65)
print(f'  Abandono de hábito de saúde em até 100 dias:')
print(f'  → Estimativa : {p_abandono_100dias*100:.0f}%')
print(f'  → IC 95%     : [{ic[0]*100:.2f}%, {ic[1]*100:.2f}%]')
print(f'  → Corrigido  : {p_corrigido*100:.1f}%')
print()
print(f'  Proporção que mantém o hábito além de 100 dias: {p_mantem*100:.0f}%')
print('=' * 65)
```

```python
# --- VISUALIZACAO ---

# GRAFICO 1 — Curva de abandono ao longo do tempo (ilustrativa, padrao curvilinear)
fig1, ax1 = plt.subplots(figsize=(12, 8))

dias = np.linspace(0, 100, 200)
# Curva curvilinea: abandono mais abrupto no inicio, depois desacelera
# Calibrada para terminar em 70% aos 100 dias
abandono_acumulado = p_abandono_100dias * (1 - np.exp(-dias / 25))

ax1.plot(dias, abandono_acumulado * 100, color='#c0392b', linewidth=3)
ax1.fill_between(dias, abandono_acumulado * 100, alpha=0.15, color='#c0392b')
ax1.axhline(y=p_abandono_100dias * 100, color='#888888', linestyle='--', linewidth=1.5,
            label=f'{p_abandono_100dias*100:.0f}% de abandono aos 100 dias')

ax1.set_xlabel('Dias desde o início do hábito', fontsize=12)
ax1.set_ylabel('Proporção acumulada que abandonou (%)', fontsize=12)
ax1.set_title('O abandono é abrupto no início, não gradual\nRevisão sistemática 2024 (JMIR) — 18 estudos, N=525.824',
              fontsize=13, pad=15)
ax1.legend(fontsize=11)
ax1.set_ylim(0, 85)

plt.tight_layout()
plt.savefig('dia-017-grafico-01-curva-abandono.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Grafico 1 salvo!')


# GRAFICO 2 — Mantem vs abandona
fig2, ax2 = plt.subplots(figsize=(10, 10))

labels2 = ['Abandonaram\nem até 100 dias', 'Mantiveram\nalém de 100 dias']
sizes2 = [p_abandono_100dias * 100, p_mantem * 100]
cores2 = ['#c0392b', '#2a8a82']

wedges, texts, autotexts = ax2.pie(sizes2, labels=labels2, colors=cores2,
                                     autopct='%1.0f%%', startangle=90,
                                     textprops={'fontsize': 15, 'fontweight': 'bold'},
                                     wedgeprops={'alpha': 0.85})

ax2.set_title('Sete de cada dez pessoas abandonam\no hábito de saúde nos primeiros 100 dias',
              fontsize=14, pad=20)

plt.tight_layout()
plt.savefig('dia-017-grafico-02-proporcao.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Grafico 2 salvo!')


# GRAFICO 3 — Distribuicao Beta
fig3, ax3 = plt.subplots(figsize=(12, 8))

x = np.linspace(ic[0] * 0.998, ic[1] * 1.002, 1000)
y = dist.pdf(x)

ax3.plot(x * 100, y, color='#c0392b', linewidth=3)
ax3.fill_between(x * 100, y, alpha=0.2, color='#c0392b')
ax3.axvline(x=ic[0] * 100, color='#c8a84b', linestyle='--', linewidth=2,
            label=f'IC 95%: [{ic[0]*100:.2f}%, {ic[1]*100:.2f}%]')
ax3.axvline(x=ic[1] * 100, color='#c8a84b', linestyle='--', linewidth=2)
ax3.axvline(x=p_abandono_100dias * 100, color='#c0392b', linewidth=2,
            label=f'Estimativa central: {p_abandono_100dias*100:.0f}%')
ax3.axvline(x=p_corrigido * 100, color='#2a8a82', linestyle='-.',
            linewidth=2, label=f'Corrigido (x0.80): {p_corrigido*100:.1f}%')

ax3.set_xlabel('Probabilidade de abandono em até 100 dias (%)', fontsize=12)
ax3.set_ylabel('Densidade', fontsize=12)
ax3.set_title('Distribuição Beta — Abandono de hábitos de saúde\nRevisão sistemática 2024 (JMIR) — N=525.824',
              fontsize=13, pad=15)
ax3.legend(fontsize=11)

plt.tight_layout()
plt.savefig('dia-017-grafico-03-bayesiano.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Grafico 3 salvo!')
```

### 💡 O Insight

Uma revisão sistemática que consolidou **18 estudos com 525.824 participantes** encontrou que **70% das pessoas abandonam um hábito de saúde digital em até 100 dias** — e o padrão de abandono é **curvilinear**: mais abrupto logo no início, não uma queda gradual e constante.

Esse padrão de desistência súbita, em vez de gradual, é exatamente o que a psicologia chama de "what the hell effect": uma falha pontual — pular um treino, comer fora da dieta, esquecer o remédio — não é vivida como um detalhe contornável. É vivida como prova de que o esforço inteiro já fracassou, abrindo caminho para o abandono total.

A implicação prática é direta: o que mais ameaça um hábito de saúde não é a primeira falha. É a interpretação que você dá a ela. Tratar um deslize como evidência de fracasso total — em vez de um dia isolado dentro de uma tentativa maior — é o gatilho que transforma uma exceção em desistência.

*Na próxima vez que você falhar um dia, a pergunta que decide tudo não é "eu falhei?" — é "isso significa que preciso parar?"*

---

### ⚠️ Limitações do Modelo
- A revisão sistemática mede abandono de aplicativos de saúde como proxy comportamental — não isola estatisticamente o mecanismo psicológico do "what the hell effect" como causa direta e exclusiva
- O conceito teórico (Polivy & Herman) tem décadas de evidência experimental robusta, mas em sua maioria com amostras menores de laboratório, diferentes do N populacional usado aqui para o número de abandono
- O padrão de abandono varia significativamente por domínio: apps de tabagismo e álcool têm abandono mais rápido que apps de atividade física e saúde mental, segundo a própria revisão
- A curva de abandono apresentada no Gráfico 1 é uma aproximação ilustrativa calibrada para refletir o padrão curvilinear descrito na revisão, não os dados brutos de cada estudo individual
- O fator de correção (×0,80) é uma aproximação padrão do projeto

*A ciência é honesta sobre o que não sabe. O modelo também.*
