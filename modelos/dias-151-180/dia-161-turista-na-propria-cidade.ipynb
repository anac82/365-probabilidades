# 365 Probabilidades - Dia #161
## Qual a probabilidade de você conhecer menos a sua cidade do que um turista?

**Tipo:** Curiosidade · Comportamento  
**Data de publicação:** 2026-11-21  
**Ferramenta:** Python  
**Decisão analisada:** O passeio que está perto de casa fica para quando?  
**Hashtag:** #365Probabilidades #Dia161

---

### 📖 A História

Quase todo mundo tem um ponto turístico da própria cidade que nunca visitou.

Fica a vinte minutos de casa, aparece em todo cartão-postal, e a visita está sempre marcada para um dia desses. Às vezes ela só acontece quando chega um parente de fora e alguém precisa levar.

Em viagem acontece o contrário. Em três dias a gente vê o que o morador não viu em três anos.

Fui ver se isso aparece nos dados, e por que acontece.

---

### 📚 O Conceito

A procrastinação costuma ser estudada com tarefas desagradáveis, como o relatório, o exame ou a academia. Suzanne Shu e Ayelet Gneezy estudaram o outro lado, o adiamento de coisas boas.

A explicação delas tem duas partes. A primeira é que a gente acredita que terá mais tempo livre no futuro do que tem hoje. A segunda é que, sem prazo, qualquer dia serve, e por isso nenhum dia é escolhido. O turista tem data para ir embora. O morador não tem.

Para testar, elas compararam moradores e visitantes, acompanharam quem estava de mudança e distribuíram vales-presente com prazos curtos e longos.

---

### 🧮 O Modelo

Uma fonte, com quatro dos seus cinco experimentos, e quatro contas próprias.

1. **Morador contra visitante.** A diferença entre as médias publicadas, com intervalo reconstruído a partir do teste t.
2. **A resposta do dia.** A parcela de moradores que conhece mais pontos turísticos em outra cidade do que na sua, com intervalo de Jeffreys, e uma mesa de dez moradores sorteada 200 mil vezes (semente 42).
3. **A despedida.** As visitas de quem se mudou, comparadas com o que se esperaria se elas se espalhassem por igual pelo tempo de moradia.
4. **O prazo.** O uso de vales-presente com prazo curto e longo, contra o que as pessoas previram.

**Fonte:**
- Shu, S. B. & Gneezy, A. (2010). Procrastination of Enjoyable Experiences. *Journal of Marketing Research*, 47, 933-944. Lido no artigo completo.
  - Experimento 1: entrevistas com **199 pedestres** em Chicago (99, dos quais 44 moradores), Londres (70, dos quais 36 moradores) e Dallas (30, dos quais 29 moradores), sobre sete cidades.
  - Experimento 2: **51 ex-moradores** de Chicago, com 219 visitas a pontos turísticos datadas.
  - Experimento 4: vale-presente de uma confeitaria, **64 universitários** com o vale e 80 num grupo de previsão.
  - Experimento 5: ingresso de cinema, **240 universitários**.

**Nota metodológica sobre o fator ×0,80.** Não se aplica. Os números são contagens de visitas e de vales usados, e médias de pontos visitados. Nenhum é proporção de opinião autorrelatada. O número de pontos visitados é lembrado pelo entrevistado, e isso está nas limitações.

**Cordas declaradas:** #137 (achar que mês que vem haverá mais tempo, Zauberman & Lynch 2005), #107 (falácia do planejamento) e #044 (procrastinação). O objeto aqui é outro, o adiamento do que é agradável.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

VERMELHO, DOURADO, VERDE, CINZA = '#c0392b', '#c8a84b', '#1a5f5a', '#6b6a64'
SEMENTE, N_SIM = 42, 200_000
rng = np.random.default_rng(SEMENTE)

def jeffreys(k, n, nivel=0.95):
    """Intervalo de Jeffreys para uma proporção."""
    a = (1 - nivel) / 2
    return stats.beta.ppf([a, 1 - a], k + 0.5, n - k + 0.5)

def sorteia(k, n):
    return rng.beta(k + 0.5, n - k + 0.5, N_SIM)

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Shu, S. B. & Gneezy, A. (2010). Procrastination of Enjoyable Experiences.
# Journal of Marketing Research, 47, 933-944. Lido no artigo completo.

# Experimento 1: 199 pedestres entrevistados em Chicago, Londres e Dallas
n_entrevistados = 199
# Pontos turísticos visitados na própria cidade, por tempo de moradia
moradores = {'menos de 1 ano': 3.0, '1 ano': 4.5, '2 anos': 5.0, '3 anos ou mais': 4.8}
# Pontos turísticos visitados numa cidade de fora, por duração da visita
visitantes = {'até 1 dia': 1.4, '2 a 3 dias': 1.7, '4 a 7 dias': 3.8, '2 semanas': 4.4, '3 semanas': 5.5}
# Comparação publicada: visitante de duas semanas contra morador de menos de um ano
media_visitante_2sem, media_morador_1ano, t_exp1, gl_exp1 = 4.4, 3.1, 3.3, 59
# Moradores que conheciam mais pontos turísticos em outra cidade do que na sua (percentual e total publicados)
por_cidade = {'Londres': (61, 36), 'Chicago': (57, 44), 'Dallas': (55, 29)}
pct_visitas_com_hospede = 60          # % das visitas do morador feitas com alguém de fora

# Experimento 2: 51 ex-moradores de Chicago, moradia média de 3 anos
n_ex_moradores = 51
anos_moradia_media = 3
visitas_total, visitas_ultimos_6_meses, visitas_ultimas_2_semanas = 219, 87, 39
pontos_chicago_antes_da_reta_final, pontos_por_cidade_de_fora = 2.56, 4.5   # t(24) = 4,15, p < 0,001

# Experimento 4: vale-presente de uma confeitaria, 64 universitários (32 por prazo)
vale_usado = {'3 semanas': (10, 32), '2 meses': (2, 32)}
vale_previsto = {'3 semanas': 50, '2 meses': 68}        # % que outro grupo previu que usaria (N = 80)
# Experimento 5: ingresso de cinema, 240 universitários (120 por prazo)
cinema_usado = {'2 semanas': (59, 120), '6 semanas': (42, 120)}

aplica_fator_080 = False   # contagens e escolhas observadas, não proporção de pesquisa de opinião

print("=" * 70)
print("  DADOS DA LITERATURA - TURISTA NA PRÓPRIA CIDADE")
print("=" * 70)
print(f"\n  Shu & Gneezy (2010), Experimento 1, N = {n_entrevistados}")
print("  Pontos turísticos visitados pelo morador, por tempo de moradia:")
for k, v in moradores.items():
    print(f"  → {k:<16} {v:.1f}".replace('.', ','))
print("  Pontos turísticos visitados pelo visitante, por duração da visita:")
for k, v in visitantes.items():
    print(f"  → {k:<16} {v:.1f}".replace('.', ','))
print("  Moradores que conheciam mais pontos em outra cidade do que na sua:")
for k, (p, n) in por_cidade.items():
    print(f"  → {k:<8} {p}% de {n}")
print(f"\n  Experimento 2, N = {n_ex_moradores} ex-moradores de Chicago")
print(f"  → {visitas_ultimos_6_meses} de {visitas_total} visitas nos últimos seis meses, {visitas_ultimas_2_semanas} nas duas últimas semanas")
print("\n  Experimentos 4 e 5, vales com prazo curto e longo")
for k, (u, n) in vale_usado.items():
    print(f"  → Confeitaria, {k:<10} usaram {u} de {n}  ·  previsão de uso {vale_previsto[k]}%")
for k, (u, n) in cinema_usado.items():
    print(f"  → Cinema, {k:<10}      usaram {u} de {n}")
print(f"\n  Fator ×0,80 aplicado: {'sim' if aplica_fator_080 else 'não'}")
print("=" * 70)
```

```python
# --- O MODELO ---
# Quatro contas próprias sobre os números publicados. Semente 42, 200 mil sorteios.

# Camada 1 · Morador contra visitante
dif_media = media_visitante_2sem - media_morador_1ano
ep_dif = dif_media / t_exp1                       # erro padrão reconstruído a partir do t publicado
ic_dif = dif_media + np.array([-1, 1]) * stats.t.ppf(0.975, gl_exp1) * ep_dif
p_visitante_ve_mais = 1 - stats.t.cdf(0, gl_exp1, loc=dif_media, scale=ep_dif)

# Camada 2 · A resposta do dia: o morador conhece mais outra cidade do que a sua?
cont = {c: (int(round(p / 100 * n)), n) for c, (p, n) in por_cidade.items()}   # contagens reconstruídas
k_tot = sum(k for k, n in cont.values()); n_tot = sum(n for k, n in cont.values())
p_resposta = k_tot / n_tot
ic_resposta = jeffreys(k_tot, n_tot)
sim_resposta = sorteia(k_tot, n_tot)
p_maioria = (sim_resposta > 0.5).mean()
ic_cidade = {c: jeffreys(k, n) for c, (k, n) in cont.items()}
# numa mesa de dez moradores, quantos conhecem mais outra cidade
mesa = rng.binomial(10, sim_resposta)
mesa_tipica = np.percentile(mesa, [10, 50, 90])

# Camada 3 · A despedida
meses_moradia = anos_moradia_media * 12
fatia_6m_obs = visitas_ultimos_6_meses / visitas_total
fatia_6m_esperada = 6 / meses_moradia                         # se as visitas se espalhassem por igual
fatia_2s_obs = visitas_ultimas_2_semanas / visitas_total
fatia_2s_esperada = 2 / (anos_moradia_media * 52)
ic_6m = jeffreys(visitas_ultimos_6_meses, visitas_total)
ic_2s = jeffreys(visitas_ultimas_2_semanas, visitas_total)
vezes_6m = fatia_6m_obs / fatia_6m_esperada
vezes_2s = fatia_2s_obs / fatia_2s_esperada

# Camada 4 · O prazo
def compara(curto, longo):
    a = sorteia(*curto); b = sorteia(*longo)
    d = (a - b) * 100
    return curto[0] / curto[1] * 100, longo[0] / longo[1] * 100, np.median(d), *np.percentile(d, [2.5, 97.5]), (d > 0).mean()
vale = compara(vale_usado['3 semanas'], vale_usado['2 meses'])
cinema = compara(cinema_usado['2 semanas'], cinema_usado['6 semanas'])

fmt = lambda x, c=1: f"{x:.{c}f}".replace('.', ',')
print("=" * 70)
print("  MODELO - RESULTADO")
print("=" * 70)
print("\n  CAMADA 1 · Morador contra visitante")
print(f"  → Visitante de duas semanas {fmt(media_visitante_2sem)} pontos  ·  morador há menos de um ano {fmt(media_morador_1ano)}")
print(f"  → Diferença {fmt(dif_media)} [{fmt(ic_dif[0])}; {fmt(ic_dif[1])}]  ·  P(visitante vê mais) = {fmt(p_visitante_ve_mais*100, 1)}%")
print("\n  CAMADA 2 · A resposta do dia")
for c, (k, n) in cont.items():
    print(f"  → {c:<8} {k} de {n} = {fmt(k/n*100, 0)}% [{fmt(ic_cidade[c][0]*100, 0)}; {fmt(ic_cidade[c][1]*100, 0)}]")
print(f"  → As três cidades: {k_tot} de {n_tot} = {fmt(p_resposta*100)}% [{fmt(ic_resposta[0]*100)}; {fmt(ic_resposta[1]*100)}]")
print(f"  → P(mais da metade dos moradores conhece mais outra cidade) = {fmt(p_maioria*100, 0)}%")
print(f"  → Numa mesa de dez moradores: {mesa_tipica[1]:.0f} (de {mesa_tipica[0]:.0f} a {mesa_tipica[2]:.0f} em 80% das mesas)")
print("\n  CAMADA 3 · A despedida")
print(f"  → Últimos seis meses: {fmt(fatia_6m_obs*100)}% das visitas [{fmt(ic_6m[0]*100)}; {fmt(ic_6m[1]*100)}], contra {fmt(fatia_6m_esperada*100)}% se fossem espalhadas por igual ({fmt(vezes_6m)} vezes)")
print(f"  → Duas últimas semanas: {fmt(fatia_2s_obs*100)}% [{fmt(ic_2s[0]*100)}; {fmt(ic_2s[1]*100)}], contra {fmt(fatia_2s_esperada*100)}% ({fmt(vezes_2s, 0)} vezes)")
print(f"  → Antes da reta final: {fmt(pontos_chicago_antes_da_reta_final, 2)} pontos em Chicago contra {fmt(pontos_por_cidade_de_fora)} por cidade visitada")
print("\n  CAMADA 4 · O prazo")
print(f"  → Confeitaria: {fmt(vale[0], 0)}% com três semanas, {fmt(vale[1], 0)}% com dois meses  ·  diferença {fmt(vale[2], 0)} pp [{fmt(vale[3], 0)}; {fmt(vale[4], 0)}]  ·  P(prazo curto vence) = {fmt(vale[5]*100)}%")
print(f"  → Previsão de outro grupo: {vale_previsto['3 semanas']}% e {vale_previsto['2 meses']}%")
print(f"  → Cinema: {fmt(cinema[0], 0)}% com duas semanas, {fmt(cinema[1], 0)}% com seis  ·  diferença {fmt(cinema[2], 0)} pp [{fmt(cinema[3], 0)}; {fmt(cinema[4], 0)}]  ·  P(prazo curto vence) = {fmt(cinema[5]*100)}%")
print("=" * 70)
```

```python
# --- VISUALIZAÇÃO ---

# ── GRÁFICO 1 · Morador e visitante ──
fig1, (axa, axb) = plt.subplots(1, 2, figsize=(14, 7), sharey=True)
for ax, d, cor, tit in [(axa, visitantes, DOURADO, 'Quem está de visita\npor duração da visita'), (axb, moradores, VERDE, 'Quem mora na cidade\npor tempo de moradia')]:
    b = ax.bar(list(d.keys()), list(d.values()), color=cor, alpha=0.85, width=0.6)
    for r, v in zip(b, d.values()):
        ax.text(r.get_x() + r.get_width() / 2, v + 0.12, f'{v:.1f}'.replace('.', ','), ha='center', fontsize=14, fontweight='bold')
    ax.set_title(tit, fontsize=13, pad=10)
    ax.tick_params(axis='x', labelsize=11)
axa.set_ylabel('Pontos turísticos visitados (média)', fontsize=12)
axa.set_ylim(0, 6.6)
fig1.suptitle('Em duas semanas, o visitante vê mais do que o morador em um ano', fontsize=15)
plt.figtext(0.5, 0.01, 'Fonte: Shu & Gneezy, 2010 (199 entrevistados em Chicago, Londres e Dallas) | #365Probabilidades', ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.savefig('dia-161-grafico-01-morador-e-visitante.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 · A resposta do dia ──
fig2, ax2 = plt.subplots(figsize=(12, 8))
nomes = list(cont.keys()) + ['As três cidades']
vals = [k / n * 100 for k, n in cont.values()] + [p_resposta * 100]
ics = [ic_cidade[c] * 100 for c in cont] + [ic_resposta * 100]
cores = [CINZA] * 3 + [VERMELHO]
for i, (v, ic, cor) in enumerate(zip(vals, ics, cores)):
    ax2.barh(i, v, color=cor, alpha=0.85, height=0.55)
    ax2.errorbar(v, i, xerr=[[v - ic[0]], [ic[1] - v]], color='#222', capsize=7, lw=1.8)
    ax2.text(ic[1] + 1.5, i, f'{v:.0f}%', va='center', fontsize=15, fontweight='bold')
ax2.axvline(50, color='#222', ls='--', lw=1.3)
ax2.text(50.6, -0.72, 'metade', fontsize=11, color='#333')
ax2.set_yticks(range(4)); ax2.set_yticklabels(nomes, fontsize=13); ax2.invert_yaxis()
ax2.set_xlim(0, 100); ax2.set_ylim(3.6, -0.9)
ax2.set_xlabel('Moradores que conheciam mais pontos turísticos em outra cidade do que na sua (%)', fontsize=12)
ax2.set_title('A maioria dos moradores conhece melhor uma cidade de fora\n'
              f'{k_tot} de {n_tot} moradores nas três cidades', fontsize=14, pad=15)
plt.figtext(0.5, 0.01, 'Fonte: Shu & Gneezy, 2010. Contagens reconstruídas dos percentuais; barras com intervalo de Jeffreys de 95% | #365Probabilidades', ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=[0, 0.03, 1, 1])
plt.savefig('dia-161-grafico-02-a-resposta.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 · A despedida ──
fig3, ax3 = plt.subplots(figsize=(12, 8))
rot = ['Últimos seis meses', 'Duas últimas semanas']
obs = [fatia_6m_obs * 100, fatia_2s_obs * 100]
esp = [fatia_6m_esperada * 100, fatia_2s_esperada * 100]
icd = [ic_6m * 100, ic_2s * 100]
x = np.arange(2); larg = 0.36
ax3.bar(x - larg / 2, esp, larg, color=CINZA, alpha=0.8, label='se as visitas se espalhassem por igual')
ax3.bar(x + larg / 2, obs, larg, color=VERMELHO, alpha=0.85, label='o que aconteceu')
for i in range(2):
    ax3.errorbar(x[i] + larg / 2, obs[i], yerr=[[obs[i] - icd[i][0]], [icd[i][1] - obs[i]]], color='#222', capsize=7, lw=1.8)
    ax3.text(x[i] - larg / 2, esp[i] + 1, f'{esp[i]:.1f}%'.replace('.', ','), ha='center', fontsize=14, fontweight='bold')
    ax3.text(x[i] + larg / 2, icd[i][1] + 1, f'{obs[i]:.0f}%', ha='center', fontsize=15, fontweight='bold')
ax3.set_xticks(x); ax3.set_xticklabels(rot, fontsize=13)
ax3.set_ylim(0, 56)
ax3.set_ylabel('Parcela das visitas a pontos turísticos (%)', fontsize=12)
ax3.set_title('As visitas se acumulam quando a mudança tem data\n'
              f'{visitas_total} visitas de {n_ex_moradores} ex-moradores de Chicago, moradia média de {anos_moradia_media} anos', fontsize=14, pad=15)
ax3.legend(fontsize=12, loc='upper right')
plt.figtext(0.5, 0.01, 'Fonte: Shu & Gneezy, 2010. A parcela esperada é conta nossa com a moradia média | #365Probabilidades', ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=[0, 0.03, 1, 1])
plt.savefig('dia-161-grafico-03-a-despedida.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")

# ── GRÁFICO 4 · O prazo ──
fig4, ax4 = plt.subplots(figsize=(12, 8))
prazos = list(vale_usado.keys())
prev = [vale_previsto[p] for p in prazos]
real = [vale_usado[p][0] / vale_usado[p][1] * 100 for p in prazos]
icr = [jeffreys(*vale_usado[p]) * 100 for p in prazos]
ax4.bar(x - larg / 2, prev, larg, color=DOURADO, alpha=0.85, label='o que as pessoas previram')
ax4.bar(x + larg / 2, real, larg, color=VERDE, alpha=0.85, label='o que aconteceu')
for i in range(2):
    ax4.errorbar(x[i] + larg / 2, real[i], yerr=[[real[i] - icr[i][0]], [icr[i][1] - real[i]]], color='#222', capsize=7, lw=1.8)
    ax4.text(x[i] - larg / 2, prev[i] + 1.5, f'{prev[i]}%', ha='center', fontsize=15, fontweight='bold')
    ax4.text(x[i] + larg / 2, icr[i][1] + 1.5, f'{real[i]:.0f}%', ha='center', fontsize=15, fontweight='bold')
ax4.set_xticks(x); ax4.set_xticklabels([f'Prazo de {p}' for p in prazos], fontsize=13)
ax4.set_ylim(0, 85)
ax4.set_ylabel('Usaram o vale-presente (%)', fontsize=12)
ax4.set_title('Com mais prazo, as pessoas acham que vão usar mais, e usam menos\n'
              f'P(prazo curto leva mais gente a usar) = {vale[5]*100:.0f}%', fontsize=14, pad=15)
ax4.legend(fontsize=12, loc='upper left')
plt.figtext(0.5, 0.01, 'Fonte: Shu & Gneezy, 2010, Experimento 4 (64 universitários com o vale, 80 na previsão) | #365Probabilidades', ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=[0, 0.03, 1, 1])
plt.savefig('dia-161-grafico-04-o-prazo.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 4 salvo!")
```

### 💡 O Insight

**De cada dez moradores, seis conhecem mais pontos turísticos de outra cidade do que da sua.**

Foram 63 de 109 moradores de Chicago, Londres e Dallas, ou 58%, com intervalo de 48% a 67%. A chance de serem de fato a maioria é de 95%.

O visitante é mais rápido. Em duas semanas ele vê em média 4,4 pontos turísticos. Quem mora há menos de um ano viu 3,1. E seis em cada dez visitas do morador aconteceram porque havia alguém de fora para levar.

O que muda o comportamento do morador é a data. Entre quem se mudou de Chicago, 40% das visitas aconteceram nos últimos seis meses, e 18% nas duas últimas semanas. Se elas se espalhassem por igual pelos três anos de moradia, seriam 17% e 1%.

O vale-presente mostra o mesmo. Com três semanas de prazo, 31% usaram. Com dois meses, 6%. Quem foi perguntado antes previu o contrário, 50% e 68%.

Ter muito tempo para fazer uma coisa boa não aumentou a chance de ela ser feita. Diminuiu.

A pergunta que fica:

*Que passeio da sua cidade você faria neste mês se soubesse que vai se mudar no mês que vem?*

---

### ⚠️ Limitações do Modelo

- As amostras são pequenas e de conveniência. O Experimento 1 entrevistou pedestres que passavam, em três cidades de língua inglesa. Não há dado brasileiro.
- A resposta do dia vem de 109 moradores, e o intervalo de 48% a 67% inclui a metade. O artigo publica o percentual de cada cidade, não as contagens, que foram reconstruídas aqui.
- O número de pontos turísticos visitados foi lembrado pelo entrevistado. Lembrança de anos de moradia erra mais do que lembrança de uma viagem de duas semanas.
- A comparação entre morador e visitante não é entre pessoas iguais. Quem viaja escolheu aquela cidade para passear, e o morador pode ter outros interesses. Os autores tratam disso comparando a mesma pessoa em casa e fora.
- O erro padrão da diferença de médias foi reconstruído a partir do teste t publicado. A tabela do artigo registra 3,0 pontos para quem mora há menos de um ano, e o teste usa 3,1.
- Na camada da despedida, a parcela esperada é conta nossa e usa a moradia média de três anos para todos. A moradia variou de um a seis anos. O acúmulo no fim é dos autores (teste contra distribuição uniforme, p < 0,001).
- Ex-moradores que estudaram em Chicago tinham data de saída conhecida desde o início. O resultado pode não valer para quem se muda sem aviso.
- Os vales-presente foram dados a universitários e valiam cerca de seis dólares. O efeito do prazo pode ser diferente com valores maiores.
- No cinema, o efeito do prazo foi claro no conjunto e fraco em cada condição isolada. O texto usa o conjunto.
- A explicação pelo tempo livre esperado no futuro é dos autores. Os experimentos são compatíveis com ela e não a provam.

*A ciência é honesta sobre o que não sabe. O modelo também.*
