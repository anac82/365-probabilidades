## Qual a probabilidade de gastar dinheiro pra ganhar tempo te deixar mais feliz do que comprar coisas?

**Decisão analisada:** Vale mais gastar dinheiro em experiências e serviços que economizam tempo do que em produtos materiais?  
**Hashtag:** #365Probabilidades #Dia025

---

### 📖 A História

Tem uma conta que eu demorei muito tempo pra aprender a fazer.

Não é conta de banco. É uma conta mais simples — e mais difícil: quanto vale uma hora da minha vida?

Por anos, eu economizava dinheiro fazendo coisas que odiava fazer. Limpava a casa porque era "desperdício" pagar alguém. Cozinhava em série no domingo porque pedir comida era "extravagância". Fazia fila em cartório quando havia um serviço online que resolveria mais rápido — mas custava R$30.

E ao mesmo tempo, gastava dinheiro em coisas que achava que me fariam sentir melhor. Roupas que eu não precisava. Gadgets que eu usava por duas semanas. Assinaturas que esquecia de cancelar.

Pesquisadores da Harvard Business School fizeram uma pergunta que nunca havia passado pela minha cabeça: e se a gente estiver gastando dinheiro na direção errada?

---

### 📚 O Conceito: Comprar Tempo vs. Comprar Coisas

Em torno do mundo, o aumento de renda trouxe uma consequência não planejada: a sensação de que o tempo está sempre acabando. Pesquisadores chamam isso de **"time famine"** — a fome de tempo que acompanha a riqueza material crescente.

Whillans e colegas testaram uma hipótese simples: usar dinheiro para comprar tempo — pagar alguém para limpar a casa, pedir comida em vez de cozinhar, contratar um serviço que elimina uma tarefa — protegeria as pessoas dos efeitos negativos da pressão de tempo sobre o bem-estar.

A resposta veio de duas frentes: um levantamento com mais de 6 mil pessoas em 4 países, e um experimento de campo onde adultos foram randomizados para gastar $40 em compras que economizavam tempo versus compras materiais — e relataram como se sentiram depois.

---

### 🧮 O Modelo
**Fontes:**
- Whillans, A.V., Dunn, E.W., Smeets, P., Bekkers, R. & Norton, M.I. (2017) — *Buying time promotes happiness* — Proceedings of the National Academy of Sciences, 114(32), 8523-8527
  - Estudo observacional: N=6.271 (EUA, Canadá, Dinamarca, Holanda)
  - Experimento de campo: N=60 (adultos trabalhadores, $40 em 2 fins de semana consecutivos)
- Whillans, A.V., Weidman, A.C. & Dunn, E.W. (2016) — *Valuing time over money is associated with greater happiness* — Social Psychological and Personality Science, 7(3), 213-222


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print('✅ Bibliotecas carregadas')
```

```python
# --- DADOS DA LITERATURA ---
# Whillans et al. (2017) — PNAS — doi:10.1073/pnas.1706541114

# ESTUDO OBSERVACIONAL — N=6.271 — 4 países
# Pessoas que gastam dinheiro em servicos que economizam tempo
# relatam maior satisfação com a vida
# (associação, não causalidade — controlado por renda, horas trabalhadas)
n_observacional = 6271
n_paises = 4

# Proporção que gasta em serviços que economizam tempo
# (apenas 28% dos americanos gastam dinheiro em time-saving services —
#  isso independente de renda, o que surpreendeu os pesquisadores)
p_gasta_tempo = 0.28   # apenas 28% gastam em serviços que economizam tempo

# EXPERIMENTO DE CAMPO — N=60 — adultos trabalhadores
# Randomizado: $40 em compra que economiza tempo vs $40 em compra material
# Medida: felicidade relatada após cada fim de semana
# Resultado: maior felicidade após compra que economizou tempo
n_experimental = 60
# O paper reporta tamanho de efeito d de Cohen ~ 0.45 (efeito moderado)
d_cohen = 0.45  # tamanho de efeito moderado

# Proporção que prefere tempo a dinheiro (Whillans, Weidman & Dunn, 2016)
# Pessoas que valorizam tempo sobre dinheiro relatam maior felicidade
# (mesmo controlando por renda)
p_prefere_tempo = 0.48   # aproximadamente metade prefere tempo

fator_correcao = 0.80

print('=' * 65)
print('  DADOS DA LITERATURA — COMPRAR TEMPO VS COMPRAR COISAS')
print('=' * 65)
print(f'\n  Whillans et al. (2017) — PNAS:')
print(f'\n  Estudo observacional:')
print(f'  → N={n_observacional:,} pessoas em {n_paises} países (EUA, Canadá, Dinamarca, Holanda)')
print(f'  → Apenas {p_gasta_tempo*100:.0f}% gastam dinheiro em serviços que economizam tempo')
print(f'  → Quem gasta: maior satisfação com a vida (controlado por renda)')
print(f'\n  Experimento de campo (causal):')
print(f'  → N={n_experimental} adultos trabalhadores')
print(f'  → $40 em compra que economiza tempo vs $40 em compra material')
print(f'  → Resultado: mais felicidade após compra que economizou tempo')
print(f'  → Tamanho de efeito: d = {d_cohen} (efeito moderado)')
print(f'\n  Whillans, Weidman & Dunn (2016):')
print(f'  → {p_prefere_tempo*100:.0f}% preferem tempo a dinheiro em dilemas hipotéticos')
print(f'  → Essa preferência associada a mais felicidade (mesmo controlando renda)')
print(f'\n  Fator de correção: x{fator_correcao}')
print('=' * 65)
print()
print('  ⚠️  NOTA METODOLÓGICA:')
print('  O estudo observacional (N=6.271) é correlacional — não prova que')
print('  gastar em tempo CAUSA felicidade. O experimento de campo (N=60)')
print('  fornece evidência causal, mas com amostra pequena. Os dois juntos')
print('  são mais robustos do que qualquer um separado.')
```

```python
# --- O MODELO ---

# Potência estatística do experimento (N=60, d=0.45, alpha=0.05)
# Para detectar d=0.45 com N=30 por grupo, potência ≈ 0.57
# (experimento modestamente sub-potente — valor de p foi significativo, mas marginal)
n_por_grupo = n_experimental / 2
potencia_aprox = 1 - stats.norm.cdf(1.645 - d_cohen * np.sqrt(n_por_grupo / 2))

# Estimativa do ganho em bem-estar ao comprar tempo em vez de coisas
# Usando d=0.45 como referência: efeito moderado e positivo
# Probabilidade de sentir mais felicidade na semana seguinte
# P(X > 0) onde X ~ N(d, 1) = P(Z > -d) = Φ(d)
p_mais_feliz = stats.norm.cdf(d_cohen)
p_mais_feliz_corrigido = stats.norm.cdf(d_cohen * fator_correcao)

# Apenas 28% gastam em tempo — o que o resto faz com o dinheiro
p_nao_gasta_tempo = 1 - p_gasta_tempo

print('=' * 65)
print('  MODELO — RESULTADO')
print('=' * 65)
print(f'\n  Potência do experimento (N={n_experimental}, d={d_cohen}):')
print(f'  → ~{potencia_aprox*100:.0f}% de chance de detectar o efeito se ele existe')
print(f'    (modestamente sub-potente — resultado deve ser interpretado com cautela)')
print(f'\n  P(mais feliz ao gastar em tempo vs. em coisas):')
print(f'  → Estimativa (d={d_cohen}): {p_mais_feliz*100:.0f}%')
print(f'  → Corrigida (×{fator_correcao}): {p_mais_feliz_corrigido*100:.0f}%')
print(f'\n  Gap de adoção: quem gasta em tempo vs quem poderia:')
print(f'  → Apenas {p_gasta_tempo*100:.0f}% gastam em serviços que economizam tempo')
print(f'  → {p_nao_gasta_tempo*100:.0f}% não o fazem — mesmo quando têm renda para isso')
print(f'\n  PARADOXO CENTRAL:')
print(f'  → A estratégia de gasto que mais aumenta felicidade é a menos usada')
print('=' * 65)
```

```python
# --- VISUALIZACAO ---

# GRAFICO 1 — Quem gasta em tempo vs quem nao gasta (gap de adocao)
fig1, ax1 = plt.subplots(figsize=(12, 8))

categorias = ['Gasta dinheiro em serviços\nque economizam tempo', 'Não gasta — mesmo\ntendo renda para isso']
valores = [p_gasta_tempo * 100, p_nao_gasta_tempo * 100]
cores1 = ['#2a8a82', '#888888']

bars = ax1.bar(categorias, valores, color=cores1, alpha=0.85, width=0.45)
ax1.set_ylim(0, 90)
ax1.set_ylabel('Proporção (%)', fontsize=12)
ax1.set_title('A estratégia que mais aumenta felicidade é a menos usada\nWhillans et al. (2017) — N=6.271 — 4 países',
              fontsize=13, pad=15)

for bar, v in zip(bars, valores):
    ax1.text(bar.get_x() + bar.get_width() / 2, v + 1.5,
             f'{v:.0f}%', ha='center', fontweight='bold', fontsize=18)

plt.tight_layout()
plt.savefig('dia-025-grafico-01-gap-adocao.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Gráfico 1 salvo!')


# GRAFICO 2 — Experimento: compra que economiza tempo vs compra material
fig2, ax2 = plt.subplots(figsize=(12, 8))

# Ilustracao do efeito experimental (d=0.45)
x = np.linspace(-3, 4, 1000)
y_controle = stats.norm.pdf(x, loc=0, scale=1)
y_tempo = stats.norm.pdf(x, loc=d_cohen, scale=1)

ax2.plot(x, y_controle, color='#888888', linewidth=2.5, label='Compra material ($40)')
ax2.plot(x, y_tempo, color='#2a8a82', linewidth=2.5, label='Compra que economiza tempo ($40)')
ax2.fill_between(x, y_tempo, y_controle,
                 where=(y_tempo > y_controle), alpha=0.2, color='#2a8a82',
                 label=f'Vantagem da compra de tempo (d={d_cohen})')

ax2.axvline(x=0, color='#888888', linestyle='--', linewidth=1.5)
ax2.axvline(x=d_cohen, color='#2a8a82', linestyle='--', linewidth=1.5)
ax2.set_xlabel('Felicidade relatada (padronizada)', fontsize=12)
ax2.set_ylabel('Densidade', fontsize=12)
ax2.set_title('Comprar tempo torna pessoas mais felizes do que comprar coisas\nWhillans et al. (2017) — experimento de campo · N=60 · d=0.45',
              fontsize=13, pad=15)
ax2.legend(fontsize=11)

plt.tight_layout()
plt.savefig('dia-025-grafico-02-experimento.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Gráfico 2 salvo!')


# GRAFICO 3 — Por que nao compramos tempo: o paradoxo do uso do dinheiro
fig3, ax3 = plt.subplots(figsize=(12, 8))

categorias3 = ['Prefere tempo\na dinheiro\n(Whillans 2016)', 'Gasta em serviços\nque economizam\ntempo\n(Whillans 2017)',
               'Benefício estimado\n(mais feliz ao\ncomprar tempo)\n(Whillans 2017)']
valores3 = [p_prefere_tempo * 100, p_gasta_tempo * 100, p_mais_feliz * 100]
cores3 = ['#c9a14a', '#888888', '#2a8a82']

bars3 = ax3.bar(categorias3, valores3, color=cores3, alpha=0.85, width=0.45)
ax3.set_ylim(0, 90)
ax3.set_ylabel('Proporção / Probabilidade (%)', fontsize=12)
ax3.set_title('O paradoxo: sabemos que tempo importa, mas não compramos\nWhillans et al. (2016, 2017)',
              fontsize=13, pad=15)

for bar, v in zip(bars3, valores3):
    ax3.text(bar.get_x() + bar.get_width() / 2, v + 1.5,
             f'{v:.0f}%', ha='center', fontweight='bold', fontsize=16)

plt.tight_layout()
plt.savefig('dia-025-grafico-03-paradoxo.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Gráfico 3 salvo!')
```

### 💡 O Insight

Em um levantamento com **6.271 pessoas em 4 países**, indivíduos que gastavam dinheiro em serviços que economizavam tempo relataram maior satisfação com a vida — mesmo quando controlado por renda, horas trabalhadas e outras variáveis. E em um experimento de campo onde adultos foram randomizados para gastar $40 em compra que economizaria tempo versus $40 em compra material, relataram mais felicidade após a compra de tempo.

O paradoxo: **apenas 28% das pessoas** gastam dinheiro em serviços que economizam tempo — independentemente de renda. Mesmo entre pessoas com alta renda, a maioria não faz essa troca.

Por quê? Os pesquisadores sugerem que comprar tempo parece "preguiça", enquanto comprar coisas parece "conquista". Há um julgamento moral implícito em pagar alguém para fazer algo que você poderia fazer — mesmo quando essa escolha seria melhor para o seu bem-estar.

A pergunta que fica não é "quanto dinheiro você tem". É: o que você está fazendo com o tempo que o seu dinheiro poderia comprar?

*Qual é a tarefa que você odeia fazer — e que poderia pagar alguém para fazer por você — mas não paga porque parece desperdício?*

---

### ⚠️ Limitações do Modelo
- O estudo observacional (N=6.271) é correlacional — não prova que comprar tempo causa felicidade; quem já é mais feliz pode ser mais propenso a comprar tempo
- O experimento de campo (N=60) fornece evidência causal, mas com amostra pequena; a potência estatística é modesta (~57%), o que significa que o resultado deve ser replicado
- Os participantes do experimento eram adultos trabalhadores americanos — pode não generalizar para outros contextos culturais, econômicos ou de tempo disponível
- A categoria "serviços que economizam tempo" é ampla: inclui desde limpeza doméstica até delivery e transporte. O efeito pode variar por tipo de serviço e por quanto a tarefa evitada era aversiva para aquela pessoa
- O valor de $40 pode ter efeitos diferentes dependendo da renda de quem gasta — gastar $40 num serviço representa percentuais muito diferentes do orçamento de pessoas com rendas distintas
- O fator de correção (×0,80) foi aplicado ao tamanho de efeito d para o cálculo de probabilidade

*A ciência é honesta sobre o que não sabe. O modelo também.*
