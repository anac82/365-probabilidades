### 📖 A História

A guerra do retorno ao escritório já tem anos de trincheira.

De um lado, empresas convocando todo mundo de volta, convencidas de que
casa é sinônimo de sofá. Do outro, gente jurando que rende o dobro longe
do open office. Dos dois lados, munição de sobra.

Opinião, quase toda. Dado, quase nenhum.

O detalhe que pouca gente sabe: essa briga foi respondida antes de existir.

Em 2010, uma empresa chinesa de 16 mil funcionários deixou um economista
de Stanford fazer o que ninguém tinha feito: **sortear** quem trabalharia
de casa. Aniversário par, fica em casa 4 dias por semana. Ímpar, escritório.
Nove meses. Desempenho medido por sistema, ligação por ligação.

Não era pesquisa de opinião. Era experimento.

E o resultado desmontou os dois lados da trincheira ao mesmo tempo.

---

### 📚 O Conceito: Medir em Vez de Opinar

O problema de comparar quem trabalha de casa com quem trabalha no
escritório é que ninguém vai para casa por sorteio. Vai quem quer, quem
pode, quem tem filho pequeno, quem mora longe. Qualquer diferença de
desempenho pode ser das pessoas, não do lugar.

A randomização quebra isso. Aniversário par ou ímpar não escolhe ninguém:
os dois grupos ficam, em média, iguais em tudo, menos no lugar onde
trabalham. O que sobrar de diferença é efeito do lugar.

E o estudo ainda entregou um segundo conceito de presente: **seleção**.
Quando o experimento acabou e a empresa liberou a escolha, metade dos
voluntários mudou de ideia e voltou para o escritório. Quem ficou em casa
era quem funcionava em casa. E o ganho quase dobrou.

O efeito do home office existe. O efeito de saber quem deve fazer home
office é quase do mesmo tamanho.

### 🧮 O Modelo

**Fontes:**
- Bloom, N., Liang, J., Roberts, J. & Ying, Z. J. (2015) - Does Working
  from Home Work? Evidence from a Chinese Experiment - Quarterly Journal
  of Economics, 130(1), 165-218 - call center da Ctrip (Xangai),
  2010-2012. Dos 996 elegíveis, 503 se voluntariaram; 249 qualificados
  (quarto próprio, banda larga, 6+ meses de casa) foram randomizados por
  aniversário: **131 em casa, 118 no escritório**, 9 meses.
  **Fonte central.**

**Âncoras publicadas:** desempenho +13%, decomposto em +9% de minutos
trabalhados por turno (menos pausas e atestados) e +4% de ligações por
minuto (ambiente mais silencioso); rotatividade caiu cerca de metade;
satisfação subiu; a probabilidade de promoção, condicional ao
desempenho, caiu (direção reportada pelos autores; a magnitude varia por
especificação e não é ancorada aqui). Após o experimento, com escolha
livre e reseleção, o ganho subiu para ~22%.

**Fator ×0,80: não se aplica.** Experimento randomizado com desempenho
registrado por sistema. Sem autorrelato, sem correção.

**Suposição declarada:** a decomposição 13% = 9% + 4% é aproximadamente
multiplicativa (1,09 × 1,04 ≈ 1,134); o arredondamento é o dos próprios
autores. O cálculo de dias-equivalentes abaixo é conta própria sobre a
âncora publicada.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Bloom, Liang, Roberts & Ying (2015), QJE 130(1), 165-218
# Call center da Ctrip, Xangai, 2010-2012

n_elegiveis = 996
n_voluntarios = 503
n_qualificados = 249
n_casa = 131          # tratamento: 4 dias por semana em casa
n_escritorio = 118    # controle
meses = 9

efeito_desempenho = 0.13     # +13% de desempenho total
efeito_minutos = 0.09        # +9% de minutos por turno
efeito_ligacoes = 0.04       # +4% de ligações por minuto
reducao_rotatividade = 0.50  # rotatividade caiu ~metade
ganho_pos_selecao = 0.22     # ~+22% após reseleção voluntária

# Promoção condicional ao desempenho: CAIU (direção reportada;
# magnitude varia por especificação e não é ancorada aqui)

# Fator de correção: NÃO se aplica (RCT, desempenho medido por sistema)

print("=" * 65)
print("  DADOS DA LITERATURA - HOME OFFICE NUM EXPERIMENTO DE VERDADE")
print("=" * 65)
print(f"\n  O funil do experimento:")
print(f"  → Elegíveis:                        {n_elegiveis}")
print(f"  → Voluntários:                      {n_voluntarios}")
print(f"  → Qualificados e randomizados:      {n_qualificados}")
print(f"  → Casa {n_casa} · Escritório {n_escritorio} · {meses} meses")
print(f"\n  Resultados:")
print(f"  → Desempenho:                       +{efeito_desempenho*100:.0f}%")
print(f"  → ... minutos por turno:            +{efeito_minutos*100:.0f}%")
print(f"  → ... ligações por minuto:          +{efeito_ligacoes*100:.0f}%")
print(f"  → Rotatividade:                     caiu ~{reducao_rotatividade*100:.0f}%")
print(f"  → Pós-seleção voluntária:           ~+{ganho_pos_selecao*100:.0f}%")
print(f"  → Promoção condicional ao desempenho: CAIU (sem magnitude ancorada)")
print(f"\n  Fator ×0,80: NÃO aplicado (RCT, medição por sistema)")
print("=" * 65)
```

```python
# --- O MODELO ---
# A anatomia de um efeito: de onde vêm os 13%, e o que a seleção acrescenta

# Passo 1 - checagem da decomposição (aproximadamente multiplicativa)
efeito_composto = (1 + efeito_minutos) * (1 + efeito_ligacoes) - 1

# Passo 2 - conta própria: dias-equivalentes de trabalho
# +13% sobre uma semana de 5 dias úteis
dias_equivalentes_semana = 5 * efeito_desempenho
dias_para_um_dia_extra = 1 / dias_equivalentes_semana  # semanas p/ 1 dia extra

# Passo 3 - o valor de saber quem deve ficar em casa
valor_selecao = ganho_pos_selecao - efeito_desempenho

print("=" * 65)
print("  MODELO - A ANATOMIA DOS 13%")
print("=" * 65)
print(f"\n  Decomposição: (1,09 × 1,04) - 1 = {efeito_composto*100:.1f}%  ≈  13%")
print(f"  → {efeito_minutos*100:.0f}% de mais minutos: menos pausas, menos atestados")
print(f"  → {efeito_ligacoes*100:.0f}% de mais ligações por minuto: silêncio")
print(f"\n  Em dias-equivalentes (conta própria, suposição declarada):")
print(f"  → +{dias_equivalentes_semana:.2f} dia útil por semana")
print(f"  → um dia útil extra a cada ~{dias_para_um_dia_extra:.1f} semanas")
print(f"\n  O valor da seleção:")
print(f"  → Sorteados: +{efeito_desempenho*100:.0f}%")
print(f"  → Voluntários que se reescolheram: ~+{ganho_pos_selecao*100:.0f}%")
print(f"  → Saber QUEM funciona em casa vale ~{valor_selecao*100:.0f} p.p.,")
print(f"     quase o tamanho do efeito do home office em si")
print(f"\n  O CONTRAPESO:")
print(f"  → Rendeu mais, saiu menos, foi promovido menos")
print(f"     (condicional ao desempenho; direção reportada pelos autores)")
print(f"  → Longe dos olhos, longe da lista de promoção")
print("=" * 65)
```

```python
# --- VISUALIZAÇÃO - GRÁFICOS SEPARADOS ---

# ── GRÁFICO 1 - O resultado em três barras ──
fig1, ax1 = plt.subplots(figsize=(10, 6))

grupos = ['Escritório\n(controle)', 'Casa\n(sorteados)', 'Casa\n(pós-seleção voluntária)']
indices = [100, 100*(1+efeito_desempenho), 100*(1+ganho_pos_selecao)]
cores = ['#c9a14a', '#2a8a82', '#1a5f5a']

bars = ax1.bar(grupos, indices, color=cores, alpha=0.88, width=0.5)
ax1.set_ylim(90, 128)
ax1.set_ylabel('Desempenho (escritório = 100)')
ax1.set_title('Home office num experimento de verdade\n'
              'Ctrip, 9 meses, randomização por aniversário (Bloom et al., 2015)',
              fontsize=12.5, pad=15)

for bar, v in zip(bars, indices):
    ax1.text(bar.get_x() + bar.get_width()/2, v + 0.7,
             f'{v:.0f}', ha='center', fontweight='bold', fontsize=15)

ax1.annotate('+13%', xy=(1, 113), xytext=(0.62, 118),
             fontsize=13, color='#2a8a82', fontweight='bold')
ax1.annotate('~+22%', xy=(2, 122), xytext=(1.62, 125.5),
             fontsize=13, color='#1a5f5a', fontweight='bold')

plt.figtext(0.5, 0.01,
            'Fonte: Bloom, Liang, Roberts & Ying (2015), QJE 130(1) | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-087-grafico-01-resultado.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 - O funil do experimento ──
fig2, ax2 = plt.subplots(figsize=(10, 6))

etapas = ['Elegíveis', 'Voluntários', 'Qualificados\ne sorteados', 'Casa', 'Escritório']
valores = [n_elegiveis, n_voluntarios, n_qualificados, n_casa, n_escritorio]
cores2 = ['#c9a14a', '#c9a14a', '#c9a14a', '#2a8a82', '#c0392b']

bars2 = ax2.barh(etapas[::-1], valores[::-1], color=cores2[::-1], alpha=0.88, height=0.55)
ax2.set_xlabel('Pessoas')
ax2.set_title('Medir em vez de opinar: o funil do experimento\n'
              'Aniversário par fica em casa, ímpar no escritório',
              fontsize=12.5, pad=15)

for bar, v in zip(bars2, valores[::-1]):
    ax2.text(v + 12, bar.get_y() + bar.get_height()/2,
             f'{v}', va='center', fontweight='bold', fontsize=13)

ax2.set_xlim(0, 1150)

plt.figtext(0.5, 0.01,
            'Fonte: Bloom et al. (2015) | randomização por paridade do aniversário | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-087-grafico-02-funil.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 - O que subiu e o que caiu ──
fig3, ax3 = plt.subplots(figsize=(10, 6))

medidas = ['Desempenho', 'Rotatividade']
deltas = [efeito_desempenho*100, -reducao_rotatividade*100]
cores3 = ['#2a8a82', '#c9a14a']

bars3 = ax3.bar(medidas, deltas, color=cores3, alpha=0.88, width=0.42)
ax3.axhline(0, color='#333', linewidth=1)
ax3.set_ylim(-62, 28)
ax3.set_ylabel('Variação (%)')
ax3.set_title('O placar do home office, e o contrapeso\n'
              'Bloom et al. (2015), 9 meses de experimento',
              fontsize=12.5, pad=15)

for bar, v in zip(bars3, deltas):
    va = 'bottom' if v > 0 else 'top'
    off = 1.5 if v > 0 else -1.5
    ax3.text(bar.get_x() + bar.get_width()/2, v + off,
             f'{v:+.0f}%', ha='center', va=va, fontweight='bold', fontsize=15)

ax3.text(0.5, 0.10,
         'O contrapeso: promoção, condicional ao desempenho, CAIU.\n'
         'Direção reportada pelos autores; magnitude varia por especificação.\n'
         'Longe dos olhos, longe da lista de promoção.',
         transform=ax3.transAxes, ha='center', fontsize=10.5,
         color='#c0392b', style='italic')

plt.figtext(0.5, 0.01,
            'Fonte: Bloom, Liang, Roberts & Ying (2015), QJE 130(1) | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-087-grafico-03-contrapeso.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**Trabalhar de casa rendeu 13% a mais. Num sorteio, não numa opinião.**

Nove por cento vieram de mais minutos no turno: menos pausas, menos
atestados. Quatro por cento vieram do silêncio: mais ligações por minuto.
Somados, quase um dia útil extra a cada semana e meia.

A rotatividade caiu pela metade. A satisfação subiu.

E quando a empresa liberou a escolha, veio o achado mais fino:
metade dos voluntários voltou para o escritório. Quem ficou era quem
funcionava em casa. O ganho quase dobrou, para 22%.

**Saber quem deve trabalhar de casa vale quase tanto quanto
o home office em si.**

Falta o contrapeso, com o mesmo peso: quem trabalhou de casa rendeu
mais, saiu menos, e foi **menos promovido**, condicional ao desempenho.
Longe dos olhos, longe da lista de promoção.

O home office funciona. A pergunta que sobra não é sobre produtividade.
É sobre o que a sua empresa mede quando decide quem sobe.

A pergunta que fica:

*Quantas brigas na sua vida seguem abertas só porque
ninguém teve coragem de medir?*

### ⚠️ Limitações do Modelo

- Um call center, uma empresa, China de 2010-2012: tarefas mensuráveis
  por sistema; generalização para trabalho criativo ou colaborativo é
  limitada e declarada
- Efeito estimado sobre voluntários qualificados (efeito local): quem
  não se voluntariou ou não tinha quarto próprio está fora da estimativa
- A queda na promoção condicional ao desempenho tem direção reportada
  pelos autores, mas a magnitude varia por especificação e não foi
  ancorada aqui
- Os dias-equivalentes são conta própria sobre a âncora publicada,
  suposição declarada
- Existe atualização com trabalho híbrido (Bloom et al., 2024, Nature,
  N=1.612): sem perda de desempenho e menos pedidos de demissão; o
  desenho de 2015 segue sendo o experimento mais limpo do tema

*A ciência é honesta sobre o que não sabe. O modelo também.*
