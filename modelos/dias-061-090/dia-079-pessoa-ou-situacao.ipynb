### 📖 A História

Toda avaliação de pessoa funciona do mesmo jeito: alguém observa um episódio e conclui
uma característica.

Ele foi ríspido na reunião, logo é uma pessoa difícil. Ela entregou atrasado, logo é
desorganizada. Aquele candidato foi honesto quando podia não ser, logo é confiável.

Processo seletivo, avaliação de desempenho, primeira impressão em jantar de família:
todos fazem a mesma operação. Um episódio entra, um adjetivo sai.

A pergunta que quase ninguém faz é quanta informação existe de fato num episódio.

E essa pergunta tem quase cem anos.

---

### 📚 O Conceito: o que é um traço, afinal

A discussão tem nome próprio na psicologia: **debate pessoa contra situação**.

De um lado, a ideia de que existem características estáveis que a pessoa carrega de um
lugar para o outro. Do outro, a ideia de que o comportamento é, sobretudo, resposta ao
contexto.

Parece disputa filosófica. Não é. É uma pergunta empírica com resposta medida, e a
resposta acabou sendo mais interessante que qualquer um dos dois lados.

O que se descobriu é que **as duas coisas são verdade em escalas diferentes**. Um
episódio isolado carrega pouquíssima informação sobre a pessoa. A soma de muitos
episódios carrega muita.

A diferença entre esses dois níveis é matemática, e dá para calcular.

---

### 🧮 O Modelo

Um estudo fundador de 1928, uma correção dos anos 1970 e a síntese moderna, com
amostragem de experiência.

**Fontes:**
- Hartshorne, H. & May, M. A., 1928 · *Studies in the Nature of Character: Studies in
  Deceit* (Character Education Inquiry) · estudo comportamental com crianças em idade
  escolar, medindo honestidade e engano em situações reais · **mesmo entre situações
  bastante parecidas, comportamento honesto e desonesto apareceram de forma
  inconsistente** · algumas crianças colavam na prova mas não pegavam lembrancinhas de
  festa; outras quebravam regras numa competição esportiva mas não colavam · a
  conclusão dos autores foi que honestidade não seria uma entidade interna, e sim
  **função da situação**
- Mischel, W., 1968 · sustenta que correlações em torno de **0,30** entre medidas do
  mesmo traço, e entre traço e comportamento subsequente, representariam o **limite
  superior** da consistência
- Epstein, S., 1979 · "The stability of behavior" · reanálise dos estudos clássicos ·
  quando estabilidade temporal e consistência entre situações são separadas
  corretamente, as correlações entre **instâncias particulares** ficam na faixa de
  **0,2 a 0,3**, mas os **coeficientes agregados** ficam entre **0,6 e 0,9** ·
  registra ainda que **os próprios colaboradores e sucessores de Hartshorne e May
  discordaram da interpretação deles**
- Fleeson, W., 2001 · "Toward a structure- and process-integrated view of personality:
  Traits as density distributions of states" · *JPSP* 80(6), 1011-1027 · **três estudos
  de amostragem de experiência**, ao longo de **duas a três semanas** de vida cotidiana
  · a variabilidade **dentro** da pessoa é alta, e o indivíduo típico manifesta,
  regular e rotineiramente, **quase todos os níveis de todos os traços** · as
  **tendências centrais** das distribuições individuais são **quase perfeitamente
  estáveis** · a **quantidade de variabilidade** também é uma diferença individual
  estável
- Estimativas posteriores de correlação intraclasse na expressão de traços em contexto
  ficam entre **0,20 e 0,40**, o que implica que o contexto pesa de **duas a três
  vezes mais** que as diferenças estáveis entre pessoas numa observação isolada
- Mischel, W. & Shoda, Y. · assinaturas comportamentais do tipo **se isto, então
  aquilo**: a estabilidade não está no comportamento bruto, e sim no padrão de resposta
  a cada tipo de situação

**Pendência declarada:** o N do estudo de 1928 não foi localizado nas fontes
acessíveis. Sabe-se que foi um estudo comportamental de grande escala com escolares.

**Nota metodológica sobre o fator ×0.80:** não se aplica. Observação comportamental
direta em 1928, e amostragem de experiência em tempo real em 2001, não survey
retrospectivo.


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

RNG = np.random.default_rng(42)

print("Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---

# Mischel, 1968: o "teto de 0,30"
teto_mischel = 0.30

# Epstein, 1979: a correcao
r_instancia = (0.20, 0.30)      # correlacao entre instancias particulares
r_agregado  = (0.60, 0.90)      # coeficientes agregados

# Fleeson, 2001: amostragem de experiencia
semanas_fleeson = (2, 3)
n_estudos_fleeson = 3
ACHADOS_FLEESON = [
    'variabilidade DENTRO da pessoa e alta',
    'a pessoa tipica manifesta quase TODOS os niveis de TODOS os tracos',
    'a tendencia central de cada pessoa e quase perfeitamente estavel',
    'a QUANTIDADE de variabilidade tambem e estavel entre pessoas',
]

# Correlacao intraclasse posterior: quanto da variacao e entre pessoas
icc = (0.20, 0.40)

n_1928 = None   # NAO LOCALIZADO. Nao inventar.

aplica_fator_080 = False

print("=" * 72)
print("  DADOS - UM EPISODIO CONTRA UMA DISTRIBUICAO")
print("=" * 72)
print(f"\n  Hartshorne & May, 1928:")
print(f"  -> Criancas honestas numa situacao eram desonestas em outra")
print(f"  -> Conclusao deles: honestidade e 'funcao da situacao'")
print(f"  -> N: NAO LOCALIZADO nas fontes acessiveis")
print(f"\n  Mischel, 1968:")
print(f"  -> Teto de {teto_mischel:.2f} para consistencia de traco")
print(f"\n  Epstein, 1979 (a correcao):")
print(f"  -> Entre INSTANCIAS particulares: {r_instancia[0]:.1f} a {r_instancia[1]:.1f}")
print(f"  -> Coeficientes AGREGADOS:        {r_agregado[0]:.1f} a {r_agregado[1]:.1f}")
print(f"  -> Os proprios colaboradores de Hartshorne e May discordavam")
print(f"     da interpretacao situacionista.")
print(f"\n  Fleeson, 2001 ({n_estudos_fleeson} estudos, "
      f"{semanas_fleeson[0]} a {semanas_fleeson[1]} semanas):")
for a in ACHADOS_FLEESON:
    print(f"  -> {a}")
print(f"\n  Correlacao intraclasse posterior: {icc[0]:.2f} a {icc[1]:.2f}")
print(f"  -> O contexto pesa de 2 a 3 vezes mais que o traco estavel")
print(f"     numa observacao isolada.")
print(f"\n  Fator x0.80 aplicado: {aplica_fator_080}")
print("=" * 72)
```

```python
# --- O MODELO ---
# Assinatura estatistica: a formula de Spearman-Brown.
# Ela explica, com uma linha de algebra, por que um episodio prediz mal
# e vinte episodios predizem bem.
#
#     r_k = k * r / (1 + (k - 1) * r)
#
# onde r e a confiabilidade de UMA observacao e k e o numero de observacoes.

def spearman_brown(r, k):
    return (k * r) / (1 + (k - 1) * r)


r_uma = 0.25     # ponto medio da faixa de instancia isolada (Epstein)
observacoes = [1, 2, 3, 5, 10, 20, 50]

# Quantas observacoes para atingir cada patamar
def k_para(alvo, r=r_uma, teto=500):
    for k in range(1, teto):
        if spearman_brown(r, k) >= alvo:
            return k
    return None


alvos = [0.60, 0.70, 0.80, 0.90]

print("=" * 72)
print("  MODELO - POR QUE UM EPISODIO NAO DIZ NADA E VINTE DIZEM MUITO")
print("=" * 72)
print(f"\n  Formula de Spearman-Brown: r_k = k*r / (1 + (k-1)*r)")
print(f"  Confiabilidade de UMA observacao: r = {r_uma:.2f}")
print(f"     (ponto medio da faixa de 0,20 a 0,30 de Epstein)")
print(f"\n  {'observacoes':>12} | {'confiabilidade':>14}")
print(f"  {'-'*12}-+-{'-'*14}")
for k in observacoes:
    print(f"  {k:>12} | {spearman_brown(r_uma, k):>14.2f}")
print(f"\n  Quantas observacoes para chegar em cada patamar:")
for alvo in alvos:
    print(f"  -> para {alvo:.2f}: {k_para(alvo)} observacoes")
print(f"\n  LEITURA:")
print(f"  -> Uma observacao isolada: {spearman_brown(r_uma, 1):.2f}. Quase nada.")
print(f"  -> Vinte observacoes: {spearman_brown(r_uma, 20):.2f}. Muita coisa.")
print(f"  -> A faixa agregada de Epstein, de {r_agregado[0]:.1f} a"
      f" {r_agregado[1]:.1f}, e reproduzida")
print(f"     exatamente por essa formula entre"
      f" {k_para(r_agregado[0])} e {k_para(r_agregado[1])} observacoes.")
print(f"\n  -> Hartshorne e May estavam certos sobre a INSTANCIA")
print(f"     e errados sobre a CONCLUSAO.")

# Traducao pratica
print(f"\n  Em linguagem de segunda-feira:")
print(f"  -> Julgar alguem por 1 episodio: confiabilidade de"
      f" {spearman_brown(r_uma, 1):.2f}")
print(f"  -> Julgar por 1 semana de convivencia (digamos 5 episodios):"
      f" {spearman_brown(r_uma, 5):.2f}")
print(f"  -> Julgar por um trimestre inteiro (digamos 20):"
      f" {spearman_brown(r_uma, 20):.2f}")
print(f"     Numero de episodios por periodo e ilustrativo, nao publicado.")
print("=" * 72)
```

```python
# --- VISUALIZACAO ---

DOURADO = '#c8a84b'
VERMELHO = '#c0392b'
VERDE = '#2a8a82'
CINZA = '#6b6a64'

# GRAFICO 1 - Assinatura: a curva de Spearman-Brown
fig1, ax1 = plt.subplots(figsize=(12, 8))

ks = np.arange(1, 51)
curva = [spearman_brown(r_uma, k) for k in ks]
ax1.plot(ks, curva, color=DOURADO, linewidth=3.4)
ax1.fill_between(ks, curva, alpha=0.12, color=DOURADO)

ax1.axhspan(r_agregado[0], r_agregado[1], color=VERDE, alpha=0.10)
ax1.text(48, (r_agregado[0] + r_agregado[1]) / 2,
         'faixa agregada\nde Epstein, 1979', ha='right', va='center', fontsize=12,
         color=VERDE, fontweight='bold')

for k in (1, 5, 20):
    v = spearman_brown(r_uma, k)
    ax1.scatter([k], [v], s=180, color=VERMELHO if k == 1 else DOURADO, zorder=3)
    ax1.text(k + 0.8, v - 0.045, ('1 observação' if k == 1 else f'{k} observações') +
             f'\n{v:.2f}'.replace('.', ','), fontsize=12, fontweight='bold',
             color=VERMELHO if k == 1 else '#333')

ax1.set_xlim(0, 51)
ax1.set_ylim(0, 1.0)
ax1.set_xlabel('Número de observações da mesma pessoa')
ax1.set_ylabel('Confiabilidade da conclusão sobre o traço')
ax1.set_title('A assinatura estatística: por que um episódio não diz nada\n'
              'Fórmula de Spearman-Brown, com r = 0,25 para uma observação isolada',
              fontsize=15, pad=18)
ax1.text(0.5, -0.135,
         'r_k = k·r / (1 + (k−1)·r)   ·   A faixa de 0,6 a 0,9 que Epstein encontrou '
         'nos dados agregados é reproduzida por esta fórmula.',
         transform=ax1.transAxes, ha='center', fontsize=12, color=DOURADO,
         fontweight='bold')

plt.figtext(0.5, 0.005,
            'Cálculo próprio sobre a faixa publicada por Epstein, 1979'
            '  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-079-grafico-01-agregacao.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 1 salvo")

# GRAFICO 2 - A pessoa como distribuicao, nao como ponto
fig2, ax2 = plt.subplots(figsize=(12, 8))

x = np.linspace(-1, 11, 1000)
from scipy import stats as st

pessoas = {
    'Pessoa A': (6.5, 1.9, VERDE),
    'Pessoa B': (4.5, 2.1, DOURADO),
}
for nome, (media, dp, cor) in pessoas.items():
    y = st.norm(media, dp).pdf(x)
    ax2.plot(x, y, color=cor, linewidth=3)
    ax2.fill_between(x, y, alpha=0.16, color=cor)
    ax2.axvline(x=media, color=cor, linestyle=':', linewidth=2)
    ax2.text(media, y.max() * 1.06, f'{nome}\nmédia {media}'.replace('.', ','),
             ha='center', fontsize=13, fontweight='bold', color=cor)

ax2.scatter([8.4], [0.012], s=200, color=VERMELHO, zorder=4)
ax2.text(8.4, 0.035, 'um episódio\nda Pessoa B', ha='center', fontsize=12,
         color=VERMELHO, fontweight='bold')

ax2.set_xlim(-1, 11)
ax2.set_ylim(0, 0.27)
ax2.set_xlabel('Quanto o traço aparece no comportamento, num momento qualquer')
ax2.set_ylabel('Densidade')
ax2.set_title('A pessoa não é um ponto. É uma distribuição.\n'
              'Fleeson, 2001: o indivíduo típico manifesta quase todos os níveis de '
              'todos os traços',
              fontsize=15, pad=18)
ax2.text(0.5, -0.135,
         'Distribuições ilustrativas, com parâmetros declarados, para mostrar o '
         'conceito. Um único episódio pode vir de qualquer parte da curva.',
         transform=ax2.transAxes, ha='center', fontsize=11, color=CINZA, style='italic')

plt.figtext(0.5, 0.005,
            'Conceito de Fleeson, 2001, JPSP 80(6), 1011-1027  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-079-grafico-02-distribuicao.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 2 salvo")

# GRAFICO 3 - Cem anos de discussao, em tres tempos
fig3, ax3 = plt.subplots(figsize=(12, 8))
ax3.axis('off'); ax3.set_xlim(0, 10); ax3.set_ylim(0, 10)

ax3.text(5, 9.4, 'Cem anos de discussão, em três tempos', ha='center', fontsize=17,
         fontweight='bold')

blocos = [
    (7.5, '1928', 'Hartshorne & May', 'crianças honestas numa situação eram\n'
     'desonestas em outra. Conclusão: honestidade\né função da situação.', VERMELHO),
    (5.0, '1979', 'Epstein', 'instância isolada: 0,2 a 0,3\nagregado: 0,6 a 0,9\n'
     'os dois lados mediam coisas diferentes.', DOURADO),
    (2.5, '2001', 'Fleeson', 'o traço não é um valor, é uma distribuição.\n'
     'a média de cada pessoa é quase\nperfeitamente estável.', VERDE),
]
for y, ano, autor, texto, cor in blocos:
    ax3.text(1.4, y, ano, ha='center', va='center', fontsize=22, fontweight='bold',
             color=cor)
    ax3.text(1.4, y - 0.62, autor, ha='center', va='center', fontsize=11, color=CINZA)
    ax3.text(5.6, y, texto, ha='left', va='center', fontsize=13, color='#333',
             bbox=dict(boxstyle='round,pad=0.6', facecolor='white', edgecolor=cor,
                       linewidth=1.8))

ax3.text(5, 0.8,
         'Um episódio prevê mal. A distribuição inteira prevê muito bem.',
         ha='center', fontsize=15, fontweight='bold', color='#333')

plt.figtext(0.5, 0.005,
            'Fontes: Hartshorne & May, 1928 · Epstein, 1979 · Fleeson, 2001'
            '  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-079-grafico-03-tres-tempos.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 3 salvo")
```

### 💡 O Insight

Em 1928, alguém pegou um monte de crianças e mediu honestidade em situações reais.
Algumas colavam na prova mas não pegavam lembrancinhas de festa. Outras quebravam
regra em competição esportiva mas não colavam.

A conclusão dos autores foi radical: honestidade não é uma coisa que a pessoa tem, é
**função da situação**.

Quarenta anos depois, Mischel formalizou isso num teto: correlações em torno de
**0,30** seriam o limite da consistência de traço.

E aí, em 1979, Epstein foi olhar as contas.

Ele mostrou que os dois lados estavam medindo coisas diferentes. Entre **instâncias
particulares**, a correlação fica mesmo em **0,2 a 0,3**. Mas os coeficientes
**agregados** ficam entre **0,6 e 0,9**.

Isso não é sutileza estatística. É a diferença entre "não dá para saber nada" e "dá
para saber bastante".

E a razão é uma linha de álgebra que existe desde 1910, a fórmula de Spearman-Brown.
Rodei ela com r de 0,25, que é o meio da faixa das instâncias isoladas:

**Uma observação: 0,25. Cinco observações: 0,63. Vinte observações: 0,87.**

A faixa que Epstein encontrou nos dados agregados é reproduzida exatamente por essa
fórmula. Não é coincidência, é a mesma matemática vista de dois ângulos.

Ou seja: **Hartshorne e May estavam certos sobre a instância e errados sobre a
conclusão.** O erro deles não foi medir mal. Foi generalizar de uma medida ruidosa para
uma afirmação sobre a natureza humana.

E aí vem a síntese moderna, que eu acho a parte mais bonita.

Fleeson, em 2001, acompanhou pessoas por semanas, várias vezes ao dia. O que encontrou
foi que o indivíduo típico **manifesta, regular e rotineiramente, quase todos os níveis
de todos os traços**. Todo mundo já foi extrovertido, todo mundo já foi calado, todo
mundo já foi meticuloso e desleixado, dependendo do dia.

Mas as **médias** de cada pessoa são quase perfeitamente estáveis. E até o **tamanho da
variação** é estável: tem gente consistentemente mais variável que outra.

Então você não é extrovertida. Você tem uma média e um espalhamento de extroversão.

O traço não é um valor. É uma distribuição.

E isso tem uma consequência prática, de segunda-feira, que muda como se avalia gente.

Quando alguém te mostra um episódio e conclui um adjetivo, essa pessoa está usando um
instrumento com confiabilidade de 0,25 e reportando o resultado como se fosse 0,90.

Não é que ela esteja errada sobre o episódio. Ela está certa sobre o episódio e errada
sobre a pessoa.

*Quantas vezes você definiu alguém a partir de uma única observação?*

### ⚠️ Limitações do Modelo

- **O N do estudo de 1928 não foi localizado** nas fontes acessíveis. Sabe-se que foi
  um estudo comportamental de grande escala com escolares, não se sabe o tamanho.
- **A fórmula de Spearman-Brown pressupõe que as observações sejam intercambiáveis e
  independentes.** Na vida real, episódios não são: quem observa escolhe o que observa,
  e observações consecutivas se contaminam.
- **O r de 0,25 é o ponto médio de uma faixa publicada**, não um valor de um estudo
  específico. A curva serve para mostrar a ordem de grandeza da agregação, não para
  estimar a confiabilidade de nenhuma avaliação real.
- **A tradução em "episódios por período" é ilustrativa.** Nenhum estudo publica
  quantos episódios cabem numa semana de convivência.
- **As distribuições do gráfico 2 são ilustrativas**, com parâmetros declarados, e
  servem para mostrar o conceito de traço como distribuição.
- **A correlação intraclasse de 0,20 a 0,40 vem de estimativas posteriores**, e varia
  bastante conforme o traço, o contexto e a forma de medir.
- **Nada aqui diz que caráter não existe.** Diz que ele aparece na distribuição, não no
  episódio, e que quem mede um episódio mediu pouco.
- Fator ×0.80 não aplicado: observação comportamental direta e amostragem de
  experiência em tempo real.

*A ciência é honesta sobre o que não sabe. O modelo também.*
