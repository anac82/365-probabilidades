### A História

Existe grande chance de você bocejar antes de terminar este texto.

Não porque ele seja chato, espero. Mas porque bocejo pega. Basta ver alguém
bocejar, ouvir o som, às vezes só ler a palavra, e o corpo responde sem pedir
licença.

Todo mundo sabe disso. O que pouca gente sabe é que o bocejo não pega de
qualquer um da mesma forma.

Durante um ano inteiro, dois pesquisadores italianos anotaram, em silêncio,
os bocejos de mais de cem pessoas em situações comuns: no trabalho, no
restaurante, no trem, na sala de espera. Anotaram quem bocejou primeiro, quem
bocejou depois, quanto tempo levou e, principalmente, qual era a relação entre
as duas pessoas.

Das muitas coisas que eles testaram, só uma fez diferença.

---

### O Conceito: contágio motor e o gradiente do vínculo

O bocejo contagioso é uma forma de contágio motor: ver ou ouvir uma ação
dispara a mesma ação em quem observa. Ele não aparece em bebês, surge por volta
dos quatro ou cinco anos, e é mais fraco em algumas condições que afetam a
leitura social.

Por isso, há muito tempo se suspeita que ele tenha a ver com empatia. A
hipótese mais conhecida é a do **gradiente empático**: a resposta ao outro seria
mais forte quanto mais próxima for a relação, na ordem família, amigos,
conhecidos, estranhos.

O estudo italiano foi o primeiro a testar esse gradiente com bocejos de verdade,
na vida real, fora do laboratório.

---

### O Modelo

Duas fontes que medem coisas diferentes, e três camadas de conta própria: o
gradiente com incerteza, a proporção de pessoas que pega bocejo, e o que isso
significa numa mesa de jantar.

**Fontes:**
- Norscia, I. & Palagi, E. (2011). Yawn Contagion and Empathy in *Homo sapiens*.
  *PLoS ONE*, 6(12), e28472. Observação naturalista durante um ano (2010-2011),
  **109 adultos** (56 mulheres, 53 homens) de quatro continentes. 613 episódios
  registrados, **480** com bocejador identificável, contágio em **149**. Só o
  vínculo entre as pessoas previu a ocorrência (F = 17,96, p < 0,001), a
  frequência e a demora da resposta. Sexo, país, contexto, posição e se a pessoa
  viu ou só ouviu não entraram no melhor modelo.
- Bartholomew, A. J. & Cirulli, E. T. (2014). Individual Variation in
  Contagious Yawning Susceptibility Is Highly Stable and Largely Unexplained by
  Empathy or Other Known Factors. *PLoS ONE*, 9(3), e91773. **328 participantes**
  viram um vídeo de três minutos de bocejos; **222** bocejaram ao menos uma vez.
  A suscetibilidade é estável na mesma pessoa. Empatia, inteligência e hora do
  dia apareciam associadas nas análises simples, mas o efeito sumiu quando a
  idade entrou no modelo. A idade foi a única que ficou, explicando só 8%.

**O que as duas fontes medem, e por que não se contradizem.** Cirulli pergunta
**quem você é**: diante de um estranho num vídeo, que traço da pessoa prevê o
bocejo? Resposta: quase nenhum, depois de descontada a idade. Norscia pergunta **com quem você está**: diante
de pessoas reais, que relação prevê o bocejo? Resposta: a proximidade. São
perguntas diferentes, e a combinação delas é o achado mais interessante do dia.

**Ponte declarada com o #015.** Lá o tema foi contágio emocional em rede, com a
felicidade de desconhecidos afetando a sua. Aqui o contágio é motor, com fontes
distintas, e a proximidade trabalha no sentido oposto: é quem está perto que
pega.

**Fator de correção ×0,80:** não se aplica. Comportamento observado diretamente
(Norscia) e registrado em sessão controlada (Cirulli), sem autorrelato.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

SEED = 42
rng = np.random.default_rng(SEED)

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

VERMELHO = '#c0392b'
DOURADO = '#c8a84b'
VERDE = '#1a5f5a'
CINZA = '#6b6a64'

print("Bibliotecas carregadas. Semente fixa:", SEED)
```

```python
# --- DADOS DA LITERATURA ---

# Norscia & Palagi (2011), PLoS ONE 6(12): e28472
n_pessoas_norscia = 109
episodios_total = 613
episodios_analisados = 480
episodios_com_contagio = 149

# Frequencia de contagio por vinculo (Figura 2 do artigo)
# media, erro padrao, IC 95% publicado. Estranhos ficaram fora desta analise.
vinculos = ['Conhecidos', 'Amigos', 'Familia e parceiros']
freq_media = np.array([0.386, 0.519, 0.850])
freq_ep = np.array([0.058, 0.080, 0.064])
freq_ic_inf = np.array([0.267, 0.356, 0.720])
freq_ic_sup = np.array([0.505, 0.682, 0.979])
n_duplas_freq = 48

# Bartholomew & Cirulli (2014), PLoS ONE 9(3): e91773
n_cirulli = 328
bocejaram_cirulli = 222
idade_explica = 0.08

aplica_fator_080 = False

print("=" * 66)
print("  DADOS - BOCEJO CONTAGIOSO")
print("=" * 66)
print(f"\n  Norscia & Palagi, 2011 (observacao de 1 ano, {n_pessoas_norscia} adultos):")
print(f"  -> episodios analisados:        {episodios_analisados}")
print(f"  -> episodios com contagio:      {episodios_com_contagio} "
      f"({episodios_com_contagio/episodios_analisados*100:.0f}%)")
print(f"\n  Frequencia de contagio por vinculo ({n_duplas_freq} duplas):")
for v, m, a, b in zip(vinculos, freq_media, freq_ic_inf, freq_ic_sup):
    print(f"  -> {v:<22} {m*100:5.1f}%   IC 95% [{a*100:.1f}; {b*100:.1f}]")
print(f"\n  Bartholomew & Cirulli, 2014 (video em laboratorio):")
print(f"  -> bocejaram ao menos uma vez:  {bocejaram_cirulli} de {n_cirulli} "
      f"({bocejaram_cirulli/n_cirulli*100:.1f}%)")
print(f"  -> variacao explicada pela idade: {idade_explica*100:.0f}%")
print(f"  -> empatia, inteligencia, hora do dia: somem ao controlar a idade")
print(f"\n  Fator x0,80: {'SIM' if aplica_fator_080 else 'NAO'}")
print("=" * 66)
```

```python
# --- O MODELO ---
N_SIM = 200_000

# ---------------------------------------------------------------
# CAMADA 1 - O gradiente com incerteza
# Cada frequencia vira uma Beta com a MESMA media e o MESMO erro
# padrao publicados (metodo dos momentos). Isso respeita o limite
# entre 0 e 1 sem cortar a distribuicao, o que uma normal faria.
# ---------------------------------------------------------------
def beta_por_momentos(media, ep):
    k = media * (1 - media) / ep ** 2 - 1
    return media * k, (1 - media) * k

params_beta = [beta_por_momentos(m, e) for m, e in zip(freq_media, freq_ep)]
sim_freq = np.column_stack([
    stats.beta(a, b).rvs(N_SIM, random_state=rng) for a, b in params_beta
])
conhecidos, amigos, familia = sim_freq[:, 0], sim_freq[:, 1], sim_freq[:, 2]

p_familia_vence_amigos = np.mean(familia > amigos)
p_amigos_vencem_conhecidos = np.mean(amigos > conhecidos)
p_familia_vence_conhecidos = np.mean(familia > conhecidos)
p_gradiente_completo = np.mean((familia > amigos) & (amigos > conhecidos))

razao_fam_conh = familia / np.maximum(conhecidos, 1e-6)
razao_med = np.median(razao_fam_conh)
razao_ic = np.percentile(razao_fam_conh, [2.5, 97.5])

# ---------------------------------------------------------------
# CAMADA 2 - Quantas pessoas pegam bocejo
# Posterior de Jeffreys sobre 222 de 328.
# ---------------------------------------------------------------
post_suscetivel = stats.beta(bocejaram_cirulli + 0.5, n_cirulli - bocejaram_cirulli + 0.5)
susc_media = post_suscetivel.mean()
susc_ic = post_suscetivel.interval(0.95)

post_episodio = stats.beta(episodios_com_contagio + 0.5,
                           episodios_analisados - episodios_com_contagio + 0.5)
epis_media = post_episodio.mean()
epis_ic = post_episodio.interval(0.95)

# ---------------------------------------------------------------
# CAMADA 3 - A mesa de jantar
# COMPOSICAO PROPRIA, declarada: chance de uma pessoa devolver o
# bocejo = P(ser suscetivel, Cirulli) x frequencia por vinculo
# (Norscia). As duas incertezas sao sorteadas juntas.
# Pergunta: numa mesa com n pessoas alem de quem bocejou, qual a
# chance de pelo menos uma devolver o bocejo?
# ---------------------------------------------------------------
sim_susc = post_suscetivel.rvs(N_SIM, random_state=rng)
p_pessoa = sim_freq * sim_susc[:, None]          # por vinculo
p_pessoa_med = np.median(p_pessoa, axis=0)

tamanhos_mesa = np.arange(1, 9)
mesa_med = np.zeros((3, len(tamanhos_mesa)))
mesa_inf = np.zeros_like(mesa_med)
mesa_sup = np.zeros_like(mesa_med)
for j, n in enumerate(tamanhos_mesa):
    pelo_menos_um = 1 - (1 - p_pessoa) ** n
    mesa_med[:, j] = np.median(pelo_menos_um, axis=0)
    mesa_inf[:, j] = np.percentile(pelo_menos_um, 2.5, axis=0)
    mesa_sup[:, j] = np.percentile(pelo_menos_um, 97.5, axis=0)

idx4 = list(tamanhos_mesa).index(4)

print("=" * 66)
print("  CAMADA 1 - O GRADIENTE DO AFETO, COM INCERTEZA")
print("=" * 66)
print(f"\n  Em {N_SIM:,} sorteios a partir das incertezas publicadas:".replace(",", "."))
print(f"  -> familia pega mais que amigos:        {p_familia_vence_amigos*100:.1f}%")
print(f"  -> amigos pegam mais que conhecidos:    {p_amigos_vencem_conhecidos*100:.1f}%")
print(f"  -> familia pega mais que conhecidos:    {p_familia_vence_conhecidos*100:.1f}%")
print(f"  -> a ordem completa se mantem:          {p_gradiente_completo*100:.1f}%")
print(f"\n  Familia contra conhecidos: {razao_med:.1f} vezes mais contagio "
      f"[{razao_ic[0]:.1f}; {razao_ic[1]:.1f}]")

print("\n" + "=" * 66)
print("  CAMADA 2 - QUANTAS PESSOAS PEGAM BOCEJO")
print("=" * 66)
print(f"\n  Cirulli, video em laboratorio: {susc_media*100:.1f}% "
      f"[{susc_ic[0]*100:.1f}; {susc_ic[1]*100:.1f}] das pessoas")
print(f"  Norscia, vida real: {epis_media*100:.1f}% "
      f"[{epis_ic[0]*100:.1f}; {epis_ic[1]*100:.1f}] dos episodios")
print(f"\n  ATENCAO: as unidades sao diferentes. Um e proporcao de PESSOAS")
print(f"  diante de tres minutos de bocejos; o outro e proporcao de")
print(f"  EPISODIOS unicos no dia a dia. Nao se comparam os niveis.")

print("\n" + "=" * 66)
print("  CAMADA 3 - A MESA DE JANTAR (composicao propria)")
print("=" * 66)
print(f"\n  Chance de UMA pessoa devolver o bocejo:")
for v, p in zip(vinculos, p_pessoa_med):
    print(f"  -> {v:<22} {p*100:5.1f}%")
print(f"\n  Mesa com 4 pessoas alem de quem bocejou, chance de pelo menos")
print(f"  uma devolver:")
for i, v in enumerate(vinculos):
    print(f"  -> {v:<22} {mesa_med[i, idx4]*100:5.1f}%  "
          f"[{mesa_inf[i, idx4]*100:.0f}; {mesa_sup[i, idx4]*100:.0f}]")
print("=" * 66)
```

```python
# --- VISUALIZACAO ---

# GRAFICO 1 - O gradiente publicado
fig1, ax1 = plt.subplots(figsize=(10, 6))

cores = [CINZA, DOURADO, VERDE]
pos = np.arange(3)
ax1.bar(pos, freq_media * 100, color=cores, alpha=0.85, width=0.55)
ax1.errorbar(pos, freq_media * 100,
             yerr=[(freq_media - freq_ic_inf) * 100, (freq_ic_sup - freq_media) * 100],
             fmt='none', ecolor='#333', capsize=8, linewidth=1.8)
for p, m in zip(pos, freq_media):
    ax1.text(p + 0.3, m * 100 + 1.5, f'{m*100:.0f}%', fontsize=15,
             fontweight='bold', color='#333')

ax1.set_xticks(pos)
ax1.set_xticklabels(vinculos, fontsize=12)
ax1.set_ylim(0, 105)
ax1.set_ylabel('Frequencia de contagio do bocejo (%)')
ax1.set_title('Quanto mais perto, mais o bocejo pega\nNorscia & Palagi, 2011: 109 adultos observados por um ano, barras com IC 95%',
              fontsize=12, pad=15)

plt.figtext(0.5, 0.005,
            'Fonte: Norscia & Palagi, 2011, PLoS ONE | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-101-grafico-01-gradiente.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 1 salvo.")

# GRAFICO 2 - O gradiente se sustenta?
fig2, ax2 = plt.subplots(figsize=(10, 6))

for serie, cor, rot in zip([conhecidos, amigos, familia], cores, vinculos):
    ax2.hist(serie * 100, bins=90, density=True, color=cor, alpha=0.55,
             edgecolor='white', linewidth=0.3, label=rot)
def fmt_p(p):
    return 'mais de 99,9%' if p > 0.999 else f'{p*100:.1f}%'.replace('.', ',')

topo2 = ax2.get_ylim()[1]
ax2.set_ylim(0, topo2 * 1.38)
ax2.text(62, topo2 * 1.33,
         f'familia acima de amigos: {fmt_p(p_familia_vence_amigos)}\n'
         f'amigos acima de conhecidos: {fmt_p(p_amigos_vencem_conhecidos)}\n'
         f'familia acima de conhecidos: {fmt_p(p_familia_vence_conhecidos)}',
         fontsize=10.5, color='#333', va='top', ha='center')

ax2.set_xlim(0, 105)
ax2.set_xlabel('Frequencia de contagio do bocejo (%)')
ax2.set_ylabel('Densidade')
ax2.set_yticks([])
ax2.legend(loc='upper left', fontsize=10, frameon=False)
ax2.set_title('A ordem do afeto resiste a incerteza\n200.000 sorteios de Betas com as medias e erros padrao publicados (semente 42)',
              fontsize=12, pad=15)

plt.figtext(0.5, 0.005,
            'Conta propria sobre Norscia & Palagi, 2011 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-101-grafico-02-quem-ganha.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 2 salvo.")

# GRAFICO 3 - Dois tercos pegam
fig3, ax3 = plt.subplots(figsize=(10, 6))

x3 = np.linspace(0.55, 0.80, 1500)
y3 = post_suscetivel.pdf(x3)
ax3.plot(x3 * 100, y3, color=VERDE, linewidth=2.8)
ax3.fill_between(x3 * 100, y3, where=(x3 >= susc_ic[0]) & (x3 <= susc_ic[1]),
                 alpha=0.22, color=VERDE)
ax3.axvline(x=200 / 3, color=DOURADO, linestyle='--', linewidth=2)
ax3.text(200 / 3 + 0.3, max(y3) * 0.92, 'dois tercos', fontsize=11, color=DOURADO)
ax3.text(susc_ic[0] * 100 - 0.2, max(y3) * 0.35,
         f'IC 95%\n[{susc_ic[0]*100:.1f}; {susc_ic[1]*100:.1f}]',
         fontsize=10, color=VERDE, ha='right')

ax3.set_xlabel('Pessoas que bocejam diante de um video de bocejos (%)')
ax3.set_ylabel('Densidade')
ax3.set_yticks([])
ax3.set_title('Dois em cada tres pegam o bocejo, e quase nada explica quem\nBartholomew & Cirulli, 2014: 222 de 328 participantes, posterior de Jeffreys',
              fontsize=12, pad=15)

plt.figtext(0.5, 0.005,
            'Fonte: Bartholomew & Cirulli, 2014, PLoS ONE | conta propria | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-101-grafico-03-dois-tercos.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 3 salvo.")

# GRAFICO 4 - A mesa de jantar
fig4, ax4 = plt.subplots(figsize=(10, 6))

for i, (cor, rot) in enumerate(zip(cores, vinculos)):
    ax4.plot(tamanhos_mesa, mesa_med[i] * 100, color=cor, linewidth=2.8,
             marker='o', markersize=7, label=rot)
    ax4.fill_between(tamanhos_mesa, mesa_inf[i] * 100, mesa_sup[i] * 100,
                     color=cor, alpha=0.15)

ax4.set_xticks(tamanhos_mesa)
ax4.set_ylim(0, 102)
ax4.set_xlabel('Pessoas na mesa, alem de quem bocejou')
ax4.set_ylabel('Chance de pelo menos uma devolver o bocejo (%)')
ax4.legend(loc='lower right', fontsize=10, frameon=False)
ax4.set_title('Num jantar de familia, o bocejo quase nunca fica sozinho\nComposicao propria: suscetibilidade (Cirulli) x frequencia por vinculo (Norscia), faixas de 95%',
              fontsize=12, pad=15)

plt.figtext(0.5, 0.005,
            'Modelo proprio que combina dois estudos diferentes | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-101-grafico-04-a-mesa.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 4 salvo.")
```

### O Insight

**85% com a família. 52% com amigos. 39% com conhecidos.**

Essa é a frequência com que o bocejo pega, conforme a relação entre as duas
pessoas. De tudo o que os pesquisadores testaram, o vínculo foi a única coisa
que importou. Não importou se eram homens ou mulheres, de que país vinham, se
estavam no trabalho ou no jantar, se viram o bocejo ou só ouviram.

E a ordem se sustenta mesmo quando se leva a incerteza a sério: em praticamente
todos os sorteios, a família fica acima dos conhecidos.

Tem um outro estudo que completa a história de um jeito curioso. Quando
mostraram um vídeo de bocejos a 328 pessoas, dois terços bocejaram. Empatia até
parecia importar, mas o efeito sumiu quando os pesquisadores levaram a idade em
conta, e a própria idade explicou só 8% da diferença entre as pessoas.

Juntos, os dois estudos dizem uma coisa bonita: **o bocejo diz pouco sobre quem
você é, e muito sobre de quem você está perto.**

Numa mesa de família com quatro pessoas, o seu bocejo volta de alguém em cerca
de **97%** das vezes. Numa reunião com quatro conhecidos, em cerca de 70%. É uma
conta que combina os dois estudos, então vale pelo contraste, não pela casa
decimal.

A pergunta que fica:

*De quem foi o último bocejo que você pegou sem nem perceber?*

### Limitações do Modelo

- **Norscia e Palagi é observacional e não cego ao vínculo.** Os próprios
  autores observaram amigos, familiares, colegas e alunos, e sabiam a relação
  entre as pessoas enquanto anotavam. A concordância entre os dois observadores
  foi alta (kappa acima de 0,75), mas o risco de expectativa não é zero.
- **As frequências por vínculo vêm de 48 duplas**, só entre quem já tinha
  bocejado junto pelo menos uma vez, e **estranhos ficaram fora** dessa análise.
  Por isso a frequência de contágio com estranhos não aparece nos gráficos.
- **"Família" junta parentes próximos e parceiros.** Não dá para separar o
  efeito de parentesco do efeito de convivência.
- **A mesa de jantar é composição própria.** Multiplica a suscetibilidade medida
  num estudo pela frequência por vínculo medida em outro, com populações e
  métodos diferentes. É ilustração do mecanismo, não previsão.
- **Os dois estudos medem unidades diferentes.** Cirulli conta pessoas diante de
  três minutos de vídeo; Norscia conta episódios isolados no dia a dia. Os
  níveis, 68% e 31%, não se comparam.
- **O modelo do número de capa é gaussiano sobre uma proporção.** Norscia e
  Palagi modelaram a frequência de contágio, que é uma proporção entre 0 e 1,
  com um modelo linear misto gaussiano. O intervalo da família chega a 97,9%,
  encostado no teto. Um modelo binomial ou uma regressão beta seria o adequado.
  Este notebook reconstrói cada frequência como uma Beta com a mesma média e o
  mesmo erro padrão, o que respeita o limite, mas herda os erros padrão
  publicados.
- **A ordem do gradiente não foi testada par a par pelos autores.** O artigo
  reporta o teste global do vínculo, F(3; 476) = 17,96 na ocorrência, e as
  médias por categoria, sem comparações múltiplas com correção. A camada 1 deste
  notebook faz essa comparação, mas trata as três estimativas como
  independentes, o que elas não são inteiramente (há observadores em comum).
- **Há inconsistência de relato no artigo original.** O AICc do modelo de
  ocorrência aparece como 2.245,493 no texto e 2.272,933 na legenda; o da
  latência, 1.116,604 e 1.116,694. Não muda a conclusão, mas fica registrado.
- **"Empatia não previu" é ausência de significância, não prova de ausência.**
  Em Cirulli, empatia aparecia nas análises simples e sumiu ao controlar a
  idade, numa seleção stepwise com corte de p < 0,01, método conhecido por
  produzir resultados instáveis. Não há teste de equivalência. E as escalas de
  empatia e de contágio emocional foram interrompidas depois de 202 e 128
  pessoas por "clara falta de efeito", uma parada decidida olhando os dados.
- **Os bocejos de Cirulli foram contados pelos próprios participantes**, que
  leram uma descrição do fenômeno antes do vídeo. Os autores admitem que isso
  pode ter inflado os 68%; estudos controlados anteriores ficavam entre 40% e
  60%.
- **O elo com empatia continua em aberto.** O gradiente de Norscia é compatível
  com empatia, mas também com atenção e convivência: quem está perto é mais
  olhado.

*A ciência é honesta sobre o que não sabe. O modelo também.*
