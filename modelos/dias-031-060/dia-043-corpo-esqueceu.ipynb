### 📖 A História

Fazia vinte e cinco dias que eu não corria.

Hoje eu peguei a cachorra de um conhecido emprestada e saímos pra correr. Ela não sabe o que são vinte e cinco dias parados, e não estava nem aí pra isso. Viu a rua abrir na frente e disparou. Sem cálculo, sem negociação, sem aquela voz na cabeça perguntando se valia a pena.

Nem nos meus treinos de tiro eu ia tão rápido quanto ela me puxou hoje.

E eu invejei aquilo. Porque nos primeiros minutos meu corpo inteiro repetia a mesma frase: você nunca correu na vida. Pernas de chumbo, pulmão em pânico, a cabeça já montando a desculpa pra voltar pra casa e tentar de novo outro dia, que nunca chega.

A sensação era de que eu tinha perdido tudo. Que vinte e cinco dias tinham apagado meses de corrida. Que eu estava começando do zero, de novo.

Ao longo dos últimos anos eu aprendi uma coisa sobre mim: quando eu decido sair pra treinar, eu não negocio com a minha mente. A decisão vem antes da vontade. Eu calço o tênis e vou, mesmo quando cada célula pede pra ficar.

Só que hoje eu descobri que isso também enferruja. Que essa parte de mim vai precisar ser reaprendida, do mesmo jeito que as pernas. O recomeço não é só do corpo. É da disciplina que segura o corpo.

Mas a sensação de ter perdido tudo estava mentindo. E eu fui checar o quanto.

---

### 📚 O Conceito: A Sensação Superestima a Perda

Quando você para de treinar, o condicionamento cai. Isso é real. Mas a
velocidade da queda que a sua cabeça imagina não tem nada a ver com a
velocidade da queda que o seu corpo sofre.

Coyle et al. (1984) acompanharam atletas de endurance que pararam de
treinar e mediram o VO₂máx ao longo de 84 dias. A queda é rápida no
começo e depois quase estabiliza: cerca de 7% nas primeiras três semanas,
e boa parte disso é volume sanguíneo e volume sistólico, não o músculo
"esquecendo" de correr. Coisas que voltam rápido.

Ou seja: depois de 25 dias, a perda fisiológica real é pequena. A sensação
de "meu corpo nunca correu" é quase toda outra coisa — sangue, calor,
cabeça e uma percepção de esforço descalibrada.

E tem um segundo erro, gêmeo do primeiro: Ruby & Dunn (2011) mostraram que
as pessoas subestimam o quanto vão gostar de se exercitar. A sensação mente
sobre a perda antes de correr, e mente sobre o prazer depois.

### 🧮 O Modelo

**Fontes:**
- Coyle et al., 1984 — *Journal of Applied Physiology*, 57(6), 1857-1864
  N=7 atletas de endurance (treinados 10±3 anos) + 8 controles
  Queda do VO₂máx: 7% (dia 12-21), 14% (dia 56), 16% (dia 84)
- Ruby & Dunn et al., 2011 — *Health Psychology* · 4 estudos, N total=279
  Prazer previsto do exercício: 7,58 · Prazer real experimentado: 8,42 (escala 0-10)
- Reed & Ones, 2006 — *Psychology of Sport and Exercise*, 7, 477-514 (apoio)
  Revisão: exercício aeróbico agudo aumenta o afeto positivo ativado

**Nota metodológica:** o fator ×0.80 NÃO se aplica. VO₂máx e volume
sistólico são medidas fisiológicas objetivas; o prazer de Ruby & Dunn vem
de experimento com previsores vs experimentadores, não de survey
populacional autorrelatado.


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

np.random.seed(43)  # Dia #043

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---

# Coyle et al., 1984 — Journal of Applied Physiology 57(6):1857-1864
# N=7 atletas de endurance (treinados 10±3 anos) que pararam de treinar
# Queda percentual do VO2max ao longo do destreinamento
dias_coyle = np.array([0, 12, 21, 56, 84])
perda_vo2_coyle = np.array([0.0, 7.0, 7.0, 14.0, 16.0])  # % de queda

# O caso de hoje
dias_parada_ana = 25

# Ruby & Dunn et al., 2011 — Health Psychology (4 estudos, N total=279)
# Erro de previsão do prazer do exercício (Estudo 1)
prazer_previsto = 7.58   # escala 0-10
prazer_real = 8.42       # escala 0-10
n_ruby = 279

# A sensação de hoje: "meu corpo nunca correu" = perda percebida ~100%
perda_percebida = 100.0  # %

# Fator de correção: NÃO se aplica (medida fisiológica objetiva)
fator_correcao = None

print("=" * 65)
print("  DADOS DA LITERATURA — DESTREINAMENTO E VOLTA")
print("=" * 65)
print(f"\n  Coyle et al., 1984 (N=7 atletas de endurance):")
print(f"  → Queda do VO2max no dia 12:   {perda_vo2_coyle[1]:.0f}%")
print(f"  → Queda do VO2max no dia 21:   {perda_vo2_coyle[2]:.0f}%")
print(f"  → Queda do VO2max no dia 56:   {perda_vo2_coyle[3]:.0f}%")
print(f"  → Queda do VO2max no dia 84:   {perda_vo2_coyle[4]:.0f}%")
print(f"\n  Ruby & Dunn, 2011 (N={n_ruby}):")
print(f"  → Prazer previsto do exercício:  {prazer_previsto:.2f} / 10")
print(f"  → Prazer real experimentado:     {prazer_real:.2f} / 10")
print(f"\n  O caso de hoje:")
print(f"  → Dias sem treinar:              {dias_parada_ana}")
print(f"  → Perda que a SENSAÇÃO dizia:    ~{perda_percebida:.0f}%")
print("=" * 65)
```

```python
# --- O MODELO ---
# Ajuste da curva de destreinamento de Coyle e estimativa no dia 25

# Ajuste quadrático à curva de perda do VO2max (forma funcional visível)
coef = np.polyfit(dias_coyle, perda_vo2_coyle, 2)
a, b, c = coef  # perda(t) = a*t^2 + b*t + c
curva_perda = np.poly1d(coef)

# Perda real estimada no dia 25 (também confere com interpolação linear 21->56)
perda_real_dia25 = float(curva_perda(dias_parada_ana))
perda_interp_linear = 7.0 + (14.0 - 7.0) * (25 - 21) / (56 - 21)
retencao_dia25 = 100 - perda_real_dia25

# O erro de estimativa: quanto a sensação exagerou a perda
razao_erro = perda_percebida / perda_real_dia25

# O erro gêmeo: prazer subestimado
gap_prazer = prazer_real - prazer_previsto
erro_prazer_pct = gap_prazer / prazer_previsto * 100

print("=" * 65)
print("  MODELO — O QUE O CORPO REALMENTE ESQUECEU")
print("=" * 65)
print(f"\n  Curva de destreinamento ajustada (Coyle):")
print(f"  perda(t) = {a:.5f}·t² + {b:.4f}·t + {c:.3f}")
print(f"\n  No dia {dias_parada_ana}:")
print(f"  → Perda real estimada do VO2max:   {perda_real_dia25:.1f}%")
print(f"    (confere: interpolação linear = {perda_interp_linear:.1f}%)")
print(f"  → Condicionamento retido:          {retencao_dia25:.1f}%")
print(f"\n  O ERRO DE ESTIMATIVA:")
print(f"  → A sensação dizia perda de:       ~{perda_percebida:.0f}%")
print(f"  → A perda real era de:             ~{perda_real_dia25:.1f}%")
print(f"  → A sua cabeça exagerou a perda em: {razao_erro:.0f}×")
print(f"\n  O ERRO GÊMEO (Ruby & Dunn):")
print(f"  → Prazer previsto:                 {prazer_previsto:.2f}")
print(f"  → Prazer real:                     {prazer_real:.2f}")
print(f"  → Você gostou {erro_prazer_pct:.0f}% mais do que previu")
print(f"\n  RESPOSTA:")
print(f"  → Seu corpo esqueceu ~{perda_real_dia25:.0f}%.")
print(f"    Você sentiu que tinha esquecido tudo.")
print(f"    A distância entre os dois é o tamanho da mentira.")
print("=" * 65)
```

```python
# --- VISUALIZAÇÃO — GRÁFICOS SEPARADOS ---

# ── GRÁFICO 1 — O erro de estimativa: sensação vs realidade ──
fig1, ax1 = plt.subplots(figsize=(10, 6))

categorias = ['O que a\nSENSAÇÃO dizia', 'O que o corpo\nrealmente perdeu']
valores = [perda_percebida, perda_real_dia25]
cores = ['#c0392b', '#1a5f5a']

bars = ax1.bar(categorias, valores, color=cores, alpha=0.85, width=0.4)
ax1.set_ylim(0, 110)
ax1.set_ylabel('Perda de condicionamento percebida (%)')
ax1.set_title('O Erro de Estimativa de Quem Volta a Treinar\n25 dias parada — a sensação de perda vs a perda real',
              fontsize=13, pad=15)

for bar, valor in zip(bars, valores):
    ax1.text(bar.get_x() + bar.get_width()/2, valor + 2,
             f'{valor:.0f}%', ha='center', fontweight='bold', fontsize=16)

ax1.text(0.5, 0.5, f'{razao_erro:.0f}× exagerada',
         ha='center', fontsize=13, color='#c8a84b', fontweight='bold',
         transform=ax1.transAxes)

plt.figtext(0.5, 0.01,
            'Fonte: Coyle et al. 1984 — Journal of Applied Physiology | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-043-grafico-01-erro-estimativa.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 — ASSINATURA MATEMÁTICA: curva de destreinamento ──
fig2, ax2 = plt.subplots(figsize=(10, 6))

t = np.linspace(0, 84, 400)
ax2.plot(t, curva_perda(t), color='#1a5f5a', linewidth=2.5,
         label='Curva ajustada de destreinamento')
ax2.scatter(dias_coyle, perda_vo2_coyle, color='#c0392b', s=70, zorder=5,
            label='Dados de Coyle et al. 1984 (N=7)')

# Marcador do dia 25
ax2.axvline(x=dias_parada_ana, color='#c8a84b', linestyle='--', alpha=0.8)
ax2.scatter([dias_parada_ana], [perda_real_dia25], color='#c8a84b', s=130,
            zorder=6, edgecolor='#333')
ax2.annotate(f'Dia 25\n≈ {perda_real_dia25:.0f}% de perda\n{retencao_dia25:.0f}% retido',
             xy=(dias_parada_ana, perda_real_dia25),
             xytext=(dias_parada_ana + 6, perda_real_dia25 + 4.5),
             fontsize=11, color='#8a6d10', fontweight='bold')

# Equação literal
ax2.text(0.03, 0.93,
         f'perda(t) = {a:.4f}·t²  {b:+.3f}·t  {c:+.2f}',
         transform=ax2.transAxes, fontsize=11, color='#1a5f5a',
         family='monospace', fontweight='bold')

ax2.set_xlabel('Dias sem treinar')
ax2.set_ylabel('Queda do VO₂máx (%)')
ax2.set_title('A Assinatura Matemática do Destreinamento\nA perda desaba rápido e depois quase congela',
              fontsize=13, pad=15)
ax2.legend(loc='lower right', fontsize=9)
ax2.set_ylim(-1, 20)

plt.figtext(0.5, 0.01,
            'Fonte: Coyle et al. 1984 — J Appl Physiol 57(6) | curva ajustada aos pontos publicados | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-043-grafico-02-curva-destreinamento.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 — O erro gêmeo: o prazer que você não prevê ──
fig3, ax3 = plt.subplots(figsize=(10, 6))

categorias3 = ['O prazer que\nvocê PREVÊ', 'O prazer que\nvocê SENTE']
valores3 = [prazer_previsto, prazer_real]
cores3 = ['#c0392b', '#1a5f5a']

bars3 = ax3.bar(categorias3, valores3, color=cores3, alpha=0.85, width=0.4)
ax3.set_ylim(0, 10)
ax3.set_ylabel('Prazer do exercício (escala 0-10)')
ax3.set_title('O Erro Gêmeo: Você Gosta Mais do que Prevê\nRuby & Dunn 2011 — 4 estudos, N total = 279',
              fontsize=13, pad=15)

for bar, valor in zip(bars3, valores3):
    ax3.text(bar.get_x() + bar.get_width()/2, valor + 0.15,
             f'{valor:.2f}', ha='center', fontweight='bold', fontsize=16)

ax3.set_ylim(0, 11)
ax3.annotate('', xy=(1, prazer_real), xytext=(0, prazer_previsto),
             arrowprops=dict(arrowstyle='->', color='#c8a84b', lw=2))
ax3.text(0.5, 0.96, f'+{gap_prazer:.2f} ponto de prazer que a sua cabeça não previu',
         ha='center', fontsize=11, color='#8a6d10', fontweight='bold',
         transform=ax3.transAxes)

plt.figtext(0.5, 0.01,
            'Fonte: Ruby & Dunn 2011 — Health Psychology | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-043-grafico-03-humor-previsto.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**Você sentiu que tinha esquecido tudo. Seu corpo esqueceu menos de 10%.**

Vinte e cinco dias parada. A sensação dizia cem por cento de perda. A curva de
Coyle diz cerca de nove. A sua cabeça exagerou a perda em cerca de onze vezes.

E boa parte desses 9% nem é o músculo. É sangue, é volume sistólico, é
percepção de esforço descalibrada. Coisas que voltam rápido, muito mais
rápido do que levaram pra ir embora.

Depois de 25 dias, você não estava começando do zero. Você estava com uns
**91% do seu condicionamento intacto**, dentro do corpo, esperando você
lembrar que ele existia.

E tem o erro gêmeo, o mais bonito: você não só superestima a perda antes de
correr. Você subestima o prazer depois. Ruby & Dunn mediram: as pessoas
gostam mais de se exercitar do que preveem que vão gostar. A cachorra já
sabia. Você reaprendeu hoje.

A disciplina de não negociar com a própria mente também enferruja numa
parada. Mas ela volta pela mesma porta que o VO₂máx: pela repetição. Um
treino de cada vez.

A pergunta que fica:

existia só na sua cabeça?*

### ⚠️ Limitações do Modelo

- Coyle et al. 1984 tem N=7, atletas de endurance de alto nível; a magnitude
  da perda varia com idade, nível de treino e genética
- A perda no dia 25 é estimada por curva ajustada a 4 pontos publicados —
  é interpolação, não medida direta nesse dia
- A "perda percebida de 100%" representa a sensação relatada, não uma medida
- Ruby & Dunn 2011 mediu prazer por autorrelato de previsores vs experimentadores
- Reed & Ones 2006 entra como apoio qualitativo; não usei tamanho de efeito
  numérico porque não o confirmei na fonte primária
- O fator ×0.80 não foi aplicado: medidas fisiológicas objetivas e experimento
  controlado, não survey populacional

*A ciência é honesta sobre o que não sabe. O modelo também.*
