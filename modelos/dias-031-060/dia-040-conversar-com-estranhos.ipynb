### 📖 A História

Eu defenderia a solidão em qualquer tribunal.

Prefiro poucos amigos e conversas profundas a uma agenda cheia de gente. Tenho
baixa tolerância a conversa fiada. Coloco o fone antes mesmo de sentar. Sempre achei
que isso fosse refinamento: quem valoriza profundidade não perde tempo com o raso.

Talvez seja. Mas os dados me deixaram desconfortável. Porque uma coisa é escolher a
solidão sabendo o preço dela. Outra é evitar a conexão por causa de uma previsão que
está simplesmente errada.

---

### 📚 O Conceito: O Erro de Previsão Afetiva

Somos péssimos em prever como vamos nos sentir. A psicologia chama isso de
**erro de previsão afetiva** (affective forecasting): a diferença sistemática entre
o que a gente acha que vai sentir e o que de fato sente.

Nicholas Epley e Juliana Schroeder, da Universidade de Chicago, testaram esse erro
num terreno específico: falar com estranhos. A maioria das pessoas evita, convencida
de que a solidão é mais agradável. A pergunta deles foi simples: será que a intuição
está certa, ou a gente está sistematicamente errando sobre o próprio prazer?

A resposta não foi uma opinião. Foi um experimento de campo, aleatorizado, em plena
estação de trem.

---

### 🧮 O Modelo

Reproduzo a comparação central do estudo. Comutadores de trem em Chicago foram
sorteados para uma de três condições: conversar com um estranho, ficar em solidão,
ou agir como fariam normalmente. Depois relataram quão positiva foi a experiência.

Um grupo separado de comutadores foi convidado a **prever** como se sentiria em cada
condição, antes de viver a experiência. É o confronto entre a intuição e a realidade.

Traduzo o tamanho de efeito (Cohen's d) em uma linguagem mais direta: a **probabilidade
de superioridade**, ou seja, a chance de quem conversou ter tido um dia melhor do que
quem ficou calado.

**Fonte:**
- Epley, N. & Schroeder, J. (2014). *Mistakenly Seeking Solitude*. **Journal of
  Experimental Psychology: General**, 143(5), 1980-1999.
  Experimento 1a: N=97 comutadores (Metra, Chicago) · Experimento 1b (previsão): N=66.

*Nota metodológica: os números centrais são tamanhos de efeito de um experimento de
campo aleatorizado, com medida objetiva de condição. Por isso o fator de correção
×0.80, reservado a proporções de survey autorrelatado, não se aplica.*


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

# Paleta 365 Probabilidades
DOURADO = '#c8a84b'
VERMELHO = '#c0392b'
VERDE    = '#1a5f5a'
CINZA    = '#6b6a64'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Fonte: Epley & Schroeder (2014) — J. Exp. Psychol. General, 143(5)

# Experimento 1a — EXPERIÊNCIA REAL (comutadores de trem, Chicago)
N_1a          = 97       # amostra final analisada
d_real        = 0.63     # Cohen's d: conexão vs. solidão (t(94)=2.49, p=.02)
n_por_grupo   = 48       # tamanho aproximado por grupo no contraste (df=94)
conversa_min  = 14.2     # duração média da conversa (SD 11.3)
agradabilidade = 5.1     # agradabilidade média da conversa (escala 0-6)

# Experimento 1b — PREVISÃO (grupo separado, antes de viver)
N_1b          = 66
d_previsto    = 0.41     # d na DIREÇÃO ERRADA: previram conexão PIOR que solidão

# Survey prévio (203 participantes MTurk)
p_evita_estranho = 0.76  # 76% diziam que evitariam falar com um estranho no trem
p_evita_amigo    = 0.00  # 0% evitariam falar com um amigo

print("=" * 65)
print("  DADOS DA LITERATURA — CONVERSAR COM ESTRANHOS")
print("=" * 65)
print(f"  Experimento 1a (experiência real, N={N_1a}):")
print(f"  → Cohen's d (conexão vs solidão):     +{d_real}")
print(f"  → Conversa média:                      {conversa_min} min")
print(f"  → Agradabilidade da conversa (0-6):    {agradabilidade}")
print()
print(f"  Experimento 1b (previsão, N={N_1b}):")
print(f"  → d previsto (direção ERRADA):        -{d_previsto}")
print()
print(f"  Survey prévio (N=203):")
print(f"  → Evitariam falar com um ESTRANHO:     {p_evita_estranho*100:.0f}%")
print(f"  → Evitariam falar com um AMIGO:        {p_evita_amigo*100:.0f}%")
print("=" * 65)
```

```python
# --- O MODELO: da intuição à realidade ---
# Traduzimos o tamanho de efeito (Cohen's d) em PROBABILIDADE DE SUPERIORIDADE:
# a chance de uma pessoa que conversou ter tido um dia melhor
# do que uma pessoa que ficou em solidão.

# IC 95% do d real (aproximação de Hedges & Olkin)
se_d = np.sqrt((n_por_grupo + n_por_grupo) / (n_por_grupo * n_por_grupo)
               + d_real**2 / (2 * (n_por_grupo + n_por_grupo)))
ic_d = (d_real - 1.96 * se_d, d_real + 1.96 * se_d)

# Probabilidade de superioridade (common-language effect size)
# P(sup) = Φ(d / √2)
p_sup_real     = stats.norm.cdf(d_real / np.sqrt(2))       # realidade
p_sup_intuicao = stats.norm.cdf(-d_previsto / np.sqrt(2))  # o que a intuição previa

gap = p_sup_real - p_sup_intuicao

print("=" * 65)
print("  MODELO — CONVERSAR COM ESTRANHOS")
print("=" * 65)
print(f"  Tamanho de efeito real (d):            +{d_real}")
print(f"  IC 95% de d:                           [{ic_d[0]:.2f}, {ic_d[1]:.2f}]")
print()
print(f"  P(conversar > solidão) — REALIDADE:    {p_sup_real*100:.0f}%")
print(f"  P(conversar > solidão) — INTUIÇÃO:     {p_sup_intuicao*100:.0f}%")
print(f"  Gap entre realidade e intuição:        {gap*100:.0f} pontos percentuais")
print()
print(f"  A intuição não só erra a intensidade.")
print(f"  Ela erra o SINAL: aposta que a solidão vence,")
print(f"  quando na verdade quem conversa se sai melhor 2 em cada 3 vezes.")
print("=" * 65)
```

```python
# --- VISUALIZAÇÃO — 3 GRÁFICOS SEPARADOS ---

# ── GRÁFICO 1 — O PARADOXO: previsão vs. realidade ──
fig1, ax1 = plt.subplots(figsize=(10, 6))
# posicoes de positividade relativas à solidão (solidão = 0)
categorias = ['O que a intuição\nprevê', 'O que de fato\nacontece']
# intuição: conexão fica ABAIXO da solidão (-0.41); realidade: ACIMA (+0.63)
valores = [-d_previsto, d_real]
cores = [VERMELHO, VERDE]
bars = ax1.bar(categorias, valores, color=cores, alpha=0.85, width=0.45)
ax1.axhline(0, color=CINZA, linewidth=1.2)
ax1.text(0.5, 0.06, 'nível da solidão', transform=ax1.get_yaxis_transform(),
         ha='left', fontsize=9, color=CINZA)
ax1.set_ylabel('Positividade da experiência\n(em desvios-padrão, relativa à solidão)')
ax1.set_ylim(-0.75, 0.9)
ax1.set_title('O paradoxo: a intuição aposta no oposto da realidade\nConversar com um estranho no trem — Epley & Schroeder 2014',
              fontsize=13, pad=15)
for bar, v in zip(bars, valores):
    off = 0.05 if v > 0 else -0.09
    ax1.text(bar.get_x()+bar.get_width()/2, v+off, f'{v:+.2f} DP',
             ha='center', fontweight='bold', fontsize=13)
plt.figtext(0.5, 0.005,
            'Previsão N=66 · Experiência real N=97 (Metra, Chicago) | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-040-grafico-01-paradoxo.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 — ASSINATURA ESTATÍSTICA: Cohen's d com IC 95% ──
fig2, ax2 = plt.subplots(figsize=(10, 6))
ax2.errorbar(d_real, 1, xerr=[[d_real-ic_d[0]], [ic_d[1]-d_real]],
             fmt='o', color=VERDE, markersize=14, capsize=8, capthick=2,
             elinewidth=2, label='experiência real (1a)')
ax2.scatter(-d_previsto, 0.5, s=180, color=VERMELHO, marker='D', zorder=5,
            label='previsão da intuição (1b)')
ax2.axvline(0, color=CINZA, linestyle='--', linewidth=1.2)
ax2.text(0.02, 0.15, 'sem diferença', color=CINZA, fontsize=9)
ax2.set_yticks([0.5, 1])
ax2.set_yticklabels(['Intuição', 'Realidade'])
ax2.set_ylim(0.1, 1.4)
ax2.set_xlim(-0.8, 1.2)
ax2.set_xlabel("Cohen's d  (efeito de conversar vs. ficar em solidão)")
ax2.set_title(f"A ASSINATURA ESTATÍSTICA\nEfeito real d = +{d_real}  IC 95% [{ic_d[0]:.2f}, {ic_d[1]:.2f}]  ·  a intuição chuta o lado errado",
              fontsize=12.5, pad=15)
ax2.legend(loc='lower right')
plt.figtext(0.5, 0.005,
            'Tamanho de efeito de experimento de campo aleatorizado | Epley & Schroeder 2014 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-040-grafico-02-assinatura.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 — PROBABILIDADE DE SUPERIORIDADE ──
fig3, ax3 = plt.subplots(figsize=(10, 6))
x = np.linspace(-4, 4, 1000)
solidao = stats.norm.pdf(x, 0, 1)
conexao = stats.norm.pdf(x, d_real, 1)
ax3.plot(x, solidao, color=CINZA, linewidth=2, label='ficou em solidão')
ax3.fill_between(x, solidao, alpha=0.15, color=CINZA)
ax3.plot(x, conexao, color=VERDE, linewidth=2.5, label='conversou com estranho')
ax3.fill_between(x, conexao, alpha=0.25, color=VERDE)
ax3.axvline(0, color=CINZA, linestyle=':', linewidth=1)
ax3.axvline(d_real, color=VERDE, linestyle=':', linewidth=1)
ax3.set_xlabel('Positividade da experiência (desvios-padrão)')
ax3.set_ylabel('Densidade')
ax3.set_title(f'Probabilidade de superioridade: {p_sup_real*100:.0f}%\nEm 2 de cada 3 encontros, quem conversou teve um dia melhor',
              fontsize=13, pad=15)
ax3.legend(loc='upper left')
ax3.text(d_real+0.15, max(conexao)*0.5,
         f'a intuição achava\nque seria só {p_sup_intuicao*100:.0f}%',
         fontsize=10, color=VERMELHO)
plt.figtext(0.5, 0.005,
            'Common-language effect size a partir de d=0.63 | Epley & Schroeder 2014 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-040-grafico-03-superioridade.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**Na realidade, quem conversou com um estranho teve um dia melhor em 2 de cada 3 vezes (67%).
A intuição apostava no contrário: achava que a chance era de só 39%.**

O erro não é de intensidade. É de direção. A gente não subestima um pouco o prazer de
conectar. A gente aposta no sinal errado, convencida de que a solidão vence, quando os
dados mostram que ela perde.

E o custo disso é invisível, porque ninguém aprende a lição. Você prevê que vai ser
desagradável, então não fala com ninguém, então nunca descobre que estava errada. O
erro se protege sozinho.

A conversa média durou 14 minutos. Não era ninguém para a vida inteira. Era um dia
comum, num trem comum, ficando um pouco melhor do que seria.

*Quantos dias você deixou um pouco piores só para não correr o risco de que ficassem
melhores?*

---

### ⚠️ Limitações do Modelo

- A "probabilidade de superioridade" é uma tradução do tamanho de efeito (d=0,63)
  assumindo distribuições normais; é uma forma de comunicar o d, não uma contagem
  direta de encontros.
- O experimento-chave foi feito com comutadores de trem em Chicago; contextos, culturas
  e temperamentos variam, e o efeito pode ser diferente em outros ambientes.
- Amostras de campo modestas (N=97 e N=66); o padrão se repetiu em nove experimentos do
  estudo, o que reforça a robustez, mas cada experimento isolado é pequeno.
- O estudo mede uma única interação; não afirma que falar com todo estranho, sempre,
  melhora o dia.

*A ciência é honesta sobre o que não sabe. O modelo também.*
