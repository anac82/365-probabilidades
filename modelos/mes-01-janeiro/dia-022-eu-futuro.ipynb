### 📖 A História

Tem uma pergunta que ninguém me fez, mas que explica muita coisa: por que é tão fácil sacrificar o meu futuro pelo meu presente?

Adiar o treino. Não guardar dinheiro. Aceitar um emprego que paga agora mas não constrói nada. Ficar num relacionamento que consome mais do que entrega. Em todos esses momentos, eu estava fazendo uma troca — dando algo da Ana de 10 anos pra frente pra resolver um problema da Ana de agora.

E a Ana de 10 anos pra frente? Ela nunca reclamou. Porque ela não existia ainda. Pelo menos não na minha cabeça.

Pesquisadores de Stanford colocaram pessoas dentro de uma máquina de ressonância magnética e pediram: pense em você mesma daqui a 10 anos. E observaram o que o cérebro fez.

O que ele fez foi tratar a Ana do futuro como uma estranha.

---

### 📚 O Conceito: Continuidade do Eu Futuro

O filósofo Derek Parfit propôs em 1971 que, se uma pessoa não sente continuidade com quem ela vai ser no futuro, ela não tem mais razão de guardar dinheiro pra esse "eu futuro" do que de dar dinheiro a um estranho na rua.

Hal Ersner-Hershfield, então em Stanford, testou isso empiricamente. Usando fMRI, ele mediu a ativação cerebral enquanto pessoas avaliavam traços de personalidade aplicados a quatro alvos: eu-presente, eu-futuro (10 anos à frente), uma pessoa conhecida, e um estranho.

O resultado: a região do cérebro associada ao processamento do "self" — o córtex cingulado anterior rostral (rACC) e o córtex pré-frontal medial (MPFC) — se ativava forte pro eu-presente, mas caía significativamente pro eu-futuro. O padrão de ativação do eu-futuro era mais parecido com o de um estranho do que com o de si mesmo.

E quanto maior essa queda — quanto mais o cérebro tratava o eu-futuro como "outra pessoa" — mais a pessoa descontava recompensas futuras, preferindo ganhar menos agora a ganhar mais depois.

### 🧮 O Modelo
**Fontes:**
- Ersner-Hershfield, H., Wimmer, G.E. & Knutson, B. (2009) — *Saving for the future self: Neural measures of future self-continuity predict temporal discounting* — Social Cognitive and Affective Neuroscience, 4(1), 85-92 — estudo de fMRI
- Ersner-Hershfield, H., Garton, M.T., Ballard, K., Samanez-Larkin, G.R. & Knutson, B. (2009) — *Don't stop thinking about tomorrow: Individual differences in future self-continuity account for saving* — Judgment and Decision Making, 4(4), 280-286 — 3 experimentos
- Hershfield, H.E., Goldstein, D.G., Sharpe, W.F., Fox, J., Yeykelis, L., Carstensen, L.L. & Bailenson, J.N. (2011) — *Increasing saving behavior through age-progressed renderings of the future self* — Journal of Marketing Research, 48(SPL), S23-S37 — 4 estudos experimentais


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print('✅ Bibliotecas carregadas')
```

```python
# --- DADOS DA LITERATURA ---

# Ersner-Hershfield, Wimmer & Knutson (2009) — fMRI — Stanford
# Participantes avaliaram traços de personalidade aplicados a:
# eu-presente, eu-futuro (10 anos), pessoa conhecida, estranho
# Região de interesse: rACC (córtex cingulado anterior rostral)
# Resultado: ativação rACC caiu significativamente pra eu-futuro vs eu-presente
# E essa diferença prediz desconto temporal (preferir agora vs depois)

# Ersner-Hershfield et al. (2009) — "Don't stop thinking about tomorrow"
# 3 experimentos — criaram o índice de continuidade do eu futuro
# Estudo 1: N=63 (Stanford), teste-reteste com mínimo 1 semana de intervalo
r_similaridade_recompensa = 0.42  # correlação: similaridade eu futuro → escolha de recompensas atrasadas
n_estudo1 = 63

# Estudo 3: maior continuidade → maior patrimônio financeiro acumulado
# (controlando por idade e educação)

# Hershfield et al. (2011) — Journal of Marketing Research — 4 estudos
# Participantes que viram avatares envelhecidos de si mesmos (realidade virtual)
# alocaram significativamente mais dinheiro para aposentadoria
# Público: adultos 18-35 anos
# Estudo 2: grupo avatar envelhecido alocou ~6.8% do salário vs ~5.2% controle
alocacao_avatar_envelhecido_pct = 6.8
alocacao_controle_pct = 5.2
ganho_alocacao_pct = alocacao_avatar_envelhecido_pct - alocacao_controle_pct

fator_correcao = 0.80

print('=' * 65)
print('  DADOS DA LITERATURA — EU FUTURO COMO ESTRANHO')
print('=' * 65)
print(f'\n  Ersner-Hershfield, Wimmer & Knutson (2009) — fMRI:')
print(f'  → Ativação rACC cai significativamente pra eu-futuro vs eu-presente')
print(f'  → Padrão do eu-futuro mais parecido com estranho do que consigo mesmo')
print(f'  → Quanto maior a queda, mais a pessoa desconta recompensas futuras')
print(f'\n  Ersner-Hershfield et al. (2009) — 3 experimentos, N={n_estudo1} (Estudo 1):')
print(f'  → r={r_similaridade_recompensa} entre similaridade eu-futuro e escolhas de recompensas atrasadas')
print(f'  → Maior continuidade → maior patrimônio financeiro acumulado')
print(f'\n  Hershfield et al. (2011) — 4 estudos (realidade virtual):')
print(f'  → Avatar envelhecido: alocaram {alocacao_avatar_envelhecido_pct}% do salário pra aposentadoria')
print(f'  → Controle: alocaram {alocacao_controle_pct}% do salário')
print(f'  → Diferença: +{ganho_alocacao_pct:.1f}pp')
print(f'\n  Fator de correção: x{fator_correcao}')
print('=' * 65)
print()
print('  ⚠️  NOTA METODOLÓGICA:')
print('  Os estudos de fMRI usam amostras pequenas (típico da neuroimagem).')
print('  Os estudos comportamentais (Estudo 1-3 de 2009, Estudos 1-4 de 2011)')
print('  replicaram o padrão com amostras maiores e métodos diferentes,')
print('  aumentando a confiança no efeito. Os percentuais de alocação de')
print('  aposentadoria (6,8% vs 5,2%) são de um dos estudos de 2011 e podem')
print('  variar por estudo — usamos o que teve valores numéricos claros.')
```

```python
# --- O MODELO ---

# Correlação entre continuidade do eu futuro e comportamento de poupança
# r = 0.42 (Estudo 1, Ersner-Hershfield et al. 2009)
# Coeficiente de determinação: quanto da variância em poupança é explicada
r2 = r_similaridade_recompensa ** 2

# IC 95% para a correlação via transformação de Fisher
z = np.arctanh(r_similaridade_recompensa)  # Fisher z-transformation
se_z = 1 / np.sqrt(n_estudo1 - 3)
z_low = z - 1.96 * se_z
z_high = z + 1.96 * se_z
r_low = np.tanh(z_low)
r_high = np.tanh(z_high)

# Efeito do avatar envelhecido, com fator de correção
ganho_alocacao_corrigido = ganho_alocacao_pct * fator_correcao

print('=' * 65)
print('  MODELO — RESULTADO')
print('=' * 65)
print(f'\n  Correlação: continuidade eu-futuro → valorização de recompensas futuras')
print(f'  → r = {r_similaridade_recompensa}')
print(f'  → IC 95%: [{r_low:.2f}, {r_high:.2f}]')
print(f'  → r² = {r2:.2f} ({r2*100:.0f}% da variância explicada)')
print(f'\n  Efeito de "ver" o eu futuro (avatar envelhecido):')
print(f'  → Alocação para aposentadoria: +{ganho_alocacao_pct:.1f}pp')
print(f'  → Corrigido (×{fator_correcao}): +{ganho_alocacao_corrigido:.1f}pp')
print(f'\n  INTERPRETAÇÃO:')
print(f'  → Quanto mais você se sente conectada com quem vai ser daqui a 10 anos,')
print(f'    mais você guarda dinheiro, cuida da saúde e faz escolhas de longo prazo.')
print(f'  → Quando o cérebro trata o eu-futuro como estranho, ele não tem motivo')
print(f'    neurológico para sacrificar o presente em favor desse "outro".')
print('=' * 65)
```

```python
# --- VISUALIZAÇÃO ---

# GRÁFICO 1 — Padrão de ativação cerebral (ilustrativo, baseado no estudo)
fig1, ax1 = plt.subplots(figsize=(12, 8))

categorias = ['Eu presente', 'Pessoa\nconhecida', 'Eu futuro\n(10 anos)', 'Estranho']
# Valores ilustrativos do padrão qualitativo descrito no paper
# (paper não reporta valores absolutos de ativação em unidade pública;
#  o padrão é: eu-presente > conhecido > eu-futuro ≈ estranho)
ativacao_relativa = [1.0, 0.6, 0.35, 0.3]
cores1 = ['#2a8a82', '#888888', '#c0392b', '#c0392b']

bars = ax1.bar(categorias, ativacao_relativa, color=cores1, alpha=0.85, width=0.5)
ax1.set_ylim(0, 1.3)
ax1.set_ylabel('Ativação relativa no rACC (ilustrativa)', fontsize=12)
ax1.set_title('Seu cérebro trata o "eu futuro" mais como um estranho do que como você\n'
              'Ersner-Hershfield, Wimmer & Knutson (2009) — fMRI · padrão qualitativo',
              fontsize=13, pad=15)

# Anotação do achado principal
ax1.annotate('Padrão\nsimilar', xy=(2.5, 0.35), fontsize=12, ha='center',
             color='#c0392b', fontweight='bold')
ax1.annotate('', xy=(2, 0.38), xytext=(3, 0.33),
             arrowprops=dict(arrowstyle='<->', color='#c0392b', lw=2))

plt.tight_layout()
plt.savefig('dia-022-grafico-01-ativacao-cerebral.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Gráfico 1 salvo!')


# GRÁFICO 2 — Efeito do avatar envelhecido na alocação para aposentadoria
fig2, ax2 = plt.subplots(figsize=(12, 8))

categorias2 = ['Controle\n(avatar atual)', 'Avatar\nenvelhecido']
valores2 = [alocacao_controle_pct, alocacao_avatar_envelhecido_pct]
cores2 = ['#888888', '#2a8a82']

bars2 = ax2.bar(categorias2, valores2, color=cores2, alpha=0.85, width=0.45)
ax2.set_ylim(0, 10)
ax2.set_ylabel('% do salário alocado para aposentadoria', fontsize=12)
ax2.set_title('Ver o "eu futuro" muda o comportamento de poupança\n'
              'Hershfield et al. (2011) — realidade virtual · adultos 18-35 anos',
              fontsize=13, pad=15)

for bar, v in zip(bars2, valores2):
    ax2.text(bar.get_x() + bar.get_width() / 2, v + 0.2,
             f'{v}%', ha='center', fontweight='bold', fontsize=16)

ax2.annotate('', xy=(1, 6.8), xytext=(0, 5.2),
             arrowprops=dict(arrowstyle='->', color='#c8a84b', lw=2.5))
ax2.text(0.5, 6.3, f'+{ganho_alocacao_pct:.1f}pp', ha='center',
         fontsize=14, color='#c8a84b', fontweight='bold')

plt.tight_layout()
plt.savefig('dia-022-grafico-02-alocacao-aposentadoria.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Gráfico 2 salvo!')


# GRÁFICO 3 — IC 95% da correlação (Fisher z-transformation)
fig3, ax3 = plt.subplots(figsize=(12, 8))

# Distribuição amostral da correlação via Fisher z
z_vals = np.linspace(z - 4*se_z, z + 4*se_z, 1000)
r_vals = np.tanh(z_vals)
y_vals = stats.norm.pdf(z_vals, loc=z, scale=se_z)

ax3.plot(r_vals, y_vals, color='#2a8a82', linewidth=3)
ax3.fill_between(r_vals, y_vals, alpha=0.2, color='#2a8a82')
ax3.axvline(x=r_low, color='#c8a84b', linestyle='--', linewidth=2,
            label=f'IC 95%: [{r_low:.2f}, {r_high:.2f}]')
ax3.axvline(x=r_high, color='#c8a84b', linestyle='--', linewidth=2)
ax3.axvline(x=r_similaridade_recompensa, color='#2a8a82', linewidth=2,
            label=f'r = {r_similaridade_recompensa}')
ax3.axvline(x=0, color='#333333', linestyle=':', linewidth=1.5,
            label='r = 0 (sem efeito)')

ax3.set_xlabel('Correlação: similaridade eu-futuro → escolha de recompensas atrasadas', fontsize=12)
ax3.set_ylabel('Densidade', fontsize=12)
ax3.set_title('Distribuição da correlação — Continuidade do Eu Futuro\n'
              f'Ersner-Hershfield et al. (2009) — N={n_estudo1} · Fisher z-transformation',
              fontsize=13, pad=15)
ax3.legend(fontsize=11)

plt.tight_layout()
plt.savefig('dia-022-grafico-03-correlacao.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Gráfico 3 salvo!')
```

### 💡 O Insight

Quando pesquisadores pediram para pessoas pensarem em si mesmas daqui a 10 anos dentro de uma máquina de ressonância magnética, o cérebro tratou o "eu futuro" quase como um **estranho** — não como si mesmo. A região cerebral associada ao processamento do self (rACC/MPFC) se ativou significativamente menos pro eu-futuro do que pro eu-presente.

E esse padrão tem consequência direta: quanto maior essa diferença neural, mais a pessoa **descontava recompensas futuras** — preferia ganhar menos agora a ganhar mais depois. Em três experimentos, pessoas com maior "continuidade do eu futuro" acumularam mais patrimônio financeiro ao longo da vida, mesmo controlando por idade e educação (r=0,42).

A parte mais provocativa: quando participantes viram um **avatar envelhecido de si mesmos** em realidade virtual, passaram a alocar significativamente mais dinheiro para aposentadoria do que quem viu apenas seu avatar atual — uma diferença de 1,6 pontos percentuais que, acumulada ao longo de décadas, muda o desfecho financeiro de uma vida.

Isso não significa que você é egoísta por não pensar no futuro. Significa que seu cérebro está fazendo algo que faz sentido evolutivo — priorizar o presente — mas que não faz sentido financeiro, nem de saúde, nem de construção de vida.

*Da próxima vez que adiar algo "pro futuro", vale lembrar: pra quem você está adiando? Pra você mesma — ou pra uma estranha?*

### ⚠️ Limitações do Modelo
- O estudo de fMRI usa amostra pequena (típico de neuroimagem) — o achado foi replicado em estudos comportamentais com amostras maiores, mas a evidência neural em si precisa de replicação com N maior
- O Gráfico 1 é uma ilustração do padrão qualitativo descrito no paper (eu-presente > eu-futuro ≈ estranho no rACC), não os valores absolutos de ativação — o paper não reporta unidades que permitam reprodução numérica exata
- A correlação de r=0,42 entre continuidade e valorização de recompensas futuras vem de um único estudo com N=63 — o IC 95% é amplo
- Os valores de alocação para aposentadoria (6,8% vs 5,2%) são de um dos quatro estudos do paper de 2011 — outros estudos usaram medidas diferentes (ex: "quanto dinheiro você alocaria") e os números variam
- Todos os participantes dos estudos experimentais eram jovens adultos (18-35 anos) — não sabemos se o efeito do avatar envelhecido seria o mesmo em pessoas mais velhas
- O conceito filosófico de Parfit (1971) é a base teórica, mas a tradução empírica não prova que o cérebro "decide" tratar o eu-futuro como estranho — apenas que a ativação neural segue esse padrão
- O fator de correção (×0,80) foi aplicado à diferença de alocação do estudo de 2011

*A ciência é honesta sobre o que não sabe. O modelo também.*
