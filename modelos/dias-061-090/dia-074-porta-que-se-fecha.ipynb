## Qual a probabilidade de existir uma porta que se fecha aos 17 anos?

**Tipo:** Preditivo

**Decisão analisada:** Ainda dá tempo, ou o prazo já passou?

### 📖 A História

Meu filho foi morar em outro país este ano, com doze anos.

Entre as mil coisas que ocupam a cabeça de uma mãe nessa situação, uma é bem menor que
as outras e mesmo assim não sai: em quanto tempo ele vai falar como quem nasceu lá.

É uma pergunta pequena, quase frívola perto do resto. Mas é das poucas que tem
resposta em número.

E o número existe porque, em 2018, alguém conseguiu resolver um problema que travava
essa literatura havia meio século.

---

### 📚 O Conceito: por que ninguém tinha conseguido medir isso

A hipótese do período crítico é antiga: existiria uma janela de tempo em que aprender
uma língua é fácil, e depois dela deixa de ser.

O problema nunca foi de ideia. Foi de método, e ele é bonito de explicar.

Quando você mede a proficiência de um adulto que começou a aprender inglês aos 8 anos
e a compara com a de outro que começou aos 25, três coisas estão misturadas: **a idade
atual**, **a idade em que começou** e **quantos anos de prática já teve**.

Quem começou cedo teve mais tempo de prática. Então, se sai melhor, não dá para saber
se é porque começou cedo ou porque praticou mais. E como separar isso exigia amostras
gigantescas, os estudos anteriores ficavam com poder insuficiente para decidir.

Esse é o mesmo problema do #066, em outra roupa: o dado agregado não vem com a
instrução de como lê-lo. Aqui, três variáveis andam juntas e ninguém conseguia
desgrudá-las.

---

### 🧮 O Modelo

Um conjunto de dados enorme, um modelo computacional que separa as três variáveis, e
uma disputa aberta sobre o resultado.

**Fontes:**
- Hartshorne, J. K., Tenenbaum, J. B. & Pinker, S., 2018 · "A critical period for
  second language acquisition: Evidence from 2/3 million English speakers" ·
  *Cognition* 177, 263-277 · **N = 669.498 falantes nativos e não nativos de inglês** ·
  modelo computacional que estima a trajetória da capacidade de aprendizado
  **separando idade atual, idade da primeira exposição e anos de experiência**
- **Achado principal:** a capacidade de aprender gramática se preserva quase até o
  limiar da vida adulta, **17,4 anos**, e depois declina de forma constante. O formato
  é um platô seguido de declínio contínuo.
- **Achado secundário, menos citado:** proficiência de nível nativo só é alcançada por
  quem começa entre **10 e 12 anos**, porque leva cerca de **30 anos de exposição**
  para atingir o platô de aprendizado.
- van der Slik, F. e colegas, 2022 · *Language Learning* · **reanálise dos mesmos
  dados** · a conclusão de uma idade crítica única e nitidamente definida aos 17,4
  anos se baseia em resultados artificiais · um modelo **contínuo** se ajusta melhor
  para monolíngues, bilíngues e aprendizes de imersão precoce · só para não imersão e
  imersão tardia o modelo descontínuo se ajusta melhor, com idades críticas de
  **18,6 e 19,0 anos**
- Frank, M. C., 2018 · *Trends in Cognitive Sciences* 22(8), 669-671 · pede cautela
  com a novidade dos dados e do método

**Nota metodológica sobre o fator ×0.80:** não se aplica. A proficiência foi medida
por teste de gramática, não por autorrelato de quão bem a pessoa acha que fala.


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Hartshorne, Tenenbaum & Pinker, 2018, Cognition 177, 263-277

n_falantes       = 669_498
idade_critica    = 17.4      # fim do plato, modelo descontinuo
janela_nativo    = (10, 12)  # idade de inicio para atingir nivel nativo
anos_ate_plato   = 30        # tempo de exposicao necessario para atingir o plato

# van der Slik et al., 2022, Language Learning - reanalise dos MESMOS dados
# Modelo continuo se ajusta melhor para tres grupos.
# Modelo descontinuo se ajusta melhor para dois grupos, com estas idades criticas:
CRITICAS_REANALISE = {
    'Nao imersao':     18.6,
    'Imersao tardia':  19.0,
}
GRUPOS_CONTINUOS = ['Monolingues', 'Bilingues', 'Imersao precoce']

aplica_fator_080 = False

print("=" * 70)
print("  DADOS - QUANDO A CAPACIDADE DE APRENDER GRAMATICA DECLINA")
print("=" * 70)
print(f"\n  Hartshorne, Tenenbaum & Pinker, 2018 (Cognition 177):")
print(f"  -> N = {n_falantes:,}".replace(",", ".") + " falantes de ingles")
print(f"  -> O modelo separa TRES coisas que andavam juntas:")
print(f"     idade atual, idade da primeira exposicao e anos de experiencia")
print(f"  -> Plato ate {idade_critica} anos, depois declinio constante")
print(f"  -> Nivel nativo so para quem comeca entre"
      f" {janela_nativo[0]} e {janela_nativo[1]} anos")
print(f"  -> Motivo: leva cerca de {anos_ate_plato} anos de exposicao"
      f" para atingir o plato")
print(f"\n  van der Slik et al., 2022 (reanalise dos MESMOS dados):")
print(f"  -> Para {', '.join(GRUPOS_CONTINUOS)}: modelo CONTINUO se ajusta melhor")
print(f"     (ou seja, sem porta que se fecha)")
for grupo, idade in CRITICAS_REANALISE.items():
    print(f"  -> Para {grupo}: modelo descontinuo, idade critica de {idade} anos")
print(f"\n  Fator x0.80 aplicado: {aplica_fator_080}")
print("=" * 70)
```

```python
# --- O MODELO ---
# A aritmetica que quase ninguem faz, e que explica a janela dos 10 aos 12.

idades_inicio = np.arange(2, 41, 1)
idade_de_chegada = idades_inicio + anos_ate_plato

# Quem comeca antes da idade critica ainda esta na fase de plato de aprendizado.
# Quem comeca depois ja inicia com a capacidade em declinio.
comeca_no_plato = idades_inicio <= idade_critica

# Traducao: comecando em cada idade, com que idade a pessoa atinge o plato
exemplos = [5, 8, 10, 12, 17, 20, 25, 30, 40]

print("=" * 70)
print("  MODELO - A ARITMETICA DA JANELA")
print("=" * 70)
print(f"\n  A conta e simples e quase ninguem faz:")
print(f"  -> Leva cerca de {anos_ate_plato} anos de exposicao para chegar ao topo")
print(f"  -> Entao a idade em que voce chega la e:"
      f" idade de inicio + {anos_ate_plato}")
print(f"\n  {'comeca aos':>12} | {'chega ao plato aos':>19} | situacao")
print(f"  {'-'*12}-+-{'-'*19}-+-{'-'*34}")
for i in exemplos:
    chegada = i + anos_ate_plato
    if i <= janela_nativo[1]:
        situacao = "ainda alcanca nivel nativo"
    elif i <= idade_critica:
        situacao = "aprende bem, mas nao chega ao topo"
    else:
        situacao = "ja comeca com a capacidade caindo"
    print(f"  {i:>12} | {chegada:>19} | {situacao}")

print(f"\n  Por que a janela fecha em {janela_nativo[1]} e nao em"
      f" {idade_critica}:")
print(f"  -> Nao e que a pessoa pare de aprender aos {idade_critica}.")
print(f"  -> E que quem comeca depois dos {janela_nativo[1]} anos ainda esta")
print(f"     subindo quando a capacidade comeca a cair.")
print(f"  -> O prazo nao e o fim do aprendizado. E o fim do TEMPO SUFICIENTE.")

print(f"\n  E a disputa sobre a forma da curva:")
print(f"  -> Hartshorne et al.: plato ate {idade_critica}, depois queda")
print(f"  -> van der Slik et al., nos MESMOS dados: para monolingues,")
print(f"     bilingues e imersao precoce, a queda e CONTINUA desde cedo")
print(f"  -> Nas duas leituras, comecar cedo importa.")
print(f"     A briga e sobre PORTA ou RAMPA, nao sobre direcao.")
print("=" * 70)
```

```python
# --- VISUALIZACAO ---

DOURADO = '#c8a84b'
VERMELHO = '#c0392b'
VERDE = '#2a8a82'
CINZA = '#6b6a64'

# GRAFICO 1 - Porta ou rampa: as duas leituras dos MESMOS dados
fig1, ax1 = plt.subplots(figsize=(12, 8))

idades = np.linspace(2, 45, 500)

# Leitura 1: plato ate 17,4 e declinio depois (forma esquematica)
curva_porta = np.where(idades <= idade_critica, 1.0,
                       np.exp(-(idades - idade_critica) / 12))
# Leitura 2: declinio continuo desde cedo (forma esquematica)
curva_rampa = np.exp(-(idades - 2) / 26)

ax1.plot(idades, curva_porta * 100, color=VERMELHO, linewidth=3.2,
         label='Leitura 1: platô até 17,4 e depois queda')
ax1.plot(idades, curva_rampa * 100, color=VERDE, linewidth=3.2, linestyle='--',
         label='Leitura 2: declínio contínuo desde cedo')

ax1.axvline(x=idade_critica, color=VERMELHO, linestyle=':', linewidth=2, alpha=0.8)
ax1.text(idade_critica + 0.4, 96, '17,4 anos', fontsize=13, color=VERMELHO,
         fontweight='bold')

ax1.axvspan(janela_nativo[0], janela_nativo[1], color=DOURADO, alpha=0.18)
ax1.text((janela_nativo[0] + janela_nativo[1]) / 2, 24,
         'janela para\nnível nativo\n10 a 12 anos', ha='center', fontsize=11,
         color='#7a5f1f', fontweight='bold')

ax1.set_xlim(2, 45)
ax1.set_ylim(0, 108)
ax1.set_xlabel('Idade da primeira exposição à língua (anos)')
ax1.set_ylabel('Capacidade de aprender gramática (escala relativa)')
ax1.legend(frameon=False, fontsize=12, loc='lower left')
ax1.set_title('Porta ou rampa: duas leituras dos mesmos 669.498 falantes\n'
              'Hartshorne, Tenenbaum & Pinker, 2018 · reanálise de van der Slik '
              'et al., 2022',
              fontsize=14, pad=18)
ax1.text(0.5, -0.135,
         'Formas esquemáticas dos dois modelos em disputa. Não são as curvas '
         'ajustadas dos artigos.',
         transform=ax1.transAxes, ha='center', fontsize=11, color=CINZA,
         style='italic')

plt.figtext(0.5, 0.005,
            'Fontes: Cognition 177, 263-277 e Language Learning, 2022'
            '  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-074-grafico-01-porta-ou-rampa.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 1 salvo")

# GRAFICO 2 - A assinatura: a aritmetica dos 30 anos
fig2, ax2 = plt.subplots(figsize=(12, 8))

inicios = np.array(exemplos)
chegadas = inicios + anos_ate_plato
cores2 = [DOURADO if i <= janela_nativo[1] else
          (VERDE if i <= idade_critica else VERMELHO) for i in inicios]

y = np.arange(len(inicios))[::-1]
for yi, ini, cheg, cor in zip(y, inicios, chegadas, cores2):
    ax2.plot([ini, cheg], [yi, yi], color=cor, linewidth=7, alpha=0.85,
             solid_capstyle='round')
    ax2.scatter([ini], [yi], s=90, color=cor, zorder=3)
    ax2.text(cheg + 1, yi, f'{cheg} anos', va='center', fontsize=12, color=cor,
             fontweight='bold')

ax2.axvline(x=idade_critica, color='#333', linestyle='--', linewidth=2)
ax2.text(idade_critica + 0.6, y.max() + 0.55, 'capacidade começa a cair aqui',
         fontsize=12, color='#333')

ax2.set_yticks(y)
ax2.set_yticklabels([f'começa aos {i}' for i in inicios], fontsize=12)
ax2.set_xlim(0, 78)
ax2.set_ylim(-0.7, y.max() + 0.9)
ax2.set_xlabel('Idade (anos)')
ax2.set_title('A assinatura: a barra é o tempo até o topo, cerca de 30 anos\n'
              'Quem começa depois dos 12 ainda está subindo quando a capacidade '
              'começa a cair',
              fontsize=14, pad=18)
ax2.text(0.5, -0.12,
         'idade de chegada ao platô = idade de início + 30 anos de exposição',
         transform=ax2.transAxes, ha='center', fontsize=13, color=DOURADO,
         fontweight='bold')

plt.figtext(0.5, 0.005,
            'Fonte: Hartshorne, Tenenbaum & Pinker, 2018, Cognition 177, 263-277'
            '  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-074-grafico-02-aritmetica.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 2 salvo")

# GRAFICO 3 - O que o modelo conseguiu separar
fig3, ax3 = plt.subplots(figsize=(12, 7))

ax3.axis('off')
ax3.set_xlim(0, 10)
ax3.set_ylim(0, 10)
caixa = dict(boxstyle='round,pad=0.7', facecolor='white', edgecolor=CINZA,
             linewidth=1.6)

ax3.text(5, 9.2, 'O problema que travava a literatura há meio século',
         ha='center', fontsize=17, fontweight='bold')

ax3.text(2.2, 7.4, 'Idade\natual', ha='center', va='center', fontsize=14, bbox=caixa)
ax3.text(5.0, 7.4, 'Idade da\nprimeira exposição', ha='center', va='center',
         fontsize=14, bbox=caixa)
ax3.text(7.8, 7.4, 'Anos de\nexperiência', ha='center', va='center', fontsize=14,
         bbox=caixa)

ax3.annotate('', xy=(5, 5.9), xytext=(2.2, 6.7),
             arrowprops=dict(arrowstyle='->', color=CINZA, lw=2))
ax3.annotate('', xy=(5, 5.9), xytext=(5.0, 6.7),
             arrowprops=dict(arrowstyle='->', color=CINZA, lw=2))
ax3.annotate('', xy=(5, 5.9), xytext=(7.8, 6.7),
             arrowprops=dict(arrowstyle='->', color=CINZA, lw=2))

ax3.text(5, 5.2, 'As três andam juntas.\nQuem começou cedo praticou mais.',
         ha='center', va='center', fontsize=15, color=VERMELHO, fontweight='bold',
         bbox=dict(boxstyle='round,pad=0.8', facecolor='#fdf3f1',
                   edgecolor=VERMELHO, linewidth=2))

ax3.annotate('', xy=(5, 3.4), xytext=(5, 4.4),
             arrowprops=dict(arrowstyle='->', color=CINZA, lw=2.4))
ax3.text(5, 2.7,
         f'Separar as três exige amostra gigante.\n'
         f'{n_falantes:,}'.replace(",", ".") + ' pessoas foi o que bastou.',
         ha='center', va='center', fontsize=15, color=VERDE, fontweight='bold',
         bbox=dict(boxstyle='round,pad=0.8', facecolor='#f0f7f6',
                   edgecolor=VERDE, linewidth=2))

ax3.text(5, 0.9,
         'Estudos anteriores não erraram a pergunta. Não tinham gente suficiente\n'
         'para responder.',
         ha='center', fontsize=13, style='italic', color='#333')

plt.figtext(0.5, 0.005,
            'Fonte: Hartshorne, Tenenbaum & Pinker, 2018  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-074-grafico-03-tres-variaveis.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 3 salvo")
```

### 💡 O Insight

**669.498 pessoas.** É o maior estudo já feito sobre quando a capacidade de aprender
gramática declina, e ele existe porque alguém finalmente teve gente suficiente para
separar três coisas que sempre andaram juntas: a idade atual, a idade em que a pessoa
começou e quantos anos de prática ela teve.

O achado que virou manchete: a capacidade se mantém em platô até os **17,4 anos** e
depois cai de forma constante.

E o achado que quase ninguém repetiu, que é o que interessa a quem tem filho: **nível
de falante nativo só é alcançado por quem começa entre 10 e 12 anos.**

O motivo é aritmético e me pegou de surpresa. Leva cerca de **30 anos de exposição**
para chegar ao topo do aprendizado. Quem começa aos 12 chega lá por volta dos 42. Quem
começa aos 20 precisaria de tempo que a capacidade não dá mais, porque ela já entrou em
declínio antes de a conta fechar.

Ou seja, o prazo não é o fim do aprendizado. **É o fim do tempo suficiente.**

Agora, a parte em que eu seguro o número.

Em 2022, van der Slik e colegas reanalisaram **os mesmos dados** e concluíram que
aquela idade crítica única e nitidamente definida é artefato do modelo escolhido. Para
monolíngues, bilíngues e quem aprendeu por imersão precoce, um modelo de **declínio
contínuo** se ajusta melhor. Só para quem aprendeu sem imersão ou com imersão tardia o
modelo de porta se sustenta, e ainda assim com idades de 18,6 e 19,0.

A diferença entre as duas leituras não é acadêmica. Uma diz que existe uma porta que
se fecha. A outra diz que existe uma rampa que desce desde sempre.

E aqui está a única coisa que as duas leituras concordam, que é o que eu levo daqui:

**Começar cedo importa, e importa mais do que se esforçar depois.** A briga é sobre a
forma da curva, não sobre a direção dela.

Isso vale para língua. E é bem provável que valha para qualquer coisa que dependa de
acúmulo lento: instrumento, esporte, ofício. A pergunta útil não é se ainda dá tempo
de aprender. Quase sempre dá.

É se ainda dá tempo de chegar ao topo, e essa é uma pergunta diferente, com resposta
menos generosa.

*O que você começou tarde e mesmo assim valeu a pena?*

### ⚠️ Limitações do Modelo

- **A idade crítica de 17,4 anos está sob disputa.** A reanálise de van der Slik e
  colegas, em 2022, usando os mesmos dados, encontra melhor ajuste para um modelo
  contínuo em três dos cinco grupos de aprendizes. Frank, em 2018, também pediu
  cautela com a novidade do método.
- **As curvas do gráfico 1 são esquemáticas.** Elas representam a forma dos dois
  modelos em disputa, não são as curvas ajustadas publicadas nos artigos.
- **A aritmética dos 30 anos é uma leitura simplificada.** O artigo descreve o tempo
  necessário para atingir o platô de aprendizado; o cálculo linear feito aqui serve
  para mostrar por que a janela fecha antes da idade crítica, não como reprodução do
  modelo.
- **O desfecho é gramática, não idioma inteiro.** Vocabulário, sotaque e fluência
  conversacional têm trajetórias próprias e não foram medidos aqui.
- **Amostra autosselecionada.** Os dados vêm de um teste de gramática viral na
  internet, respondido por quem se interessou em fazê-lo. Não é amostra probabilística
  da população.
- **Nada aqui diz que adulto não aprende.** Diz que a trajetória até o nível nativo
  exige um tempo que começa a faltar. Aprender bem e aprender como nativo são metas
  diferentes.
- Fator ×0.80 não aplicado: proficiência medida por teste, não por autorrelato.

*A ciência é honesta sobre o que não sabe. O modelo também.*
