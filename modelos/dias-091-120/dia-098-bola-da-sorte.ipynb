# 365 Probabilidades - Dia #098
## Qual a probabilidade de o seu amuleto da sorte funcionar?

**Tipo:** Crença
**Data de publicação:** 2026-09-19
**Ferramenta:** Python
**Decisão analisada:** A superstição melhora o desempenho de verdade?
**Hashtag:** #365Probabilidades #Dia098

---

### 📖 A História

Todo mundo tem um. Inclusive você, que jura que não.

A camisa que não lava em dia de final. A caneta que passou na
prova e virou titular de todas as outras. O mesmo lugar no sofá,
porque da última vez deu certo. O "não fala nada que estraga".

E em 2010, a ciência pareceu confirmar que a gente estava certo.
Um experimento na Psychological Science entregou bolas de golfe a
28 pessoas. Para metade, o pesquisador disse: "aqui está a bola;
até agora ela tem se mostrado uma bola de sorte". A outra metade
ouviu só "esta é a bola que todos usaram".

Quem recebeu a bola "de sorte" acertou **65% das tacadas.
O controle, 48%.** O amuleto valia 1,6 tacadas em 10. O achado
virou manchete, capítulo de livro, prova científica da cueca da
sorte.

Aí, em 2014, aconteceu o formato mais limpo que a ciência tem:
**o duelo direto**. Dois pesquisadores refizeram o experimento,
pré-registrado, com o dobro e depois o quádruplo do N, e com a
própria autora original ajudando a deixar tudo idêntico: a mesma
tarefa, a mesma distância, a mesma frase.

Resultado, nas palavras deles: desempenho **"quase idêntico"**
ao controle. Duas vezes.

A bola da sorte era só uma bola.

---

### 📚 O Conceito: O Duelo Perfeito

O achado original tinha um mecanismo charmoso, e é por isso que
ele colou: a superstição não seria mágica, seria psicologia. A
crença na sorte aumentaria a confiança, e a confiança melhoraria
o desempenho. Elegante, plausível, publicável.

O que faz de 2014 um caso-modelo é a qualidade do duelo. As
replicações checaram e descartaram todas as saídas de emergência:
a manipulação pegou (os participantes registraram a "sorte" da
bola), o poder estatístico sobrava, a dificuldade da tarefa era
comparável, e os participantes acreditavam em sorte tanto quanto
os alemães de 2010. Não sobrou onde o efeito se esconder.

E há uma pista retrospectiva que o dia faz questão de ensinar:
o intervalo de confiança original era **[0,05, 1,60]**. Ele já
confessava. Com 28 pessoas, o efeito era qualquer coisa entre
quase nada e gigantesco, e o 0,83 era um ponto num intervalo do
tamanho de um campo.

O império caiu inteiro: em 2020, o experimento dos anagramas do
mesmo grupo foi replicado duas vezes, pré-registrado. Também
perto de zero.

A diferença de forma com o Dia #096, declarada: lá, uma
literatura inteira pesada em meta-análise. Aqui, o formato mais
limpo que existe: uma alegação contra a sua réplica direta.

E o fair play que merece registro: a autora original **ajudou** os
replicadores. Isso é a ciência funcionando como deveria.

---

### 🧮 O Modelo

**Fontes:**
- Damisch, L., Stoberock, B. & Mussweiler, T. (2010) - Keep your
  fingers crossed! How superstition improves performance -
  Psychological Science, 21(7), 1014-1020 - Experimento 1:
  **N=28**, 10 tacadas de 100 cm; bola "de sorte" 65% vs
  controle 48% (6,4 vs 4,8 de 10); **d=0,83 [0,05, 1,60]**,
  p=0,042. A origem.
- Calin-Jageman, R. J. & Caldwell, T. L. (2014) - Social
  Psychology, 45(3), 239-245, acesso aberto - **duas replicações
  pré-registradas** (N=124 e N=120), com cooperação da autora
  original: desempenho quase idêntico ao controle nas duas;
  objeções checadas e descartadas; heterogeneidade significativa
  na mini-meta. **Fonte central.**
- Ly, A., Etz, A., Marsman, M. & Wagenmakers, E.-J. (2019) -
  Behavior Research Methods - Bayes factor de replicação:
  **BF = 0,175**, dados da réplica ~5,7x mais prováveis sob o
  nulo. O árbitro formal.
- Dickhäuser, O. et al. (2020) - duas replicações pré-registradas
  do experimento dos anagramas do mesmo grupo: efeitos perto de
  zero. A queda do império.

**Fator ×0,80: não se aplica.** Experimentos randomizados com
desfecho contado (tacadas).

**As contas nossas:** (1) o Bayes factor de replicação traduzido
para português; (2) o preço do amuleto em tacadas: o d de cada
ato convertido de volta para a régua do experimento, usando o
desvio-padrão implícito do original; (3) a pista do intervalo:
o que um IC de amostra 28 já dizia antes da réplica dizer.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---

# Ato 1: Damisch et al. (2010), Exp. 1
n_orig = 28
putts_sorte, putts_controle = 6.4, 4.8      # de 10 (65% vs 48%)
d_orig, ic_orig = 0.83, (0.05, 1.60)
diff_putts = putts_sorte - putts_controle    # 1,6 tacadas

# Ato 2: Calin-Jageman & Caldwell (2014), pré-registradas
n_rep1, n_rep2 = 124, 120                    # (58+66) e (54+66)
d_rep = 0.0    # "desempenho quase idêntico": valor pontual aproximado, declarado
# objeções checadas: manipulação, poder, dificuldade, crença em sorte

# O árbitro: Ly et al. (2019)
bf_rep = 0.175                               # ~5,7x pró-nulo

# A queda do império: Dickhäuser et al. (2020), anagramas: ~zero

# Desvio-padrão implícito do original (para traduzir d em tacadas)
sd_putts = diff_putts / d_orig

print("=" * 65)
print("  DADOS DA LITERATURA - A BOLA DA SORTE E O DUELO")
print("=" * 65)
print(f"\n  ATO 1 (2010, N={n_orig}):")
print(f"  → Bola 'de sorte': {putts_sorte}/10  ·  controle: {putts_controle}/10")
print(f"  → d = {d_orig}  [{ic_orig[0]}, {ic_orig[1]}]  ·  p = 0,042")
print(f"  → O amuleto 'valia' {diff_putts:.1f} tacadas em 10")
print(f"\n  ATO 2 (2014, pré-registrado, N={n_rep1} e N={n_rep2}):")
print(f"  → Desempenho 'quase idêntico' ao controle, duas vezes")
print(f"  → Todas as saídas de emergência checadas e fechadas")
print(f"\n  O ÁRBITRO: BF de replicação = {bf_rep}")
print(f"  → Dados ~{1/bf_rep:.1f}x mais prováveis sob 'não existe'")
print(f"\n  Anagramas do mesmo grupo (2020): perto de zero também")
print(f"\n  Fator ×0,80: NÃO aplicado (experimentos randomizados)")
print("=" * 65)
```

```python
# --- O MODELO ---
# Conta 1: o BF traduzido
# Conta 2: o preço do amuleto em tacadas
# Conta 3: a pista do intervalo

# ═══ CONTA 2 - tacadas ═══
d_ponderado = (n_orig * d_orig + (n_rep1 + n_rep2) * d_rep) / (n_orig + n_rep1 + n_rep2)
tacadas_promessa = d_orig * sd_putts
tacadas_ponderado = d_ponderado * sd_putts
tacadas_replica = d_rep * sd_putts

# ═══ CONTA 3 - a pista ═══
se_orig = (ic_orig[1] - ic_orig[0]) / (2 * 1.96)
largura = ic_orig[1] - ic_orig[0]

print("=" * 65)
print("  MODELO - AS TRÊS CONTAS")
print("=" * 65)
print(f"\n  CONTA 1 - o Bayes factor, em português:")
print(f"  → Se o efeito existisse do tamanho anunciado, a réplica")
print(f"     dificilmente sairia assim. Se não existisse, sairia")
print(f"     exatamente assim. Razão: ~{1/bf_rep:.1f} para 1, pró-nulo.")
print(f"\n  CONTA 2 - o preço do amuleto em tacadas (de 10):")
print(f"  → A promessa de 2010:            +{tacadas_promessa:.1f} tacadas")
print(f"  → Ponderando tudo por N (aprox.): +{tacadas_ponderado:.2f} tacadas")
print(f"  → As réplicas puras:              +{tacadas_replica:.1f} tacadas")
print(f"  → O amuleto prometia 1,6 tacadas e entrega, na leitura")
print(f"     mais generosa, um sexto de tacada. Na honesta, nada.")
print(f"\n  CONTA 3 - a pista que já estava lá:")
print(f"  → IC original: [{ic_orig[0]}, {ic_orig[1]}]: largura de {largura:.2f}")
print(f"  → Com 28 pessoas, o efeito era 'qualquer coisa entre")
print(f"     quase nada e gigantesco'. A borda inferior raspava o zero.")
print(f"  → Amostra pequena + p=0,042 + IC do tamanho de um campo:")
print(f"     a certidão de nascimento já pedia desconfiança.")
print(f"\n  A LEITURA:")
print(f"  → Se o seu amuleto te acalma, esse conforto é seu, e ninguém")
print(f"     o julgou aqui. O que morreu foi a promessa de desempenho.")
print("=" * 65)
```

```python
# --- VISUALIZAÇÃO - GRÁFICOS SEPARADOS ---

# ── GRÁFICO 1 - O duelo em efeitos ──
fig1, ax1 = plt.subplots(figsize=(10, 6))

labels = ['Original 2010\n(N=28)', 'Réplica 1\n(N=124)', 'Réplica 2\n(N=120)']
ds = [d_orig, 0.0, 0.0]
ses = [se_orig, 2/np.sqrt(n_rep1), 2/np.sqrt(n_rep2)]   # SEs aproximados pelo N
xs = [0, 1, 2]
cores = ['#c9a14a', '#c0392b', '#c0392b']

ax1.axhline(0, color='#333', linewidth=1.3)
for x, d, se, cor in zip(xs, ds, ses, cores):
    ax1.plot([x, x], [d - 1.96*se, d + 1.96*se], color=cor, linewidth=2.5)
    ax1.plot([x-0.06, x+0.06], [d - 1.96*se]*2, color=cor, linewidth=2.5)
    ax1.plot([x-0.06, x+0.06], [d + 1.96*se]*2, color=cor, linewidth=2.5)
    ax1.scatter([x], [d], s=150, color=cor, zorder=3)
    ax1.text(x + 0.12, d, f'{d:.2f}', fontsize=13, fontweight='bold', va='center')

ax1.set_xticks(xs)
ax1.set_xticklabels(labels, fontsize=11.5)
ax1.set_ylabel('Tamanho de efeito (d)')
ax1.set_ylim(-0.55, 1.75)
ax1.set_title('O duelo: a alegação contra a sua réplica direta\n'
              'Mesma tarefa, mesma frase, autora original cooperando',
              fontsize=12.5, pad=15)
ax1.text(0.98, 0.03, 'intervalos das réplicas aproximados pelo N (declarado)',
         transform=ax1.transAxes, ha='right', fontsize=9, color='#777', style='italic')

plt.figtext(0.5, 0.01,
            'Fontes: Damisch et al. (2010); Calin-Jageman & Caldwell (2014) | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-098-grafico-01-duelo.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 - O preço do amuleto em tacadas ──
fig2, ax2 = plt.subplots(figsize=(10, 6))

labels2 = ['A promessa\nde 2010', 'Ponderando tudo\npor N (aprox.)', 'As réplicas\npuras']
vals2 = [tacadas_promessa, tacadas_ponderado, tacadas_replica]
cores2 = ['#c9a14a', '#6b6a64', '#c0392b']

bars = ax2.bar(labels2, vals2, color=cores2, alpha=0.88, width=0.45)
ax2.set_ylabel('Tacadas extras por 10 tentativas')
ax2.set_ylim(0, 2.0)
ax2.set_title('O preço do amuleto, em tacadas\n'
              'De 1,6 tacadas prometidas a um sexto de tacada, ou nada',
              fontsize=12.5, pad=15)
for bar, v in zip(bars, vals2):
    ax2.text(bar.get_x() + bar.get_width()/2, v + 0.045,
             f'+{v:.2f}', ha='center', fontweight='bold', fontsize=14)

plt.figtext(0.5, 0.01,
            'Conta própria: d convertido em tacadas pelo desvio-padrão implícito do original | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-098-grafico-02-tacadas.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 - A pista do intervalo ──
fig3, ax3 = plt.subplots(figsize=(10, 6))

x = np.linspace(-0.6, 2.1, 500)
y = stats.norm.pdf(x, d_orig, se_orig)
ax3.plot(x, y, color='#c9a14a', linewidth=3)
ax3.fill_between(x, y, where=(x >= ic_orig[0]) & (x <= ic_orig[1]),
                 alpha=0.18, color='#c9a14a')
ax3.axvline(0, color='#333', linewidth=1.5)
ax3.axvline(ic_orig[0], color='#c0392b', linestyle='--', linewidth=1.8)
ax3.text(ic_orig[0] + 0.03, max(y)*0.9, f'borda inferior: {ic_orig[0]}\nraspando o zero',
         fontsize=10.5, color='#c0392b')
ax3.axvline(ic_orig[1], color='#6b6a64', linestyle='--', linewidth=1.4)
ax3.text(ic_orig[1] - 0.42, max(y)*0.9, f'borda superior: {ic_orig[1]}\n"gigantesco"',
         fontsize=10.5, color='#555')
ax3.set_xlabel('Tamanho de efeito (d)')
ax3.set_ylabel('Densidade')
ax3.set_title('A pista que já estava lá: o intervalo original [0,05, 1,60]\n'
              'Com N=28, o efeito era qualquer coisa entre quase nada e enorme',
              fontsize=12, pad=15)

plt.figtext(0.5, 0.01,
            'Conta própria sobre Damisch et al. (2010) | a largura do intervalo como lição | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-098-grafico-03-pista.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**O amuleto prometia 1,6 tacadas em 10. As réplicas, com o
quádruplo de gente e a autora original ajudando, entregaram:
nada.**

E o dia deixa duas lições que valem mais que o veredito.

A primeira é o duelo perfeito como padrão-ouro: réplica direta,
pré-registrada, com todas as saídas de emergência fechadas, e com
fair play, a autora original cooperando para que o teste fosse
justo. É a ciência funcionando como deveria, e é raro o
suficiente para merecer aplauso.

A segunda é a pista que já estava lá: o intervalo original,
[0,05, 1,60], confessava desde 2010 que o efeito era qualquer
coisa entre quase nada e gigantesco. Amostra de 28, p de 0,042 e
um intervalo do tamanho de um campo: a certidão de nascimento já
pedia desconfiança. Aprender a ler a largura de um intervalo é a
vacina mais barata que existe contra manchete.

E o que sobrevive, sem inventar ciência para isso: se a sua
camisa de final te acalma, esse conforto é seu, e ninguém o
julgou aqui. O que morreu foi a promessa de desempenho: a de que
a crença faria a bola entrar.

A pergunta que fica:

*Qual crença sua sobreviveria a uma segunda rodada,
com o dobro de testemunhas?*

### ⚠️ Limitações do Modelo

- As réplicas cobrem a tarefa do golfe (e, em 2020, os
  anagramas); outros contextos de superstição não foram testados
  neste duelo
- A mini-meta dos replicadores reporta heterogeneidade
  significativa: o efeito pode variar entre contextos, sem que
  isso resgate o achado original
- Valores pontuais das réplicas aproximados de "quase idêntico"
  (efeitos perto de zero); intervalos do gráfico 1 aproximados
  pelo N, declarado
- A ponderação por N da conta 2 é aproximação didática, não
  meta-análise formal
- O Bayes factor de replicação usa a posterior do original como
  prior; outras especificações mudam o valor, não a direção
- Conforto, ansiedade e ritual como regulação emocional são
  literatura própria, não julgada aqui
- Forma declarada: 2º dia de disputa do bloco, par com o #096

*A ciência é honesta sobre o que não sabe. O modelo também.*
