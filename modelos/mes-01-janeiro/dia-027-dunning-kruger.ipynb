### 📖 A História

Nesses primeiros dias construindo o 365 Probabilidades, descobri uma coisa incômoda.

Um monte de estatística que eu tomava como verdade simplesmente não tem fonte real.

Números redondos, bonitos, fáceis de repetir. Aquele "85% disso", "você usa só 10% daquilo", "leva 21 dias pra criar um hábito". Eu já tinha repetido vários deles sem pensar duas vezes.

Aí você vai atrás da fonte primária. E ela não existe. Ou existe, mas diz outra coisa. Ou é um estudo de amostra minúscula que virou manchete e nunca foi replicado.

O mais curioso não é que os números sejam falsos. É a confiança com que são vendidos. Tem gente ganhando seguidor e dinheiro repetindo estatística inventada com a segurança de quem tem doutorado no assunto.

E a ciência tem uma explicação exata pra isso. Inclusive pra por que a pessoa que mais espalha o número errado costuma ser a que menos desconfia dele.

---

### 📚 O Conceito: Incompetência Invisível

Em 1999, dois psicólogos, Justin Kruger e David Dunning, publicaram um estudo que ficou famoso como "efeito Dunning-Kruger".

A ideia central é quase cruel na sua simetria: as mesmas habilidades que você precisa para fazer algo bem são as habilidades que você precisa para perceber que está fazendo mal.

Quem não sabe gramática não tem como perceber os próprios erros de gramática — porque perceber o erro exige justamente o conhecimento que falta. Quem não entende de lógica não consegue avaliar se o próprio raciocínio é lógico. A incompetência se esconde de quem a carrega.

O resultado é que as pessoas menos capazes numa tarefa não só erram mais — elas também são as mais confiantes de que acertaram. Falta a elas a ferramenta mental para enxergar o próprio erro.

É por isso que o guru que espalha a estatística inventada raramente hesita. Não é (só) má-fé. É que reconhecer que o número é frágil exigiria exatamente o rigor que a pessoa não tem — e, por não ter, nem sabe que está faltando.

### 🧮 O Modelo

Aqui os dados são percentis de auto-avaliação de um experimento controlado, não proporções de survey populacional. Por isso o fator de correção ×0.80 não se aplica neste dia — não há viés de amostragem populacional a corrigir; os números são os do próprio experimento.

**Fontes:**
- Kruger, J. & Dunning, D. (1999) — "Unskilled and Unaware of It: How Difficulties in Recognizing One's Own Incompetence Lead to Inflated Self-Assessments", Journal of Personality and Social Psychology, 77(6), 1121-1134 — 4 estudos, tarefas de raciocínio lógico, gramática e humor
- Krueger, J. & Mueller, R. A. (2002) — crítica metodológica (regressão à média + efeito melhor-que-a-média), JPSP 82(2), 180-188
- Nuhfer et al. (2016) — reanálise apontando artefato estatístico no formato clássico do gráfico


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Kruger & Dunning (1999), JPSP 77(6):1121-1134
# Estudo de raciocínio lógico — percentis por quartil de desempenho

# Quartil INFERIOR (piores desempenhos)
perc_real_inferior = 12       # percentil real médio
perc_autoaval_inferior = 62   # percentil que eles ACHAVAM que tinham

# Quartil SUPERIOR (melhores desempenhos)
perc_real_superior = 86       # percentil real médio
perc_autoaval_superior = 74   # percentil que eles achavam (subestimaram)

# Tamanho dos erros
erro_inferior = perc_autoaval_inferior - perc_real_inferior   # superestimação
erro_superior = perc_real_superior - perc_autoaval_superior   # subestimação

print("=" * 65)
print("  DADOS DA LITERATURA — DUNNING-KRUGER (1999)")
print("=" * 65)
print(f"\n  QUARTIL INFERIOR (piores):")
print(f"  → Desempenho real:        {perc_real_inferior}º percentil")
print(f"  → Auto-avaliação:         {perc_autoaval_inferior}º percentil")
print(f"  → Superestimação:         +{erro_inferior} pontos percentuais")
print(f"\n  QUARTIL SUPERIOR (melhores):")
print(f"  → Desempenho real:        {perc_real_superior}º percentil")
print(f"  → Auto-avaliação:         {perc_autoaval_superior}º percentil")
print(f"  → Subestimação:           -{erro_superior} pontos percentuais")
print(f"\n  A tesoura: os piores erram muito pra cima,")
print(f"  os melhores erram pouco pra baixo.")
print("=" * 65)
```

```python
# --- O MODELO ---
# A assimetria da confiança: quem sabe menos erra mais... e para cima

razao_assimetria = erro_inferior / erro_superior

# "Distância até a realidade" — o quanto a auto-imagem do pior desempenho
# está descolada do desempenho real
gap_pior = perc_autoaval_inferior - perc_real_inferior

print("=" * 65)
print("  MODELO — A CONFIANÇA DE QUEM MAIS ERRA")
print("=" * 65)
print(f"\n  Quem estava no {perc_real_inferior}º percentil real")
print(f"  se enxergava no {perc_autoaval_inferior}º percentil.")
print(f"\n  → Descolamento da realidade:   {gap_pior} pontos percentuais")
print(f"\n  Assimetria do erro:")
print(f"  → Piores superestimam:   +{erro_inferior}pp")
print(f"  → Melhores subestimam:   -{erro_superior}pp")
print(f"  → Os piores erram {razao_assimetria:.1f}x mais (e na direção da confiança)")
print(f"\n  LEITURA:")
print(f"  → A confiança alta num assunto que não se domina")
print(f"     não é prova de acerto — é justamente o sintoma previsto.")
print(f"  → A habilidade que falta pra acertar é a mesma que")
print(f"     falta pra perceber que se está errando.")
print("=" * 65)
```

```python
# --- VISUALIZAÇÃO ---

cor_real = '#2a8a82'
cor_aval = '#c0392b'
cor_ouro = '#c8a84b'

# ── GRÁFICO 1 — A tesoura: real vs auto-avaliação por quartil ──
fig1, ax1 = plt.subplots(figsize=(10, 6))

quartis = ['Inferior\n(piores)', '2º quartil', '3º quartil', 'Superior\n(melhores)']
# Valores reais aproximados por quartil (estudo de raciocínio lógico)
real = [12, 37, 62, 86]
# Auto-avaliação: notavelmente plana — todos se acham por volta de 60-75
autoaval = [62, 65, 70, 74]

x = np.arange(len(quartis))
ax1.plot(x, real, 'o-', color=cor_real, linewidth=2.5, markersize=9, label='Desempenho real')
ax1.plot(x, autoaval, 's--', color=cor_aval, linewidth=2.5, markersize=9, label='Auto-avaliação')

ax1.fill_between(x, real, autoaval, where=[a > r for a, r in zip(autoaval, real)],
                 alpha=0.12, color=cor_aval)

ax1.set_xticks(x)
ax1.set_xticklabels(quartis)
ax1.set_ylabel('Percentil')
ax1.set_ylim(0, 100)
ax1.set_title('A tesoura de Dunning-Kruger\nQuem sabe menos se avalia quase igual a quem sabe mais',
              fontsize=13, pad=15)
ax1.legend(loc='lower right', frameon=False)

ax1.annotate('gap de 50pp', xy=(0, 37), xytext=(0.4, 40),
             fontsize=10, color=cor_aval, fontweight='bold',
             arrowprops=dict(arrowstyle='-', color=cor_aval, lw=1))

plt.figtext(0.5, 0.01,
            'Fonte: Kruger & Dunning (1999), JPSP 77(6) — estudo de raciocínio lógico | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-027-grafico-01-tesoura.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 — O tamanho e a direção do erro ──
fig2, ax2 = plt.subplots(figsize=(10, 6))

grupos = ['Quartil inferior\n(piores)', 'Quartil superior\n(melhores)']
erros = [erro_inferior, -erro_superior]  # positivo = pra cima, negativo = pra baixo
cores = [cor_aval, cor_real]

bars = ax2.bar(grupos, erros, color=cores, alpha=0.85, width=0.45)
ax2.axhline(y=0, color='#333', linewidth=1)
ax2.set_ylabel('Erro de auto-avaliação (pontos percentuais)')
ax2.set_ylim(-25, 60)
ax2.set_title('A assimetria da confiança\nOs piores superestimam muito; os melhores subestimam pouco',
              fontsize=13, pad=15)

for bar, e in zip(bars, erros):
    offset = 2 if e > 0 else -4
    ax2.text(bar.get_x() + bar.get_width()/2, e + offset,
             f'{"+" if e > 0 else ""}{e}pp', ha='center',
             fontweight='bold', fontsize=15)

ax2.text(0, 54, 'superestima', ha='center', fontsize=10, color=cor_aval, style='italic')
ax2.text(1, -20, 'subestima', ha='center', fontsize=10, color=cor_real, style='italic')

plt.figtext(0.5, 0.01,
            'Fonte: Kruger & Dunning (1999), JPSP 77(6) | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-027-grafico-02-assimetria.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 — Distribuição Beta ilustrativa ──
# Onde a auto-avaliação do quartil inferior "acredita" estar (62º percentil)
# vs. onde de fato está (12º), como distribuições ilustrativas.
fig3, ax3 = plt.subplots(figsize=(10, 6))

x = np.linspace(0, 1, 1000)

# "Crença" centrada em ~0.62; "realidade" centrada em ~0.12
dist_crenca = stats.beta(6.2, 3.8)
dist_real = stats.beta(1.8, 12.0)

ax3.plot(x*100, dist_crenca.pdf(x), color=cor_aval, linewidth=2.5, label='Onde acham que estão (~62º)')
ax3.fill_between(x*100, dist_crenca.pdf(x), alpha=0.18, color=cor_aval)
ax3.plot(x*100, dist_real.pdf(x), color=cor_real, linewidth=2.5, label='Onde realmente estão (~12º)')
ax3.fill_between(x*100, dist_real.pdf(x), alpha=0.18, color=cor_real)

ax3.axvline(x=62, color=cor_aval, linestyle='--', alpha=0.7)
ax3.axvline(x=12, color=cor_real, linestyle='--', alpha=0.7)

ax3.set_xlabel('Percentil de desempenho')
ax3.set_ylabel('Densidade (ilustrativa)')
ax3.set_title('A distância entre a auto-imagem e a realidade\nQuartil inferior — distribuições ilustrativas',
              fontsize=12, pad=15)
ax3.legend(loc='upper center', frameon=False)

plt.figtext(0.5, 0.01,
            'Distribuições ilustrativas centradas nos percentis de Kruger & Dunning (1999) | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-027-grafico-03-bayesiano.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**Quem ficou entre os piores (12º percentil real) se avaliou no 62º percentil.**

Ou seja: as pessoas de pior desempenho não se achavam medianas. Elas se achavam claramente acima da média. Erraram a própria nota por cerca de 50 pontos percentuais — pra cima.

E o outro lado da tesoura é igualmente revelador: quem estava de fato no topo (86º percentil) se subestimou, chutando algo em torno do 74º. Quem sabe, desconfia. Quem não sabe, não tem como desconfiar.

É esse o mecanismo por trás de tanta coisa que circula como verdade na internet. Não é que as pessoas mintam de propósito. É que a incompetência num assunto rouba justamente a capacidade de perceber a própria incompetência naquele assunto. O número falso é repassado com a maior convicção — e a convicção é sintoma, não prova.

A defesa não é ficar inseguro sobre tudo. É outra: quanto mais certeza imediata você sente sobre um assunto que nunca estudou a fundo, mais vale a pena checar a fonte antes de repassar.

*Qual foi a última vez que você repassou um número, uma "verdade" ou um conselho — com toda a confiança do mundo — sem nunca ter visto a fonte original?*

### ⚠️ Limitações do Modelo

- O efeito Dunning-Kruger clássico sofreu críticas metodológicas relevantes. Krueger & Mueller (2002) e Nuhfer et al. (2016) argumentam que boa parte do padrão do gráfico decorre de regressão à média combinada com o efeito "melhor que a média" — ou seja, parte do achado é artefato estatístico, não só metacognição.
- A famosa curva do "Monte Stupid" (confiança que dispara no nível mais baixo de competência) **não existe no paper original** — é uma criação popular posterior.
- O estudo original usou estudantes de Cornell em tarefas cognitivas abstratas; generalizar para "todo mundo, em tudo" vai além do que os dados mostram.
- Os percentis usados aqui são os do estudo de raciocínio lógico; humor e gramática têm valores um pouco diferentes.

*A ciência é honesta sobre o que não sabe. O modelo também.*
