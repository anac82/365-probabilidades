### 📖 A História

Esse tema começou ontem, numa conversa com uma amiga que fazia aniversário.

A pauta era maratona, os ciclos de treino, e de repente estávamos falando de felicidade. De quanto colocamos nas coisas que ainda não aconteceram. De tudo o que é feito no pré, e de como, muitas vezes, o que chega não justifica a energia que a gente gastou esperando.

A promoção que demorou tanto que, quando veio, já parecia devida. A viagem planejada por um ano que, no terceiro dia, era só uma viagem. A mensagem esperada que chegou e não fez o que devia fazer.

E todo mundo já ganhou o que não esperava e passou o dia inteiro leve por causa de uma coisa pequena.

O prêmio não mudou de tamanho. O que mudou foi o que a gente esperava dele.

---

### 📚 O Conceito: a felicidade é a distância

Em 2014, um grupo de Londres resolveu medir a felicidade do momento com a mesma ferramenta que a neurociência usa para medir aprendizado: o erro de previsão. A diferença entre o que você esperava e o que veio.

Montaram um jogo de apostas. Em cada rodada a pessoa escolhe entre um valor garantido e uma aposta. Vê o resultado. E, a cada poucas rodadas, responde: quão feliz você está agora?

Primeiro com 26 pessoas dentro de um scanner. Depois com 18.420 pessoas num aplicativo de celular, no mundo inteiro.

O resultado: quanto a pessoa tinha ganhado no total não previa a felicidade. O que previa era a soma de três coisas, cada uma pesando menos conforme envelhece: o que ela ganhou na certa, o que ela esperava ganhar, e a diferença entre o que veio e o que esperava.

O mesmo dinheiro na conta. Felicidades diferentes. A expectativa decide.

**Fronteira com o #080 (esteira hedônica):** lá o assunto era o nível da felicidade meses e anos depois de um evento grande, e o quanto ele volta para onde estava. Aqui é a felicidade do momento, e a distância entre o que veio e o que se esperava. Escalas de tempo diferentes, mecanismos diferentes (adaptação lá, expectativa aqui). Os dois podem ser verdade ao mesmo tempo.

### 🧮 O Modelo

A equação publicada, literal, e uma simulação com ela: duas pessoas que jogam o mesmo jogo, terminam com o mesmo dinheiro, e são medidas pela mesma régua de felicidade. Uma esperava pouco. A outra esperava muito.

**Fontes:**
- Rutledge, R. B., Skandali, N., Dayan, P. & Dolan, R. J. (2014). *A computational and neural model of momentary subjective well-being.* PNAS, 111(33), 12252-12257. **26 participantes** no laboratório (com fMRI) e **18.420** no aplicativo *The Great Brain Experiment*. A riqueza acumulada no jogo não prevê a felicidade do momento; a combinação de expectativas recentes e erros de previsão prevê, e a atividade do estriado ventral acompanha a mesma conta.
- Blain, B. & Rutledge, R. B. (2020). *Momentary subjective well-being depends on learning and not reward.* eLife, 9:e57977. **N=75.** A felicidade depois de ganhar não depende do tamanho do prêmio, só da surpresa de ter ganhado; a equação explica **58% a 62%** da variação do humor; fator de esquecimento **γ = 0,59 a 0,63**.

**A equação (Rutledge 2014):**

$$\text{Felicidade}(t) = w_0 + w_1 \sum_{j=1}^{t} \gamma^{t-j}\, CR_j + w_2 \sum_{j=1}^{t} \gamma^{t-j}\, EV_j + w_3 \sum_{j=1}^{t} \gamma^{t-j}\, RPE_j$$

onde CR é o valor garantido quando a pessoa não aposta, EV é o valor esperado da aposta quando aposta, RPE é o erro de previsão (resultado menos EV), e γ é o fator de esquecimento: cada rodada pesa γ vezes menos que a seguinte.

**Nota metodológica:** ×0,80 não se aplica. Escala de humor contra resultado objetivo do jogo, dentro da mesma pessoa, com N de 18.420.

**Pendência declarada:** os pesos e o r² exatos do artigo de 2014 não foram lidos no texto completo (acesso bloqueado na validação); a simulação usa γ = 0,6 do eLife 2020 e pesos ilustrativos declarados. A simulação demonstra o mecanismo, não estima os parâmetros (alerta 25 do livro-caixa).


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

DOURADO = '#c9a14a'
VERMELHO = '#c0392b'
VERDE = '#2a8a82'
CINZA = '#6b6a64'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---

# RUTLEDGE, SKANDALI, DAYAN & DOLAN 2014 - PNAS 111(33)
n_lab = 26            # laboratório, com fMRI
n_app = 18420         # aplicativo The Great Brain Experiment
riqueza_preve_felicidade = False   # "overall wealth accumulated was not a good predictor of happiness"
# termos da equação: CR (valor garantido), EV (valor esperado da aposta), RPE (resultado - EV), γ (esquecimento)

# BLAIN & RUTLEDGE 2020 - eLife 9:e57977 (mesmo laboratório, paradigma vizinho)
n_blain = 75
r2_blain = (0.58, 0.62)    # variação do humor explicada pela equação (estável, volátil)
gamma_blain = (0.59, 0.63)  # fator de esquecimento

# PARÂMETROS DA SIMULAÇÃO (premissa declarada, não estimativa)
gamma = 0.60
w_cr, w_ev, w_rpe = 1.0, 1.0, 1.0   # pesos iguais: a simulação testa a forma, não a magnitude relativa

fator_correcao = 0.80
aplica_fator = False

print("=" * 65)
print("  DADOS DA LITERATURA - A FELICIDADE É A DISTÂNCIA")
print("=" * 65)
print(f"\n  RUTLEDGE 2014 (PNAS): {n_lab} no laboratório + {n_app:,} no aplicativo".replace(',', '.'))
print(f"  → A riqueza acumulada prevê a felicidade do momento? {'sim' if riqueza_preve_felicidade else 'NÃO'}")
print(f"  → O que prevê: expectativa recente + erro de previsão (resultado menos expectativa)")
print(f"\n  BLAIN & RUTLEDGE 2020 (eLife): N={n_blain}")
print(f"  → A equação explica {r2_blain[0]*100:.0f}% a {r2_blain[1]*100:.0f}% da variação do humor")
print(f"  → Fator de esquecimento γ = {gamma_blain[0]} a {gamma_blain[1]}")
print(f"\n  SIMULAÇÃO: γ = {gamma}, pesos {w_cr}/{w_ev}/{w_rpe} (premissa declarada)")
print(f"  ×0,80: não se aplica")
print("=" * 65)
```

```python
# --- O MODELO ---
# (a) A SIMULAÇÃO: duas pessoas, o mesmo jogo, o mesmo dinheiro no fim, expectativas diferentes
# (b) OS DUELOS: cada uma no seu jogo; a pessimista termina mais feliz? e mais feliz ganhando menos?
# (c) A ANATOMIA: quanto da felicidade vem de cada termo da equação

rng = np.random.default_rng(42)

def felicidade(cr, ev, rpe, gamma, w=(1.0, 1.0, 1.0)):
    # soma com esquecimento: cada rodada pesa gamma vezes menos que a seguinte
    t = len(cr)
    pesos = gamma ** np.arange(t)[::-1]   # gamma^(t-j)
    h = np.zeros(t)
    for i in range(t):
        p = gamma ** np.arange(i, -1, -1)
        h[i] = w[0]*np.sum(p*cr[:i+1]) + w[1]*np.sum(p*ev[:i+1]) + w[2]*np.sum(p*rpe[:i+1])
    return h

# (a) O MESMO JOGO
# 30 rodadas. Em todas a pessoa aposta (para isolar expectativa e surpresa). O resultado de cada
# aposta é o MESMO para as duas: só muda o que cada uma esperava.
n_rodadas = 30
resultados = rng.choice([0, 10], size=n_rodadas)      # a aposta paga 10 ou 0
# Pessoa A espera pouco: acha que a aposta vale 3 (30% de chance de ganhar 10)
# Pessoa B espera muito: acha que a aposta vale 7 (70% de chance)
ev_A = np.full(n_rodadas, 3.0)
ev_B = np.full(n_rodadas, 7.0)
cr = np.zeros(n_rodadas)                                # ninguém pegou o valor garantido
rpe_A = resultados - ev_A
rpe_B = resultados - ev_B
h_A = felicidade(cr, ev_A, rpe_A, gamma)
h_B = felicidade(cr, ev_B, rpe_B, gamma)
ganho_total = resultados.sum()
dif_media = np.mean(h_A - h_B)
# Nota: com pesos iguais em EV e RPE, EV + RPE = resultado, e as duas curvas coincidem.
# É por isso que a assimetria só aparece quando os pesos são diferentes: no artigo, w_RPE > w_EV.
# Rodamos a versão com o peso da surpresa maior que o da expectativa (premissa declarada).
w_assim = (1.0, 0.5, 1.0)   # expectativa pesa metade da surpresa
h_A2 = felicidade(cr, ev_A, rpe_A, gamma, w_assim)
h_B2 = felicidade(cr, ev_B, rpe_B, gamma, w_assim)
dif_media2 = np.mean(h_A2 - h_B2)

# (b) OS DUELOS
# 100 mil pares. Cada par joga o MESMO jogo (mesmos resultados). Expectativa de cada um sorteada
# entre 2 e 8. Quem esperava menos termina mais feliz? (felicidade média das últimas 10 rodadas)
N_PARES = 100_000
vence_menos = 0
gap = np.zeros(N_PARES)
for k in range(N_PARES):
    res = rng.choice([0, 10], size=n_rodadas)
    e1, e2 = rng.uniform(2, 8, size=2)
    h1 = felicidade(cr, np.full(n_rodadas, e1), res - e1, gamma, w_assim)[-10:].mean()
    h2 = felicidade(cr, np.full(n_rodadas, e2), res - e2, gamma, w_assim)[-10:].mean()
    quem_esperava_menos = h1 if e1 < e2 else h2
    quem_esperava_mais  = h2 if e1 < e2 else h1
    gap[k] = quem_esperava_menos - quem_esperava_mais
    vence_menos += quem_esperava_menos > quem_esperava_mais
p_vence_menos = vence_menos / N_PARES

# (b) OS DUELOS
# Cada pessoa joga o SEU jogo (resultados independentes, 50% de chance de ganhar 10 em cada aposta).
# A pessoa pessimista espera 3 por aposta; a otimista espera 7. Felicidade = média das últimas 10 rodadas.
# Duelo 1: em quantos pares a pessimista termina mais feliz?
# Duelo 2 (o que importa): em quantos pares a pessimista termina mais feliz TENDO GANHADO MENOS dinheiro?
N_PARES = 100_000
def joga(exp, n=30):
    res = rng.choice([0, 10], size=(N_PARES, n))
    ev = np.full(n, exp)
    # felicidade vetorizada: soma com esquecimento sobre as últimas rodadas
    pesos = gamma ** np.arange(n)[::-1]
    h = np.zeros(N_PARES)
    for i in range(n - 10, n):
        p = gamma ** np.arange(i, -1, -1)
        h += w_assim[1]*np.sum(p*ev[:i+1]) + w_assim[2]*(res[:, :i+1] - ev[:i+1]) @ p
    return h / 10, res.sum(axis=1)
h_pess, ganho_pess = joga(3.0)
h_otim, ganho_otim = joga(7.0)
p_pess_mais_feliz = np.mean(h_pess > h_otim)
ganhou_menos = ganho_pess < ganho_otim
p_pess_feliz_e_pobre = np.mean((h_pess > h_otim) & ganhou_menos)
p_pess_feliz_dado_pobre = np.mean(h_pess[ganhou_menos] > h_otim[ganhou_menos])
gap = h_pess - h_otim

# (c) A ANATOMIA: numa rodada típica, quanto pesa cada termo (Monte Carlo)
N_RODADAS_MC = 200_000
ev_mc = rng.uniform(2, 8, N_RODADAS_MC)
res_mc = rng.choice([0, 10], N_RODADAS_MC)
rpe_mc = res_mc - ev_mc
var_ev = np.var(w_assim[1] * ev_mc)
var_rpe = np.var(w_assim[2] * rpe_mc)
share_rpe = var_rpe / (var_ev + var_rpe)

print("=" * 65)
print("  MODELO - O MESMO DINHEIRO, FELICIDADES DIFERENTES")
print("=" * 65)
print(f"\n  (a) A SIMULAÇÃO: {n_rodadas} rodadas, resultados idênticos, ganho final R$ {ganho_total} para as duas")
print(f"  → Pessoa A esperava 3 por aposta; pessoa B esperava 7")
print(f"  → Com expectativa e surpresa pesando igual: diferença média de felicidade = {dif_media:+.2f} (as curvas coincidem)")
print(f"  → Com a surpresa pesando o dobro da expectativa: A mais feliz que B por {dif_media2:+.1f} pontos em média")
print(f"  → Mesmo jogo. Mesmo dinheiro. A régua muda o resultado.")
print(f"\n  (b) OS DUELOS: {N_PARES:,} pares, cada um no seu jogo (50% de chance por aposta)".replace(',', '.').replace('pares.', 'pares,'))
print(f"  → Pessimista (espera 3) contra otimista (espera 7)")
print(f"  → A pessimista termina mais feliz em {p_pess_mais_feliz*100:.0f} de 100 pares")
print(f"  → Em {ganhou_menos.mean()*100:.0f} de 100 pares a pessimista ganhou MENOS dinheiro; nesses, ela termina mais feliz em {p_pess_feliz_dado_pobre*100:.0f} de 100")
print(f"  → Mais feliz E mais pobre ao mesmo tempo: {p_pess_feliz_e_pobre*100:.0f} de 100 pares")
print(f"  → Vantagem média: {np.mean(gap):+.1f} pontos; IC 95% [{np.percentile(gap, 2.5):+.1f}, {np.percentile(gap, 97.5):+.1f}]")
print(f"\n  (c) A ANATOMIA: de onde vem a variação da felicidade numa rodada")
print(f"  → Surpresa (RPE): {share_rpe*100:.0f}%   Expectativa (EV): {(1-share_rpe)*100:.0f}%")
print(f"\n  RESPOSTA DO DIA")
print(f"  → Quem esperava menos termina mais feliz em {p_pess_mais_feliz*100:.0f} de 100 duelos,")
print(f"     e em {p_pess_feliz_dado_pobre*100:.0f} de 100 mesmo quando ganhou menos dinheiro.")
print(f"  → Não é o que aconteceu. É a distância entre o que aconteceu e o que você esperava.")
print("=" * 65)
```

```python
# --- VISUALIZAÇÃO - GRÁFICOS SEPARADOS ---

# ── GRÁFICO 1 - O mesmo jogo, duas felicidades ──
fig1, ax1 = plt.subplots(figsize=(11, 6.5))
rod = np.arange(1, n_rodadas + 1)
ax1.plot(rod, h_A2, color=VERDE, lw=3, marker='o', ms=5, label='Esperava pouco (3 por aposta)')
ax1.plot(rod, h_B2, color=VERMELHO, lw=3, marker='o', ms=5, label='Esperava muito (7 por aposta)')
ax1.fill_between(rod, h_A2, h_B2, where=h_A2 >= h_B2, color=VERDE, alpha=0.10)
for i, r in enumerate(resultados):
    ax1.text(rod[i], min(h_A2.min(), h_B2.min()) - 1.2, '●' if r == 10 else '○', ha='center', fontsize=8, color=CINZA)
ax1.text(rod[0], min(h_A2.min(), h_B2.min()) - 2.6, '● ganhou 10   ○ ganhou 0   (os mesmos resultados para as duas)', fontsize=9, color=CINZA)
ax1.set_xlabel('Rodada')
ax1.set_ylabel('Felicidade do momento (equação de Rutledge, unidades arbitrárias)')
ax1.set_title(f'O mesmo jogo, o mesmo dinheiro (R$ {ganho_total}), duas felicidades\n'
              f'Diferença média de {dif_media2:+.1f} pontos a favor de quem esperava menos', fontsize=13, pad=15)
ax1.legend(frameon=False, loc='upper left')
plt.figtext(0.5, 0.01, 'Simulação com a equação de Rutledge et al. 2014 · γ=0,6 · surpresa pesa o dobro da expectativa (premissa) · semente 42 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-106-grafico-01-mesmo-jogo.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 - A equação (assinatura) ──
fig2, ax2 = plt.subplots(figsize=(11, 6.5))
ax2.axis('off')
ax2.text(0.5, 0.80, r'$\mathrm{Felicidade}(t) = w_0 + w_1\sum_{j\leq t}\gamma^{\,t-j}\,CR_j + w_2\sum_{j\leq t}\gamma^{\,t-j}\,EV_j + w_3\sum_{j\leq t}\gamma^{\,t-j}\,RPE_j$',
         ha='center', fontsize=17, transform=ax2.transAxes)
linhas = [
    ('CR', 'o que você ganhou na certa (não apostou)', CINZA),
    ('EV', 'o que você esperava ganhar (apostou)', VERMELHO),
    ('RPE', 'a surpresa: o que veio menos o que esperava', VERDE),
    ('γ', f'o esquecimento: cada rodada pesa {gamma:.0%} da seguinte', DOURADO),
]
for i, (sym, txt, cor) in enumerate(linhas):
    ax2.text(0.18, 0.58 - i*0.11, sym, fontsize=16, fontweight='bold', color=cor, transform=ax2.transAxes)
    ax2.text(0.27, 0.58 - i*0.11, txt, fontsize=13, transform=ax2.transAxes)
ax2.text(0.5, 0.08, 'O dinheiro acumulado não está na equação. Não precisou estar: não previa nada.',
         ha='center', fontsize=12, style='italic', color=CINZA, transform=ax2.transAxes)
ax2.set_title('A assinatura matemática: a equação da felicidade do momento\nRutledge, Skandali, Dayan & Dolan 2014, PNAS · 26 no laboratório, 18.420 no aplicativo', fontsize=13, pad=15)
plt.figtext(0.5, 0.01, 'Blain & Rutledge 2020: a equação explica 58% a 62% da variação do humor, γ ≈ 0,6 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-106-grafico-02-equacao.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 - Os duelos ──
fig3, ax3 = plt.subplots(figsize=(11, 6.5))
ax3.hist(gap, bins=80, color=VERDE, alpha=0.75)
ax3.axvline(0, color='#333', lw=1.5)
ax3.axvline(np.mean(gap), color=DOURADO, lw=2.5)
ax3.text(np.mean(gap) + 0.15, ax3.get_ylim()[1]*0.9, f'média {np.mean(gap):+.1f}', color=DOURADO, fontsize=12, fontweight='bold')
ax3.set_xlabel('Felicidade da pessimista (espera 3) menos a da otimista (espera 7), cada uma no seu jogo')
ax3.set_ylabel('Pares')
ax3.set_title(f'100 mil duelos: a pessimista termina mais feliz em {p_pess_mais_feliz*100:.0f} de 100\n'
              f'e em {p_pess_feliz_dado_pobre*100:.0f} de 100 mesmo quando ganhou menos dinheiro', fontsize=13, pad=15)
plt.figtext(0.5, 0.01, 'Jogos independentes, 50% de chance por aposta · felicidade média das últimas 10 rodadas · semente 42 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-106-grafico-03-duelos.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**Quem esperava menos termina mais feliz em 84 de 100 duelos. E em 69 de 100 mesmo quando ganhou menos dinheiro. A felicidade não é o que aconteceu. É a distância entre o que aconteceu e o que você esperava.**

Vinte e seis pessoas num scanner e dezoito mil num celular deram a mesma resposta: o total ganho não prevê a felicidade do momento. O que prevê é a expectativa recente e a surpresa, e as duas envelhecem rápido: cada rodada pesa 60% da seguinte.

Isso explica a promoção que não fez nada e a coisa pequena que fez o dia. Não é ingratidão. É a equação. E não é a esteira hedônica do #080: aquilo é sobre voltar ao nível; isto é sobre a distância, no momento.

E explica por que baixar a expectativa não é conselho de derrotado: é o único termo da conta que está na sua mão antes do resultado chegar.

*Da última vez que você ficou feliz de verdade, o que você estava esperando?*

### ⚠️ Limitações do Modelo

- A simulação usa a forma da equação publicada com pesos ilustrativos (surpresa pesando o dobro da expectativa) e γ = 0,6 do eLife 2020. Os pesos e o r² exatos do artigo de 2014 não foram lidos no texto completo. É demonstração de mecanismo, não estimativa.
- Com pesos iguais para expectativa e surpresa, as duas curvas coincidem por construção (expectativa mais surpresa é o resultado). A assimetria do dia depende de a surpresa pesar mais, que é o que os autores reportam; a magnitude é premissa.
- Nos duelos, a chance real de ganhar é 50% para as duas, e a pessimista (espera 3) está mais perto da verdade que a otimista (espera 7). O dia mostra o custo de esperar demais, não a virtude de esperar de menos: com chance real de 50%, quem espera 5 está calibrado.
- Na vida real a expectativa muda o que a pessoa escolhe, e o jogo muda junto; aqui as duas apostam sempre.
- O jogo mede dinheiro pequeno em rodadas de segundos. Felicidade de vida não está no estudo; humor do momento está.
- O aplicativo é amostra autosselecionada (quem baixa um jogo de neurociência).
- Não confundir com a esteira hedônica do #080: aquele dia mede nível ao longo de anos (adaptação); este mede o momento (expectativa e surpresa). Corda declarada no texto.

*A ciência é honesta sobre o que não sabe. O modelo também.*
