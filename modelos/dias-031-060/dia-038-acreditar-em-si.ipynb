## Qual a probabilidade de acreditar que vai dar certo mudar o resultado?

**Tipo:** Preditivo / Comportamental

**Ferramenta:** Python
**Decisão analisada:** Vale a pena confiar que algo vai dar certo antes de ter provas?
**Hashtag:** #365Probabilidades #Dia038

---

### 📖 A História

Tem uma frase que todo mundo repete e ninguém prova.

"Acredite em você e o resultado vem."

Eu também já acreditei nisso sem checar. Na reinvenção de carreira,
depois de sair de um lugar onde eu já sabia todas as respostas,
a confiança não veio antes. Veio depois de alguns acertos pequenos.

E isso muda a pergunta inteira.

Será que a crença veio primeiro e trouxe o resultado?
Ou o resultado veio primeiro e trouxe a crença?

A ciência testou essa pergunta com um método específico
para separar causa de consequência ao longo do tempo.
E a resposta é mais interessante do que a frase de efeito.

---

### 📚 O Conceito: Autoeficácia e o Problema do Ovo e da Galinha

Autoeficácia é a crença na própria capacidade de executar
uma ação e alcançar um resultado (Bandura, 1977).

O problema é que a maioria dos estudos mede autoeficácia
e desempenho no mesmo momento, o que não diz se a crença
causa o resultado, se o resultado causa a crença, ou os dois.

Um desenho estatístico chamado painel cross-lagged resolve isso:
mede as duas variáveis em dois momentos no tempo e testa qual
caminho (crença para resultado futuro, ou resultado para crença
futura) é mais forte.

---

### 🧮 O Modelo

**Fontes:**
- Multon, Brown & Lent, 1991 — Journal of Counseling Psychology
  Meta-análise · 38 amostras · N=4.998 · r=0.38 (autoeficácia-desempenho acadêmico)
  Moderador por idade: r=0.35 (universitários) vs r=0.21 (crianças, k=2 amostras)
- Talsma, Schüz, Schwarzer & Norris, 2018 — Learning and Individual Differences
  Meta-análise cross-lagged · k=11 estudos (1984-2016)
  Teste de diferença dos caminhos recíprocos: χ²=37.47, p<0.0001
- Moritz, Feltz, Fahrbach & Mack, 2000 — Research Quarterly for Exercise and Sport
  Meta-análise · 45 estudos, 102 correlações · r=0.38 (autoeficácia-desempenho esportivo)


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---

# Multon, Brown & Lent, 1991 — Journal of Counseling Psychology
# Meta-análise · 38 amostras · N=4.998 (correlação geral autoeficácia-desempenho acadêmico)
r_geral_academico = 0.38
n_multon = 4998
k_multon = 38

# Moderador por idade (mesma meta-análise, subgrupos)
r_universitarios = 0.35   # k não reportado separadamente na fonte consultada
r_criancas = 0.21         # k=2 amostras — heterogeneidade alta, amostra pequena

# Moritz, Feltz, Fahrbach & Mack, 2000 — Research Quarterly for Exercise and Sport
# Meta-análise · 45 estudos, 102 correlações
r_esporte = 0.38
k_moritz = 45
n_moritz = None  # N pooled não reportado nos resumos consultados — DECLARADO COMO LIMITAÇÃO

# Talsma, Schüz, Schwarzer & Norris, 2018 — Learning and Individual Differences
# Meta-análise cross-lagged · k=11 estudos (1984-2016)
chi2_reciprocidade = 37.47  # teste de diferença entre os dois caminhos direcionais
k_talsma = 11

# Fator de correção
# NÃO SE APLICA — estes são coeficientes de correlação de meta-análises,
# não proporções de survey populacional autorrelatado.
fator_correcao_aplicavel = False

print("=" * 65)
print("  DADOS DA LITERATURA — CRENÇA E RESULTADO")
print("=" * 65)
print(f"\n  Multon, Brown & Lent, 1991 (N={n_multon:,}, k={k_multon} amostras):")
print(f"  → r geral (autoeficácia-desempenho acadêmico): {r_geral_academico:.2f}")
print(f"  → r universitários:                             {r_universitarios:.2f}")
print(f"  → r crianças (k=2, amostra pequena):            {r_criancas:.2f}")
print(f"\n  Moritz et al., 2000 (k={k_moritz} estudos, 102 correlações):")
print(f"  → r (autoeficácia-desempenho esportivo):        {r_esporte:.2f}")
print(f"\n  Talsma et al., 2018 (k={k_talsma} estudos, painel cross-lagged):")
print(f"  → χ² diferença entre caminhos recíprocos:       {chi2_reciprocidade:.2f} (p<0.0001)")
print(f"\n  Fator de correção ×0.80 aplicável? {fator_correcao_aplicavel}")
print("=" * 65)
```

```python
# --- O MODELO ---
# Assinatura estatística: transformação z de Fisher + IC 95%
# Fonte com N real disponível: Multon, Brown & Lent, 1991 (N=4.998)

# Passo 1 — Transformação z de Fisher para o r geral (N real conhecido)
z_geral = np.arctanh(r_geral_academico)
se_z = 1 / np.sqrt(n_multon - 3)
ic_z = (z_geral - 1.96 * se_z, z_geral + 1.96 * se_z)
ic_r = (np.tanh(ic_z[0]), np.tanh(ic_z[1]))

# Passo 2 — Comparação convergente entre domínios (acadêmico vs esportivo)
# Ambos convergem para r≈0.38 em adultos, em domínios completamente diferentes
convergencia = abs(r_geral_academico - r_esporte)

# Passo 3 — Efeito da maturidade cognitiva (crianças vs adultos)
gap_idade = r_universitarios - r_criancas

# Passo 4 — Variância explicada (r²) — o que a crença de fato "explica"
variancia_explicada = r_geral_academico ** 2

# Passo 5 — O paradoxo central
# Talsma et al. (2018) provam que os dois caminhos direcionais
# (crença -> resultado futuro E resultado -> crença futura)
# são estatisticamente diferentes entre si (χ² significativo),
# ou seja: reciprocidade real, não uma via de mão única.
# Isso significa que r=0.38 mede ASSOCIAÇÃO, não uma seta causal única.

print("=" * 65)
print("  MODELO — CRENÇA E RESULTADO")
print("=" * 65)
print(f"\n  Correlação geral (Multon, Brown & Lent, 1991):")
print(f"  → r = {r_geral_academico:.2f}")
print(f"  → IC 95% (Fisher z, N={n_multon:,}): [{ic_r[0]:.3f}, {ic_r[1]:.3f}]")
print(f"  → Variância explicada (r²):          {variancia_explicada*100:.1f}%")
print(f"\n  Convergência entre domínios:")
print(f"  → r acadêmico = {r_geral_academico:.2f} | r esportivo = {r_esporte:.2f}")
print(f"  → Diferença:    {convergencia:.2f} (praticamente idêntico)")
print(f"\n  Efeito da maturidade cognitiva:")
print(f"  → r crianças = {r_criancas:.2f} | r universitários = {r_universitarios:.2f}")
print(f"  → Gap:          +{gap_idade:.2f}")
print(f"\n  PARADOXO CENTRAL (Talsma et al., 2018):")
print(f"  → χ² = {chi2_reciprocidade:.2f}, p<0.0001")
print(f"  → Os caminhos crença→resultado e resultado→crença")
print(f"     são diferentes entre si, mas AMBOS são reais.")
print(f"  → A crença muda o resultado. O resultado também já")
print(f"     tinha mudado a crença antes.")
print("=" * 65)
```

```python
# --- VISUALIZAÇÃO — GRÁFICOS SEPARADOS ---

# ── GRÁFICO 1 — Convergência entre domínios (forest-style) ──
fig1, ax1 = plt.subplots(figsize=(10, 6))

labels = ['Crianças\n(desempenho acadêmico)\nMulton et al. 1991, k=2',
          'Adultos\n(desempenho acadêmico)\nMulton et al. 1991',
          'Adultos\n(desempenho esportivo)\nMoritz et al. 2000, k=45']
valores = [r_criancas, r_universitarios, r_esporte]
cores = ['#c8a84b', '#1a5f5a', '#1a5f5a']

y_pos = np.arange(len(labels))
ax1.hlines(y_pos, 0, valores, color=cores, linewidth=2, alpha=0.6)
ax1.scatter(valores, y_pos, color=cores, s=200, zorder=3)

for y, v in zip(y_pos, valores):
    ax1.text(v + 0.02, y, f'r={v:.2f}', va='center', fontweight='bold', fontsize=12)

ax1.set_yticks(y_pos)
ax1.set_yticklabels(labels, fontsize=10)
ax1.set_xlim(0, 0.55)
ax1.set_xlabel('Correlação (r) entre crença e desempenho')
ax1.set_title('Crença e Resultado Convergem Entre Domínios Diferentes\nMulton, Brown & Lent 1991 · Moritz et al. 2000',
              fontsize=13, pad=15)
ax1.axvline(x=0.38, color='#c0392b', linestyle='--', alpha=0.5, linewidth=1.5)
ax1.text(0.38, -0.7, 'r=0.38\n(convergência)', ha='center', fontsize=9, color='#c0392b')

plt.figtext(0.5, 0.01,
            'Fonte: Multon, Brown & Lent 1991 · Moritz, Feltz, Fahrbach & Mack 2000 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-038-grafico-01-convergencia.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 — Assinatura estatística: Fisher z e IC 95% ──
fig2, ax2 = plt.subplots(figsize=(10, 6))

x = np.linspace(0.30, 0.46, 1000)
z_x = np.arctanh(x)
y = stats.norm.pdf(z_x, loc=z_geral, scale=se_z) / (1 - x**2)  # densidade transformada de volta pra escala de r

ax2.plot(x, y, color='#1a5f5a', linewidth=2.5)
ax2.fill_between(x, y, alpha=0.2, color='#1a5f5a')
ax2.axvline(x=ic_r[0], color='#c8a84b', linestyle='--', linewidth=2)
ax2.axvline(x=ic_r[1], color='#c8a84b', linestyle='--', linewidth=2)
ax2.axvline(x=r_geral_academico, color='#333', linewidth=2)

ax2.text(ic_r[0] - 0.005, max(y)*0.5, f'{ic_r[0]:.3f}', ha='right', fontsize=10, color='#c8a84b')
ax2.text(ic_r[1] + 0.005, max(y)*0.5, f'{ic_r[1]:.3f}', ha='left', fontsize=10, color='#c8a84b')
ax2.text(r_geral_academico, max(y)*1.05, f'r={r_geral_academico:.2f}', ha='center', fontsize=11, fontweight='bold')

ax2.set_xlabel('Correlação (r) entre crença e desempenho acadêmico')
ax2.set_ylabel('Densidade (via transformação z de Fisher)')
ax2.set_title(f'Assinatura Estatística — Transformação z de Fisher\nMulton, Brown & Lent 1991 (N={n_multon:,}) · IC 95%',
              fontsize=12, pad=15)

plt.figtext(0.5, 0.01,
            'Fonte: Multon, Brown & Lent, 1991 — Journal of Counseling Psychology | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-038-grafico-02-fisher-z.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 — O paradoxo recíproco (esquema de caminhos) ──
fig3, ax3 = plt.subplots(figsize=(10, 6))
ax3.set_xlim(0, 10)
ax3.set_ylim(0, 6)
ax3.axis('off')

# Caixas
box_crenca = dict(boxstyle='round,pad=0.6', facecolor='#1a5f5a', edgecolor='none')
box_resultado = dict(boxstyle='round,pad=0.6', facecolor='#c0392b', edgecolor='none')

ax3.text(2, 4.5, 'CRENÇA\n(autoeficácia)', ha='center', va='center', fontsize=13,
         color='white', fontweight='bold', bbox=box_crenca)
ax3.text(8, 4.5, 'RESULTADO\n(desempenho)', ha='center', va='center', fontsize=13,
         color='white', fontweight='bold', bbox=box_resultado)

ax3.text(2, 1.5, 'CRENÇA\n(t+1)', ha='center', va='center', fontsize=13,
         color='white', fontweight='bold', bbox=box_crenca)
ax3.text(8, 1.5, 'RESULTADO\n(t+1)', ha='center', va='center', fontsize=13,
         color='white', fontweight='bold', bbox=box_resultado)

# Setas cruzadas (cross-lagged)
ax3.annotate('', xy=(7, 2.0), xytext=(3, 4.2),
             arrowprops=dict(arrowstyle='->', color='#1a5f5a', lw=2.5))
ax3.annotate('', xy=(3, 2.0), xytext=(7, 4.2),
             arrowprops=dict(arrowstyle='->', color='#c0392b', lw=2.5))

ax3.text(5, 3.5, 'crença → resultado futuro', ha='center', fontsize=10, color='#1a5f5a', style='italic')
ax3.text(5, 2.6, 'resultado → crença futura', ha='center', fontsize=10, color='#c0392b', style='italic')

ax3.text(5, 0.3, f'χ² = {chi2_reciprocidade:.2f}, p<0.0001 — os dois caminhos são diferentes entre si, mas ambos são reais',
         ha='center', fontsize=10, color='#333')
ax3.set_title('O Paradoxo Recíproco — Painel Cross-Lagged\nTalsma, Schüz, Schwarzer & Norris, 2018 (k=11 estudos)',
              fontsize=12, pad=10)

plt.figtext(0.5, 0.01,
            'Fonte: Talsma, Schüz, Schwarzer & Norris, 2018 — Learning and Individual Differences | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-038-grafico-03-paradoxo-reciproco.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**A crença muda o resultado. O resultado também já tinha mudado a crença antes.**

Não é uma seta. É um ciclo.

Acreditar que vai dar certo tem uma correlação real e replicada
com o resultado, r≈0.38, o mesmo número aparecendo em desempenho
acadêmico e em desempenho esportivo, dois domínios sem nenhuma
relação entre si.

Mas o painel cross-lagged de Talsma et al. (2018) mostra algo
que a frase de efeito nunca admite: o caminho contrário também
é estatisticamente diferente e real. O resultado de ontem constrói
a crença de hoje, que constrói o resultado de amanhã.

Isso significa que "acreditar em você" não é o ponto de partida
mágico. É o ponto onde você está depois de alguns acertos pequenos
que talvez você nem tenha registrado como prova.

A pergunta que fica:

*Que resultado pequeno você já teve, que ainda não virou crença?*

---

### ⚠️ Limitações do Modelo

- r=0.38 explica cerca de 14% da variância no desempenho. É uma
  associação moderada, não uma garantia individual
- N pooled da meta-análise de Moritz et al. (2000) não estava
  disponível nos resumos consultados — usado apenas como
  replicação convergente do r, sem IC 95% próprio
- O subgrupo de crianças em Multon et al. (1991) tem apenas
  k=2 amostras — alta incerteza, não generalizável
- Cross-lagged mostra reciprocidade, não elimina toda
  confundidora (ex: traços de personalidade estáveis podem
  influenciar os dois lados ao mesmo tempo)

*A ciência é honesta sobre o que não sabe. O modelo também.*
