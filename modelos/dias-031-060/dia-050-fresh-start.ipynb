### 📖 A História

Eu sempre esperei o momento certo para recomeçar.

A próxima segunda. O primeiro dia do mês. O ano novo. Depois das férias. Quando as coisas acalmarem. Já perdi a conta de quantas vezes adiei um recomeço à espera de uma data que parecia mais digna, mais redonda, mais oficial para carregar o peso de uma virada.

Parte disso é fuga, eu já sei. Ontem mesmo escrevi sobre como a gente adia por emoção, não por preguiça.

Mas tem outra coisa que os dados me mostraram, e que eu não esperava. Essas datas não são só desculpa. Elas funcionam de verdade. O calendário tem alavancas reais, e a gente costuma escolher as fracas achando que são as fortes.

---

### 📚 O Conceito: O Efeito Fresh Start

Hengchen Dai, Katherine Milkman e Jason Riis (2014) descreveram o que chamaram de **efeito fresh start**: marcos temporais, como o começo de uma semana, de um mês ou de um ano, aumentam a motivação para perseguir objetivos.

O mecanismo é sutil e bonito. Um marco cria uma descontinuidade na linha do tempo. Ele separa, na sua cabeça, o eu antigo e imperfeito do eu novo que está começando agora. Essa distância psicológica faz você se sentir uma página virada, e páginas viradas puxam comportamento aspiracional: ir à academia, começar a dieta, retomar o projeto.

Você não recomeça porque a data é mágica. Você recomeça porque, por um instante, o calendário te deixa ser outra pessoa.

---

### 🧮 O Modelo

Uso os dados de frequência de academia do estudo (registros de catraca, não autorrelato) para medir quanto cada marco temporal aumenta a probabilidade de alguém aparecer, e comparo a força de cada data.

**Fonte:**
- Dai, H., Milkman, K. L. & Riis, J. (2014). *The Fresh Start Effect: Temporal Landmarks Motivate Aspirational Behavior*. **Management Science**.
  Idas à academia: N=11.912 alunos, 442 dias. Também: buscas por "dieta" no Google (3.104 dias) e 66.062 contratos de meta no stickK.

**Nota metodológica:** os aumentos são medidas comportamentais objetivas (frequência real registrada), não proporções de survey autorrelatado. Por isso o fator ×0.80 não se aplica.


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

# Paleta 365 Probabilidades
DOURADO = '#c8a84b'
VERMELHO = '#c0392b'
VERDE    = '#1a5f5a'
CINZA    = '#6b6a64'

np.random.seed(45)  # Dia #045

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Dai, Milkman & Riis (2014), Management Science
# Aumento (%) nas idas à academia vs um dia comum (N=11.912 alunos, 442 dias)

marcos = {
    'Início de semestre':        47.1,
    'Início de semana (segunda)':33.4,
    'Depois de pausa escolar':   24.3,
    'Início de mês':             14.4,
    'Início de ano (1º jan)':    11.6,
    'Depois do aniversário':      7.5,
}
N_academia = 11912
dias_academia = 442
N_stickk = 66062  # contratos de meta
N_google_dias = 3104

print("=" * 66)
print("  DADOS DA LITERATURA — EFEITO FRESH START")
print("=" * 66)
print(f"  Dai, Milkman & Riis 2014 · academia: N={N_academia:,} alunos, {dias_academia} dias")
print("  Aumento nas idas à academia vs um dia comum:")
for nome, pct in sorted(marcos.items(), key=lambda x: -x[1]):
    print(f"    {nome:<28} +{pct:.1f}%")
print("=" * 66)
```

```python
# --- O MODELO: quanto cada recomeço realmente puxa ---

# Razão de taxa (rate ratio): quão mais provável aparecer no marco vs dia comum
rr = {nome: 1 + pct/100 for nome, pct in marcos.items()}

rr_segunda = rr['Início de semana (segunda)']
rr_janeiro = rr['Início de ano (1º jan)']
vantagem_seg_vs_jan = marcos['Início de semana (segunda)'] - marcos['Início de ano (1º jan)']
# Razão dos empurrões (aumentos vs dia comum): quão maior é o "boost" da segunda
razao_empurrao = marcos['Início de semana (segunda)'] / marcos['Início de ano (1º jan)']

print("=" * 66)
print("  MODELO — A FORÇA DE CADA RECOMEÇO")
print("=" * 66)
print("  Razão de taxa (chance de aparecer vs dia comum):")
for nome, pct in sorted(marcos.items(), key=lambda x: -x[1]):
    print(f"    {nome:<28} {rr[nome]:.2f}×")
print()
print(f"  A SEGUNDA-FEIRA:")
print(f"  → Começar numa segunda deixa você {rr_segunda:.2f}× mais")
print(f"    propenso a aparecer do que num dia comum (+{marcos['Início de semana (segunda)']:.1f}%).")
print()
print(f"  A ARMADILHA DA DATA PERFEITA:")
print(f"  → 1º de janeiro:      +{marcos['Início de ano (1º jan)']:.1f}%")
print(f"  → Próxima segunda:    +{marcos['Início de semana (segunda)']:.1f}%")
print(f"  → O empurrão da próxima segunda é ~{razao_empurrao:.1f}× o do ano novo")
print(f"    ({vantagem_seg_vs_jan:.1f} pontos percentuais a mais).")
print()
print(f"  RESPOSTA:")
print(f"  → Segunda-feira é, sim, mais poderosa do que você pensa.")
print(f"    E ela chega toda semana. Você não precisa esperar janeiro.")
print("=" * 66)
```

```python
# --- VISUALIZAÇÃO — 3 GRÁFICOS SEPARADOS ---

# ── GRÁFICO 1 — Um dia comum vs uma segunda-feira ──
fig1, ax1 = plt.subplots(figsize=(10, 6))
cats = ['Um dia\nqualquer', 'O início de\numa semana']
vals = [100, 100 * rr_segunda]
bars = ax1.bar(cats, vals, color=[CINZA, DOURADO], alpha=0.9, width=0.45)
ax1.set_ylabel('Idas à academia (dia comum = 100)')
ax1.set_ylim(0, 150)
ax1.set_title('Segunda-feira é mais poderosa do que você pensa\nIdas à academia — Dai, Milkman & Riis 2014 (N=11.912)',
              fontsize=13, pad=15)
for bar, v in zip(bars, vals):
    ax1.text(bar.get_x()+bar.get_width()/2, v+2, f'{v:.0f}',
             ha='center', fontweight='bold', fontsize=16)
ax1.annotate('', xy=(1, 100*rr_segunda), xytext=(0, 100),
             arrowprops=dict(arrowstyle='->', color=VERMELHO, lw=2))
ax1.text(0.5, 122, '+33,4%', ha='center', fontsize=14, color=VERMELHO, fontweight='bold')
plt.figtext(0.5, 0.005,
            'Registros de catraca, não autorrelato | Dai et al. 2014 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-045-grafico-01-segunda.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 — ASSINATURA ESTATÍSTICA: o cardápio de recomeços ──
fig2, ax2 = plt.subplots(figsize=(10, 6.5))
itens = sorted(marcos.items(), key=lambda x: x[1])
nomes = [k for k, _ in itens]
rrs = [1 + v/100 for _, v in itens]
ys = np.arange(len(nomes))
for i, (nome, pct) in enumerate(itens):
    destaque = 'segunda' in nome.lower()
    cor = DOURADO if destaque else VERDE
    ax2.plot([1.0, 1 + pct/100], [i, i], color=cor, lw=3, alpha=0.85, zorder=2)
    ax2.scatter([1 + pct/100], [i], color=cor, s=110, zorder=3, edgecolor='#333')
    ax2.text(1 + pct/100 + 0.012, i, f'{1+pct/100:.2f}×  (+{pct:.1f}%)',
             va='center', fontsize=11, color='#333')
ax2.axvline(1.0, color=VERMELHO, ls='--', lw=1.5, alpha=0.7)
ax2.text(1.0, len(nomes)-0.35, 'dia comum', color=VERMELHO, fontsize=10, ha='center')
ax2.set_yticks(ys); ax2.set_yticklabels(nomes)
ax2.set_xlabel('Razão de taxa — chance de aparecer vs um dia comum')
ax2.set_xlim(0.95, 1.62)
ax2.set_title('A ASSINATURA ESTATÍSTICA\nO cardápio de recomeços: nem toda data puxa igual',
              fontsize=13, pad=15)
plt.figtext(0.5, 0.005,
            'Tamanho de efeito por marco temporal · Dai et al. 2014 (N=11.912) | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-045-grafico-02-cardapio.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 — Não precisa esperar o ano novo ──
fig3, ax3 = plt.subplots(figsize=(10, 6))
cats3 = ['Esperar o\n1º de janeiro', 'A próxima\nsegunda-feira']
vals3 = [marcos['Início de ano (1º jan)'], marcos['Início de semana (segunda)']]
bars3 = ax3.bar(cats3, vals3, color=[CINZA, DOURADO], alpha=0.9, width=0.45)
ax3.set_ylabel('Aumento nas idas à academia (%)')
ax3.set_ylim(0, 42)
ax3.set_title('Você não precisa esperar a data perfeita\nA próxima segunda pesa mais que o ano novo — Dai et al. 2014',
              fontsize=13, pad=15)
for bar, v in zip(bars3, vals3):
    ax3.text(bar.get_x()+bar.get_width()/2, v+0.8, f'+{v:.1f}%',
             ha='center', fontweight='bold', fontsize=16)
ax3.text(0.5, 0.80, f'~{razao_empurrao:.1f}× o empurrão', ha='center',
         fontsize=13, color=VERMELHO, fontweight='bold', transform=ax3.transAxes)
plt.figtext(0.5, 0.005,
            'Início de ano +11,6% vs início de semana +33,4% | Dai et al. 2014 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-045-grafico-03-nao-espere-janeiro.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**Começar numa segunda-feira deixa você 33% mais propenso a aparecer do que num dia comum.**

Não porque a segunda tenha mágica. Mas porque ela cria uma quebra: o eu que não treinou semana passada fica do outro lado da linha, e um eu novo começa agora. Essa distância psicológica é combustível de verdade, medido em catracas de academia, não em intenção.

E aqui está a parte que me pegou. A gente costuma esperar a data mais imponente para recomeçar, o 1º de janeiro. Mas o ano novo puxa só +11,6%. A próxima segunda-feira puxa +33,4%, quase o triplo. A data mais poderosa não é a mais rara. É a que volta toda semana.

Ou seja: enquanto você espera o momento perfeito, passam por você 52 recomeços por ano, um a cada segunda. E cada um deles é mais forte do que aquele que você está esperando lá em janeiro.

O momento certo para recomeçar não é uma data no futuro. É a próxima que chegar.

*Qual recomeço você está adiando à espera de uma data mais digna, quando a próxima segunda já serviria?*

### ⚠️ Limitações do Modelo

- Os dados de academia vêm de uma população universitária (N=11.912); a magnitude pode variar por idade, cultura e tipo de hábito.
- Fresh start eleva a intenção e a iniciativa de começar; não garante manutenção do hábito no longo prazo.
- Os aumentos são relativos a um dia comum de base; efeitos podem se sobrepor (uma segunda que também é 1º do mês, por exemplo).
- As razões de taxa apresentadas são tamanhos de efeito do estudo; não recalculei IC a partir dos microdados.
- O fator ×0.80 não foi aplicado: são medidas comportamentais objetivas, não survey autorrelatado.
