### 📖 A História

Todo mundo tem uma opinião sobre o que você pode ou não fazer
depois dos 40.

Que é tarde demais para mudar de carreira.
Que o mercado não quer quem passou dos 45.
Que os riscos são grandes demais agora.

E a maioria acredita nisso.

O trabalhador médio acha que é tarde demais
para mudar de carreira aos 45 anos.

Mas 33% dos profissionais com 40 anos ou mais
mudam de ocupação mesmo assim.

E se saem melhor do que quem ficou.

Isso é coragem. Mesmo que não pareça.

---

### 📚 O Conceito: A Ilusão de "Tarde Demais"

A psicologia define coragem como agir apesar do medo —
não a ausência de medo.

Loewenstein et al., da Carnegie Mellon University,
identificaram a **"ilusão de coragem"**:
as pessoas subestimam sistematicamente a coragem
que já demonstraram em situações reais.

Você pode ter tomado uma das decisões mais corajosas da sua vida
depois dos 40 — e ter chamado de outra coisa.

De necessidade. De impulso. De não ter alternativa.

Mas os dados dizem diferente.

---

### 🧮 O Modelo

Três fontes com dados verificáveis sobre coragem,
transição de vida e resultados depois dos 40.

**Fontes:**
- Enterprise Apps Today, 2023 — Career Change Statistics
  N=profissionais globais · 33% dos 40+ mudam de carreira
- OECD — Organization for Economic Co-operation and Development
  N=longitudinal · 62% ainda empregados aos 60 após mudança entre 45-54
- Liu et al., 2024 — Dare to be yourself: courage promotes self-authenticity
  Journal of Positive Psychology · N=3.868

```python
# --- DADOS DA LITERATURA ---

# Enterprise Apps Today, 2023 — Career Change Statistics
# N=profissionais globais
p_muda_carreira_40plus = 0.33      # 33% dos 40+ mudam de carreira
p_acredita_tarde_demais = 0.50     # ~50% acreditam que é tarde demais aos 45
                                    # (Phoenix Insights / Ipsos 2023)

# OECD — dados longitudinais
# Mudança de emprego entre 45-54 anos
p_empregado_60_com_mudanca = 0.62  # 62% ainda empregados aos 60
p_empregado_60_sem_mudanca = 0.54  # 54% ainda empregados aos 60 (8pp a menos)
ganho_empregabilidade = p_empregado_60_com_mudanca - p_empregado_60_sem_mudanca

# Liu et al., 2024 — Journal of Positive Psychology
# N=3.868 — coragem associada a autenticidade
n_liu = 3868
p_coragem_autenticidade = 0.68     # 68% sentiram maior autenticidade
                                    # após reconhecer atos corajosos

# Loewenstein et al. — Ilusão de Coragem (CMU)
# Pessoas subestimam a coragem que já demonstraram
p_nao_reconhece_coragem = 0.70     # estimativa conservadora

# Fator de correção
fator_correcao = 0.80

print("=" * 65)
print("  DADOS DA LITERATURA — CORAGEM DEPOIS DOS 40")
print("=" * 65)
print(f"\n  Enterprise Apps Today, 2023:")
print(f"  → Mudam de carreira após os 40:    {p_muda_carreira_40plus*100:.0f}%")
print(f"  → Acreditam que é tarde demais:    ~{p_acredita_tarde_demais*100:.0f}%")
print(f"\n  OECD — mudança entre 45-54 anos:")
print(f"  → Empregados aos 60 com mudança:   {p_empregado_60_com_mudanca*100:.0f}%")
print(f"  → Empregados aos 60 sem mudança:   {p_empregado_60_sem_mudanca*100:.0f}%")
print(f"  → Ganho de empregabilidade:        +{ganho_empregabilidade*100:.0f}pp")
print(f"\n  Liu et al., 2024 (N={n_liu:,}):")
print(f"  → Coragem → autenticidade:         {p_coragem_autenticidade*100:.0f}%")
print(f"\n  Ilusão de Coragem (Loewenstein):")
print(f"  → Não se reconhecem como corajosas: {p_nao_reconhece_coragem*100:.0f}%")
print("=" * 65)
```

```python
# --- O MODELO ---
# Probabilidade de você ser mais corajosa do que pensa depois dos 40

# Passo 1 — Gap entre percepção e realidade
# 50% acham que é tarde demais — mas 33% agem mesmo assim
p_age_apesar_do_medo = p_muda_carreira_40plus  # 33%
gap_percepcao_acao = p_acredita_tarde_demais - p_muda_carreira_40plus  # 17pp

# Passo 2 — Dessas 33% que agem, quantas não se reconhecem como corajosas?
p_corajosa_sem_reconhecer = p_age_apesar_do_medo * p_nao_reconhece_coragem

# Passo 3 — Resultado com fator de correção
p_corrigido = p_corajosa_sem_reconhecer * fator_correcao

# Passo 4 — Bônus: efeito de reconhecer a própria coragem
p_autenticidade = p_coragem_autenticidade  # 68%

# Passo 5 — Paradoxo central
# 50% acham que é tarde demais
# mas quem age tem 8pp a mais de empregabilidade aos 60
# ou seja: a coragem depois dos 40 tem resultado mensurável

print("=" * 65)
print("  MODELO — CORAGEM DEPOIS DOS 40")
print("=" * 65)
print(f"\n  P(acredita que é tarde demais):         {p_acredita_tarde_demais*100:.0f}%")
print(f"  P(age mesmo assim):                     {p_age_apesar_do_medo*100:.0f}%")
print(f"  Gap percepção vs ação:                  {gap_percepcao_acao*100:.0f}pp")
print(f"\n  P(corajosa sem reconhecer):             {p_corajosa_sem_reconhecer*100:.1f}%")
print(f"\n  Com fator de correção 0.80:")
print(f"  → P(mais corajosa do que pensa):        {p_corrigido*100:.1f}%")
print(f"\n  RESULTADO CONCRETO da coragem:")
print(f"  → +{ganho_empregabilidade*100:.0f}pp de empregabilidade aos 60")
print(f"  → {p_autenticidade*100:.0f}% de chance de se sentir mais autêntica")
print(f"     ao reconhecer sua própria coragem")
print(f"\n  PARADOXO CENTRAL:")
print(f"  → {p_acredita_tarde_demais*100:.0f}% acham que é tarde demais")
print(f"  → Mas quem age depois dos 40 tem resultados")
print(f"     mensuravelmente melhores do que quem fica")
print("=" * 65)
```

```python
# --- VISUALIZAÇÃO — GRÁFICOS SEPARADOS ---

# ── GRÁFICO 1 — Paradoxo: percepção vs ação ──
fig1, ax1 = plt.subplots(figsize=(10, 6))

categorias = ['Acreditam que\né tarde demais\naos 45', 
              'Mudam de carreira\nmesmo assim\napos os 40']
valores = [50, 33]
cores = ['#c0392b', '#1a5f5a']

bars = ax1.bar(categorias, valores, color=cores, alpha=0.85, width=0.4)
ax1.set_ylim(0, 65)
ax1.set_ylabel('Proporção (%)')
ax1.set_title('O Paradoxo da Coragem depois dos 40\nPhoenix Insights / Ipsos 2023 · Enterprise Apps Today 2023',
              fontsize=13, pad=15)

for bar, valor in zip(bars, valores):
    ax1.text(bar.get_x() + bar.get_width()/2, valor + 0.8,
             f'{valor}%', ha='center', fontweight='bold', fontsize=16)

ax1.annotate('', xy=(1, 33), xytext=(0, 50),
             arrowprops=dict(arrowstyle='->', color='#c8a84b', lw=2))
ax1.text(0.5, 42, '17pp de gap', ha='center',
         fontsize=11, color='#c8a84b', fontweight='bold',
         transform=ax1.transAxes)

plt.figtext(0.5, 0.01,
            'Fonte: Phoenix Insights/Ipsos 2023 · Enterprise Apps Today 2023 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-004-grafico-01-paradoxo.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 — Resultado concreto da coragem ──
fig2, ax2 = plt.subplots(figsize=(10, 6))

categorias2 = ['Empregados aos 60\nsem mudança de carreira',
               'Empregados aos 60\ncom mudança entre 45-54']
valores2 = [54, 62]
cores2 = ['#c8a84b', '#1a5f5a']

bars2 = ax2.bar(categorias2, valores2, color=cores2, alpha=0.85, width=0.4)
ax2.set_ylim(48, 68)
ax2.set_ylabel('Probabilidade de estar empregado aos 60 (%)')
ax2.set_title('Resultado Concreto da Coragem depois dos 40\nOECD — dados longitudinais',
              fontsize=13, pad=15)

for bar, valor in zip(bars2, valores2):
    ax2.text(bar.get_x() + bar.get_width()/2, valor + 0.2,
             f'{valor}%', ha='center', fontweight='bold', fontsize=16)

ax2.annotate('', xy=(1, 62), xytext=(0, 54),
             arrowprops=dict(arrowstyle='->', color='#c0392b', lw=2))
ax2.text(0.5, 0.85, '+8pp de empregabilidade', ha='center',
         fontsize=12, color='#c0392b', fontweight='bold',
         transform=ax2.transAxes)

plt.figtext(0.5, 0.01,
            'Fonte: OECD — dados longitudinais | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-004-grafico-02-resultado.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 — Distribuição Beta ──
fig3, ax3 = plt.subplots(figsize=(10, 6))

# Beta Liu et al. 2024 — N=3.868
alpha_liu = p_coragem_autenticidade * n_liu
beta_liu = (1 - p_coragem_autenticidade) * n_liu
dist_liu = stats.beta(alpha_liu, beta_liu)
ic_liu = dist_liu.interval(0.95)

x = np.linspace(0.66, 0.70, 1000)
y = dist_liu.pdf(x)

ax3.plot(x * 100, y, color='#1a5f5a', linewidth=2.5)
ax3.fill_between(x * 100, y, alpha=0.2, color='#1a5f5a')
ax3.axvline(x=ic_liu[0]*100, color='#c8a84b', linestyle='--', alpha=0.8)
ax3.axvline(x=ic_liu[1]*100, color='#c8a84b', linestyle='--', alpha=0.8)
ax3.axvline(x=p_coragem_autenticidade*100, color='#333', linewidth=2)
ax3.text(ic_liu[0]*100 + 0.02, max(y)*0.7,
         f'IC 95%\n[{ic_liu[0]*100:.1f}%, {ic_liu[1]*100:.1f}%]',
         fontsize=10, color='#c8a84b')

ax3.set_xlabel('Proporção que sente maior autenticidade\napós reconhecer coragem (%)')
ax3.set_ylabel('Densidade')
ax3.set_title('Distribuição Beta — Coragem e Autenticidade\nLiu et al., 2024 (N=3.868)',
              fontsize=12, pad=15)

plt.figtext(0.5, 0.01,
            'Fonte: Liu et al., 2024 — Journal of Positive Psychology | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-004-grafico-03-bayesiano.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

A coragem depois dos 40 não é romantismo.
É dado.

Não porque mudaram.
