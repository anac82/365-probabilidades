### A História

Existe uma conta que quase ninguém faz.

Quando alguém vende uma ação que subiu, a frase pronta é que realizou o
lucro. Quando segura uma que caiu, a frase pronta é que está esperando
recuperar. As duas soam sensatas. Ditas juntas, na mesma boca, no mesmo
dia, elas descrevem uma regra estranha: vender o que está dando certo e
segurar o que está dando errado.

Em 1998, um pesquisador americano conseguiu os registros de negociação de
10 mil contas de uma corretora, sete anos de histórico, e resolveu medir
isso direito. A pergunta dele não era quantas vendas foram com lucro, que
é uma pergunta enganosa: em mercado de alta, qualquer um vende mais
vencedores, porque tem mais vencedores na carteira. A pergunta certa é
outra. Das oportunidades de vender com lucro, quantas foram usadas? E das
oportunidades de vender com prejuízo, quantas foram usadas?

Vinte anos depois, do outro lado do equador, a mesma conta foi refeita com
os dados de todo o mercado de ações brasileiro.

O número bateu na segunda casa.

---

### O Conceito: o ponto de referência

O comportamento tem nome desde 1985, efeito disposição, e tem uma
explicação emprestada da teoria do prospecto de Kahneman e Tversky, de
1979.

A ideia central é que as pessoas não avaliam o dinheiro pelo tamanho do
patrimônio, e sim pela distância até um ponto de referência, que em geral
é o preço que pagaram. Acima dele, no campo dos ganhos, a curva de valor é
côncava e a pessoa fica avessa ao risco: quer garantir o que tem. Abaixo
dele, no campo das perdas, a curva é convexa e a pessoa fica tolerante ao
risco: prefere apostar na recuperação a assumir a perda.

Daí sai a assimetria. O mesmo movimento de preço pesa diferente conforme o
lado da linha em que ele acontece.

Vale registrar que essa explicação é disputada. Há trabalhos mostrando que
a teoria do prospecto, sozinha, só prevê o efeito disposição em condições
bem específicas, e há quem proponha que o que importa é a utilidade do ato
de realizar, não a curva. O comportamento é robusto. A explicação, não
tanto.

---

### O Modelo

Duas bases administrativas, nenhum questionário, e três camadas de conta
própria: a assimetria com incerteza real, o encontro entre os dois
mercados e a virada de dezembro.

**Fontes:**
- Odean, T. (1998). Are Investors Reluctant to Realize Their Losses?
  *The Journal of Finance*, 53(5), 1775-1798. Registros de 10.000 contas
  de uma corretora de desconto americana, 1987 a 1993, 162.948 negócios.

  Disposição: Propensão à Venda de Investidores Individuais e
  Institucionais. *Revista Brasileira de Economia*, 73(1), 97-119. Base
  cedida pela B3 com o mercado inteiro, janeiro de 2012 a outubro de 2014,
  cerca de 60 milhões de operações e 537.845 investidores cadastrados,
  dos quais 505.233 pessoas físicas.
- Kahneman, D. & Tversky, A. (1979). Prospect Theory: An Analysis of
  Decision under Risk. *Econometrica*, 47(2), 263-291. Entra pelo achado
  específico da função de valor com ponto de referência, não por
  autoridade. Reuso de nome declarado: Kahneman já apareceu nos dias

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

SEED = 42
rng = np.random.default_rng(SEED)

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

VERMELHO = '#c0392b'
DOURADO = '#c8a84b'
VERDE = '#1a5f5a'
CINZA = '#6b6a64'

def mil(n):
    return f"{n:,}".replace(",", ".")

print("Bibliotecas carregadas. Semente fixa:", SEED)
```

```python
# --- DADOS DA LITERATURA ---

# ------------------------------------------------------------------
# Odean (1998), Journal of Finance 53(5) - Tabela I
# 10.000 contas, corretora de desconto americana, 1987-1993
# ------------------------------------------------------------------
us_gr_ano, us_gp_ano = 13_883, 79_658      # ganhos realizados e potenciais
us_pr_ano, us_pp_ano = 11_930, 110_348     # perdas realizadas e potenciais

us_gr_dez, us_gp_dez = 866, 7_131          # dezembro
us_pr_dez, us_pp_dez = 1_555, 10_604

# janeiro a novembro sai por subtracao (o artigo publica os dois blocos)
us_gr_jn, us_gp_jn = us_gr_ano - us_gr_dez, us_gp_ano - us_gp_dez
us_pr_jn, us_pp_jn = us_pr_ano - us_pr_dez, us_pp_ano - us_pp_dez

# Teste alternativo do proprio artigo, por conta (1.893 contas)
us_conta_pgr, us_conta_plr, us_conta_t = 0.57, 0.36, 19
us_dolar_pgr, us_dolar_plr, us_dolar_t = 0.58, 0.42, 13

# Retornos em excesso depois da decisao (Tabela VI)
horizontes = [84, 252, 504]
exc_vendidos = [0.0047, 0.0235, 0.0645]    # vencedores que foram vendidos
exc_mantidos = [-0.0056, -0.0106, 0.0287]  # perdedores que ficaram na carteira
exc_p = [0.002, 0.001, 0.014]

# ------------------------------------------------------------------
# Prates, da Costa Jr. & Santos (2019), RBE 73(1) - Tabela 5
# Mercado inteiro da B3, jan/2012 a out/2014. Contagens em milhares.
# ------------------------------------------------------------------
br_gr, br_gp = 5_108_000, 147_097_000      # pessoa fisica
br_pr, br_pp = 3_996_000, 174_285_000

br_tipos = [
    ("Pessoa fisica",                 0.0336, 0.0224,  0.0111),
    ("Sociedades",                    0.0274, 0.0198,  0.0075),
    ("PF profissional de mercado",    0.0394, 0.0328,  0.0066),
    ("Distribuidoras",                0.0343, 0.0331,  0.0012),
    ("Clubes de investimento",        0.0163, 0.0151,  0.0011),
    ("Fundos de previdencia",         0.0083, 0.0086, -0.0004),
    ("Fundos mutuos",                 0.0137, 0.0148, -0.0011),
    ("Estrangeiro PF",                0.0126, 0.0144, -0.0018),
    ("Sociedades corretoras",         0.0187, 0.0214, -0.0028),
    ("Corretoras de outras pracas",   0.0240, 0.0270, -0.0030),
    ("Bancos comerciais",             0.0157, 0.0197, -0.0040),
    ("Bancos de investimento",        0.0240, 0.0314, -0.0074),
]

br_retorno_medio = {"Pessoa fisica": 1.03, "Fundos mutuos": 3.34,
                    "Fundos de previdencia": 7.01}

aplica_fator_080 = False

print("=" * 68)
print("  DADOS - EFEITO DISPOSICAO EM DOIS MERCADOS")
print("=" * 68)
print(f"\n  EUA, Odean 1998 (10.000 contas, 1987-1993):")
print(f"  -> ganhos:  {mil(us_gr_ano)} realizados / {mil(us_gp_ano)} potenciais")
print(f"  -> perdas:  {mil(us_pr_ano)} realizadas / {mil(us_pp_ano)} potenciais")
print(f"  -> dezembro: {mil(us_gr_dez)} e {mil(us_pr_dez)} realizados")
print(f"\n  BRASIL, Prates et al. 2019 (mercado inteiro da B3, 2012-2014):")
print(f"  -> pessoa fisica, ganhos:  {mil(br_gr)} realizados / {mil(br_gp)} potenciais")
print(f"  -> pessoa fisica, perdas:  {mil(br_pr)} realizadas / {mil(br_pp)} potenciais")
print(f"  -> tipos de investidor analisados: {len(br_tipos)}")
print(f"\n  Fator x0,80 aplicado: {'SIM' if aplica_fator_080 else 'NAO'}")
print("=" * 68)
```

```python
# --- O MODELO ---
# Tres camadas de conta propria sobre as duas bases.

N_SIM = 200_000

def posterior(realizados, potenciais):
    "Posterior de Jeffreys para a proporcao realizada."
    return stats.beta(realizados + 0.5, potenciais + 0.5)

def sorteia(realizados, potenciais, n=N_SIM):
    return posterior(realizados, potenciais).rvs(n, random_state=rng)

# ---------------------------------------------------------------
# CAMADA 1 - A assimetria com incerteza real (EUA)
# ---------------------------------------------------------------
us_pgr_ano = us_gr_ano / (us_gr_ano + us_gp_ano)
us_plr_ano = us_pr_ano / (us_pr_ano + us_pp_ano)

sim_us_pgr = sorteia(us_gr_ano, us_gp_ano)
sim_us_plr = sorteia(us_pr_ano, us_pp_ano)
sim_us_razao = sim_us_pgr / sim_us_plr
us_razao_med = np.median(sim_us_razao)
us_razao_ic = np.percentile(sim_us_razao, [2.5, 97.5])

# Sensibilidade a dependencia, aplicada AOS DOIS PAISES.
# Odean avisa que as decisoes nao sao independentes e que isso infla
# o teste. No Brasil a dependencia e provavelmente maior ainda: cada
# posicao em carteira gera uma observacao por DIA. Dividir as contagens
# por um efeito de desenho d equivale a dizer que so 1 em cada d
# observacoes traz informacao nova.
deffs = [1, 10, 100, 1000]
MARGEM_EQUIV = 0.10   # margem de equivalencia entre as razoes dos dois paises

def razao_com_desconto(gr, gp, pr, pp, d, n=100_000):
    a = sorteia(gr / d, gp / d, n)
    b = sorteia(pr / d, pp / d, n)
    return a / b

sens = {}
for d in deffs:
    r_us = razao_com_desconto(us_gr_ano, us_gp_ano, us_pr_ano, us_pp_ano, d)
    r_br = razao_com_desconto(br_gr, br_gp, br_pr, br_pp, d)
    dif = r_us - r_br
    sens[d] = dict(
        us_ic=np.percentile(r_us, [2.5, 97.5]), us_p1=np.mean(r_us > 1),
        br_ic=np.percentile(r_br, [2.5, 97.5]), br_p1=np.mean(r_br > 1),
        dif_ic=np.percentile(dif, [2.5, 97.5]),
        p_equiv=np.mean(np.abs(dif) < MARGEM_EQUIV),
    )

# ---------------------------------------------------------------
# CAMADA 2 - O encontro dos dois mercados
# ---------------------------------------------------------------
br_pgr = br_gr / (br_gr + br_gp)
br_ppr = br_pr / (br_pr + br_pp)

sim_br_pgr = sorteia(br_gr, br_gp)
sim_br_ppr = sorteia(br_pr, br_pp)
sim_br_razao = sim_br_pgr / sim_br_ppr
br_razao_med = np.median(sim_br_razao)
br_razao_ic = np.percentile(sim_br_razao, [2.5, 97.5])

dif_razoes = sim_us_razao - sim_br_razao
dif_med = np.median(dif_razoes)
dif_ic = np.percentile(dif_razoes, [2.5, 97.5])

# Relativo contra absoluto: a razao de 1,5 e relativa. Em pontos
# percentuais por oportunidade, a distancia e bem diferente.
us_dif_abs = (us_gr_ano / (us_gr_ano + us_gp_ano) - us_pr_ano / (us_pr_ano + us_pp_ano)) * 100
br_dif_abs = (br_pgr - br_ppr) * 100

# ---------------------------------------------------------------
# CAMADA 3 - Dezembro como controle interno (EUA)
# ---------------------------------------------------------------
sim_pgr_jn = sorteia(us_gr_jn, us_gp_jn)
sim_plr_jn = sorteia(us_pr_jn, us_pp_jn)
sim_pgr_dz = sorteia(us_gr_dez, us_gp_dez)
sim_plr_dz = sorteia(us_pr_dez, us_pp_dez)

cd_jn = sim_pgr_jn - sim_plr_jn          # coeficiente de disposicao jan-nov
cd_dz = sim_pgr_dz - sim_plr_dz          # dezembro
virada = cd_jn - cd_dz                   # de quanto o comportamento vira
virada_med = np.median(virada)
virada_ic = np.percentile(virada, [2.5, 97.5])
p_dez_inverte = np.mean(cd_dz < 0)

razao_jn = np.median(sim_pgr_jn / sim_plr_jn)
razao_dz = np.median(sim_pgr_dz / sim_plr_dz)

# ---------------------------------------------------------------
# Traducoes
# ---------------------------------------------------------------
us_por_100 = (us_pgr_ano * 100, us_plr_ano * 100)
br_por_100 = (br_pgr * 100, br_ppr * 100)
custo_252 = exc_vendidos[1] - exc_mantidos[1]

print("=" * 68)
print("  CAMADA 1 - A ASSIMETRIA, COM INCERTEZA (EUA, 1987-1993)")
print("=" * 68)
print(f"\n  Proporcao de ganhos realizados:  {us_pgr_ano*100:.1f}%")
print(f"  Proporcao de perdas realizadas:  {us_plr_ano*100:.1f}%")
print(f"  Razao entre elas:                {us_razao_med:.3f}")
print(f"  IC 95% da razao:                 [{us_razao_ic[0]:.3f}, {us_razao_ic[1]:.3f}]")
print(f"\n  Em 100 oportunidades de vender, o investidor usa:")
print(f"  -> {us_por_100[0]:.0f} das que estao no lucro")
print(f"  -> {us_por_100[1]:.0f} das que estao no prejuizo")
print(f"\n  Sensibilidade a dependencia (so 1 em cada d observacoes informa):")
for d in deffs:
    z = sens[d]
    print(f"  -> d = {d:>4}: razao EUA [{z['us_ic'][0]:.2f}, {z['us_ic'][1]:.2f}], "
          f"P(razao>1) = {z['us_p1']*100:.1f}%")
print(f"\n  Teste alternativo do proprio artigo, por conta (1.893 contas):")
print(f"  -> PGR {us_conta_pgr:.2f} contra PLR {us_conta_plr:.2f}, t = {us_conta_t}")
print(f"  -> medido em dinheiro: {us_dolar_pgr:.2f} contra {us_dolar_plr:.2f}, t = {us_dolar_t}")

print("\n" + "=" * 68)
print("  CAMADA 2 - O ENCONTRO DOS DOIS MERCADOS")
print("=" * 68)
print(f"\n  EUA 1987-1993, 10.000 contas:     razao {us_razao_med:.3f} "
      f"[{us_razao_ic[0]:.3f}, {us_razao_ic[1]:.3f}]")
print(f"  BRASIL 2012-2014, mercado inteiro: razao {br_razao_med:.3f} "
      f"[{br_razao_ic[0]:.3f}, {br_razao_ic[1]:.3f}]")
print(f"  Diferenca entre as duas razoes:    {dif_med:.3f} "
      f"[{dif_ic[0]:.3f}, {dif_ic[1]:.3f}]")
print(f"\n  ATENCAO: intervalo que atravessa o zero NAO prova igualdade.")
print(f"  Teste de equivalencia: P(|diferenca| < {MARGEM_EQUIV:.2f}), com o mesmo")
print(f"  desconto de dependencia aplicado aos dois paises:")
for d in deffs:
    z = sens[d]
    print(f"  -> d = {d:>4}: EUA [{z['us_ic'][0]:.2f}, {z['us_ic'][1]:.2f}]  "
          f"Brasil [{z['br_ic'][0]:.2f}, {z['br_ic'][1]:.2f}]  "
          f"P(equivalentes) = {z['p_equiv']*100:.0f}%")
print(f"\n  Leitura: com dependencia leve ou moderada, as razoes sao")
print(f"  equivalentes. Com dependencia forte, o dado americano fica largo")
print(f"  demais para afirmar equivalencia; so da para dizer compativeis.")
print(f"\n  Relativo contra absoluto:")
print(f"  -> EUA:    {us_dif_abs:.1f} pontos percentuais de diferenca por oportunidade")
print(f"  -> Brasil: {br_dif_abs:.1f} ponto percentual de diferenca por oportunidade")
print(f"  A razao de 1,5 e relativa. A distancia absoluta e pequena no Brasil.")
print(f"\n  Em 100 oportunidades, a pessoa fisica brasileira usa:")
print(f"  -> {br_por_100[0]:.1f} das que estao no lucro")
print(f"  -> {br_por_100[1]:.1f} das que estao no prejuizo")
print(f"\n  ATENCAO: os NIVEIS nao sao comparaveis entre os dois estudos")
print(f"  (dependem do tamanho da carteira e da frequencia de observacao).")
print(f"  O que se compara e a RAZAO entre eles.")
print(f"\n  Coeficiente de disposicao por tipo de investidor no Brasil:")
for nome, pgr, ppr, cd in br_tipos:
    marca = "sim" if cd > 0 else "nao"
    print(f"  -> {nome:<28} CD = {cd:+.4f}   efeito: {marca}")
print(f"\n  Retorno medio por venda: pessoa fisica {br_retorno_medio['Pessoa fisica']:.2f}%, "
      f"fundos mutuos {br_retorno_medio['Fundos mutuos']:.2f}%, "
      f"previdencia {br_retorno_medio['Fundos de previdencia']:.2f}%")

print("\n" + "=" * 68)
print("  CAMADA 3 - DEZEMBRO, O MES EM QUE A REGRA VIRA (EUA)")
print("=" * 68)
print(f"\n  Janeiro a novembro: razao {razao_jn:.2f}  (vende mais o que subiu)")
print(f"  Dezembro:           razao {razao_dz:.2f}  (vende mais o que caiu)")
print(f"  Virada no coeficiente de disposicao: {virada_med*100:.1f} p.p. "
      f"[{virada_ic[0]*100:.1f}, {virada_ic[1]*100:.1f}]")
print(f"  P(o coeficiente fica negativo em dezembro): {p_dez_inverte*100:.1f}%")
print(f"\n  A mesma pessoa, a mesma carteira. O que muda e o calendario")
print(f"  do imposto. O comportamento nao e um traco fixo.")

print("\n" + "=" * 68)
print("  O QUE ISSO CUSTOU (EUA)")
print("=" * 68)
print(f"\n  Retorno em excesso depois da decisao:")
for h, v, m, p in zip(horizontes, exc_vendidos, exc_mantidos, exc_p):
    print(f"  -> {h:>3} dias: vendidos {v*100:+.2f}% contra mantidos {m*100:+.2f}%"
          f"  (diferenca {(v-m)*100:+.2f} p.p., p = {p})")
print(f"\n  Em um ano de bolsa, {custo_252*100:.2f} pontos percentuais separam")
print(f"  o que foi vendido do que foi mantido. Na direcao errada.")
print("=" * 68)
```

### O Insight

**1,52 nos Estados Unidos. 1,50 no Brasil.**

Uma ação que subiu tem cerca de uma vez e meia mais chance de ser vendida
do que uma que caiu. O primeiro número vem de 10 mil contas americanas
entre 1987 e 1993. O segundo vem do mercado de ações brasileiro inteiro,
entre 2012 e 2014, com cerca de 60 milhões de operações. Dois países, duas
décadas, dois desenhos de mercado, e a diferença entre as duas razões é de
0,02. Um intervalo que atravessa o zero não prova igualdade, então o modelo
faz um teste de equivalência: com dependência leve ou moderada entre as
decisões, as duas razões ficam dentro de uma margem de 0,10 uma da outra.
Com dependência forte, só dá para dizer que são compatíveis.

Não é opinião sobre dinheiro. É contagem de botão apertado.

Um cuidado de leitura: "uma vez e meia" é uma comparação relativa. Em pontos
percentuais, a distância é de cerca de 5 pontos nos Estados Unidos e de 1
ponto no Brasil, porque lá cada oportunidade é contada por dia de carteira.
O padrão é o mesmo; o tamanho absoluto depende de como se conta.

E tem uma coisa que separa quem faz isso de quem não faz. No Brasil, o
efeito é claro na pessoa física e **desaparece nos institucionais**: fundos
mútuos, fundos de previdência e bancos de investimento apresentam
coeficiente negativo, ou seja, aceitam a perda com mais facilidade do que
realizam o ganho. Os mesmos que, na mesma base, tiveram os maiores retornos
médios por operação. A pessoa física teve o menor.

O detalhe mais bonito está em dezembro. Nos dados americanos, a regra
inverte: naquele mês o investidor vende mais o que caiu do que o que subiu,
porque o prejuízo abate imposto e o ano está acabando. A virada é de quase
8 pontos percentuais no coeficiente.

A mesma pessoa. A mesma carteira. O que mudou foi o calendário.

Isso desmonta a leitura preguiçosa de que segurar o perdedor é um traço de
personalidade. Não é. É uma conta que só é feita quando existe um prazo e
uma licença para fazê-la.

A pergunta que fica:

*O que você está segurando só porque desfazer significaria admitir que já
não vale o que custou?*

### Limitações do Modelo

- **Os níveis não são comparáveis entre os dois estudos.** Nos Estados
  Unidos as proporções são 14,8% e 9,8%; no Brasil, 3,4% e 2,2%. Isso não
  significa que o brasileiro negocie sete vezes menos: as proporções
  dependem do tamanho da carteira e da frequência com que as oportunidades
  são contadas. O que se compara aqui é a razão entre elas, e só ela.
- **Independência, nos dois artigos.** O teste z de Odean e o qui-quadrado
  de Prates et al. tratam cada ganho ou perda em carteira como observação
  independente. Não são: a mesma posição aparece dia após dia, e no Brasil
  cada dia de carteira gera uma observação. Odean avisa que isso infla a
  estatística de teste. O modelo refaz as posteriores dos dois países
  dividindo as contagens por 10, 100 e 1.000, e reporta também o teste do
  próprio Odean feito por conta (1.893 contas, t = 19).
- **"Igual" exige teste de equivalência.** Diferença com intervalo que passa
  pelo zero é ausência de diferença detectada, não prova de igualdade. O
  modelo calcula a probabilidade de as razões ficarem a menos de 0,10 uma da
  outra. Sob dependência forte essa probabilidade cai bastante, e o texto
  diz isso.
- **Relativo contra absoluto.** A razão de 1,5 é relativa. A diferença
  absoluta é de cerca de 5 pontos percentuais nos Estados Unidos e de 1,1
  ponto no Brasil. E com dezenas de milhões de observações, tudo sai
  estatisticamente significativo, inclusive coeficientes institucionais de
  −0,0004: significância, ali, não informa tamanho.
- **O aviso do próprio Odean sobre os meses.** Ele publica a razão mês a mês
  só como figura, e escreve que, nas partições em que os testes se aproximam
  do limiar de significância, os resultados devem ser lidos com ceticismo. Por
  isso o gráfico do calendário mostra apenas janeiro e dezembro, que estão no
  texto, e a média de janeiro a novembro calculada sobre as contagens.
- **Construção da carteira.** Nos dois estudos, o preço de compra é uma
  proxy do ponto de referência, e operações anteriores ao início da base
  ficam de fora. No estudo brasileiro os dados são saldo diário por
  investidor, com preço médio por FIFO, o que aproxima e não reproduz cada
  ordem individual.
- **Seleção.** A base americana é de uma corretora de desconto, com viés de
  sobrevivência declarado pelo autor. A base brasileira é o mercado inteiro,
  mas cobre dois anos e dez meses, e descarta compras sem venda
  correspondente dentro da janela. Isso tende a tirar da base justamente as
  posições seguradas até o fim, os perdedores esquecidos, o que provavelmente
  **subestima** o efeito no Brasil.
- **O custo é associação, não experimento.** A diferença de 3,41 pontos
  percentuais em um ano entre vendidos e mantidos é medida ex post, e parte
  dela é atribuída a momento de preço, não ao erro do investidor.
- **A explicação é disputada.** A teoria do prospecto organiza bem a
  história, mas há trabalhos mostrando que ela só prevê o efeito em
  condições específicas. O comportamento é sólido. O porquê, menos.

*A ciência é honesta sobre o que não sabe. O modelo também.*
