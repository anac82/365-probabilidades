### 📖 A História

Tem uma coisa que quase ninguém percebe acontecendo, porque ela acontece devagar
demais.

Em algum momento, entre os vinte e poucos e os trinta e poucos, a agenda para de
crescer. Não tem um dia em que isso é anunciado. Simplesmente, um ano você conheceu
cinco pessoas novas e no seguinte conheceu duas, e depois nenhuma, e ninguém reparou.

A gente costuma explicar isso com falta de tempo, filho pequeno, trabalho, mudança de
cidade. Cada um tem a sua versão, e todas soam individuais.

O detalhe é que a curva é a mesma para quase todo mundo, e ela tem forma conhecida.

---

### 📚 O Conceito: rede não é uma coisa só

O erro mais comum ao pensar em vida social é tratar tudo como um bloco: quantas pessoas
eu conheço.

A pesquisa separa isso em camadas diferentes, e é aí que fica interessante, porque
**cada camada tem uma trajetória própria ao longo da vida**.

A **rede global** é o total de pessoas com quem se tem contato.

A **rede pessoal** é o grupo de quem de fato importa, com quem se conversa sobre coisas
que importam.

A **rede de amizade** são os amigos, especificamente.

E a **rede familiar** são os laços de parentesco.

Somar as quatro numa curva só esconde exatamente o que o dia tem de interessante,
porque elas não caminham juntas.

### 🧮 O Modelo

Uma meta-análise grande de estudos sobre mudança de rede social ao longo da vida.

**Fontes:**
- Wrzus, C., Hänel, M., Wagner, J. & Neyer, F. J., 2013 · "Social network changes and
  life events across the life span: a meta-analysis" · *Psychological Bulletin* 139(1),
  53-80 · **277 estudos**, **N=177.635 participantes**, da adolescência à velhice
- **Rede global:** cresce até a **vida adulta jovem** e **decresce continuamente**
  daí em diante
- **Rede pessoal e rede de amizade:** **decrescem ao longo da vida adulta**
- **Rede familiar:** **estável em tamanho da adolescência à velhice**
- **Redes de colegas de trabalho e de vizinhos:** relevantes apenas em **faixas etárias
  específicas**, e não ao longo de toda a vida

**Nota metodológica sobre o fator ×0.80:** não se aplica. Meta-análise de medidas de
tamanho de rede social, e não proporção de survey autorrelatado sobre comportamento.


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Wrzus, Hanel, Wagner & Neyer, 2013, Psychological Bulletin 139(1), 53-80

n_estudos = 277
n_participantes = 177_635
faixa = 'da adolescencia a velhice'

# Trajetoria de cada tipo de rede, conforme publicado
TRAJETORIAS = {
    'Rede global': 'cresce ate a vida adulta jovem e decresce continuamente',
    'Rede pessoal': 'decresce ao longo da vida adulta',
    'Rede de amizade': 'decresce ao longo da vida adulta',
    'Rede familiar': 'ESTAVEL em tamanho da adolescencia a velhice',
    'Colegas de trabalho': 'relevante apenas em faixas etarias especificas',
    'Vizinhos': 'relevante apenas em faixas etarias especificas',
}

aplica_fator_080 = False

print("=" * 72)
print("  DADOS - COMO A REDE SOCIAL MUDA AO LONGO DA VIDA")
print("=" * 72)
print(f"\n  Wrzus, Hanel, Wagner & Neyer, 2013 (Psych Bulletin 139(1), 53-80):")
print(f"  -> {n_estudos} estudos · N = {n_participantes:,}".replace(",", ".") +
      f" · {faixa}")
print(f"\n  {'tipo de rede':<22} trajetoria")
print(f"  {'-'*22} {'-'*48}")
for nome, traj in TRAJETORIAS.items():
    print(f"  {nome:<22} {traj}")

decrescem = [k for k, v in TRAJETORIAS.items() if 'decresce' in v]
estaveis = [k for k, v in TRAJETORIAS.items() if 'ESTAVEL' in v]
print(f"\n  -> Redes que ENCOLHEM na vida adulta: {len(decrescem)}")
print(f"     ({', '.join(decrescem)})")
print(f"  -> Redes ESTAVEIS: {len(estaveis)} ({', '.join(estaveis)})")
print(f"\n  Fator x0.80 aplicado: {aplica_fator_080}")
print("=" * 72)
```

```python
# --- O MODELO ---
# A meta-analise publica a FORMA das trajetorias, nao uma equacao com parametros.
# Entao a assinatura deste dia e a TAXA DE MANUTENCAO: se a rede encolhe,
# quantos vinculos novos por ano seriam necessarios para manter o tamanho?

PICO = 30          # "vida adulta jovem". PREMISSA DECLARADA.
INICIO = 15        # inicio da janela de formacao de rede
FIM = 80           # horizonte de vida adulta considerado

# 1) A aritmetica da janela
anos_ate_pico = PICO - INICIO
anos_depois = FIM - PICO
prop_depois = anos_depois / (FIM - INICIO)

grade_pico = [25, 28, 30, 33, 35]

# 2) ASSINATURA: taxa de reposicao necessaria.
# Premissa declarada: a rede de amizade cai de forma aproximadamente constante
# ao longo da vida adulta, perdendo uma fracao F do tamanho do pico ate os 80.
# Testamos varios valores de F, porque a meta-analise nao publica a magnitude.
TAMANHO_PICO = 20          # amigos no pico. VALOR ILUSTRATIVO DECLARADO.
FRACOES = [0.40, 0.50, 0.60, 0.70]   # quanto da rede se perde ate os 80

print("=" * 72)
print("  MODELO - A ARITMETICA DA JANELA E A TAXA DE MANUTENCAO")
print("=" * 72)
print(f"\n  PREMISSA DECLARADA: 'vida adulta jovem' tratada como {PICO} anos,")
print(f"  janela de formacao dos {INICIO} aos {PICO} e horizonte ate {FIM}.")
print(f"  A meta-analise publica a FORMA da curva, nao a idade exata do pico")
print(f"  nem o tamanho das redes.")

print(f"\n  1) O tamanho da janela de crescimento:")
print(f"  -> Cresce por cerca de {anos_ate_pico} anos, dos {INICIO} aos {PICO}.")
print(f"  -> Decresce por cerca de {anos_depois} anos, dos {PICO} aos {FIM}.")
print(f"  -> Ou seja, {prop_depois*100:.0f}% da vida adulta acontece DEPOIS do pico.")

print(f"\n  2) Sensibilidade a premissa do pico:")
for p_ in grade_pico:
    depois = (FIM - p_) / (FIM - INICIO)
    print(f"  -> Se o pico fosse aos {p_}: {depois*100:.0f}% da vida vem depois")
print(f"  -> A conclusao qualitativa nao muda em nenhum ponto da faixa.")

print(f"\n  3) ASSINATURA: quantos vinculos por ano para NAO encolher?")
print(f"     Tamanho no pico assumido: {TAMANHO_PICO} amigos (valor ilustrativo).")
print(f"\n  {'perda ate os 80':>16} | {'amigos perdidos':>16} | {'por ano':>9} | "
      f"{'1 novo a cada':>14}")
print(f"  {'-'*16}-+-{'-'*16}-+-{'-'*9}-+-{'-'*14}")
for f in FRACOES:
    perdidos = TAMANHO_PICO * f
    por_ano = perdidos / anos_depois
    meses = 12 / por_ano
    print(f"  {f*100:>15.0f}% | {perdidos:>16.1f} | {por_ano:>9.2f} | "
          f"{meses:>10.0f} meses")

f_central = 0.55
perdidos_c = TAMANHO_PICO * f_central
por_ano_c = perdidos_c / anos_depois
meses_c = 12 / por_ano_c

print(f"\n  -> Num cenario central, com perda de {f_central*100:.0f}% ate os {FIM}:")
print(f"     seria preciso um vinculo novo a cada {meses_c:.0f} meses")
print(f"     apenas para MANTER o tamanho da rede.")
print(f"  -> Nao para aumentar. Para ficar parado.")

print(f"\n  4) A leitura:")
print(f"  -> A taxa parece pequena, e e justamente esse o problema.")
print(f"  -> Um vinculo a cada quatro ou cinco anos e algo que ninguem")
print(f"     percebe que deixou de fazer.")
print(f"  -> O encolhimento nao acontece por um evento. Acontece por omissao.")

print(f"\n  5) E o que NAO encolhe:")
print(f"  -> A rede familiar fica estavel em tamanho da adolescencia a velhice.")
print(f"  -> Ela e a unica que nao exige reposicao para continuar existindo.")
print("=" * 72)
```

```python
# --- VISUALIZACAO ---

DOURADO = '#c8a84b'
VERMELHO = '#c0392b'
VERDE = '#2a8a82'
CINZA = '#6b6a64'

# GRAFICO 1 - As trajetorias, em forma esquematica
fig1, ax1 = plt.subplots(figsize=(12, 8))

idade = np.linspace(15, 80, 500)


def curva_global(x):
    # cresce ate o pico e decresce continuamente depois
    return np.where(x <= PICO,
                    0.45 + 0.55 * (x - 15) / (PICO - 15),
                    1.0 - 0.55 * (x - PICO) / (80 - PICO))


def curva_amizade(x):
    # decresce ao longo da vida adulta
    return np.where(x <= 22, 0.85 + 0.15 * (x - 15) / 7,
                    1.0 - 0.62 * (x - 22) / (80 - 22))


def curva_familia(x):
    return np.full_like(x, 0.72)


ax1.plot(idade, curva_global(idade) * 100, color=DOURADO, linewidth=3.4,
         label='Rede global')
ax1.plot(idade, curva_amizade(idade) * 100, color=VERMELHO, linewidth=3.4,
         label='Rede de amizade')
ax1.plot(idade, curva_familia(idade) * 100, color=VERDE, linewidth=3.4,
         label='Rede familiar')

ax1.axvline(x=PICO, color=CINZA, linestyle=':', linewidth=2)
ax1.text(PICO + 0.8, 104, 'vida adulta jovem', fontsize=12, color=CINZA)

ax1.set_xlim(15, 80)
ax1.set_ylim(0, 115)
ax1.set_yticks([])
ax1.set_xlabel('Idade (anos)')
ax1.set_ylabel('Tamanho da rede (escala relativa)')
ax1.legend(frameon=False, fontsize=13, loc='lower left')
ax1.set_title('Três redes, três trajetórias diferentes\n'
              'Wrzus, Hänel, Wagner & Neyer, 2013 · ' + f'{n_estudos} estudos · N=' +
              f'{n_participantes:,}'.replace(",", "."),
              fontsize=15, pad=18)
ax1.text(0.5, -0.13,
         'Formas esquemáticas das trajetórias descritas na meta-análise. '
         'Não são curvas ajustadas publicadas, e a escala é relativa.',
         transform=ax1.transAxes, ha='center', fontsize=11, color=CINZA, style='italic')

plt.figtext(0.5, 0.005,
            'Fonte: Wrzus et al., 2013, Psychological Bulletin 139(1), 53-80'
            '  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia083_g1_trajetorias.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 1 salvo")

# GRAFICO 2 - Assinatura: a taxa de manutencao
fig2, ax2 = plt.subplots(figsize=(12, 8))

fracoes_plot = np.linspace(0.30, 0.80, 200)
meses_plot = [12 / ((TAMANHO_PICO * f) / anos_depois) for f in fracoes_plot]

ax2.plot(fracoes_plot * 100, meses_plot, color=DOURADO, linewidth=3.4)
ax2.fill_between(fracoes_plot * 100, meses_plot, alpha=0.12, color=DOURADO)

for f in FRACOES:
    m = 12 / ((TAMANHO_PICO * f) / anos_depois)
    ax2.scatter([f * 100], [m], s=170, color=DOURADO, zorder=3)
    ax2.text(f * 100, m + 1.1, f'{m:.0f} meses', ha='center', fontsize=13,
             fontweight='bold', color='#333')

m_c = 12 / ((TAMANHO_PICO * f_central) / anos_depois)
ax2.scatter([f_central * 100], [m_c], s=260, color=VERMELHO, zorder=4)
ax2.axvline(x=f_central * 100, color=VERMELHO, linestyle=':', linewidth=2)
ax2.text(f_central * 100 + 1.2, m_c + 4.5,
         'cenário central\num vínculo novo\na cada ' + f'{m_c:.0f}' + ' meses',
         fontsize=13, fontweight='bold', color=VERMELHO)

ax2.set_xlim(29, 81)
ax2.set_ylim(0, max(meses_plot) * 1.25)
ax2.set_xlabel('Quanto da rede de amizade se perde entre o pico e os 80 anos (%)')
ax2.set_ylabel('Intervalo entre vínculos novos, só para manter o tamanho (meses)')
ax2.set_title('A assinatura estatística: a taxa de manutenção\n'
              'Quantos vínculos novos por ano seriam necessários para a rede não '
              'encolher',
              fontsize=15, pad=18)
ax2.text(0.5, -0.135,
         f'Premissa declarada: {TAMANHO_PICO} amigos no pico, perda aproximadamente '
         f'constante ao longo de {anos_depois} anos.\nA meta-análise publica a forma '
         'da trajetória, não o tamanho das redes. A faixa testada cobre de 30% a 80% '
         'de perda.',
         transform=ax2.transAxes, ha='center', fontsize=11, color=CINZA, style='italic')

plt.figtext(0.5, 0.005,
            'Cálculo próprio sobre a forma de trajetória publicada'
            '  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia083_g2_manutencao.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 2 salvo")

# GRAFICO 3 - O que encolhe e o que nao encolhe
fig3, ax3 = plt.subplots(figsize=(12, 8))

nomes = list(TRAJETORIAS.keys())
y = np.arange(len(nomes))[::-1]
for yi, nome in zip(y, nomes):
    traj = TRAJETORIAS[nome]
    if 'ESTAVEL' in traj:
        cor, marca = VERDE, 'estável'
    elif 'decresce' in traj:
        cor, marca = VERMELHO, 'encolhe'
    else:
        cor, marca = CINZA, 'só em certas idades'
    largura = 1.0 if marca != 'só em certas idades' else 0.45
    ax3.barh([yi], [largura], color=cor, alpha=0.85, height=0.55)
    ax3.text(largura + 0.05, yi, marca, va='center', fontsize=14, color=cor,
             fontweight='bold')

ax3.set_yticks(y)
ax3.set_yticklabels(nomes, fontsize=13)
ax3.set_xlim(0, 1.75)
ax3.set_xticks([])
ax3.grid(False)
ax3.set_title('O que encolhe, e o que não encolhe\n'
              'Uma única rede fica estável da adolescência à velhice',
              fontsize=15, pad=18)
ax3.text(0.5, -0.1,
         'A rede familiar não cresce, mas também não cai. É a única com essa '
         'trajetória entre as seis medidas.',
         transform=ax3.transAxes, ha='center', fontsize=12, color=VERDE,
         fontweight='bold')

plt.figtext(0.5, 0.005,
            'Fonte: Wrzus et al., 2013  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia083_g3_o_que_fica.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 3 salvo")

# --- CONFIRMACAO: onde os arquivos foram salvos ---
import os
print("\n" + "=" * 72)
print("  ARQUIVOS GERADOS")
print("=" * 72)
print("  Pasta de trabalho:", os.getcwd())
for f in sorted(os.listdir()):
    if f.startswith("dia083_") and f.endswith(".png"):
        print(f"  -> {f}  ({os.path.getsize(f)/1024:.0f} KB)")
print("=" * 72)
```

### 💡 O Insight

Duzentos e setenta e sete estudos. Cento e setenta e sete mil seiscentas e trinta e
cinco pessoas. Da adolescência à velhice.

**A rede social global cresce até a vida adulta jovem e decresce continuamente daí em
diante.**

E a rede de amizades, especificamente, **decresce ao longo de toda a vida adulta**.

Não é uma fase ruim. Não é o seu ano difícil. É a forma da curva, e ela é a mesma para
quase todo mundo.

Fiz a aritmética disso, e ela é desconfortável.

Se a rede cresce até por volta dos trinta e cai depois, então a fase de **crescimento**
dura cerca de quinze anos, e a fase de **encolhimento** dura os outros cinquenta.

**Setenta e sete por cento da vida adulta acontece depois do pico.**

Testei essa conta com o pico em 25, 28, 30, 33 e 35 anos. A proporção fica entre 69% e
85%, e a conclusão não muda em nenhum ponto da faixa.

Mas o número que eu achei mais revelador foi outro, e ele responde uma pergunta
prática: **quantos vínculos novos por ano seriam necessários apenas para a rede não
encolher?**

Assumindo vinte amigos no pico, valor ilustrativo declarado, e uma perda de cerca de
55% ao longo dos cinquenta anos seguintes, a resposta é **um vínculo novo a cada 55
meses**.

Menos de um a cada quatro anos. Só para ficar parado.

E é justamente por ser tão pouco que ninguém percebe. O encolhimento não acontece por
um evento, por uma briga ou por uma mudança de cidade. Acontece por omissão, numa taxa
baixa demais para ser notada.

Agora a parte que salva o dia de ser apenas triste.

**Entre as seis redes medidas, uma fica estável da adolescência à velhice: a familiar.**

Ela não cresce. Mas também não cai.

Enquanto tudo o mais encolhe, e enquanto os colegas de trabalho e os vizinhos aparecem
e somem conforme a fase, a rede de parentesco fica onde está.

Isso me parece uma informação prática, e não um consolo.

Se a maior parte da sua vida adulta será vivida com a rede encolhendo, o esforço de
manter alguém não é romantismo: é o único jeito de a conta não fechar em zero. E se a
única camada que resiste sozinha é a familiar, talvez valha reparar em quem já está
ali sem precisar ser convidado.

A turma provavelmente já está formada. O que ainda dá para escolher é quem fica.

*Quem entrou na sua vida depois dos trinta, e como isso aconteceu?*

### ⚠️ Limitações do Modelo

- **A meta-análise publica a forma das trajetórias, não uma equação com parâmetros.**
  As curvas do gráfico 1 são esquemáticas, com escala relativa, e representam as
  trajetórias descritas. Não são curvas ajustadas publicadas.
- **A idade exata do pico não é publicada como número único.** Tratei "vida adulta
  jovem" como 30 anos, que é premissa declarada, e testei de 25 a 35 anos com análise
  de sensibilidade.
- **A taxa de manutenção é cálculo próprio, com premissa declarada.** O tamanho de
  vinte amigos no pico é valor ilustrativo, e a perda de 55% é o ponto central de uma
  faixa testada de 30% a 80%. A meta-análise publica a forma da trajetória, não o
  tamanho das redes.
- **Tamanho de rede não é qualidade de rede.** Uma rede menor pode ser melhor, e nada
  aqui mede satisfação, intimidade ou apoio recebido.
- **Meta-análise de estudos majoritariamente transversais**, o que significa que parte
  do que parece mudança com a idade pode ser diferença entre gerações.
- **A rede familiar estável em tamanho não quer dizer estável em qualidade.** Laços de
  parentesco continuam existindo mesmo quando estão ruins ou distantes, e isso conta
  como rede.
- **Dados de 2013**, anteriores à consolidação das redes sociais digitais no formato
  atual. O que conta como contato pode ter mudado desde então.
- Fator ×0.80 não aplicado.

*A ciência é honesta sobre o que não sabe. O modelo também.*
