### 📖 A História

Escrevo isto de outro fuso horário, com a sensação familiar de estar acordada na hora
errada.

Quem viaja muito repete uma coisa que soa a superstição: **ir para leste é pior que ir
para oeste.** Mesmo número de fusos, mesma duração de voo, e mesmo assim um lado cobra
mais caro que o outro.

Isso parece o tipo de frase que a gente inventa para explicar um mal-estar qualquer.
Só que dessa vez a superstição tem mecanismo, tem modelo matemático, e tem uma tabela
com onze mil jogos de basquete confirmando.

E a raiz de tudo é um número esquisito que quase ninguém conhece.

---

### 📚 O Conceito: 24,2 horas

O corpo humano tem um relógio interno, no núcleo supraquiasmático, e quando ele é
medido em isolamento, sem sol, sem relógio e sem pistas sociais, ele **não roda em 24
horas**. Roda em torno de **24,2 horas**, com estimativas na literatura indo de 23,8 a
24,6.

Parece detalhe. Não é.

Como o relógio roda um pouco **mais lento** que o dia real, ele tem uma tendência
natural a atrasar. Todo dia a luz da manhã o empurra de volta uns minutos para o
lugar. É por isso que ficar acordado até mais tarde é fácil e acordar mais cedo é uma
briga.

Agora aplique isso a uma viagem.

Voar para **oeste** alonga o seu dia. Você precisa **atrasar** o relógio, que é
justamente para onde ele já quer ir.

Voar para **leste** encurta o seu dia. Você precisa **adiantar** o relógio, contra a
tendência dele.

A assimetria do jet lag não é psicológica. É consequência aritmética de o período ser
maior que 24.

---

### 🧮 O Modelo

Um mecanismo fisiológico, um modelo matemático e um teste empírico grande.

**Fontes:**
- **Período circadiano endógeno:** estimativas do ritmo humano em livre curso ficam
  entre **23,8 e 24,6 horas**, com a média em torno de **24,2**, e a maioria acima de
  24. Ajustar-se a um dia mais curto (avanço de fase) é tipicamente mais difícil do
  que a um dia mais longo (atraso de fase).
- Lu, Z., Klein-Cardeña, K., Lee, S., Antonsen, T. M., Girvan, M. & Ott, E., 2016 ·
  *Chaos* 26(9), 094811 · modelo matemático de células do núcleo supraquiasmático ·
  recuperação de viagem para leste cerca de **50% mais longa** que a equivalente para
  oeste
- Leota, J., Hoffman, D., Czeisler, M., Mascaro, L., Drummond, S. P. A., Anderson, C.,
  Rajaratnam, S. M. W. & Facer-Childs, E. R., 2022 · *Frontiers in Physiology* 13,
  892681 · **11.481 jogos da NBA**, dez temporadas de 2011/2012 a 2020/2021 · modelos
  mistos multiníveis, fator fixo de direção do jet lag e fatores aleatórios de time,
  adversário e horário do jogo
- **Resultados nos times da casa**, para jet lag a leste, mas não a oeste:
  vitórias **Δ = −6,03% (p = 0,051, marginal)**, saldo de pontos **Δ = −1,29
  (p = 0,015)**, saldo de rebotes **Δ = −1,29 (p < 0,0001)**, diferencial de
  aproveitamento efetivo **Δ = −1,2% (p < 0,01)**
- **Dose:** com 2 horas a leste, saldo de pontos **Δ = −4,53 (p < 0,05)**; com 1 hora,
  **Δ = −0,72 (p = 0,07)**
- **Tradução dos autores:** os 6,03% equivalem a **2,47 vitórias a menos em casa** numa
  temporada de 41 jogos em casa
- **Sem associação para os times visitantes**, o que os autores atribuem ao ambiente
  controlado das viagens de time
- **Literatura divergente:** McHill e Chinoy, 2020, encontraram associação negativa
  para viagem a **oeste** na NBA; Winter e colegas, 2009, encontraram associação
  **positiva** para leste no beisebol

**Referência interna:** o **#047** tratou de percepção subjetiva de tempo, que é objeto
diferente. Aqui o assunto é relógio biológico, não sensação de duração.

**Nota metodológica sobre o fator ×0.80:** não se aplica. Registro objetivo de jogos e
de desfechos esportivos, sem autorrelato.


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---

# Periodo circadiano endogeno humano
periodo_medio = 24.2
periodo_faixa = (23.8, 24.6)

# Lu et al., 2016, Chaos 26(9), 094811
recuperacao_leste_a_mais = 0.50   # ~50% mais longa que a equivalente a oeste

# Leota et al., 2022, Frontiers in Physiology 13, 892681
n_jogos      = 11_481
n_temporadas = 10
jogos_casa   = 41                 # por time, por temporada

# Efeitos nos times DA CASA, jet lag a leste (nao a oeste)
EFEITOS_CASA = {
    'Percentual de vitorias':          (-6.03, '%',       0.051),
    'Saldo de pontos':                 (-1.29, ' pontos', 0.015),
    'Saldo de rebotes':                (-1.29, ' rebotes', 0.0001),
    'Aproveitamento efetivo (dif.)':   (-1.20, '%',       0.01),
}

# Dose: magnitude do jet lag a leste e saldo de pontos
DOSE = {1: (-0.72, 0.07), 2: (-4.53, 0.05)}

# Times visitantes: nenhuma associacao significativa
visitantes_significativo = False

aplica_fator_080 = False

print("=" * 70)
print("  DADOS - O RELOGIO DE 24,2 HORAS E O PRECO DA DIRECAO")
print("=" * 70)
print(f"\n  Periodo circadiano endogeno humano:")
print(f"  -> Media em torno de {periodo_medio} h")
print(f"  -> Faixa das estimativas: {periodo_faixa[0]} a {periodo_faixa[1]} h")
print(f"  -> Como roda ACIMA de 24, atrasar e facil e adiantar e dificil")
print(f"\n  Lu et al., 2016 (modelo matematico do nucleo supraquiasmatico):")
print(f"  -> Recuperacao a leste cerca de"
      f" {recuperacao_leste_a_mais*100:.0f}% mais longa que a oeste")
print(f"\n  Leota et al., 2022 ({n_jogos:,}".replace(",", ".") +
      f" jogos, {n_temporadas} temporadas):")
print(f"  Efeitos nos times DA CASA, jet lag a leste:")
for nome, (delta, unidade, p) in EFEITOS_CASA.items():
    marca = "significativo" if p < 0.05 else "MARGINAL"
    print(f"  -> {nome:<32} {delta:>6.2f}{unidade:<9}"
          f" p = {p:<7} {marca}")
print(f"\n  Dose (saldo de pontos por hora de jet lag a leste):")
for horas, (delta, p) in DOSE.items():
    print(f"  -> {horas} h: {delta:>6.2f} pontos   p = {p}")
print(f"\n  Times visitantes: associacao significativa? {visitantes_significativo}")
print(f"\n  Fator x0.80 aplicado: {aplica_fator_080}")
print("=" * 70)
```

```python
# --- O MODELO ---
# 1) A aritmetica do relogio: quanto ele precisa se mover, por dia, e para onde.

deriva_diaria_min = (periodo_medio - 24) * 60      # minutos que o relogio atrasa por dia
deriva_faixa_min = [(p - 24) * 60 for p in periodo_faixa]

# 2) A traducao dos 6,03% em vitorias, feita pelos autores
delta_vitorias_pct = EFEITOS_CASA['Percentual de vitorias'][0]
vitorias_perdidas = abs(delta_vitorias_pct) / 100 * jogos_casa

# 3) O que e significativo e o que nao e
significativos = [n for n, (_, _, p) in EFEITOS_CASA.items() if p < 0.05]
marginais = [n for n, (_, _, p) in EFEITOS_CASA.items() if p >= 0.05]

# 4) Dose: quanto piora por hora adicional
horas = sorted(DOSE)
por_hora = (DOSE[2][0] - DOSE[1][0]) / (2 - 1)

print("=" * 70)
print("  MODELO - POR QUE UM LADO CUSTA MAIS QUE O OUTRO")
print("=" * 70)
print(f"\n  1) A aritmetica do relogio:")
print(f"  -> Periodo de {periodo_medio} h significa que o relogio atrasa")
print(f"     cerca de {deriva_diaria_min:.0f} minutos por dia se ficar solto.")
print(f"  -> A faixa das estimativas vai de {deriva_faixa_min[0]:.0f} minutos"
      f" (relogio ADIANTA) a +{deriva_faixa_min[1]:.0f} minutos (ATRASA).")
print(f"     A maioria das pessoas fica do lado que atrasa.")
print(f"  -> Viajar para OESTE pede atraso: o relogio ja quer isso.")
print(f"  -> Viajar para LESTE pede adianto: o relogio resiste.")
print(f"  -> Por isso a recuperacao a leste e cerca de"
      f" {recuperacao_leste_a_mais*100:.0f}% mais longa.")
print(f"\n  2) A traducao em vitorias (feita pelos autores):")
print(f"  -> {abs(delta_vitorias_pct):.2f}% de {jogos_casa} jogos em casa")
print(f"     = {vitorias_perdidas:.2f} vitorias a menos por temporada")
print(f"\n  3) O que resiste ao criterio de significancia:")
for nome in significativos:
    delta, unidade, p = EFEITOS_CASA[nome]
    print(f"  -> {nome}: {delta:.2f}{unidade}, p = {p}  SIM")
for nome in marginais:
    delta, unidade, p = EFEITOS_CASA[nome]
    print(f"  -> {nome}: {delta:.2f}{unidade}, p = {p}  MARGINAL")
print(f"  -> Ou seja: o DESEMPENHO piora com folga estatistica.")
print(f"     A VITORIA acompanha, mas nao alcanca o criterio convencional.")
print(f"     Vitoria depende de mil outras coisas alem de desempenho.")
print(f"\n  4) A dose importa:")
for h in horas:
    d, p = DOSE[h]
    print(f"  -> {h} h de jet lag a leste: {d:.2f} pontos de saldo (p = {p})")
print(f"  -> Da primeira para a segunda hora, a piora aumenta"
      f" {abs(por_hora):.2f} pontos.")
print(f"     Nao e linear: a segunda hora custa muito mais que a primeira.")
print("=" * 70)
```

```python
# --- VISUALIZACAO ---

DOURADO = '#c8a84b'
VERMELHO = '#c0392b'
VERDE = '#2a8a82'
CINZA = '#6b6a64'

# GRAFICO 1 - O relogio que nao tem 24 horas
fig1, ax1 = plt.subplots(figsize=(12, 7))

ax1.barh([0], [periodo_faixa[1] - periodo_faixa[0]], left=periodo_faixa[0],
         height=0.26, color=VERDE, alpha=0.45, edgecolor=VERDE, linewidth=2.5)
ax1.scatter([periodo_medio], [0], s=320, color=VERDE, zorder=3)
ax1.axvline(x=24.0, color=VERMELHO, linestyle='--', linewidth=2.5)

ax1.text(periodo_medio, 0.24, f'{periodo_medio} h'.replace('.', ','), ha='center',
         fontsize=26, fontweight='bold', color=VERDE)
ax1.text(24.0, -0.30, 'o dia\ntem 24 h', ha='center', va='top', fontsize=14,
         color=VERMELHO, fontweight='bold')
ax1.text((periodo_faixa[0] + periodo_faixa[1]) / 2, -0.20,
         f'faixa das estimativas: {periodo_faixa[0]} a {periodo_faixa[1]} h'
         .replace('.', ','),
         ha='center', fontsize=11, color=CINZA, style='italic')

ax1.set_xlim(23.6, 24.9)
ax1.set_ylim(-0.62, 0.55)
ax1.set_yticks([])
ax1.set_xlabel('Período do relógio biológico humano, medido em isolamento (horas)')
ax1.set_title('O seu relógio interno não tem 24 horas\n'
              'Ele roda um pouco mais devagar que o planeta',
              fontsize=15, pad=18)

plt.figtext(0.5, 0.005,
            'Faixa conforme a literatura de ritmo circadiano  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-076-grafico-01-relogio.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 1 salvo")

# GRAFICO 2 - Assinatura: o que e significativo e o que e marginal
fig2, ax2 = plt.subplots(figsize=(12, 8))

nomes = list(EFEITOS_CASA.keys())
y = np.arange(len(nomes))[::-1]
for yi, nome in zip(y, nomes):
    delta, unidade, p = EFEITOS_CASA[nome]
    cor = VERDE if p < 0.05 else CINZA
    ax2.barh([yi], [delta], color=cor, alpha=0.88, height=0.5)
    ax2.text(delta - 0.12, yi, f'{delta:.2f}'.replace('.', ',') + unidade,
             va='center', ha='right', fontsize=15, fontweight='bold', color=cor)
    marca = f'p = {p}'.replace('.', ',') + ('' if p < 0.05 else '   MARGINAL')
    ax2.text(0.12, yi, marca, va='center', fontsize=12, color=cor,
             fontweight='bold' if p >= 0.05 else 'normal')

ax2.axvline(x=0, color='#333', linewidth=1.6)
ax2.set_yticks(y)
ax2.set_yticklabels(nomes, fontsize=12)
ax2.set_xlim(-7.6, 3.4)
ax2.set_xticks([])
ax2.grid(False)
ax2.set_title('A assinatura estatística: o desempenho cai com folga, '
              'a vitória não alcança\n'
              + f'Leota et al., 2022 · {n_jogos:,}'.replace(",", ".").replace(
                  'Leota et al.. 2022', 'Leota et al., 2022')
              + f' jogos da NBA em {n_temporadas} temporadas',
              fontsize=15, pad=18)
ax2.text(0.5, -0.09,
         'Efeitos do jet lag a leste sobre times da casa. Para times visitantes, '
         'nenhuma associação foi significativa.',
         transform=ax2.transAxes, ha='center', fontsize=11, color=CINZA, style='italic')

plt.figtext(0.5, 0.005,
            'Fonte: Leota et al., 2022, Frontiers in Physiology 13, 892681'
            '  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-076-grafico-02-significancia.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 2 salvo")

# GRAFICO 3 - A dose: a segunda hora custa muito mais que a primeira
fig3, ax3 = plt.subplots(figsize=(12, 7))

hs = [0] + horas
saldos = [0.0] + [DOSE[h][0] for h in horas]
ax3.plot(hs, saldos, marker='o', markersize=15, linewidth=3.2, color=VERMELHO)
for h, s in zip(hs, saldos):
    if h == 0:
        continue
    ax3.text(h, s - 0.55, f'{s:.2f}'.replace('.', ',') + ' pontos',
             ha='center' if h == 1 else 'right',
             fontsize=16, fontweight='bold', color=VERMELHO)
    ax3.text(h, s + 0.30, f'p = {DOSE[h][1]}'.replace('.', ','), ha='center',
             fontsize=11, color=CINZA)

ax3.axhline(y=0, color='#333', linestyle='--', linewidth=1.5)
ax3.set_xticks(hs)
ax3.set_xticklabels(['sem jet lag', '1 hora a leste', '2 horas a leste'], fontsize=13)
ax3.set_ylim(-6.4, 1.4)
ax3.set_ylabel('Saldo de pontos do time da casa')
ax3.set_title('A segunda hora custa muito mais que a primeira\n'
              'Uma hora a leste custa 0,72 ponto. Duas horas custam 4,53.',
              fontsize=15, pad=18)
ax3.text(0.5, -0.145,
         'Cada ponto é uma comparação publicada, não uma curva ajustada. '
         'Nada entre eles foi estimado.',
         transform=ax3.transAxes, ha='center', fontsize=11, color=CINZA, style='italic')

plt.figtext(0.5, 0.005,
            'Fonte: Leota et al., 2022  |  #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-076-grafico-03-dose.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 3 salvo")
```

### 💡 O Insight

**Duas vírgula quarenta e sete vitórias.**

É o que um time da NBA jogando em casa deixa na mesa, por temporada, quando chega de
uma viagem para **leste**. Não é o efeito de viajar. É o efeito da **direção**.

Onze mil quatrocentos e oitenta e um jogos, dez temporadas, com time, adversário e
horário do jogo controlados no modelo.

E a raiz disso é um número que quase ninguém conhece: **o relógio biológico humano roda
em cerca de 24,2 horas.**

Doze minutos a mais que o planeta. Todo dia.

Como ele roda um pouco mais devagar, ele tem uma tendência natural a atrasar, e a luz
da manhã o empurra de volta ao lugar. É por isso que virar a noite é fácil e acordar
cedo é uma briga, para praticamente todo mundo.

Voar para oeste **alonga** o dia, e pede exatamente o atraso que o relógio já quer
fazer. Voar para leste **encurta** o dia, e pede um adianto contra a corrente. O modelo
matemático de Lu e colegas estima que a recuperação a leste leva cerca de 50% mais
tempo.

Agora, duas honestidades.

**A primeira:** o número da manchete, os 6,03% de vitórias, tem **p = 0,051**. Não
alcança o critério convencional, e os autores registram como marginal. O que alcança
com folga são os desfechos de desempenho: rebotes com p abaixo de 0,0001, aproveitamento
de arremesso com p abaixo de 0,01.

A leitura correta, então, é mais fina e mais interessante: **o desempenho piora de forma
consistente, e a vitória acompanha sem alcançar significância.** Faz sentido, porque
vitória depende de mil coisas além de desempenho.

**A segunda:** a literatura não é unânime. Outros grupos encontraram associação negativa
para viagem a oeste na NBA, e associação positiva para leste no beisebol. O mecanismo é
sólido, a tradução dele em placar é bagunçada.

O que eu levo daqui não é sobre esporte.

É que existe uma diferença de doze minutos entre o seu corpo e o mundo, todo dia, e ela
é grande o suficiente para aparecer no placar de uma liga profissional depois de onze
mil jogos.

E ela explica uma coisa banal que a gente costuma tratar como falha de caráter: acordar
cedo é difícil porque você está pedindo ao seu corpo para fazer a coisa contra a qual
ele foi construído. Não é preguiça. É aritmética.

*O que você chama de indisciplina e talvez seja só o seu relógio?*

### ⚠️ Limitações do Modelo

- **Estudo observacional de dados de jogos.** Ninguém sorteou a direção da viagem.
  Associação, não causa.
- **O achado de vitórias é marginal (p = 0,051)** e está a um passo do critério
  convencional. Os desfechos de desempenho são bem mais robustos, e é neles que a
  conclusão se apoia.
- **A literatura é divergente.** McHill e Chinoy, 2020, encontraram associação negativa
  para viagem a oeste na NBA. Winter e colegas, 2009, encontraram associação positiva
  para leste no beisebol. O padrão fisiológico é consistente; o efeito sobre resultado
  esportivo, não.
- **Sem associação para os times visitantes.** Os autores atribuem ao ambiente
  controlado das viagens de time, mas isso é interpretação, não achado, e enfraquece a
  generalização para viajantes comuns.
- **Atletas profissionais não são amostra da população.** Idade, condicionamento, sono
  monitorado e apoio de equipe tornam esse grupo bem diferente de quem pega um voo de
  férias.
- **O gráfico da dose tem apenas os dois pontos publicados.** Nada entre eles foi
  estimado, e a comparação de 1 hora tem p = 0,07.
- **O período de 24,2 horas é uma média.** A faixa das estimativas vai de 23,8 a 24,6,
  e cronotipo individual altera bastante essa conta. Quem tem período mais curto sofre
  menos indo para leste.
- O modelo de Lu e colegas é **matemático**, não é medida direta em viajantes.
- Fator ×0.80 não aplicado: registro objetivo de jogos e desfechos esportivos.

*A ciência é honesta sobre o que não sabe. O modelo também.*
