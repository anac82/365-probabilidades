### 📖 A História

Eu sou a pessoa que resolve.

Sempre fui. A amiga que você liga às três da manhã. A que ouve o problema inteiro, respira, e devolve o caminho com uma calma quase irritante. E eu adoro isso, confesso. Me dá um prazer meio suspeito ser útil, ter a resposta, enxergar a saída que a pessoa no meio do furacão não consegue ver.

O problema é quando o furacão é meu.

Aí, misteriosamente, toda aquela lucidez evapora. A mesma mulher que distribui clareza a torto e a direito fica rodando em círculos, remoendo a mesma decisão, sem conseguir aplicar em si um décimo do bom senso que aplica nos outros.

Durante muito tempo eu achei que isso era um defeito meu. Até descobrir que tem nome, tem experimento, e tem até uma saída.

---

### 📚 O Conceito: O Paradoxo de Salomão

O rei Salomão é lembrado como o homem mais sábio de sua época. Reis viajavam para ouvir seus conselhos. E, ao mesmo tempo, a vida pessoal dele foi um desastre de decisões: excessos, alianças ruins, um filho criado sem limites que fragmentou o reino.

Igor Grossmann e Ethan Kross deram nome a esse contraste: o **paradoxo de Salomão**. A gente raciocina com muito mais sabedoria sobre os problemas dos outros do que sobre os próprios. Não por falta de inteligência, mas por excesso de proximidade. Quando o problema é seu, você está dentro dele, e de dentro não dá pra ver a forma inteira.

A parte boa é que eles também encontraram a saída, e ela é quase constrangedora de tão simples.

---

### 🧮 O Modelo

Uso os escores de sabedoria de raciocínio do estudo para medir o tamanho da diferença entre pensar sobre o problema de um amigo e sobre o próprio, calculo a probabilidade de você ser mais lúcida sobre a vida do outro, e mostro o quanto o auto-distanciamento fecha essa diferença.

**Fonte:**
- Grossmann, I. & Kross, E. (2014). *Exploring Solomon's Paradox: Self-Distancing Eliminates the Self-Other Asymmetry in Wise Reasoning*. **Psychological Science**, 25(8), 1571-1580.
  Três estudos, N total = 693 (inclui adultos de 20-40 e de 60-80 anos).
  Estudo 1: raciocínio sobre o outro M=0,49 (DP 0,69) vs sobre si M=-0,51 (DP 1,03); ηp²=0,25.
  Auto-distanciamento (3ª pessoa) elimina a assimetria: d=0,15 vs raciocinar sobre o outro.

**Nota metodológica:** são escores de sabedoria de raciocínio de experimentos, não proporções de survey autorrelatado. Por isso o fator ×0.80 não se aplica.


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

# Paleta 365 Probabilidades
DOURADO = '#c8a84b'
VERMELHO = '#c0392b'
VERDE    = '#1a5f5a'
CINZA    = '#6b6a64'

np.random.seed(46)  # Dia #046

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Grossmann & Kross (2014), Psychological Science — Estudo 1 (N=104)
# Escore de sabedoria de raciocínio (padronizado)

m_outro, dp_outro = 0.49, 0.69   # raciocinando sobre o problema de um amigo
m_self,  dp_self  = -0.51, 1.03  # raciocinando sobre o próprio problema
eta2_p = 0.25                    # ηp² da assimetria (efeito grande)

N_total = 693
# Auto-distanciamento (Estudo 2): 3ª pessoa quase elimina a diferença
d_distanciado_vs_outro = 0.15    # ~ zero de diferença

print("=" * 66)
print("  DADOS DA LITERATURA — O PARADOXO DE SALOMÃO")
print("=" * 66)
print(f"  Grossmann & Kross 2014 · N total = {N_total}")
print(f"  Sabedoria de raciocínio (escore padronizado):")
print(f"  → Sobre o problema de um AMIGO:  {m_outro:+.2f}  (DP {dp_outro})")
print(f"  → Sobre o SEU próprio problema:  {m_self:+.2f}  (DP {dp_self})")
print(f"  → Tamanho da assimetria:         ηp² = {eta2_p} (grande)")
print()
print(f"  Auto-distanciamento (pensar em 3ª pessoa):")
print(f"  → Diferença vs raciocinar sobre o outro: d = {d_distanciado_vs_outro} (~zero)")
print("=" * 66)
```

```python
# --- O MODELO: quão mais sábia você é sobre a vida dos outros ---

# Cohen's d da assimetria self vs outro (Estudo 1)
dp_pooled = np.sqrt((dp_outro**2 + dp_self**2) / 2)
d_assimetria = (m_outro - m_self) / dp_pooled

# Monte Carlo — P(mais sábia sobre o problema do OUTRO do que sobre o SEU)
n_sim = 100_000
amostra_outro = np.random.normal(m_outro, dp_outro, n_sim)
amostra_self  = np.random.normal(m_self,  dp_self,  n_sim)
p_mais_sabia_outro = np.mean(amostra_outro > amostra_self)

# O quanto o distanciamento recupera
reducao_gap = (d_assimetria - d_distanciado_vs_outro) / d_assimetria

print("=" * 66)
print("  MODELO — A LUCIDEZ QUE VOCÊ RESERVA PROS OUTROS")
print("=" * 66)
print(f"  Distância entre pensar no problema do amigo e no seu:")
print(f"  → Cohen's d = {d_assimetria:.2f}  (efeito grande)")
print()
print(f"  Monte Carlo ({n_sim:,} simulações):")
print(f"  → P(você ser mais sábia sobre o problema do outro")
print(f"     do que sobre o seu): {p_mais_sabia_outro*100:.0f}%")
print()
print(f"  A SAÍDA — pensar em 3ª pessoa:")
print(f"  → A distância cai de d={d_assimetria:.2f} para d={d_distanciado_vs_outro}")
print(f"  → O auto-distanciamento recupera ~{reducao_gap*100:.0f}% da sabedoria")
print(f"     que você normalmente reserva pros outros.")
print()
print(f"  RESPOSTA:")
print(f"  → Você decide, sim, melhor sobre a vida dos outros ({p_mais_sabia_outro*100:.0f}%).")
print(f"    E consegue pegar essa lucidez emprestada pra você")
print(f"    só mudando a pessoa do verbo: 'o que ELA faria?'")
print("=" * 66)
```

```python
# --- VISUALIZAÇÃO — 3 GRÁFICOS SEPARADOS ---

# ── GRÁFICO 1 — A assimetria: o outro vs você ──
fig1, ax1 = plt.subplots(figsize=(10, 6))
cats = ['O SEU\npróprio problema', 'O problema\nde um amigo']
vals = [m_self, m_outro]
cores = [VERMELHO, VERDE]
bars = ax1.bar(cats, vals, color=cores, alpha=0.9, width=0.45)
ax1.axhline(0, color=CINZA, lw=1)
ax1.set_ylabel('Sabedoria de raciocínio (escore padronizado)')
ax1.set_ylim(-0.8, 0.8)
ax1.set_title('Você é mais sábia sobre a vida dos outros\nParadoxo de Salomão — Grossmann & Kross 2014 (Estudo 1)',
              fontsize=13, pad=15)
for bar, v in zip(bars, vals):
    off = 0.06 if v >= 0 else -0.10
    ax1.text(bar.get_x()+bar.get_width()/2, v+off, f'{v:+.2f}',
             ha='center', fontweight='bold', fontsize=16)
plt.figtext(0.5, 0.005,
            'Diferença grande (ηp²=0,25) · você raciocina melhor sobre o problema alheio | Grossmann & Kross 2014 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-046-grafico-01-assimetria.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 — ASSINATURA ESTATÍSTICA: as duas distribuições + Monte Carlo ──
fig2, ax2 = plt.subplots(figsize=(10, 6))
x = np.linspace(-4, 3, 1000)
y_self  = (1/(dp_self*np.sqrt(2*np.pi)))  * np.exp(-0.5*((x-m_self)/dp_self)**2)
y_outro = (1/(dp_outro*np.sqrt(2*np.pi))) * np.exp(-0.5*((x-m_outro)/dp_outro)**2)
ax2.plot(x, y_self, color=VERMELHO, lw=2.5, label=f'Sobre o SEU problema (M={m_self})')
ax2.fill_between(x, y_self, alpha=0.18, color=VERMELHO)
ax2.plot(x, y_outro, color=VERDE, lw=2.5, label=f'Sobre o problema de um AMIGO (M={m_outro})')
ax2.fill_between(x, y_outro, alpha=0.18, color=VERDE)
ax2.axvline(m_self, color=VERMELHO, ls='--', alpha=0.6)
ax2.axvline(m_outro, color=VERDE, ls='--', alpha=0.6)
ax2.text(-3.7, ax2.get_ylim()[1]*0.82,
         f'P(mais sábia\nsobre o do outro)\n= {p_mais_sabia_outro*100:.0f}%\nd = {d_assimetria:.2f}',
         fontsize=11, color='#8a6d10', fontweight='bold')
ax2.set_xlabel('Sabedoria de raciocínio (escore padronizado)')
ax2.set_ylabel('Densidade')
ax2.set_title('A ASSINATURA ESTATÍSTICA\nDuas versões suas: a conselheira e a protagonista',
              fontsize=13, pad=15)
ax2.legend(loc='upper right', fontsize=9)
plt.figtext(0.5, 0.005,
            'Monte Carlo (100 mil simulações) sobre as distribuições do Estudo 1 | Grossmann & Kross 2014 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-046-grafico-02-assinatura.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 — A saída: pensar em 3ª pessoa fecha a distância ──
fig3, ax3 = plt.subplots(figsize=(10, 6))
cats3 = ['Imersa no\nseu problema\n(1ª pessoa)', 'Olhando de fora\n(3ª pessoa)']
vals3 = [d_assimetria, d_distanciado_vs_outro]
cores3 = [VERMELHO, VERDE]
bars3 = ax3.bar(cats3, vals3, color=cores3, alpha=0.9, width=0.45)
ax3.set_ylabel("Distância até a sua sabedoria de conselheira (Cohen's d)")
ax3.set_ylim(0, 1.35)
ax3.set_title('A saída cabe numa mudança de pronome\nAuto-distanciamento quase zera a diferença — Grossmann & Kross 2014',
              fontsize=13, pad=15)
for bar, v in zip(bars3, vals3):
    ax3.text(bar.get_x()+bar.get_width()/2, v+0.03, f'd = {v:.2f}',
             ha='center', fontweight='bold', fontsize=16)
ax3.annotate('', xy=(1, d_distanciado_vs_outro+0.05), xytext=(0, d_assimetria-0.05),
             arrowprops=dict(arrowstyle='->', color=DOURADO, lw=2))
ax3.text(0.5, 0.72, 'a distância\nquase some', ha='center',
         fontsize=12, color='#8a6d10', fontweight='bold', transform=ax3.transAxes)
plt.figtext(0.5, 0.005,
            '"O que ELA faria?" recupera a lucidez que você reserva pros outros | Grossmann & Kross 2014 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-046-grafico-03-distancia.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**Você tem cerca de 79% de chance de ser mais lúcida sobre o problema de um amigo do que sobre o seu.**

Não é falta de inteligência. É excesso de proximidade. Quando a vida é a sua, você está dentro do furacão, e de dentro dá pra sentir o vento, mas não dá pra ver o formato. A diferença entre a sua sabedoria de conselheira e a de protagonista é grande, quase um desvio-padrão inteiro (d ≈ 1,1).

E aqui está o achado que me deixou de queixo caído, porque a solução é ridícula de simples. Quando você pensa sobre o seu próprio dilema em terceira pessoa, "o que a Ana faria?", "o que eu diria a uma amiga vivendo exatamente isto?", o distanciamento devolve quase toda aquela lucidez. A diferença desaba de d ≈ 1,1 para d ≈ 0,15. Quase some.

Salomão aconselhava reinos e afundava a própria casa. Eu adoro resolver a vida dos outros e travo na minha. Talvez você também. A boa notícia é que a conselheira sábia que mora em você não foi embora quando o problema virou seu. Ela só precisa que você troque a pessoa do verbo.

*Que conselho você daria, sem hesitar, a uma amiga vivendo exatamente a sua vida agora?*

### ⚠️ Limitações do Modelo

- Os Estudos 1 e 2 usaram amostras majoritariamente jovens; o Estudo 3 confirmou o padrão também em adultos de 60-80 anos.
- Sabedoria de raciocínio foi medida por respostas codificadas a cenários (reconhecer limites do próprio conhecimento, considerar outras perspectivas, buscar meio-termo), não é uma medida absoluta de "acerto".
- O efeito do auto-distanciamento é robusto, mas de tamanho médio; não é uma varinha mágica.
- A probabilidade de ~79% vem das distribuições do Estudo 1 (Monte Carlo); é uma estimativa, não um valor fixo.
- O fator ×0.80 não foi aplicado: escores de experimento, não survey autorrelatado.

*A ciência é honesta sobre o que não sabe. O modelo também.*
