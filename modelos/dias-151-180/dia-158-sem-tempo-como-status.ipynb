## Qual a probabilidade de "estou sem tempo" ser um símbolo de status?

**Tipo:** Comportamental (experimental)  
**Data de publicação:** 2026-11-18  
**Ferramenta:** Python  
**Decisão analisada:** O que eu comunico quando digo que estou sem tempo?  
**Hashtag:** #365Probabilidades #Dia158 #ProbabilidadesDoTempo

---

### 📖 A História

[ABERTURA DA ANA - uma linha sua entra aqui]

"Na correria" virou resposta padrão para "como você está?". Quase ninguém responde que está com tempo sobrando.

Fui ver o que essa resposta comunica a quem ouve.

---

### 📚 O Conceito

Durante muito tempo, mostrar tempo livre foi sinal de riqueza. Quem podia não trabalhar exibia isso.

Três pesquisadoras de marketing (Silvia Bellezza, Neeru Paharia e Anat Keinan) propuseram que em alguns lugares o sinal se inverteu. A pessoa que vive ocupada no trabalho passou a ser vista como alguém competente e ambicioso, e por isso disputado. Tempo escasso viraria prova de que a pessoa vale muito.

O teste é simples. Um grupo de participantes lê sobre alguém que trabalha muito e não tem tempo. Outro grupo, sorteado, lê sobre alguém com tempo livre. Os dois grupos dão uma nota de 1 a 7 para o status dessa pessoa (posição social, riqueza e renda). Como a única diferença entre os grupos é o texto lido, a diferença nas notas vem da agenda descrita.

---

### 🧮 O Modelo

O artigo publica a média e o desvio-padrão de cada grupo. Em cima deles entram três contas próprias.

1. **Os cinco testes americanos num efeito só.** Cada teste vira uma diferença padronizada (g de Hedges), e os cinco são combinados por efeitos aleatórios, com intervalo de confiança, heterogeneidade e intervalo de predição.
2. **Os duelos.** O efeito combinado é traduzido na pergunta do dia. Em 100 confrontos entre um ocupado e alguém com tempo livre, em quantos o ocupado recebe a nota maior. A conta é duelos = Φ(g / √2), conferida por simulação direta (200 mil confrontos, semente 42). A Itália entra ao lado, com o sinal invertido.
3. **O que muda na impressão.** No Estudo 1A, o tamanho da diferença em competência, em ser disputado, em status e em simpatia.

**Fontes:**
- Bellezza, S., Paharia, N. & Keinan, A. (2017). Conspicuous Consumption of Time: When Busyness and Lack of Leisure Time Become a Status Symbol. *Journal of Consumer Research*, 44(1), 118-138. doi:10.1093/jcr/ucw076. Estudos 1A (N = 307), 1B (N = 112), 2A (N = 300), 2B (N = 302) e 3 (112 americanos e 98 italianos); estudo-piloto com 1.100 tweets. Lido no artigo completo, na cópia da Columbia Business School.
- Holthoff, L. C. & Scheiben, C. (2019). Conspicuous Consumption of Time: A Replication. *Journal of Marketing Behavior*, 3(4), 355-362. Réplica na Alemanha (n = 425 no Estudo 1). O efeito se repete, e lazer exibido rende ainda mais status. Lido no resumo.

**Nota metodológica sobre o fator ×0,80.** Não se aplica. Os números são médias de escala em experimentos com sorteio de condição, e os 12% do estudo-piloto são contagem de tweets, não proporção de pesquisa de opinião.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

SEMENTE = 42
N_SIM = 200_000
rng = np.random.default_rng(SEMENTE)

VERMELHO = '#c0392b'
DOURADO = '#c8a84b'
VERDE = '#1a5f5a'
CINZA = '#7f8c8d'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Bellezza, Paharia & Keinan (2017), Journal of Consumer Research 44(1), 118-138
# Nota de status de 1 a 7. Cada linha: (rótulo, N total, média ocupado, DP, média com tempo livre, DP)
# O artigo não publica o tamanho de cada grupo. O modelo supõe metade em cada condição.

estudos_eua = [
    ('1A · posts de Facebook (N=307)',        307, 3.70, 1.02, 3.40, 1.23),
    ('1B · carta de um amigo (N=112)',        112, 3.99, 1.08, 3.52, 0.99),
    ('2A · muitas ou poucas horas (N=300)',   300, 4.07, 1.13, 3.51, 1.24),
    ('2B · muitas ou poucas horas (N=302)',   302, 3.92, 1.31, 3.29, 1.20),
    ('3 · americanos (N=112)',                112, 4.62, 0.89, 3.95, 1.75),
]
estudo_italia = ('3 · italianos (N=98)', 98, 4.54, 0.85, 5.21, 1.35)

# Estudo 1A: o que muda na impressão sobre a pessoa (ocupado contra com tempo livre)
# (rótulo, média ocupado, DP, média com tempo livre, DP)
impressoes_1a = [
    ('Competência e ambição', 4.88, 1.02, 3.24, 1.11),
    ('Ser disputado',         3.99, 1.16, 2.68, 1.17),
    ('Status',                3.70, 1.02, 3.40, 1.23),
    ('Simpatia e honestidade', 4.44, 0.75, 4.35, 0.83),
]
n_1a = 307

# Estudo-piloto: tweets de celebridades que se gabam com falsa modéstia
n_tweets = 1100
p_tweets_trabalho = 0.12          # "cerca de 12%", contagem exata não publicada

# Crença de que trabalho leva à ascensão social (piloto, 30 pessoas por país, escala de 1 a 7)
mobilidade_eua = 4.91
mobilidade_italia = 3.98

# Holthoff & Scheiben (2019), Journal of Marketing Behavior 3(4), 355-362: réplica na Alemanha
n_alemanha_estudo1 = 425

aplica_fator_080 = False

print("=" * 70)
print("  DADOS DA LITERATURA - 'ESTOU SEM TEMPO' COMO SÍMBOLO DE STATUS")
print("=" * 70)
print("\n  Bellezza, Paharia & Keinan (2017), JCR 44(1). Nota de status de 1 a 7:")
for rot, n, mo, so, ml, sl in estudos_eua + [estudo_italia]:
    print(f"  → {rot:<38} ocupado {mo:.2f} (DP {so:.2f})  ·  tempo livre {ml:.2f} (DP {sl:.2f})")
print(f"\n  Tweets analisados: " + f"{n_tweets:,}".replace(',', '.') + f", cerca de {p_tweets_trabalho*100:.0f}% sobre trabalho demais e falta de tempo")
print(f"  Crença na ascensão pelo trabalho (1 a 7): EUA {mobilidade_eua:.2f}  ·  Itália {mobilidade_italia:.2f}")
print(f"  Réplica alemã (Holthoff & Scheiben 2019): n = {n_alemanha_estudo1} no Estudo 1")
print(f"\n  Fator ×0,80 aplicado: {'sim' if aplica_fator_080 else 'não'}")
print("=" * 70)
```

```python
# --- O MODELO ---
# Três contas próprias sobre as médias e os desvios publicados. Semente 42, 200 mil sorteios.

def efeito(n, m1, s1, m2, s2):
    # diferença padronizada (g de Hedges) e variância, com metade da amostra em cada grupo
    n1 = n2 = n / 2
    sp = np.sqrt(((n1 - 1) * s1**2 + (n2 - 1) * s2**2) / (n1 + n2 - 2))
    d = (m1 - m2) / sp
    j = 1 - 3 / (4 * (n1 + n2 - 2) - 1)
    g = j * d
    var = j**2 * ((n1 + n2) / (n1 * n2) + d**2 / (2 * (n1 + n2)))
    return g, var

def duelos(g):
    # em quantos de 100 confrontos o ocupado recebe a nota maior
    return stats.norm.cdf(g / np.sqrt(2)) * 100

# ---------- Camada 1: os cinco testes americanos num efeito só ----------
g_eua = np.array([efeito(*e[1:])[0] for e in estudos_eua])
v_eua = np.array([efeito(*e[1:])[1] for e in estudos_eua])
k = len(g_eua)

w = 1 / v_eua
g_fixo = np.sum(w * g_eua) / np.sum(w)
Q = np.sum(w * (g_eua - g_fixo) ** 2)
C = np.sum(w) - np.sum(w**2) / np.sum(w)
tau2 = max(0.0, (Q - (k - 1)) / C)                      # DerSimonian-Laird
i2 = max(0.0, (Q - (k - 1)) / Q) * 100 if Q > 0 else 0.0
p_q = 1 - stats.chi2.cdf(Q, k - 1)

w_al = 1 / (v_eua + tau2)
g_comb = np.sum(w_al * g_eua) / np.sum(w_al)
ep_comb = np.sqrt(1 / np.sum(w_al))
ic_comb = (g_comb - 1.96 * ep_comb, g_comb + 1.96 * ep_comb)
t_crit = stats.t.ppf(0.975, k - 2)
ip_comb = (g_comb - t_crit * np.sqrt(tau2 + ep_comb**2), g_comb + t_crit * np.sqrt(tau2 + ep_comb**2))
dif_media_pontos = np.sum(w_al * np.array([e[2] - e[4] for e in estudos_eua])) / np.sum(w_al)

# ---------- Camada 2: os duelos ----------
duelos_eua = duelos(g_comb)
g_sim = rng.normal(g_comb, ep_comb, N_SIM)
ic_duelos_eua = np.percentile(duelos(g_sim), [2.5, 97.5])
p_efeito_positivo = (g_sim > 0).mean()

g_it, v_it = efeito(*estudo_italia[1:])
ic_it = (g_it - 1.96 * np.sqrt(v_it), g_it + 1.96 * np.sqrt(v_it))
duelos_it = duelos(g_it)
g_it_sim = rng.normal(g_it, np.sqrt(v_it), N_SIM)
ic_duelos_it = np.percentile(duelos(g_it_sim), [2.5, 97.5])

# simulação direta dos confrontos, com as médias e desvios do Estudo 3
def confronto(m1, s1, m2, s2):
    a = rng.normal(m1, s1, N_SIM)
    b = rng.normal(m2, s2, N_SIM)
    return (a > b).mean() * 100

sim_eua_3 = confronto(*estudos_eua[4][2:])
sim_it_3 = confronto(*estudo_italia[2:])

# ---------- Camada 3: o que muda na impressão (Estudo 1A) ----------
escada = [(rot, efeito(n_1a, mo, so, ml, sl)[0]) for rot, mo, so, ml, sl in impressoes_1a]

print("=" * 70)
print("  MODELO - RESULTADO")
print("=" * 70)
print("\n  CAMADA 1 · Os cinco testes americanos")
for e, g, v in zip(estudos_eua, g_eua, v_eua):
    print(f"  → {e[0]:<38} g = {g:+.2f} [{g-1.96*np.sqrt(v):+.2f}; {g+1.96*np.sqrt(v):+.2f}]")
print(f"  → EFEITO COMBINADO (efeitos aleatórios): g = {g_comb:.2f}, IC 95% [{ic_comb[0]:.2f}; {ic_comb[1]:.2f}]")
print(f"  → Heterogeneidade: Q = {Q:.2f} (gl = {k-1}, p = {p_q:.2f}), I² = {i2:.0f}%, tau² = {tau2:.3f}")
print(f"  → Intervalo de predição para um novo teste: [{ip_comb[0]:.2f}; {ip_comb[1]:.2f}]")
print(f"  → Diferença média na escala de 1 a 7: {dif_media_pontos:.2f} ponto")
print(f"  → P(efeito positivo): {p_efeito_positivo*100:.1f}%")
print("\n  CAMADA 2 · Os duelos (em 100 confrontos, quem recebe a nota maior)")
print(f"  → Estados Unidos: o ocupado vence {duelos_eua:.0f} de 100 [{ic_duelos_eua[0]:.0f}; {ic_duelos_eua[1]:.0f}]")
print(f"  → Itália: g = {g_it:+.2f} [{ic_it[0]:+.2f}; {ic_it[1]:+.2f}], o ocupado vence {duelos_it:.0f} de 100 [{ic_duelos_it[0]:.0f}; {ic_duelos_it[1]:.0f}]")
print(f"  → Simulação direta do Estudo 3: EUA {sim_eua_3:.0f} de 100  ·  Itália {sim_it_3:.0f} de 100")
print("\n  CAMADA 3 · O que muda na impressão (Estudo 1A, N = 307)")
for rot, g in escada:
    print(f"  → {rot:<24} g = {g:+.2f}  ·  {duelos(g):.0f} de 100 duelos")
print("=" * 70)
```

```python
# --- VISUALIZAÇÃO - GRÁFICOS SEPARADOS ---

# ── GRÁFICO 1 · Os cinco testes e a Itália ──
fig1, ax1 = plt.subplots(figsize=(12, 8))
rotulos = [e[0] for e in estudos_eua] + ['EUA · efeito combinado', estudo_italia[0]]
gs = list(g_eua) + [g_comb, g_it]
eps = list(np.sqrt(v_eua)) + [ep_comb, np.sqrt(v_it)]
cores = [VERDE] * k + [DOURADO, VERMELHO]
ys = np.arange(len(gs))[::-1]
for y, g, ep, c, rot in zip(ys, gs, eps, cores, rotulos):
    ax1.errorbar(g, y, xerr=1.96 * ep, fmt='D' if 'combinado' in rot else 'o', color=c,
                 markersize=12 if 'combinado' in rot else 9, capsize=5, lw=2)
    ax1.text(1.42, y, f'{g:+.2f} [{g-1.96*ep:+.2f}; {g+1.96*ep:+.2f}]', va='center', fontsize=11, family='monospace')
ax1.axvline(0, color='#222', lw=1.2)
ax1.set_yticks(ys)
ax1.set_yticklabels(rotulos, fontsize=11)
ax1.set_xlim(-1.2, 2.3)
ax1.set_xlabel('Diferença padronizada na nota de status (g de Hedges)\n← tempo livre vale mais          ocupado vale mais →', fontsize=11)
ax1.set_title('Nos Estados Unidos o ocupado é visto com mais status. Na Itália, o contrário\n'
              f'Efeito combinado g = {g_comb:.2f}, IC 95% [{ic_comb[0]:.2f}; {ic_comb[1]:.2f}]', fontsize=14, pad=15)
plt.figtext(0.5, 0.01, 'Fonte: Bellezza, Paharia & Keinan, 2017, JCR 44(1). Efeitos calculados das médias e desvios publicados | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=[0, 0.03, 1, 1])
plt.savefig('dia-158-grafico-01-cinco-testes.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 · Os duelos ──
fig2, ax2 = plt.subplots(figsize=(12, 8))
nomes = ['Estados Unidos\n(cinco testes combinados)', 'Itália\n(Estudo 3)']
vals = [duelos_eua, duelos_it]
ics = [ic_duelos_eua, ic_duelos_it]
barras = ax2.bar(nomes, vals, color=[VERDE, VERMELHO], alpha=0.85, width=0.5)
for i, (v, ic) in enumerate(zip(vals, ics)):
    ax2.errorbar(i, v, yerr=[[v - ic[0]], [ic[1] - v]], color='#222', capsize=8, lw=2)
    ax2.text(i, ic[1] + 2, f'{v:.0f} de 100', ha='center', fontsize=20, fontweight='bold')
ax2.axhline(50, color=DOURADO, ls='--', lw=2)
ax2.text(1.48, 51, 'empate (50)', ha='right', fontsize=11, color=DOURADO, fontweight='bold')
ax2.set_ylim(0, 85)
ax2.set_ylabel('Confrontos em que o ocupado recebe a nota maior de status', fontsize=12)
ax2.set_title('Em 100 confrontos entre um ocupado e alguém com tempo livre,\nquantas vezes o ocupado é visto com mais status', fontsize=14, pad=15)
ax2.text(0.5, 0.06, 'duelos = Φ(g / √2)', transform=ax2.transAxes, ha='center', fontsize=13, family='monospace',
         color='white', bbox=dict(boxstyle='round', facecolor='#333', edgecolor='none'))
plt.figtext(0.5, 0.01, 'Modelo próprio sobre Bellezza, Paharia & Keinan, 2017. Barras com IC 95% | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=[0, 0.03, 1, 1])
plt.savefig('dia-158-grafico-02-duelos.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 · O que muda na impressão ──
fig3, ax3 = plt.subplots(figsize=(12, 8))
rot3 = [r for r, _ in escada][::-1]
g3 = [g for _, g in escada][::-1]
cores3 = [CINZA, VERMELHO, DOURADO, VERDE]
ax3.barh(rot3, g3, color=cores3, alpha=0.85, height=0.55)
for i, g in enumerate(g3):
    ax3.text(g + 0.03, i, f'g = {g:.2f}  ·  {duelos(g):.0f} de 100 duelos', va='center', fontsize=12, fontweight='bold')
ax3.set_xlim(0, 2.2)
ax3.set_xlabel('Diferença padronizada entre o ocupado e a pessoa com tempo livre', fontsize=12)
ax3.set_title('A agenda cheia muda muito a impressão de competência e pouco a de status\n'
              'Estudo 1A, 307 americanos avaliando posts de Facebook', fontsize=14, pad=15)
plt.figtext(0.5, 0.01, 'Fonte: Bellezza, Paharia & Keinan, 2017, Estudo 1A | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=[0, 0.03, 1, 1])
plt.savefig('dia-158-grafico-03-impressao.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**Nos Estados Unidos, o ocupado é visto com mais status em 62 de cada 100 confrontos.**

O efeito combinado dos cinco testes é g = 0,42 (IC 95% de 0,30 a 0,54), cerca de meio ponto numa escala de 1 a 7. Os cinco testes concordam entre si. É uma vantagem real e modesta, 12 confrontos acima do empate.

Na Itália o resultado inverte. Lá o ocupado vence só 34 de cada 100 confrontos, e quem tem tempo livre é visto com mais status.

O que a agenda cheia muda de verdade é a impressão de competência. No Estudo 1A, o ocupado é visto como mais competente e ambicioso em 86 de cada 100 confrontos, e como mais rico e bem posicionado em 57.

Então a resposta depende de quem ouve. "Estou sem tempo" comunica status onde as pessoas acreditam que trabalho duro leva à ascensão. Onde essa crença é mais fraca, comunica o contrário.

A pergunta que fica:

*Quando você diz que está sem tempo, o que espera que a outra pessoa entenda?*

### ⚠️ Limitações do Modelo

- As pessoas avaliadas são personagens inventados, descritos em posts, cartas e textos curtos. O estudo mede a impressão de quem lê, não o status real de quem vive ocupado.
- O artigo não publica o tamanho de cada grupo. Os efeitos foram calculados supondo metade da amostra em cada condição, o que é próximo do que os autores descrevem.
- No Estudo 1A a nota de status é um item só (posição social). Nos outros quatro testes é a média de três itens (posição social, riqueza e renda). O efeito do 1A é o menor dos cinco.
- Os cinco testes americanos saíram do mesmo artigo e do mesmo grupo de pesquisa. A concordância entre eles (I² = 0%) não substitui réplica independente. A réplica alemã de 2019 foi lida só no resumo e não entra na conta.
- O intervalo de predição (0,23 a 0,61) vale para um novo teste parecido com esses, com americanos avaliando textos. Não vale para outros países.
- A Itália é um teste só, com 98 pessoas. O intervalo do efeito vai de −0,99 a −0,19.
- A tradução em duelos supõe notas com distribuição normal. A simulação direta do Estudo 3 dá 63 e 34 de 100, perto da fórmula.
- As amostras são de painéis online e de universitários, em 2015 e 2016. Não há dado do Brasil.
- A explicação pela crença na ascensão social vem da moderação medida nos Estudos 2A e 2B, com uma escala de confiabilidade baixa no 2A (α = 0,59). É associação dentro do experimento, não causa demonstrada.
- Os 12% de tweets sobre trabalho demais vêm de uma leitura de 1.100 tweets de celebridades, sem contagem exata publicada. Entram como contexto.

*A ciência é honesta sobre o que não sabe. O modelo também.*
