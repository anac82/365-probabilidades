### 📖 A História

Você já deve ter visto a frase: a cada sete anos, as pessoas ao seu redor mudam, menos as
mais próximas. Circula como ditado, às vezes alemão, às vezes chinês, às vezes com cinco
anos no lugar de sete. A origem é um estudo holandês cujo release, em 2009, saiu com o
título "metade dos seus amigos se perde em sete anos". E o estudo mede exatamente o
círculo mais próximo. A frase acertou uma metade.

Faça duas listas. A de hoje: as pessoas para quem você ligaria se precisasse conversar
sobre algo sério, e as que chamaria para ajudar num conserto em casa. E a de sete anos
atrás, as mesmas perguntas. Quantos nomes se repetem?

Um grupo de sociólogos holandeses fez exatamente isso, com a mesma gente, duas vezes,
sete anos de distância. O tamanho das listas não mudou. Os nomes, sim.

---

### 📚 O Conceito

A sociologia chama de **rede pessoal** o conjunto de pessoas com quem alguém troca apoio.
A parte mais íntima dela, os **confidentes** e os que ajudam na prática, costuma ter três
a cinco nomes. O que o estudo mede é a **rotatividade** dessa parte: quem permanece, quem
sai, e por quê.

A hipótese dos autores vem de Scott Feld: relações nascem e se sustentam em **contextos
de encontro** (o trabalho, a família, a vizinhança, o clube). Quando o contexto some, a
ocasião de se ver some junto, e a relação esfria sem que ninguém decida nada. A alternativa
seria a explicação que a gente costuma dar: as pessoas saem da nossa vida por desgaste,
por briga, porque a relação mudou. O estudo pesa as duas.

### 🧮 O Modelo

**Tabela de fontes**

| Wellman, Wong, Tindall & Nazer (1997), *Social Networks* 19 | Canadá, Toronto (East York) | 33 pessoas, 162 laços íntimos | ~10 anos | 27% dos laços íntimos seguiam íntimos; citado na tese de Mollenhorst |

**Fonte principal:**
- **Mollenhorst, G., Völker, B. & Flap, H. (2014).** Changes in personal relationships:
  How social contexts affect the emergence and discontinuation of relationships.
  *Social Networks*, 37, 65-80. doi 10.1016/j.socnet.2013.12.003. Texto fechado; o mesmo
  estudo é o capítulo 6 da tese de Mollenhorst (Utrecht, 2009, *Networks in contexts*),
  lida no texto completo (tabelas 6.1 a 6.5).
  - Survey of the Social Networks of the Dutch. Onda 1 em 1999/2000, **1.007 pessoas** de
    18 a 65 anos, amostra representativa da Holanda. Onda 2 em 2007, **600 reentrevistadas**.
  - Rede medida por duas perguntas: com quem discutiu assuntos pessoais importantes nos
    últimos seis meses, e a quem pediria ajuda num conserto em casa.
  - **2.131 pessoas nomeadas** em 1999/2000. Em 2007, **646 (30%) cumpriam a mesma
    função**; outros 18% seguiam na rede num papel menor; **48% ficaram, 52% saíram**.
  - Dos que saíram, 977 com resposta: **60% ainda estavam em contato**. Motivo mais comum:
    falta de ocasião de encontro (**52,9%**); mudança na relação (briga, divórcio),
    **12,2%**.
  - Por onde se conheceram: dos **130 conhecidos no trabalho, 11 continuaram**.

**Segunda camada:**
- **Fischer, C. S. & Offer, S. (2020).** Who is dropped and why? Methodological and
  substantive accounts for network loss. *Social Networks*, 61, 78-86. doi
  10.1016/j.socnet.2019.08.008. Lido no resumo e na ficha da publicação. Mesma pergunta,
  outro país, intervalo de um ano em vez de sete: metade dos nomes some da lista, mas
  grande parte por esquecimento, e os parentes próximos são os que ficam.
- **Wellman, B., Wong, R. Y., Tindall, D. & Nazer, N. (1997).** A decade of network change.
  *Social Networks*, 19(1), 27-50. Citado na tese de Mollenhorst: 27% de 162 laços íntimos
  de 33 moradores de Toronto seguiam íntimos dez anos depois.

**Modelo próprio:**
- Posterior Beta (Jeffreys) para cada proporção de laços: quem continua, quem fica na
  rede, quem some por falta de ocasião, quem some por briga, e a continuação por contexto.
- A sua rede de quatro: simulação de quantas pessoas sobram em sete anos, com a taxa de
  permanência medida, e a chance de a lista zerar.
- Duas rodadas de sete anos, para ver o que sobra em catorze.
- Três países na mesma régua: permanência por ano na Holanda, na Califórnia e em Toronto.
- Quem entrou e quem saiu, por tipo de laço: a composição da rede em 1999/2000 e em 2007
  (parceiro, parentes, amigos, colegas, vizinhos).
- **Sem ×0,80:** são laços nomeados e reencontrados em duas ondas de um painel, não
  proporções de opinião autorrelatada. O 18% "na rede com outra função" vem do texto, não
  de uma tabela, e a contagem correspondente é reconstruída (declarado).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

VERMELHO, DOURADO, VERDE, CINZA = '#c0392b', '#c8a84b', '#1a5f5a', '#6b6a64'
SEMENTE = 42
ger = np.random.default_rng(SEMENTE)

def vir(x, casas=1):
    return f"{x:.{casas}f}".replace(".", ",")

def pct(x, casas=1):
    return vir(100 * x, casas) + "%"

def mil(n):
    return f"{int(round(n)):,}".replace(",", ".")

def jeffreys(k, n):
    d = stats.beta(k + 0.5, n - k + 0.5)
    return d, d.mean(), d.interval(0.95)

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Mollenhorst, Völker & Flap (2014), Social Networks 37; tese de Mollenhorst (2009), cap. 6, tabelas 6.1 a 6.5

N_ONDA1, N_ONDA2 = 1007, 600            # pessoas entrevistadas em 1999/2000 e reentrevistadas em 2007
TAMANHO_REDE = {"1999/2000": 3.58, "2007": 4.15}   # média de nomes nas duas perguntas
COMPOSICAO = {   # tabela 6.1: média de nomes por tipo de laço, 1999/2000 e 2007
    "parceiro": (0.47, 0.51), "parentes": (1.14, 1.28), "amigos": (1.36, 1.16),
    "colegas": (0.29, 0.26), "vizinhos": (0.56, 0.90), "outros": (0.14, 0.34)}
OR_PARENTE = 1.6         # tabela 6.4
# Fischer & Offer 2020 (Califórnia, UCNets, ~1.000 pessoas, 1 ano)
FO = {"N": 1000, "anos": 1, "p_nao_repetido": 0.50, "p_esquecido_entre_sumidos": 0.40, "perda_real_max": 0.20}
# Wellman et al. 1997 (Toronto, 33 pessoas, 162 laços íntimos, ~10 anos)
WELLMAN = {"lacos": 162, "intimos_depois": 44, "anos": 10}   # 27% de 162 = 43,7, arredondado

# Os laços nomeados em 1999/2000 e o que aconteceu com eles em 2007
N_LACOS = 2131                           # pessoas nomeadas em 1999/2000 (com contexto informado)
MESMA_FUNCAO = 646                       # em 2007, ainda confidente ou ajuda prática (30%)
P_OUTRA_FUNCAO = 0.18                    # ficaram na rede em outro papel, do texto; contagem reconstruída abaixo
OUTRA_FUNCAO = int(round(P_OUTRA_FUNCAO * N_LACOS))
FICARAM = MESMA_FUNCAO + OUTRA_FUNCAO    # 48%
SAIRAM = N_LACOS - FICARAM

# Dos que saíram da rede, 977 com resposta sobre o porquê (até 3 respostas) e sobre o contato
N_SAIRAM_RESP = 977
AINDA_EM_CONTATO = 591                   # saíram da rede íntima, mas seguem em contato
SEM_CONTATO = 348
MOTIVOS = {                              # tabela 6.5, total
    "contato foi rareando": 213,
    "não dividem mais o mesmo lugar": 199,
    "outro motivo": 185,
    "a relação mudou (briga, divórcio, sem necessidade)": 120,
    "sem resposta": 105,
    "esqueci de mencionar, continua importante": 78,
    "morreu": 67,
    "mudou de casa": 29,
    "distância grande demais": 9,
}
FALTA_DE_OCASIAO = 213 + 199 + 29 + 9 + 67    # 52,9%, soma dos próprios autores
RELACAO_MUDOU = 120                            # 12,2%

# Tabela 6.2: por onde se encontravam em 1999/2000 -> total, continuaram em 2007
CONTEXTO = {   # (total, continuaram)
    "em casa": (1363, 445),
    "na vizinhança": (246, 77),
    "com/via família": (194, 67),
    "no trabalho": (130, 11),
    "via um amigo": (47, 14),
    "no bar, na saída": (40, 10),
    "no clube, na associação": (38, 10),
    "na escola": (11, 3),
    "outro lugar": (62, 9),
}
OR_TRABALHO = 0.27     # tabela 6.4, modelo multinível com controles
OR_PARCEIRO = 4.5

# Eventos de vida nos sete anos (N = 582)
EVENTOS = {"trocou de emprego": 0.281, "mudou de casa": 0.213, "parou de trabalhar": 0.152,
           "teve filho": 0.101, "casou ou foi morar junto": 0.090, "nenhum evento": 0.250}

print("=" * 70)
print("  DADOS DA LITERATURA — SETE ANOS DEPOIS")
print("=" * 70)
print(f"\n  SSND: {mil(N_ONDA1)} pessoas em 1999/2000, {N_ONDA2} reentrevistadas em 2007")
print(f"  Tamanho médio da rede íntima: {vir(TAMANHO_REDE['1999/2000'], 2)} → {vir(TAMANHO_REDE['2007'], 2)} nomes")
print(f"\n  {mil(N_LACOS)} pessoas nomeadas em 1999/2000. Em 2007:")
print(f"  → mesma função:            {MESMA_FUNCAO:>5d}  ({pct(MESMA_FUNCAO / N_LACOS)})")
print(f"  → na rede, outra função:   {OUTRA_FUNCAO:>5d}  ({pct(OUTRA_FUNCAO / N_LACOS)}, reconstruído dos 18% do texto)")
print(f"  → saíram da rede:          {SAIRAM:>5d}  ({pct(SAIRAM / N_LACOS)})")
print(f"\n  Dos que saíram ({N_SAIRAM_RESP} com resposta):")
print(f"  → ainda em contato:        {AINDA_EM_CONTATO}  ({pct(AINDA_EM_CONTATO / N_SAIRAM_RESP)})")
print(f"  → falta de ocasião:        {FALTA_DE_OCASIAO}  ({pct(FALTA_DE_OCASIAO / N_SAIRAM_RESP)})")
print(f"  → a relação mudou:         {RELACAO_MUDOU}  ({pct(RELACAO_MUDOU / N_SAIRAM_RESP)})")
print(f"\n  Por onde se encontravam em 1999/2000 → continuaram em 2007:")
for k, (t, c) in CONTEXTO.items():
    print(f"  → {k:<26s} {c:>4d} de {t:<5d} ({pct(c / t)})")
print(f"\n  Composição média da rede (1999/2000 → 2007): " + "; ".join(f"{k} {vir(a, 2)} → {vir(b, 2)}" for k, (a, b) in COMPOSICAO.items()))
print(f"\n  Eventos de vida em sete anos: " + "; ".join(f"{k} {pct(v, 0)}" for k, v in EVENTOS.items()))
print(f"\n  Fischer & Offer 2020 (Califórnia, ~{mil(FO['N'])} pessoas, {FO['anos']} ano): {pct(FO['p_nao_repetido'], 0)} dos nomes não repetidos; {pct(FO['p_esquecido_entre_sumidos'], 0)} destes por esquecimento; perda real < {pct(FO['perda_real_max'], 0)}")
print(f"  Wellman et al. 1997 (Toronto, {WELLMAN['anos']} anos): {WELLMAN['intimos_depois']} de {WELLMAN['lacos']} laços íntimos seguiam íntimos ({pct(WELLMAN['intimos_depois'] / WELLMAN['lacos'], 0)})")
print("=" * 70)
```

```python
# --- O MODELO ---

# 1. As proporções com incerteza (Jeffreys)
d_mesma, m_mesma, ic_mesma = jeffreys(MESMA_FUNCAO, N_LACOS)
d_fica, m_fica, ic_fica = jeffreys(FICARAM, N_LACOS)
d_contato, m_contato, ic_contato = jeffreys(AINDA_EM_CONTATO, N_SAIRAM_RESP)
d_ocasiao, m_ocasiao, ic_ocasiao = jeffreys(FALTA_DE_OCASIAO, N_SAIRAM_RESP)
d_briga, m_briga, ic_briga = jeffreys(RELACAO_MUDOU, N_SAIRAM_RESP)
razao_ocasiao_briga = FALTA_DE_OCASIAO / RELACAO_MUDOU

# 2. Por contexto: continuação com IC
ctx = pd.DataFrame([(k, t, c) for k, (t, c) in CONTEXTO.items()], columns=["contexto", "total", "continuaram"])
ctx["p"] = ctx["continuaram"] / ctx["total"]
ctx["lo"] = [jeffreys(c, t)[2][0] for c, t in zip(ctx["continuaram"], ctx["total"])]
ctx["hi"] = [jeffreys(c, t)[2][1] for c, t in zip(ctx["continuaram"], ctx["total"])]
p_trabalho = CONTEXTO["no trabalho"][1] / CONTEXTO["no trabalho"][0]
p_familia = CONTEXTO["com/via família"][1] / CONTEXTO["com/via família"][0]
p_casa = CONTEXTO["em casa"][1] / CONTEXTO["em casa"][0]
# chance de um laço de trabalho ficar, contra um laço de família: razão simples e a posterior da diferença
amostra_trab = jeffreys(11, 130)[0].rvs(200000, random_state=ger)
amostra_fam = jeffreys(67, 194)[0].rvs(200000, random_state=ger)
p_fam_maior = (amostra_fam > amostra_trab).mean()

# 3. A sua rede de quatro, sete anos depois
# cada laço permanece com a probabilidade medida; a incerteza da proporção entra sorteando p da posterior
SIMS = 200000
REDE = 4
p_sim = d_mesma.rvs(SIMS, random_state=ger)
sobram = ger.binomial(REDE, p_sim)
dist_sobram = np.bincount(sobram, minlength=REDE + 1) / SIMS
p_zera = dist_sobram[0]
p_maioria_fica = dist_sobram[REDE // 2 + 1:].sum()      # 3 ou 4 de 4
esperado = REDE * m_mesma
# a mesma conta com "ficar na rede" (48%), a leitura generosa
p_sim48 = d_fica.rvs(SIMS, random_state=ger)
sobram48 = ger.binomial(REDE, p_sim48)
dist_sobram48 = np.bincount(sobram48, minlength=REDE + 1) / SIMS
p_zera48 = dist_sobram48[0]

# 4. Duas rodadas: catorze anos, supondo a mesma taxa (régua, não previsão)
p_14 = m_mesma ** 2
sobram14 = ger.binomial(sobram, d_mesma.rvs(SIMS, random_state=ger))
p_zera14 = (sobram14 == 0).mean()

# 6. Três países, três intervalos: a permanência por ano, como régua comum (taxa constante, declarada)
def perm_anual(p, anos):
    return p ** (1 / anos)
REGUA = {
    "Holanda, mesma função, 7 anos (Mollenhorst)": (m_mesma, 7),
    "Holanda, ainda na rede, 7 anos (Mollenhorst)": (m_fica, 7),
    "Califórnia, nome repetido, 1 ano (Fischer & Offer)": (1 - FO["p_nao_repetido"], 1),
    "Califórnia, perda real corrigida, 1 ano (Fischer & Offer)": (1 - FO["perda_real_max"], 1),
    "Toronto, íntimo segue íntimo, 10 anos (Wellman)": (WELLMAN["intimos_depois"] / WELLMAN["lacos"], 10),
}
regua_anual = {k: perm_anual(p, a) for k, (p, a) in REGUA.items()}
P_ESQUECIDO_NL = MOTIVOS["esqueci de mencionar, continua importante"] / N_SAIRAM_RESP

# 7. Quem entrou e quem saiu, por tipo de laço
comp = pd.DataFrame(COMPOSICAO, index=["1999/2000", "2007"]).T
comp["variacao"] = comp["2007"] - comp["1999/2000"]
comp["variacao_pct"] = comp["variacao"] / comp["1999/2000"]

print("=" * 70)
print("  MODELO — SETE ANOS DEPOIS")
print("=" * 70)
print(f"\n  1. De cada pessoa com quem você conta hoje, em sete anos:")
print(f"     ainda na mesma função:   {pct(m_mesma)}  IC 95% [{pct(ic_mesma[0])}; {pct(ic_mesma[1])}]")
print(f"     ainda na sua rede:       {pct(m_fica)}  IC 95% [{pct(ic_fica[0])}; {pct(ic_fica[1])}]")
print(f"     fora da rede:            {pct(1 - m_fica)}")
print(f"\n  2. Quem saiu da rede:")
print(f"     ainda em contato:        {pct(m_contato)}  [{pct(ic_contato[0])}; {pct(ic_contato[1])}]")
print(f"     saiu por falta de ocasião: {pct(m_ocasiao)}  [{pct(ic_ocasiao[0])}; {pct(ic_ocasiao[1])}]")
print(f"     saiu porque a relação mudou: {pct(m_briga)}  [{pct(ic_briga[0])}; {pct(ic_briga[1])}]")
print(f"     falta de ocasião / briga: {vir(razao_ocasiao_briga)} vezes")
print(f"\n  3. Por onde a pessoa foi conhecida:")
for _, r in ctx.sort_values("p", ascending=False).iterrows():
    print(f"     {r['contexto']:<26s} {pct(r['p'])}  [{pct(r['lo'])}; {pct(r['hi'])}]  ({r['continuaram']} de {r['total']})")
print(f"     laço de família fica mais que laço de trabalho em {pct(p_fam_maior)} das amostras da posterior")
print(f"     família / trabalho: {vir(p_familia / p_trabalho)} vezes; OR do trabalho no modelo com controles: {vir(OR_TRABALHO, 2)}")
print(f"\n  4. A sua rede de {REDE}, sete anos depois (mesma função, {mil(SIMS)} sorteios):")
for k in range(REDE + 1):
    print(f"     sobram {k}: {pct(dist_sobram[k])}")
print(f"     esperado: {vir(esperado)} pessoas; P(ninguém sobra) = {pct(p_zera)}; P(3 ou 4 sobram) = {pct(p_maioria_fica)}")
print(f"     na leitura generosa (ficar na rede, 48%): P(ninguém sobra) = {pct(p_zera48)}")
print(f"\n  5. Catorze anos, duas rodadas na mesma taxa (régua):")
print(f"     cada pessoa: {pct(p_14)}; P(ninguém dos {REDE} sobra) = {pct(p_zera14)}")
print(f"\n  6. Três países na mesma régua (permanência por ano, supondo taxa constante):")
for k, v in regua_anual.items():
    print(f"     {k:<58s} {pct(v)} ao ano")
print(f"     na Holanda, {pct(P_ESQUECIDO_NL)} dos que saíram foram 'esqueci de mencionar'; na Califórnia, {pct(FO['p_esquecido_entre_sumidos'], 0)}")
print(f"\n  7. Quem entrou e quem saiu (média de nomes por pessoa):")
for k, r in comp.iterrows():
    print(f"     {k:<10s} {vir(r['1999/2000'], 2)} → {vir(r['2007'], 2)}  ({'+' if r['variacao'] >= 0 else ''}{vir(r['variacao_pct'] * 100, 0)}%)")
print(f"     parceiro fica {vir(OR_PARCEIRO)} vezes mais que um 'outro'; parente, {vir(OR_PARENTE)} vez")
print("=" * 70)
```

### 💡 O Insight

**De cada dez pessoas com quem você conta hoje, três ainda farão o mesmo papel daqui a
sete anos. Cinco ainda estarão por perto. As outras terão saído, e a maioria sem nenhuma
briga.**

O ditado acertou uma metade. Os mais próximos ficam quando são família. O parceiro fica
4,5 vezes mais do que qualquer outro laço, e os parentes e vizinhos ganham espaço na
lista. Os mais próximos que a gente escolheu, os amigos e os colegas, são os que saem.
Não porque importam menos, mas porque o lugar onde a gente se via acaba primeiro.

O que o número não quer dizer é que as pessoas se perdem. Só 12% saem porque a relação
mudou. Seis em cada dez que saíram da lista ainda estavam em contato. E, na Califórnia,
quatro em cada dez nomes que tinham sumido tinham sido só esquecidos. Nesses casos, a
saudade não é de alguém que se foi. É do lugar que a pessoa ocupava. Às vezes nem isso,
é só da ocasião.

*Quem estava na sua lista há sete anos e hoje não está? Vocês brigaram, ou só ficaram
sem ocasião de se ver?*

### ⚠️ Limitações do Modelo

- A rede medida é a íntima, por duas perguntas (confidentes e ajuda prática). Amigos de
  outros tipos não entram, e os autores avisam que delimitar a rede por uma pergunta só
  exagera a rotatividade. As duas listas de 1999/2000 e de 2007 foram cruzadas mostrando
  ao entrevistado os nomes antigos, o que reduz esse erro.
- Holanda, 1999 a 2007, antes das redes sociais como as conhecemos. A proporção de quem
  sai da rede mas segue em contato pode ser outra hoje.
- O 18% "na rede com outra função" aparece no texto, não numa tabela; a contagem
  correspondente foi reconstruída, e o 48% é a soma declarada pelos autores.
- Os motivos de saída são autorrelatados pelo entrevistado, até três por laço, e a soma
  "falta de ocasião" é uma classificação dos autores.
- Fischer & Offer mostram que parte da rotatividade medida é esquecimento, não perda: na
  Califórnia, 40% dos nomes "sumidos" foram esquecidos, e a perda real fica abaixo de 20%
  ao ano. Na Holanda, 7,9% dos que saíram eram "esqueci de mencionar", e a lista antiga
  era mostrada ao entrevistado, o que reduz o erro; ainda assim, o 52% de saída é um
  teto, não um piso.
- A régua "por ano" supõe taxa constante, o que ignora que laços novos saem mais do que
  os antigos; serve para comparar países, não para prever.
- A simulação da rede de quatro trata cada pessoa como independente e usa a taxa média.
  Laços de família ficam mais, laços de trabalho muito menos; a média esconde isso.
- A rodada de catorze anos repete a mesma taxa, o que é régua, não previsão.
- Cordas com dias anteriores: o #083 (Wrzus et al. 2013, a rede encolhe com a idade) e o
  #024 (três a cinco amigos próximos). Este dia mede a troca de nomes, não o tamanho.

*A ciência é honesta sobre o que não sabe. O modelo também.*
