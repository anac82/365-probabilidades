### O Conceito: quando a intuicao merece credito

A briga entre intuicao e metodo tem uma resposta melhor do que "depende", e ela veio de um artigo escrito por dois pesquisadores que passaram a carreira discordando.

Daniel Kahneman defendia que a intuicao humana e cheia de vieses previsiveis. Gary Klein documentava bombeiros e enfermeiras tomando decisoes intuitivas brilhantes sob pressao.

Em 2009 eles escreveram juntos um artigo com o subtitulo mais honesto que ja vi: *A Failure to Disagree*. Um fracasso em discordar.

O que eles concluiram e que a pergunta "posso confiar na minha intuicao?" esta mal formulada. A intuicao nao e uma caracteristica sua. Ela e uma competencia especifica de cada assunto, e depende de duas condicoes:

**1. O ambiente precisa ter regularidade suficiente para ser previsivel.**

**2. Voce precisa ter tido oportunidade de aprender essa regularidade, com pratica longa e retorno rapido sobre seus acertos e erros.**

Xadrez tem as duas. Bombeiro em incendio estrutural tem as duas. Previsao de bolsa no longo prazo nao tem a primeira. Escolher com quem casar nao tem a segunda.

E o achado mais desconfortavel dos dois:

**a sua confianca subjetiva nao indica se voce esta certa.**

A sensacao de "eu sei" tem exatamente a mesma intensidade quando a intuicao foi treinada e quando ela e so barulho.

### O Modelo

A fonte principal e uma meta-analise que colocou especialistas humanos contra formulas, sobre os mesmos casos, em 136 estudos.

Modelo estatistico: distribuicao Beta sobre a proporcao de vitorias de cada lado, com IC 95%, mais a distribuicao dos 136 tamanhos de efeito com mediana e quartis.

**Fontes:**
- Grove, W. M., Zald, D. H., Lebow, B. S., Snitz, B. E., & Nelson, C. (2000). Clinical versus mechanical prediction: A meta-analysis. *Psychological Assessment*, 12(1), 19-30.
  N = 136 estudos comparando julgamento clinico contra previsao mecanica.
- Kahneman, D., & Klein, G. (2009). Conditions for Intuitive Expertise: A Failure to Disagree. *American Psychologist*, 64(6), 515-526.
  Artigo teorico. Entra como conceito, nao como fonte numerica.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Grove, Zald, Lebow, Snitz & Nelson (2000)
# Psychological Assessment, 12(1), 19-30

n_estudos = 136

# Placar: cada estudo compara especialista humano vs formula, sobre os mesmos casos
# Criterio do artigo: ES < -0.1 favorece o clinico
#                     ES entre -0.1 e 0.1 = equivalentes
#                     ES > 0.1 favorece a formula
k_formula_vence = 63       # 47% dos estudos
k_empate = 65              # 48%
k_intuicao_vence = 8       # 6%

# Distribuicao dos tamanhos de efeito (positivo favorece a formula)
es_medio_ponderado = 0.086
es_dp = 0.12
es_mediana = 0.080
es_q1 = -0.008
es_q3 = 0.20
es_minimo = -0.30          # estudo onde a intuicao mais ganhou
es_maximo = 0.74           # estudo onde a formula mais ganhou

# Heterogeneidade
QT = 1635.2
gl_QT = 135

# Ganho medio de acuracia da formula reportado no artigo
ganho_acuracia_formula = 0.10   # cerca de 10%

# NOTA METODOLOGICA: fator 0.80 NAO se aplica.
# E meta-analise de tamanhos de efeito, nao proporcao de survey autorrelatado.
fator_correcao = None

print("=" * 68)
print("  DADOS DA LITERATURA - INTUICAO CONTRA FORMULA")
print("=" * 68)
print(f"\n  Grove et al. (2000) - Psychological Assessment 12(1)")
print(f"  {n_estudos} estudos, mesmos casos julgados pelos dois metodos")
print(f"\n  O PLACAR")
print(f"  {'-'*54}")
print(f"  A formula venceu claramente     {k_formula_vence:>4} estudos   {k_formula_vence/n_estudos*100:>5.1f}%")
print(f"  Empate tecnico                  {k_empate:>4} estudos   {k_empate/n_estudos*100:>5.1f}%")
print(f"  A intuicao venceu claramente    {k_intuicao_vence:>4} estudos   {k_intuicao_vence/n_estudos*100:>5.1f}%")
print(f"  {'-'*54}")
print(f"  Total                           {n_estudos:>4} estudos")
print(f"\n  A formula venceu {k_formula_vence/k_intuicao_vence:.1f} vezes mais do que a intuicao.")
print(f"\n  TAMANHOS DE EFEITO (positivo = formula melhor)")
print(f"  Media ponderada: {es_medio_ponderado:.3f}  (DP {es_dp:.2f})")
print(f"  Mediana:         {es_mediana:.3f}")
print(f"  Quartis:         [{es_q1:.3f} , {es_q3:.2f}]")
print(f"  Amplitude:       [{es_minimo:.2f} , {es_maximo:.2f}]")
print(f"\n  Heterogeneidade: QT = {QT}, gl = {gl_QT}, p < 0.0001")
print(f"  Os estudos NAO sao todos iguais. Existem contextos")
print(f"  em que a intuicao vai bem, e o modelo precisa dizer isso.")
print(f"\n  Ganho medio de acuracia da formula: cerca de {ganho_acuracia_formula*100:.0f}%")
print(f"\n  Fator 0.80: NAO aplicado (meta-analise de tamanhos de efeito)")
print("=" * 68)
```

```python
# --- O MODELO ---
# Distribuicao Beta sobre cada proporcao do placar, com IC 95%

def beta_placar(k, n):
    d = stats.beta(k, n - k)
    return d, d.interval(0.95)

beta_intuicao, ic_intuicao = beta_placar(k_intuicao_vence, n_estudos)
beta_formula, ic_formula = beta_placar(k_formula_vence, n_estudos)
beta_empate, ic_empate = beta_placar(k_empate, n_estudos)

p_intuicao = k_intuicao_vence / n_estudos
p_formula = k_formula_vence / n_estudos
p_empate = k_empate / n_estudos

# Probabilidade de a intuicao NAO perder (vencer ou empatar)
p_nao_perde = (k_intuicao_vence + k_empate) / n_estudos
beta_nao_perde, ic_nao_perde = beta_placar(k_intuicao_vence + k_empate, n_estudos)

# Quantas vezes voce precisaria disputar para vencer pelo menos uma vez
# P(pelo menos uma vitoria em m disputas) = 1 - (1 - p)^m
ms = np.array([1, 5, 10, 20, 50])
p_ao_menos_uma = 1 - (1 - p_intuicao) ** ms

print("=" * 68)
print("  MODELO - A CHANCE DA INTUICAO")
print("=" * 68)
print(f"\n  ASSINATURA ESTATISTICA")
print(f"  Beta(k, n-k) sobre o placar de {n_estudos} estudos")
print(f"\n  {'':<26} | {'estimativa':>10} | {'IC 95%':>18}")
print(f"  {'-'*26}-+-{'-'*10}-+-{'-'*18}")
print(f"  {'A formula vence':<26} | {p_formula*100:>9.1f}% | [{ic_formula[0]*100:>5.1f}%, {ic_formula[1]*100:>5.1f}%]")
print(f"  {'Empate':<26} | {p_empate*100:>9.1f}% | [{ic_empate[0]*100:>5.1f}%, {ic_empate[1]*100:>5.1f}%]")
print(f"  {'A INTUICAO VENCE':<26} | {p_intuicao*100:>9.1f}% | [{ic_intuicao[0]*100:>5.2f}%, {ic_intuicao[1]*100:>5.2f}%]")
print(f"\n  Mesmo no cenario mais generoso do intervalo,")
print(f"  a intuicao vence em {ic_intuicao[1]*100:.1f}% das vezes.")
print(f"\n  A INTUICAO NAO PERDER (vencer ou empatar)")
print(f"  {p_nao_perde*100:.1f}%  IC 95% [{ic_nao_perde[0]*100:.1f}%, {ic_nao_perde[1]*100:.1f}%]")
print(f"  Esse e o numero honesto: na maioria das vezes a intuicao")
print(f"  nao e pior. Ela so raramente e melhor.")
print(f"\n  P(pelo menos uma vitoria da intuicao em m disputas)")
for m, p in zip(ms, p_ao_menos_uma):
    print(f"    m = {m:>3}  ->  {p*100:>5.1f}%")
print("=" * 68)
```

```python
# --- VISUALIZACAO ---

# GRAFICO 1 - O placar dos 136 estudos
fig1, ax1 = plt.subplots(figsize=(10, 6))

rotulos = ['A FORMULA\nvenceu', 'EMPATE', 'A INTUICAO\nvenceu']
valores = [k_formula_vence, k_empate, k_intuicao_vence]
cores = ['#1a5f5a', '#c8a84b', '#c0392b']

bars = ax1.bar(rotulos, valores, color=cores, alpha=0.88, width=0.5)
ax1.set_ylabel('Numero de estudos')
ax1.set_ylim(0, 82)
ax1.set_title(f'O placar de {n_estudos} estudos\nGrove, Zald, Lebow, Snitz & Nelson, 2000 - Psychological Assessment',
              fontsize=13, pad=15)

for bar, valor in zip(bars, valores):
    ax1.text(bar.get_x() + bar.get_width()/2, valor + 1.8,
             f'{valor}', ha='center', fontweight='bold', fontsize=22)
    ax1.text(bar.get_x() + bar.get_width()/2, valor + 8,
             f'{valor/n_estudos*100:.0f}%', ha='center', fontsize=13, color='gray')

ax1.text(2, 30, 'em 136 disputas,\na intuicao ganhou\n8 vezes',
         ha='center', fontsize=12, color='#c0392b', fontweight='bold')

plt.figtext(0.5, 0.005,
            'Fonte: Grove et al., 2000 - Psychological Assessment 12(1), 19-30 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-039-grafico-01-placar.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 1 salvo")

# GRAFICO 2 - Distribuicao dos tamanhos de efeito
fig2, ax2 = plt.subplots(figsize=(10, 6))

grid_es = np.linspace(es_minimo - 0.05, es_maximo + 0.05, 900)
curva = stats.norm(es_medio_ponderado, es_dp).pdf(grid_es)

ax2.plot(grid_es, curva, color='#1a5f5a', linewidth=2.5)
ax2.fill_between(grid_es, curva, where=(grid_es > 0), alpha=0.22, color='#1a5f5a')
ax2.fill_between(grid_es, curva, where=(grid_es <= 0), alpha=0.22, color='#c0392b')

ax2.axvline(x=0, color='#333', linewidth=2)
ax2.axvline(x=es_mediana, color='#1a5f5a', linewidth=2, linestyle=':')
ax2.axvline(x=es_q1, color='#c8a84b', linestyle='--', linewidth=1.8)
ax2.axvline(x=es_q3, color='#c8a84b', linestyle='--', linewidth=1.8)

topo = curva.max()
ax2.text(0.008, topo*0.97, 'empate', fontsize=11, color='#333')
ax2.text(es_mediana + 0.012, topo*0.80, f'mediana {es_mediana:.2f}',
         fontsize=11, color='#1a5f5a', fontweight='bold')
ax2.text(es_q3 + 0.012, topo*0.55, f'quartis\n[{es_q1:.2f} , {es_q3:.2f}]',
         fontsize=10, color='#c8a84b')
ax2.text(-0.25, topo*0.30, 'intuicao\nmelhor', fontsize=11, color='#c0392b',
         ha='center', fontweight='bold')
ax2.text(0.55, topo*0.30, 'formula\nmelhor', fontsize=11, color='#1a5f5a',
         ha='center', fontweight='bold')

ax2.set_xlabel('Tamanho de efeito (positivo favorece a formula)')
ax2.set_ylabel('Densidade')
ax2.set_title(f'A distribuicao inteira, nao so a media\nMedia ponderada {es_medio_ponderado:.3f} (DP {es_dp:.2f}) - amplitude de {es_minimo:.2f} a {es_maximo:.2f}',
              fontsize=13, pad=15)

plt.figtext(0.5, 0.005,
            f'Heterogeneidade QT = {QT}, gl = {gl_QT}, p < 0.0001 - os estudos nao sao equivalentes | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-039-grafico-02-efeitos.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 2 salvo")

# GRAFICO 3 - Assinatura estatistica: Beta da vitoria da intuicao
fig3, ax3 = plt.subplots(figsize=(10, 6))

grid_p = np.linspace(0.005, 0.20, 1200)
ax3.plot(grid_p * 100, beta_intuicao.pdf(grid_p), color='#c0392b', linewidth=3)
ax3.fill_between(grid_p * 100, beta_intuicao.pdf(grid_p), alpha=0.22, color='#c0392b')

ax3.axvline(x=ic_intuicao[0]*100, color='#c8a84b', linestyle='--', linewidth=2)
ax3.axvline(x=ic_intuicao[1]*100, color='#c8a84b', linestyle='--', linewidth=2)
ax3.axvline(x=p_intuicao*100, color='#333', linewidth=2)

topo3 = beta_intuicao.pdf(grid_p).max()
ax3.text(p_intuicao*100 + 0.4, topo3*0.90,
         f'{p_intuicao*100:.1f}%\n8 de 136', fontsize=14,
         color='#333', fontweight='bold')
ax3.text(ic_intuicao[1]*100 + 0.4, topo3*0.45,
         f'IC 95%\n[{ic_intuicao[0]*100:.1f}% , {ic_intuicao[1]*100:.1f}%]',
         fontsize=11, color='#c8a84b')

ax3.set_xlabel('Proporcao de estudos em que a intuicao venceu (%)')
ax3.set_ylabel('Densidade')
ax3.set_title(f'Assinatura estatistica - Beta({k_intuicao_vence}, {n_estudos - k_intuicao_vence})\n'
              f'Mesmo no teto do intervalo, a intuicao vence em {ic_intuicao[1]*100:.1f}% das disputas',
              fontsize=13, pad=15)

plt.figtext(0.5, 0.005,
            'Derivado do placar de Grove et al., 2000 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-039-grafico-03-bayesiano.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 3 salvo")
```

### O Insight

**Em 136 disputas entre o especialista e a formula, a intuicao venceu 8 vezes.**

Cinco virgula nove por cento. Com IC 95% de [2,6% , 10,4%]. Mesmo no cenario mais generoso do intervalo, a intuicao ganha em uma disputa a cada dez.

Mas tem um numero que quase ninguem cita, e ele muda a conversa.

Em 65 dos 136 estudos deu empate. Somando com as 8 vitorias, **a intuicao nao foi pior em 53,7% das vezes**.

Ou seja: o problema da intuicao nao e ela errar sempre. Ela quase nunca e catastrofica.

O problema e que ela quase nunca e melhor. E ela custa mais caro, porque cansa, hesita, e muda de opiniao conforme o seu dia.

Agora, a parte que salva.

Kahneman e Klein mostraram que a intuicao nao e um traco seu. Ela e uma competencia especifica de cada assunto, e ela precisa de duas coisas para existir: um ambiente com regularidade, e tempo suficiente de pratica com retorno rapido sobre erros e acertos.

Isso reformula a pergunta inteira.

Nao e "eu sou uma pessoa intuitiva ou analitica".

E **"a minha intuicao ja viveu esta situacao vezes suficientes para ter aprendido alguma coisa?"**

Minha intuicao sobre risco de credito e otima. Dezoito anos de repeticao, milhares de casos, retorno claro sobre cada erro. Ambiente regular, aprendizado possivel.

Minha intuicao sobre virar a propria vida do avesso nao tem esse treino. Isso acontece uma vez. Nao tem repeticao, nao tem feedback rapido, nao tem historico para calibrar.

E o detalhe cruel: nos dois casos a sensacao de certeza e identica.

Voce nao consegue sentir a diferenca entre uma intuicao treinada e um palpite bonito. Por dentro, os dois dao o mesmo aperto no peito.

A pergunta que fica:

*A sua intuicao ja viveu essa situacao vezes suficientes para ter aprendido alguma coisa?*

### Limitacoes do Modelo

- Grove et al. compara especialistas contra formulas em tarefas de previsao (diagnostico, prognostico, desempenho). Nao mediu decisoes pessoais de vida. Aplicar o resultado a "seguir a intuicao" no cotidiano e uma extrapolacao minha, e ela precisa ser lida como analogia, nao como medida direta.
- A heterogeneidade e altissima: QT = 1635,2 com 135 graus de liberdade, p < 0,0001. Os 136 estudos nao sao equivalentes entre si. A media esconde contextos muito diferentes, e por isso o grafico 2 mostra a distribuicao inteira e nao so a media.
- A classificacao em vitoria, empate e derrota usa os cortes do proprio artigo (efeito abaixo de -0,1, entre -0,1 e 0,1, e acima de 0,1). Cortes diferentes dariam placares diferentes.
- As distribuicoes Beta tratam os 136 estudos como observacoes independentes de mesmo peso, o que e uma simplificacao: estudos tem tamanhos de amostra diferentes.
- Kahneman & Klein (2009) e artigo teorico, de revisao e discussao. Entra aqui como arcabouco conceitual, nao como fonte de dado. As duas condicoes que eles propoem sao qualitativas, nao tem limiar numerico.
- Meta-analise publicada em 2000, reunindo estudos anteriores. Modelos preditivos mudaram muito desde entao, provavelmente ampliando a vantagem da formula.
