### 📖 A História

Tem gente que carrega o ano da formatura como se fosse uma sentença.

Formou em ano de crise. Aceitou o que tinha. Ganhou menos do que a turma
que saiu dois anos antes. E arquivou a conclusão de que começou perdendo
e vai terminar perdendo.

A primeira parte é verdade. A segunda, não.

Um dos estudos mais completos já feitos sobre isso seguiu formados por
duas décadas, cruzando universidade, empregador e imposto de renda.
E encontrou as duas coisas: o azar de entrada existe, é caro,
e tem prazo de validade.

Quem começou no ano errado perdeu cerca de **9% dos ganhos** no início.
Metade dessa cicatriz sumiu em 5 anos.
Em 8 a 10 anos, ela desapareceu por completo.

E o mecanismo da recuperação não foi esperar.
Foi se mover.

A sorte de entrada não é sentença. É condição inicial.
E condição inicial, em qualquer sistema dinâmico, é o que o movimento corrige.

---

### 📚 O Conceito: Histerese e a Meia-Vida do Azar

Economistas chamam de **histerese** quando um choque temporário deixa
efeito permanente. A pergunta do estudo é exatamente essa: recessão na
formatura é choque que passa ou marca que fica?

A resposta média: passa. O perfil de perda decai ano a ano, como uma
meia-vida. E cerca de **30% da reversão** vem de um único comportamento:
trocar de empresa. Quem se forma em recessão começa em empresas menores,
que pagam menos, e recupera migrando para empresas melhores.

A resposta honesta: passa **na média**. Formados mais favorecidos trocam
de empresa rápido e saram antes; nos menos favorecidos, a perda pode
persistir. O custo é real e desigual, e essa desigualdade fica visível
no gráfico 3, não escondida no rodapé.

No Dia #053 a sorte decidia quem chega ao topo. Hoje é o outro lado da
mesma moeda: a má sorte de entrada tem data para perder o efeito,
desde que exista movimento.

### 🧮 O Modelo

**Fontes:**
- Oreopoulos, P., von Wachter, T. & Heisz, A. (2012) - The Short- and
  Long-Term Career Effects of Graduating in a Recession - American
  Economic Journal: Applied Economics, 4(1), 1-29 - base pareada
  universidade-empregador-empregado cobrindo cerca de 70% dos formados
  universitários canadenses de 1976 a 1995, com registros fiscais e de
  folha de 1982 a 1999. **Fonte central.** N individual único não é
  publicado no artigo; a cobertura administrativa é declarada como
  medida de tamanho (ver Limitações).

**Âncoras publicadas:** alta de 5 p.p. no desemprego na formatura gera
perda inicial de ~9% nos ganhos; metade da perda desaparece em ~5 anos;
os coeficientes deixam de ser estatisticamente distinguíveis de zero
após 8 a 10 anos; mobilidade entre empresas explica ~30% da reversão.

**Fator ×0,80: não se aplica.** Registros administrativos de imposto e
folha de pagamento, medida objetiva. Sem autorrelato, sem correção.

**Suposição declarada:** o decaimento exponencial abaixo é uma
aproximação estilizada ajustada às três âncoras publicadas
(t=0: -9%; t=5: -4,5%; t=8 a 10: indistinguível de zero). O artigo
reporta coeficientes de estudo de evento por ano de experiência, não
uma curva paramétrica. A exponencial nunca toca o zero; o que zera é a
significância estatística.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Oreopoulos, von Wachter & Heisz (2012), AEJ: Applied, 4(1), 1-29
# Base administrativa canadense, 1976-1999, ~70% dos formados universitários

choque_pp = 5                 # alta de 5 p.p. no desemprego na formatura
perda_inicial = 0.09          # ~9% de perda inicial nos ganhos
anos_meia_vida = 5            # metade da perda some em ~5 anos
anos_zero_estatistico = 10    # coeficientes indistinguíveis de zero em 8-10 anos
fracao_mobilidade = 0.30      # ~30% da reversão vem de troca de empresa

# Fator de correção: NÃO se aplica
# (registro administrativo objetivo de imposto e folha, sem autorrelato)

print("=" * 65)
print("  DADOS DA LITERATURA - FORMAR-SE EM RECESSÃO")
print("=" * 65)
print(f"\n  Oreopoulos, von Wachter & Heisz (2012):")
print(f"  → Choque analisado:                 +{choque_pp} p.p. de desemprego")
print(f"  → Perda inicial nos ganhos:         -{perda_inicial*100:.0f}%")
print(f"  → Metade da perda desaparece em:    ~{anos_meia_vida} anos")
print(f"  → Zero estatístico em:              8 a {anos_zero_estatistico} anos")
print(f"  → Reversão explicada por mudar de empresa: ~{fracao_mobilidade*100:.0f}%")
print(f"\n  Fator ×0,80: NÃO aplicado (medida objetiva, registro fiscal)")
print("=" * 65)
```

```python
# --- O MODELO ---
# A meia-vida do azar: decaimento exponencial ajustado às âncoras publicadas
# SUPOSIÇÃO DECLARADA: aproximação estilizada; o artigo reporta coeficientes
# de estudo de evento, não uma curva paramétrica.

# perda(t) = perda_inicial * exp(-lambda * t), com meia-vida de 5 anos
lam = np.log(2) / anos_meia_vida

t = np.arange(0, 13)
perda_t = perda_inicial * np.exp(-lam * t)

# Custo acumulado da década (integral da curva de 0 a 10 anos),
# em pontos percentuais-ano de salário
custo_acumulado = perda_inicial / lam * (1 - np.exp(-lam * anos_zero_estatistico))

print("=" * 65)
print("  MODELO - A MEIA-VIDA DO AZAR DE ENTRADA")
print("=" * 65)
print(f"\n  Equação: perda(t) = 9% · e^(-λt),  λ = ln(2)/5 = {lam:.4f}")
print(f"\n  Perda estimada por ano de experiência:")
for ano in [0, 1, 3, 5, 8, 10]:
    marca = "  ← metade" if ano == anos_meia_vida else ("  ← zero estatístico" if ano == anos_zero_estatistico else "")
    print(f"  → Ano {ano:>2}: -{perda_inicial*np.exp(-lam*ano)*100:4.1f}%{marca}")
print(f"\n  Custo acumulado da década (área sob a curva):")
print(f"  → {custo_acumulado*100:.0f} p.p.-ano ≈ metade de um salário anual,")
print(f"     diluída ao longo de 10 anos")
print(f"\n  Do que é feita a recuperação:")
print(f"  → ~{fracao_mobilidade*100:.0f}% de trocar de empresa (mobilidade)")
print(f"  → ~{(1-fracao_mobilidade)*100:.0f}% de crescimento restante (dentro da firma e outros)")
print(f"\n  LEITURA HONESTA:")
print(f"  → A média sara em 10 anos. A cauda inferior pode não sarar.")
print(f"  → Recuperação não é espera: é movimento entre empresas.")
print("=" * 65)
```

```python
# --- VISUALIZAÇÃO - GRÁFICOS SEPARADOS ---

# ── GRÁFICO 1 - A curva de recuperação ──
fig1, ax1 = plt.subplots(figsize=(10, 6))

t_fino = np.linspace(0, 12, 500)
curva = perda_inicial * np.exp(-lam * t_fino) * 100

ax1.plot(t_fino, -curva, color='#c0392b', linewidth=3, zorder=3)
ax1.fill_between(t_fino, -curva, 0, alpha=0.12, color='#c0392b')
ax1.axhline(0, color='#333', linewidth=1)

ancoras_t = [0, 5, 10]
ancoras_v = [-9.0, -4.5, -perda_inicial*np.exp(-lam*10)*100]
ax1.scatter(ancoras_t, ancoras_v, s=90, color='#c9a14a', zorder=4,
            edgecolors='#333', linewidths=1)

ax1.annotate('-9%\nno começo', xy=(0, -9), xytext=(1.2, -8.6),
             fontsize=11, fontweight='bold', color='#c0392b')
ax1.annotate('metade em 5 anos', xy=(5, -4.5), xytext=(5.4, -5.6),
             fontsize=11, color='#333')
ax1.annotate('zero estatístico\nem 8-10 anos', xy=(10, ancoras_v[2]),
             xytext=(8.2, -3.2), fontsize=11, color='#2a8a82',
             fontweight='bold')

ax1.set_xlabel('Anos de experiência desde a formatura')
ax1.set_ylabel('Perda nos ganhos (%)')
ax1.set_ylim(-10.5, 1.0)
ax1.set_title('A meia-vida do azar: perda(t) = 9% · e^(-λt), λ = ln(2)/5\n'
              'Curva estilizada sobre as âncoras de Oreopoulos et al. (2012)',
              fontsize=12.5, pad=15)

plt.figtext(0.5, 0.01,
            'Fonte: Oreopoulos, von Wachter & Heisz (2012), AEJ: Applied 4(1) | curva: suposição declarada | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-086-grafico-01-meia-vida.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 - Do que é feita a recuperação ──
fig2, ax2 = plt.subplots(figsize=(10, 6))

partes = ['Trocar de empresa\n(mobilidade)', 'Crescimento restante\n(dentro da firma e outros)']
valores = [fracao_mobilidade*100, (1-fracao_mobilidade)*100]
cores = ['#2a8a82', '#c9a14a']

bars = ax2.barh(partes, valores, color=cores, alpha=0.88, height=0.45)
ax2.set_xlim(0, 80)
ax2.set_xlabel('Participação na reversão da perda inicial (%)')
ax2.set_title('A recuperação não é espera: é movimento\n'
              'Decomposição da reversão em Oreopoulos et al. (2012)',
              fontsize=12.5, pad=15)

for bar, v in zip(bars, valores):
    ax2.text(v + 1.5, bar.get_y() + bar.get_height()/2,
             f'~{v:.0f}%', va='center', fontweight='bold', fontsize=15)

ax2.text(0.98, 0.06,
         'Quem se forma em recessão começa em empresas menores,\n'
         'que pagam menos, e recupera migrando para empresas melhores.',
         transform=ax2.transAxes, ha='right', fontsize=10, color='#555',
         style='italic')

plt.figtext(0.5, 0.01,
            'Fonte: Oreopoulos, von Wachter & Heisz (2012) | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-086-grafico-02-mecanismo.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 - A média sara, a desigualdade fica ──
# ESTILIZADO E DECLARADO: o artigo mostra recuperação rápida no topo da
# distribuição de ganhos previstos e perda persistente na base.
fig3, ax3 = plt.subplots(figsize=(10, 6))

lam_topo = np.log(2) / 3      # topo: recuperação mais rápida (estilizado)
curva_topo = perda_inicial * np.exp(-lam_topo * t_fino) * 100
curva_media = perda_inicial * np.exp(-lam * t_fino) * 100
curva_base = (4.0 + 5.0 * np.exp(-lam * t_fino))   # base: não converge a zero (estilizado)

ax3.plot(t_fino, -curva_topo, color='#2a8a82', linewidth=2.5,
         label='Mais favorecidos (recuperam antes)')
ax3.plot(t_fino, -curva_media, color='#c9a14a', linewidth=2.5,
         label='Média (zero em 8-10 anos)')
ax3.plot(t_fino, -curva_base, color='#c0392b', linewidth=2.5, linestyle='--',
         label='Menos favorecidos (perda pode persistir)')
ax3.axhline(0, color='#333', linewidth=1)

ax3.set_xlabel('Anos de experiência desde a formatura')
ax3.set_ylabel('Perda nos ganhos (%)')
ax3.set_ylim(-10.5, 1.0)
ax3.legend(loc='lower right', frameon=False, fontsize=10.5)
ax3.set_title('O custo é substancial e desigual\n'
              'Curvas ESTILIZADAS sobre a heterogeneidade de Oreopoulos et al. (2012)',
              fontsize=12.5, pad=15)

plt.figtext(0.5, 0.01,
            'Fonte: Oreopoulos, von Wachter & Heisz (2012) | curvas estilizadas, suposição declarada | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-086-grafico-03-desigualdade.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**Começar no ano errado custa 9%. E some em dez anos.**

Metade da cicatriz desaparece em cinco.
O resto, entre oito e dez.

O custo total da década, somado, equivale a cerca de
**metade de um salário anual**, diluída em dez anos.
É caro. Não é sentença.

E a parte que ninguém conta: a recuperação tem método.
Cerca de **30% dela vem de um único movimento: trocar de empresa.**
Quem se formou em recessão começou em empresas menores, que pagavam
menos. Quem sarou mais rápido não esperou o mercado melhorar.
Migrou.

A leitura honesta: a média sara, e a média esconde gente.
Formados com mais recursos trocaram de empresa rápido e saram antes.
Nos menos favorecidos, a perda pode ficar.
O azar de entrada tem prazo de validade, mas o prazo não corre
igual para todo mundo.

Condição inicial não é destino.
É só o ponto de onde o movimento parte.

A pergunta que fica:

*Qual condição inicial você ainda está tratando como sentença,
dez anos depois de ela ter vencido?*

### ⚠️ Limitações do Modelo

- Base canadense, formados universitários, 1976 a 1999: generalização a
  outros países, níveis de formação e décadas é limitada, e o contexto
  institucional (mobilidade, proteção social) difere do Brasil
- O artigo não publica um N individual único; a cobertura declarada é de
  ~70% dos formados universitários canadenses do período, cruzada com
  registros fiscais de 1982 a 1999
- A curva exponencial dos gráficos 1 e 3 é aproximação estilizada das
  âncoras publicadas; o estudo reporta coeficientes de estudo de evento
  por ano de experiência, com incerteza que a curva não mostra
- "Zero em 8-10 anos" é zero estatístico: os coeficientes deixam de ser
  distinguíveis de zero, o que não é o mesmo que perda exatamente nula
- A recuperação média esconde heterogeneidade: no grupo de menor ganho
  previsto, o próprio artigo documenta perdas que podem ser permanentes
- Estudo observacional com identificação por variação regional e de
  coorte: linguagem causal aqui é mais defensável que em coorte simples,
  mas segue dependente das hipóteses de identificação

*A ciência é honesta sobre o que não sabe. O modelo também.*
