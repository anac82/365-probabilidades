## Qual a probabilidade de a felicidade estar na mala ainda fechada?

**Tipo:** Bem-estar
**Data de publicação:** 2026-09-12
**Ferramenta:** Python
**Decisão analisada:** Se o depois não sustenta, o que exatamente eu compro quando compro uma viagem?
**Hashtag:** #365Probabilidades #Dia091

---

### 📖 A História

Existe um tipo de volta que ninguém posta.

A mala fica mais leve e mais pesada ao mesmo tempo. O assento ao
lado, vazio. Eu conheço essa volta: é a de visitar um filho que foi morar
em outro país e voltar para casa sozinha.

Nessa volta, a ciência das férias parece uma piada de mau gosto.
Mas foi exatamente nela que eu fui entender o que um estudo
holandês mediu em 1.530 pessoas: **a felicidade da viagem não mora
no depois.** O depois, para quase todo mundo, é estatisticamente
igual a nunca ter viajado.

A felicidade mora no antes.

Quem tem uma viagem marcada é mensuravelmente mais feliz do que
quem não tem, semanas antes de fechar a mala. A antecipação é o
único pedaço da viagem que paga adiantado.

E isso muda o que se faz com a dor da volta: não dá para esticar
o depois. Mas dá para marcar o próximo antes.

---

### 📚 O Conceito: A Assimetria da Antecipação

A teoria do ponto de ajuste diz que a felicidade tem um nível
basal teimoso: eventos a empurram para cima ou para baixo, e ela
volta. As férias são o exemplo de manual: a rotina reabsorve tudo
em semanas.

O que o estudo de Nawijn mostrou é que a reabsorção é assimétrica:

**O antes funciona.** Viajantes carregam felicidade extra antes da
viagem, provavelmente pela antecipação: o roteiro, o countdown, a
mala mental sendo feita.

**O depois quase não funciona.** Na volta, viajantes e não
viajantes são indistinguíveis. A única exceção: viagens avaliadas
como muito relaxadas seguram um brilho por até duas semanas, e em
oito ele desaparece por completo.

**E a duração é irrelevante.** Férias longas não deixam mais feliz
que curtas. Nem o tamanho da viagem, nem os dias desde a volta:
nada disso move o ponteiro.

No Dia #047, o paradoxo das férias era outro: a memória estica a
duração em retrospecto. Hoje é o complemento: a felicidade encolhe
na mesma volta em que a memória estica.

---

### 🧮 O Modelo

**Fontes:**
- Nawijn, J., Marchand, M. A., Veenhoven, R. & Vingerhoets, A. J.
  (2010) - Vacationers Happier, but Most not Happier After a
  Holiday - Applied Research in Quality of Life, 5(1), 35-47,
  acesso aberto - desenho pré/pós com **N=1.530 holandeses**
  (974 viajantes, 556 não viajantes). Pré-viagem: viajantes
  M=2,25 (EP=0,08) contra M=2,07 (EP=0,09) dos que ficaram,
  F(3,531)=3,39, p=0,018, **η²=0,02**. Pós-viagem: sem diferença geral. Viagem muito
  relaxada: M=2,64, com brilho de até 2 semanas, extinto em 8.
  Duração da estadia: sem efeito. **Fonte central.**

**Fator ×0,80: não se aplica.** A âncora é comparação de médias
com tamanho de efeito (η²), não proporção de survey.

**Honestidade estrutural do dia:** η²=0,02 é minúsculo e vai
declarado como minúsculo, no texto e no gráfico. O dia não vende
milagre da antecipação; vende a assimetria: o pouco que as férias
dão de felicidade, dão antes.

**Suposições declaradas do modelo:** (1) a posterior bayesiana do
degrau usa as médias e erros-padrão publicados com aproximação
normal e prior plana; o teste original do artigo é multivariado
(p=0,018), e a aproximação fica declarada. (2) A aritmética das
antecipações assume que o degrau de +0,18 vale durante a janela de
antecipação; como o artigo não publica o comprimento dessa janela,
ela vira parâmetro de sensibilidade (2, 4 e 8 semanas). (3) A curva
de decaimento do gráfico 2 é estilizada sobre as âncoras publicadas.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Nawijn, Marchand, Veenhoven & Vingerhoets (2010), ARQOL 5(1), 35-47

n_total = 1530
n_viajantes, n_ficaram = 974, 556

# Pré-viagem (felicidade média)
m_pre_viajantes = 2.25    # quem tem viagem marcada
m_pre_ficaram   = 2.07    # quem não tem
eta2_pre = 0.02           # minúsculo, e declarado como minúsculo
p_pre = 0.018

# Pós-viagem: sem diferença geral entre viajantes e não viajantes

# A exceção: por tipo de viagem (pós-viagem)
m_muito_relaxada = 2.64
m_relaxada       = 2.16
m_estresse_neutra = 1.07  # ATENÇÃO: subgrupos pequenos, sem conclusão firme
m_pre_muito_relaxada = 2.48  # basal do grupo muito relaxada
semanas_brilho, semanas_extincao = 2, 8

# Os nulos elegantes
# duração da estadia: F(8,718)=1,59, n.s. → férias longas não rendem mais
# dias desde a volta: F(10,718)=0,90, n.s.

print("=" * 65)
print("  DADOS DA LITERATURA - A VIAGEM ANTES DA VIAGEM")
print("=" * 65)
print(f"\n  Nawijn et al. (2010): N={n_total} ({n_viajantes} viajantes, {n_ficaram} ficaram)")
print(f"\n  ANTES da viagem:")
print(f"  → Com viagem marcada:   M = {m_pre_viajantes}")
print(f"  → Sem viagem marcada:   M = {m_pre_ficaram}")
print(f"  → p = {p_pre} · η² = {eta2_pre}  ← minúsculo, e declarado")
print(f"\n  DEPOIS da viagem:")
print(f"  → Viajantes × não viajantes: SEM diferença")
print(f"\n  A exceção (pós, por tipo de viagem):")
print(f"  → Muito relaxada:  M = {m_muito_relaxada}  (brilho de ~{semanas_brilho} semanas,")
print(f"     extinto em {semanas_extincao}; basal do grupo: {m_pre_muito_relaxada})")
print(f"  → Relaxada:        M = {m_relaxada}")
print(f"  → Estressante/neutra: M = {m_estresse_neutra}  (subgrupos pequenos, sem conclusão)")
print(f"\n  Os nulos elegantes:")
print(f"  → Duração da estadia: sem efeito (férias longas não rendem mais)")
print(f"  → Dias desde a volta: sem efeito")
print(f"\n  Fator ×0,80: NÃO aplicado (comparação de médias, η²)")
print("=" * 65)
```

```python
# --- O MODELO ---
# Parte 1: posterior bayesiana do degrau da antecipação
# Parte 2: a aritmética das antecipações (datas contra dias)

# ═══ PARTE 1 - BAYES: o degrau é real? ═══
# Aproximação normal com as médias e EPs publicados, prior plana.
# (O teste original é multivariado, p=0,018; aproximação declarada.)
ep_viajantes, ep_ficaram = 0.08, 0.09
delta = m_pre_viajantes - m_pre_ficaram           # 0,18
ep_delta = np.sqrt(ep_viajantes**2 + ep_ficaram**2)

p_degrau_real = stats.norm.cdf(delta / ep_delta)  # P(δ > 0 | dados)
ic_crivel = (delta - 1.96*ep_delta, delta + 1.96*ep_delta)

# ═══ PARTE 2 - A ARITMÉTICA DAS ANTECIPAÇÕES ═══
# Estratégia A: UMA viagem longa  → 1 janela de antecipação
# Estratégia B: DUAS viagens curtas (mesmo total de dias) → 2 janelas
# Moeda: pontos-semana de felicidade acima do basal, num ano.
# O depois vale zero para viagem comum (achado do estudo); na melhor
# hipótese (muito relaxada), vale a área do brilho que morre em 8 semanas.

brilho0 = m_muito_relaxada - m_pre_muito_relaxada          # 0,16
lam = np.log(brilho0 / 0.01) / semanas_extincao
area_brilho = brilho0/lam * (1 - np.exp(-lam*semanas_extincao))

janelas = [2, 4, 8]   # sensibilidade: semanas de antecipação
print("=" * 65)
print("  MODELO - BAYES + A ARITMÉTICA DAS ANTECIPAÇÕES")
print("=" * 65)
print(f"\n  PARTE 1 - A posterior do degrau (aproximação declarada):")
print(f"  → δ = {delta:.2f} ponto · EP = {ep_delta:.3f}")
print(f"  → P(degrau > 0 | dados) = {p_degrau_real*100:.0f}%")
print(f"  → IC de credibilidade 95%: [{ic_crivel[0]:+.2f}, {ic_crivel[1]:+.2f}]")
print(f"  → 9 chances em 10 de o degrau ser real; 1 em 10 de não ser nada.")
print(f"     O intervalo roça o zero, e isso fica dito.")
print(f"\n  PARTE 2 - Datas contra dias (pontos-semana num ano):")
print(f"  → Área do brilho pós (melhor hipótese): {area_brilho:.2f} ponto-semana")
print(f"\n  {'janela':>8} | {'1 viagem longa':>16} | {'2 curtas':>10} | razão")
for W in janelas:
    a_comum  = W*delta
    b_comum  = 2*W*delta
    a_relax  = W*delta + area_brilho
    b_relax  = 2*W*delta + 2*area_brilho
    print(f"  {W:>5} sem | comum: {a_comum:>5.2f}       | {b_comum:>7.2f}    | {b_comum/a_comum:.1f}x")
    print(f"  {'':>8} | relax: {a_relax:>5.2f}       | {b_relax:>7.2f}    | {b_relax/a_relax:.1f}x")
print(f"\n  → Duas viagens curtas rendem ~2x os pontos-semana de UMA longa,")
print(f"     em QUALQUER janela plausível e nos dois cenários.")
print(f"  → A conclusão não depende da suposição: é robusta à sensibilidade.")
print(f"\n  A LEITURA:")
print(f"  → O estudo dá as peças; o modelo monta a aposta:")
print(f"     quem quer felicidade compra datas, não dias.")
print("=" * 65)
```

```python
# --- VISUALIZAÇÃO - GRÁFICOS SEPARADOS ---

# ── GRÁFICO 1 - A assimetria: antes e depois ──
fig1, ax1 = plt.subplots(figsize=(10, 6))

x = np.arange(2)
largura = 0.32
antes = [m_pre_viajantes, m_pre_ficaram]
depois = [2.16, 2.16]  # pós: indistinguíveis (nível ilustrativo comum, declarado)

b1 = ax1.bar(x - largura/2, antes, largura, color='#2a8a82', alpha=0.88,
             label='Antes da viagem')
b2 = ax1.bar(x + largura/2, depois, largura, color='#c9a14a', alpha=0.55,
             label='Depois (sem diferença; nível ilustrativo)')
ax1.set_xticks(x)
ax1.set_xticklabels(['Com viagem marcada', 'Sem viagem'])
ax1.set_ylim(0, 3.1)
ax1.set_ylabel('Felicidade média')
ax1.set_title('A assimetria: a felicidade mora no antes\n'
              'p=0,018 antes; nenhuma diferença depois · η²=0,02, minúsculo e declarado',
              fontsize=12.5, pad=15)
ax1.legend(loc='lower right', frameon=False, fontsize=10.5)

for bar, v in zip(b1, antes):
    ax1.text(bar.get_x() + bar.get_width()/2, v + 0.05,
             f'{v:.2f}', ha='center', fontweight='bold', fontsize=13)

plt.figtext(0.5, 0.01,
            'Fonte: Nawijn et al. (2010), N=1.530 | pós ilustrativo: o dado é a ausência de diferença | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-091-grafico-01-assimetria.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 - O brilho que evapora (curva estilizada, declarada) ──
fig2, ax2 = plt.subplots(figsize=(10, 6))

semanas = np.linspace(0, 10, 400)
# decaimento estilizado: elevado nas 2 primeiras semanas, basal em 8
brilho_inicial = m_muito_relaxada - m_pre_muito_relaxada  # 0,16
lam = np.log(brilho_inicial / 0.01) / semanas_extincao
curva = m_pre_muito_relaxada + brilho_inicial * np.exp(-lam * semanas)

ax2.plot(semanas, curva, color='#c0392b', linewidth=3)
ax2.axhline(m_pre_muito_relaxada, color='#333', linestyle='--', linewidth=1.5)
ax2.text(9.9, m_pre_muito_relaxada + 0.006, 'nível basal do grupo (2,48)',
         fontsize=10, color='#333', ha='right')
ax2.axvspan(0, 2, color='#c9a14a', alpha=0.12)
ax2.text(1, m_muito_relaxada + 0.01, 'o brilho:\n~2 semanas', fontsize=10.5,
         color='#8a6d2f', ha='center')
ax2.annotate('extinto em 8 semanas', xy=(8, m_pre_muito_relaxada + 0.012),
             xytext=(6.1, 2.56), fontsize=10.5, color='#c0392b',
             arrowprops=dict(arrowstyle='->', color='#c0392b', lw=1.5))
ax2.set_xlabel('Semanas após a volta')
ax2.set_ylabel('Felicidade média')
ax2.set_ylim(2.44, 2.70)
ax2.set_title('Mesmo a viagem muito relaxada: o brilho evapora\n'
              'Curva ESTILIZADA sobre as âncoras publicadas (2,64 → basal 2,48)',
              fontsize=12.5, pad=15)

plt.figtext(0.5, 0.01,
            'Fonte: Nawijn et al. (2010) | forma da curva: suposição declarada | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-091-grafico-02-brilho.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 - A aritmética das antecipações ──
fig3, ax3 = plt.subplots(figsize=(10, 6))

x = np.arange(len(janelas))
larg = 0.32
uma_longa = [W*delta + area_brilho for W in janelas]
duas_curtas = [2*W*delta + 2*area_brilho for W in janelas]

b1 = ax3.bar(x - larg/2, uma_longa, larg, color='#c9a14a', alpha=0.88,
             label='Uma viagem longa')
b2 = ax3.bar(x + larg/2, duas_curtas, larg, color='#2a8a82', alpha=0.88,
             label='Duas curtas (mesmo total de dias)')
ax3.set_xticks(x)
ax3.set_xticklabels([f'janela de\n{W} semanas' for W in janelas])
ax3.set_ylabel('Pontos-semana de felicidade num ano')
ax3.set_title('A aritmética das antecipações: datas valem mais que dias\n'
              'Simulação com suposições declaradas · robusta à janela de antecipação',
              fontsize=12.5, pad=15)
ax3.legend(loc='upper left', frameon=False, fontsize=11)

for bars in (b1, b2):
    for bar in bars:
        v = bar.get_height()
        ax3.text(bar.get_x() + bar.get_width()/2, v + 0.06,
                 f'{v:.1f}', ha='center', fontweight='bold', fontsize=12)

for xi, (a, b) in enumerate(zip(uma_longa, duas_curtas)):
    ax3.text(xi, b + 0.55, f'{b/a:.1f}x', ha='center', fontsize=13,
             color='#1a5f5a', fontweight='bold')

plt.figtext(0.5, 0.01,
            'Modelo próprio sobre as âncoras de Nawijn et al. (2010) | cenário muito relaxada; viagem comum dá razão 2,0x exata | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-091-grafico-03-aritmetica.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**A felicidade da viagem mora na mala ainda fechada.**

Quem tem viagem marcada é mensuravelmente mais feliz do que quem
não tem, antes de sair de casa: 2,25 contra 2,07, p=0,018. O efeito
é minúsculo, η²=0,02, e fica declarado como minúsculo. Não é
êxtase. É um degrau.

Mas é um degrau assimétrico. Na volta, quem viajou e quem ficou
são estatisticamente a mesma pessoa. Só a viagem muito relaxada
segura um brilho, por duas semanas, e em oito ele evapora por
completo.

A posterior bayesiana do degrau: **93% de probabilidade de ele ser
real**, com o intervalo de credibilidade roçando o zero, e isso
fica dito. Nove chances em dez.

E o modelo do dia monta a aposta que o estudo só insinua: com o
degrau valendo durante a janela de antecipação, **duas viagens
curtas rendem o dobro dos pontos-semana de felicidade de uma
longa com o mesmo total de dias**, em qualquer janela plausível
(2, 4 ou 8 semanas) e nos dois cenários. **Quem quer felicidade
de viagem compra datas no calendário, não dias a mais na mesma
viagem.**

Para quem faz a volta difícil, a que ninguém posta, a ciência
oferece o único consolo que ela tem em estoque: o depois não se
estica para ninguém. Mas o antes se agenda. A próxima data marcada
começa a pagar no dia em que entra no calendário.

A pergunta que fica:

*Qual é a próxima data que você precisa colocar no calendário,
não pela viagem, mas pela espera?*

### ⚠️ Limitações do Modelo

- η²=0,02 é minúsculo: a viagem marcada explica 2% da variância de
  felicidade pré-viagem; o claim do dia é a assimetria, não a
  magnitude
- Amostra holandesa de 2010, felicidade por autorrelato em desenho
  observacional: antecipação como explicação é plausível e
  declarada como interpretação dos autores, não causa demonstrada
- Os subgrupos de viagem estressante ou neutra eram pequenos
  demais para conclusão, e ficam fora das âncoras do dia
- A curva do gráfico 2 é estilizada sobre os pontos publicados
  (brilho de ~2 semanas, extinção em 8); a forma exata do
  decaimento não é reportada no artigo
- A posterior bayesiana usa aproximação normal com prior plana
  sobre as médias e EPs publicados; o teste original é multivariado
  (p=0,018), e a aproximação fica declarada
- A janela de antecipação não é publicada pelo artigo; o modelo a
  trata como parâmetro de sensibilidade (2, 4 e 8 semanas), e a
  conclusão da razão ~2x sobrevive em toda a faixa
- O gráfico 3 é modelo próprio, não medida: soma retângulos sobre
  as âncoras publicadas, com o cenário da viagem comum dando razão
  exata de 2,0x
- Ponte declarada com o Dia #047: lá a memória esticava a duração
  em retrospecto; aqui a felicidade encolhe na volta. Claims
  complementares, fontes distintas

*A ciência é honesta sobre o que não sabe. O modelo também.*
