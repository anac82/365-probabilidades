### 📖 A História

Por doze anos, cuidei do meu filho sozinha.

Hoje ele embarca pra morar com o pai, nos Estados Unidos. E eu estou profundamente abalada.

Não é a separação do pai dele — essa aconteceu há muito tempo. É a ideia de não ter mais o Edgar no meu dia a dia. Ele, que ao longo desses anos foi virando, além de filho, um grande amigo. A gente faz tanta coisa junto.

A dor que eu sinto agora é do tamanho desses doze anos. Não é exagero. Não é fraqueza. É a conta exata do amor que a gente construiu.

E é aqui que eu preciso ser honesta sobre o que a ciência diz — e sobre o que ela não diz.

Ela não diz que essa dor é grande demais, ou que vai passar rápido, ou que eu deveria sentir menos. Nada disso.

O que ela diz é outra coisa, mais silenciosa: que existe, dentro de mim, uma capacidade de atravessar isso que eu não consigo enxergar agora, no meio da dor. E que eu quase sempre subestimo essa capacidade — justamente quando mais preciso dela.

### 📚 O Conceito: O Sistema Imunológico Psicológico

Em 1998, Daniel Gilbert e Timothy Wilson, de Harvard e da Universidade da Virgínia, descreveram algo que chamaram de "viés de durabilidade".

A ideia: quando imaginamos como vamos nos sentir depois de um evento difícil, tendemos a superestimar quanto tempo o sofrimento vai durar. Prevemos que ficaremos arrasados por muito mais tempo do que de fato ficamos.

O motivo não é que a dor seja pequena. É que temos um mecanismo que os pesquisadores apelidaram de "sistema imunológico psicológico" — um conjunto de processos que, sem a gente perceber, vai dando sentido às coisas, reorganizando, ajudando a seguir. Ele trabalha nos bastidores, quieto.

E porque trabalha quieto, a gente não o enxerga quando faz suas previsões. Gilbert e Wilson chamaram esse ponto cego de "negligência imunológica": no momento em que imaginamos o futuro difícil, esquecemos de contar com a nossa própria capacidade de nos recuperar.

Não é sobre a dor ser menor. É sobre nós sermos mais resilientes do que a versão assustada de nós mesmos consegue prever.

### 🧮 O Modelo

**Fontes:**
- Gilbert, D. T., Pinel, E. C., Wilson, T. D., Blumberg, S. J. & Wheatley, T. P. (1998) — "Immune Neglect: A Source of Durability Bias in Affective Forecasting", Journal of Personality and Social Psychology, 75(3), 617-638
- Wilson, T. D. & Gilbert, D. T. (2005) — "Affective Forecasting: Knowing What to Want", Current Directions in Psychological Science, 14(3), 131-134
- Sobre o Paradoxo da Região Beta: Gilbert et al. (2004) — a recuperação de dores mais intensas pode ser mais rápida do que a de incômodos menores, porque a intensidade aciona os mecanismos de enfrentamento

**Nota metodológica:** a previsão afetiva é um campo da psicologia social, área que passou pela chamada crise de replicação. O viés de durabilidade em si é um achado robusto, replicado em contextos variados, mas a magnitude exata do efeito varia de estudo para estudo. Por isso, aplico o fator de correção conservador do projeto (×0.80) sobre a estimativa central — e trato os números como ordem de grandeza, não como precisão cirúrgica.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Gilbert & Wilson e cols. — viés de durabilidade na previsão afetiva
# Padrão consistente: as pessoas PREVEEM sofrer mais intensamente e por mais
# tempo do que REALMENTE sofrem após eventos negativos.

# Valores ILUSTRATIVOS do padrão típico observado na literatura
# (curvas de intensidade emocional prevista vs. vivida ao longo do tempo).
# Não são de um único estudo — representam a forma geral do achado.

# Intensidade emocional negativa (0 a 1), previsão vs. experiência real
tempo_semanas = np.array([0, 1, 2, 4, 8, 12])

# O que as pessoas PREVEEM sentir (declínio lento)
previsto = np.array([1.00, 0.92, 0.85, 0.75, 0.65, 0.58])

# O que as pessoas REALMENTE sentem (declínio mais rápido — sistema imune psicológico)
real = np.array([1.00, 0.78, 0.60, 0.42, 0.30, 0.24])

# Fator de correção do projeto (campo com debate de replicação)
fator_correcao = 0.80

# "Gap de resiliência": quanto a previsão superestima o sofrimento, na média do período
gap_medio = np.mean(previsto - real)
gap_corrigido = gap_medio * fator_correcao

print("=" * 65)
print("  DADOS — VIÉS DE DURABILIDADE (ilustrativo do padrão)")
print("=" * 65)
print(f"\n  Intensidade emocional negativa ao longo do tempo (0-1):")
print(f"  {'Semana':>8} {'Previsto':>10} {'Real':>8}")
for t, p, r in zip(tempo_semanas, previsto, real):
    print(f"  {t:>8} {p:>10.2f} {r:>8.2f}")
print(f"\n  Gap médio (previsto − real):        {gap_medio:.2f}")
print(f"  Gap médio corrigido (×0.80):        {gap_corrigido:.2f}")
print(f"\n  Leitura: na média, as pessoas preveem sofrer bem mais")
print(f"  (e por mais tempo) do que realmente sofrem.")
print("=" * 65)
```

```python
# --- O MODELO ---
# Traduzindo o padrão numa leitura acessível, com honestidade sobre o que é

# Redução emocional em 8 semelhante: quanto a intensidade caiu, prevista vs real
queda_prevista_8sem = (previsto[0] - previsto[4]) / previsto[0]
queda_real_8sem = (real[0] - real[4]) / real[0]

# Aplicando fator de correção conservador à "vantagem" da recuperação real
vantagem_bruta = queda_real_8sem - queda_prevista_8sem
vantagem_corrigida = vantagem_bruta * fator_correcao

print("=" * 65)
print("  MODELO — VOCÊ AGUENTA MELHOR DO QUE PREVÊ")
print("=" * 65)
print(f"\n  Em cerca de 8 semanas (ilustrativo):")
print(f"  → Queda de intensidade que as pessoas PREVEEM:  {queda_prevista_8sem*100:.0f}%")
print(f"  → Queda de intensidade que REALMENTE ocorre:    {queda_real_8sem*100:.0f}%")
print(f"\n  → As pessoas se recuperam mais do que imaginam.")
print(f"     Diferença (corrigida ×0.80):                 {vantagem_corrigida*100:.0f} pontos")
print(f"\n  IMPORTANTE — o que este número NÃO diz:")
print(f"  → Não diz que a dor é pequena.")
print(f"  → Não diz que a saudade vai sumir.")
print(f"  → Não promete prazo pra ninguém.")
print(f"\n  O que ele sugere:")
print(f"  → A versão assustada de nós subestima a nossa")
print(f"     capacidade de atravessar o que tememos.")
print("=" * 65)
```

```python
# --- VISUALIZAÇÃO ---

cor_prev = '#c0392b'   # previsto (o que tememos)
cor_real = '#2a8a82'   # real (o que acontece)
cor_ouro = '#c8a84b'

# ── GRÁFICO 1 — Previsto vs Real ao longo do tempo ──
fig1, ax1 = plt.subplots(figsize=(11, 6.5))

ax1.plot(tempo_semanas, previsto, 'o-', color=cor_prev, linewidth=2.5,
         markersize=8, label='O que prevemos sentir')
ax1.plot(tempo_semanas, real, 's-', color=cor_real, linewidth=2.5,
         markersize=8, label='O que realmente sentimos')
ax1.fill_between(tempo_semanas, previsto, real, alpha=0.12, color=cor_ouro)

ax1.set_xlabel('Tempo após o evento (semanas)')
ax1.set_ylabel('Intensidade emocional negativa (ilustrativa)')
ax1.set_ylim(0, 1.05)
ax1.set_title('O viés de durabilidade\nPrevemos sofrer mais tempo do que de fato sofremos',
              fontsize=13, pad=15)
ax1.legend(frameon=False, loc='upper right')

ax1.annotate('nossa resiliência\nmora nessa faixa',
             xy=(4, 0.58), xytext=(5.5, 0.72),
             fontsize=10, color='#8a7020', ha='center',
             arrowprops=dict(arrowstyle='->', color='#c8a84b', lw=1.5))

plt.figtext(0.5, 0.005,
            'Curva ilustrativa do padrão descrito por Gilbert & Wilson (1998) — não é dado ponto a ponto | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-029-grafico-01-durabilidade.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 — O Paradoxo da Região Beta ──
fig2, ax2 = plt.subplots(figsize=(11, 6.5))

t = np.linspace(0, 12, 200)
# Dor pequena: não aciona enfrentamento forte — decai devagar
dor_pequena = 0.45 * np.exp(-t / 9.0)
# Dor grande: aciona mecanismos de enfrentamento — decai mais rápido, cruza a pequena
dor_grande = 1.0 * np.exp(-t / 3.2)

ax2.plot(t, dor_pequena, color=cor_ouro, linewidth=3, label='Incômodo menor (enfrentamento não é acionado)')
ax2.plot(t, dor_grande, color=cor_real, linewidth=3, label='Dor intensa (aciona o enfrentamento)')

# Marcar o cruzamento
idx = np.argwhere(np.diff(np.sign(dor_grande - dor_pequena))).flatten()
if len(idx) > 0:
    xc = t[idx[0]]
    yc = dor_grande[idx[0]]
    ax2.plot(xc, yc, 'o', color='#333', markersize=8)
    ax2.annotate('a dor grande\njá passou a pequena',
                 xy=(xc, yc), xytext=(xc + 1.5, yc + 0.18),
                 fontsize=10, color='#333', ha='left',
                 arrowprops=dict(arrowstyle='->', color='#333', lw=1.2))

ax2.set_xlabel('Tempo (semanas)')
ax2.set_ylabel('Intensidade da dor (ilustrativa)')
ax2.set_ylim(0, 1.05)
ax2.set_title('O Paradoxo da Região Beta\nÀs vezes a dor intensa passa mais rápido que o incômodo pequeno',
              fontsize=13, pad=15)
ax2.legend(frameon=False, loc='upper right')

plt.figtext(0.5, 0.005,
            'Ilustração conceitual do paradoxo descrito por Gilbert et al. (2004) | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-029-grafico-02-regiao-beta.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 — Previsto vs Real: a recuperação em 8 semanas ──
fig3, ax3 = plt.subplots(figsize=(10, 6.5))

categorias = ['Recuperação que\nPREVEMOS', 'Recuperação que\nREALMENTE ocorre']
valores = [queda_prevista_8sem * 100, queda_real_8sem * 100]
cores = [cor_prev, cor_real]

bars = ax3.bar(categorias, valores, color=cores, alpha=0.85, width=0.45)
ax3.set_ylabel('Quanto a intensidade emocional caiu (%)')
ax3.set_ylim(0, 80)
ax3.set_title('Em cerca de 8 semanas (ilustrativo)\nRecuperamos mais do que a versão assustada de nós prevê',
              fontsize=13, pad=15)

for bar, v in zip(bars, valores):
    ax3.text(bar.get_x() + bar.get_width()/2, v + 1.5,
             f'{v:.0f}%', ha='center', fontweight='bold', fontsize=16)

plt.figtext(0.5, 0.005,
            'Valores ilustrativos do padrão de Gilbert & Wilson (1998) | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-029-grafico-03-recuperacao.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**A pesquisa mostra, de forma consistente, que subestimamos a nossa capacidade de nos recuperar de eventos difíceis.**

Não porque a dor não seja real. Mas porque, no momento em que imaginamos o futuro, não conseguimos ver o trabalho silencioso que a nossa mente vai fazer pra nos ajudar a atravessar.

Tem ainda uma camada contraintuitiva nisso, que os pesquisadores chamam de Paradoxo da Região Beta: às vezes nos recuperamos mais rápido de uma dor grande do que de um incômodo pequeno. Porque a dor grande aciona todos os mecanismos de enfrentamento de uma vez — e o incômodo pequeno, não. A intensidade, por mais cruel que pareça agora, também é o que liga a nossa capacidade de seguir.

Eu escrevo isso hoje sem saber ainda como vou me sentir daqui a um mês, ou a um ano. A saudade do Edgar no dia a dia é real e vai continuar sendo. A ciência não promete que ela some.

O que ela sugere é mais modesto, e talvez por isso mais confiável: que a dor de hoje, por mais real que seja, não é uma boa vidente. Ela enxerga o presente com nitidez, mas erra ao projetar por quanto tempo vai durar. Que a distância não apaga doze anos. E que o vínculo que fez essa dor tão grande é o mesmo que vai continuar existindo, do outro lado do mundo.

*Quando a dor de uma perda te diz que vai ser sempre assim, o quanto você confia nessa previsão — e o quanto ela pode estar te enganando?*

### ⚠️ Limitações do Modelo

- O viés de durabilidade foi estudado majoritariamente com eventos como términos de relacionamento, resultados de eleições e frustrações profissionais — não com luto profundo, separações familiares prolongadas ou perdas de grande magnitude. Generalizar diretamente para essas dores vai além do que os dados sustentam.
- A pesquisa descreve tendências médias de população. Não é uma previsão sobre nenhuma pessoa específica, nem uma promessa de que qualquer dor vai diminuir num prazo determinado.
- Este modelo não trata de luto complicado nem de depressão, que têm dinâmicas próprias e podem exigir apoio profissional. Se a dor se tornar incapacitante ou persistente, procurar ajuda especializada é o caminho — não "esperar o viés agir".
- O campo da previsão afetiva passou pela crise de replicação; o efeito central é robusto, mas magnitudes específicas devem ser lidas com cautela.

*A ciência é honesta sobre o que não sabe. O modelo também.*

*Este é um tema sensível. Se você estiver passando por uma perda ou sofrimento difícil de carregar sozinho, falar com alguém de confiança ou com um profissional pode ajudar — e não há nada de fraco nisso.*
