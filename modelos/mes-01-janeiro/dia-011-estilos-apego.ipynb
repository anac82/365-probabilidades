### 📖 A História

Tem gente que constrói relacionamentos como quem constrói uma casa: com paciência, com confiança de que vai durar, sem precisar checar o alicerce todo dia. E tem gente que vive com a sensação de que tudo pode desmoronar a qualquer momento — mesmo quando nada deu motivo para isso.

A diferença entre essas duas formas de amar não é sorte, nem personalidade no sentido vago da palavra. Tem nome: estilo de apego. E foi estudado por décadas, com amostras enormes, em populações inteiras.

O que a ciência encontrou é direto: o jeito como você se conecta com quem ama não é só sobre como você se sente. É sobre quanto tempo, estatisticamente, esse relacionamento tende a durar.

---

### 📚 O Conceito: Teoria do Apego

A Teoria do Apego, originada com John Bowlby e expandida por Mary Ainsworth, propõe que os vínculos formados na infância com cuidadores criam um "modelo de trabalho" interno sobre como relacionamentos funcionam — modelo que se estende para a vida adulta e molda como nos conectamos romanticamente.

Três estilos principais emergem na vida adulta:

- **Apego seguro**: confortável com intimidade e autonomia. Confia que o parceiro estará disponível, não precisa testar constantemente o vínculo.
- **Apego ansioso**: busca proximidade intensa, teme abandono, interpreta sinais neutros como rejeição.
- **Apego evitativo**: valoriza independência a ponto de evitar intimidade profunda, desconforto com dependência emocional.

Diferente de um traço fixo, o estilo de apego pode mudar ao longo da vida através de relacionamentos reparadores, terapia e autoconhecimento — mas tende a se manter relativamente estável sem intervenção deliberada.

---

### 🧮 O Modelo
**Fontes:**
- National Comorbidity Survey Replication (NCS-R) — N=5.645 adultos americanos, amostra nacional representativa — prevalência de estilos de apego
- Madigan, S. et al. (2023) — *Attachment styles and romantic relationship outcomes* — Psychological Bulletin, 149(1-2) — duração média de relacionamentos por estilo de apego


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print('✅ Bibliotecas carregadas')
```

```python
# --- DADOS DA LITERATURA ---

# NCS-R — N=5.645 adultos (amostra nacional representativa EUA)
p_seguro     = 0.635
p_evitativo  = 0.222
p_ansioso    = 0.055
n_ncsr        = 5645

# Madigan et al. 2023 — Psychological Bulletin
# Duracao media de relacionamentos romanticos por estilo de apego (anos)
duracao_seguro    = 10.02
duracao_evitativo = 5.97
duracao_ansioso   = 4.86

fator_correcao = 0.80

print('=' * 65)
print('  DADOS DA LITERATURA — ESTILOS DE APEGO')
print('=' * 65)
print(f'\n  NCS-R (N={n_ncsr:,}, amostra nacional representativa):')
print(f'  → Apego seguro     : {p_seguro*100:.1f}%')
print(f'  → Apego evitativo  : {p_evitativo*100:.1f}%')
print(f'  → Apego ansioso    : {p_ansioso*100:.1f}%')
print(f'\n  Madigan et al. 2023 (Psychological Bulletin):')
print(f'  → Duracao media — Seguro    : {duracao_seguro} anos')
print(f'  → Duracao media — Evitativo : {duracao_evitativo} anos')
print(f'  → Duracao media — Ansioso   : {duracao_ansioso} anos')
print(f'\n  Fator de correcao aplicado: x{fator_correcao}')
print('=' * 65)
```

```python
# --- O MODELO ---
# Distribuicao Beta com IC 95% para prevalencia dos estilos

def beta_ic(p, n, conf=0.95):
    alpha = p * n
    beta_  = (1 - p) * n
    dist   = stats.beta(alpha, beta_)
    return dist.interval(conf), dist

ic_seguro,    dist_seguro    = beta_ic(p_seguro, n_ncsr)
ic_evitativo, dist_evitativo = beta_ic(p_evitativo, n_ncsr)
ic_ansioso,   dist_ansioso   = beta_ic(p_ansioso, n_ncsr)

p_seguro_corrigido    = p_seguro * fator_correcao
p_evitativo_corrigido = p_evitativo * fator_correcao
p_ansioso_corrigido   = p_ansioso * fator_correcao

# Razao de duracao entre estilos
razao_seguro_ansioso    = duracao_seguro / duracao_ansioso
razao_seguro_evitativo  = duracao_seguro / duracao_evitativo
ganho_anos_ansioso      = duracao_seguro - duracao_ansioso
ganho_anos_evitativo    = duracao_seguro - duracao_evitativo

# --- CAMADA ADICIONAL: duracao media ESPERADA na populacao geral ---
# Combina prevalencia (NCS-R) + duracao por estilo (Madigan et al. 2023)
# Esta e uma estimativa DERIVADA, nao reportada diretamente em nenhuma das duas fontes

p_total_classificado = p_seguro + p_evitativo + p_ansioso  # 8.8% nao classificado fica fora
p_seguro_norm    = p_seguro / p_total_classificado
p_evitativo_norm = p_evitativo / p_total_classificado
p_ansioso_norm   = p_ansioso / p_total_classificado

duracao_media_ponderada = (p_seguro_norm * duracao_seguro +
                            p_evitativo_norm * duracao_evitativo +
                            p_ansioso_norm * duracao_ansioso)

ganho_seguro_sobre_media = duracao_seguro - duracao_media_ponderada
perda_ansioso_sobre_media = duracao_media_ponderada - duracao_ansioso

print('=' * 65)
print('  MODELO — RESULTADO')
print('=' * 65)
print(f'  Apego seguro:')
print(f'  → Estimativa : {p_seguro*100:.1f}%')
print(f'  → IC 95%     : [{ic_seguro[0]*100:.1f}%, {ic_seguro[1]*100:.1f}%]')
print(f'  → Corrigido  : {p_seguro_corrigido*100:.1f}%')
print(f'\n  Apego evitativo:')
print(f'  → Estimativa : {p_evitativo*100:.1f}%')
print(f'  → IC 95%     : [{ic_evitativo[0]*100:.1f}%, {ic_evitativo[1]*100:.1f}%]')
print(f'  → Corrigido  : {p_evitativo_corrigido*100:.1f}%')
print(f'\n  Apego ansioso:')
print(f'  → Estimativa : {p_ansioso*100:.1f}%')
print(f'  → IC 95%     : [{ic_ansioso[0]*100:.1f}%, {ic_ansioso[1]*100:.1f}%]')
print(f'  → Corrigido  : {p_ansioso_corrigido*100:.1f}%')
print(f'\n  RAZAO DE DURACAO:')
print(f'  → Seguro dura {razao_seguro_ansioso:.2f}x mais que ansioso (+{ganho_anos_ansioso:.1f} anos)')
print(f'  → Seguro dura {razao_seguro_evitativo:.2f}x mais que evitativo (+{ganho_anos_evitativo:.1f} anos)')
print(f'\n  DURACAO MEDIA ESPERADA NA POPULACAO GERAL (modelo derivado):')
print(f'  → Ponderando pela prevalencia de cada estilo: {duracao_media_ponderada:.2f} anos')
print(f'  → Quem tem apego seguro ganha +{ganho_seguro_sobre_media:.2f} anos sobre essa media')
print(f'  → Quem tem apego ansioso perde -{perda_ansioso_sobre_media:.2f} anos sobre essa media')
print('=' * 65)
```

```python
# --- VISUALIZACAO ---

# GRAFICO 1 — Prevalencia dos estilos de apego
fig1, ax1 = plt.subplots(figsize=(12, 8))

estilos = ['Seguro', 'Evitativo', 'Ansioso']
prevalencias = [p_seguro * 100, p_evitativo * 100, p_ansioso * 100]
ics_prev = [ic_seguro, ic_evitativo, ic_ansioso]
cores1 = ['#2a8a82', '#c8a84b', '#c0392b']

bars = ax1.bar(estilos, prevalencias, color=cores1, alpha=0.85, width=0.5)

yerr_low  = [p - ic[0]*100 for p, ic in zip(prevalencias, ics_prev)]
yerr_high = [ic[1]*100 - p for p, ic in zip(prevalencias, ics_prev)]
ax1.errorbar(range(len(estilos)), prevalencias, yerr=[yerr_low, yerr_high],
             fmt='none', color='#333333', capsize=6, linewidth=2)

ax1.set_ylim(0, 75)
ax1.set_ylabel('Prevalência na população adulta (%)', fontsize=12)
ax1.set_title('Distribuição dos estilos de apego em adultos\nNCS-R — N=5.645, amostra nacional representativa (EUA)',
              fontsize=13, pad=15)

for bar, p in zip(bars, prevalencias):
    ax1.text(bar.get_x() + bar.get_width() / 2, p + 3,
             f'{p:.1f}%', ha='center', fontweight='bold', fontsize=14)

plt.tight_layout()
plt.savefig('dia-011-grafico-01-prevalencia.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Grafico 1 salvo!')


# GRAFICO 2 — Duracao media de relacionamentos por estilo
fig2, ax2 = plt.subplots(figsize=(12, 8))

estilos2 = ['Ansioso', 'Evitativo', 'Seguro']
duracoes = [duracao_ansioso, duracao_evitativo, duracao_seguro]
cores2 = ['#c0392b', '#c8a84b', '#2a8a82']

bars2 = ax2.barh(estilos2, duracoes, color=cores2, alpha=0.85, height=0.5)
ax2.set_xlabel('Duração média do relacionamento (anos)', fontsize=12)
ax2.set_xlim(0, 12)
ax2.set_title('Apego seguro: relacionamentos duram quase o dobro\nMadigan et al. (2023) — Psychological Bulletin',
              fontsize=13, pad=15)

for bar, d in zip(bars2, duracoes):
    ax2.text(bar.get_width() + 0.2, bar.get_y() + bar.get_height() / 2,
             f'{d:.2f} anos', va='center', fontweight='bold', fontsize=12)

plt.tight_layout()
plt.savefig('dia-011-grafico-02-duracao.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Grafico 2 salvo!')


# GRAFICO 3 — Distribuicao Beta (apego seguro)
fig3, ax3 = plt.subplots(figsize=(12, 8))

x = np.linspace(ic_seguro[0] * 0.99, ic_seguro[1] * 1.01, 1000)
y = dist_seguro.pdf(x)

ax3.plot(x * 100, y, color='#2a8a82', linewidth=3)
ax3.fill_between(x * 100, y, alpha=0.2, color='#2a8a82')
ax3.axvline(x=ic_seguro[0] * 100, color='#c8a84b', linestyle='--', linewidth=2,
            label=f'IC 95%: [{ic_seguro[0]*100:.1f}%, {ic_seguro[1]*100:.1f}%]')
ax3.axvline(x=ic_seguro[1] * 100, color='#c8a84b', linestyle='--', linewidth=2)
ax3.axvline(x=p_seguro * 100, color='#2a8a82', linewidth=2,
            label=f'Estimativa central: {p_seguro*100:.1f}%')
ax3.axvline(x=p_seguro_corrigido * 100, color='#c0392b', linestyle='-.',
            linewidth=2, label=f'Corrigido (x0.80): {p_seguro_corrigido*100:.1f}%')

ax3.set_xlabel('Probabilidade de apego seguro na população (%)', fontsize=12)
ax3.set_ylabel('Densidade', fontsize=12)
ax3.set_title('Distribuição Beta — Prevalência de apego seguro\nNCS-R — N=5.645, amostra nacional representativa',
              fontsize=13, pad=15)
ax3.legend(fontsize=11)

plt.tight_layout()
plt.savefig('dia-011-grafico-03-bayesiano.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Grafico 3 salvo!')


# GRAFICO 4 — Duracao media esperada na populacao (modelo derivado)
fig4, ax4 = plt.subplots(figsize=(12, 8))

categorias4 = ['Apego\nansioso', 'Média da\npopulação\n(ponderada)', 'Apego\nseguro']
valores4 = [duracao_ansioso, duracao_media_ponderada, duracao_seguro]
cores4 = ['#c0392b', '#c8a84b', '#2a8a82']

bars4 = ax4.bar(categorias4, valores4, color=cores4, alpha=0.85, width=0.5)
ax4.set_ylim(0, 12)
ax4.set_ylabel('Duração esperada do relacionamento (anos)', fontsize=12)
ax4.set_title('Quanto seu estilo de apego te afasta da média da população\nModelo derivado: prevalência (NCS-R) × duração (Madigan et al. 2023)',
              fontsize=13, pad=15)

for bar, v in zip(bars4, valores4):
    ax4.text(bar.get_x() + bar.get_width() / 2, v + 0.3,
             f'{v:.2f} anos', ha='center', fontweight='bold', fontsize=13)

ax4.annotate('', xy=(2, duracao_seguro), xytext=(1, duracao_media_ponderada),
             arrowprops=dict(arrowstyle='->', color='#2a8a82', lw=2))
ax4.text(1.5, (duracao_seguro + duracao_media_ponderada)/2 + 0.3,
         f'+{ganho_seguro_sobre_media:.1f} anos', ha='center',
         fontsize=10, color='#2a8a82', fontweight='bold')

plt.tight_layout()
plt.savefig('dia-011-grafico-04-media-ponderada.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Grafico 4 salvo!')
```

### 💡 O Insight

**63,5% dos adultos têm apego seguro** — IC 95%: [62,2%, 64,8%], em amostra nacional representativa de 5.645 pessoas. A maioria das pessoas, na verdade, já tem a base para relacionamentos duradouros.

Mas a diferença para quem não tem é grande: relacionamentos de pessoas com apego seguro duram em média **10,02 anos** — quase o **dobro** dos 4,86 anos de pessoas com apego ansioso, e 1,68 vezes mais que os 5,97 anos de pessoas com apego evitativo.

Combinando a prevalência de cada estilo com sua duração típica, a **duração média esperada de um relacionamento na população geral é de 8,72 anos**. Isso significa que ter apego seguro coloca você **1,3 anos acima** dessa média — e apego ansioso te deixa **3,86 anos abaixo** dela.

Isso não significa que apego define um destino fixo. Significa que o padrão de conexão que você aprendeu — muitas vezes antes dos 5 anos de idade — continua moldando, estatisticamente, quanto tempo seus vínculos românticos tendem a durar. E também significa que isso pode mudar: estilo de apego não é sentença, é um padrão que pode ser revisado.

---

### ⚠️ Limitações do Modelo
- Os dados de prevalência são de amostra americana — a distribuição pode variar em outros contextos culturais
- "Duração do relacionamento" é uma média populacional — não prediz o destino de uma relação específica
- A duração média ponderada (8,72 anos) é uma estimativa derivada pelo projeto, combinando duas fontes distintas — não um número reportado diretamente em nenhuma delas
- Causalidade não é unidirecional: relacionamentos ruins também podem reforçar padrões de apego inseguro, não apenas o contrário
- Estilo de apego pode mudar ao longo da vida através de relacionamentos reparadores e terapia — os números capturam um retrato médio, não um destino fixo
- O fator de correção (×0,80) é uma aproximação padrão do projeto

*A ciência é honesta sobre o que não sabe. O modelo também.*
