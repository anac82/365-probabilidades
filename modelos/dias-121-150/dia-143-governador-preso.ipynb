### 📖 A História

O tema de hoje surgiu de uma conversa com a minha chefe nos dias em que acompanhamos as
apurações do primeiro turno para presidente e para governador. O que ela me trouxe à
tona foi a recência e a frequência com que os governadores do estado acabam com algum
envolvimento em prisões.

Quando comecei a pesquisar, foi comum encontrar, em postagens e sites, a frase de que
todo governador acaba preso. E então eu fui contar. Não os políticos em geral, que
ninguém consegue conferir um a um, e sim um universo fechado, as pessoas eleitas para
governar o estado desde a volta das eleições diretas, em 1982.

Uma regra antes do primeiro número. O evento contado é ter recebido ao menos uma ordem
judicial de prisão, cumprida. O que veio depois, anulação, absolvição, arquivamento ou
condenação, está registrado nome a nome no notebook, mas não muda a contagem. A pergunta
é quantos governadores do Rio um juiz mandou prender. Não é quantos foram condenados, e
os dois não se confundem.

---

### 📚 O Conceito

Com nove pessoas, a estatística não tem muito o que fazer, e é isso que o dia mostra. A
proporção observada é um número; o que ela consegue dizer sobre "a chance" é outro. A
posterior de Jeffreys para cinco em nove tem largura de quase sessenta pontos. O achado
não é só o 56%; é que, com N = 9, 30% e 80% são compatíveis com o que se viu.

A segunda régua é a base de comparação. O SISDEPEN conta quantas pessoas estão privadas de
liberdade num dado dia; dividido pela população adulta, dá a fração de brasileiros presos
num instante. Não é "já foi preso alguma vez na vida", que ninguém mede no Brasil. Entra
como ordem de grandeza, declarada.

### 🧮 O Modelo

**Universo e registros (fontes abertas, verificadas uma a uma em 27/10/2026):**

- Os outros três estados do Sudeste, governadores eleitos desde 1982: São Paulo (7

- Base de comparação: SISDEPEN, 17º ciclo (jul-dez/2024), 905.316 pessoas sob sanção em

**Modelo próprio:**
- Posterior Beta (Jeffreys) para 5 de 9 e para 5 de 13, com a largura do intervalo como
  resultado.
- Quanto N seria preciso para o intervalo ter 10 pontos de largura.
- O Rio contra SP, MG e ES: teste exato de Fisher e comparação das posteriores (5 de 9
  contra 1 de 23).
- Razão entre a fração observada e a fração de adultos presos num instante, com a
  ressalva de que são medidas diferentes.
- **Sem ×0,80:** registro judicial, não autorrelato.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

VERMELHO, DOURADO, VERDE, CINZA = '#c0392b', '#c8a84b', '#1a5f5a', '#6b6a64'
SEMENTE = 42
ger = np.random.default_rng(SEMENTE)

def vir(x, casas=1):
    return f"{x:.{casas}f}".replace(".", ",")

def pct(x, casas=1):
    return vir(100 * x, casas) + "%"

def mil(n):
    return f"{int(round(n)):,}".replace(",", ".")

def beta_j(k, n):
    d = stats.beta(k + 0.5, n - k + 0.5)
    return d, d.mean(), d.interval(0.95)

print("✅ Bibliotecas carregadas")
```

```python
# --- OS DADOS ---
# Universo: quem governou o Rio de Janeiro desde 1983. Registros conferidos em 27/10/2026.
gov = pd.DataFrame([
    # nome, eleito, inicio, fim, preso, prisoes, primeira_prisao, dias_preso_total_aprox, desfecho
    ("Leonel Brizola",      True,  1983, 1994, False, 0, None,         0,    "sem registro"),
    ("Moreira Franco",      True,  1987, 1991, True,  1, "2019-03-21", 4,    "decisões anuladas pelo STF; denúncia arquivada"),
    ("Nilo Batista",        False, 1994, 1995, False, 0, None,         0,    "vice; sem registro"),
    ("Marcello Alencar",    True,  1995, 1999, False, 0, None,         0,    "sem registro; morreu em 2014"),
    ("Anthony Garotinho",   True,  1999, 2002, True,  4, "2016-11-16", None, "casos anulados; inelegível em 2018"),
    ("Benedita da Silva",   False, 2002, 2003, False, 0, None,         0,    "vice; sem registro"),
    ("Rosinha Garotinho",   True,  2003, 2007, True,  3, "2017-11-22", None, "casos anulados"),
    ("Sérgio Cabral",       True,  2007, 2014, True,  1, "2016-11-17", 2223, "condenado em mais de 20 ações, parte revista"),
    ("Francisco Dornelles", False, 2016, 2016, False, 0, None,         0,    "vice, interino; sem registro"),
    ("Luiz Fernando Pezão", True,  2014, 2018, True,  1, "2018-11-29", 376,  "condenado em 1ª instância; absolvido pelo TRF-2"),
    ("Wilson Witzel",       True,  2019, 2020, False, 0, None,         0,    "afastado pelo STJ; impeachment em 2021; prisão negada"),
    ("Cláudio Castro",      True,  2020, 2026, False, 0, None,         0,    "renúncia em 2026; inelegível pelo TSE; buscas da PF"),
    ("Thiago Pampolha",     False, 2026, 2026, False, 0, None,         0,    "vice; sem registro"),
], columns=["nome", "eleito", "inicio", "fim", "preso", "prisoes", "primeira_prisao", "dias_preso", "desfecho"])

eleitos = gov[gov["eleito"]]
N_ELEITOS, K_ELEITOS = len(eleitos), int(eleitos["preso"].sum())
N_TODOS, K_TODOS = len(gov), int(gov["preso"].sum())
N_PRISOES = int(gov["prisoes"].sum())

# Eleito em 2026, ainda sem posse: Douglas Ruas (PL), proclamado pelo TRE-RJ em 15/10/2026 (50,88% dos
# válidos, primeiro turno), posse em 01/01/2027. Entra como sensibilidade, não na conta de quem governou.
ELEITO_2026 = "Douglas Ruas"
N_ELEITOS_COM_2026 = N_ELEITOS + 1

# Os outros três estados do Sudeste: governadores ELEITOS desde 1982 (registros conferidos em 27/10/2026)
sudeste = pd.DataFrame([
    ("SP", "Franco Montoro", 1983, False, ""), ("SP", "Orestes Quércia", 1987, False, "processado, sem prisão"),
    ("SP", "Luiz Antônio Fleury", 1991, False, ""), ("SP", "Mário Covas", 1995, False, ""),
    ("SP", "Geraldo Alckmin", 2003, False, "investigado, sem prisão"), ("SP", "José Serra", 2007, False, ""),
    ("SP", "Tarcísio de Freitas", 2023, False, ""),
    ("MG", "Tancredo Neves", 1983, False, ""), ("MG", "Newton Cardoso", 1987, False, ""),
    ("MG", "Hélio Garcia", 1991, False, ""), ("MG", "Eduardo Azeredo", 1995, True, "preso em 23/05/2018 após condenação pelo TJMG (20 anos, mensalão tucano)"),
    ("MG", "Itamar Franco", 1999, False, ""), ("MG", "Aécio Neves", 2003, False, "pedido de prisão negado pelo STF em 2017, como senador"),
    ("MG", "Antonio Anastasia", 2011, False, ""), ("MG", "Fernando Pimentel", 2015, False, "denunciado, sem prisão"),
    ("MG", "Romeu Zema", 2019, False, ""),
    ("ES", "Gerson Camata", 1983, False, ""), ("ES", "Max Mauro", 1987, False, ""), ("ES", "Albuíno Azeredo", 1991, False, ""),
    ("ES", "Vitor Buaiz", 1995, False, ""), ("ES", "José Ignácio Ferreira", 1999, False, "investigado, sem prisão"),
    ("ES", "Paulo Hartung", 2003, False, ""), ("ES", "Renato Casagrande", 2011, False, ""),
], columns=["uf", "nome", "primeiro_mandato", "preso", "registro"])
por_uf = sudeste.groupby("uf").agg(n=("nome", "size"), presos=("preso", "sum")).reset_index()
por_uf = pd.concat([pd.DataFrame({"uf": ["RJ"], "n": [N_ELEITOS], "presos": [K_ELEITOS]}), por_uf], ignore_index=True)
N_OUTROS, K_OUTROS = int(sudeste.shape[0]), int(sudeste["preso"].sum())

# Base de comparação: SISDEPEN 17º ciclo (31/12/2024) e Censo 2022 (18 anos ou mais)
SISDEPEN_TOTAL = 905316
SISDEPEN_CELAS = 670265
POP_18MAIS_2022 = 155_000_000   # ordem de grandeza do Censo 2022, declarada; conferir tabela SIDRA na montagem final
p_adulto_preso = SISDEPEN_CELAS / POP_18MAIS_2022
p_adulto_sancao = SISDEPEN_TOTAL / POP_18MAIS_2022

print("=" * 70)
print("  OS DADOS — QUEM GOVERNOU O RIO DESDE 1983")
print("=" * 70)
for _, r in gov.iterrows():
    tag = "ELEITO " if r["eleito"] else "vice   "
    print(f"  {tag} {r['nome']:<22s} {r['inicio']}-{str(r['fim'])[2:]}  {'PRESO ' + str(r['prisoes']) + 'x' if r['preso'] else 'não preso':<12s} {r['desfecho']}")
print(f"\n  Eleito em 2026, posse em 2027, fora da conta: {ELEITO_2026}")
print(f"  Eleitos: {N_ELEITOS}, presos {K_ELEITOS}. Todos que ocuparam o cargo: {N_TODOS}, presos {K_TODOS}. Prisões registradas: {N_PRISOES}.")
print("\n  Sudeste, governadores eleitos desde 1982:")
for _, r in por_uf.iterrows():
    print(f"  → {r['uf']}: {int(r['presos'])} de {int(r['n'])}")
print(f"  SISDEPEN 31/12/2024: {mil(SISDEPEN_TOTAL)} sob sanção, {mil(SISDEPEN_CELAS)} em celas; adultos ≈ {mil(POP_18MAIS_2022)}")
print(f"  → fração de adultos em celas num instante: {pct(p_adulto_preso, 2)}; sob sanção: {pct(p_adulto_sancao, 2)}")
print("=" * 70)
```

```python
# --- O MODELO ---

# 1. A proporção e o que ela consegue dizer: posterior de Jeffreys
d_el, m_el, ic_el = beta_j(K_ELEITOS, N_ELEITOS)
d_td, m_td, ic_td = beta_j(K_TODOS, N_TODOS)
larg_el = ic_el[1] - ic_el[0]; larg_td = ic_td[1] - ic_td[0]
p_el_acima_50 = 1 - d_el.cdf(0.5)
p_el_acima_30 = 1 - d_el.cdf(0.3); p_el_abaixo_80 = d_el.cdf(0.8)

# 1b. Sensibilidade: incluindo o eleito de 2026, que ainda não tomou posse
d_el10, m_el10, ic_el10 = beta_j(K_ELEITOS, N_ELEITOS_COM_2026)
p_fisher_10 = stats.fisher_exact([[K_ELEITOS, N_ELEITOS_COM_2026 - K_ELEITOS], [1, 22]], alternative="greater")[1]

# 2. Quanto N seria preciso para o intervalo ter 10 pontos, mantendo a proporção
def largura(n, p=K_ELEITOS / N_ELEITOS):
    d = stats.beta(p * n + 0.5, (1 - p) * n + 0.5); lo, hi = d.interval(0.95); return hi - lo
n_para_10 = next(n for n in range(9, 5000) if largura(n) <= 0.10)
n_para_20 = next(n for n in range(9, 5000) if largura(n) <= 0.20)

# 3. Se a chance verdadeira fosse a dos adultos num instante, qual a chance de ver 5 em 9?
p_5_de_9_base = stats.binom(N_ELEITOS, p_adulto_preso).sf(K_ELEITOS - 1)
# e com uma chance "alta" qualquer, de 10%
p_5_de_9_dez = stats.binom(N_ELEITOS, 0.10).sf(K_ELEITOS - 1)
razao = m_el / p_adulto_preso

# 4. Sorteio de nove governadores a partir da posterior: quantos presos (preditiva)
SIMS = 200000
pred = stats.binom(9, d_el.rvs(SIMS, random_state=ger)).rvs(random_state=ger)
pred_ic = np.percentile(pred, [2.5, 97.5])

# 5. O Rio contra o resto do Sudeste: 5 de 9 contra 1 de 23
tabela = np.array([[K_ELEITOS, N_ELEITOS - K_ELEITOS], [K_OUTROS, N_OUTROS - K_OUTROS]])
p_fisher = stats.fisher_exact(tabela, alternative="greater")[1]
d_out, m_out, ic_out = beta_j(K_OUTROS, N_OUTROS)
amostras_rj = d_el.rvs(SIMS, random_state=ger); amostras_out = d_out.rvs(SIMS, random_state=ger)
p_rj_maior = (amostras_rj > amostras_out).mean()
razao_sudeste = np.median(amostras_rj / amostras_out)
p_5_de_9_sudeste = stats.binom(N_ELEITOS, m_out).sf(K_ELEITOS - 1)

print("=" * 70)
print("  MODELO — UM GOVERNADOR DO RIO SER PRESO")
print("=" * 70)
print(f"\n  1. Eleitos: {K_ELEITOS} de {N_ELEITOS} = {pct(K_ELEITOS / N_ELEITOS)}; posterior {pct(m_el)}  IC 95% [{pct(ic_el[0])}; {pct(ic_el[1])}], largura {vir(larg_el * 100, 0)} pontos")
print(f"     P(chance acima de 50%) = {pct(p_el_acima_50)}; P(acima de 30%) = {pct(p_el_acima_30)}; P(abaixo de 80%) = {pct(p_el_abaixo_80)}")
print(f"     Com o eleito de 2026 (posse em 2027): {K_ELEITOS} de {N_ELEITOS_COM_2026} = {pct(K_ELEITOS / N_ELEITOS_COM_2026)}; posterior {pct(m_el10)} [{pct(ic_el10[0])}; {pct(ic_el10[1])}]; Fisher contra os vizinhos p = {p_fisher_10:.4f}")
print(f"     Todos que ocuparam o cargo: {K_TODOS} de {N_TODOS} = {pct(K_TODOS / N_TODOS)}; posterior {pct(m_td)} [{pct(ic_td[0])}; {pct(ic_td[1])}], largura {vir(larg_td * 100, 0)} pontos")
print(f"\n  2. Para o intervalo ter 20 pontos seriam precisos {n_para_20} governadores; para 10 pontos, {n_para_10}")
print(f"     (o Rio elege um a cada quatro anos: {n_para_10 * 4} anos)")
print(f"\n  3. Fração de adultos em celas num instante: {pct(p_adulto_preso, 2)}; a dos governadores eleitos é {vir(razao, 0)} vezes maior")
print(f"     Se a chance de um governador fosse a de um adulto qualquer, ver 5 em 9 teria probabilidade {p_5_de_9_base:.1e}")
print(f"     Mesmo com chance de 10%, ver 5 em 9 teria probabilidade {pct(p_5_de_9_dez, 2)}")
print(f"\n  5. Rio contra SP, MG e ES: {K_ELEITOS} de {N_ELEITOS} contra {K_OUTROS} de {N_OUTROS}")
print(f"     outros três estados: posterior {pct(m_out)} [{pct(ic_out[0])}; {pct(ic_out[1])}]")
print(f"     teste exato de Fisher (Rio maior): p = {p_fisher:.4f}; P(chance do Rio maior que a dos vizinhos) = {pct(p_rj_maior)}")
print(f"     razão mediana entre as chances: {vir(razao_sudeste, 0)} vezes; com a chance dos vizinhos, ver 5 em 9 teria {pct(p_5_de_9_sudeste, 2)}")
print(f"\n  4. Os próximos nove, pela posterior: {vir(pred.mean())} presos em média, IC 95% [{int(pred_ic[0])}; {int(pred_ic[1])}]")
print("=" * 70)
```

### 💡 O Insight

**Entre os eleitos desde 1982, cinco dos nove governadores do Rio foram presos. Nos
outros três estados do Sudeste, um em 23.**

O 56% é frequência histórica, não previsão. Com nove pessoas, a estatística não diz se a
chance é de um em três ou de quatro em cinco. O que ela diz, com p = 0,003, é que o Rio
é diferente dos vizinhos, e que a diferença é grande o bastante para merecer explicação.

A explicação, o modelo não dá. Ele conta ordens de prisão cumpridas, e uma ordem de
prisão é o encontro de duas coisas: o que a pessoa fez e o que o sistema de justiça
decidiu fazer com ela. O Rio teve mais governadores envolvidos em crimes, ou mais
governadores alcançados por operações que terminaram em prisão? Os dados deste dia não
separam uma coisa da outra. Prisão é o evento contado; condenação é outro, e quatro dos
cinco tiveram as prisões anuladas, arquivadas ou foram absolvidos depois.

*Cinco em nove contra um em 23: o que você acha que explica a diferença?*

### ⚠️ Limitações do Modelo

- Os registros vêm de decisões noticiadas (STF, STJ, TRF-2, TSE) e foram conferidos um a
  um em 27/10/2026, sem consulta aos autos. Datas e contagens de prisões repetidas
  (Garotinho e Rosinha) variam entre fontes; o dia conta pessoas, não prisões.
- O evento contado é "ao menos uma ordem judicial de prisão cumprida", de qualquer
  tipo (preventiva, temporária, após condenação); nunca é lido como culpa. Os desfechos
  estão na tabela por nome.
- O universo é uma escolha: eleitos que tomaram posse (9) e todos que ocuparam o cargo
  (13). O eleito de 2026 ainda não governou; com ele, 5 em 10, e o teste contra os
  vizinhos continua do mesmo lado. As três contas estão no notebook.
- A comparação com o SISDEPEN cruza "já foi preso na vida" com "está preso num instante".
  Fica no notebook como contexto e não entra no texto nem no carrossel.
- O modelo conta ordens de prisão cumpridas e não separa duas causas possíveis para a
  diferença entre o Rio e os vizinhos: mais crime, ou mais operações judiciais que
  terminaram em prisão. O teste mede a diferença; não explica a causa.
- O 56% é frequência histórica de nove pessoas. A preditiva dos "próximos nove" é
  exercício sobre o passado, não previsão sobre quem vier a governar.
- A população adulta entra como ordem de grandeza do Censo 2022; a tabela exata será
  registrada na montagem final.
- Para SP, MG e ES, "nenhuma prisão localizada" significa busca nome a nome em fontes
  abertas, não certidão. Uma prisão não noticiada mudaria a conta de 1 em 23 para 2 em
  23, e o resultado do teste não mudaria de lado.
- Nove pessoas. Tudo o que se conclui aqui tem essa largura.
