### 📖 A História

A vida inteira migrou para a tela, e ninguém perguntou o preço.

O contrato veio por e-mail. A prova do seu filho é num tablet.
O relatório que decide o trimestre está num PDF. O livro virou
aplicativo. A migração aconteceu sem plebiscito, na base do
frete grátis e da conveniência.

E a pergunta que ficou para trás: a gente entende a mesma coisa
lendo na tela?

Cinquenta e quatro estudos, mais de 170 mil pessoas e duas
décadas de dados respondem: **não exatamente**.

O papel vence. Por pouco, mas vence. E vence justamente onde
mais importa: nos textos que informam, quando o relógio corre.

O detalhe que transforma o achado em soco vem no fim:
a vantagem do papel **cresceu** ao longo dos anos.
A geração que nasceu na tela lê pior nela, não melhor.

---

### 📚 O Conceito: O Tamanho Honesto de um Efeito

O efeito é g=0,21. Em estatística, isso é pequeno, e este projeto
não infla número pequeno: hoje o modelo faz o contrário, traduz
o 0,21 para o tamanho real dele, com as distribuições sobrepostas
à vista.

O que torna o achado sólido não é o tamanho. São três outras coisas:

**Convergência:** dois desenhos metodológicos diferentes dentro da
meta-análise chegam ao mesmo número (0,21 e 0,21), e uma
meta-análise independente de outro grupo (Kong, Seo & Zhai, 2018)
chega, de novo, a 0,21.

**Especificidade:** o efeito aparece em textos informativos e some
na narrativa pura. Aparece sob pressão de tempo e encolhe na
leitura sem pressa. Efeito que liga e desliga onde a teoria prevê
é mais crível que efeito difuso.

**A direção do tempo:** de 2000 a 2017, a vantagem do papel
aumentou. Se fosse só falta de costume com a tela, deveria ter
encolhido. Não encolheu.

**E o achado mais pesado: idade não protege.** O nível educacional
não modera o efeito. Crianças do fundamental, adolescentes e
universitários pagam o mesmo preço. A geração que nasceu na tela
não desenvolveu defesa contra ela.

O mecanismo proposto pelo campo é a **hipótese do raso**: a tela é
o ambiente dos ganhos rápidos, do like, da rolagem infinita, e o
cérebro aprende a tratá-la como lugar de leitura rasa. Sob pressão,
esse hábito cobra: os estudos do grupo de Ackerman mostram leitores
de tela mais **superconfiantes**, achando que entenderam mais do
que entenderam. E a mesma linha de pesquisa aponta a saída: tarefas
simples de profundidade (escrever palavras-chave do texto, tratar a
tarefa como central) **eliminam a inferioridade da tela**. O meio
não é destino. É hábito, e hábito se treina.

### 🧮 O Modelo

**Fontes:**
- Delgado, P., Vargas, C., Ackerman, R. & Salmerón, L. (2018) -
  Don't throw away your printed books: A meta-analysis on the
  effects of reading media on reading comprehension - Educational
  Research Review, 25, 23-38 - **54 estudos**, 76 comparações,
  **N=171.055**, publicados de 2000 a 2017. Modelo de efeitos
  aleatórios com ajuste de Knapp-Hartung. Vantagem do papel:
  **g=-0,21 [IC 95%: -0,28, -0,14]** entre participantes (k=56,
  após remoção de 2 outliers) e **dc=-0,21 [-0,37, -0,06]** dentro
  de participantes (k=18). Heterogeneidade I²=72,2 (Q=208,96,
  p<0,001); **intervalo de predição [-0,56, +0,14]**. Sem viés de
  publicação detectado (fail-safe N=1.727 contra critério de 290;
  Egger p=0,39). **Fonte central.**
- Kong, Y., Seo, Y. S. & Zhai, L. (2018) - Computers & Education,
  123 - meta-análise independente, 17 estudos - mesma estimativa
  pontual: **-0,21**. Apoio.

**Moderadores com valores publicados (Tabela 1 do artigo):**
tempo limitado g=-0,26 [-0,35, -0,16] contra ritmo próprio
g=-0,09 [-0,22, +0,05] (QB=4,12, p=0,04); texto informativo
g=-0,27 [-0,36, -0,18] e misto g=-0,30 [-0,40, -0,21] contra
narrativa pura g=+0,01 [-0,20, +0,20] (QB=7,00, p<0,05, R²=31%);
ano de publicação: **β=-0,01 por ano (p=0,03, R²=64%)**.

**O efeito em contexto, pelos próprios autores:** o crescimento
anual de compreensão no ensino fundamental é estimado em 0,32 por
ano; intervenções de leitura rendem em média 0,45. O efeito do
meio equivale a **2/3 de um ano escolar** de crescimento e à
**metade de uma intervenção** de reforço.

**Fator ×0,80: não se aplica.** Meta-análise de tamanhos de efeito
sobre desempenho medido em testes de compreensão. Sem correção.

**Suposições declaradas:** as traduções do efeito (probabilidade
de superioridade, sobreposição) são contas próprias sobre g=0,21,
assumindo normalidade. Os sinais do artigo (negativos = vantagem
do papel) são apresentados aqui como vantagem positiva do papel,
com a convenção declarada.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Delgado, Vargas, Ackerman & Salmerón (2018), Educ. Research Review 25
# 54 estudos, 76 comparações, N=171.055, 2000 a 2017
# Convenção: valores POSITIVOS = vantagem do papel (sinal invertido do artigo)

n_estudos, n_comparacoes, n_total = 54, 76, 171055

# Efeitos principais, com IC 95%
g_entre,  ic_entre  = 0.21, (0.14, 0.28)   # k=56, entre participantes
g_intra,  ic_intra  = 0.21, (0.06, 0.37)   # k=18, dentro de participantes
g_kong              = 0.21                  # Kong, Seo & Zhai 2018, independente
g_sem_cinza, ic_sem_cinza = 0.19, (0.11, 0.27)  # sensibilidade: só publicados, k=38

# Heterogeneidade e predição (entre participantes)
I2, Q, p_Q = 72.24, 208.96, 0.001
pred_int = (-0.14, 0.56)   # intervalo de predição (convenção positiva)

# Viés de publicação
failsafe_N, failsafe_criterio, egger_p = 1727, 290, 0.39

# Moderadores (Tabela 1, IC 95%, convenção positiva)
g_tempo_limitado, ic_tempo_limitado = 0.26, (0.16, 0.35)
g_ritmo_proprio,  ic_ritmo_proprio  = 0.09, (-0.05, 0.22)   # atravessa o zero
g_informativo,    ic_informativo    = 0.27, (0.18, 0.36)
g_misto,          ic_misto          = 0.30, (0.21, 0.40)
g_narrativa,      ic_narrativa      = -0.01, (-0.20, 0.20)  # zero

# Moderadores não significativos, mas com direção informativa (Tabela 1)
g_computador, ic_computador = 0.23, (0.15, 0.31)   # significativo
g_portatil,   ic_portatil   = 0.12, (-0.03, 0.27)  # tablet/e-reader: n.s.
g_com_rolagem, ic_com_rolagem = 0.25, (0.16, 0.33) # significativo
g_sem_rolagem, ic_sem_rolagem = 0.13, (-0.01, 0.27) # marginal
# Nível educacional: NÃO modera (QB n.s.): fundamental, médio e
# universitário pagam o mesmo preço. Idade não protege.

# Ano de publicação (meta-regressão)
beta_ano, p_ano, R2_ano = 0.01, 0.03, 0.64  # vantagem cresce 0,01 DP/ano

# O efeito em contexto (referências citadas pelos autores)
crescimento_anual_fundamental = 0.32   # Luyten et al. 2017
efeito_intervencao_leitura = 0.45      # Scammacca et al. 2015

print("=" * 65)
print("  DADOS DA LITERATURA - PAPEL CONTRA TELA")
print("=" * 65)
print(f"\n  Delgado et al. (2018): {n_estudos} estudos, {n_comparacoes} comparações, N={n_total:,}")
print(f"  → Entre participantes (k=56):  g = {g_entre:.2f}  [{ic_entre[0]:.2f}, {ic_entre[1]:.2f}]")
print(f"  → Dentro de participantes:     d = {g_intra:.2f}  [{ic_intra[0]:.2f}, {ic_intra[1]:.2f}]")
print(f"  → Só estudos publicados:       g = {g_sem_cinza:.2f}  [{ic_sem_cinza[0]:.2f}, {ic_sem_cinza[1]:.2f}]")
print(f"  → Kong et al. (2018), indep.:      {g_kong:.2f}")
print(f"\n  Heterogeneidade: I² = {I2:.1f} · Q = {Q:.2f}, p < {p_Q}")
print(f"  Intervalo de predição: [{pred_int[0]:.2f}, {pred_int[1]:.2f}]")
print(f"  → em parte das populações o efeito pode ser trivial ou nulo")
print(f"\n  Viés de publicação: fail-safe N = {failsafe_N:,} (critério: {failsafe_criterio})")
print(f"  Egger p = {egger_p} · publicados vs não publicados: sem diferença")
print(f"\n  MODERADORES (o efeito liga e desliga):")
print(f"  → Tempo limitado:   g = {g_tempo_limitado:.2f}  [{ic_tempo_limitado[0]:.2f}, {ic_tempo_limitado[1]:.2f}]")
print(f"  → Ritmo próprio:    g = {g_ritmo_proprio:.2f}  [{ic_ritmo_proprio[0]:.2f}, {ic_ritmo_proprio[1]:.2f}]  ← atravessa o zero")
print(f"  → Informativo:      g = {g_informativo:.2f}  [{ic_informativo[0]:.2f}, {ic_informativo[1]:.2f}]")
print(f"  → Misto:            g = {g_misto:.2f}  [{ic_misto[0]:.2f}, {ic_misto[1]:.2f}]")
print(f"  → Narrativa pura:   g = {g_narrativa:.2f}  [{ic_narrativa[0]:.2f}, {ic_narrativa[1]:.2f}]  ← zero")
print(f"\n  PISTAS (direção, sem significância formal entre categorias):")
print(f"  → Computador:  g = {g_computador:.2f} [{ic_computador[0]:.2f}, {ic_computador[1]:.2f}]  · portátil: {g_portatil:.2f} [{ic_portatil[0]:.2f}, {ic_portatil[1]:.2f}] n.s.")
print(f"  → Com rolagem: g = {g_com_rolagem:.2f} [{ic_com_rolagem[0]:.2f}, {ic_com_rolagem[1]:.2f}]  · sem: {g_sem_rolagem:.2f} [{ic_sem_rolagem[0]:.2f}, {ic_sem_rolagem[1]:.2f}]")
print(f"\n  IDADE NÃO PROTEGE: nível educacional não modera o efeito.")
print(f"  Fundamental, médio e universitário pagam o mesmo preço.")
print(f"\n  ANO: vantagem do papel cresce {beta_ano:.2f} DP/ano (p={p_ano}, R²={R2_ano:.0%})")
print(f"\n  Fator ×0,80: NÃO aplicado (meta de tamanhos de efeito)")
print("=" * 65)
```

```python
# --- O MODELO ---
# O tamanho honesto de g=0,21: traduções (contas próprias) e contexto (do artigo)

g = g_entre

# Traduções (contas próprias, assumindo normalidade)
cles = stats.norm.cdf(g / np.sqrt(2))     # prob. de superioridade num duelo
u3 = stats.norm.cdf(g)                    # fração da tela abaixo da mediana do papel
overlap = 2 * stats.norm.cdf(-g / 2)      # sobreposição das distribuições

# Contexto (âncoras citadas pelo próprio artigo)
fracao_ano_escolar = g / crescimento_anual_fundamental
fracao_intervencao = g / efeito_intervencao_leitura

# Projeção do moderador de ano (ilustrativa, sobre a meta-regressão publicada)
anos_projecao = 17  # 2000 a 2017, janela do próprio estudo

print("=" * 65)
print("  MODELO - O TAMANHO HONESTO DE g = 0,21 [0,14, 0,28]")
print("=" * 65)
print(f"\n  TRADUÇÕES (contas próprias, suposição declarada):")
print(f"  → Num duelo aleatório, o papel vence {cles*100:.0f}% das vezes")
print(f"     (uma moeda honesta venceria 50%)")
print(f"  → {u3*100:.0f}% dos leitores de tela abaixo da mediana do papel")
print(f"  → Sobreposição das distribuições: {overlap*100:.0f}%")
print(f"\n  CONTEXTO (âncoras do próprio artigo):")
print(f"  → O efeito equivale a {fracao_ano_escolar*100:.0f}% do crescimento anual")
print(f"     de compreensão no fundamental (0,21 de 0,32): ~2/3 de um ano escolar")
print(f"  → E a {fracao_intervencao*100:.0f}% de uma intervenção de reforço de leitura")
print(f"\n  A HONESTIDADE MÁXIMA (intervalo de predição):")
print(f"  → [{pred_int[0]:.2f}, {pred_int[1]:.2f}]: em 95% das populações o efeito real")
print(f"     cai nessa faixa. Em algumas é grande; em outras, trivial ou nulo.")
print(f"     A heterogeneidade (I²={I2:.0f}) é parte do achado, não um defeito.")
print(f"\n  POR QUE O DIA EXISTE:")
print(f"  → O efeito liga onde importa: informativo [0,18·0,36], relógio [0,16·0,35]")
print(f"  → E desliga onde não: narrativa [-0,20·0,20], ritmo próprio [-0,05·0,22]")
print(f"  → E cresce {beta_ano:.2f} DP/ano há {anos_projecao} anos (R²={R2_ano:.0%}).")
print(f"     Familiaridade com a tela não resolveu. Isso é o soco.")
print("=" * 65)
```

### 💡 O Insight

**O papel vence a tela: g=0,21 [0,14, 0,28]. Pouco. E cada vez mais.**

Num duelo aleatório, o leitor de papel compreende melhor 56 vezes
em 100. Uma moeda venceria 50. É esse o tamanho, e ele fica à
vista nos gráficos, com as curvas quase sobrepostas. Mas o próprio
artigo dá a régua certa: 0,21 equivale a **dois terços de um ano
escolar** de crescimento em compreensão, e à metade de uma
intervenção de reforço de leitura.

O que faz o achado importar não é o tamanho. É onde ele liga,
agora com os intervalos à vista: **texto informativo, g=0,27
[0,18, 0,36]; relógio correndo, g=0,26 [0,16, 0,35].** O contrato,
a prova, o relatório. Exatamente o que a vida migrou para a tela.
E onde desliga: narrativa pura, g=0,01 [-0,20, 0,20]; leitura sem
pressa, g=0,09 [-0,05, 0,22]. Os dois atravessam o zero. O romance
no aplicativo segue absolvido, com intervalo de confiança.

E é a direção do tempo que transforma estatística em soco:
**a vantagem do papel cresceu 0,01 DP por ano, por 17 anos
(p=0,03), e o ano de publicação sozinho explica 64% da variância
entre estudos.** Se fosse falta de costume, os nativos digitais
teriam apagado a diferença. Fizeram o contrário.

E o achado que muda o tamanho da conversa: **idade não protege.**
O efeito é o mesmo no fundamental, no médio e na universidade.
Enquanto isso, escolas digitalizam acervos e vestibulares migram
para a tela, medindo compreensão exatamente no meio e na condição
em que ela rende menos.

A boa notícia existe e fica registrada: a linha de pesquisa que
originou a meta mostra que estratégias simples de profundidade,
como escrever palavras-chave do que se leu, eliminam a
inferioridade da tela. O meio não é destino. É hábito.

A migração para a tela não foi errada. Foi sem leitura do contrato.

A pergunta que fica:

*Que leitura importante da sua vida você entregou à tela
sem perguntar o preço?*

### ⚠️ Limitações do Modelo

- g=0,21 é pequeno e está declarado como pequeno; as traduções
  (56 em 100 duelos, 58% abaixo da mediana, 92% de sobreposição)
  são contas próprias assumindo normalidade
- O intervalo de predição [-0,14, +0,56] atravessa o zero: em
  parte das populações o efeito real pode ser trivial ou nulo, e
  a heterogeneidade é alta (I²=72). O que se afirma é a média e
  as condições que a modulam, não uma lei universal
- 97% dos participantes vêm de 4 megaestudos; a sensibilidade
  excluindo os 4 mantém g=-0,22, e excluindo a literatura não
  publicada mantém g=-0,19: o resultado não depende de nenhum
  subconjunto
- Estudos majoritariamente com textos curtos e leitura única em
  laboratório; leitura prolongada de livros é menos representada
- Computador contra portátil e rolagem contra ausência de rolagem
  são pistas: as diferenças entre categorias não atingiram
  significância formal (QB n.s.) e entram como direção, não veredito
- A eliminação da inferioridade por estratégias de profundidade vem
  dos estudos primários do grupo de Ackerman citados pela meta, não
  da meta em si
- "Tela" agrega dispositivos e épocas diferentes (2000 a 2017);
  os aparelhos mudaram desde então
- Heterogeneidade entre estudos é esperada em meta-análises deste
  tamanho; o resultado descreve a média, não cada leitor
- Ponte declarada com o Dia #037: lá, o hábito de leitura e
  sobrevivência; aqui, o meio e a compreensão. Claims distintos
