### 📖 A História

Passei anos orgulhosa da minha capacidade de fazer várias coisas ao mesmo tempo. Reunião com um olho na planilha, e-mail respondido enquanto alguém apresentava, três conversas acontecendo em paralelo.

Achava que isso era produtividade. Era velocidade.

Um experimento de laboratório testou exatamente isso — e a resposta não foi o que eu esperava. Não foi "multitarefa te deixa mais lenta". Foi mais interessante, e mais desconfortável, do que isso.

---

### 📚 O Conceito: Custo de Troca de Tarefa

O cérebro não processa duas tarefas complexas ao mesmo tempo — ele alterna entre elas. Pesquisadores chamam isso de **custo de troca de tarefa** (task-switching cost), com dois estágios: "goal shifting" (decidir fazer isso agora em vez daquilo) e "rule activation" (desligar as regras de uma tarefa, ligar as regras da outra). Esse modelo, proposto por Rubinstein, Meyer e Evans (2001), é a base conceitual de praticamente tudo que vem depois na pesquisa sobre multitarefa.

Mas o experimento mais direto sobre o assunto — pessoas reais, interrompidas no meio de uma tarefa real — encontrou algo que a maioria dos resumos populares sobre multitarefa não conta: as pessoas **não ficaram mais lentas**. Ficaram mais rápidas. O preço não apareceu no relógio. Apareceu em como elas se sentiram fazendo o trabalho.

---

### 🧮 O Modelo
**Fontes:**
- Rubinstein, J.S., Meyer, D.E. & Evans, J.E. (2001) — *Executive Control of Cognitive Processes in Task Switching* — Journal of Experimental Psychology: Human Perception and Performance, 27(4), 763-797 (base conceitual, sem dado quantitativo usado no modelo abaixo)
- Mark, G., Gudith, D. & Klocke, U. (2008) — *The Cost of Interrupted Work: More Speed and Stress* — CHI '08: Proceedings of the SIGCHI Conference on Human Factors in Computing Systems, 107-110 — N=48
- David Meyer, citado pela American Psychological Association (apa.org/topics/research/multitasking) e em entrevista (Anderson, 2001) — estimativa verbal de 20% a 40% de perda de tempo produtivo, sem N nem estudo controlado por trás


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print('✅ Bibliotecas carregadas')
```

```python
# --- DADOS DA LITERATURA ---
# Mark, Gudith & Klocke (2008) — CHI Proceedings — N=48 (universitarios alemaes, media 26 anos)
# Tarefa: responder 12 e-mails simulando um gestor de RH, interrompidos por telefone/IM a cada ~2min

n_estudo = 48

# Tempo para concluir a tarefa completa (minutos) — Tabela 1 do paper
tempo_sem_interrupcao   = 22.77  # condicao baseline (sem interrupcao)
tempo_interrupcao_mesmo_contexto = 20.31  # interrupcao do mesmo assunto da tarefa
tempo_interrupcao_contexto_diferente = 20.60  # interrupcao de assunto diferente

# Numero de palavras por e-mail (mostra parte do motivo da velocidade: escreveram menos)
palavras_sem_interrupcao = 31.49
palavras_interrupcao_mesmo = 29.17
palavras_interrupcao_diferente = 30.16

# Medidas de carga psicologica (escala NASA-TLX, 1-20) — Tabela 3 do paper
estresse_sem_interrupcao = 6.92
estresse_interrupcao_mesmo = 9.46
estresse_interrupcao_diferente = 9.13

frustracao_sem_interrupcao = 4.73
frustracao_interrupcao_mesmo = 6.63
frustracao_interrupcao_diferente = 6.48

esforco_sem_interrupcao = 9.50
esforco_interrupcao_mesmo = 11.04
esforco_interrupcao_diferente = 11.52

carga_mental_sem_interrupcao = 10.02
carga_mental_interrupcao_mesmo = 10.83
carga_mental_interrupcao_diferente = 11.50

# Estimativa verbal de David Meyer (via APA / entrevista) — NAO vem deste estudo
perda_produtiva_min_pct = 0.20
perda_produtiva_max_pct = 0.40

fator_correcao = 0.80

print('=' * 65)
print('  DADOS DA LITERATURA — CUSTO DE FAZER DUAS COISAS AO MESMO TEMPO')
print('=' * 65)
print(f'\n  Mark, Gudith & Klocke (2008), N={n_estudo}:')
print(f'  → Tempo de tarefa sem interrupção: {tempo_sem_interrupcao:.2f} min')
print(f'  → Tempo de tarefa com interrupção (mesmo contexto): {tempo_interrupcao_mesmo_contexto:.2f} min')
print(f'  → Tempo de tarefa com interrupção (contexto diferente): {tempo_interrupcao_contexto_diferente:.2f} min')
print(f'\n  Carga psicológica (escala 1-20), sem vs com interrupção:')
print(f'  → Estresse: {estresse_sem_interrupcao:.2f} → {estresse_interrupcao_mesmo:.2f}/{estresse_interrupcao_diferente:.2f}')
print(f'  → Frustração: {frustracao_sem_interrupcao:.2f} → {frustracao_interrupcao_mesmo:.2f}/{frustracao_interrupcao_diferente:.2f}')
print(f'  → Esforço: {esforco_sem_interrupcao:.2f} → {esforco_interrupcao_mesmo:.2f}/{esforco_interrupcao_diferente:.2f}')
print(f'  → Carga mental: {carga_mental_sem_interrupcao:.2f} → {carga_mental_interrupcao_mesmo:.2f}/{carga_mental_interrupcao_diferente:.2f}')
print(f'\n  Estimativa verbal de David Meyer (via APA/entrevista, fora deste estudo):')
print(f'  → Perda de tempo produtivo ao alternar tarefas: {perda_produtiva_min_pct*100:.0f}% a {perda_produtiva_max_pct*100:.0f}%')
print(f'\n  Fator de correção aplicado (apenas à estimativa de Meyer): x{fator_correcao}')
print('=' * 65)
print()
print('  ⚠️  NOTA METODOLÓGICA:')
print('  A famosa estatística de "23 minutos para retomar o foco", muito')
print('  citada na internet como vinda deste mesmo estudo (Mark, Gudith &')
print('  Klocke, 2008), NÃO aparece em nenhuma parte do paper original —')
print('  conferido linha por linha. Investigações independentes rastrearam')
print('  a origem real dessa frase a uma entrevista de Gloria Mark à Gallup')
print('  em 2006, sem publicação revisada por pares por trás do número.')
print('  Por isso ela NÃO é usada neste modelo. Os dados abaixo vêm direto')
print('  da Tabela 1 e Tabela 3 do paper de 2008, lido na íntegra.')
```

```python
# --- O MODELO ---

# Tempo medio das duas condicoes COM interrupcao
tempo_medio_interrompido = (tempo_interrupcao_mesmo_contexto + tempo_interrupcao_contexto_diferente) / 2
ganho_velocidade_pct = (tempo_sem_interrupcao - tempo_medio_interrompido) / tempo_sem_interrupcao

# Reducao no numero de palavras (parte da explicacao da velocidade: escreveram menos)
palavras_medio_interrompido = (palavras_interrupcao_mesmo + palavras_interrupcao_diferente) / 2
reducao_palavras_pct = (palavras_sem_interrupcao - palavras_medio_interrompido) / palavras_sem_interrupcao

# Aumento percentual nas medidas de carga psicologica (media das duas condicoes interrompidas)
estresse_medio_interrompido = (estresse_interrupcao_mesmo + estresse_interrupcao_diferente) / 2
aumento_estresse_pct = (estresse_medio_interrompido - estresse_sem_interrupcao) / estresse_sem_interrupcao

frustracao_medio_interrompido = (frustracao_interrupcao_mesmo + frustracao_interrupcao_diferente) / 2
aumento_frustracao_pct = (frustracao_medio_interrompido - frustracao_sem_interrupcao) / frustracao_sem_interrupcao

esforco_medio_interrompido = (esforco_interrupcao_mesmo + esforco_interrupcao_diferente) / 2
aumento_esforco_pct = (esforco_medio_interrompido - esforco_sem_interrupcao) / esforco_sem_interrupcao

carga_mental_medio_interrompido = (carga_mental_interrupcao_mesmo + carga_mental_interrupcao_diferente) / 2
aumento_carga_mental_pct = (carga_mental_medio_interrompido - carga_mental_sem_interrupcao) / carga_mental_sem_interrupcao

# Estimativa de Meyer, com fator de correcao
perda_produtiva_corrigida_min_pct = perda_produtiva_min_pct * fator_correcao
perda_produtiva_corrigida_max_pct = perda_produtiva_max_pct * fator_correcao

print('=' * 65)
print('  MODELO — RESULTADO')
print('=' * 65)
print(f'  Velocidade:')
print(f'  → Tarefas interrompidas foram concluídas {ganho_velocidade_pct*100:.1f}% mais rápido')
print(f'  → Parte da explicação: e-mails {reducao_palavras_pct*100:.1f}% mais curtos quando interrompido')
print()
print(f'  Custo psicológico (aumento vs. sem interrupção):')
print(f'  → Estresse: +{aumento_estresse_pct*100:.0f}%')
print(f'  → Frustração: +{aumento_frustracao_pct*100:.0f}%')
print(f'  → Esforço percebido: +{aumento_esforco_pct*100:.0f}%')
print(f'  → Carga mental: +{aumento_carga_mental_pct*100:.0f}%')
print()
print(f'  Estimativa separada de Meyer (qualitativa, não deste estudo):')
print(f'  → Original: {perda_produtiva_min_pct*100:.0f}% a {perda_produtiva_max_pct*100:.0f}%')
print(f'  → Corrigida (×{fator_correcao}): {perda_produtiva_corrigida_min_pct*100:.0f}% a {perda_produtiva_corrigida_max_pct*100:.0f}%')
print('=' * 65)
```

```python
# --- VISUALIZACAO ---

# GRAFICO 1 — Tempo de conclusao da tarefa, por condicao
fig1, ax1 = plt.subplots(figsize=(12, 8))

categorias = ['Sem\ninterrupção', 'Interrupção\n(mesmo contexto)', 'Interrupção\n(contexto diferente)']
valores = [tempo_sem_interrupcao, tempo_interrupcao_mesmo_contexto, tempo_interrupcao_contexto_diferente]
cores1 = ['#888888', '#2a8a82', '#2a8a82']

bars = ax1.bar(categorias, valores, color=cores1, alpha=0.85, width=0.5)
ax1.set_ylim(0, 28)
ax1.set_ylabel('Tempo para concluir a tarefa (min)', fontsize=12)
ax1.set_title('Tarefas interrompidas foram concluídas mais rápido\nMark, Gudith & Klocke (2008), N=48',
              fontsize=13, pad=15)

for bar, v in zip(bars, valores):
    ax1.text(bar.get_x() + bar.get_width() / 2, v + 0.5,
             f'{v:.1f} min', ha='center', fontweight='bold', fontsize=14)

plt.tight_layout()
plt.savefig('dia-020-grafico-01-tempo-tarefa.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Grafico 1 salvo!')


# GRAFICO 2 — Carga psicologica, sem vs com interrupcao (medias)
fig2, ax2 = plt.subplots(figsize=(12, 8))

dimensoes = ['Estresse', 'Frustração', 'Esforço', 'Carga mental']
sem_interrupcao_vals = [estresse_sem_interrupcao, frustracao_sem_interrupcao, esforco_sem_interrupcao, carga_mental_sem_interrupcao]
com_interrupcao_vals = [estresse_medio_interrompido, frustracao_medio_interrompido, esforco_medio_interrompido, carga_mental_medio_interrompido]

x = np.arange(len(dimensoes))
largura = 0.35

ax2.bar(x - largura/2, sem_interrupcao_vals, largura, label='Sem interrupção', color='#888888', alpha=0.85)
ax2.bar(x + largura/2, com_interrupcao_vals, largura, label='Com interrupção (média)', color='#c0392b', alpha=0.85)

ax2.set_ylabel('Escala NASA-TLX (1-20)', fontsize=12)
ax2.set_xticks(x)
ax2.set_xticklabels(dimensoes)
ax2.set_title('Mais rápido não é de graça\nMark, Gudith & Klocke (2008), N=48',
              fontsize=13, pad=15)
ax2.legend(fontsize=11)

plt.tight_layout()
plt.savefig('dia-020-grafico-02-carga-psicologica.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Grafico 2 salvo!')


# GRAFICO 3 — Estimativa separada de Meyer (qualitativa, fora do estudo principal)
fig3, ax3 = plt.subplots(figsize=(12, 8))

categorias3 = ['Estimativa original\n(Meyer, via APA)', 'Estimativa corrigida\n(×0,80)']
minimos = [perda_produtiva_min_pct * 100, perda_produtiva_corrigida_min_pct * 100]
maximos = [perda_produtiva_max_pct * 100, perda_produtiva_corrigida_max_pct * 100]
centros = [(mi + ma) / 2 for mi, ma in zip(minimos, maximos)]
erros = [(ma - mi) / 2 for mi, ma in zip(minimos, maximos)]
cores3 = ['#c8a84b', '#c8a84b']

bars3 = ax3.bar(categorias3, centros, yerr=erros, color=cores3, alpha=0.85, width=0.45,
                capsize=12, error_kw={'elinewidth': 2.5, 'capthick': 2.5})
ax3.set_ylim(0, 50)
ax3.set_ylabel('Perda estimada de tempo produtivo (%)', fontsize=12)
ax3.set_title('Uma estimativa separada, não deste estudo\nDavid Meyer, citado pela APA — sem N nem estudo controlado',
              fontsize=13, pad=15)

for bar, mi, ma in zip(bars3, minimos, maximos):
    ax3.text(bar.get_x() + bar.get_width() / 2, ma + 3,
             f'{mi:.0f}%–{ma:.0f}%', ha='center', fontweight='bold', fontsize=14)

plt.tight_layout()
plt.savefig('dia-020-grafico-03-estimativa-meyer.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Grafico 3 salvo!')
```

### 💡 O Insight

Em um experimento com 48 pessoas respondendo e-mails de trabalho, quem foi interrompido por telefone ou mensagem **completou a tarefa mais rápido**, não mais lento, do que quem trabalhou sem interrupção. Não por mágica: parte da explicação é que as pessoas escreveram e-mails mais curtos quando interrompidas, compensando o tempo perdido.

O preço não apareceu no relógio. Apareceu em como as pessoas relataram se sentir: estresse, frustração e esforço percebido todos significativamente mais altos em quem foi interrompido — mesmo sem diferença de qualidade no trabalho entregue (sem mais erros, sem perda de educação nos e-mails).

Isso muda a pergunta que vale fazer. Não é "multitarefa te deixa mais lenta?" — pelo menos neste experimento, a resposta foi não. É "o que você está disposta a pagar pra manter esse ritmo?" porque o custo é real, só que ele é cobrado em você, não no cronômetro.

*Da próxima vez que sentir orgulho de ter feito três coisas ao mesmo tempo, vale perguntar: mais rápido em quê — e a que custo?*

### ⚠️ Limitações do Modelo
- O estudo foi conduzido em laboratório, com uma tarefa específica (responder e-mails) e um público específico (universitários alemães, média de 26 anos) — pode não generalizar para todo tipo de trabalho ou perfil profissional
- O "ganho de velocidade" não é gratuito: ele veio acompanhado de aumento real e estatisticamente significativo em estresse, frustração e esforço percebido — não é correto interpretar isso como "interromper as pessoas é bom"
- Parte da velocidade extra se explica por e-mails mais curtos, não necessariamente por mais eficiência — as pessoas podem ter sacrificado profundidade, não só tempo
- A interrupção neste estudo era curta e previsível (a cada ~2 minutos, de um "supervisor"); isso é diferente de interrupções longas, frequentes ou imprevisíveis do dia a dia real
- A estimativa de 20%-40% de David Meyer é uma declaração verbal de um pesquisador, citada pela APA e em entrevista — não é um resultado experimental com N e IC 95%, e não vem deste estudo específico
- A popular estatística de "23 minutos para retomar o foco", frequentemente atribuída a este mesmo estudo (Mark, Gudith & Klocke, 2008), não foi encontrada em nenhuma parte do paper original e, por investigações externas, parece não ter publicação revisada por pares por trás dela — por isso não foi usada aqui
- O fator de correção (×0,80) foi aplicado apenas à estimativa de Meyer; os dados de Mark et al. já são valores medidos diretamente, sem correção

*A ciência é honesta sobre o que não sabe. O modelo também.*
