### O Conceito: o sistema inteiro num número

VO₂máx é o volume máximo de oxigênio que o seu corpo consegue
captar, transportar e usar por minuto.

E é por isso que ele sabe tanto: é o único número que testa
a cadeia inteira de uma vez. Pulmão que capta, coração que
bombeia, sangue que transporta, músculo e mitocôndria que usam.
Pressão mede um elo. Glicose mede outro. O fôlego testa o sistema.

Quando esse número cai, alguma parte da cadeia envelheceu.
E o Baltimore Longitudinal Study of Aging passou décadas
medindo e remedindo as mesmas pessoas para descobrir
como essa queda acontece.

A resposta mudou o que se sabia sobre o relógio.

### O Modelo

**Fonte central (o relógio):**
- Fleg, Morrell, Bos, Brant, Talbot, Wright & Lakatta, 2005
  Accelerated longitudinal decline of aerobic capacity in healthy
  older adults — Circulation, 112(5), 674-682
  Baltimore Longitudinal Study of Aging: N=810 (375 mulheres,
  435 homens), 21 a 87 anos, VO₂pico medido em série na mesma
  pessoa, seguimento mediano de 7,9 anos, sem doença cardíaca
  Declínio ACELERA: de 3-6% por década (20-30 anos)
  para mais de 20% por década (70+), nos dois sexos,
  em todos os quartis de atividade física
  Componente que cai: o pulso de oxigênio (O₂ por batimento);
  a frequência cardíaca máxima cai só 4-6% por década

**Régua de apoio (por que o relógio importa):**
- Kodama et al., 2009 — JAMA, 301(19) — 33 coortes saudáveis,
  N=102.980: cada 1 MET a mais, RR=0,87 [0,84, 0,90];
  grupo baixo começa abaixo de 7,9 METs (27,7 ml/kg/min)


```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---

# Fleg et al., 2005 — Circulation, 112(5) — Baltimore Longitudinal Study
n_total = 810
n_mulheres, n_homens = 375, 435
idades = "21 a 87"
seguimento = 7.9   # anos, mediano

# Taxas de declinio do VO2pico POR DECADA de idade (achado central):
# aceleram de 3-6% (20s-30s) para >20% (70s+), nos dois sexos,
# em todos os quartis de atividade fisica autorrelatada
taxa_20s_30s = (0.03, 0.06)   # 3 a 6% por decada
taxa_70s = 0.20               # mais de 20% por decada
# Regua de bolso da literatura: ~10% por decada a partir dos 30 (media)

# Decomposicao (achado fino):
queda_fc_max_decada = (0.04, 0.06)   # frequencia cardiaca maxima: 4-6%/decada
# O que despenca e o PULSO DE OXIGENIO (O2 entregue por batimento),
# que espelha a queda do VO2pico

# Kodama et al., 2009 — regua de apoio (populacoes saudaveis)
rr_por_met = (0.87, 0.84, 0.90)   # cada 1 MET a mais
limiar_baixo_mets = 7.9           # inicio do grupo baixo
limiar_baixo_mlkg = limiar_baixo_mets * 3.5   # 27,65 ml/kg/min

# Fator ×0,80: NAO aplicado (VO2 medido em esteira, longitudinal)
aplica_fator_080 = False

print("=" * 65)
print("  DADOS DA LITERATURA — O RELOGIO DO FOLEGO")
print("=" * 65)
print(f"\n  Fleg et al., 2005 (Baltimore Longitudinal Study):")
print(f"  N={n_total} ({n_mulheres} mulheres, {n_homens} homens), {idades} anos")
print(f"  Medidas seriadas na MESMA pessoa; seguimento {seguimento} anos")
print(f"\n  O ponteiro acelera (queda do VO2pico por decada de idade):")
print(f"  Aos 20-30 anos:   {taxa_20s_30s[0]*100:.0f} a {taxa_20s_30s[1]*100:.0f}% por decada")
print(f"  Depois dos 70:    mais de {taxa_70s*100:.0f}% por decada")
print(f"  Em todos os quartis de atividade fisica")
print(f"\n  A decomposicao do relogio:")
print(f"  FC maxima cai so {queda_fc_max_decada[0]*100:.0f}-{queda_fc_max_decada[1]*100:.0f}% por decada")
print(f"  O que despenca e o PULSO DE OXIGENIO: quanto cada batida entrega")
print(f"\n  Regua de apoio (Kodama 2009, 33 coortes saudaveis):")
print(f"  Cada 1 MET a mais: RR = {rr_por_met[0]:.2f} [{rr_por_met[1]:.2f}, {rr_por_met[2]:.2f}]")
print(f"  Linha do grupo baixo: {limiar_baixo_mets} METs ({limiar_baixo_mlkg:.1f} ml/kg/min)")
print(f"\n  Fator ×0,80: NAO aplicado (medidas objetivas, longitudinais)")
print("=" * 65)
```

```python
# --- O MODELO ---
# O relogio que acelera, e quanto a altura compra de tempo

# Taxas por decada (estilizacao DECLARADA, interpolando as ancoras do Fleg:
# 4,5% na media dos 20s-30s ate 25% nos 80s; ancoras publicadas em
# negrito: 3-6% nos 20s-30s e >20% nos 70s)
decadas = np.array([25, 35, 45, 55, 65, 75, 85])
taxas   = np.array([0.045, 0.06, 0.10, 0.14, 0.18, 0.22, 0.25])

def trajetoria(vo2_aos_30, idades_alvo):
    # aplica as taxas por decada multiplicativamente, interpolando ao ano
    vo2 = np.zeros_like(idades_alvo, dtype=float)
    for i, idade in enumerate(idades_alvo):
        v = vo2_aos_30
        a = 30.0
        while a < idade:
            passo = min(1.0, idade - a)
            idx = np.searchsorted(decadas, a + 5) - 0
            taxa_anual = np.interp(a, decadas, taxas) / 10.0
            v *= (1 - taxa_anual) ** passo
            a += passo
        vo2[i] = v
    return vo2

idades_curva = np.linspace(30, 90, 400)

# Duas alturas ILUSTRATIVAS aos 30 anos (mesma taxa, alturas diferentes):
vo2_treinada  = 42.0   # ml/kg/min
vo2_sedentaria = 33.0

curva_alta  = trajetoria(vo2_treinada, idades_curva)
curva_baixa = trajetoria(vo2_sedentaria, idades_curva)

def idade_cruzamento(curva):
    abaixo = np.where(curva < limiar_baixo_mlkg)[0]
    return idades_curva[abaixo[0]] if len(abaixo) else np.inf

cruza_alta  = idade_cruzamento(curva_alta)
cruza_baixa = idade_cruzamento(curva_baixa)
anos_comprados = cruza_alta - cruza_baixa

print("=" * 65)
print("  MODELO — A ALTURA COMPRA TEMPO")
print("=" * 65)
print(f"\n  A mesma taxa de queda para todos (Fleg: independe do quartil")
print(f"  de atividade). O que o treino muda e a ALTURA de partida.")
print(f"\n  Duas trajetorias ilustrativas (mesmo relogio, alturas diferentes):")
print(f"  Partida aos 30: {vo2_treinada:.0f} vs {vo2_sedentaria:.0f} ml/kg/min")
print(f"\n  Quando cada uma cruza a linha do grupo baixo do Kodama")
print(f"  ({limiar_baixo_mlkg:.1f} ml/kg/min = {limiar_baixo_mets} METs):")
print(f"  Trajetoria baixa:  aos {cruza_baixa:.0f} anos")
print(f"  Trajetoria alta:   aos {cruza_alta:.0f} anos")
print(f"\n  A ALTURA COMPROU {anos_comprados:.0f} ANOS acima da linha.")
print(f"\n  O relogio ninguem para. A altura de onde se desce, sim.")
print("=" * 65)
```

```python
# --- VISUALIZACAO — GRAFICOS SEPARADOS ---

# GRAFICO 1 — O ponteiro acelera
fig1, ax1 = plt.subplots(figsize=(10, 6))
labels = ['20s', '30s', '40s', '50s', '60s', '70s', '80s']
cores = ['#2a8a82' if t < 0.10 else '#c8a84b' if t < 0.18 else '#c0392b' for t in taxas]
bars = ax1.bar(labels, taxas*100, color=cores, alpha=0.9, width=0.62)
for b, t in zip(bars, taxas):
    ax1.text(b.get_x()+b.get_width()/2, t*100+0.5, f'{t*100:.0f}%',
             ha='center', fontweight='bold', fontsize=12)
ax1.set_ylabel('Queda do VO2pico por decada (%)')
ax1.set_xlabel('Decada de vida')
ax1.set_title('O ponteiro acelera — a queda do folego por decada\nFleg et al. 2005 (longitudinal, N=810): ancoras 3-6% e >20%; degraus intermediarios interpolados',
              fontsize=12, pad=15)
plt.figtext(0.5, 0.01, 'Fonte: Fleg et al., 2005, Circulation | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-099-grafico-01-relogio.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 1 salvo!")

# GRAFICO 2 — A altura compra tempo
fig2, ax2 = plt.subplots(figsize=(10, 6))
ax2.plot(idades_curva, curva_alta, color='#2a8a82', linewidth=3,
         label=f'Partida alta (42 ml/kg aos 30)')
ax2.plot(idades_curva, curva_baixa, color='#c0392b', linewidth=3,
         label=f'Partida baixa (33 ml/kg aos 30)')
ax2.axhline(y=limiar_baixo_mlkg, color='#c8a84b', linestyle='--', linewidth=2)
ax2.text(31, limiar_baixo_mlkg-1.6, f'linha do grupo baixo (Kodama): {limiar_baixo_mets} METs',
         fontsize=10, color='#8a6d2f')
for cruza, cor in [(cruza_baixa, '#c0392b'), (cruza_alta, '#2a8a82')]:
    ax2.axvline(x=cruza, color=cor, linestyle=':', linewidth=1.5, alpha=0.8)
ax2.annotate(f'{anos_comprados:.0f} anos comprados\npela altura de partida',
             xy=((cruza_alta+cruza_baixa)/2, limiar_baixo_mlkg+0.6),
             xytext=(56, 36), fontsize=12, color='#8a6d2f', fontweight='bold',
             arrowprops=dict(arrowstyle='->', color='#c8a84b', lw=1.5))
ax2.set_xlabel('Idade')
ax2.set_ylabel('VO2pico (ml/kg/min)')
ax2.set_title('A altura compra tempo — mesmo relogio, partidas diferentes\nTrajetorias ilustrativas com as taxas do Fleg; limiar do Kodama (suposicoes declaradas)',
              fontsize=12, pad=15)
ax2.legend(loc='upper right', frameon=False)
plt.figtext(0.5, 0.01, 'Conta propria sobre Fleg 2005 e Kodama 2009 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-099-grafico-02-altura.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 2 salvo!")

# GRAFICO 3 — A decomposicao: cada batida entrega menos
fig3, ax3 = plt.subplots(figsize=(10, 6))
comp = ['Frequencia cardiaca\nmaxima', 'Pulso de oxigenio\n(O2 por batimento)']
quedas = [5, 14]   # %/decada: FC 4-6 (ponto medio 5); pulso espelha o VO2 (~media da vida adulta)
cores3 = ['#2a8a82', '#c0392b']
bars3 = ax3.bar(comp, quedas, color=cores3, alpha=0.9, width=0.45)
for b, q in zip(bars3, quedas):
    ax3.text(b.get_x()+b.get_width()/2, q+0.3, f'~{q}%/decada',
             ha='center', fontweight='bold', fontsize=13)
ax3.set_ylabel('Queda tipica por decada (%)')
ax3.set_ylim(0, 18)
ax3.set_title('O relogio nao para de bater — cada batida entrega menos\nFleg et al. 2005: FC maxima cai 4-6%/decada; o pulso de O2 espelha a queda do VO2pico',
              fontsize=12, pad=15)
ax3.text(0.5, 0.04, 'Barras em valores tipicos declarados: FC no ponto medio do intervalo publicado;\npulso de O2 na media adulta da queda do VO2 (espelhamento reportado pelos autores)',
         transform=ax3.transAxes, ha='center', fontsize=9, color='#555')
plt.figtext(0.5, 0.01, 'Fonte: Fleg et al., 2005, Circulation | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-099-grafico-03-pulso.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 3 salvo!")
```

### O Insight

**O fôlego cai para todo mundo. E cai cada vez mais rápido: uns 5% por década na juventude, mais de 20% por década depois dos 70.**

Treinar não impede essa queda. Isso os dados dizem com clareza, e é a parte que ninguém gosta de ouvir.

Mas a conta importante não é a velocidade da queda. É o ponto de partida.

Quem chega aos 30, 40, 50 com o fôlego lá em cima desce a mesma ladeira que todo mundo, só que começando de muito mais alto. Na conta deste notebook, essa diferença de partida vale **16 anos a mais** longe da zona de risco. Dezesseis anos comprados não com genética, mas com treino acumulado.

E tem um detalhe que muda como eu vejo o próprio corpo: com a idade, o coração quase não perde ritmo. O que ele perde é força de entrega. Ele continua batendo como sempre bateu, só que cada batida leva menos oxigênio para o resto do corpo.

Então, quando o fôlego encurta na escada, não é o corpo desistindo. É o corpo avisando de que altura você está descendo.

*A queda é igual para todo mundo. A altura, não. De onde você quer descer?*

### Limitacoes do Modelo

- As taxas por decada intermediarias sao interpoladas entre as
  ancoras publicadas (3-6% e >20%); os degraus exatos por decada
  estao no artigo original e a interpolacao e declarada no grafico
- As duas trajetorias sao ILUSTRATIVAS: alturas de partida
  escolhidas para o exemplo; a taxa igual entre elas segue o achado
  do Fleg (declinio semelhante em todos os quartis de atividade)
- O limiar de 7,9 METs e o corte do grupo baixo do Kodama, nao uma
  fronteira biologica; cruza-lo nao e um evento, e uma regiao
- Coorte de Baltimore: saudavel, comunitaria, majoritariamente
  branca e escolarizada; generalizacao declarada com cautela
- Associacao, nao causa, na regua do Kodama; VO2 estimado por
  esteira, nao por analise de gases, em parte das coortes
- Cordas declaradas: #043 (destreinamento: o que acontece quando
  o treino para), #018 e #092 (eixo mortalidade, reguas distintas)
