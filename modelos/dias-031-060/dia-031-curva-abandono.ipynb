### 📚 O Conceito: A Curva de Abandono

Existe uma linha inteira de pesquisa em psicologia comportamental sobre o que acontece com as pessoas depois que elas fazem uma promessa a si mesmas.

A pergunta central: entre a promessa e o resultado, o que erode?

John Norcross, da Universidade de Scranton, é o pesquisador que mais estudou isso. Em 2002, publicou no Journal of Clinical Psychology um estudo prospectivo com 200 pessoas que fizeram resoluções de Ano Novo. Ele acompanhou cada uma por dois anos, medindo quem ainda mantinha o compromisso a cada ponto do tempo.

O achado central não é que muita gente desiste. É que a desistência segue uma curva previsível.

E essa curva foi replicada em 2020 por Oscarsson e colegas, num estudo sueco muito maior (N=1.066), publicado no PLoS ONE. Os números batem quase perfeitamente com Norcross, o que é raro em psicologia.

Ou seja: quase vinte anos separam os dois estudos, e a natureza humana permaneceu constante. A curva de abandono de projetos pessoais é um dos padrões mais estáveis da literatura.

Além disso, Oscarsson descobriu algo interessante sobre a **natureza** da meta. Metas de aproximação ("vou correr três vezes por semana") têm taxa de sucesso significativamente maior que metas de evitação ("vou parar de comer açúcar"). A diferença chega a 12 pontos percentuais no acompanhamento de um ano.

O motivo hipotetizado é que metas de aproximação criam comportamento visível, mensurável, comemorável. Metas de evitação criam uma vigilância permanente sobre o que **não** fazer, o que é psicologicamente exaustivo.

### 🧮 O Modelo

**Fontes principais:**
- Norcross, J. C., Mrykalo, M. S. & Blagys, M. D. (2002) — "Auld Lang Syne: Success predictors, change processes, and self-reported outcomes of New Year's resolvers and nonresolvers." *Journal of Clinical Psychology*, 58(4), 397-405 — **N=200**, coorte prospectiva, 2 anos de acompanhamento
- Oscarsson, M., Carlbring, P., Andersson, G. & Rozental, A. (2020) — "A large-scale experiment on New Year's resolutions: Approach-oriented goals are more successful than avoidance-oriented goals." *PLoS ONE* 15(12):e0234097 — **N=1.066**, coorte sueca, 1 ano de acompanhamento
- Lally, P., van Jaarsveld, C. H. M., Potts, H. W. W. & Wardle, J. (2010) — "How are habits formed: Modelling habit formation in the real world." *European Journal of Social Psychology*, 40(6), 998-1009 — sobre tempo médio de formação de hábitos

**Nota metodológica:** aplico o fator de correção ×0.80 neste dia. Os números são proporções de survey autorrelatadas em coortes prospectivas, categoria clássica para aplicação do fator conservador do projeto. Aplico o fator sobre a taxa de sobrevivência em 6 meses (40% → ~32%), que é o número que uso como referência central no insight.

Uso a convergência entre Norcross (~40%) e Oscarsson (~40%) para o ponto de 6 meses. Como os dois convergiram, essa média é um dos números mais bem sustentados da literatura.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Norcross, Mrykalo & Blagys (2002), J Clin Psychology 58(4):397-405
# Oscarsson, Carlbring, Andersson & Rozental (2020), PLoS ONE 15(12)

n_norcross = 200
n_oscarsson = 1066

# Taxa de sobrevivência ao longo do tempo (média dos dois estudos)
tempo_dias = np.array([0, 7, 30, 90, 180, 730])
tempo_labels = ['inicio', '1 semana', '1 mes', '3 meses', '6 meses', '2 anos']
sobrevivencia = np.array([1.00, 0.77, 0.55, 0.43, 0.40, 0.19])

# Achado adicional (Oscarsson 2020): metas de aproximacao vs evitacao
p_sucesso_aproximacao = 0.59
p_sucesso_evitacao = 0.47
gap_aproximacao = p_sucesso_aproximacao - p_sucesso_evitacao

# Fator de correcao do projeto (survey autorrelato)
fator_correcao = 0.80

# Aplicacao do fator ao ponto central (6 meses)
p_6meses_bruto = 0.40
p_6meses_corrigido = p_6meses_bruto * fator_correcao

print("=" * 68)
print("  DADOS — CURVA DE ABANDONO DE PROJETOS PESSOAIS")
print("=" * 68)
print(f"\n  Fontes:")
print(f"    Norcross et al. (2002)   — N={n_norcross}")
print(f"    Oscarsson et al. (2020)  — N={n_oscarsson:,}")

print(f"\n  TAXA DE SOBREVIVENCIA AO LONGO DO TEMPO:")
for label, s in zip(tempo_labels, sobrevivencia):
    print(f"    {label:<12}    {s*100:>4.0f}%")

print(f"\n  FATOR DE CORRECAO (aplicado ao ponto de 6 meses):")
print(f"  Bruto:      {p_6meses_bruto*100:.0f}%")
print(f"  Corrigido:  ~{p_6meses_corrigido*100:.0f}%  (x{fator_correcao})")

print(f"\n  META DE APROXIMACAO vs EVITACAO (Oscarsson, 1 ano):")
print(f"  Aproximacao ('vou fazer X'):    {p_sucesso_aproximacao*100:.0f}%")
print(f"  Evitacao ('vou parar de Y'):    {p_sucesso_evitacao*100:.0f}%")
print(f"  Diferenca:                      +{gap_aproximacao*100:.0f}pp a favor de aproximacao")
print("=" * 68)
```

```python
# --- O MODELO ---
# A curva de mortalidade e onde estou hoje (dia 31)

dia_atual = 31
sobrevivencia_dia_atual = np.interp(dia_atual, tempo_dias, sobrevivencia)

# "Vale da duvida" — entre 7 e 90 dias
queda_vale = sobrevivencia[1] - sobrevivencia[3]

print("=" * 68)
print("  MODELO — ONDE EU ESTOU HOJE (dia 31)")
print("=" * 68)
print(f"\n  Sobrevivencia esperada aproximada no dia {dia_atual}:")
print(f"  {sobrevivencia_dia_atual*100:.0f}% dos projetos pessoais ainda ativos")
print(f"  Ou seja: quase metade dos que comecaram ja sairam.")

print(f"\n  VALE DA DUVIDA (entre 1 semana e 3 meses):")
print(f"  Queda de {queda_vale*100:.0f} pontos percentuais na sobrevivencia")
print(f"  E a fase mais ingreme da curva de mortalidade")
print(f"  Entusiasmo inicial passou, resultados ainda nao vieram")

print(f"\n  A PROTECAO POSSIVEL:")
print(f"  Metas de APROXIMACAO: +{gap_aproximacao*100:.0f}pp de chance de sucesso")
print(f"    em relacao a metas de EVITACAO")
print(f"  'Vou publicar hoje' e aproximacao. Cria evidencia.")
print(f"  'Vou parar de procrastinar' e evitacao. Cria vigilancia.")

print(f"\n  A LEITURA DIFICIL:")
print(f"  Estar no dia 31 nao me protege dos proximos vales.")
print(f"  A cada semana, um pouco mais da coorte original sai.")
print(f"  O objetivo nao e chegar aos 365. E publicar amanha.")
print("=" * 68)
```

```python
# --- VISUALIZACAO ---

cor_sobrevive = '#2a8a82'
cor_saiu = '#c0392b'
cor_ouro = '#c8a84b'
cor_neutra = '#9c9b94'

# GRAFICO 1 — A curva de sobrevivencia ao longo do tempo
fig1, ax1 = plt.subplots(figsize=(11, 6.5))

tempo_smooth = np.linspace(0, 730, 400)
sobrevivencia_smooth = np.interp(tempo_smooth, tempo_dias, sobrevivencia)

ax1.plot(tempo_smooth, sobrevivencia_smooth*100, color=cor_sobrevive, linewidth=3)
ax1.fill_between(tempo_smooth, sobrevivencia_smooth*100, alpha=0.15, color=cor_sobrevive)

for t, s, label in zip(tempo_dias, sobrevivencia, tempo_labels):
    ax1.plot(t, s*100, 'o', color=cor_sobrevive, markersize=9)
    ax1.text(t, s*100 + 4, f'{s*100:.0f}%', ha='center', fontweight='bold', fontsize=11)
    ax1.text(t, -6, label, ha='center', fontsize=9, color='#666')

sob_hoje = np.interp(31, tempo_dias, sobrevivencia)
ax1.axvline(x=31, color=cor_ouro, linestyle='--', linewidth=2, alpha=0.9)
ax1.plot(31, sob_hoje*100, 's', color=cor_ouro, markersize=13, zorder=5)
ax1.annotate(f'estou aqui\n(dia 31)\n~{sob_hoje*100:.0f}% ativos',
             xy=(31, sob_hoje*100), xytext=(110, 80),
             fontsize=11, color='#8a7020', fontweight='bold', ha='center',
             arrowprops=dict(arrowstyle='->', color=cor_ouro, lw=1.5))

ax1.set_xlabel('Tempo desde o inicio do projeto (dias)')
ax1.set_ylabel('% ainda mantendo o projeto')
ax1.set_ylim(-12, 108)
ax1.set_xlim(-20, 760)
ax1.set_xticks([])
ax1.set_title('A curva de mortalidade dos projetos pessoais\nMedia de Norcross (2002, N=200) e Oscarsson et al. (2020, N=1.066)',
              fontsize=13, pad=15)

plt.figtext(0.5, 0.005,
            'Curva media baseada em coortes prospectivas — numeros autorrelatados | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-031-grafico-01-curva-abandono.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 1 salvo!")

# GRAFICO 2 — O vale da duvida
fig2, ax2 = plt.subplots(figsize=(11, 6.5))

dias_vale = np.linspace(7, 90, 200)
sob_vale = np.interp(dias_vale, tempo_dias, sobrevivencia)

ax2.plot(dias_vale, sob_vale*100, color=cor_saiu, linewidth=3)
ax2.fill_between(dias_vale, sob_vale*100, alpha=0.18, color=cor_saiu)
ax2.axvspan(21, 90, alpha=0.08, color=cor_saiu)

for t, s, label in [(7, 0.77, '1 semana'),
                     (30, 0.55, '1 mes'),
                     (90, 0.43, '3 meses')]:
    ax2.plot(t, s*100, 'o', color=cor_saiu, markersize=9)
    ax2.text(t, s*100 + 2.5, f'{s*100:.0f}%', ha='center', fontweight='bold', fontsize=13)
    ax2.text(t, 35, label, ha='center', fontsize=10, color='#555', style='italic')

ax2.axvline(x=31, color=cor_ouro, linestyle='--', linewidth=2, alpha=0.9)
ax2.plot(31, np.interp(31, tempo_dias, sobrevivencia)*100, 's',
         color=cor_ouro, markersize=13, zorder=5)
ax2.text(31, 68, 'dia 31',
         ha='center', fontsize=11, color='#8a7020', fontweight='bold')

ax2.text(55, 47, 'vale da duvida',
         ha='center', fontsize=13, color=cor_saiu, style='italic', fontweight='bold')

ax2.set_xlabel('Dias desde o inicio')
ax2.set_ylabel('% ainda mantendo o projeto')
ax2.set_ylim(30, 85)
ax2.set_xlim(7, 90)
ax2.set_title('Zoom no vale da duvida\nA fase mais ingreme da queda: entre 3 semanas e 3 meses',
              fontsize=13, pad=15)

plt.figtext(0.5, 0.005,
            'Norcross (2002) e Oscarsson (2020) — queda de ~34pp em pouco mais de 2 meses | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-031-grafico-02-vale-duvida.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 2 salvo!")

# GRAFICO 3 — Aproximacao vs Evitacao
fig3, ax3 = plt.subplots(figsize=(11, 6.5))

categorias = ['Metas de EVITACAO\n"vou parar de..."', 'Metas de APROXIMACAO\n"vou fazer..."']
valores = [p_sucesso_evitacao*100, p_sucesso_aproximacao*100]
cores = [cor_saiu, cor_sobrevive]

bars = ax3.bar(categorias, valores, color=cores, alpha=0.88, width=0.5)
ax3.set_ylabel('Taxa de sucesso em 1 ano (%)')
ax3.set_ylim(0, 75)
ax3.set_title('A natureza da meta importa\nMetas de aproximacao superam metas de evitacao em ~12 pontos',
              fontsize=13, pad=15)

for bar, v in zip(bars, valores):
    ax3.text(bar.get_x() + bar.get_width()/2, v + 1.5,
             f'{v:.0f}%', ha='center', fontweight='bold', fontsize=17)

ax3.annotate('', xy=(1, p_sucesso_aproximacao*100), xytext=(0, p_sucesso_evitacao*100),
             arrowprops=dict(arrowstyle='->', color=cor_ouro, lw=2))
ax3.text(0.5, 0.78, f'+{gap_aproximacao*100:.0f}pp de vantagem',
         ha='center', fontsize=11, color='#8a7020', fontweight='bold',
         transform=ax3.transAxes)

plt.figtext(0.5, 0.005,
            'Fonte: Oscarsson et al. (2020), PLoS ONE — N=1.066 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-031-grafico-03-aproximacao.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 3 salvo!")
```

### 💡 O Insight

**Aproximadamente 32% dos projetos pessoais sobrevivem aos primeiros seis meses.**

Três em cada dez. E este é o número corrigido com o fator conservador do projeto. O número bruto da literatura é 40%, mas todo autorrelato tem viés de otimismo, então prefiro trabalhar com a versão mais defensável.

Mesmo assim: sete em cada dez desistem antes dos 180 dias.

O detalhe mais desconfortável está na forma da curva. Não é uma queda gradual. É uma escada. A primeira semana é fácil (77% ainda dentro). O primeiro mês já elimina quase metade dos que começaram (55%). O terceiro mês elimina mais (43%). Depois disso, a curva se estabiliza um pouco, mas continua descendo.

A fase entre a terceira semana e o terceiro mês é a mais perigosa. É o "vale da dúvida": o entusiasmo inicial já passou, a rotina ainda não se firmou, e os resultados concretos ainda estão longe.

Hoje é o dia trinta e um do meu projeto. Estou entrando nesse vale.

E é aqui que a pesquisa aponta uma direção prática. Oscarsson e colegas descobriram que metas de aproximação superam metas de evitação em cerca de 12 pontos percentuais de taxa de sucesso. "Vou correr três vezes por semana" é uma meta de aproximação. "Vou parar de comer açúcar" é uma meta de evitação. A primeira cria comportamento comemorável. A segunda cria vigilância exaustiva.

O 365 Probabilidades é, por acaso ou por sorte, uma meta de aproximação: produzir um modelo por dia. Cada dia é um pequeno "sim" concreto, não um "não" negativo repetido. Isso pode explicar, em parte, por que eu ainda estou aqui no dia trinta e um.

E a honestidade que este projeto exige: eu não sei se chego aos 365. A curva de mortalidade continua descendo até os 60, 90, 180 dias. A cada semana, um pouco mais de gente sai. Publicar hoje não é garantia de nada.

Mas há uma coisa que a ciência da persistência ensina, e que virou minha regra: o objetivo não é chegar aos 365. É publicar amanhã.

### ⚠️ Limitações do Modelo

- Os estudos de Norcross e Oscarsson são sobre resoluções de Ano Novo, um subconjunto específico de projetos pessoais (muitas vezes ligadas à saúde e comportamento). Generalizar para outros tipos de projeto (criativo, profissional, artístico) exige cautela.
- A taxa de sobrevivência varia enormemente por tipo de projeto, por presença de accountability social e por natureza da meta (aproximação vs evitação). Um projeto público (como este) tem taxa de sobrevivência maior do que um projeto privado, por conta da pressão social.
- Todos os dados são autorrelatados. Aplico ×0.80 sobre o número principal para compensar viés de otimismo.
- Amostras majoritariamente norte-americanas e europeias. Cultura afeta persistência.
- A curva mostrada é média populacional. Não descreve nenhum indivíduo específico. Você pode estar na cauda de 19% que chega aos 2 anos, ou nos 55% que caem antes de 1 mês. A média não é o seu destino.
