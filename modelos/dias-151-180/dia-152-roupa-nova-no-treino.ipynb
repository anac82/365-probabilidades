### 📖 A História

Esses dias, depois de uma longa viagem de carro, acordei sem vontade de treinar nem de voltar à rotina. Já tenho um hábito para quando a preguiça aparece e, ainda assim, às vezes eu desisto.

O hábito é não pensar. No início da semana, decido os dias de treino. Quando o despertador toca, não negocio com a minha mente. Nada de "só mais dez minutinhos" ou "amanhã a gente vai". Isso tem me ajudado muito.

Mas também achei um novo motivador. Não é dos mais sustentáveis, mas naquele dia funcionou. Numa viagem ao exterior, comprei um tênis super moderno e guardei a estreia dele para o dia em que nem o "não negociar" desse conta. Aquele foi o dia. A força veio da estreia do tênis novo, com toda aquela tecnologia.

Fiquei pensando no que exatamente aquele tênis mudou. Se foi o meu corpo que correu melhor, ou se fui eu que saí de casa achando que ia correr melhor.

### 📚 O Conceito: o placebo que a gente veste

Um placebo é algo que funciona porque a pessoa acredita que vai funcionar. Ele é mais
conhecido em remédios, mas pode aparecer em qualquer coisa que venha com uma promessa,
inclusive num tênis.

Em 2012, Hajo Adam e Adam Galinsky chamaram de "cognição vestida" a ideia de que a roupa
muda o jeito de pensar de quem a veste. No experimento mais famoso, vestir um jaleco
apresentado como de médico melhorou a atenção. Em 2019, Devin Burns e colegas refizeram esse
teste com registro prévio e não acharam efeito nenhum. Em 2025, uma meta-análise com 40
estudos, assinada pelos próprios autores do jaleco e um colega, concluiu que os estudos
antigos não se sustentam, mas que os feitos depois de 2015 indicam um efeito real da roupa no
que a pessoa pensa e sente.

Para separar o corpo da cabeça num treino de verdade, é preciso o mesmo tênis com duas
histórias diferentes. Foi o que um grupo da Nova Zelândia fez em 2025.

### 🧮 O Modelo

**Fontes:**
- Hébert-Losier, K., Pfister, A., Finlayson, S. J., Esculier, J. F., Lamb, P. & Beaven, C. M.
  (2025). Are super shoes a super placebo? A randomised crossover trial in female recreational
  runners. *Footwear Science*, doi 10.1080/19424280.2025.2458330. **N = 24** corredoras
  amadoras, ensaio cruzado e sorteado, 4 corridas de 6 minutos. O mesmo tênis (Nike ZoomX
  Vaporfly Next% 2) apresentado como tênis de elite ou, pintado de preto, como cópia
  falsificada. Consumo de oxigênio −0,05 ± 0,47 mL/kg/min (d = −0,02); gasto de energia
  −0,02 ± 0,17 W/kg (d = −0,03); biomecânica sem diferença. Conforto +14,6 ± 15,0 mm numa
  escala de 100 (d = 0,94); 87,5% preferiram o tênis "de elite"; corrida mais agradável,
  mais fácil, desempenho percebido melhor e menor risco de lesão percebido (d de 0,72 a 1,16).
- Adam, H. & Galinsky, A. D. (2012). Enclothed cognition. *Journal of Experimental Social
  Psychology*, 48(4), 918-925.
- Burns, D. M., Fox, E. L., Greenstein, M., Olbricht, G. R. & Montgomery, D. (2019). An old task
  in new clothes. *Journal of Experimental Social Psychology*, 83, 150-156. Réplica direta com
  registro prévio, sem efeito do jaleco.
- Horton, C. B., Adam, H. & Galinsky, A. D. (2025). Evaluating the evidence for enclothed
  cognition. *Personality and Social Psychology Bulletin*, 51(2), 203-221. 40 estudos, 105
  efeitos, N = 3.789. Números consolidados não conferidos no texto completo.
- **Curiosidade, a cor do uniforme:** Hill, R. A. & Barton, R. A. (2005). Red enhances human
  performance in contests. *Nature*, 435, 293. Olimpíada de 2004, quatro esportes de luta,
  cor sorteada: o vermelho venceu em mais disputas (χ² = 4,19, p = 0,041), em 16 de 21
  rodadas e em 19 de 29 categorias de peso. Pollet, T. V. & Peperkoorn, L. S. (2013), *Frontiers
  in Psychology*, 4:643: no UFC, quem lutou de calção vermelho venceu 27 de 61 lutas
  (p = 0,443). Fortunato, L. & Clauset, A. (2016), preprint bioRxiv 10.1101/086710: os dados
  de 2008 não repetem o efeito, e a vantagem de 2004 pode vir da estrutura dos chaveamentos.
  Hill & Barton responderam defendendo o efeito em lutas equilibradas.

**Desenho e família do teste:** ensaio cruzado e sorteado (cada corredora é o próprio
controle). O tamanho de efeito d de medidas pareadas recebe um intervalo de 95% pela
aproximação usual (erro-padrão de raiz de 1/n + d²/2n). Cada d vira uma probabilidade de duelo
(a chance de a corredora correr melhor, ou se sentir melhor, com o tênis "de elite" do que com
o mesmo tênis "falsificado") pela fórmula Φ(d/√2), com Monte Carlo sobre a incerteza. A
preferência recebe uma posterior Beta de Jeffreys. Na curiosidade da cor, as contagens recebem
posteriores Beta de Jeffreys.

**Sem ×0,80.** São medidas de laboratório e escolhas observadas num experimento, não
proporções de pesquisa autorrelatada.

**Cordas declaradas:** #070 (o remédio que você sabe ser falso, placebo aberto), #068 (a
euforia da corrida) e #073 (correr e o joelho). Objetos distintos.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

SEMENTE = 152
rng = np.random.default_rng(SEMENTE)
SORTEIOS = 200_000

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

VERMELHO, DOURADO, VERDE, CINZA, AZUL = '#c0392b', '#c8a84b', '#1a5f5a', '#6b6a64', '#2c5d8f'
RODAPE = ('Fonte: Hébert-Losier et al. 2025, Footwear Science · Hill & Barton 2005, Nature · '
          'Pollet & Peperkoorn 2013 · cálculo próprio | #365Probabilidades')

def vir(x, casas=1):
    return f"{x:.{casas}f}".replace(".", ",")

def jeffreys(k, n):
    return stats.beta(k + 0.5, n - k + 0.5)

print("✅ Bibliotecas carregadas · semente", SEMENTE)
```

```python
# --- DADOS DA LITERATURA ---
N_CORREDORAS = 24

# Hébert-Losier et al. 2025: o mesmo tênis, apresentado como "de elite" ou como "falsificado"
MEDIDAS = [
    # (nome, d, tipo)   tipo: 'corpo' ou 'cabeça'
    ('Consumo de oxigênio', -0.02, 'corpo'),
    ('Gasto de energia', -0.03, 'corpo'),
    ('Conforto', 0.94, 'cabeça'),
    ('Percepções (menor valor)', 0.72, 'cabeça'),
    ('Percepções (maior valor)', 1.16, 'cabeça'),
]
DIFS = {'Consumo de oxigênio (mL/kg/min)': (-0.05, 0.47), 'Gasto de energia (W/kg)': (-0.02, 0.17),
        'Conforto (mm, escala de 100)': (14.6, 15.0)}
PREFEREM_ELITE = 21   # 87,5% de 24

# Curiosidade: o vermelho nas lutas
RODADAS_2004 = (16, 21)        # rodadas com mais vencedores de vermelho
CATEGORIAS_2004 = (19, 29)     # categorias de peso com mais vitórias de vermelho
UFC_2013 = (27, 61)            # lutas vencidas por quem usava calção vermelho

print("=" * 72)
print("  DADOS · O MESMO TÊNIS, DUAS HISTÓRIAS (N = 24 corredoras)")
print("=" * 72)
for nome, (m, dp) in DIFS.items():
    print(f"  {nome:<34} diferença média {vir(m, 2):>6} ± {vir(dp, 2)} (DP)")
for nome, d, tipo in MEDIDAS:
    print(f"  {nome:<34} d = {vir(d, 2):>5}   [{tipo}]")
print(f"  Preferiram o tênis 'de elite': {PREFEREM_ELITE} de {N_CORREDORAS} = {vir(PREFEREM_ELITE / N_CORREDORAS * 100)}%")
print("\n  O VERMELHO NAS LUTAS")
print(f"  Olimpíada 2004 · rodadas com mais vitórias do vermelho: {RODADAS_2004[0]} de {RODADAS_2004[1]}")
print(f"  Olimpíada 2004 · categorias com mais vitórias do vermelho: {CATEGORIAS_2004[0]} de {CATEGORIAS_2004[1]}")
print(f"  UFC 2010-2012 · lutas vencidas de calção vermelho: {UFC_2013[0]} de {UFC_2013[1]}")
print("=" * 72)
```

```python
# --- O MODELO ---
def ic_d(d, n=N_CORREDORAS):
    se = np.sqrt(1 / n + d ** 2 / (2 * n))
    return d - 1.96 * se, d + 1.96 * se, se

def duelo(d, se, sorteios=SORTEIOS):
    amostra = rng.normal(d, se, sorteios)
    p = stats.norm.cdf(amostra / np.sqrt(2))
    return p.mean(), np.percentile(p, [2.5, 97.5])

# CAMADA 1 · O corpo e a cabeça, em tamanho de efeito
print("=" * 72)
print("  CAMADA 1 · O CORPO E A CABEÇA (d com IC 95%)")
print("=" * 72)
RES = {}
for nome, d, tipo in MEDIDAS:
    lo, hi, se = ic_d(d)
    p, (plo, phi) = duelo(d, se)
    RES[nome] = {'d': d, 'lo': lo, 'hi': hi, 'se': se, 'p': p, 'plo': plo, 'phi': phi, 'tipo': tipo}
    print(f"  {nome:<28} d = {vir(d, 2):>5}  IC 95% [{vir(lo, 2)}; {vir(hi, 2)}]")

# CAMADA 2 · Em duelos
print("\n" + "=" * 72)
print("  CAMADA 2 · EM DUELOS (Monte Carlo com " + f"{SORTEIOS:,}".replace(",", ".") + " sorteios)")
print("=" * 72)
print("  Chance de a mesma corredora ir melhor com o tênis 'de elite' do que com o 'falsificado'")
for nome, r in RES.items():
    print(f"  {nome:<28} {vir(r['p'] * 100):>5}%  [{vir(r['plo'] * 100)}; {vir(r['phi'] * 100)}]   (sem efeito = 50%)")
corpo = np.mean([RES[k]['p'] for k in RES if RES[k]['tipo'] == 'corpo'])
print(f"\n  No corpo, cerca de {round(corpo * 100)} em 100. No conforto, cerca de {round(RES['Conforto']['p'] * 100)} em 100.")

# CAMADA 3 · A preferência
pref = jeffreys(PREFEREM_ELITE, N_CORREDORAS)
lo, hi = pref.interval(0.95)
P_MAIORIA_FORTE = pref.sf(0.75)
print("\n" + "=" * 72)
print("  CAMADA 3 · QUAL TÊNIS ELAS PREFERIRAM")
print("=" * 72)
print(f"  Posterior de Jeffreys: {vir(pref.mean() * 100)}%  IC 95% [{vir(lo * 100)}; {vir(hi * 100)}]")
print(f"  P(mais de 3 em cada 4 preferem o tênis 'de elite'): {vir(P_MAIORIA_FORTE * 100)}%")
print("  Lembrando: era o mesmo tênis.")

# CAMADA 4 · A curiosidade do vermelho
print("\n" + "=" * 72)
print("  CAMADA 4 · O VERMELHO NAS LUTAS")
print("=" * 72)
VERM = {}
for nome, (k, n) in (('Olimpíada 2004, rodadas', RODADAS_2004), ('Olimpíada 2004, categorias', CATEGORIAS_2004),
                     ('UFC, lutas', UFC_2013)):
    post = jeffreys(k, n)
    a, b = post.interval(0.95)
    VERM[nome] = (k / n, a, b, post.sf(0.5))
    print(f"  {nome:<28} {k} de {n} = {vir(k / n * 100)}%  IC 95% [{vir(a * 100)}; {vir(b * 100)}]  P(vermelho > 50%) = {vir(post.sf(0.5) * 100)}%")
print("  Em 2004, o vermelho parecia ganhar. Em 2008 e no UFC, o efeito sumiu. A disputa continua aberta.")
print("=" * 72)
```

```python
# --- VISUALIZAÇÃO ---

# GRÁFICO 1 · O corpo e a cabeça
fig1, ax1 = plt.subplots(figsize=(12, 7))
nomes = list(RES)
y = np.arange(len(nomes))[::-1]
for yi, nome in zip(y, nomes):
    r = RES[nome]
    cor = CINZA if r['tipo'] == 'corpo' else DOURADO
    ax1.plot([r['lo'], r['hi']], [yi, yi], color=cor, linewidth=3)
    ax1.plot(r['d'], yi, 'o', color=cor, markersize=11)
    ax1.text(r['hi'] + 0.05, yi, f"d = {vir(r['d'], 2)}", va='center', fontsize=11)
ax1.axvline(0, color='#333', linewidth=1.2)
ax1.axvspan(-0.2, 0.2, color=CINZA, alpha=0.08)
ax1.text(0, -0.75, 'efeito desprezível', ha='center', fontsize=10, color=CINZA)
ax1.set_ylim(-0.95, len(nomes) - 0.5)
ax1.set_yticks(y)
ax1.set_yticklabels(nomes, fontsize=11)
ax1.set_xlim(-1, 2.1)
ax1.set_xlabel('Tamanho do efeito d (tênis "de elite" menos tênis "falsificado") · IC 95%', fontsize=11)
ax1.set_title('O mesmo tênis, duas histórias: o corpo não muda, a sensação muda muito\n'
              '24 corredoras amadoras, ensaio cruzado e sorteado', fontsize=13, pad=14)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=8.5, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-152-grafico-01-corpo-e-cabeca.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# GRÁFICO 2 · Em duelos
fig2, ax2 = plt.subplots(figsize=(12, 7))
sel = ['Consumo de oxigênio', 'Gasto de energia', 'Conforto']
x = np.arange(len(sel))
v = [RES[s]['p'] * 100 for s in sel]
err = [[RES[s]['p'] * 100 - RES[s]['plo'] * 100 for s in sel], [RES[s]['phi'] * 100 - RES[s]['p'] * 100 for s in sel]]
ax2.bar(x, v, color=[CINZA, CINZA, DOURADO], width=0.55)
ax2.errorbar(x, v, yerr=err, fmt='none', ecolor='#333', capsize=6)
for i, val in enumerate(v):
    ax2.text(i, val + err[1][i] + 1.5, f"{round(val)} em 100", ha='center', fontsize=14, fontweight='bold')
ax2.axhline(50, color=VERMELHO, linestyle='--', linewidth=1.8)
ax2.text(0.5, 51.5, 'se o tênis não importasse (50)', color=VERMELHO, fontsize=10, ha='center')
ax2.set_xticks(x)
ax2.set_xticklabels(['gasta menos\noxigênio', 'gasta menos\nenergia', 'sente mais\nconforto'], fontsize=12)
ax2.set_ylim(0, 100)
ax2.set_ylabel('Chance de ir melhor com o tênis "de elite" (em 100)', fontsize=11)
ax2.set_title('Em 100 corridas com o tênis "de elite", quantas vão melhor do que com o "falsificado"\n'
              'Monte Carlo com ' + f'{SORTEIOS:,}'.replace(",", ".") + ' sorteios sobre o tamanho do efeito', fontsize=13, pad=14)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=8.5, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-152-grafico-02-em-duelos.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# GRÁFICO 3 · A preferência
fig3, ax3 = plt.subplots(figsize=(12, 7))
xs = np.linspace(0.4, 1, 600)
ax3.plot(xs * 100, pref.pdf(xs), color=DOURADO, linewidth=3)
ax3.fill_between(xs * 100, pref.pdf(xs), where=(xs >= lo) & (xs <= hi), color=DOURADO, alpha=0.25)
ax3.axvline(PREFEREM_ELITE / N_CORREDORAS * 100, color='#333', linewidth=1.5)
ax3.axvline(50, color=VERMELHO, linestyle='--', linewidth=1.5)
ax3.text(50.5, max(pref.pdf(xs)) * 0.9, 'se fosse indiferente', color=VERMELHO, fontsize=10)
ax3.text(PREFEREM_ELITE / N_CORREDORAS * 100 + 0.8, max(pref.pdf(xs)) * 0.6,
         f"21 de 24 preferiram\no tênis 'de elite'\nIC 95% [{vir(lo * 100, 0)}; {vir(hi * 100, 0)}]", fontsize=11)
ax3.set_xlabel('Proporção que prefere o tênis "de elite" (%)', fontsize=12)
ax3.set_ylabel('Densidade da posterior', fontsize=12)
ax3.set_title('Era o mesmo tênis, e quase todas escolheram o que tinha a história melhor\n'
              'posterior Beta de Jeffreys', fontsize=13, pad=14)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=8.5, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-152-grafico-03-a-preferencia.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")

# GRÁFICO 4 · O vermelho
fig4, ax4 = plt.subplots(figsize=(12, 6))
nomes_v = list(VERM)
y = np.arange(len(nomes_v))[::-1]
for yi, nome in zip(y, nomes_v):
    p, a, b, _ = VERM[nome]
    ax4.plot([a * 100, b * 100], [yi, yi], color=VERMELHO, linewidth=3)
    ax4.plot(p * 100, yi, 'o', color=VERMELHO, markersize=11)
    ax4.text(b * 100 + 1, yi, f"{vir(p * 100, 0)}%", va='center', fontsize=12)
ax4.axvline(50, color='#333', linestyle='--', linewidth=1.5)
ax4.set_yticks(y)
ax4.set_yticklabels(nomes_v, fontsize=11)
ax4.set_xlim(20, 100)
ax4.set_xlabel('Parte vencida por quem estava de vermelho (%) · IC 95%', fontsize=11)
ax4.set_title('A cor do uniforme: o vermelho pareceu ganhar em 2004, mas o efeito não se repetiu\n'
              'Hill & Barton 2005 · Pollet & Peperkoorn 2013', fontsize=13, pad=14)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=8.5, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-152-grafico-04-o-vermelho.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 4 salvo!")
```

### 💡 O Insight

**Com o mesmo tênis apresentado como de elite, o corpo gastou a mesma energia. Em 100
corridas, ele iria melhor em cerca de 49, o mesmo que se o tênis não importasse. Mas a
sensação mudou muito. O conforto foi maior em cerca de 74 de 100, e 21 de 24 corredoras
preferiram o tênis "de elite", que era o mesmo.**

O tênis novo quase não muda o que o corpo faz. Ele muda como o treino é vivido. A corrida fica
mais gostosa, parece mais fácil, e a pessoa sai achando que foi melhor. Isso não é pouca coisa,
porque gostar do treino é uma das razões para voltar a ele.

A cor do uniforme segue o mesmo caminho. Na Olimpíada de 2004, quem lutou de vermelho pareceu
ganhar mais. Nos dados de 2008 e no UFC, o efeito sumiu.

*Qual foi a última coisa que você comprou para treinar, e o que ela mudou de verdade?*

### ⚠️ Limitações do Modelo

- O experimento tem 24 corredoras amadoras, todas mulheres, num só laboratório da Nova Zelândia.
  O resultado pode ser diferente com homens, atletas ou corridas longas.
- Foram medidas a economia de corrida e a biomecânica em trechos de 6 minutos, não o tempo numa
  prova. Um efeito pequeno sobre o tempo de prova não pode ser descartado.
- O artigo traz as percepções como uma faixa de d (0,72 a 1,16), sem o valor de cada uma.
  Mostramos os dois extremos.
- O intervalo de d usa uma aproximação padrão para medidas pareadas, e o duelo supõe
  distribuição normal das diferenças. São leituras do modelo, não sorteios de corredoras reais.
- O tênis usado já é de alta performance. O estudo não compara tênis velho com tênis novo, e
  sim a mesma tecnologia com histórias diferentes.
- Os números consolidados da meta-análise da cognição vestida (Horton, Adam & Galinsky 2025)
  não foram conferidos no texto completo e por isso entram só como conclusão.
- A curiosidade do vermelho é uma disputa aberta. O artigo de 2016 que contesta o efeito é um
  preprint, e os autores originais responderam defendendo o achado em lutas equilibradas.
- Sem ×0,80, porque são medidas de laboratório e escolhas observadas.

*A ciência é honesta sobre o que não sabe. O modelo também.*
