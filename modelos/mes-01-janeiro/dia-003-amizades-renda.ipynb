### 📖 A História

Quando meu filho nasceu, comecei a tomar decisões que nunca imaginei
que fossem financeiras.

Qual escola escolher. Em qual bairro morar. Quais atividades oferecer.
Quais círculos sociais cultivar.

Achava que estava escolhendo valores, segurança, oportunidades de
aprendizado. Não sabia que estava escolhendo, também, a renda futura
do meu filho.

A ciência já mediu isso. E o número é inequívoco.

As amizades que uma criança faz — o nível econômico das famílias com
quem convive — é o preditor mais forte de mobilidade de renda já
identificado pela ciência. Mais forte do que a escola. Mais forte do
que o emprego dos pais. Mais forte do que o bairro.

Toda mãe toma essa decisão. Quase nenhuma sabe que está tomando.

---

### 📚 O Conceito: Conectividade Econômica

Chetty definiu um indicador chamado **Economic Connectedness** —
a proporção de amigos de alta renda que uma criança de baixa renda possui.

A descoberta: essa métrica é o **preditor mais forte de mobilidade
econômica ascendente** já identificado pela ciência.

Mais forte do que a qualidade da escola.
Mais forte do que o emprego dos pais.
Mais forte do que a localização geográfica.

Quando você escolhe onde seu filho vai estudar, brincar e crescer —
você está escolhendo, também, quem serão os amigos dele.

E isso tem um valor financeiro mensurável.

```python
# 365 Probabilidades — Dia #003
# Qual a probabilidade de suas amizades estarem determinando sua renda?
# Fonte: Chetty et al., 2022 — Nature (N=21 bilhões de amizadas | 72 milhões de usuários)

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Configuração visual padrão do projeto
plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Chetty et al., 2022 — Nature
# N = 21 bilhões de amizades | 72 milhões de usuários Facebook EUA

# Resultado principal
ganho_renda_amizades_ricas = 0.20  # +20% de renda na vida adulta
# para pessoas de baixa renda que crescem com amigos de alta renda

# Segregação econômica nas redes sociais
# Redes são altamente segregadas por renda
p_amigos_mesma_faixa = 0.77  # 77% dos amigos são da mesma faixa de renda
p_amigos_faixa_diferente = 0.23  # apenas 23% cruzam faixas de renda

# Economic Connectedness (EC) — métrica de Chetty
# EC = proporção de amigos de alta renda entre pessoas de baixa renda
# Média nacional EUA: 0.78 (abaixo de 1.0 = segregado)
ec_nacional = 0.78
ec_ideal = 1.0  # sem segregação — metade dos amigos de alta renda

# Impacto por grau de conectividade econômica
# Baseado nos dados do Social Capital Atlas
impacto_ec = {
    'EC muito baixo (< 0.5)':   -0.15,  # -15% de renda potencial
    'EC baixo (0.5–0.75)':      -0.08,  # -8% de renda potencial
    'EC médio (0.75–1.0)':       0.00,  # referência
    'EC alto (1.0–1.25)':       +0.10,  # +10% de renda potencial
    'EC muito alto (> 1.25)':   +0.20   # +20% de renda potencial
}

# Fator de correção conservador
fator_correcao = 0.80

print("=" * 65)
print("  DADOS DA LITERATURA — AMIZADES E RENDA")
print("  Chetty et al., 2022 — Nature (N=21 bilhões de amizades)")
print("=" * 65)
print(f"\n  Resultado principal:")
print(f"  → Amigos de alta renda aumentam renda adulta em: +{ganho_renda_amizades_ricas*100:.0f}%")
print(f"\n  Segregação nas redes sociais:")
print(f"  → {p_amigos_mesma_faixa*100:.0f}% dos amigos são da mesma faixa de renda")
print(f"  → Apenas {p_amigos_faixa_diferente*100:.0f}% cruzam faixas econômicas")
print(f"\n  Economic Connectedness (EC) nacional EUA:")
print(f"  → Média: {ec_nacional} (ideal = {ec_ideal})")
print(f"\n  Impacto por nível de EC:")
for nivel, impacto in impacto_ec.items():
    sinal = "+" if impacto >= 0 else ""
    print(f"  → {nivel:<30} {sinal}{impacto*100:.0f}% de renda")
print("=" * 65)
```

```python
# --- O MODELO ---
# Probabilidade de suas amizades estarem limitando sua renda

# Passo 1 — Probabilidade de ter rede segregada economicamente
# 77% dos amigos são da mesma faixa — logo a maioria das pessoas
# está em redes homogêneas
p_rede_segregada = p_amigos_mesma_faixa  # 77%

# Passo 2 — Dado rede segregada, probabilidade de estar abaixo da média de EC
# EC nacional = 0.78 — abaixo de 1.0 significa sub-representação de amigos ricos
p_abaixo_ec_ideal = 1 - ec_nacional  # 22% abaixo do ideal
p_rede_limitante = p_rede_segregada * (1 - ec_nacional / ec_ideal)

# Passo 3 — Impacto financeiro médio de uma rede segregada
# Pessoas com EC baixo perdem entre 8% e 15% de renda potencial
perda_renda_media = abs(impacto_ec['EC baixo (0.5–0.75)'])  # 8%
perda_renda_max = abs(impacto_ec['EC muito baixo (< 0.5)'])  # 15%

# Passo 4 — Ganho potencial de expandir a rede
ganho_potencial = impacto_ec['EC muito alto (> 1.25)']  # +20%

# Corrigindo
p_rede_limitante_corrigido = p_rede_limitante * fator_correcao
p_rede_segregada_corrigida = p_rede_segregada * fator_correcao

print("=" * 65)
print("  MODELO — AMIZADES E PROBABILIDADE DE LIMITAÇÃO DE RENDA")
print("=" * 65)
print(f"\n  P(rede segregada economicamente):     {p_rede_segregada*100:.0f}%")
print(f"  P(EC abaixo do ideal):                {p_abaixo_ec_ideal*100:.0f}%")
print(f"  P(rede limitando renda):              {p_rede_limitante*100:.0f}%")
print(f"\n  Com fator de correção 0.80:")
print(f"  → P(rede segregada):                  {p_rede_segregada_corrigida*100:.0f}%")
print(f"  → P(rede limitando renda):            {p_rede_limitante_corrigido*100:.0f}%")
print(f"\n  Impacto financeiro estimado:")
print(f"  → Perda média com EC baixo:           -{perda_renda_media*100:.0f}% de renda")
print(f"  → Perda máxima com EC muito baixo:    -{perda_renda_max*100:.0f}% de renda")
print(f"  → Ganho potencial com EC muito alto:  +{ganho_potencial*100:.0f}% de renda")
print(f"\n  CONCLUSÃO:")
print(f"  → {p_rede_segregada_corrigida*100:.0f}% vivem em redes economicamente segregadas")
print(f"  → Para quem tem baixa renda, isso custa até -{perda_renda_max*100:.0f}% de renda potencial")
print(f"  → Expandir a rede para EC alto vale +{ganho_potencial*100:.0f}% de renda na vida adulta")
print(f"\n  FATO IRREFUTÁVEL (Chetty et al., 2022 — Nature):")
print(f"  → N = 21 bilhões de amizades | 72 milhões de usuários")
print(f"  → Conectividade econômica é o preditor mais forte")
print(f"     de mobilidade de renda já identificado pela ciência")
print(f"  → Mais forte do que qualidade das escolas,")
print(f"     emprego dos pais ou localização geográfica")
print("=" * 65)
```

```python
# --- CAMADA 2 — MODELO ESTATÍSTICO ---
# Distribuição Beta para modelar incerteza nas proporções
# Chetty et al., 2022 — N efetivo conservador
# O N real é 72 milhões — usamos 100.000 como N efetivo conservador

n_chetty = 100000

# Distribuição Beta para segregação
dist_segregada = stats.beta(
    p_amigos_mesma_faixa * n_chetty,
    (1 - p_amigos_mesma_faixa) * n_chetty
)

# Distribuição Beta para EC abaixo do ideal
dist_ec_baixo = stats.beta(
    p_abaixo_ec_ideal * n_chetty,
    (1 - p_abaixo_ec_ideal) * n_chetty
)

# Intervalos de confiança 95%
ic_segregada = dist_segregada.interval(0.95)
ic_ec_baixo = dist_ec_baixo.interval(0.95)

# Ganho de renda com IC — baseado no resultado principal de Chetty
# +20% com N=72 milhões — intervalo é extremamente estreito
# Usamos estimativa de erro padrão de 2pp (reportado no paper)
ganho_medio = 0.20
erro_padrao = 0.02
ic_ganho = (ganho_medio - 1.96 * erro_padrao,
            ganho_medio + 1.96 * erro_padrao)

print("=" * 65)
print("  MODELO BAYESIANO — INTERVALO DE CONFIANÇA 95%")
print("  Chetty et al., 2022 — Nature")
print("=" * 65)
print(f"\n  Segregação econômica nas redes:")
print(f"  → Estimativa central: {p_amigos_mesma_faixa*100:.0f}%")
print(f"  → IC 95%: [{ic_segregada[0]*100:.2f}%, {ic_segregada[1]*100:.2f}%]")
print(f"\n  EC abaixo do ideal:")
print(f"  → Estimativa central: {p_abaixo_ec_ideal*100:.0f}%")
print(f"  → IC 95%: [{ic_ec_baixo[0]*100:.2f}%, {ic_ec_baixo[1]*100:.2f}%]")
print(f"\n  Ganho de renda com alta conectividade econômica:")
print(f"  → Estimativa central: +{ganho_medio*100:.0f}%")
print(f"  → IC 95%: [+{ic_ganho[0]*100:.1f}%, +{ic_ganho[1]*100:.1f}%]")
print(f"\n  CONCLUSÃO COM INCERTEZA:")
print(f"  → Entre {ic_segregada[0]*fator_correcao*100:.0f}% e {ic_segregada[1]*fator_correcao*100:.0f}%")
print(f"     vivem em redes economicamente segregadas")
print(f"  → Ganho de renda garantido com alta EC:")
print(f"     entre +{ic_ganho[0]*100:.1f}% e +{ic_ganho[1]*100:.1f}%")
print(f"  → Com 95% de confiança")
print("=" * 65)
```

### 💡 O Insight

**77% das pessoas vivem em redes economicamente segregadas.**

Não por mal — por gravidade social. Nos aproximamos de quem
é parecido com a gente. É natural. É humano.

Mas tem um custo financeiro mensurável.

Para quem tem baixa renda e rede segregada, esse custo é
entre **8% e 15% de renda potencial** ao longo da vida.

E o ganho de reverter isso — com 95% de confiança —
está entre **+16.1% e +23.9% de renda na vida adulta.**

O estudo de Chetty é o maior já conduzido sobre redes sociais:
21 bilhões de amizades. 72 milhões de pessoas.
Publicado na Nature em 2022.

E a conclusão é inequívoca:

Conectividade econômica é o preditor mais forte de
mobilidade de renda já identificado pela ciência.

Mais forte do que a escola que você frequentou.
Mais forte do que o emprego dos seus pais.
Mais forte do que onde você cresceu.
