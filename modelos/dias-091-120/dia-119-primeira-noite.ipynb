### 📖 A História

Todo mundo já dormiu mal na primeira noite de hotel.

O colchão é bom, o quarto é silencioso, o cansaço é grande. E mesmo assim o sono vem
mais tarde, mais leve, e qualquer barulho no corredor acorda. Na segunda noite, some.

Os laboratórios de sono sabem disso há sessenta anos. Chamam de efeito da primeira
noite, e a regra é descartar a primeira noite de qualquer estudo, porque ela não vale.
O que ninguém sabia era o mecanismo.

Em 2016, um grupo da Brown University pôs pessoas para dormir com neuroimagem e achou:
não é o cérebro inteiro que dorme pior. É uma metade.

---

### 📚 O Conceito: sono com um hemisfério de sentinela

Golfinhos e algumas aves dormem com um hemisfério de cada vez: metade do cérebro
descansa, a outra vigia. Humanos não fazem isso, ou não faziam até alguém medir a
primeira noite num lugar novo.

O que o estudo mostra é uma versão discreta da mesma coisa: na primeira noite em
ambiente desconhecido, o hemisfério esquerdo dorme menos profundamente que o direito,
reage mais a sons estranhos e acorda a pessoa mais rápido. Na segunda noite, a
assimetria desaparece. Os autores chamam de vigia noturna, e leem como mecanismo de
proteção, não como defeito.

---

### 🧮 O Modelo

Quatro camadas.

1. **A assimetria, em gente.** O tamanho de efeito publicado (d = 0,8) traduzido em
   duelos: em quantas pessoas de cada cem a metade esquerda dorme menos que a direita
   na primeira noite, com o intervalo que n = 11 impõe. E a correlação com a demora
   para dormir, com IC por transformação z de Fisher.
2. **A vigia.** Mais de 80% dos despertares da primeira noite vieram de sons
   apresentados ao hemisfério esquerdo. Monte Carlo de uma noite com sons estranhos
   sob duas hipóteses (vigia contra acaso), semente declarada.
3. **O poder do desenho.** Análise de poder do experimento principal: quanta chance um
   estudo com 11 pessoas tinha de achar o próprio efeito, e qual o menor efeito que
   ele enxergaria.
4. **Quanto custa, para quem viaja.** Os minutos perdidos na primeira noite (N = 45,
   2024) somados ao longo de um ano de noites fora de casa, com a incerteza publicada.

**Fontes:**
- Tamaki, M., Bang, J. W., Watanabe, T. & Sasaki, Y. (2016). "Night Watch in One Brain
  Hemisphere during Sleep Associated with the First-Night Effect in Humans". *Current
  Biology* 26(9), 1190-1194. **35 jovens saudáveis** em três experimentos, duas sessões
  de sono com uma semana de intervalo. Experimento 1 (n = 11, MEG + polissonografia):
  atividade de ondas lentas da rede de modo padrão no hemisfério esquerdo menor que no
  direito na noite 1 (t₁₀ = 2,59, p = 0,027, **d = 0,8**) e menor que no próprio
  esquerdo na noite 2 (t₁₀ = 2,69, p = 0,023, d = 0,8); direito sem diferença entre
  noites (p = 0,355); assimetria correlacionada com a latência de sono na noite 1
  (**r = −0,68**, p = 0,022) e não na noite 2 (r = 0,03). Experimento 2 (n = 13,
  sons desviantes monaurais): resposta cerebral maior no esquerdo na noite 1 (d = 0,9);
  **mais de 80% dos despertares** seguiram sons ao hemisfério esquerdo (Wilcoxon
  z₁₂ = 3,22, p = 0,001), dominância que some na noite 2 (p = 0,751). Experimento 3
  (n = 11, bater o dedo ao ouvir o som): mais pessoas acordadas pelo lado esquerdo na
  noite 1 (McNemar p = 0,033) e reação mais rápida (z₁₀ = 2,31, p = 0,021). Lido no PDF.
- Wick, A. Z., Combertaldi, S. L. & Rasch, B. (2024). "The first-night effect of sleep
  occurs over nonconsecutive nights in unfamiliar and familiar environments". *Sleep*
  47(10), zsae179. Estudo 1: **N = 45** (18 a 31 anos), duas noites de laboratório com
  uma semana de intervalo: latência **20,4 ± 2,1 contra 14,0 ± 2,2 min** (F₁,₄₄ = 8,33,
  p = 0,006); sono total **438,1 ± 5,1 contra 450,7 ± 4,1 min** (F₁,₄₃ = 4,26,
  p = 0,045); valores são média ± EP. Estudo 2 (N = 30): o efeito aparece também em
  casa, mais fraco.
- Ding, L., Chen, B., Dai, Y. & Li, Y. (2022). "A meta-analysis of the first-night
  effect in healthy individuals for the full age spectrum". *Sleep Medicine* 89,
  159-165. **53 estudos, N = 1.422**, 9 a 85 anos: latência, despertares e latência de
  REM maiores; sono total, eficiência e REM menores; mais N1 (todos p ≤ 0,001); sono
  profundo sem diferença; efeito menor nos 20 a 29 anos. **Só o resumo foi acessível;
  as magnitudes agrupadas não foram consultadas e não entram no modelo.**
- Sprajcer, M., Gupta, C., Roach, G. & Sargent, C. (2022). "Can we put the first night

  despertares ou eficiência; **mais N1 e menos N2 e N3** na noite 1. Entra como
  contexto: em laboratório moderno a quantidade de sono pode não mudar, e a
  profundidade muda.

  study of sleep". *Psychophysiology* 2(3), 263-266. Origem do termo; sem número usado.

**Nota metodológica:** o fator ×0,80 não se aplica. Polissonografia, neuroimagem e

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

SEMENTE = 42
SORTEIOS = 200_000

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

VERMELHO, DOURADO, VERDE, CINZA = '#c0392b', '#c8a84b', '#1a5f5a', '#6b6a64'

def mil(n):
    return f"{int(n):,}".replace(",", ".")

def vir(x, casas=1):
    return f"{x:.{casas}f}".replace(".", ",")

print("✅ Bibliotecas carregadas · semente", SEMENTE)
```

```python
# --- DADOS DA LITERATURA ---

# Tamaki et al. (2016), Current Biology 26(9), 1190-1194 — lido no PDF
N_EXP1, N_EXP2, N_EXP3 = 11, 13, 11
D_ASSIMETRIA = 0.8            # esquerdo < direito na noite 1 (t10 = 2,59)
T_ASSIMETRIA = 2.59
R_LATENCIA = -0.68            # assimetria × latência de sono, noite 1 (p = 0,022)
P_DESPERTAR_ESQ = 0.80        # "mais de 80%" dos despertares vieram do hemisfério esquerdo
Z_DESPERTAR = 3.22            # Wilcoxon, p = 0,001

# Wick, Combertaldi & Rasch (2024), Sleep 47(10) — Estudo 1, média ± EP
N_WICK = 45
LAT_N1, LAT_N2, EP_LAT1, EP_LAT2 = 20.37, 14.01, 2.05, 2.19      # minutos
F_LAT = 8.33
TST_N1, TST_N2 = 438.05, 450.65                                    # minutos
F_TST = 4.26
N_TST = 44                                                         # F(1,43)

# Sprajcer et al. (2022), Chronobiology International 39(12) — contexto
N_SPRAJCER = 124

# Ding et al. (2022), Sleep Medicine 89 — só direção (resumo)
K_DING, N_DING = 53, 1_422

aplica_fator_080 = False      # polissonografia e neuroimagem

print("=" * 70)
print("  DADOS DA LITERATURA")
print("=" * 70)
print(f"\n  TAMAKI 2016 · {N_EXP1 + N_EXP2 + N_EXP3} jovens saudáveis em três experimentos")
print(f"  → noite 1: hemisfério esquerdo dorme menos profundo que o direito, d = {vir(D_ASSIMETRIA)} (n = {N_EXP1})")
print(f"  → quanto maior a assimetria, mais demora para dormir: r = {vir(R_LATENCIA, 2)} (só na noite 1)")
print(f"  → mais de {P_DESPERTAR_ESQ*100:.0f}% dos despertares vieram do lado esquerdo (n = {N_EXP2}); na noite 2, some")
print(f"\n  WICK 2024 · N = {N_WICK} · duas noites de laboratório com uma semana de intervalo")
print(f"  → para adormecer: {vir(LAT_N1)} contra {vir(LAT_N2)} min (+{vir(LAT_N1-LAT_N2)} min, p = 0,006)")
print(f"  → sono total: {vir(TST_N1)} contra {vir(TST_N2)} min ({vir(TST_N1-TST_N2)} min, p = 0,045)")
print(f"\n  DING 2022 · meta-análise, {K_DING} estudos, N = {mil(N_DING)}: direção confirmada, magnitude não consultada")
print(f"  SPRAJCER 2022 · N = {N_SPRAJCER}: quantidade de sono igual, sono mais leve na noite 1 (contexto)")
print(f"\n  ×0,80: não se aplica. Medida de laboratório.")
print("=" * 70)
```

```python
# --- O MODELO ---

# ── CAMADA 1 · a assimetria, em gente ──
# CLES para diferença pareada: P(esquerdo < direito numa pessoa) = Φ(d/√2).
# IC do d pareado com n = 11: EP(d) ≈ √(1/n + d²/(2n)).
ep_d = np.sqrt(1 / N_EXP1 + D_ASSIMETRIA ** 2 / (2 * N_EXP1))
d_lo, d_hi = D_ASSIMETRIA - 1.96 * ep_d, D_ASSIMETRIA + 1.96 * ep_d
cles = lambda d: stats.norm.cdf(d / np.sqrt(2))
duelos, duelos_lo, duelos_hi = cles(D_ASSIMETRIA), cles(d_lo), cles(d_hi)
# IC da correlação por z de Fisher (n = 11)
zr = np.arctanh(R_LATENCIA); ep_z = 1 / np.sqrt(N_EXP1 - 3)
r_lo, r_hi = np.tanh(zr - 1.96 * ep_z), np.tanh(zr + 1.96 * ep_z)

print("=" * 70)
print("  CAMADA 1 · A ASSIMETRIA, EM GENTE")
print("=" * 70)
print(f"\n  d = {vir(D_ASSIMETRIA)} com n = {N_EXP1}: IC 95% [{vir(d_lo, 2)}; {vir(d_hi, 2)}]")
print(f"  Duelos: em {vir(duelos*100, 0)} de cada 100 pessoas a metade esquerda dorme menos que a direita")
print(f"  na primeira noite. IC 95% [{vir(duelos_lo*100, 0)}; {vir(duelos_hi*100, 0)}]. Com 11 pessoas, o intervalo é largo.")
print(f"\n  Assimetria × demora para dormir: r = {vir(R_LATENCIA, 2)}  IC 95% [{vir(r_lo, 2)}; {vir(r_hi, 2)}]")
print(f"  Variância compartilhada: {vir(R_LATENCIA**2*100, 0)}%. Na noite 2, r = 0,03: a ligação some.")

# ── CAMADA 2 · a vigia ──
# Uma noite com k sons estranhos durante o sono profundo. Sob a hipótese de vigia,
# cada despertar vem do lado esquerdo com p = 0,80; sob o acaso, 0,50.
g = np.random.default_rng(SEMENTE)
K_SONS = 10
esq_vigia = g.binomial(K_SONS, P_DESPERTAR_ESQ, SORTEIOS)
esq_acaso = g.binomial(K_SONS, 0.5, SORTEIOS)
# Quão raro seria o 80% se fosse acaso? P(≥ 8 de 10 pelo mesmo lado | 0,5)
p_acaso_8 = stats.binom.sf(7, K_SONS, 0.5)

print("\n" + "=" * 70)
print("  CAMADA 2 · A VIGIA")
print("=" * 70)
print(f"\n  Mais de {P_DESPERTAR_ESQ*100:.0f}% dos despertares da noite 1 vieram do hemisfério esquerdo")
print(f"  (Wilcoxon z = {vir(Z_DESPERTAR, 2)}, p = 0,001, n = {N_EXP2}). Na noite 2, {vir(50)}% a {vir(50)}%.")
print(f"\n  Noite simulada com {K_SONS} sons estranhos ({mil(SORTEIOS)} noites):")
print(f"  → com vigia (p = 0,80): mediana de {np.median(esq_vigia):.0f} despertares pelo esquerdo,"
      f" P(≥ 8 de 10) = {vir((esq_vigia>=8).mean()*100)}%")
print(f"  → no acaso  (p = 0,50): mediana de {np.median(esq_acaso):.0f},"
      f" P(≥ 8 de 10) = {vir((esq_acaso>=8).mean()*100)}%  (exato: {vir(p_acaso_8*100)}%)")
print(f"  Se os dois lados vigiassem igual, 8 despertares em 10 pelo mesmo lado aconteceriam")
print(f"  em {vir(p_acaso_8*100)}% das noites. Aconteceu na primeira noite e sumiu na segunda.")

# ── CAMADA 3 · o poder do desenho ──
# Teste t pareado, n = 11, alfa = 0,05 bilateral.
def poder_pareado(d, n, alpha=0.05):
    df = n - 1
    tcrit = stats.t.ppf(1 - alpha / 2, df)
    nc = d * np.sqrt(n)
    p = stats.nct.sf(tcrit, df, nc) + stats.nct.cdf(-tcrit, df, nc)
    return float(np.nan_to_num(p, nan=1.0))
poder_obs = poder_pareado(D_ASSIMETRIA, N_EXP1)
from scipy import optimize
d_mde = optimize.brentq(lambda d: poder_pareado(d, N_EXP1) - 0.80, 0.05, 2.5)
n_para_80 = next(n for n in range(4, 200) if poder_pareado(D_ASSIMETRIA, n) >= 0.80)

print("\n" + "=" * 70)
print("  CAMADA 3 · O PODER DO DESENHO (n = 11, teste t pareado)")
print("=" * 70)
print(f"\n  Poder para achar d = {vir(D_ASSIMETRIA)} com 11 pessoas: {vir(poder_obs*100, 0)}%")
print(f"  Menor efeito detectável com 80% de poder:  d = {vir(d_mde, 2)}")
print(f"  Pessoas necessárias para 80% de poder em d = {vir(D_ASSIMETRIA)}: {n_para_80}")
print(f"\n  O estudo tinha cerca de dois terços de chance de achar o próprio efeito. Achou, e")
print(f"  os experimentos 2 e 3 (n = 13 e 11) apontaram na mesma direção com medidas")
print(f"  diferentes. É convergência, não replicação independente. Fica declarado.")

# ── CAMADA 4 · quanto custa, para quem viaja ──
# Diferenças do Wick 2024 com incerteza: EP da diferença pareada derivado do F.
t_lat = np.sqrt(F_LAT); ep_dif_lat = (LAT_N1 - LAT_N2) / t_lat
t_tst = np.sqrt(F_TST); ep_dif_tst = abs(TST_N1 - TST_N2) / t_tst
d_lat, d_tst = t_lat / np.sqrt(N_WICK), t_tst / np.sqrt(N_TST)
g2 = np.random.default_rng(SEMENTE + 1)
print("\n" + "=" * 70)
print("  CAMADA 4 · QUANTO CUSTA, PARA QUEM VIAJA (Wick 2024, N = 45)")
print("=" * 70)
print(f"\n  Primeira noite: +{vir(LAT_N1-LAT_N2)} min para adormecer (EP {vir(ep_dif_lat)}, d = {vir(d_lat, 2)})")
print(f"                  {vir(TST_N1-TST_N2)} min de sono (EP {vir(ep_dif_tst)}, d = {vir(d_tst, 2)})")
print(f"\n  {'noites fora por ano':>20} | {'sono perdido/ano':>18} | {'IC 95%':>16}")
print("  " + "-" * 62)
custo = {}
for noites in [6, 12, 30, 60]:
    dif = g2.normal(TST_N1 - TST_N2, ep_dif_tst, SORTEIOS)      # min por primeira noite
    horas = -dif * noites / 60
    lo, hi = np.percentile(horas, [2.5, 97.5])
    custo[noites] = (horas.mean(), lo, hi)
    print(f"  {noites:>20} | {vir(horas.mean()):>15} h | [{vir(lo)}; {vir(hi)}]")
print("  " + "-" * 62)
print(f"  É pouco por noite e não zera com a viagem. Numa amostra maior e mais recente")
print(f"  (Sprajcer 2022, N = 124) a quantidade não mudou; o que mudou foi a profundidade.")
print(f"  O custo é aproximado; o mecanismo é o achado.")
print("=" * 70)
```

### 💡 O Insight

**Na primeira noite num lugar novo, metade do seu cérebro fica de vigia. Em 71 de
cada 100 pessoas, é a esquerda.**

O hemisfério esquerdo dorme menos profundamente que o direito na primeira noite
(d = 0,8, IC [0,12; 1,48] com 11 pessoas, intervalo largo e declarado), e quanto maior
a assimetria, mais a pessoa demora para dormir (r = −0,68). Na segunda noite, a
assimetria some e a correlação vai a zero.

É essa metade que acorda você. **Mais de 80% dos despertares** da primeira noite vieram
de sons apresentados ao hemisfério esquerdo. Se os dois lados vigiassem igual, oito
despertares em dez pelo mesmo lado aconteceriam em 5,5% das noites. Aconteceu na
primeira, sumiu na segunda. E quem acorda pelo lado de vigia acorda mais rápido.

O estudo é pequeno: 35 pessoas em três experimentos, e o principal tinha cerca de
67% de chance de achar o próprio efeito. Achou, e os outros dois apontaram na mesma
direção com medidas diferentes. Convergência, não replicação independente.

Quanto custa? Numa amostra de 45 pessoas, a primeira noite tem 6 minutos a mais para
adormecer e 13 minutos a menos de sono. Para quem passa 30 noites por ano fora de casa,
são cerca de 6 horas de sono por ano, só nas primeiras noites. Pouco por noite,
e não zera com a prática.

Os autores leem a vigia como proteção, não como defeito: em ambiente desconhecido, uma
metade do cérebro fica de sentinela, como fazem golfinhos e aves, numa versão discreta.
A noite mal dormida no hotel não é você desaprendendo a dormir. É o seu cérebro fazendo
plantão.

*Você já chegou um dia antes de algo importante só para dormir a primeira noite ruim
antes da que conta?*

### ⚠️ Limitações do Modelo

- **N pequeno no mecanismo.** Tamaki 2016 tem 11, 13 e 11 pessoas. O d = 0,8 vem com
  IC de 0,12 a 1,48, e o poder do experimento principal era de 67%. Os três
  experimentos convergem, mas são do mesmo grupo e do mesmo laboratório; não há
  replicação independente conhecida da assimetria.
- **Os 80% são "mais de 80%", média entre pessoas.** As contagens brutas de despertares
  por pessoa estão no material suplementar e não foram consultadas; a simulação da
  camada 2 usa 0,80 como taxa e serve para dar escala ao acaso, não para reestimar.
- **A magnitude do custo vem de N = 45** (Wick 2024) e é conta de bolso: minutos por
  primeira noite vezes noites por ano, sem adaptação ao longo do ano. A meta-análise
  (53 estudos) confirma a direção, mas suas magnitudes agrupadas não foram consultadas.
  Uma amostra maior e mais recente (Sprajcer 2022, N = 124) não achou diferença na
  quantidade de sono, só na profundidade: o custo em horas pode ser menor do que a
  camada 4 mostra; o mecanismo, não.
- **Jovens saudáveis, em laboratório.** Nenhum estudo aqui mede hotel, casa de amigo ou
  viagem real; o "lugar novo" é um laboratório de sono, com eletrodos. O efeito aparece
  também em casa (Wick, Estudo 2), mais fraco.
- **A leitura de "proteção" é dos autores.** O dado mostra assimetria, despertares e
  reação mais rápida; a função evolutiva é interpretação, e o texto a atribui a eles.

*A ciência é honesta sobre o que não sabe. O modelo também.*
