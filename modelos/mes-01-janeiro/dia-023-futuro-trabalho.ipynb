### 📖 A História

Trabalhei quase duas décadas no mercado financeiro. Vi planilhas substituírem relatórios manuais, algoritmos substituírem traders, e modelos de machine learning substituírem analistas de crédito que levavam semanas pra fazer o que uma máquina faz em minutos.

Nunca pensei que fosse acontecer comigo. Até que começou a acontecer ao meu redor.

A manchete que mais circulou nos últimos anos é simples e assustadora: quase metade dos empregos está em risco de ser automatizada. É o tipo de número que paralisa. Que faz você se perguntar se vale a pena investir numa carreira que talvez não exista em 10 anos.

Mas esse número tem uma história por trás — e ela é mais complicada, mais honesta, e mais útil do que a manchete sugere.

---

### 📚 O Conceito: Automação de Ocupações vs. Automação de Tarefas

Em 2013, dois pesquisadores de Oxford — Carl Benedikt Frey e Michael Osborne — usaram um classificador de machine learning para avaliar 702 ocupações nos EUA e estimaram que **47% dos empregos** estavam em alto risco de automação.

O número virou manchete no mundo inteiro. E moldou o debate sobre o futuro do trabalho por uma década.

Mas em 2016, uma equipe da OECD refez a análise com uma mudança metodológica crucial: em vez de classificar **ocupações inteiras** como automatizáveis ou não, eles olharam para as **tarefas** dentro de cada ocupação. Um contador não faz só cálculos — ele também conversa com clientes, interpreta contextos, negocia. Um médico não faz só diagnósticos — ele acolhe, explica, decide em cenários de incerteza.

Resultado: quando você olha por tarefas, o número cai de 47% para **9%**.

A diferença entre esses dois números não é um erro — é uma pergunta diferente. "Quantas ocupações têm tarefas automatizáveis?" (muitas) é diferente de "Quantas ocupações podem ser inteiramente substituídas?" (poucas).

Em 2023, o Goldman Sachs adicionou a IA generativa à equação e estimou que **300 milhões de empregos** poderiam ser "expostos" à automação. Em 2025, o mesmo time publicou um follow-up: o impacto real no mercado de trabalho era, até aquele momento, **negligível**.

### 🧮 O Modelo
**Fontes:**
- Frey, C.B. & Osborne, M.A. (2017) — *The Future of Employment: How Susceptible Are Jobs to Computerisation?* — Technological Forecasting and Social Change, 114, 254-280 — 702 ocupações analisadas
- Arntz, M., Gregory, T. & Zierahn, U. (2016) — *The Risk of Automation for Jobs in OECD Countries* — OECD Social, Employment and Migration Working Papers, No. 189 — abordagem por tarefas, 21 países
- Briggs, J. & Kodnani, D. (2023) — *The Potentially Large Effects of Artificial Intelligence on Economic Growth* — Goldman Sachs Global Economics Analyst, 26 March 2023 — 900+ ocupações analisadas
- Briggs, J. & Dong, S. (2025) — Goldman Sachs follow-up — impacto no mercado de trabalho ainda negligível


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print('✅ Bibliotecas carregadas')
```

```python
# --- DADOS DA LITERATURA ---

# Frey & Osborne (2017) — Oxford
# 702 ocupacoes classificadas via Gaussian Process Classifier
n_ocupacoes_fo = 702
p_alto_risco_fo = 0.47  # 47% em alto risco (>70% de probabilidade de automacao)

# Arntz, Gregory & Zierahn (2016) — OECD
# Mesmos dados, abordagem por TAREFAS em vez de ocupacoes inteiras
# 21 paises da OECD
p_alto_risco_oecd = 0.09  # 9% em alto risco quando olha por tarefas
n_paises_oecd = 21

# Goldman Sachs (marco 2023) — Briggs & Kodnani
# Analise de 900+ ocupacoes, foco em IA generativa
n_ocupacoes_gs = 900
empregos_expostos_milhoes = 300  # 300 milhoes de empregos expostos globalmente
p_expostos_algum_grau = 0.67    # 2/3 das ocupacoes nos EUA expostas a algum grau
p_substituivel_trabalho = 0.25  # ate 1/4 do trabalho poderia ser substituido
aumento_pib_global_pct = 7      # potencial aumento do PIB global

# Goldman Sachs (marco 2025) — Briggs & Dong — follow-up
# "Aggregate labor market impacts are still negligible"
# Impacto real esperado a partir de 2027

fator_correcao = 0.80

print('=' * 65)
print('  DADOS DA LITERATURA — AUTOMACAO E O FUTURO DO TRABALHO')
print('=' * 65)
print(f'\n  Frey & Osborne (2017) — Oxford:')
print(f'  → {n_ocupacoes_fo} ocupacoes analisadas')
print(f'  → {p_alto_risco_fo*100:.0f}% em alto risco de automacao (abordagem por ocupacao)')
print(f'\n  Arntz, Gregory & Zierahn (2016) — OECD:')
print(f'  → Mesmos dados, abordagem por TAREFAS')
print(f'  → {p_alto_risco_oecd*100:.0f}% em alto risco (abordagem por tarefa)')
print(f'  → {n_paises_oecd} paises analisados')
print(f'\n  Goldman Sachs (2023) — IA generativa:')
print(f'  → {empregos_expostos_milhoes} milhoes de empregos expostos globalmente')
print(f'  → {p_expostos_algum_grau*100:.0f}% das ocupacoes expostas a algum grau')
print(f'  → Ate {p_substituivel_trabalho*100:.0f}% do trabalho poderia ser substituido')
print(f'\n  Goldman Sachs (2025) — follow-up:')
print(f'  → Impacto real no mercado de trabalho: negligivel ate agora')
print(f'  → Impacto esperado a partir de 2027')
print(f'\n  Fator de correcao: x{fator_correcao}')
print('=' * 65)
print()
print('  ⚠️  NOTA METODOLOGICA:')
print('  A diferenca entre 47% e 9% nao e um erro — e uma pergunta')
print('  diferente. Frey & Osborne perguntaram "quantas ocupacoes tem')
print('  tarefas automatizaveis?" A OECD perguntou "quantas ocupacoes')
print('  podem ser inteiramente substituidas?" As duas respostas sao')
print('  validas; elas medem coisas diferentes.')
```

```python
# --- O MODELO ---

# Faixa real do risco, com correcao
p_risco_baixo = p_alto_risco_oecd   # 9% — abordagem por tarefas
p_risco_alto = p_alto_risco_fo      # 47% — abordagem por ocupacao
p_risco_medio = (p_risco_baixo + p_risco_alto) / 2  # ponto medio

p_risco_baixo_corrigido = p_risco_baixo * fator_correcao
p_risco_alto_corrigido = p_risco_alto * fator_correcao
p_risco_medio_corrigido = p_risco_medio * fator_correcao

# Goldman Sachs: exposicao vs substituicao
p_exposto_corrigido = p_expostos_algum_grau * fator_correcao
p_substituivel_corrigido = p_substituivel_trabalho * fator_correcao

print('=' * 65)
print('  MODELO — RESULTADO')
print('=' * 65)
print(f'\n  Faixa do risco de automacao:')
print(f'  → Abordagem por ocupacao (Frey & Osborne): {p_risco_alto*100:.0f}%')
print(f'  → Abordagem por tarefa (OECD):             {p_risco_baixo*100:.0f}%')
print(f'  → Ponto medio:                             {p_risco_medio*100:.0f}%')
print(f'\n  Com fator de correcao (x{fator_correcao}):')
print(f'  → Alto: {p_risco_alto_corrigido*100:.0f}%')
print(f'  → Baixo: {p_risco_baixo_corrigido*100:.1f}%')
print(f'  → Medio: {p_risco_medio_corrigido*100:.1f}%')
print(f'\n  Goldman Sachs — IA generativa:')
print(f'  → Expostos: {p_expostos_algum_grau*100:.0f}% → corrigido: {p_exposto_corrigido*100:.0f}%')
print(f'  → Substituivel: {p_substituivel_trabalho*100:.0f}% → corrigido: {p_substituivel_corrigido*100:.0f}%')
print(f'\n  INTERPRETACAO:')
print(f'  → "Exposto" nao significa "eliminado"')
print(f'  → Historicamente, automacao cria mais empregos do que destroi')
print(f'  → O impacto real da IA generativa ainda nao se materializou')
print('=' * 65)
```

```python
# --- VISUALIZACAO ---

# GRAFICO 1 — A mesma pergunta, duas respostas: 47% vs 9%
fig1, ax1 = plt.subplots(figsize=(12, 8))

categorias = ['Por ocupacao inteira\n(Frey & Osborne, 2017)', 'Por tarefas dentro\nda ocupacao\n(OECD, 2016)']
valores = [p_alto_risco_fo * 100, p_alto_risco_oecd * 100]
cores1 = ['#c0392b', '#2a8a82']

bars = ax1.bar(categorias, valores, color=cores1, alpha=0.85, width=0.45)
ax1.set_ylim(0, 60)
ax1.set_ylabel('Empregos em alto risco de automacao (%)', fontsize=12)
ax1.set_title('A mesma pergunta, duas respostas\nQuanto do trabalho esta em risco depende de como voce mede',
              fontsize=13, pad=15)

for bar, v in zip(bars, valores):
    ax1.text(bar.get_x() + bar.get_width() / 2, v + 1.5,
             f'{v:.0f}%', ha='center', fontweight='bold', fontsize=18)

ax1.annotate('', xy=(1, 9), xytext=(0, 47),
             arrowprops=dict(arrowstyle='->', color='#c8a84b', lw=2.5))
ax1.text(0.5, 30, '38pp de\ndiferenca', ha='center',
         fontsize=13, color='#c8a84b', fontweight='bold')

plt.tight_layout()
plt.savefig('dia-023-grafico-01-ocupacao-vs-tarefa.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Grafico 1 salvo!')


# GRAFICO 2 — Goldman Sachs: exposto vs substituivel
fig2, ax2 = plt.subplots(figsize=(12, 8))

categorias2 = ['Exposto a algum\ngrau de automacao', 'Trabalho que poderia\nser substituido']
valores2 = [p_expostos_algum_grau * 100, p_substituivel_trabalho * 100]
cores2 = ['#c8a84b', '#c0392b']

bars2 = ax2.bar(categorias2, valores2, color=cores2, alpha=0.85, width=0.45)
ax2.set_ylim(0, 85)
ax2.set_ylabel('%', fontsize=12)
ax2.set_title('"Exposto" nao e a mesma coisa que "eliminado"\nGoldman Sachs (2023) — Briggs & Kodnani — 900+ ocupacoes',
              fontsize=13, pad=15)

for bar, v in zip(bars2, valores2):
    ax2.text(bar.get_x() + bar.get_width() / 2, v + 2,
             f'{v:.0f}%', ha='center', fontweight='bold', fontsize=18)

plt.tight_layout()
plt.savefig('dia-023-grafico-02-exposto-vs-substituivel.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Grafico 2 salvo!')


# GRAFICO 3 — Linha do tempo: previsao vs realidade
fig3, ax3 = plt.subplots(figsize=(14, 8))

anos = [2013, 2016, 2023, 2025, 2027]
labels = [
    'Frey & Osborne:\n47% em risco',
    'OECD:\n9% em risco\n(por tarefas)',
    'Goldman Sachs:\n300M expostos\n(IA generativa)',
    'Goldman Sachs:\nimpacto ainda\nnegligivel',
    'Goldman Sachs:\nimpacto esperado\n(previsao)'
]
cores3 = ['#c0392b', '#2a8a82', '#c8a84b', '#888888', '#888888']
marcadores = ['o', 'o', 'o', 'o', 's']

for i, (ano, label, cor, marc) in enumerate(zip(anos, labels, cores3, marcadores)):
    ax3.scatter(ano, 0, s=200, color=cor, zorder=3, marker=marc)
    offset_y = 0.3 if i % 2 == 0 else -0.4
    ax3.annotate(label, xy=(ano, 0), xytext=(ano, offset_y),
                 fontsize=11, ha='center', va='center',
                 arrowprops=dict(arrowstyle='-', color=cor, lw=1.5),
                 color=cor, fontweight='bold')

ax3.axhline(y=0, color='#333333', linewidth=2)
ax3.set_xlim(2011, 2029)
ax3.set_ylim(-1, 1)
ax3.set_xlabel('Ano', fontsize=12)
ax3.set_yticks([])
ax3.set_title('A previsao que ainda nao se cumpriu\nLinha do tempo: automacao e emprego',
              fontsize=13, pad=15)

plt.tight_layout()
plt.savefig('dia-023-grafico-03-linha-do-tempo.png', dpi=150, bbox_inches='tight')
plt.close()
print('✅ Grafico 3 salvo!')
```

### 💡 O Insight

Em 2013, dois pesquisadores de Oxford estimaram que **47% dos empregos nos EUA** estavam em alto risco de automação. O número virou manchete no mundo inteiro e moldou uma década de medo sobre o futuro do trabalho.

Mas quando a OECD refez a análise três anos depois, olhando por **tarefas** dentro de cada ocupação em vez de ocupações inteiras, o número caiu para **9%**. A diferença de 38 pontos percentuais não é um erro — é uma pergunta diferente. "Quantas ocupações têm tarefas automatizáveis?" é diferente de "Quantas ocupações podem ser inteiramente substituídas?"

Em 2023, o Goldman Sachs adicionou a IA generativa e estimou que **300 milhões de empregos** poderiam ser "expostos". Mas em 2025, o mesmo time publicou: o impacto real no mercado de trabalho era, até aquele momento, **negligível**. O desemprego não subiu. A produtividade não explodiu. Os empregos que deveriam ter desaparecido ainda estavam lá.

Isso não significa que a automação não vai mudar o trabalho — vai, e já está mudando. Mas a forma como ela muda é diferente do que a manchete de 47% sugere. Ela não elimina ocupações inteiras de uma vez. Ela transforma tarefas dentro de ocupações — e cria outras que não existiam antes.

*A pergunta que importa não é "meu trabalho vai existir daqui a 10 anos?" É: "as tarefas que eu faço hoje vão existir — e quais novas tarefas eu preciso aprender a fazer?"*

### ⚠️ Limitações do Modelo
- A estimativa de 47% de Frey & Osborne usa uma abordagem por ocupação inteira que superestima o risco, segundo a OECD — mas foi publicada em revista peer-reviewed e é amplamente citada
- A estimativa de 9% da OECD corrige pela heterogeneidade de tarefas, mas pode subestimar o ritmo de avanço da IA — foi feita antes da IA generativa
- O relatório do Goldman Sachs (2023) não é um estudo peer-reviewed — é uma análise de banco de investimento com interesses próprios no mercado de tecnologia
- O follow-up de 2025 mostra que o impacto não se materializou ainda, mas isso não significa que não vai se materializar — pode simplesmente estar atrasado
- Nenhuma dessas fontes pode prever com certeza o que vai acontecer daqui a 10 anos — são extrapolações baseadas em tecnologia atual
- O fator de correção (×0,80) foi aplicado às estimativas percentuais

*A ciência é honesta sobre o que não sabe. O modelo também.*
