### 📖 A História

A proposta está na mesa. A cabeça está cheia. E alguém, com a
melhor das intenções, entrega o conselho mais universal do
repertório humano: dorme sobre isso. A noite é boa conselheira.
Amanhã você decide melhor.

Você adia. Todo mundo adia. E por vinte anos, esse conselho teve
até certidão científica: um estudo publicado na Science em 2006
dizia que, para decisões complexas, **se distrair do problema
produzia escolhas melhores do que pensar nele**. O inconsciente,
dizia a teoria, trabalharia por você. Virou manchete, virou
livro, virou slide de consultor.

Aí a ciência fez o que faz de melhor quando amadurece: voltou
para conferir.

Sessenta e um efeitos publicados foram reunidos e pesados.
Uma réplica gigante, com 399 pessoas e poder estatístico de
sobra até para efeitos pequenos, refez o experimento original.

Resultado da réplica: **nada**. Efeito de -0,01.

E o conjunto da literatura, corrigido pelo viés de publicação:
**0,018**, com o intervalo abraçando o zero dos dois lados.

O conselho continua simpático. O efeito não existe.

---

### 📚 O Conceito: A Mágica do Inconsciente Esperto

A alegação original tinha nome técnico, "deliberação sem
atenção": diante de escolhas com muitos atributos, pensar
conscientemente atrapalharia, e um período de distração deixaria
o inconsciente organizar tudo e entregar a melhor resposta.

O campo se dividiu em duas explicações para os resultados
irregulares que se acumulavam. A **conta dos moderadores**: o
efeito existe, mas só aparece sob condições específicas. E a
**conta da confiabilidade**: o efeito não existe, e os achados
positivos eram ruído de experimentos pequenos passando pelo
filtro da publicação.

A réplica de 399 pessoas foi desenhada para arbitrar, com poder
de 99,7% até para o efeito pequeno da meta favorável. O veredito
pesou para a segunda conta. E o Bayes factor deu a medida: os
dados são **quase 8 vezes mais prováveis** num mundo em que o
efeito não existe.

Duas fronteiras para o dia não ser mal lido:

**Dormir de verdade não está em julgamento.** O que morreu é o
truque da distração, a mágica do inconsciente esperto. Os efeitos
do sono real sobre memória e cansaço são outra literatura.

**E a ponte com o Dia #013 é declarada:** lá, a pausa e a mente
vagando ajudavam a CRIATIVIDADE (incubação, Baird 2012, fontes
distintas). Aqui, o teste é DECIDIR MELHOR entre opções. Criar e
decidir são construtos diferentes, e o veredito diverge sem
contradição.

### 🧮 O Modelo

**Fontes:**
- Nieuwenstein, M. R., Wierenga, T., Morey, R. D., Wicherts,
  J. M., Blom, T. N., Wagenmakers, E.-J. & van Rijn, H. (2015) -
  On making the right choice: A meta-analysis and large-scale
  replication attempt of the unconscious thought advantage -
  Judgment and Decision Making, 10(1), 1-17, acesso aberto -
  meta com critérios estritos: **61 efeitos, 31 estudos, 45 não
  significativos**; corrigida por viés de publicação:
  **g=0,018 [-0,10, 0,14]**; réplica **N=399** (196 deliberação,
  203 distração), poder 0,997: **g=-0,01 (EP 0,10)**; Bayes
  factor ~7,8 a favor do nulo. **Fonte central.**
- Strick, M., Dijksterhuis, A. et al. (2011) - Social Cognition -
  a meta dos proponentes, 92 estudos: g=0,224 [0,145, 0,303].
  O outro lado, declarado.
- Dijksterhuis, A. et al. (2006) - Science - a origem:
  experimentos fundadores com d=0,69 e amostras minúsculas
  (ex.: N=30).
- Vlaev, I. et al. - meta em decisão médica: efeito nulo também
  entre profissionais. Apoio.

**Conflitos declarados dos dois lados:** a meta favorável tem o
autor original entre os autores; a meta estrita e a réplica vêm
do time cético (incluindo Wagenmakers). Os dois lados na mesa,
com a réplica pré-planejada como árbitro.

**Fator ×0,80: não se aplica.** Meta-análises e experimentos
controlados.

**As contas nossas:** (1) posterior sobre o efeito corrigido;
(2) a escada em duelos: cada degrau da história traduzido em
"de 100 confrontos entre distraído e deliberador, quantos o
distraído vence" (probabilidade de superioridade, suposição de
normalidade declarada).


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---

# A escada do encolhimento
d_fundadores = 0.69          # os 6 estudos originais (Dijksterhuis, Science 2006)
g_strick, ic_strick = 0.224, (0.145, 0.303)   # meta dos proponentes, 92 estudos
g_estrita, ic_estrita = 0.018, (-0.10, 0.14)  # Nieuwenstein 2015, corrigida por viés
g_replica, se_replica = -0.01, 0.10           # réplica N=399, poder 0,997

n_efeitos, n_estudos, n_ns = 61, 31, 45       # 45 de 61 efeitos: não significativos
n_replica = 399
bf_nulo = 7.8                                  # Bayes factor pró-nulo

print("=" * 65)
print("  DADOS DA LITERATURA - A ESCADA DO ENCOLHIMENTO")
print("=" * 65)
print(f"\n  1. Os fundadores (Science 2006, amostras ~30):   d = {d_fundadores}")
print(f"  2. Meta dos proponentes (92 estudos):            g = {g_strick}  [{ic_strick[0]}, {ic_strick[1]}]")
print(f"  3. Meta estrita, corrigida por viés:             g = {g_estrita}  [{ic_estrita[0]}, {ic_estrita[1]}]")
print(f"     ({n_efeitos} efeitos, {n_estudos} estudos, {n_ns} não significativos)")
print(f"  4. Réplica pré-planejada (N={n_replica}, poder 0,997): g = {g_replica}")
print(f"\n  Bayes factor: dados ~{bf_nulo}x mais prováveis sob 'o efeito não existe'")
print(f"  Bônus: meta em decisão médica: nulo também")
print(f"\n  Fator ×0,80: NÃO aplicado (meta-análises e experimentos)")
print("=" * 65)
```

```python
# --- O MODELO ---
# Conta 1: posterior sobre o efeito corrigido
# Conta 2: a escada em duelos (probabilidade de superioridade)

# ═══ CONTA 1 - a posterior ═══
se_estrita = (ic_estrita[1] - ic_estrita[0]) / (2 * 1.96)
p_positivo = 1 - stats.norm.cdf((0 - g_estrita) / se_estrita)
p_pequeno_real = 1 - stats.norm.cdf((0.2 - g_estrita) / se_estrita)

# ═══ CONTA 2 - a escada em duelos ═══
def duelos(d):
    return stats.norm.cdf(d / np.sqrt(2)) * 100

escada = {
    'Fundadores (2006)': duelos(d_fundadores),
    'Meta favorável (2011)': duelos(g_strick),
    'Meta estrita (2015)': duelos(g_estrita),
    'Réplica N=399': duelos(g_replica),
}

print("=" * 65)
print("  MODELO - AS DUAS CONTAS")
print("=" * 65)
print(f"\n  CONTA 1 - o que sobrou do efeito (posterior, aprox. declarada):")
print(f"  → P(efeito > 0)                    = {p_positivo*100:.0f}%   ← moeda")
print(f"  → P(efeito ao menos pequeno, ≥0,2) = {p_pequeno_real*100:.2f}%")
print(f"\n  CONTA 2 - a escada em duelos:")
print(f"  De 100 confrontos, quantos o 'distraído' vence?")
for nome, v in escada.items():
    print(f"  → {nome:<24} {v:.1f} de 100")
print(f"\n  → O conselho nasceu vencendo 69 de 100 duelos")
print(f"     e terminou vencendo 49,7: uma moeda com meio ponto")
print(f"     de vício, para o lado errado.")
print(f"\n  O BAYES FACTOR, em português:")
print(f"  → Se o efeito existisse, dados como estes seriam raros.")
print(f"     Se não existisse, seriam esperados. A razão: ~{bf_nulo}x")
print(f"     a favor do 'não existe'.")
print(f"\n  A LEITURA:")
print(f"  → Se a noite ajuda, é pelo descanso e pela distância")
print(f"     emocional, mecanismos que não estavam em julgamento.")
print(f"  → O que morreu foi a mágica do inconsciente esperto.")
print("=" * 65)
```

```python
# --- VISUALIZAÇÃO - GRÁFICOS SEPARADOS ---

# ── GRÁFICO 1 - A escada do encolhimento ──
fig1, ax1 = plt.subplots(figsize=(10, 6.5))

labels = ['Fundadores\nScience 2006\n(amostras ~30)', 'Meta favorável\n2011\n(92 estudos)',
          'Meta estrita\ncorrigida, 2015\n(61 efeitos)', 'Réplica\npré-planejada\n(N=399)']
pontos = [d_fundadores, g_strick, g_estrita, g_replica]
ics = [None, ic_strick, ic_estrita, (g_replica - 1.96*se_replica, g_replica + 1.96*se_replica)]
xs = [0, 1, 2, 3]
cores = ['#c9a14a', '#c9a14a', '#c0392b', '#c0392b']

ax1.axhline(0, color='#333', linewidth=1.2)
for x, pt, ic, cor in zip(xs, pontos, ics, cores):
    if ic:
        ax1.plot([x, x], [ic[0], ic[1]], color=cor, linewidth=2.5)
        ax1.plot([x-0.06, x+0.06], [ic[0], ic[0]], color=cor, linewidth=2.5)
        ax1.plot([x-0.06, x+0.06], [ic[1], ic[1]], color=cor, linewidth=2.5)
    ax1.scatter([x], [pt], s=140, color=cor, zorder=3)
    ax1.text(x, pt + 0.055, f'{pt:+.3f}' if abs(pt) < 0.1 else f'{pt:+.2f}',
             ha='center', fontweight='bold', fontsize=12.5)

ax1.set_xticks(xs)
ax1.set_xticklabels(labels, fontsize=10)
ax1.set_ylabel('Tamanho de efeito (d / g)')
ax1.set_ylim(-0.32, 0.92)
ax1.set_title('A escada do encolhimento: de 0,69 a nada em nove anos\n'
              'Quanto maior e mais rigoroso o teste, menor o efeito',
              fontsize=12.5, pad=15)

plt.figtext(0.5, 0.01,
            'Fontes: Dijksterhuis et al. (2006); Strick et al. (2011); Nieuwenstein et al. (2015) | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-096-grafico-01-escada.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 - A posterior do que sobrou ──
fig2, ax2 = plt.subplots(figsize=(10, 6))

x = np.linspace(-0.25, 0.30, 500)
y = stats.norm.pdf(x, g_estrita, se_estrita)
ax2.plot(x, y, color='#c0392b', linewidth=3)
ax2.fill_between(x, y, where=(x > 0), alpha=0.18, color='#c9a14a')
ax2.fill_between(x, y, where=(x <= 0), alpha=0.18, color='#6b6a64')
ax2.axvline(0, color='#333', linewidth=1.5)
ax2.axvline(0.2, color='#2a8a82', linestyle='--', linewidth=1.8)
ax2.text(0.203, max(y)*0.85, 'efeito "pequeno\nmas real" (0,2):\nP = 0,15%',
         fontsize=10.5, color='#1a5f5a')
ax2.text(g_estrita, max(y)*1.03, f'g = {g_estrita}', ha='center',
         fontsize=12, fontweight='bold', color='#c0392b')
ax2.text(-0.13, max(y)*0.45, f'P(efeito > 0)\n= {p_positivo*100:.0f}%\numa moeda',
         fontsize=11.5, color='#444', ha='center')
ax2.set_xlabel('Tamanho do efeito (g)')
ax2.set_ylabel('Densidade')
ax2.set_title('O que sobrou do efeito, com incerteza honesta\n'
              'Posterior sobre a meta corrigida por viés de publicação',
              fontsize=12.5, pad=15)

plt.figtext(0.5, 0.01,
            'Conta própria sobre Nieuwenstein et al. (2015) | aproximação normal declarada | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-096-grafico-02-posterior.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 - A escada em duelos ──
fig3, ax3 = plt.subplots(figsize=(10, 6))

nomes = list(escada.keys())
vals = list(escada.values())
cores3 = ['#c9a14a', '#c9a14a', '#c0392b', '#c0392b']

bars = ax3.bar(range(4), vals, color=cores3, alpha=0.88, width=0.5)
ax3.axhline(50, color='#333', linestyle='--', linewidth=1.5)
ax3.text(3.42, 50.8, 'moeda: 50', fontsize=10.5, color='#333', ha='right')
ax3.set_xticks(range(4))
ax3.set_xticklabels([n.replace(' (', '\n(') for n in nomes], fontsize=10)
ax3.set_ylim(40, 75)
ax3.set_ylabel('Duelos vencidos pelo "distraído" (de 100)')
ax3.set_title('A escada em duelos: o conselho encolhendo até virar moeda\n'
              'Tradução própria (probabilidade de superioridade)',
              fontsize=12.5, pad=15)

for bar, v in zip(bars, vals):
    ax3.text(bar.get_x() + bar.get_width()/2, v + 0.6,
             f'{v:.1f}', ha='center', fontweight='bold', fontsize=14)

plt.figtext(0.5, 0.01,
            'Conta própria | suposição de normalidade declarada | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-096-grafico-03-duelos.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**"Durma sobre o assunto" nasceu vencendo 69 de 100 duelos.
Terminou vencendo 49,7: uma moeda com meio ponto de vício,
para o lado errado.**

A escada do encolhimento é a biografia clássica de um efeito que
não existia: amostras de 30 pessoas e manchete na Science; uma
meta dos proponentes segurando um efeito pequeno; a meta estrita
corrigida encontrando 0,018 com o intervalo abraçando o zero; e a
réplica de 399 pessoas, com poder de sobra, entregando -0,01.
A chance de sobrar ao menos um efeito pequeno: **0,15%**.

O que sobrevive do conselho, sobrevive por outros caminhos: se a
noite ajuda, é pelo descanso e pela distância emocional,
mecanismos que não estavam em julgamento. O que morreu foi a
mágica específica: a ideia de que o seu inconsciente decide
melhor que você.

E a fronteira com o Dia #013 fica registrada: a pausa que ajudava
a criatividade segue de pé, com fontes próprias. Criar e decidir
são construtos diferentes, e a ciência tem o direito de dar
vereditos diferentes para cada um.

A pergunta que fica:

*Quantos conselhos universais da sua vida têm N=30
na certidão de nascimento?*

### ⚠️ Limitações do Modelo

- A disputa acadêmica não está formalmente encerrada: a conta dos
  moderadores segue defendida por proponentes; a réplica
  pré-planejada com poder 0,997 pesa fortemente contra, e o dia
  declara para que lado pesa
- O Bayes factor depende do modelo da hipótese alternativa
  (construído sobre os estudos fundadores); com outras
  especificações, o valor muda, a direção não
- Conflitos dos dois lados declarados: a meta favorável tem o
  autor original entre os autores; a estrita vem do time cético
- Tarefas de laboratório (escolher carros e apartamentos por
  atributos), não decisões de vida com consequências
- A escada em duelos assume normalidade e usa o ponto central de
  cada estimativa; é tradução, não novo teste
- O sono real e seus efeitos sobre memória e cansaço não foram
  testados aqui: a fronteira fica explícita para o conselho não
  ser enterrado além do que os dados autorizam

*A ciência é honesta sobre o que não sabe. O modelo também.*
