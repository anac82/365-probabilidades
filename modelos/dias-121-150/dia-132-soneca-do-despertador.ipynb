### 📚 O Conceito

A soneca do despertador tem má fama. A ideia corrente é que ela fragmenta o sono, rouba
descanso e deixa a pessoa mais cansada do que se levantasse no primeiro toque. Até 2022,
quase nada disso tinha sido medido em gente de verdade, manhã por manhã.

Um grupo da Universidade de Notre Dame acompanhou 450 profissionais em tempo integral com
um relógio no pulso e uma pergunta por dia sobre como tinham acordado. Com isso dá para
responder duas perguntas diferentes. Quantas pessoas se dizem adeptas da soneca? E, numa
manhã qualquer, quantas de fato a apertaram? A primeira é hábito declarado. A segunda é
registro do dia. A pergunta do título é a segunda.

---

### 🧮 O Modelo

Quatro camadas. A fonte central dá as contagens e as razões de chances. As contas são nossas.

1. **Amanhã cedo.** A fatia das manhãs com soneca, com o intervalo contado de dois jeitos
   (por manhã e por pessoa), ao lado do alarme único e do acordar sem alarme.
2. **Os adeptos.** Quem se declara adepto, com e sem o fator ×0,80, e o que a declaração
   diz sobre a manhã.
3. **Homens, mulheres e a idade.** As razões de chances do artigo transformadas em
   probabilidade por sexo e idade, ancoradas na proporção observada entre as mulheres.
4. **Os passos.** Cada mil passos por dia e a chance de ser adepto, na mesma reconstrução.

**Fonte central:**
- Mattingly, S. M., Martinez, G., Young, J., Cain, M. K. & Striegel, A. (2022). "Snoozing:
  an examination of a common method of waking". *Sleep* 45(10), zsac184. **N = 450**
  profissionais em tempo integral nos Estados Unidos (253 homens e 197 mulheres), com
  relógio vestível. **257 (57%) se declaram adeptos da soneca.** Em **1.994 manhãs de
  385 pessoas**, houve soneca em **620 (31%)**, alarme único em 600 (30%) e nenhum alarme
  em 774 (39%). Regressão logística para ser adepto: homens com razão de chances **0,51
  [0,32; 0,81]**; cada ano de idade, 0,98 (p = 0,029); cada mil passos por dia, 0,89
  (p = 0,008). Passos por dia: adeptos 7.249, não adeptos 8.390, média 7.749. Sono por tipo
  de manhã: sem alarme 8,74 h, alarme único 7,83 h, soneca 7,95 h. Adeptos e não adeptos
  dormem o mesmo (8,15 h e 8,21 h) e não relatam mais privação de sono. **A tabela 1 do
  artigo soma 260 adeptos (133 homens e 127 mulheres) e o texto diz 257. Uso 257, do
  texto, e a divisão por sexo da tabela.** Os intervalos das razões de chances de idade e
  passos não foram lidos.

**Fonte de apoio:**
- Sundelin, T., Landry, S. & Axelsson, J. (2023). "Is snoozing losing? Why intermittent
  morning alarms are used and how they affect sleep, cognition, cortisol, and mood".
  *Journal of Sleep Research*, doi 10.1111/jsr.14054. Survey com **1.732** pessoas, mais
  soneca entre os mais jovens e os de cronotipo tardio. Laboratório com **31** adeptos:
  30 minutos de soneca custaram cerca de 6 minutos de sono, evitaram despertar do sono
  profundo e melhoraram ou não afetaram os testes ao levantar, sem efeito em cortisol e
  humor. **Lido no resumo e em fonte secundária. Os percentuais que circulam (69% usam
  soneca ou alarmes múltiplos, 60% voltam a dormir entre toques) não foram conferidos na
  primária e não entram na conta.**

**Nota metodológica.** O fator ×0,80 se aplica aos 57% de adeptos, que é hábito declarado
em questionário. Não se aplica às 620 manhãs em 1.994, que são registro diário cruzado
com o relógio, nem às horas de sono medidas.

**Vizinhança interna:** #119 (sono, 10/10) e #076 (período circadiano), com outras fontes.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

SEMENTE = 42
rng = np.random.default_rng(SEMENTE)

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

VERMELHO, DOURADO, VERDE, CINZA = '#c0392b', '#c8a84b', '#1a5f5a', '#6b6a64'
RODAPE = 'Fonte: Mattingly et al. 2022, Sleep 45(10) · cálculo próprio | #365Probabilidades'

def vir(x, casas=1):
    return f"{x:.{casas}f}".replace(".", ",")

def pct(x, casas=1):
    return vir(100 * x, casas) + "%"

def mil(n):
    return f"{int(n):,}".replace(",", ".")

def jeffreys(k, n):
    d = stats.beta(k + 0.5, n - k + 0.5)
    return d, d.interval(0.95)

print("✅ Bibliotecas carregadas · semente", SEMENTE)
```

```python
# --- DADOS DA LITERATURA ---
# Mattingly et al. (2022), Sleep 45(10), zsac184
N = 450
N_HOMENS, N_MULHERES = 253, 197
ADEPTOS = 257                      # texto do artigo (a tabela 1 soma 260)
ADEPTOS_H, ADEPTOS_M = 133, 127    # tabela 1
MANHAS, PESSOAS_MANHAS = 1_994, 385
SONECA, ALARME_UNICO, SEM_ALARME = 620, 600, 774
RC_HOMEM, RC_HOMEM_IC = 0.51, (0.32, 0.81)   # razão de chances, homens contra mulheres
RC_ANO = 0.98                      # por ano de idade (p = 0,029); intervalo não lido
RC_MIL_PASSOS = 0.89               # por mil passos a mais por dia (p = 0,008); intervalo não lido
PASSOS_ADEPTOS, PASSOS_NAO, PASSOS_MEDIA = 7_249, 8_390, 7_749
IDADE_MEDIA = 36.3
SONO_H = {'sem alarme': 8.74, 'alarme único': 7.83, 'soneca': 7.95}
SONO_ADEPTOS, SONO_NAO = 8.15, 8.21
FATOR = 0.80

# Sundelin, Landry & Axelsson (2023), J Sleep Res, doi 10.1111/jsr.14054
SUNDELIN_SURVEY, SUNDELIN_LAB, SUNDELIN_SONO_PERDIDO_MIN = 1_732, 31, 6

print("=" * 72)
print("  DADOS · MATTINGLY ET AL. (2022)")
print("=" * 72)
print(f"\n  {N} profissionais ({N_HOMENS} homens, {N_MULHERES} mulheres) · idade média {vir(IDADE_MEDIA)} anos")
print(f"  Adeptos da soneca: {ADEPTOS} de {N} = {pct(ADEPTOS / N)}  (tabela 1 soma {ADEPTOS_H + ADEPTOS_M})")
print(f"    mulheres {ADEPTOS_M}/{N_MULHERES} = {pct(ADEPTOS_M / N_MULHERES)} · homens {ADEPTOS_H}/{N_HOMENS} = {pct(ADEPTOS_H / N_HOMENS)}")
print(f"\n  {mil(MANHAS)} manhãs de {PESSOAS_MANHAS} pessoas:")
print(f"    soneca        {SONECA:>4} = {pct(SONECA / MANHAS)}")
print(f"    alarme único  {ALARME_UNICO:>4} = {pct(ALARME_UNICO / MANHAS)}")
print(f"    sem alarme    {SEM_ALARME:>4} = {pct(SEM_ALARME / MANHAS)}")
print(f"\n  Razões de chances para ser adepto:")
print(f"    homem contra mulher      {vir(RC_HOMEM, 2)} [{vir(RC_HOMEM_IC[0], 2)}; {vir(RC_HOMEM_IC[1], 2)}]")
print(f"    por ano de idade         {vir(RC_ANO, 2)}  (p = 0,029)")
print(f"    por mil passos por dia   {vir(RC_MIL_PASSOS, 2)}  (p = 0,008)")
print(f"  Passos por dia: adeptos {mil(PASSOS_ADEPTOS)} · não adeptos {mil(PASSOS_NAO)} · média {mil(PASSOS_MEDIA)}")
print(f"\n  Horas de sono: " + " · ".join(f"{k} {vir(v, 2)} h" for k, v in SONO_H.items()))
print(f"  Adeptos {vir(SONO_ADEPTOS, 2)} h · não adeptos {vir(SONO_NAO, 2)} h · sem diferença em privação relatada")
print(f"\n  Sundelin et al. (2023): survey com {mil(SUNDELIN_SURVEY)} pessoas; laboratório com {SUNDELIN_LAB} adeptos,")
print(f"  30 min de soneca custaram cerca de {SUNDELIN_SONO_PERDIDO_MIN} min de sono, sem efeito em cortisol e humor.")
print(f"\n  ×0,80: só nos {pct(ADEPTOS / N, 0)} de adeptos (hábito declarado). Não nas manhãs registradas.")
print("=" * 72)
```

```python
# --- O MODELO ---

# ── CAMADA 1 · amanhã cedo ──
p_soneca = SONECA / MANHAS
d_manha, ic_manha = jeffreys(SONECA, MANHAS)
# intervalo conservador: as manhãs de uma mesma pessoa se parecem; conto como se fossem 385 observações
k_pessoa = p_soneca * PESSOAS_MANHAS
d_pessoa, ic_pessoa = jeffreys(k_pessoa, PESSOAS_MANHAS)
print("=" * 72)
print("  CAMADA 1 · AMANHÃ CEDO")
print("=" * 72)
print(f"\n  P(apertar a soneca numa manhã) = {SONECA}/{MANHAS} = {pct(p_soneca)}")
print(f"  Intervalo contando cada manhã ({MANHAS}):  [{pct(ic_manha[0])}; {pct(ic_manha[1])}]")
print(f"  Intervalo contando cada pessoa ({PESSOAS_MANHAS}): [{pct(ic_pessoa[0])}; {pct(ic_pessoa[1])}]")
print("  O primeiro é estreito demais, porque as manhãs de uma mesma pessoa se parecem.")
print("  O segundo é o limite conservador. O valor real fica entre os dois.")
for nome, k in (('alarme único', ALARME_UNICO), ('sem alarme', SEM_ALARME)):
    _, ic = jeffreys(k, MANHAS)
    print(f"  {nome:>13}: {pct(k / MANHAS)}  [{pct(ic[0])}; {pct(ic[1])}]")
print(f"  Em cem manhãs, {vir(100 * p_soneca, 0)} começam com soneca, {vir(100 * ALARME_UNICO / MANHAS, 0)} com um toque só e {vir(100 * SEM_ALARME / MANHAS, 0)} sem alarme.")

# ── CAMADA 2 · os adeptos ──
p_adepto = ADEPTOS / N
d_adepto, ic_adepto = jeffreys(ADEPTOS, N)
p_adepto_80 = p_adepto * FATOR
# se só os adeptos apertam a soneca, com que frequência eles a usam?
p_soneca_dado_adepto = p_soneca / p_adepto
print("\n" + "=" * 72)
print("  CAMADA 2 · OS ADEPTOS")
print("=" * 72)
print(f"\n  Adeptos declarados: {pct(p_adepto)}  IC 95% [{pct(ic_adepto[0])}; {pct(ic_adepto[1])}]")
print(f"  Com o fator ×0,80 (convenção do projeto para hábito declarado): {pct(p_adepto_80)}")
print(f"  Se só os adeptos apertam a soneca, eles a apertam em {pct(p_soneca_dado_adepto)} das manhãs.")
print(f"  Com {pct(p_adepto_80, 0)} de adeptos, seria em {pct(p_soneca / p_adepto_80)} das manhãs. Limite, não dado.")
print(f"  Adeptos dormem {vir(SONO_ADEPTOS, 2)} h e não adeptos {vir(SONO_NAO, 2)} h. A soneca não rouba a noite.")

# ── CAMADA 3 · homens, mulheres e a idade ──
# Reconstrução: ancoro as chances na proporção observada entre as mulheres, na idade média da
# amostra, e aplico as razões de chances ajustadas do artigo. Premissa declarada: as razões
# ajustadas valem sobre a proporção bruta das mulheres.
odds_m = (ADEPTOS_M / N_MULHERES) / (1 - ADEPTOS_M / N_MULHERES)
def p_adepto_de(sexo, idade, passos=PASSOS_MEDIA):
    o = odds_m
    if sexo == 'homem':
        o *= RC_HOMEM
    o *= RC_ANO ** (idade - IDADE_MEDIA)
    o *= RC_MIL_PASSOS ** ((passos - PASSOS_MEDIA) / 1000)
    return o / (1 + o)
print("\n" + "=" * 72)
print("  CAMADA 3 · HOMENS, MULHERES E A IDADE")
print("=" * 72)
print(f"\n  Âncora: mulheres, {vir(IDADE_MEDIA)} anos, {mil(PASSOS_MEDIA)} passos: {pct(ADEPTOS_M / N_MULHERES)} adeptas")
print(f"  Homens na mesma idade, pela razão de chances {vir(RC_HOMEM, 2)}: {pct(p_adepto_de('homem', IDADE_MEDIA))}")
print(f"  Dez anos a mais tiram {pct(1 - RC_ANO ** 10, 0)} das chances (0,98 elevado a 10 = {vir(RC_ANO ** 10, 2)}).")
print(f"\n  {'idade':>6} {'mulheres':>10} {'homens':>10}")
for idade in (20, 25, 30, 36, 40, 44, 50, 55, 60):
    print(f"  {idade:>6} {pct(p_adepto_de('mulher', idade)):>10} {pct(p_adepto_de('homem', idade)):>10}" + ("   ← eu" if idade == 44 else ""))
print("  Premissa: razões de chances ajustadas aplicadas sobre a proporção bruta das mulheres.")
print("  O intervalo da razão por idade não foi lido; a curva é estimativa pontual.")

# ── CAMADA 4 · os passos ──
print("\n" + "=" * 72)
print("  CAMADA 4 · OS PASSOS")
print("=" * 72)
print(f"\n  Cada mil passos a mais por dia tira {pct(1 - RC_MIL_PASSOS, 0)} das chances de ser adepto.")
print(f"  Adeptos andam {mil(PASSOS_ADEPTOS)} passos por dia; não adeptos, {mil(PASSOS_NAO)}. Diferença de {mil(PASSOS_NAO - PASSOS_ADEPTOS)}.")
print(f"\n  {'passos/dia':>11} {'mulher 36':>10} {'homem 36':>10} {'mulher 44':>10}")
for passos in (3000, 5000, 7749, 10000, 12000, 15000):
    print(f"  {mil(passos):>11} {pct(p_adepto_de('mulher', 36.3, passos)):>10} {pct(p_adepto_de('homem', 36.3, passos)):>10} {pct(p_adepto_de('mulher', 44, passos)):>10}")
p_eu = p_adepto_de('mulher', 44, 12000)
print(f"\n  Uma mulher de 44 anos com 12 mil passos por dia: {pct(p_eu)}, perto de um homem de 36 na média de passos.")
print("  É associação, não causa. Quem anda mais pode acordar mais cedo, e não o contrário.")
print("=" * 72)
```

```python
# --- VISUALIZAÇÃO ---

# ── GRÁFICO 1 · as três manhãs ──
fig1, ax1 = plt.subplots(figsize=(11, 6))
nomes = ['sem alarme', 'alarme único', 'soneca']
ks = [SEM_ALARME, ALARME_UNICO, SONECA]
cores = [CINZA, DOURADO, VERMELHO]
for i, (nome, k, cor) in enumerate(zip(nomes, ks, cores)):
    p = k / MANHAS
    _, ic = jeffreys(k, MANHAS)
    _, icp = jeffreys(p * PESSOAS_MANHAS, PESSOAS_MANHAS)
    ax1.barh(i, 100 * p, color=cor, alpha=0.85, height=0.55)
    ax1.plot([100 * icp[0], 100 * icp[1]], [i, i], color='black', lw=1.2)
    ax1.plot([100 * ic[0], 100 * ic[1]], [i, i], color='black', lw=4)
    ax1.text(100 * p + 1.2, i + 0.33, f"{pct(p)}  ·  {k} de {mil(MANHAS)}", fontsize=11, va='center')
ax1.set_yticks(range(3)); ax1.set_yticklabels(nomes, fontsize=12)
ax1.set_xlim(0, 52); ax1.set_xlabel('Manhãs (%)')
ax1.set_title('Como as manhãs começam\n1.994 manhãs de 385 pessoas · linha grossa: intervalo por manhã · linha fina: por pessoa',
              fontsize=13, pad=14)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-132-grafico-01-as-tres-manhas.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 · homens, mulheres e a idade ──
fig2, ax2 = plt.subplots(figsize=(11, 6))
idades = np.arange(20, 61)
ax2.plot(idades, [100 * p_adepto_de('mulher', i) for i in idades], color=VERMELHO, lw=3, label='mulheres')
ax2.plot(idades, [100 * p_adepto_de('homem', i) for i in idades], color=VERDE, lw=3, label='homens')
ax2.scatter([IDADE_MEDIA], [100 * ADEPTOS_M / N_MULHERES], color=VERMELHO, s=90, zorder=5, edgecolor='white')
ax2.scatter([IDADE_MEDIA], [100 * p_adepto_de('homem', IDADE_MEDIA)], color=VERDE, s=90, zorder=5, edgecolor='white')
ax2.axvline(44, color=DOURADO, ls='--', lw=1.5)
ax2.text(44.4, 22, f"44 anos\nmulheres {pct(p_adepto_de('mulher', 44))}\nhomens {pct(p_adepto_de('homem', 44))}", fontsize=10, color=DOURADO)
ax2.text(IDADE_MEDIA + 0.4, 100 * ADEPTOS_M / N_MULHERES + 2.5, 'âncora: 127 de 197 mulheres', fontsize=9, color=VERMELHO)
ax2.set_ylim(15, 85); ax2.set_xlabel('Idade'); ax2.set_ylabel('Chance de ser adepto da soneca (%)')
ax2.legend(frameon=False, fontsize=11)
ax2.set_title('Quem se declara adepto da soneca, por sexo e idade\nrazões de chances 0,51 (homens) e 0,98 por ano, sobre a proporção observada nas mulheres',
              fontsize=13, pad=14)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-132-grafico-02-homens-mulheres-idade.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 · os passos ──
fig3, ax3 = plt.subplots(figsize=(11, 6))
passos = np.arange(2000, 16001, 250)
ax3.plot(passos, [100 * p_adepto_de('mulher', IDADE_MEDIA, s) for s in passos], color=VERMELHO, lw=3, label='mulher de 36 anos')
ax3.plot(passos, [100 * p_adepto_de('homem', IDADE_MEDIA, s) for s in passos], color=VERDE, lw=3, label='homem de 36 anos')
ax3.plot(passos, [100 * p_adepto_de('mulher', 44, s) for s in passos], color=DOURADO, lw=2.5, ls='--', label='mulher de 44 anos')
ax3.axvline(PASSOS_ADEPTOS, color=CINZA, ls=':', lw=1.5); ax3.axvline(PASSOS_NAO, color=CINZA, ls=':', lw=1.5)
ax3.text(PASSOS_ADEPTOS - 150, 80, f'adeptos\n{mil(PASSOS_ADEPTOS)}', ha='right', fontsize=9, color=CINZA)
ax3.text(PASSOS_NAO + 150, 80, f'não adeptos\n{mil(PASSOS_NAO)}', ha='left', fontsize=9, color=CINZA)
ax3.set_ylim(15, 85); ax3.set_xlabel('Passos por dia'); ax3.set_ylabel('Chance de ser adepto da soneca (%)')
ax3.xaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: mil(v)))
ax3.legend(frameon=False, fontsize=11)
ax3.set_title('Cada mil passos por dia tira 11% das chances de ser adepto\nassociação, não causa',
              fontsize=13, pad=14)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-132-grafico-03-os-passos.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")

# ── GRÁFICO 4 · a assinatura: Beta dos adeptos, com e sem ×0,80 ──
fig4, ax4 = plt.subplots(figsize=(11, 6))
x = np.linspace(0.40, 0.70, 1000)
ax4.plot(100 * x, d_adepto.pdf(x), color=VERMELHO, lw=3)
ax4.fill_between(100 * x, d_adepto.pdf(x), alpha=0.18, color=VERMELHO)
ax4.axvline(100 * ic_adepto[0], color=DOURADO, ls='--', lw=2); ax4.axvline(100 * ic_adepto[1], color=DOURADO, ls='--', lw=2)
ax4.axvline(100 * p_adepto_80, color=VERDE, lw=2.5)
ax4.text(100 * p_adepto_80 - 0.6, d_adepto.pdf(p_adepto).max() * 0.85, f'×0,80\n{pct(p_adepto_80)}', ha='right', fontsize=11, color=VERDE)
ax4.text(100 * ic_adepto[1] + 0.5, d_adepto.pdf(p_adepto).max() * 0.6, f'IC 95%\n[{pct(ic_adepto[0])}; {pct(ic_adepto[1])}]', fontsize=11, color=DOURADO)
ax4.set_xlabel('Adeptos da soneca (%)'); ax4.set_ylabel('Densidade')
ax4.yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: vir(v, 1)))
ax4.set_title('Distribuição Beta · adeptos declarados da soneca\n257 de 450, posterior de Jeffreys · a linha verde é a convenção ×0,80 do projeto',
              fontsize=13, pad=14)
plt.figtext(0.5, 0.01, RODAPE, ha='center', fontsize=9, color='gray')
plt.tight_layout(rect=(0, 0.03, 1, 1))
plt.savefig('dia-132-grafico-04-bayesiano.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 4 salvo!")
```

### 💡 O Insight

**A chance de você apertar a soneca amanhã cedo é de 31%.** Entre 29% e 33% contando
cada manhã, entre 27% e 36% contando cada pessoa.

Em cem manhãs, 31 começam com soneca, 30 com um toque só e 39 sem alarme nenhum. Mais da
metade das pessoas se declara adepta, 57%, ou 46% com a convenção do projeto. Os adeptos
apertam a soneca em pouco mais da metade das manhãs deles, e dormem o mesmo que os outros.
A soneca não rouba a noite. Custa cerca de seis minutos, pelo laboratório sueco.

Quem aperta tem perfil. As mulheres são mais adeptas que os homens, com o dobro das chances.
Cada dez anos de idade tiram 18% das chances. E cada mil passos por dia tiram 11%. Uma
mulher de 44 anos que anda 12 mil passos tem a chance de um homem de 36 que anda a média.
É associação, e a seta pode ir nos dois sentidos.

### ⚠️ Limitações do Modelo

- **Uma amostra, um país, um ano.** 450 profissionais americanos com relógio vestível, em
  2022. Idade média de 36 anos. Nenhum dado brasileiro foi localizado.
- **As manhãs se repetem por pessoa.** O intervalo por manhã é estreito demais e o
  intervalo por pessoa é o limite conservador. O real fica entre os dois.
- **A tabela 1 do artigo soma 260 adeptos e o texto diz 257.** Uso 257. A diferença move
