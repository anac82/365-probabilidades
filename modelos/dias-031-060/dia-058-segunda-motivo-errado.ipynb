### 📖 A História

Todo domingo à noite, alguma coisa aperta.

O peito pesa um pouco. O sono custa a chegar.
E o nome que se dá pra isso é sempre o mesmo: segunda-feira.

A segunda virou o vilão oficial da semana. A gente reclama dela no
grupo do trabalho, nos memes, no café. Culpa a segunda pelo mau humor,
pela preguiça, pela vontade de não sair da cama.

58% das pessoas dizem que é o pior dia da semana.

Mas e se a segunda estiver sendo acusada do crime errado?

E se o problema dela não estiver na sua cabeça, e sim no seu corpo?

---

### 📚 O Conceito: O Réu Errado

A gente acha que odeia a segunda porque ela estraga o humor.
Só que quando se olham os dados de humor de centenas de milhares de
pessoas, a segunda não é pior que a terça, a quarta ou a quinta.

O que existe não é um "efeito segunda", é um efeito fim de semana:
o sábado e o domingo são tão bons que qualquer dia útil parece cinza
por comparação. Tire a sexta da conta e a tal "Blue Monday" some.

Ou seja: o motivo que a gente dá para temer a segunda — o humor —
quase não se sustenta nos números.

O que se sustenta é outra coisa. Enquanto a cabeça culpa a segunda
por um crime que ela não cometeu, o corpo vive uma segunda-feira bem
mais dura, e em silêncio: mais infarto, mais cortisol, mais desgaste.

A segunda não é um problema de humor. É um problema de fisiologia.

---

### 🧮 O Modelo

Três fontes com dados verificáveis: o que a gente acha, o que o humor
realmente diz, e o que o corpo cobra.

**Fontes:**
- YouGov, 2021 — *Most and least favorite day of the week* —
  N=4.241 adultos dos EUA · 58% apontam a segunda como o pior dia
- Stone, Schneider & Harter, 2012 — *Day-of-week mood patterns in the
  United States* — Journal of Positive Psychology · N=340.000 (Gallup) ·
  humor de segunda não difere do meio de semana (efeito fim de semana)
- Byrne & Laffan, 2023 — STEMI na Irlanda (British Cardiovascular
  Society) — N=10.528 pacientes · +13% de chance de infarto grave na
  segunda
- Chandola et al., 2025 — *Journal of Affective Disorders* — N≈3.500
  idosos · +23% de cortisol capilar em quem sente ansiedade na segunda


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---

# YouGov, 2021 — Most and least favorite day of the week
# N=4.241 adultos (EUA) — AUTORRELATO -> recebe fator ×0.80
n_yougov = 4241
p_crenca_pior_dia = 0.58        # 58% acham a segunda o pior dia da semana

# Stone, Schneider & Harter, 2012 — Journal of Positive Psychology
# N=340.000 (Gallup) — humor de segunda NÃO difere do meio de semana
n_stone = 340000
dif_humor_segunda = 0.00        # diferença de humor segunda vs terça/quarta/quinta
                                # (não significativa; o que existe é efeito fim de semana)

# Byrne & Laffan, 2023 — STEMI, British Cardiovascular Society
# N=10.528 pacientes — medida OBJETIVA (coorte), sem ×0.80
n_stemi = 10528
aumento_infarto_segunda = 0.13  # +13% de chance de infarto grave na segunda

# Chandola et al., 2025 — Journal of Affective Disorders
# N≈3.500 idosos — medida BIOLÓGICA (cortisol capilar), sem ×0.80
n_cortisol = 3500
aumento_cortisol_segunda = 0.23 # +23% de cortisol em quem sente ansiedade na segunda
frac_fisiologica = 0.75         # 75% do efeito vem da amplificação fisiológica
frac_ansiedade = 0.25           # 25% vem de sentir mais ansiedade

# Fator de correção
fator_correcao = 0.80

print("=" * 65)
print("  DADOS DA LITERATURA — A SEGUNDA-FEIRA")
print("=" * 65)
print(f"\n  O QUE A GENTE ACHA (YouGov 2021, N={n_yougov:,}):")
print(f"  → Acham a segunda o pior dia:        {p_crenca_pior_dia*100:.0f}%")
print(f"\n  O QUE O HUMOR DIZ (Stone 2012, N={n_stone:,}):")
print(f"  → Diferença de humor segunda vs      {dif_humor_segunda*100:.0f} pp")
print(f"    meio de semana (não significativa)")
print(f"\n  O QUE O CORPO COBRA:")
print(f"  → Infarto grave na segunda (N={n_stemi:,}): +{aumento_infarto_segunda*100:.0f}%")
print(f"  → Cortisol na segunda (N={n_cortisol:,}):    +{aumento_cortisol_segunda*100:.0f}%")
print(f"    (75% fisiológico, 25% ansiedade)")
print("=" * 65)
```

```python
# --- O MODELO ---
# Qual a probabilidade de você odiar a segunda pelo MOTIVO ERRADO?

# Passo 1 — A crença (autorrelato) com fator de correção
p_crenca_corrigida = p_crenca_pior_dia * fator_correcao

# Passo 2 — O motivo que se dá (a cabeça) não se sustenta
# Com N=340.000, a diferença de humor da segunda é ~0.
# Logo, quem culpa a segunda pelo humor está atribuindo ao réu errado.
motivo_humor_valido = dif_humor_segunda > 0.02   # limiar de relevância prática
premissa_emocional_falsa = not motivo_humor_valido

# Passo 3 — O motivo real (o corpo) é objetivo e mensurável
risco_corpo = np.mean([aumento_infarto_segunda, aumento_cortisol_segunda])

# Passo 4 — Probabilidade de odiar a segunda pelo motivo errado
# = odeia a segunda (crença corrigida) E o motivo declarado (humor) é falso
p_motivo_errado = p_crenca_corrigida * (1.0 if premissa_emocional_falsa else 0.0)

print("=" * 65)
print("  MODELO — ODIAR A SEGUNDA PELO MOTIVO ERRADO")
print("=" * 65)
print(f"\n  P(acha a segunda o pior dia):           {p_crenca_pior_dia*100:.0f}%")
print(f"  Com fator de correção ×0.80:            {p_crenca_corrigida*100:.1f}%")
print(f"\n  Premissa emocional (o humor piora?):    {'FALSA' if premissa_emocional_falsa else 'válida'}")
print(f"  → diferença real de humor: {dif_humor_segunda*100:.0f} pp (N={n_stone:,})")
print(f"\n  → P(odeia a segunda pelo MOTIVO ERRADO): {p_motivo_errado*100:.1f}%")
print(f"\n  O MOTIVO REAL (o corpo):")
print(f"  → +{aumento_infarto_segunda*100:.0f}% de infarto grave na segunda")
print(f"  → +{aumento_cortisol_segunda*100:.0f}% de cortisol na segunda")
print(f"  → risco médio do corpo na segunda:       +{risco_corpo*100:.0f}%")
print(f"\n  PARADOXO CENTRAL:")
print(f"  → A cabeça culpa a segunda por um crime que ela não cometeu.")
print(f"  → O corpo vive a segunda de verdade — e cobra em silêncio.")
print("=" * 65)
```

```python
# --- VISUALIZAÇÃO — GRÁFICOS SEPARADOS ---

# ── GRÁFICO 1 — O mito: a cabeça culpa, o humor não confirma ──
fig1, ax1 = plt.subplots(figsize=(10, 6))

categorias = ['Acham a segunda\no pior dia\n(YouGov, N=4.241)',
              'Diferença real de humor:\nsegunda vs meio de semana\n(Stone, N=340.000)']
valores = [58, 0]
cores = ['#c0392b', '#2a8a82']

bars = ax1.bar(categorias, valores, color=cores, alpha=0.85, width=0.45)
ax1.set_ylim(0, 68)
ax1.set_ylabel('Proporção / diferença (%)')
ax1.set_title('O Mito da Segunda: a cabeça culpa, o humor não confirma\nYouGov 2021 · Stone, Schneider & Harter 2012',
              fontsize=13, pad=15)

ax1.text(0, 58 + 1.2, '58%', ha='center', fontweight='bold', fontsize=17)
ax1.text(1, 2.0, '≈ 0 pp\n(não significativo)', ha='center',
         fontweight='bold', fontsize=12, color='#2a8a82')

plt.figtext(0.5, 0.01,
            'Fonte: YouGov 2021 (N=4.241) · Stone et al. 2012 (N=340.000) | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-058-grafico-01-mito.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 — O motivo real: o corpo cobra em silêncio ──
fig2, ax2 = plt.subplots(figsize=(10, 6))

categorias2 = ['Infarto grave (STEMI)\nna segunda\n(Byrne, N=10.528)',
               'Cortisol elevado\nna segunda\n(Chandola, N≈3.500)']
valores2 = [13, 23]
cores2 = ['#c0392b', '#c8a84b']

bars2 = ax2.bar(categorias2, valores2, color=cores2, alpha=0.85, width=0.45)
ax2.set_ylim(0, 28)
ax2.set_ylabel('Aumento na segunda-feira vs outros dias (%)')
ax2.set_title('O Motivo Real: o corpo vive a segunda de verdade\nByrne & Laffan 2023 · Chandola et al. 2025',
              fontsize=13, pad=15)

for bar, valor in zip(bars2, valores2):
    ax2.text(bar.get_x() + bar.get_width()/2, valor + 0.5,
             f'+{valor}%', ha='center', fontweight='bold', fontsize=16)

plt.figtext(0.5, 0.01,
            'Fonte: Byrne & Laffan 2023 (STEMI, N=10.528) · Chandola et al. 2025 (N≈3.500) | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-058-grafico-02-corpo.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 — A Assinatura Estatística: Distribuição Beta da crença ──
fig3, ax3 = plt.subplots(figsize=(10, 6))

alpha_yg = p_crenca_pior_dia * n_yougov
beta_yg = (1 - p_crenca_pior_dia) * n_yougov
dist_yg = stats.beta(alpha_yg, beta_yg)
ic_yg = dist_yg.interval(0.95)

x = np.linspace(ic_yg[0] * 0.985, ic_yg[1] * 1.015, 1000)
y = dist_yg.pdf(x)

ax3.plot(x * 100, y, color='#c0392b', linewidth=2.5)
ax3.fill_between(x * 100, y, alpha=0.2, color='#c0392b')
ax3.axvline(x=ic_yg[0] * 100, color='#c8a84b', linestyle='--', linewidth=2, alpha=0.9)
ax3.axvline(x=ic_yg[1] * 100, color='#c8a84b', linestyle='--', linewidth=2, alpha=0.9)
ax3.axvline(x=p_crenca_pior_dia * 100, color='#333', linewidth=2)

ax3.text(p_crenca_pior_dia * 100, max(y) * 0.55,
         f'  IC 95%\n  [{ic_yg[0]*100:.1f}%, {ic_yg[1]*100:.1f}%]',
         fontsize=11, color='#c8a84b', fontweight='bold')

ax3.set_xlabel('Proporção que aponta a segunda como o pior dia (%)')
ax3.set_ylabel('Densidade')
ax3.set_title('A Assinatura Estatística — Distribuição Beta da crença\nYouGov 2021 (N=4.241)',
              fontsize=13, pad=15)

plt.figtext(0.5, 0.01,
            'Fonte: YouGov 2021 (N=4.241) — proporção autorrelatada | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-058-grafico-03-bayesiano.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**58% das pessoas acham a segunda o pior dia da semana.**
**Com correção estatística, cerca de 46% carregam esse medo — pelo motivo errado.**

A gente jura que odeia a segunda porque ela estraga o humor.
Mas o humor de segunda, medido em 340.000 pessoas, não é pior
que o de terça, quarta ou quinta. A "Blue Monday" é quase só
saudade do fim de semana.

O que a segunda cobra de verdade não aparece no seu humor.
Aparece no seu corpo: **+13% de infarto grave** e **+23% de cortisol**.

Enquanto a cabeça reclama de um crime que a segunda não cometeu,
o corpo vive a segunda real, e paga a conta em silêncio.

Talvez o problema nunca tenha sido gostar ou não gostar da segunda.
Talvez seja aprender a atravessá-la com menos desgaste.

A pergunta que fica:

*Da próxima segunda que te pesar, o que está reclamando: a sua
cabeça, ou o seu corpo?*

---

### ⚠️ Limitações do Modelo

- Os 58% (YouGov 2021) são autorrelato; por isso o fator ×0.80.
  O ×0.80 foi aplicado apenas a essa proporção de crença, e não
  aos dados de infarto (coorte) nem de cortisol (medida biológica).
- Stone et al. 2012 mede humor médio populacional; não exclui que
  indivíduos específicos tenham piora real de humor na segunda.
- Byrne & Laffan 2023 (STEMI) é associação retrospectiva, não causa;
  o padrão pode variar por país e sistema de saúde.
- Chandola et al. 2025 mediu cortisol capilar em idosos; +23% é
  aumento relativo, não proporção, e a amostra não generaliza direto
  para todas as idades.
- O pico de infarto na segunda tem causas múltiplas (retomada de
  atividade, cronobiologia), não apenas o "ódio à segunda".

*A ciência é honesta sobre o que não sabe. O modelo também.*
