### 📖 A História

Semana passada, um sorteio mandou metade de um call center para
casa, e quem foi rendeu 13% a mais. Hoje o assunto é o escritório
que sobrou.

Especificamente, o escritório aberto: aquele em que as paredes
caíram para "aproximar as pessoas". A promessa tem vinte anos e
uma lógica sedutora: sem divisórias, mais encontros; mais
encontros, mais colaboração. Empresas inteiras foram reformadas
em cima dessa frase.

Que ninguém tinha testado.

Até que dois pesquisadores de Harvard penduraram um crachá-sensor
no pescoço de 152 pessoas em duas multinacionais que estavam
exatamente naquele momento: derrubando as divisórias da sede.
O crachá registrava quem estava de frente para quem, quem falava
e quem ouvia, sem gravar conteúdo. Os servidores entregaram os
e-mails e as mensagens. Antes e depois da reforma.

Resultado: com as paredes no chão, a conversa cara a cara
**caiu cerca de 70%**. De 5,8 horas por pessoa por dia
para 1,7.

As pessoas não se aproximaram. Elas colocaram fones de ouvido,
construíram as paredes que a empresa tirou, e mudaram a conversa
para o texto.

---

### 📚 O Conceito: A Retração Social

O achado tem nome no artigo: em vez de colaboração vibrante, a
arquitetura aberta disparou *"uma resposta humana natural de
retração social"*.

A mecânica é quase óbvia depois de dita: privacidade não é
capricho, é a condição para certos tipos de conversa. O feedback
delicado, a dúvida que expõe, a negociação. Quando todo mundo vê
e ouve todo mundo, essas conversas não acontecem em voz alta.
Migram para onde há paredes de novo: a caixa de entrada.

E a migração rebaixa o canal. A voz carrega tom, hesitação,
correção em tempo real. O texto carrega o que sobrou.

O parentesco de método com o Dia #087 é declarado e proposital:
lá, um sorteio; aqui, sensores. As duas pontas do par "onde você
trabalha" foram respondidas do mesmo jeito, **medindo em vez de
perguntando**. E as respostas, lado a lado, formam uma ironia:
a casa rendeu mais, e o escritório desenhado para conversar
conversou 70% menos.

### 🧮 O Modelo

**Fontes:**
- Bernstein, E. S. & Turban, S. (2018) - The impact of the 'open'
  workspace on human collaboration - Philosophical Transactions
  of the Royal Society B, 373(1753), 20170239 - dois estudos de
  campo em sedes Fortune 500, crachás sociométricos
  (infravermelho, microfone sem conteúdo, acelerômetro,
  Bluetooth) + logs de e-mail e mensageiro dos servidores.
  **Estudo 1: N=52**, 3 semanas antes + 3 depois: conversa cara a
  cara de **5,8 para 1,7 horas/dia** (queda de ~71%); e-mail
  **+56%**; mensagens instantâneas **+67%**. **Estudo 2: N=100**
  (~45% do andar), 8 semanas antes + 8 depois, iniciando 2 meses
  após a reforma: queda de ~70% replicada; e-mail +22% a +50%.
  Primeiro estudo a medir, e não perguntar. **Fonte central.**

**Fator ×0,80: não se aplica.** Sensores e logs de servidor;
nenhum autorrelato.

**Contagem por pares (díades), do próprio artigo:** no Estudo 1,
643 díades conversaram menos cara a cara depois da reforma e 141
conversaram mais (784 pares com interação medida); no e-mail, 222
díades diminuíram e 374 aumentaram (596 pares). No Estudo 2, 1.830
díades acompanhadas.

**O modelo, em duas camadas (semente fixa, `seed=42`):**

1. **Bayesiana, sobre contagem real.** Beta com prior de Jeffreys
   sobre as díades: a probabilidade de um par de colegas conversar
   menos depois que as paredes caem, e de trocar mais e-mail, com
   IC 95%.
2. **Simulação.** Um escritório de 52 pessoas: para cada colega seu,
   sorteio da posterior se a conversa entre vocês cai ou não. Sai a
   distribuição de "com quantos dos seus 51 colegas você passa a
   conversar menos". É um "e se" com parâmetros medidos, não uma
   auditoria de ninguém.

**A conta de bolso (suposições declaradas):** o orçamento anual de
conversa. As 4,1 horas diárias evaporadas são projetadas para o
ano com sensibilidade em dias úteis (200, 220 e 240), e
traduzidas em expediente semanal. Projeção linear declarada:
assume que a queda medida em semanas se sustenta no ano, o que os
dados de 16 semanas do Estudo 2 sugerem mas não garantem.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Bernstein & Turban (2018), Phil. Trans. R. Soc. B 373(1753)

# Estudo 1 (OpenCo1)
n1 = 52
h_antes, h_depois = 5.8, 1.7        # horas de conversa cara a cara / dia
queda_f2f_1 = (h_antes - h_depois) / h_antes
email_delta_1 = 0.56                 # +56% de e-mails
im_delta_1 = 0.67                    # +67% de mensagens instantâneas

# Contagem por díades (pares de colegas), Estudo 1
diades_f2f_caiu, diades_f2f_subiu = 643, 141     # 784 pares com interação medida
diades_email_caiu, diades_email_subiu = 222, 374 # 596 pares com e-mail
n_diades_2 = 1830                                # Estudo 2: pares acompanhados

# Estudo 2 (OpenCo2)
n2 = 100                             # ~45% do andar
queda_f2f_2 = 0.70                   # ~70%, replicada
email_delta_2 = (0.22, 0.50)         # +22% a +50%
# desenho: 8 semanas antes + 8 depois, começando 2 meses após a
# reforma (mata a objeção do "período de adaptação")

# Produtividade: queda QUALITATIVA relatada por executivos
# (declarada como relato, não como medida)

print("=" * 65)
print("  DADOS DA LITERATURA - O ESCRITÓRIO ABERTO, MEDIDO")
print("=" * 65)
print(f"\n  Estudo 1 (N={n1}, 3+3 semanas, crachás sociométricos):")
print(f"  → Conversa cara a cara:  {h_antes} → {h_depois} horas/dia  ({queda_f2f_1*100:.0f}% de queda)")
print(f"  → E-mail:                +{email_delta_1*100:.0f}%")
print(f"  → Mensagens (IM):        +{im_delta_1*100:.0f}%")
print(f"  → Díades: {diades_f2f_caiu} conversaram menos, {diades_f2f_subiu} mais;")
print(f"            {diades_email_subiu} trocaram mais e-mail, {diades_email_caiu} menos")
print(f"\n  Estudo 2 (N={n2}, 8+8 semanas, início 2 meses após a reforma):")
print(f"  → Queda da conversa:     ~{queda_f2f_2*100:.0f}%  (replicada)")
print(f"  → E-mail:                +{email_delta_2[0]*100:.0f}% a +{email_delta_2[1]*100:.0f}%")
print(f"\n  Produtividade: queda relatada por executivos (qualitativa)")
print(f"\n  Fator ×0,80: NÃO aplicado (sensores e logs de servidor)")
print("=" * 65)
```

```python
# --- O MODELO ---
# Duas camadas: (1) bayesiana sobre as díades; (2) simulação do seu escritório
# Depois, a conta de bolso do orçamento anual de conversa

rng = np.random.default_rng(42)
N_SIM = 100_000

# ────────────────────────────────────────────────────────────────
# CAMADA 1 - BAYESIANA: um par de colegas conversa menos depois da reforma?
# Contagem real do artigo, prior de Jeffreys Beta(0,5 · 0,5)
# ────────────────────────────────────────────────────────────────
n_diades_f2f = diades_f2f_caiu + diades_f2f_subiu          # 784
post_f2f = stats.beta(0.5 + diades_f2f_caiu, 0.5 + diades_f2f_subiu)
ic_f2f = post_f2f.interval(0.95)
p_f2f_caiu = post_f2f.mean()

n_diades_email = diades_email_caiu + diades_email_subiu    # 596
post_email = stats.beta(0.5 + diades_email_subiu, 0.5 + diades_email_caiu)
ic_email = post_email.interval(0.95)
p_email_subiu = post_email.mean()

# A "troca de canal" num par: P(conversa cai) × P(e-mail sobe), assumindo
# independência entre os dois eventos (suposição declarada: o artigo não
# publica a tabela cruzada por díade)
amostra_f2f = post_f2f.rvs(N_SIM, random_state=rng)
amostra_email = post_email.rvs(N_SIM, random_state=rng)
troca = amostra_f2f * amostra_email
ic_troca = np.percentile(troca, [2.5, 97.5])

# ────────────────────────────────────────────────────────────────
# CAMADA 2 - SIMULAÇÃO: o seu escritório de 52 pessoas
# Para cada colega com quem você conversava, a conversa cai com probabilidade
# sorteada da posterior. 100.000 escritórios.
# ────────────────────────────────────────────────────────────────
colegas = round(2 * n_diades_f2f / n1)                    # ≈30: colegas com quem havia conversa medida
p_sorteado = post_f2f.rvs(N_SIM, random_state=rng)
colegas_menos = rng.binomial(colegas, p_sorteado)          # com quantos você conversa menos
ic_colegas = np.percentile(colegas_menos, [2.5, 97.5])
media_colegas = colegas_menos.mean()
p_pelo_menos_24 = np.mean(colegas_menos >= 24)          # 80% dos 30
p_pelo_menos_27 = np.mean(colegas_menos >= 27)          # 90% dos 30
p_metade_ou_menos = np.mean(colegas_menos <= colegas / 2)

# ────────────────────────────────────────────────────────────────
# A CONTA DE BOLSO: o orçamento anual de conversa (projeção declarada)
# ────────────────────────────────────────────────────────────────
horas_perdidas_dia = h_antes - h_depois          # 4,1 horas/dia
horas_perdidas_semana = horas_perdidas_dia * 5   # 20,5 horas/semana
dias_uteis = np.array([200, 220, 240])           # sensibilidade
horas_perdidas_ano = horas_perdidas_dia * dias_uteis
fracao_expediente = horas_perdidas_semana / 40

print("=" * 65)
print("  MODELO - CAMADA 1: BAYES SOBRE AS DÍADES")
print("=" * 65)
print(f"\n  Conversa cara a cara ({n_diades_f2f} pares com interação medida):")
print(f"  → P(um par de colegas conversa MENOS após a reforma) = {p_f2f_caiu*100:.1f}%")
print(f"     IC 95%: [{ic_f2f[0]*100:.1f}%, {ic_f2f[1]*100:.1f}%]")
print(f"\n  E-mail ({n_diades_email} pares com troca medida):")
print(f"  → P(um par de colegas troca MAIS e-mail após a reforma) = {p_email_subiu*100:.1f}%")
print(f"     IC 95%: [{ic_email[0]*100:.1f}%, {ic_email[1]*100:.1f}%]")
print(f"\n  A retração social num par (independência assumida, declarada):")
print(f"  → P(conversa cai E e-mail sobe) = {troca.mean()*100:.1f}%")
print(f"     IC 95%: [{ic_troca[0]*100:.1f}%, {ic_troca[1]*100:.1f}%]")

print("\n" + "=" * 65)
print("  MODELO - CAMADA 2: O SEU ESCRITÓRIO (seed 42, 100.000 sorteios)")
print("=" * 65)
print(f"\n  Escritório de {n1} pessoas; você conversava com ~{colegas} delas. Paredes no chão:")
print(f"  → Com quantos você passa a conversar MENOS:  {media_colegas:.0f} de {colegas}")
print(f"     IC 95%: [{ic_colegas[0]:.0f}, {ic_colegas[1]:.0f}]")
print(f"  → P(pelo menos 24 dos {colegas}):   {p_pelo_menos_24*100:.0f}%")
print(f"  → P(pelo menos 27 dos {colegas}):   {p_pelo_menos_27*100:.0f}%")
print(f"  → P(metade ou menos):          {p_metade_ou_menos*100:.2f}%")
print(f"  → A reforma que prometia aproximar afastou quatro em cada cinco pares.")

print("\n" + "=" * 65)
print("  A CONTA DE BOLSO: O ORÇAMENTO ANUAL DE CONVERSA")
print("=" * 65)
print(f"\n  Conversa evaporada por pessoa: {horas_perdidas_dia:.1f} horas/dia")
print(f"  → {horas_perdidas_semana:.1f} horas/semana ({fracao_expediente*100:.0f}% de um expediente de 40h)")
print(f"\n  No ano (sensibilidade em dias úteis):")
for d, h in zip(dias_uteis, horas_perdidas_ano):
    print(f"  → {d} dias úteis:  {h:,.0f} horas de conversa/ano".replace(',', '.'))
print(f"\n  A TROCA QUE NÃO FECHA:")
print(f"  → Voz:   -{queda_f2f_1*100:.0f}%")
print(f"  → Texto: +{email_delta_1*100:.0f}% (e-mail) e +{im_delta_1*100:.0f}% (IM)")
print(f"  → Direções opostas, moedas diferentes: minutos de voz não")
print(f"     viram mensagens na mesma taxa, e o canal rebaixa.")
print(f"\n  A LEITURA:")
print(f"  → A empresa derrubou paredes para comprar conversa")
print(f"     e recebeu ~900 horas a menos dela, por pessoa, por ano.")
print(f"  → As pessoas reconstruíram as paredes: de fone e de texto.")
print("=" * 65)
```

```python
# --- VISUALIZAÇÃO - GRÁFICOS SEPARADOS ---

# ── GRÁFICO 1 - As horas por dia ──
fig1, ax1 = plt.subplots(figsize=(10, 6))

grupos = ['Antes\n(divisórias)', 'Depois\n(aberto)']
vals = [h_antes, h_depois]
bars = ax1.bar(grupos, vals, color=['#c9a14a', '#c0392b'], alpha=0.88, width=0.42)
ax1.set_ylabel('Conversa cara a cara (horas por pessoa/dia)')
ax1.set_ylim(0, 7)
ax1.set_title('As paredes caíram, a conversa foi junto\n'
              'Estudo 1: N=52, crachás sociométricos (Bernstein & Turban, 2018)',
              fontsize=12.5, pad=15)

for bar, v in zip(bars, vals):
    ax1.text(bar.get_x() + bar.get_width()/2, v + 0.12,
             f'{v}h', ha='center', fontweight='bold', fontsize=17)

ax1.annotate('', xy=(1, 1.9), xytext=(0, 5.6),
             arrowprops=dict(arrowstyle='->', color='#c0392b', lw=2))
ax1.text(0.5, 4.1, f'-{queda_f2f_1*100:.0f}%', ha='center',
         fontsize=16, color='#c0392b', fontweight='bold')

plt.figtext(0.5, 0.01,
            'Fonte: Bernstein & Turban (2018), Phil. Trans. R. Soc. B | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-093-grafico-01-horas.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 - A troca: voz cai, texto sobe ──
fig2, ax2 = plt.subplots(figsize=(10, 6))

canais = ['Conversa\ncara a cara', 'E-mail', 'Mensagens\ninstantâneas']
deltas = [-queda_f2f_1*100, email_delta_1*100, im_delta_1*100]
cores2 = ['#c0392b', '#2a8a82', '#2a8a82']

bars2 = ax2.bar(canais, deltas, color=cores2, alpha=0.88, width=0.45)
ax2.axhline(0, color='#333', linewidth=1.2)
ax2.set_ylabel('Variação após a reforma (%)')
ax2.set_ylim(-92, 92)
ax2.set_title('A retração social: a conversa migrou para o texto\n'
              'Estudo 2 (N=100) replica: queda de ~70%, e-mail +22% a +50%',
              fontsize=12.5, pad=15)

for bar, v in zip(bars2, deltas):
    va = 'bottom' if v > 0 else 'top'
    off = 3 if v > 0 else -3
    ax2.text(bar.get_x() + bar.get_width()/2, v + off,
             f'{v:+.0f}%', ha='center', va=va, fontweight='bold', fontsize=15)

ax2.text(0.98, 0.06, 'direções opostas, moedas diferentes:\na troca não fecha em conteúdo',
         transform=ax2.transAxes, ha='right', fontsize=10, color='#555', style='italic')

plt.figtext(0.5, 0.01,
            'Fonte: Bernstein & Turban (2018) | logs de servidor, sem autorrelato | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-093-grafico-02-troca.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 - Bayes e simulação: as díades e o seu escritório ──
fig3, (ax3a, ax3b) = plt.subplots(1, 2, figsize=(12, 5.5))

# Painel A: posteriores por díade
x = np.linspace(0.50, 0.95, 1000)
ax3a.plot(x*100, post_f2f.pdf(x), color='#c0392b', linewidth=2.5,
          label=f'Conversa cai: {p_f2f_caiu*100:.0f}% [{ic_f2f[0]*100:.0f}, {ic_f2f[1]*100:.0f}]')
ax3a.fill_between(x*100, post_f2f.pdf(x), alpha=0.18, color='#c0392b')
ax3a.plot(x*100, post_email.pdf(x), color='#2a8a82', linewidth=2.5,
          label=f'E-mail sobe: {p_email_subiu*100:.0f}% [{ic_email[0]*100:.0f}, {ic_email[1]*100:.0f}]')
ax3a.fill_between(x*100, post_email.pdf(x), alpha=0.18, color='#2a8a82')
ax3a.axvline(50, color='#333', linewidth=1.2, linestyle=':')
ax3a.text(50.8, ax3a.get_ylim()[1]*0.02, 'moeda', fontsize=9, color='#555')
ax3a.set_xlabel('Probabilidade para um par de colegas (%)')
ax3a.set_ylabel('Densidade a posteriori')
ax3a.set_yticks([])
ax3a.legend(frameon=False, fontsize=9.5, loc='upper left')
ax3a.set_title(f'Por par de colegas: {n_diades_f2f} díades (voz), {n_diades_email} (e-mail)',
               fontsize=11.5)

# Painel B: o seu escritório
vals, counts = np.unique(colegas_menos, return_counts=True)
ax3b.bar(vals, counts / N_SIM, color='#c9a14a', alpha=0.9, width=0.85)
ax3b.axvline(colegas / 2, color='#333', linewidth=1.2, linestyle=':')
ax3b.text(colegas/2 + 0.4, ax3b.get_ylim()[1]*0.45, 'metade', fontsize=9, color='#555')
ax3b.text(0.03, 0.95,
          f'média {media_colegas:.0f} de {colegas}\nIC 95% [{ic_colegas[0]:.0f}, {ic_colegas[1]:.0f}]\n'
          f'P(≥ 24) = {p_pelo_menos_24*100:.0f}%\nP(≤ metade) = {p_metade_ou_menos*100:.2f}%',
          transform=ax3b.transAxes, va='top', fontsize=10, fontfamily='monospace',
          bbox=dict(boxstyle='round', facecolor='white', edgecolor='#ccc'))
ax3b.set_xlabel(f'Colegas (de ~{colegas}) com quem você passa a conversar menos')
ax3b.set_xticks(range(10, colegas + 1, 5))
ax3b.set_yticks([])
ax3b.set_xlim(10, colegas + 1)
ax3b.set_title('O seu escritório: 100.000 sorteios da posterior', fontsize=11.5)

fig3.suptitle('A retração social, par a par: quatro em cada cinco duplas conversam menos\n'
              'Beta com prior de Jeffreys sobre as díades do Estudo 1 · Monte Carlo com seed 42',
              fontsize=12.5, y=1.02)
plt.figtext(0.5, -0.02,
            'Fonte: Bernstein & Turban (2018) | contagem por díade do próprio artigo | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-093-grafico-03-diades.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")

# ── GRÁFICO 4 - O orçamento anual de conversa ──
fig4, ax4 = plt.subplots(figsize=(10, 6))

labels3 = [f'{d} dias úteis' for d in dias_uteis]
bars3 = ax4.bar(labels3, horas_perdidas_ano, color='#c9a14a', alpha=0.88, width=0.45)
ax4.set_ylabel('Horas de conversa perdidas por pessoa/ano')
ax4.set_ylim(0, 1150)
ax4.set_title('O orçamento anual de conversa (conta de bolso, projeção declarada)\n'
              f'{horas_perdidas_dia:.1f}h/dia · {horas_perdidas_semana:.1f}h/semana: meio expediente semanal evaporou',
              fontsize=12, pad=15)

for bar, v in zip(bars3, horas_perdidas_ano):
    ax4.text(bar.get_x() + bar.get_width()/2, v + 22,
             f'{v:,.0f}h'.replace(',','.'), ha='center', fontweight='bold', fontsize=14)

plt.figtext(0.5, 0.01,
            'Modelo próprio sobre as horas do Estudo 1 | projeção linear com sensibilidade | #365Probabilidades',
            ha='center', fontsize=8.5, color='gray')
plt.tight_layout()
plt.savefig('dia-093-grafico-04-orcamento.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 4 salvo!")
```

### 💡 O Insight

**Derrubaram as paredes para comprar conversa. A conversa caiu 70%.**

De 5,8 horas por pessoa por dia para 1,7.

Par a par, com a contagem do próprio artigo: **82% das duplas de
colegas passaram a conversar menos**, com intervalo de 79% a 85%.
Num escritório de 52 pessoas, dos cerca de 30 colegas com quem você
de fato conversava, a simulação diz que você passa a conversar menos
com **25**, entre 20 e 28. A chance de a reforma ter poupado metade
das suas conversas é zero em 100.000 sorteios.

E a conta de bolso, com projeção declarada: **cerca de 900 horas de
conversa por pessoa por ano**, meio expediente semanal, evaporadas
na reforma que prometia o contrário.

E não foi para o silêncio: foi para o texto. E-mail +56%,
mensagens +67%. As pessoas fizeram o que pessoas fazem quando a
privacidade acaba: reconstruíram as paredes com fones de ouvido e
caixa de entrada. O artigo chama de retração social. Qualquer um
que já trabalhou num aberto chama de terça-feira.

A ironia do par desta semana, com o Dia #087, fica registrada:
medindo em vez de perguntando, a casa rendeu 13% a mais, e o
escritório desenhado para conversar conversou 70% menos.
O problema nunca foi onde as pessoas trabalham. É o que o lugar
faz com elas.

As honestidades: duas empresas, 152 pessoas, sem grupo de
controle, e a queda de produtividade é relato de executivos, não
medida. O que os sensores garantem é a interação. O resto é
inferência, e fica declarado.

### ⚠️ Limitações do Modelo

- Duas empresas, N total de 152: generalização setorial limitada
  e declarada
- Desenho pré/pós no mesmo grupo, sem grupo de controle
  randomizado; o Estudo 2 mitiga o período de adaptação (inicia
  2 meses após a reforma), mas não elimina tendências temporais
- Interação não é produtividade: a queda de produtividade é
  relato qualitativo de executivos, declarado como relato
- Voz medida em minutos, texto em mensagens: moedas diferentes;
  o modelo afirma a direção da troca, não equivalência de conteúdo
- A Beta usa as díades com interação medida (784 de 1.326 pares
  possíveis no Estudo 1); pares que nunca conversaram ficam fora
  por construção. Díades não são totalmente independentes (cada
  pessoa está em várias), o que estreita o IC um pouco além do justo
- A "troca de canal" num par assume independência entre queda de
  voz e alta de e-mail; o artigo não publica a tabela cruzada
- A simulação do escritório é um "e se" sobre a posterior: mede uma
  propriedade do modelo, não uma empresa real
- O orçamento anual é projeção linear própria sobre 6 a 16
  semanas de medição, com sensibilidade em dias úteis; assume
  persistência da queda
- Par declarado com o Dia #087 (Bloom et al., 2015): mesmo eixo
  "onde você trabalha", fontes e desenhos independentes

*A ciência é honesta sobre o que não sabe. O modelo também.*
