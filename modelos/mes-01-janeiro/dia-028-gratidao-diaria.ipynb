### 📖 A História

De uns tempos pra cá, comecei a escrever antes de dormir.

Não é bem um diário. É mais um registro de sentimentos e sensações do dia — e, junto, uma lista pequena de motivos pra agradecer.

Foi num ritmo de vida tenso, dos que passam correndo. E é justamente aí que a coisa me surpreendeu: quando você senta pra escrever, o dia parece se fixar. O que passaria despercebido no caos, de repente, fica.

Eu não tinha ideia de quantos motivos pra agradecer cabiam num dia comum.

Buscar o Edgar na escola e, no meio de um turbilhão de coisas na cabeça, ouvir as histórias engraçadas que ele conta no banco de trás. Coisas pequenas. Que numa semana corrida eu nem registraria — e que, escritas, viram âncora.

Sem saber, eu tinha começado a fazer quase exatamente o que dois pesquisadores testaram num experimento clássico. E o que eles descobriram tem um detalhe que muda tudo: não é a gratidão em si que faz diferença. É a frequência dela.

---

### 📚 O Conceito: Contar Bênçãos, Diariamente

Em 2003, Robert Emmons e Michael McCullough publicaram um dos estudos mais influentes da psicologia positiva: "Counting Blessings Versus Burdens".

A pergunta era direta: listar motivos de gratidão, de forma deliberada, melhora mesmo o bem-estar — ou isso é só sabedoria de livro de autoajuda?

Eles fizeram três experimentos, sempre com o mesmo desenho básico: dividir pessoas aleatoriamente em grupos, e pedir que umas registrassem motivos de gratidão, enquanto outras registrassem aborrecimentos ou eventos neutros. Depois, mediam humor, sintomas físicos, sono, avaliação de vida.

A descoberta mais interessante não foi *que* a gratidão funciona. Foi *quando* ela funciona. E a resposta tem a ver com frequência — exatamente o ponto em que a minha escrita da noite se encaixou sem eu perceber.

---

### 🧮 O Modelo

Os três estudos de Emmons & McCullough (2003) diferem numa variável decisiva: a frequência do registro. Ao alinhá-los, aparece um padrão claro — o efeito sobre o afeto positivo só emerge quando a prática passa de semanal para diária.

Aqui os números são resultados diretos de um experimento controlado (médias de fator de afeto e tamanhos de efeito reais do paper), não proporções extrapoladas para a população. Por isso o fator de correção ×0.80 do projeto não se aplica neste dia.

**Fontes:**
- Emmons, R. A. & McCullough, M. E. (2003) — "Counting Blessings Versus Burdens: An Experimental Investigation of Gratitude and Subjective Well-Being in Daily Life", Journal of Personality and Social Psychology, 84(2), 377-389
  - Estudo 1: registro semanal, N=192 (universitários)
  - Estudo 2: registro diário, N=157 (universitários)
  - Estudo 3: registro diário por 21 dias, N=65 (adultos com doença neuromuscular)
- Froh, J. J. et al. (2008) — replicação com adolescentes, Journal of School Psychology, 46, 213-233


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("✅ Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Emmons & McCullough (2003), JPSP 84(2):377-389
# Fator de AFETO POSITIVO por estudo (gratidão vs. comparação), com frequência do registro

# Estudo 1 — SEMANAL (N=192): afeto positivo NÃO significativo
e1_gratidao = 0.18
e1_controle = -0.03   # grupo "eventos neutros"
e1_sig = False
e1_F = 1.73

# Estudo 2 — DIÁRIO (N=157): afeto positivo significativo
e2_gratidao = 0.24
e2_controle = -0.26   # grupo "aborrecimentos"
e2_sig = True
e2_F = 3.28

# Estudo 3 — DIÁRIO 21 dias (N=65): afeto positivo significativo (mais robusto)
e3_gratidao = 0.35
e3_controle = -0.25   # grupo controle puro
e3_sig = True
e3_F = 5.18
e3_d = 0.56           # Cohen's d

# Bônus do Estudo 3
e3_sono_gratidao = 7.58
e3_sono_controle = 7.06

print("=" * 65)
print("  DADOS — GRATIDÃO E AFETO POSITIVO (Emmons & McCullough 2003)")
print("=" * 65)
print(f"\n  ESTUDO 1 — registro SEMANAL (N=192)")
print(f"  → Gratidão: {e1_gratidao:+.2f}   Controle: {e1_controle:+.2f}")
print(f"  → Afeto positivo significativo? {'SIM' if e1_sig else 'NÃO'} (F={e1_F})")
print(f"\n  ESTUDO 2 — registro DIÁRIO (N=157)")
print(f"  → Gratidão: {e2_gratidao:+.2f}   Aborrecimentos: {e2_controle:+.2f}")
print(f"  → Afeto positivo significativo? {'SIM' if e2_sig else 'NÃO'} (F={e2_F}, p<.05)")
print(f"\n  ESTUDO 3 — registro DIÁRIO 21 dias (N=65)")
print(f"  → Gratidão: {e3_gratidao:+.2f}   Controle: {e3_controle:+.2f}")
print(f"  → Afeto positivo significativo? {'SIM' if e3_sig else 'NÃO'} (F={e3_F}, p=.026, d={e3_d})")
print(f"  → Bônus: sono {e3_sono_gratidao}h vs {e3_sono_controle}h")
print("=" * 65)
```

```python
# --- O MODELO ---
# A dose importa: o efeito no afeto positivo surge quando o registro vira diário

gap_e1 = e1_gratidao - e1_controle
gap_e2 = e2_gratidao - e2_controle
gap_e3 = e3_gratidao - e3_controle

print("=" * 65)
print("  MODELO — SEMANAL vs DIÁRIO")
print("=" * 65)
print(f"\n  Diferença gratidão − comparação, no fator de afeto positivo:")
print(f"  → Estudo 1 (SEMANAL):  {gap_e1:+.2f}   → não significativo")
print(f"  → Estudo 2 (DIÁRIO):   {gap_e2:+.2f}   → significativo")
print(f"  → Estudo 3 (DIÁRIO):   {gap_e3:+.2f}   → significativo (d={e3_d})")
print(f"\n  LEITURA:")
print(f"  → A mesma intervenção (listar motivos de gratidão)")
print(f"     quase não moveu o afeto positivo quando era SEMANAL,")
print(f"     mas moveu de forma consistente quando virou DIÁRIA.")
print(f"  → Não é o gesto isolado. É a frequência.")
print(f"\n  Sono no Estudo 3 (efeito colateral positivo):")
print(f"  → +{(e3_sono_gratidao - e3_sono_controle)*60:.0f} minutos de sono por noite, em média")
print("=" * 65)
```

```python
# --- VISUALIZAÇÃO ---

cor_grat = '#2a8a82'
cor_ctrl = '#c0392b'
cor_ouro = '#c8a84b'

# ── GRÁFICO 1 — Afeto positivo: gratidão vs comparação, por frequência ──
fig1, ax1 = plt.subplots(figsize=(11, 6))

estudos = ['Estudo 1\n(SEMANAL)', 'Estudo 2\n(DIÁRIO)', 'Estudo 3\n(DIÁRIO, 21 dias)']
grat = [e1_gratidao, e2_gratidao, e3_gratidao]
ctrl = [e1_controle, e2_controle, e3_controle]

x = np.arange(len(estudos))
largura = 0.35
b1 = ax1.bar(x - largura/2, grat, largura, label='Grupo gratidão', color=cor_grat, alpha=0.85)
b2 = ax1.bar(x + largura/2, ctrl, largura, label='Grupo comparação', color=cor_ctrl, alpha=0.85)

ax1.axhline(y=0, color='#333', linewidth=1)
ax1.set_ylabel('Fator de afeto positivo (escore padronizado)')
ax1.set_xticks(x)
ax1.set_xticklabels(estudos)
ax1.set_title('Gratidão e afeto positivo: a diferença aparece no registro diário\nEmmons & McCullough (2003)',
              fontsize=13, pad=15)
ax1.legend(frameon=False, loc='upper right')

# marcar não-significância do estudo 1
ax1.text(0, 0.30, 'n.s.', ha='center', fontsize=12, color='gray', style='italic')
ax1.text(1, 0.34, 'p<.05', ha='center', fontsize=11, color=cor_grat, fontweight='bold')
ax1.text(2, 0.45, 'p=.026\nd=0,56', ha='center', fontsize=11, color=cor_grat, fontweight='bold')

plt.figtext(0.5, 0.005,
            'Fonte: Emmons & McCullough (2003), JPSP 84(2) | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-028-grafico-01-frequencia.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 1 salvo!")

# ── GRÁFICO 2 — O tamanho do gap (gratidão − comparação) por frequência ──
fig2, ax2 = plt.subplots(figsize=(10, 6))

labels = ['Estudo 1\nSEMANAL', 'Estudo 2\nDIÁRIO', 'Estudo 3\nDIÁRIO 21d']
gaps = [e1_gratidao - e1_controle, e2_gratidao - e2_controle, e3_gratidao - e3_controle]
cores = ['#b0b0b0', cor_grat, cor_grat]

bars = ax2.bar(labels, gaps, color=cores, alpha=0.85, width=0.5)
ax2.set_ylabel('Vantagem da gratidão no afeto positivo\n(diferença de escore)')
ax2.set_ylim(0, 0.75)
ax2.set_title('A dose faz o efeito\nQuanto a gratidão supera o grupo de comparação',
              fontsize=13, pad=15)

for bar, g, sig in zip(bars, gaps, [False, True, True]):
    rotulo = f'{g:.2f}' + ('' if sig else '  (n.s.)')
    ax2.text(bar.get_x() + bar.get_width()/2, g + 0.02,
             rotulo, ha='center', fontweight='bold', fontsize=13)

plt.figtext(0.5, 0.005,
            'Fonte: Emmons & McCullough (2003), JPSP 84(2) | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-028-grafico-02-dose.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 2 salvo!")

# ── GRÁFICO 3 — Efeitos colaterais do registro diário (Estudo 3) ──
fig3, ax3 = plt.subplots(figsize=(10, 6))

medidas = ['Afeto positivo\n(escore)', 'Afeto negativo\n(escore)', 'Sono\n(horas/noite)']
gratidao_vals = [0.35, -0.26, 7.58]
controle_vals = [-0.25, 0.26, 7.06]

# normalizar sono para caber junto? Não — usar dois eixos seria confuso.
# Mostrar afeto (escores) e sono separado via texto. Aqui: só afetos no gráfico principal.
medidas_af = ['Afeto positivo', 'Afeto negativo']
g_af = [0.35, -0.26]
c_af = [-0.25, 0.26]

x = np.arange(len(medidas_af))
largura = 0.35
ax3.bar(x - largura/2, g_af, largura, label='Grupo gratidão', color=cor_grat, alpha=0.85)
ax3.bar(x + largura/2, c_af, largura, label='Grupo controle', color=cor_ctrl, alpha=0.85)
ax3.axhline(y=0, color='#333', linewidth=1)
ax3.set_xticks(x)
ax3.set_xticklabels(medidas_af)
ax3.set_ylabel('Escore padronizado do fator')
ax3.set_title('Estudo 3 — registro diário por 21 dias\nMais afeto positivo, menos afeto negativo (e +31 min de sono/noite)',
              fontsize=12, pad=15)
ax3.legend(frameon=False, loc='upper right')

plt.figtext(0.5, 0.005,
            'Fonte: Emmons & McCullough (2003), Estudo 3, N=65 | sono: 7,58h vs 7,06h | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-028-grafico-03-colaterais.png', dpi=150, bbox_inches='tight')
plt.close()
print("✅ Gráfico 3 salvo!")
```

### 💡 O Insight

**A mesma prática que não moveu o afeto positivo quando era semanal, moveu quando virou diária.**

No Estudo 1, com registro uma vez por semana, o grupo de gratidão teve avaliações de vida melhores e menos queixas físicas — mas o afeto positivo em si não mudou de forma significativa (F=1,73, ns).

Quando o registro passou a ser diário (Estudos 2 e 3), o afeto positivo subiu de forma clara: no Estudo 3, o mais rigoroso, o grupo de gratidão marcou +0,35 no fator de afeto positivo contra −0,25 do controle (d=0,56). O mesmo grupo ainda dormiu mais (7,58h contra 7,06h) e relatou menos afeto negativo.

Não é que a gratidão semanal seja inútil. É que a versão que mexe no como você se sente, no dia a dia, é a diária. A dose importa mais que o gesto.

Foi exatamente o que aconteceu comigo sem eu planejar: não foi agradecer de vez em quando. Foi sentar toda noite e escrever — inclusive nos dias em que parecia que não havia muito o que registrar. O ato de procurar, todo dia, é que fixa.

Uma honestidade necessária: nem todos os efeitos apareceram em todos os estudos. Sintomas físicos e exercício, por exemplo, não se replicaram de forma consistente. O achado sólido é sobre afeto positivo com prática diária — não uma cura para tudo.

*Se você tivesse que anotar agora três motivos pequenos pra agradecer hoje, quantos você lembraria — e quantos teria deixado passar sem escrever?*

---

### ⚠️ Limitações do Modelo

- Os Estudos 1 e 2 usaram universitários; o Estudo 3, adultos com doença neuromuscular. A generalização para "qualquer pessoa" tem limites.
- Nem todos os desfechos se replicaram: sintomas físicos e tempo de exercício apareceram no Estudo 1 mas não nos demais. O achado robusto é o de afeto positivo com registro diário.
- Os autores não sabem por quanto tempo o efeito se sustenta — os estudos duraram de 2 a 3 semanas.
- Auto-relato de bem-estar carrega viés de percepção; o Estudo 3 mitigou isso com relatos de cônjuges, mas é uma medida indireta.

*A ciência é honesta sobre o que não sabe. O modelo também.*
