### O Conceito: o efeito holofote

Em 2000, Thomas Gilovich, Victoria Medvec e Kenneth Savitsky publicaram no Journal of Personality and Social Psychology um experimento que virou classico.

Eles pediram a estudantes de Cornell que vestissem uma camiseta constrangedora, com a foto do Barry Manilow, e entrassem numa sala com outras pessoas. Depois perguntaram: quantos ali conseguiriam dizer quem estava estampado na sua camiseta?

Os alvos estimaram 46%.

Os observadores foram testados de verdade. Acertaram 23%.

Exatamente metade.

O mecanismo, testado nos estudos 4 e 5 do mesmo artigo, e ancoragem e ajuste insuficiente. Voce ancora na sua propria experiencia interna, que e vivida, detalhada e barulhenta, e tenta ajustar para a perspectiva do outro. Mas ajusta pouco. O holofote que voce sente aceso sobre voce e um holofote que so voce enxerga.

O estudo 2 mostrou que isso nao e sobre vergonha. Quando a camiseta era de alguem admiravel, Bob Marley, Martin Luther King, Jerry Seinfeld, a estimativa foi de 48%. Praticamente identica. Ou seja: nao e o constrangimento que infla o numero. E o egocentrismo perceptual, que funciona igual para o que te orgulha e para o que te envergonha.

### O Modelo

A pergunta do dia nao e quantos repararam. E quantas chances existem de **ninguem** ter reparado.

Isso e um problema binomial simples. Se cada observador tem probabilidade p de notar, e existem k observadores independentes, entao:

$$P(\text{ninguem reparou}) = (1 - p)^k$$

Rodo esse modelo duas vezes. Uma com o p que voce acredita (46%) e outra com o p que os dados mediram (23%). A distancia entre as duas curvas e o tamanho da sua ilusao.

**Fonte:**
- Gilovich, T., Medvec, V. H., & Savitsky, K. (2000). The Spotlight Effect in Social Judgment: An Egocentric Bias in Estimates of the Salience of One's Own Actions and Appearance. *Journal of Personality and Social Psychology*, 78(2), 211-222.
  N = 109 participantes no total. Estudo 1: 15 alvos, 64 observadores, 30 controles. Experimento com medida objetiva de reconhecimento.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

print("Bibliotecas carregadas")
```

```python
# --- DADOS DA LITERATURA ---
# Gilovich, Medvec & Savitsky (2000)
# Journal of Personality and Social Psychology, 78(2), 211-222
# Estudo 1: 15 alvos, 64 observadores, 30 controles

n_alvos = 15          # participantes que vestiram a camiseta
n_observadores = 64   # participantes testados de fato

p_estimado = 0.46     # media da estimativa dos alvos (camiseta constrangedora)
p_real = 0.23         # proporcao de observadores que de fato identificaram
p_estimado_positiva = 0.48  # Estudo 2: camiseta admiravel (Marley, MLK, Seinfeld)

razao_ilusao = p_estimado / p_real

# NOTA METODOLOGICA: o fator de correcao 0.80 NAO se aplica aqui.
# Ele vale apenas para proporcoes de survey populacional autorrelatado.
# Este e um experimento com medida objetiva do desfecho: os observadores
# foram testados de verdade, nao perguntados sobre si mesmos.
fator_correcao = None

print("=" * 65)
print("  DADOS DA LITERATURA - EFEITO HOLOFOTE")
print("=" * 65)
print(f"\n  Gilovich, Medvec & Savitsky (2000) - JPSP 78(2)")
print(f"  Estudo 1: {n_alvos} alvos, {n_observadores} observadores")
print(f"\n  Camiseta constrangedora (Barry Manilow):")
print(f"  -> Estimativa dos alvos:              {p_estimado*100:.0f}%")
print(f"  -> Reconhecimento real medido:        {p_real*100:.0f}%")
print(f"  -> Razao da ilusao:                   {razao_ilusao:.1f}x")
print(f"\n  Estudo 2 - camiseta admiravel:")
print(f"  -> Estimativa dos alvos:              {p_estimado_positiva*100:.0f}%")
print(f"  -> Conclusao: o vies independe da valencia")
print(f"\n  Fator 0.80: NAO aplicado (experimento com medida objetiva)")
print("=" * 65)
```

```python
# --- O MODELO ---
# P(ninguem reparou) = (1 - p)^k, para k observadores independentes

# Passo 1 - Distribuicao Beta do reconhecimento real (N=64 observadores)
alpha_real = p_real * n_observadores
beta_real = (1 - p_real) * n_observadores
dist_real = stats.beta(alpha_real, beta_real)
ic_real = dist_real.interval(0.95)

# Passo 2 - Modelo binomial: probabilidade de ninguem ter reparado
ks = np.array([1, 3, 5, 10, 20])
p_ninguem_real = (1 - p_real) ** ks
p_ninguem_crenca = (1 - p_estimado) ** ks

# Passo 3 - Cenario de referencia: uma sala pequena, 5 pessoas
k_ref = 5
p_ninguem_ref = (1 - p_real) ** k_ref
p_ninguem_ref_crenca = (1 - p_estimado) ** k_ref
razao_ref = p_ninguem_ref / p_ninguem_ref_crenca

# Passo 4 - Propagacao da incerteza da Beta para o modelo binomial
amostras_p = dist_real.rvs(20000, random_state=42)
amostras_ninguem = (1 - amostras_p) ** k_ref
ic_ninguem = np.percentile(amostras_ninguem, [2.5, 97.5])

print("=" * 65)
print("  MODELO - EFEITO HOLOFOTE")
print("=" * 65)
print(f"\n  ASSINATURA ESTATISTICA")
print(f"  Beta({alpha_real:.1f}, {beta_real:.1f}) - reconhecimento real")
print(f"  Estimativa central: {p_real*100:.0f}%")
print(f"  IC 95%: [{ic_real[0]*100:.1f}%, {ic_real[1]*100:.1f}%]")
print(f"\n  Note: mesmo no limite superior do IC ({ic_real[1]*100:.1f}%),")
print(f"  o reconhecimento real fica abaixo dos {p_estimado*100:.0f}% estimados.")
print(f"\n  P(NINGUEM REPAROU) = (1 - p)^k")
print(f"\n  {'k':>4} | {'com dado real':>14} | {'com sua crenca':>15}")
print(f"  {'-'*4}-+-{'-'*14}-+-{'-'*15}")
for k, pr, pc in zip(ks, p_ninguem_real, p_ninguem_crenca):
    print(f"  {k:>4} | {pr*100:>13.1f}% | {pc*100:>14.1f}%")
print(f"\n  CENARIO DE REFERENCIA - sala com {k_ref} pessoas:")
print(f"  -> P(ninguem reparou) real:      {p_ninguem_ref*100:.1f}%")
print(f"     IC 95%: [{ic_ninguem[0]*100:.1f}%, {ic_ninguem[1]*100:.1f}%]")
print(f"  -> P(ninguem reparou) na sua cabeca: {p_ninguem_ref_crenca*100:.1f}%")
print(f"  -> Voce subestima essa chance em {razao_ref:.1f}x")
print("=" * 65)
```

```python
# --- VISUALIZACAO ---

# GRAFICO 1 - Estimado vs real
fig1, ax1 = plt.subplots(figsize=(10, 6))

categorias = ['O quanto voce acha\nque repararam',
              'O quanto de fato\nrepararam']
valores = [p_estimado * 100, p_real * 100]
cores = ['#c0392b', '#2a8a82']

bars = ax1.bar(categorias, valores, color=cores, alpha=0.85, width=0.4)
ax1.set_ylim(0, 60)
ax1.set_ylabel('Proporcao de observadores (%)')
ax1.set_title('O Efeito Holofote\nGilovich, Medvec & Savitsky, 2000 - JPSP (15 alvos, 64 observadores)',
              fontsize=13, pad=15)

for bar, valor in zip(bars, valores):
    ax1.text(bar.get_x() + bar.get_width()/2, valor + 1,
             f'{valor:.0f}%', ha='center', fontweight='bold', fontsize=18)

ax1.annotate('', xy=(1, p_real*100), xytext=(0, p_estimado*100),
             arrowprops=dict(arrowstyle='->', color='#c8a84b', lw=2))
ax1.text(0.5, 0.62, f'{razao_ilusao:.1f}x mais\ndo que a realidade', ha='center',
         fontsize=12, color='#c8a84b', fontweight='bold',
         transform=ax1.transAxes)

plt.figtext(0.5, 0.005,
            'Fonte: Gilovich et al., 2000 - Journal of Personality and Social Psychology | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-038-grafico-01-holofote.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 1 salvo")

# GRAFICO 2 - P(ninguem reparou) pelo tamanho da plateia
fig2, ax2 = plt.subplots(figsize=(10, 6))

k_grid = np.arange(1, 21)
curva_real = (1 - p_real) ** k_grid * 100
curva_crenca = (1 - p_estimado) ** k_grid * 100

ax2.plot(k_grid, curva_real, color='#2a8a82', linewidth=3,
         label=f'Com o dado real (p = {p_real*100:.0f}%)')
ax2.plot(k_grid, curva_crenca, color='#c0392b', linewidth=3, linestyle='--',
         label=f'Com a sua crenca (p = {p_estimado*100:.0f}%)')
ax2.fill_between(k_grid, curva_crenca, curva_real, alpha=0.15, color='#c8a84b')

ax2.scatter([k_ref], [p_ninguem_ref*100], color='#2a8a82', s=90, zorder=5)
ax2.scatter([k_ref], [p_ninguem_ref_crenca*100], color='#c0392b', s=90, zorder=5)
ax2.annotate(f'{p_ninguem_ref*100:.0f}%',
             xy=(k_ref, p_ninguem_ref*100), xytext=(k_ref+1.2, p_ninguem_ref*100+7),
             fontsize=13, fontweight='bold', color='#2a8a82')
ax2.annotate(f'{p_ninguem_ref_crenca*100:.0f}%',
             xy=(k_ref, p_ninguem_ref_crenca*100), xytext=(k_ref+1.2, p_ninguem_ref_crenca*100-9),
             fontsize=13, fontweight='bold', color='#c0392b')

ax2.set_xlabel('Numero de pessoas na sala (k)')
ax2.set_ylabel('P(ninguem reparou) em %')
ax2.set_title('A probabilidade de ninguem ter reparado\n$P = (1 - p)^k$ - modelo binomial',
              fontsize=13, pad=15)
ax2.set_xticks(np.arange(1, 21, 2))
ax2.legend(frameon=False, fontsize=10)

plt.figtext(0.5, 0.005,
            'Modelo binomial sobre Gilovich et al., 2000 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-038-grafico-02-binomial.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 2 salvo")

# GRAFICO 3 - Assinatura estatistica: Beta com IC 95%
fig3, ax3 = plt.subplots(figsize=(10, 6))

x = np.linspace(0.05, 0.60, 1000)
y = dist_real.pdf(x)

ax3.plot(x * 100, y, color='#2a8a82', linewidth=3)
ax3.fill_between(x * 100, y, alpha=0.2, color='#2a8a82')
ax3.axvline(x=ic_real[0]*100, color='#c8a84b', linestyle='--', linewidth=2)
ax3.axvline(x=ic_real[1]*100, color='#c8a84b', linestyle='--', linewidth=2)
ax3.axvline(x=p_estimado*100, color='#c0392b', linewidth=2.5)

ax3.text(p_estimado*100 + 0.6, max(y)*0.80,
         f'O que voce\nacredita: {p_estimado*100:.0f}%',
         fontsize=11, color='#c0392b', fontweight='bold')
ax3.text(ic_real[0]*100 - 3.5, max(y)*0.45,
         f'IC 95%\n[{ic_real[0]*100:.1f}%, {ic_real[1]*100:.1f}%]',
         fontsize=10, color='#c8a84b', ha='right')

ax3.set_xlabel('Proporcao de observadores que de fato identificaram (%)')
ax3.set_ylabel('Densidade')
ax3.set_title('Assinatura estatistica - Distribuicao Beta do reconhecimento real\n'
              f'Beta({alpha_real:.1f}, {beta_real:.1f}) - Gilovich et al., 2000 (64 observadores)',
              fontsize=12, pad=15)

plt.figtext(0.5, 0.005,
            'Fonte: Gilovich et al., 2000 - JPSP 78(2), 211-222 | #365Probabilidades',
            ha='center', fontsize=9, color='gray')
plt.tight_layout()
plt.savefig('dia-038-grafico-03-bayesiano.png', dpi=150, bbox_inches='tight')
plt.close()
print("Grafico 3 salvo")
```

### O Insight

**Voce acha que 46% repararam. Repararam 23%.**

Exatamente metade. E essa metade e a parte facil.

A parte que muda a decisao e outra. Numa sala com cinco pessoas, a probabilidade de **ninguem** ter reparado naquilo que te constrangeu e de **27%**, com IC 95% entre 13% e 48%.

Na sua cabeca, essa mesma probabilidade e de 5%.

Voce subestima em quase seis vezes a chance de ninguem ter visto nada.

E tem um detalhe que desmonta a explicacao facil. Quando a camiseta era de alguem admiravel, a superestimacao foi identica: 48%. Ou seja, isso nao e vergonha, nao e inseguranca, nao e falta de autoestima. E um erro de calculo perceptual. Voce ancora na sua experiencia interna, que e altissima resolucao, e ajusta pouco demais para a perspectiva de quem esta do lado de fora.

O holofote esta aceso. So que a lampada esta dentro de voce.

### Limitacoes do Modelo

- O fator de correcao 0.80 nao foi aplicado. Ele vale apenas para proporcoes de survey populacional autorrelatado, e este e um experimento com medida objetiva do desfecho: os observadores foram testados de fato.
- N pequeno na condicao de alvos (15 participantes). O IC 95% foi calculado sobre os 64 observadores, que e a medida objetiva. A estimativa de 46% e uma media de 15 autorrelatos, e o desvio-padrao nao esta disponivel publicamente, por isso ela entra no modelo como ponto, sem intervalo.
- O modelo binomial assume observadores independentes e mesma probabilidade p para cada um. Na vida real ha contagio de atencao: se uma pessoa olha, outras olham.
- O desfecho medido foi reconhecimento sob teste direto, nao percepcao espontanea no cotidiano. Isso torna o 23% uma estimativa conservadora: fora do laboratorio, ninguem para para te examinar.
- Amostra de universitarios norte-americanos em 2000. Generalizacao para outros contextos culturais nao esta testada aqui.
